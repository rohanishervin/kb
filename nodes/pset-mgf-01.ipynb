# Problem Set: Moment Generating Functions



## Problem 1: MGF and Moments of the Binomial Distribution

**Problem Statement:**

Let $X \sim \text{Bin}(n, p)$, where $n \in \mathbb{N}$ and $p \in (0, 1)$ with $q = 1 - p$.

1. Derive the [moment generating function](moment-generating-functions.ipynb) $M_X(t) = \mathbb{E}[e^{tX}]$.  
2. Using $M_X(t)$, compute the first moment $\mathbb{E}[X]$ and the second moment $\mathbb{E}[X^2]$.  
3. Determine the [variance](variance.ipynb) $\text{Var}(X) = \mathbb{E}[X^2] - (\mathbb{E}[X])^2$.

::: {.callout-tip collapse="true"}
## Solution

**1. Derivation of the MGF:**

The [probability mass function](probability-mass-function.ipynb) of $X$ is:

$$
P(X = x) = \binom{n}{x} p^x (1-p)^{n-x}, \quad x \in \{0, 1, \dots, n\}
$$

By definition of the [moment generating function](moment-generating-functions.ipynb):

$$
\begin{aligned}
M_X(t) &= \mathbb{E}[e^{tX}] = \sum_{x=0}^n e^{tx} P(X = x) \\
&= \sum_{x=0}^n e^{tx} \binom{n}{x} p^x (1-p)^{n-x} \\
&= \sum_{x=0}^n \binom{n}{x} (p e^t)^x (1-p)^{n-x}
\end{aligned}
$$

Applying the [Binomial Theorem](binomial-theorem.ipynb) $\sum_{x=0}^n \binom{n}{x} a^x b^{n-x} = (a + b)^n$ with $a = p e^t$ and $b = 1 - p = q$:

$$
M_X(t) = (p e^t + 1 - p)^n = (p e^t + q)^n
$$

**2. First and Second Moments:**

The moments are obtained by differentiating $M_X(t)$ and evaluating at $t = 0$:

- **First Moment $\mathbb{E}[X]$:**

  $$
  M_X'(t) = \frac{d}{dt} (p e^t + q)^n = n(p e^t + q)^{n-1} \cdot p e^t
  $$

  Evaluating at $t = 0$:

  $$
  \mathbb{E}[X] = M_X'(0) = n(p e^0 + q)^{n-1} p e^0 = n(p + q)^{n-1} p = n(1)^{n-1} p = np
  $$

- **Second Moment $\mathbb{E}[X^2]$:**

  Applying the product rule to $M_X'(t) = n(p e^t + q)^{n-1} (p e^t)$:

  $$
  M_X''(t) = n(n-1)(p e^t + q)^{n-2} (p e^t)^2 + n(p e^t + q)^{n-1} (p e^t)
  $$

  Evaluating at $t = 0$:

  $$
  \begin{aligned}
  \mathbb{E}[X^2] = M_X''(0) &= n(n-1)(p + q)^{n-2} p^2 + n(p + q)^{n-1} p \\
  &= n(n-1)p^2 + np \\
  &= n^2 p^2 - np^2 + np
  \end{aligned}
  $$

**3. Variance:**

$$
\begin{aligned}
\text{Var}(X) &= \mathbb{E}[X^2] - (\mathbb{E}[X])^2 \\
&= (n^2 p^2 - np^2 + np) - (np)^2 \\
&= np - np^2 \\
&= np(1 - p) = npq
\end{aligned}
$$

$\blacksquare$
:::



## Problem 2: MGF and Moments of the Poisson Distribution

**Problem Statement:**

Let $X \sim \text{Poisson}(\lambda)$ with parameter $\lambda > 0$.

1. Derive the [moment generating function](moment-generating-functions.ipynb) $M_X(t) = \mathbb{E}[e^{tX}]$.  
2. Using $M_X(t)$, compute the first moment $\mathbb{E}[X]$ and the second moment $\mathbb{E}[X^2]$.  
3. Verify that $\mathbb{E}[X] = \lambda$ and $\text{Var}(X) = \lambda$.

::: {.callout-tip collapse="true"}
## Solution

**1. Derivation of the MGF:**

The [probability mass function](probability-mass-function.ipynb) of $X$ is:

$$
P(X = x) = \frac{e^{-\lambda} \lambda^x}{x!}, \quad x \in \{0, 1, 2, \dots\}
$$

By definition of the [moment generating function](moment-generating-functions.ipynb):

$$
M_X(t) = \mathbb{E}[e^{tX}] = \sum_{x=0}^\infty e^{tx} \frac{e^{-\lambda} \lambda^x}{x!} = \sum_{x=0}^\infty \frac{(\lambda e^t)^x e^{-\lambda}}{x!}
$$

Recognizing that for any parameter $\mu > 0$, $\sum_{x=0}^\infty \frac{\mu^x e^{-\mu}}{x!} = 1 \implies \sum_{x=0}^\infty \frac{\mu^x}{x!} = e^\mu$, we set $\mu = \lambda e^t$:

$$
\sum_{x=0}^\infty \frac{(\lambda e^t)^x e^{-\lambda e^t}}{x!} = 1 \implies \sum_{x=0}^\infty \frac{(\lambda e^t)^x e^{-\lambda}}{x!} = \frac{e^{-\lambda}}{e^{-\lambda e^t}} = e^{\lambda(e^t - 1)}
$$

Therefore:

$$
M_X(t) = e^{\lambda(e^t - 1)}
$$

**2. First and Second Moments:**

- **First Moment $\mathbb{E}[X]$:**

  $$
  M_X'(t) = \frac{d}{dt} e^{\lambda(e^t - 1)} = \lambda e^t e^{\lambda(e^t - 1)}
  $$

  Evaluating at $t = 0$:

  $$
  \mathbb{E}[X] = M_X'(0) = \lambda e^0 e^{\lambda(e^0 - 1)} = \lambda
  $$

- **Second Moment $\mathbb{E}[X^2]$:**

  Differentiating $M_X'(t)$ using the product rule:

  $$
  M_X''(t) = \frac{d}{dt}\left( \lambda e^t e^{\lambda(e^t - 1)} \right) = \lambda e^t e^{\lambda(e^t - 1)} + (\lambda e^t)^2 e^{\lambda(e^t - 1)}
  $$

  Evaluating at $t = 0$:

  $$
  \mathbb{E}[X^2] = M_X''(0) = \lambda e^0 e^0 + (\lambda e^0)^2 e^0 = \lambda + \lambda^2
  $$

**3. Variance:**

$$
\text{Var}(X) = \mathbb{E}[X^2] - (\mathbb{E}[X])^2 = (\lambda + \lambda^2) - \lambda^2 = \lambda
$$

Thus, $\mathbb{E}[X] = \lambda$ and $\text{Var}(X) = \lambda$.

$\blacksquare$
:::



## Problem 3: MGF and Moments of the Exponential Distribution

**Problem Statement:**

Let $X \sim \text{Expo}(\lambda)$ with rate parameter $\lambda > 0$, having [probability density function](probability-density-function.ipynb) $f_X(x) = \lambda e^{-\lambda x}$ for $x \ge 0$.

1. Derive the [moment generating function](moment-generating-functions.ipynb) $M_X(t) = \mathbb{E}[e^{tX}]$, stating its domain of convergence.  
2. Using $M_X(t)$, compute the first moment $\mathbb{E}[X]$ and the second moment $\mathbb{E}[X^2]$.  
3. Verify that $\mathbb{E}[X] = \frac{1}{\lambda}$ and $\text{Var}(X) = \frac{1}{\lambda^2}$.

::: {.callout-tip collapse="true"}
## Solution

**1. Derivation of the MGF:**

By definition of the [moment generating function](moment-generating-functions.ipynb):

$$
M_X(t) = \mathbb{E}[e^{tX}] = \int_0^\infty e^{tx} \lambda e^{-\lambda x} \, dx = \int_0^\infty \lambda e^{-(\lambda - t)x} \, dx
$$

For the integral to converge, the exponent must remain negative as $x \to \infty$, which requires $\lambda - t > 0$, or equivalently $t < \lambda$. Under this condition:

$$
M_X(t) = \frac{\lambda}{\lambda - t} \int_0^\infty (\lambda - t) e^{-(\lambda - t)x} \, dx
$$

Since $(\lambda - t) e^{-(\lambda - t)x}$ is the PDF of an $\text{Expo}(\lambda - t)$ distribution on $[0, \infty)$, its integral over the entire support equals $1$:

$$
M_X(t) = \frac{\lambda}{\lambda - t}, \quad \text{for } t < \lambda
$$

**2. First and Second Moments:**

- **First Moment $\mathbb{E}[X]$:**

  $$
  M_X'(t) = \frac{d}{dt} \left[ \lambda (\lambda - t)^{-1} \right] = \lambda (-1)(\lambda - t)^{-2}(-1) = \frac{\lambda}{(\lambda - t)^2}
  $$

  Evaluating at $t = 0$:

  $$
  \mathbb{E}[X] = M_X'(0) = \frac{\lambda}{(\lambda - 0)^2} = \frac{\lambda}{\lambda^2} = \frac{1}{\lambda}
  $$

- **Second Moment $\mathbb{E}[X^2]$:**

  $$
  M_X''(t) = \frac{d}{dt} \left[ \lambda (\lambda - t)^{-2} \right] = \lambda (-2)(\lambda - t)^{-3}(-1) = \frac{2\lambda}{(\lambda - t)^3}
  $$

  Evaluating at $t = 0$:

  $$
  \mathbb{E}[X^2] = M_X''(0) = \frac{2\lambda}{(\lambda - 0)^3} = \frac{2\lambda}{\lambda^3} = \frac{2}{\lambda^2}
  $$

**3. Variance:**

$$
\text{Var}(X) = \mathbb{E}[X^2] - (\mathbb{E}[X])^2 = \frac{2}{\lambda^2} - \left( \frac{1}{\lambda} \right)^2 = \frac{2}{\lambda^2} - \frac{1}{\lambda^2} = \frac{1}{\lambda^2}
$$

$\blacksquare$
:::



## Problem 4: MGF and Moments of the Normal Distribution

**Problem Statement:**

**(a)** Let $Z \sim \mathcal{N}(0, 1)$ be a standard [Normal distribution](normal-distribution.ipynb) with PDF $\varphi(z) = \frac{1}{\sqrt{2\pi}} e^{-z^2 / 2}$.  
1. Derive the [moment generating function](moment-generating-functions.ipynb) $M_Z(t) = \mathbb{E}[e^{tZ}]$.  
2. Using $M_Z(t)$, compute $\mathbb{E}[Z]$ and verify that $\text{Var}(Z) = 1$.

**(b)** Let $X \sim \mathcal{N}(\mu, \sigma^2)$. Express $X = \mu + \sigma Z$ and use the properties of moment generating functions to determine $M_X(t)$.

::: {.callout-tip collapse="true"}
## Solution

**(a)**

**1. Derivation of $M_Z(t)$:**

By definition of the [moment generating function](moment-generating-functions.ipynb):

$$
M_Z(t) = \mathbb{E}[e^{tZ}] = \int_{-\infty}^\infty e^{tz} \frac{1}{\sqrt{2\pi}} e^{-z^2 / 2} \, dz = \frac{1}{\sqrt{2\pi}} \int_{-\infty}^\infty \exp\left( -\frac{z^2 - 2tz}{2} \right) \, dz
$$

Completing the square in the exponent:

$$
-\frac{z^2 - 2tz}{2} = -\frac{(z - t)^2 - t^2}{2} = -\frac{(z - t)^2}{2} + \frac{t^2}{2}
$$

Factoring out the term $e^{t^2 / 2}$ (which does not depend on $z$):

$$
M_Z(t) = e^{t^2 / 2} \int_{-\infty}^\infty \frac{1}{\sqrt{2\pi}} e^{-(z - t)^2 / 2} \, dz
$$

The remaining integrand is the PDF of a $\mathcal{N}(t, 1)$ random variable, so its integral over $\mathbb{R}$ equals $1$:

$$
M_Z(t) = e^{t^2 / 2}
$$

**2. Moments and Variance:**

- **First Moment $\mathbb{E}[Z]$:**

  $$
  M_Z'(t) = \frac{d}{dt} e^{t^2 / 2} = t e^{t^2 / 2}
  $$

  Evaluating at $t = 0$:

  $$
  \mathbb{E}[Z] = M_Z'(0) = 0 \cdot e^0 = 0
  $$

- **Second Moment $\mathbb{E}[Z^2]$:**

  $$
  M_Z''(t) = \frac{d}{dt} \left( t e^{t^2 / 2} \right) = e^{t^2 / 2} + t^2 e^{t^2 / 2}
  $$

  Evaluating at $t = 0$:

  $$
  \mathbb{E}[Z^2] = M_Z''(0) = e^0 + 0^2 e^0 = 1
  $$

- **Variance:**

  $$
  \text{Var}(Z) = \mathbb{E}[Z^2] - (\mathbb{E}[Z])^2 = 1 - 0^2 = 1
  $$

**(b)**

Any normal random variable $X \sim \mathcal{N}(\mu, \sigma^2)$ can be expressed as $X = \mu + \sigma Z$, where $Z \sim \mathcal{N}(0, 1)$.  
Using the linear transformation property of MGFs ($M_{aZ + b}(t) = e^{bt} M_Z(at)$):

$$
M_X(t) = \mathbb{E}[e^{t(\mu + \sigma Z)}] = \mathbb{E}[e^{\mu t} e^{(\sigma t) Z}] = e^{\mu t} \mathbb{E}[e^{(\sigma t) Z}] = e^{\mu t} M_Z(\sigma t)
$$

Substituting $M_Z(s) = e^{s^2 / 2}$ with $s = \sigma t$:

$$
M_X(t) = e^{\mu t} e^{(\sigma t)^2 / 2} = e^{\mu t + \frac{1}{2}\sigma^2 t^2}
$$

$\blacksquare$
:::
