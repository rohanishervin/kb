# Multiple Integrals

**Prerequisites:** [The Riemann Integral](riemann-integral.ipynb) · [Properties of the Riemann Integral](properties-of-the-riemann-integral.ipynb) · [Uniform Continuity and Compactness](uniform-continuity.ipynb) · [Compactness](compactness.ipynb) · [The Intermediate Value Theorem](intermediate-value-theorem.ipynb)

The [Riemann integral](riemann-integral.ipynb) of one variable generalizes by replacing intervals with boxes and lengths with volumes. For $f \ge 0$ on a region of the plane, the integral is the volume under the graph; if $f$ is a density, it is the total mass.

## 1\. Integrals over Boxes

::: {#def-box-partition}
## Boxes and Partitions

A (closed) **box** is $R = [a_1, b_1]\times\dots\times[a_n, b_n]$, with **volume** $\operatorname{vol}(R) = \prod_i(b_i - a_i)$ (length, area, volume for $n = 1, 2, 3$). A **partition** $\mathcal{P}$ of $R$ consists of partitions of each $[a_i, b_i]$; it divides $R$ into finitely many **subboxes** $Q$, whose volumes add up to $\operatorname{vol}(R)$ (expand the product $\prod_i\sum_j\Delta x_{ij}$). The **mesh** $\|\mathcal{P}\|$ is the largest subinterval length, and $\mathcal{P}'$ **refines** $\mathcal{P}$ if each of its coordinate partitions refines that of $\mathcal{P}$.
:::

::: {#def-multiple-integral}
## Darboux Sums and the Integral

For bounded $f : R \to \mathbb{R}$ let $m_Q = \inf_Qf$, $M_Q = \sup_Qf$ and
$$L(f, \mathcal{P}) = \sum_Qm_Q\operatorname{vol}(Q), \qquad U(f, \mathcal{P}) = \sum_QM_Q\operatorname{vol}(Q).$$
As in one variable, refining raises $L$ and lowers $U$, and $L(f, \mathcal{P}) \le U(f, \mathcal{Q})$ for any two partitions ([same proof](riemann-integral.ipynb#lem-refinement), with subintervals replaced by subboxes). $f$ is **integrable** on $R$ if $\sup_{\mathcal{P}}L(f, \mathcal{P}) = \inf_{\mathcal{P}}U(f, \mathcal{P})$; the common value is $\int_Rf$, also written $\int\cdots\int_Rf\,dV$ or $\iint_Rf\,dA$ for $n = 2$. Equivalently ([Riemann's criterion](riemann-integral.ipynb#thm-riemann-criterion)), for each $\varepsilon > 0$ there is $\mathcal{P}$ with $U(f, \mathcal{P}) - L(f, \mathcal{P}) < \varepsilon$.
:::

The proofs of the [elementary properties](properties-of-the-riemann-integral.ipynb) carry over word for word: the integral is linear, monotone ($f \le g \Rightarrow \int f \le \int g$), $|\int f| \le \int|f|$, $\int_R k = k\operatorname{vol}(R)$, and it is additive when $R$ is split into two boxes along a hyperplane $x_i = c$. Also, Riemann sums $\sum_Qf(\xi_Q)\operatorname{vol}(Q)$, $\xi_Q \in Q$, tend to $\int_Rf$ as $\|\mathcal{P}\| \to 0$, as in [one variable](properties-of-the-riemann-integral.ipynb#thm-riemann-sums).

::: {#thm-continuous-integrable-box}
## Continuous Functions Are Integrable

Every continuous $f : R \to \mathbb{R}$ on a box is integrable.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-continuous-integrable-box
$R$ is compact, so $f$ is [uniformly continuous](uniform-continuity.ipynb): given $\varepsilon > 0$ there is $\delta > 0$ with $|f(x) - f(y)| < \varepsilon/\operatorname{vol}(R)$ when $|x - y| < \delta$. Choose $\mathcal{P}$ with $\sqrt n\,\|\mathcal{P}\| < \delta$; then any two points of a subbox are within $\delta$, so $M_Q - m_Q \le \varepsilon/\operatorname{vol}(R)$ and $U - L \le \varepsilon$. $\blacksquare$
:::

## 2\. Sets of Measure Zero

::: {#def-measure-zero}
## Measure Zero

$Z \subseteq \mathbb{R}^n$ has **($n$-dimensional) measure zero** if for every $\varepsilon > 0$ there are boxes $R_1, R_2, \dots$ (finitely or countably many) with $Z \subseteq \bigcup_iR_i$ and $\sum_i\operatorname{vol}(R_i) < \varepsilon$.
:::

::: {#prp-measure-zero-examples}
## Examples of Measure Zero

1. Subsets of measure-zero sets, finite sets and countable sets have measure zero; so does a countable union of measure-zero sets.
2. The graph $\{(x, \varphi(x)) : x \in R'\}$ of a continuous function on a box $R' \subseteq \mathbb{R}^{n-1}$ has measure zero in $\mathbb{R}^n$; so do subsets of hyperplanes and spheres.
3. If $Z$ is compact and has measure zero, it can be covered by **finitely many** boxes of total volume $< \varepsilon$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-measure-zero-examples
(1) Cover the $i$-th point (or set) with boxes of total volume $< \varepsilon/2^i$; the total is $< \varepsilon$.
(2) Let $V = \operatorname{vol}(R')$. By uniform continuity choose a partition of $R'$ into subboxes $R'_j$ on which $\varphi$ varies by less than $\varepsilon/V$; the graph over $R'_j$ lies in the box $R'_j\times[m_j, m_j + \varepsilon/V]$, and these have total volume $\sum_j\operatorname{vol}(R'_j)\varepsilon/V = \varepsilon$. A bounded piece of a hyperplane lies in the graph of an affine function over a box (after renumbering coordinates); the sphere $|x| = r$ lies in the union of the graphs of $\pm\sqrt{\max(r^2 - |x'|^2, 0)}$, continuous on the box $[-r, r]^{n-1}$. Unbounded hyperplanes are countable unions of bounded pieces.
(3) Enlarge each covering box $R_i$ to an open box of volume $< \operatorname{vol}(R_i) + \varepsilon/2^i$; finitely many of these cover the compact $Z$, with total volume $< 2\varepsilon$. $\blacksquare$
:::

::: {#thm-integrable-off-measure-zero}
## Integrability Criterion

Let $f$ be bounded on a box $R$ and continuous at every point of $R\setminus Z$, where $Z$ is compact of measure zero. Then $f$ is integrable on $R$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-integrable-off-measure-zero
Let $|f| \le M$ and $\varepsilon > 0$. By @prp-measure-zero-examples (3), $Z$ is covered by finitely many open boxes $B_1, \dots, B_k$ of total volume $< \varepsilon$. Let $\mathcal{P}_0$ be the partition of $R$ generated by the coordinates of the corners of the $B_j$ (clipped to $R$). Every subbox of $\mathcal{P}_0$ meeting some $B_j$ lies in $\overline B_j$; these subboxes have total volume $< \varepsilon$ and contribute at most $2M\varepsilon$ to $U - L$. The other subboxes are disjoint from $\bigcup_jB_j \supseteq Z$; their union $K$ is compact, $f$ is continuous on $K$, hence uniformly continuous there, and refining $\mathcal{P}_0$ as in the proof of @thm-continuous-integrable-box makes their contribution to $U - L$ at most $\varepsilon$. So $U - L < (2M + 1)\varepsilon$ for a suitable partition. $\blacksquare$
:::

The general statement, **Lebesgue's criterion** (a bounded $f$ on a box is integrable iff its set of discontinuity points has measure zero), is stated without proof, as in [one variable](riemann-integral.ipynb#thm-lebesgue-criterion).

## 3\. Integrals over Regions

::: {#def-admissible-region}
## Admissible Regions and Volume

A bounded set $D \subseteq \mathbb{R}^n$ is an **admissible region** if its boundary $\partial D$ has measure zero. For bounded $f : D \to \mathbb{R}$ let $\tilde f = f$ on $D$ and $\tilde f = 0$ outside $D$, and define
$$\int_Df = \int_R\tilde f$$
for a box $R \supseteq D$, if this integral exists (it does not depend on $R$, by additivity, since $\tilde f = 0$ on the added boxes). The **volume** of $D$ is $\operatorname{vol}(D) = \int_D1$.
:::

::: {#thm-continuous-on-region}
## Continuous Functions on Admissible Regions

If $D$ is admissible and $f$ is bounded and continuous on $D$, then $\int_Df$ exists. In particular $\operatorname{vol}(D)$ exists.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-continuous-on-region
At interior points of $D$, $\tilde f = f$ near the point, and at points outside $\overline D$, $\tilde f = 0$ near the point; so $\tilde f$ is continuous off $\partial D$, which is closed and bounded (compact) of measure zero. Apply @thm-integrable-off-measure-zero. $\blacksquare$
:::

By @prp-measure-zero-examples, balls, boxes, and regions bounded by finitely many graphs of continuous functions (such as $\{a \le x \le b,\ \alpha(x) \le y \le \beta(x)\}$ with $\alpha, \beta$ continuous) are admissible.

::: {#prp-region-integral-properties}
## Properties on Regions

For admissible $D$ and $f, g$ integrable on $D$: $\int_D(f + g) = \int_Df + \int_Dg$, $\int_Dcf = c\int_Df$; $f \le g$ implies $\int_Df \le \int_Dg$; $\int_Dk = k\operatorname{vol}(D)$. If $D = D'\cup D''$ with $D'$, $D''$ admissible and $D'\cap D''$ of measure zero, and $f$ is continuous and bounded on $D$, then $\int_Df = \int_{D'}f + \int_{D''}f$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-region-integral-properties
The first three follow from the box case applied to $\tilde f, \tilde g$. For additivity, $\tilde f = \tilde f_{D'} + \tilde f_{D''} - \tilde f_{D'\cap D''}$, where each term is $f$ on the indicated set and $0$ elsewhere. The last term vanishes off $Z = D'\cap D''$. A set of measure zero contains no ball, so $Z \subseteq \partial D'\cup\partial D''$, a compact set of measure zero; covering it by finitely many boxes of total volume $< \varepsilon$ (@prp-measure-zero-examples) and using their corners in the partition, the lower and upper sums of this term lie within $\pm M\varepsilon$ ($|f| \le M$). So it is integrable with integral $0$. $\blacksquare$
:::

::: {#thm-mean-value-multiple}
## Mean Value Theorem for Integrals

Let $D$ be a compact admissible region that is **path-connected** (any two points are joined by a continuous curve in $D$) and $f : D \to \mathbb{R}$ continuous. Then $\int_Df = f(P)\operatorname{vol}(D)$ for some $P \in D$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-mean-value-multiple
If $\operatorname{vol}(D) = 0$ any $P$ works. Otherwise, $f$ attains its minimum $m = f(A)$ and maximum $M = f(B)$ on the compact $D$, and by monotonicity $m \le \frac{1}{\operatorname{vol}(D)}\int_Df \le M$. Join $A$ to $B$ by a continuous curve $\gamma : [0, 1] \to D$; by the [IVT](intermediate-value-theorem.ipynb) applied to $f\circ\gamma$ there is $c$ with $f(\gamma(c)) = \frac{1}{\operatorname{vol}(D)}\int_Df$; take $P = \gamma(c)$. $\blacksquare$
:::

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

f = lambda x, y: 1 - (x**2 + y**2) / 2          # decreasing in x and y on [0,1]^2
for k in [4, 16, 64, 256]:
    g = np.linspace(0, 1, k + 1); h = 1 / k
    X, Y = np.meshgrid(g[:-1], g[:-1])
    U, L = np.sum(f(X, Y)) * h * h, np.sum(f(X + h, Y + h)) * h * h
    print(f"k = {k:3d}: L = {L:.6f}, U = {U:.6f}, U - L = {U - L:.6f}")
print("exact value: 2/3 =", 2 / 3)

fig = plt.figure(figsize=(7, 5.2))
ax = fig.add_subplot(projection="3d")
k = 6; g = np.linspace(0, 1, k + 1); h = 1 / k
X, Y = np.meshgrid(g[:-1], g[:-1])
ax.bar3d(X.ravel(), Y.ravel(), 0 * X.ravel(), h, h, f(X, Y).ravel(), color=BLUE, alpha=0.35, edgecolor="white", linewidth=0.3)
s = np.linspace(0, 1, 30); SX, SY = np.meshgrid(s, s)
ax.plot_surface(SX, SY, f(SX, SY), color=ORANGE, alpha=0.6, linewidth=0)
ax.set_title(r"Upper sum for $1 - (x^2 + y^2)/2$ on $[0,1]^2$, $6\times6$ partition")
plt.tight_layout(); plt.show()
```
