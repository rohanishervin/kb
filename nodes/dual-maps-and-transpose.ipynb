# Dual Maps and the Transpose

**Prerequisites:** [Dual Spaces](dual-spaces.ipynb) · [Annihilators](annihilators.ipynb) · [The Matrix of a Linear Map](matrix-of-a-linear-map.ipynb)

A linear map $f : V \to W$ moves vectors forward, from $V$ to $W$. It also moves functionals **backward**: a functional on $W$ becomes a functional on $V$ by first applying $f$. This backward map $f^*$ is the **dual map**, and in dual bases its matrix is the **transpose** of the matrix of $f$. Duality then explains a fact that looks like a coincidence in coordinates: the rows and the columns of any matrix span spaces of the same dimension.

Throughout, $V$ and $W$ are finite-dimensional [vector spaces](vector-spaces.ipynb) over a field $K$ with [duals](dual-spaces.ipynb#def-dual-space) $V^*$ and $W^*$, and $f : V \to W$ is a [linear map](linear-maps.ipynb#def-linear-map).

## 1\. The Dual Map

::: {#def-dual-map}
## Dual Map

The **dual map** (or **transpose**) of $f$ is
$$f^* : W^* \to V^*, \qquad f^*(\psi) = \psi \circ f,$$
that is, $(f^*\psi)(v) = \psi(f(v))$ for $\psi \in W^*$ and $v \in V$.
:::

::: {.callout-note}
## Intuition: Pulling Back a Measurement
$\psi$ is a way of measuring vectors of $W$. To measure a vector $v \in V$ with it, first send $v$ to $W$ by $f$, then measure. This is why $f^*$ goes in the opposite direction to $f$.
:::

::: {#prp-dual-map-linear}
## The Dual Map Is Well Defined and Linear

$f^*(\psi)$ is a linear functional on $V$, and $f^* : W^* \to V^*$ is linear.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-dual-map-linear
$f^*(\psi) = \psi \circ f$ is a [composite of linear maps](linear-maps.ipynb#prp-composition-linearity), hence linear. For $\psi, \chi \in W^*$, $a \in K$ and $v \in V$:
$$f^*(a\psi + \chi)(v) = (a\psi + \chi)(f(v)) = a\,\psi(f(v)) + \chi(f(v)) = (a f^*(\psi) + f^*(\chi))(v). \qquad \blacksquare$$
:::

::: {#thm-dual-matrix-transpose}
## The Matrix of $f^*$ Is the Transpose

Let $\mathcal{B} = (v_1, \dots, v_n)$ and $\mathcal{C} = (w_1, \dots, w_m)$ be bases of $V$ and $W$, with [dual bases](dual-spaces.ipynb#def-dual-basis) $\mathcal{B}^* = (v_1^*, \dots, v_n^*)$ and $\mathcal{C}^* = (w_1^*, \dots, w_m^*)$. If $A = [f]_{\mathcal{B}}^{\mathcal{C}}$, then
$$[f^*]_{\mathcal{C}^*}^{\mathcal{B}^*} = A^{\mathsf T}.$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-dual-matrix-transpose
Write $A = (a_{ij})$, so $f(v_i) = \sum_l a_{li} w_l$, and let $B = (b_{ij}) = [f^*]_{\mathcal{C}^*}^{\mathcal{B}^*}$, so $f^*(w_j^*) = \sum_k b_{kj} v_k^*$. Evaluate both sides at $v_i$. The right side gives $\sum_k b_{kj} v_k^*(v_i) = b_{ij}$. The left side gives
$$f^*(w_j^*)(v_i) = w_j^*(f(v_i)) = w_j^*\Big(\sum_l a_{li} w_l\Big) = a_{ji}.$$
Hence $b_{ij} = a_{ji}$, i.e. $B = A^{\mathsf T}$. $\blacksquare$
:::

::: {#exm-dual-map}
## Pulling Back a Functional

Let $f : \mathbb{R}^3 \to \mathbb{R}^2$, $f(x, y, z) = (x + 2y,\ y + z)$, with matrix $A = \begin{pmatrix} 1 & 2 & 0 \\ 0 & 1 & 1 \end{pmatrix}$. A functional on $\mathbb{R}^2$ has the form $\psi(y_1, y_2) = p\,y_1 + q\,y_2$. Then
$$f^*(\psi)(x, y, z) = p(x + 2y) + q(y + z) = p\,x + (2p + q)\,y + q\,z.$$
The coefficient vector $(p,\ 2p + q,\ q)$ of $f^*(\psi)$ is $A^{\mathsf T} \binom{p}{q}$, as the theorem predicts.
:::

## 2\. Algebraic Properties

::: {#thm-dual-composition}
## Duals Reverse Composition

For linear maps $V_1 \xrightarrow{f} V_2 \xrightarrow{g} V_3$: $(g \circ f)^* = f^* \circ g^*$. In matrices, $(BA)^{\mathsf T} = A^{\mathsf T} B^{\mathsf T}$. Also $(\mathrm{id}_V)^* = \mathrm{id}_{V^*}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-dual-composition
For $\chi \in V_3^*$: $(g \circ f)^*(\chi) = \chi \circ g \circ f = (g^*\chi) \circ f = f^*(g^*(\chi))$. Clearly $\mathrm{id}^*(\varphi) = \varphi \circ \mathrm{id} = \varphi$. The matrix identity follows by taking matrices in dual bases with @thm-dual-matrix-transpose and the [composition theorem](matrix-of-a-linear-map.ipynb#thm-matrix-multiplication-composition): every pair of matrices $A, B$ of compatible sizes arises this way. $\blacksquare$
:::

::: {#cor-dual-invertibility}
## Inverses and Transposes

If $f$ is invertible, so is $f^*$, and $(f^*)^{-1} = (f^{-1})^*$. For an invertible matrix, $(A^{\mathsf T})^{-1} = (A^{-1})^{\mathsf T}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-dual-invertibility
By @thm-dual-composition, $(f^{-1})^* \circ f^* = (f \circ f^{-1})^* = \mathrm{id}_{W^*}$ and $f^* \circ (f^{-1})^* = (f^{-1} \circ f)^* = \mathrm{id}_{V^*}$. $\blacksquare$
:::

## 3\. Kernels, Images and Annihilators

::: {#thm-kernel-dual-annihilator-image}
## The Kernel of $f^*$

$\ker f^* = (\operatorname{im} f)^\perp$, the [annihilator](annihilators.ipynb#def-annihilator) of the image.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-kernel-dual-annihilator-image
$\psi \in \ker f^*$ $\iff$ $\psi(f(v)) = 0$ for every $v \in V$ $\iff$ $\psi$ vanishes on every element of $\operatorname{im} f$ $\iff$ $\psi \in (\operatorname{im} f)^\perp$. $\blacksquare$
:::

::: {#thm-row-rank-equals-column-rank}
## $f$ and $f^*$ Have the Same Rank

$\operatorname{rank} f^* = \operatorname{rank} f$. In matrix terms: **the row rank of any matrix equals its column rank**, $\operatorname{rank} A^{\mathsf T} = \operatorname{rank} A$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-row-rank-equals-column-rank
Let $m = \dim W = \dim W^*$. By [rank-nullity](kernel-image-and-rank-nullity.ipynb#thm-rank-nullity) for $f^*$, @thm-kernel-dual-annihilator-image and the [dimension of an annihilator](annihilators.ipynb#cor-annihilator-dim),
$$\operatorname{rank} f^* = m - \dim \ker f^* = m - \dim (\operatorname{im} f)^\perp = m - (m - \operatorname{rank} f) = \operatorname{rank} f.$$
For a matrix $A$, the column rank (dimension of the span of the columns) is the rank of $x \mapsto A x$, and the row rank is the column rank of $A^{\mathsf T}$, the matrix of the dual map. $\blacksquare$
:::

::: {.callout-note}
## Why This Is Remarkable
The rows of a $3 \times 100$ matrix live in $K^{100}$, its columns in $K^3$, and nothing about the entries suggests a relation between the two spans. Duality supplies one without row reduction or inner products: the rows describe the dual map, and a map and its dual lose exactly the same number of dimensions.
:::

::: {#thm-image-dual-annihilator-kernel}
## The Image of $f^*$

$\operatorname{im} f^* = (\ker f)^\perp$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-image-dual-annihilator-kernel
**Inclusion.** If $\varphi = f^*(\psi)$ and $v \in \ker f$, then $\varphi(v) = \psi(f(v)) = \psi(0) = 0$; so $\operatorname{im} f^* \subseteq (\ker f)^\perp$.

**Dimensions.** Let $n = \dim V$. By @thm-row-rank-equals-column-rank, rank-nullity and the [dimension of an annihilator](annihilators.ipynb#cor-annihilator-dim),
$$\dim \operatorname{im} f^* = \operatorname{rank} f = n - \dim \ker f = \dim (\ker f)^\perp.$$
A subspace of full dimension [is everything](basis-and-dimension.ipynb#thm-subspace-dimension). $\blacksquare$
:::

::: {.callout-note}
## Linear Systems Read Through Duality
$A x = b$ is solvable iff $b \in \operatorname{im} A$. Applying the double annihilator to @thm-kernel-dual-annihilator-image gives the classical criterion: $A x = b$ is solvable **iff $y^{\mathsf T} b = 0$ for every $y$ with $y^{\mathsf T} A = 0$**, i.e. iff every linear combination of the equations that cancels the left-hand sides also cancels the right-hand side.
:::

::: {#lem-double-dual-naturality}
## Naturality of the Double Dual

With the [natural isomorphisms](dual-spaces.ipynb#thm-natural-iso) $\iota_V : V \to V^{**}$ and $\iota_W : W \to W^{**}$, one has $f^{**} \circ \iota_V = \iota_W \circ f$. So, after identifying each space with its double dual, $f^{**} = f$; for matrices, $(A^{\mathsf T})^{\mathsf T} = A$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-double-dual-naturality
For $v \in V$ and $\psi \in W^*$:
$$f^{**}(\iota_V(v))(\psi) = \iota_V(v)(f^*\psi) = (f^*\psi)(v) = \psi(f(v)) = \iota_W(f(v))(\psi). \qquad \blacksquare$$
:::

## 4\. Rows as Functionals

On $K^n$ it is natural to write vectors as **columns** and functionals as **rows**: the functional $x \mapsto y_1 x_1 + \dots + y_n x_n$ is the row $(y_1, \dots, y_n)$, and evaluating it is the product
$$\begin{pmatrix} y_1 & \cdots & y_n \end{pmatrix}\begin{pmatrix} x_1 \\ \vdots \\ x_n \end{pmatrix} = y_1 x_1 + \dots + y_n x_n.$$
The standard dual basis consists of the unit rows. Pulling back a row $y$ along $x \mapsto A x$ gives the row $y A$; written as a column this is $A^{\mathsf T} y^{\mathsf T}$. So the rows of $A$ are the pullbacks of the coordinate functionals of $K^m$, and transposition is the bookkeeping that converts between columns (vectors) and rows (functionals).

The code checks the theorems on a $3 \times 4$ matrix whose third row is the sum of the first two.


```python

import numpy as np

A = np.array([[1.0, 2.0, 0.0, -1.0],
              [2.0, 4.0, 1.0,  1.0],
              [3.0, 6.0, 1.0,  0.0]])      # row 3 = row 1 + row 2
m, n = A.shape
r = np.linalg.matrix_rank(A)
print("column rank =", r, "  row rank =", np.linalg.matrix_rank(A.T))

# ker f* = (im f)^perp: rows y with y A = 0
_, s, Vt = np.linalg.svd(A.T)
Y = Vt[r:].T                               # basis of ker(A^T), one column per functional
print("dim ker f* =", Y.shape[1], " = m - rank =", m - r)
print("y^T A = 0 for these y:", np.allclose(Y.T @ A, 0))

# solvability test: A x = b solvable iff y^T b = 0 for all y in ker(A^T)
for b in (A @ np.array([1.0, -1.0, 2.0, 0.5]), np.array([1.0, 0.0, 0.0])):
    print(f"b = {np.round(b, 2)}: y^T b = {np.round(Y.T @ b, 3)} -> solvable: {np.allclose(Y.T @ b, 0)}")
```
