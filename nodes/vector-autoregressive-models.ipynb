# Vector Autoregressive Models (VAR)

## 1\. The VAR(p) Model

The Vector Autoregressive (VAR) model is the [multivariate](multivariate-time-series.ipynb) extension of the univariate [Autoregressive (AR)](autoregressive-models.ipynb) model. It is one of the most widely used models for analyzing multivariate time series due to its simplicity and flexibility.

::: {#def-var-p}
## Vector Autoregressive Model: VAR(p)
Let $\vec{r}_t$ be a $k$-dimensional vector of time series variables (e.g., asset returns) at time $t$. A VAR(p) model relates $\vec{r}_t$ to its past $p$ lags:
$$
\vec{r}_t = \vec{\Phi}_0 + \Phi_1 \vec{r}_{t-1} + \Phi_2 \vec{r}_{t-2} + \dots + \Phi_p \vec{r}_{t-p} + \vec{a}_t
$$
where:
- $\vec{\Phi}_0$ is a $k \times 1$ constant vector.
- $\Phi_i$ (for $i = 1, \dots, p$) are $k \times k$ parameter matrices.
- $\vec{a}_t = (a_{1t}, a_{2t}, \dots, a_{kt})^T$ is a sequence of iid random shock vectors (multivariate white noise).
- $E[\vec{a}_t] = \vec{0}$ and $\text{Cov}(\vec{a}_t) = \Sigma$.
:::

## 2\. Model Fitting and Estimation

The parameters of a VAR(p) model include the constant vector $\vec{\Phi}_0$, the autoregressive coefficient matrices $\Phi_1, \dots, \Phi_p$, and the noise covariance matrix $\Sigma$.

The fitting process involves:
1. **Identification:** Determining the lag order $p$, often using information criteria such as [AIC or BIC](autoregressive-models.ipynb).
2. **Estimation:** Estimating the parameter matrices.

::: {.callout-note}
## Estimation: OLS vs. MLE
A key property of the standard VAR model is that every equation has exactly the same explanatory variables (the lagged vectors $\vec{r}_{t-1}, \dots, \vec{r}_{t-p}$). 
Because the regressors are identical across all $k$ equations, applying **Ordinary Least Squares (OLS)** equation-by-equation is statistically efficient and yields the exact same estimates as **Maximum Likelihood Estimation (MLE)**. This makes the VAR model computationally very easy to estimate.
:::
