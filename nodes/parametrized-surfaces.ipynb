# Parametrized Surfaces and Surface Area

**Prerequisites:** [The Cross Product](cross-product.ipynb) · [The Inverse Function Theorem and Curvilinear Coordinates](inverse-function-theorem.ipynb) · [Change of Variables in Multiple Integrals](change-of-variables-multiple-integrals.ipynb)

Curves are described by one parameter; surfaces need two. Throughout, $W$ is a region of the $uv$-plane and $\varphi$ is $C^1$ on an open set containing it.

## 1\. Smooth Parametrized Surfaces

::: {#def-parametrized-surface}
## Parametrized Surface, Normal

A **parametrized surface** is a map $\varphi : W \to \mathbb{R}^3$, $\mathbf r = \varphi(u, v) = (x(u, v), y(u, v), z(u, v))$. Its partial derivatives
$$\mathbf r_u = \Big(\frac{\partial x}{\partial u}, \frac{\partial y}{\partial u}, \frac{\partial z}{\partial u}\Big), \qquad \mathbf r_v = \Big(\frac{\partial x}{\partial v}, \frac{\partial y}{\partial v}, \frac{\partial z}{\partial v}\Big)$$
are the velocities of the **coordinate curves** $u \mapsto \varphi(u, v_0)$ and $v \mapsto \varphi(u_0, v)$, hence tangent to the surface. $\varphi$ is **smooth** if $\mathbf r_u\times\mathbf r_v \ne 0$ on $W$ (equivalently, $\mathbf r_u$, $\mathbf r_v$ are independent); then $\mathbf n = \dfrac{\mathbf r_u\times\mathbf r_v}{|\mathbf r_u\times\mathbf r_v|}$ is the **standard unit normal** and $\mathbf r_u$, $\mathbf r_v$ span the **tangent plane**.
:::

::: {#prp-surface-locally-graph}
## Smooth Surfaces Are Locally Graphs

If $\varphi$ is smooth at $(u_0, v_0)$, there is an open set $D \ni (u_0, v_0)$ on which $\varphi$ is injective and $\varphi(D)$ is the graph of a $C^1$ function over an open set of one of the coordinate planes.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-surface-locally-graph
The components of $\mathbf r_u\times\mathbf r_v$ are the $2\times2$ minors $\frac{\partial(y, z)}{\partial(u, v)}$, $\frac{\partial(z, x)}{\partial(u, v)}$, $\frac{\partial(x, y)}{\partial(u, v)}$ of $D\varphi$; one of them, say $\det\frac{\partial(x, y)}{\partial(u, v)}$, is non-zero at $(u_0, v_0)$. By the [inverse function theorem](inverse-function-theorem.ipynb#thm-inverse-function), $(u, v) \mapsto (x(u, v), y(u, v))$ maps an open $D \ni (u_0, v_0)$ bijectively onto an open $D'$ with $C^1$ inverse $\psi$. Then $\varphi$ is injective on $D$ (already its first two components are) and $\varphi(D) = \{(x, y, z(\psi(x, y))) : (x, y) \in D'\}$. $\blacksquare$
:::

::: {#exm-surfaces}
## Smooth Surfaces

1. **Graphs.** $\varphi(u, v) = (u, v, f(u, v))$ for $C^1$ $f$: $\mathbf r_u = (1, 0, f_u)$, $\mathbf r_v = (0, 1, f_v)$, $\mathbf r_u\times\mathbf r_v = (-f_u, -f_v, 1) \ne 0$, $\mathbf n = \frac{(-f_u, -f_v, 1)}{\sqrt{1 + f_u^2 + f_v^2}}$ (upward).
2. **Sphere.** $\varphi(u, v) = (R\sin u\cos v, R\sin u\sin v, R\cos u)$, $0 \le u \le \pi$, $0 \le v \le 2\pi$ (spherical coordinates $u = \varphi$, $v = \theta$): $\mathbf r_u\times\mathbf r_v = R^2\sin u\,(\sin u\cos v, \sin u\sin v, \cos u)$, an outward normal of length $R^2\sin u$. Smoothness fails at the poles $u = 0, \pi$, a defect of the parametrization (each pole is the image of a whole segment), not of the sphere.
3. **Torus.** For $a > b > 0$, $\varphi(u, v) = ((a + b\cos u)\cos v, (a + b\cos u)\sin v, b\sin u)$, $0 \le u, v \le 2\pi$, obtained by rotating the circle of radius $b$ about $(a, 0)$ in the $xz$-plane around the $z$-axis: $\mathbf r_u\times\mathbf r_v = -b(a + b\cos u)(\cos u\cos v, \cos u\sin v, \sin u)$, of length $b(a + b\cos u) > 0$, pointing into the solid torus.
4. **Hyperboloid.** $\varphi(u, v) = (a\cosh u\cos v, b\cosh u\sin v, c\sinh u)$ lies on $\frac{x^2}{a^2} + \frac{y^2}{b^2} - \frac{z^2}{c^2} = 1$, and $\mathbf r_u\times\mathbf r_v = (-bc\cosh^2u\cos v, -ac\cosh^2u\sin v, ab\sinh u\cosh u) \ne 0$.
:::

::: {#exm-degenerate-surfaces}
## When $\mathbf r_u\times\mathbf r_v = 0$

A constant map has image a point; $(f(u, v), f(u, v), f(u, v))$ has image in a line; the cone $(u\cos v, u\sin v, u)$ has $\mathbf r_u\times\mathbf r_v = (-u\cos v, -u\sin v, u)$, vanishing at the vertex $u = 0$, where the cone has no tangent plane; and $(u, v|v|, v^2)$, whose image is the creased surface $z = |y|$, is $C^1$ with $\mathbf r_v = 0$ along the crease $v = 0$.
:::

## 2\. Surface Area

A small rectangle $[u, u + \Delta u]\times[v, v + \Delta v]$ is mapped approximately onto the parallelogram spanned by $\mathbf r_u\Delta u$ and $\mathbf r_v\Delta v$, whose area is $|\mathbf r_u\times\mathbf r_v|\,\Delta u\,\Delta v$ by the [geometry of the cross product](cross-product.ipynb#thm-cross-geometry).

::: {#def-surface-area}
## Area Element and Surface Area

The **area element** of $\varphi$ is $dS = |\mathbf r_u\times\mathbf r_v|\,du\,dv$. If $\varphi$ is injective on the admissible region $D$ (possibly except on a set of measure zero), the **area** of $\varphi(D)$ is
$$\operatorname{area}\varphi(D) = \iint_DdS = \iint_D|\mathbf r_u\times\mathbf r_v|\,du\,dv .$$
:::

::: {#prp-first-fundamental-form}
## The Area Element via $E, F, G$

With $E = \mathbf r_u\cdot\mathbf r_u$, $F = \mathbf r_u\cdot\mathbf r_v$, $G = \mathbf r_v\cdot\mathbf r_v$: $\ dS = \sqrt{EG - F^2}\,du\,dv$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-first-fundamental-form
By the [identity](cross-product.ipynb#prp-cross-algebra) $(a\times b)\cdot(c\times d) = (a\cdot c)(b\cdot d) - (a\cdot d)(b\cdot c)$ with $a = c = \mathbf r_u$, $b = d = \mathbf r_v$. $\blacksquare$
:::

::: {#prp-reparametrization-surface}
## Reparametrization

Let $\tau : W' \to W$ be a $C^1$ bijection with $\det D\tau \ne 0$, $(u, v) = \tau(u', v')$, and $\tilde\varphi = \varphi\circ\tau$. Then
$$\mathbf r_{u'}\times\mathbf r_{v'} = \det\frac{\partial(u, v)}{\partial(u', v')}\;\mathbf r_u\times\mathbf r_v .$$
Hence $\tilde\varphi$ is smooth iff $\varphi$ is, the area is unchanged, and the standard normal is preserved if $\det D\tau > 0$ (**orientation-preserving** reparametrization) and reversed if $\det D\tau < 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-reparametrization-surface
By the chain rule $\mathbf r_{u'} = u_{u'}\mathbf r_u + v_{u'}\mathbf r_v$ and $\mathbf r_{v'} = u_{v'}\mathbf r_u + v_{v'}\mathbf r_v$. Since the cross product is bilinear with $\mathbf r_u\times\mathbf r_u = \mathbf r_v\times\mathbf r_v = 0$ and $\mathbf r_v\times\mathbf r_u = -\mathbf r_u\times\mathbf r_v$, we get $\mathbf r_{u'}\times\mathbf r_{v'} = (u_{u'}v_{v'} - v_{u'}u_{v'})\,\mathbf r_u\times\mathbf r_v$. For the area use the [change of variables formula](change-of-variables-multiple-integrals.ipynb#thm-change-of-variables): $\iint_{W'}|\mathbf r_u\times\mathbf r_v|\circ\tau\,|\det D\tau| = \iint_W|\mathbf r_u\times\mathbf r_v|$. $\blacksquare$
:::

On a connected smooth parametrized surface, $\det D\tau$ has constant sign (it is continuous and never zero), so a surface admits exactly two choices of unit normal compatible with smooth reparametrizations: its two **orientations**.

::: {#exm-surface-areas}
## Areas

1. For a graph, $dS = \sqrt{1 + f_u^2 + f_v^2}\,du\,dv$.
2. **Sphere:** $\int_0^{2\pi}\int_0^\pi R^2\sin u\,du\,dv = 4\pi R^2$.
3. **Torus:** $\int_0^{2\pi}\int_0^{2\pi}b(a + b\cos u)\,du\,dv = 4\pi^2ab$.
4. **The surface of the solid** $x^2 + y^2 \le 1$, $x^2 + z^2 \le 1$. The part on the cylinder $x^2 + z^2 = 1$ consists of the graphs $z = \pm\sqrt{1 - x^2}$ over the disk $x^2 + y^2 \le 1$, each of area
$$\iint_{x^2+y^2\le1}\sqrt{1 + \tfrac{x^2}{1 - x^2}}\,dA = \int_{-1}^1\int_{-\sqrt{1-x^2}}^{\sqrt{1-x^2}}\frac{dy}{\sqrt{1 - x^2}}\,dx = \int_{-1}^12\,dx = 4 .$$
By symmetry the part on the other cylinder is the same, so the total area is $16$ (an improper integral, since the integrand is unbounded near $x = \pm1$).
:::

```python

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

u, v, R, a, b = sp.symbols("u v R a b", positive=True)
def area_elem(phi):
    ru, rv = phi.diff(u), phi.diff(v)
    E, F, G = ru.dot(ru), ru.dot(rv), rv.dot(rv)
    return sp.simplify(sp.sqrt(sp.factor(sp.simplify(E * G - F**2))))
sph = sp.Matrix([R * sp.sin(u) * sp.cos(v), R * sp.sin(u) * sp.sin(v), R * sp.cos(u)])
tor = sp.Matrix([(a + b * sp.cos(u)) * sp.cos(v), (a + b * sp.cos(u)) * sp.sin(v), b * sp.sin(u)])
es = area_elem(sph); et = b * (a + b * sp.cos(u))
print("sphere dS:", es, "| area:", sp.integrate(R**2 * sp.sin(u), (u, 0, sp.pi), (v, 0, 2 * sp.pi)))
ru, rv = tor.diff(u), tor.diff(v)
print("torus |r_u x r_v|^2 - b^2(a + b cos u)^2 =", sp.simplify(ru.cross(rv).dot(ru.cross(rv)) - et**2), "| area:", sp.integrate(et, (u, 0, 2 * sp.pi), (v, 0, 2 * sp.pi)))

fig = plt.figure(figsize=(7, 5.4))
ax = fig.add_subplot(projection="3d")
A, B = 2.0, 0.7
U, V = np.meshgrid(np.linspace(0, 2 * np.pi, 50), np.linspace(0, 2 * np.pi, 70))
X, Y, Z = (A + B * np.cos(U)) * np.cos(V), (A + B * np.cos(U)) * np.sin(V), B * np.sin(U)
ax.plot_surface(X, Y, Z, color=BLUE, alpha=0.35, linewidth=0.2, edgecolor="white")
for uu in np.linspace(0, 2 * np.pi, 7)[:-1]:
    for vv in np.linspace(0, 2 * np.pi, 9)[:-1]:
        p = np.array([(A + B * np.cos(uu)) * np.cos(vv), (A + B * np.cos(uu)) * np.sin(vv), B * np.sin(uu)])
        n = np.array([np.cos(uu) * np.cos(vv), np.cos(uu) * np.sin(vv), np.sin(uu)])
        ax.quiver(*p, *(0.5 * n), color=ORANGE, lw=1, arrow_length_ratio=0.25)
ax.set_box_aspect((1, 1, 0.4)); ax.set_title(r"Torus with the outward normal $-\mathbf{n}$")
plt.tight_layout(); plt.show()
```
