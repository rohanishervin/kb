# Partial and Directional Derivatives

**Prerequisites:** [Limits and Continuity of Functions of Several Variables](limits-and-continuity-in-Rn.ipynb) · [The Derivative](derivative.ipynb#def-derivative) · [Graphs, Level Sets and Quadric Surfaces](graphs-and-level-sets.ipynb)

Let $S \subseteq \mathbb{R}^n$, $f : S \to \mathbb{R}^m$ and $a$ an **interior point** of $S$ (some ball about $a$ lies in $S$). The quotient $\frac{f(a + h) - f(a)}{h}$ makes no sense for a vector $h$, so we first differentiate along lines.

## 1\. Definitions

::: {#def-directional-derivative}
## Directional Derivative

For $v \in \mathbb{R}^n$ the **derivative of $f$ at $a$ along $v$** is
$$D_vf(a) = \lim_{h\to0}\frac{f(a + hv) - f(a)}{h} \qquad (h \in \mathbb{R}),$$
if the limit exists, i.e. the derivative at $0$ of the curve $t \mapsto f(a + tv)$. For a unit vector $u$, $D_uf(a)$ is the **directional derivative** in the direction $u$: the instantaneous rate of change of $f$ when moving from $a$ with unit speed in direction $u$.
:::

Computed componentwise, $D_vf(a) = (D_vf_1(a), \dots, D_vf_m(a))$ ([curves](parametrized-curves.ipynb#prp-curve-componentwise)). For $f : \mathbb{R}^2 \to \mathbb{R}$ and a unit vector $u$, $t \mapsto f(a + tu)$ is the section of the graph $z = f(x, y)$ by the vertical plane through $a$ parallel to $u$, and $D_uf(a)$ is the slope of that section at $a$.

::: {#prp-directional-homogeneous}
## Scaling the Direction

If $D_vf(a)$ exists and $c \in \mathbb{R}$, then $D_{cv}f(a) = c\,D_vf(a)$. In particular, for $v \ne 0$, $D_vf(a) = |v|\,D_{v/|v|}f(a)$, and $D_{-u}f(a) = -D_uf(a)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-directional-homogeneous
For $c = 0$ both sides are $0$. For $c \ne 0$ put $k = ch$: $\frac{f(a + hcv) - f(a)}{h} = c\,\frac{f(a + kv) - f(a)}{k}$, and $k \to 0$ iff $h \to 0$. $\blacksquare$
:::

::: {#def-partial-derivative}
## Partial Derivatives

The **partial derivative** of $f$ with respect to $x_j$ at $a$ is $\dfrac{\partial f}{\partial x_j}(a) = D_{e_j}f(a)$, i.e.
$$\frac{\partial f}{\partial x_j}(a) = \lim_{h\to0}\frac{f(a_1, \dots, a_j + h, \dots, a_n) - f(a_1, \dots, a_n)}{h} .$$
Other notations: $\partial_jf(a)$, $D_jf(a)$, $f_{x_j}(a)$, $\frac{\partial y}{\partial x_j}(a)$ for $y = f(x)$.
:::

So $\frac{\partial f}{\partial x_j}$ is computed by treating all variables except $x_j$ as constants and differentiating in one variable.

## 2\. Examples

::: {#exm-partials}
## Computing Partial Derivatives

1. $f(x, y, z) = x^2y - z^3$: $\ f_x = 2xy$, $f_y = x^2$, $f_z = -3z^2$.
2. $f(x, y, z) = (xyz^2, 2y - z)$: $\ \frac{\partial f}{\partial x} = (yz^2, 0)$, $\frac{\partial f}{\partial y} = (xz^2, 2)$, $\frac{\partial f}{\partial z} = (2xyz, -1)$.
:::

::: {#exm-saddle-directions}
## Directional Derivatives of a Saddle

$f(x, y) = x^2 - y^2$ at $a = (1, 0)$, $u = (\cos\theta, \sin\theta)$:
$$\frac{f(1 + h\cos\theta, h\sin\theta) - f(1, 0)}{h} = 2\cos\theta + h(\cos^2\theta - \sin^2\theta) \to 2\cos\theta .$$
For $\theta = 0$ the section is the parabola $z = (1 + h)^2$ with slope $2$; for $\theta = \frac\pi4$ it is the line $z = 1 + \sqrt2h$; for $\theta = \frac\pi2$ it is the parabola $z = 1 - h^2$ with slope $0$.
:::

::: {#exm-all-directions-not-linear}
## All Directional Derivatives Exist, but Are Not Linear in $u$

$f(x, y) = \dfrac{x^2y}{x^2 + y^2}$ for $(x, y) \ne 0$, $f(0, 0) = 0$. Since $f = 0$ on both axes, $f_x(0, 0) = f_y(0, 0) = 0$; for $(x, y) \ne 0$,
$$f_y = \frac{x^2(x^2 + y^2) - 2y\cdot x^2y}{(x^2 + y^2)^2} = \frac{x^4 - x^2y^2}{(x^2 + y^2)^2} .$$
For $u = (\cos\theta, \sin\theta)$, $\frac{f(hu) - f(0)}{h} = \frac{h^3\cos^2\theta\sin\theta}{h^3} = \cos^2\theta\sin\theta$, so every $D_uf(0)$ exists. But $u \mapsto D_uf(0) = \cos^2\theta\sin\theta$ is not of the form $c_1\cos\theta + c_2\sin\theta$: the tangent lines of the sections do not lie in one plane.
:::

::: {#exm-directional-not-continuous}
## All Directional Derivatives Zero, yet Discontinuous

Let $f(x, y) = 1$ if $y = x^2$ and $x \ne 0$, and $f = 0$ elsewhere. A line through $0$ meets the parabola at most at one point other than $0$, at positive distance from $0$; so $f(hu) = 0$ for small $h$, and $D_uf(0) = 0$ for every $u$. Yet $f(t, t^2) = 1 \not\to 0 = f(0)$: $f$ is not continuous at $0$.
:::

Directional derivatives therefore do not capture "differentiability" in several variables; the right notion, approximation by a linear map, is the subject of the [next note](differentiability-in-Rn.ipynb).

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

fig = plt.figure(figsize=(12, 4.6))
ax = fig.add_subplot(1, 2, 1, projection="3d")
x = np.linspace(-0.2, 2.2, 40); y = np.linspace(-1.2, 1.2, 40); X, Y = np.meshgrid(x, y)
ax.plot_surface(X, Y, X**2 - Y**2, color=BLUE, alpha=0.3, linewidth=0)
h = np.linspace(-1, 1, 50)
for th, col, lab in [(0, ORANGE, "0"), (np.pi / 4, AQUA, r"\pi/4"), (np.pi / 2, VIOLET, r"\pi/2")]:
    px, py = 1 + h * np.cos(th), h * np.sin(th)
    ax.plot(px, py, px**2 - py**2, color=col, lw=2.5, label=f"$\\theta = {lab}$, slope ${round(2 * np.cos(th), 3) + 0:g}$")
ax.legend(fontsize=8, loc="upper left"); ax.set_title(r"Sections of $z = x^2 - y^2$ through $(1, 0)$")

ax2 = fig.add_subplot(1, 2, 2)
th = np.linspace(0, 2 * np.pi, 400)
ax2.plot(th, np.cos(th)**2 * np.sin(th), color=ORANGE, lw=2, label=r"$D_uf(0) = \cos^2\theta\sin\theta$")
ax2.plot(th, 0 * th, color=MUTED, lw=1, ls="--", label=r"$f_x(0)\cos\theta + f_y(0)\sin\theta = 0$")
ax2.set_xlabel(r"$\theta$"); ax2.legend(fontsize=8, loc="lower left")
ax2.set_title(r"$f = x^2y/(x^2 + y^2)$: directional derivatives at $0$")
plt.tight_layout(); plt.show()
```
