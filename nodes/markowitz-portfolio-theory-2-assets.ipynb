# Markowitz Portfolio Theory: The Two-Asset Case

Modern Portfolio Theory (MPT), pioneered by Harry Markowitz (1952), provides the foundational mathematical framework for asset allocation under uncertainty. By combining individual risky assets into a portfolio, investors can achieve diversification—reducing total portfolio risk below the weighted average risk of individual components.

## 1\. Portfolio Weights and Budget Constraint

Consider a single-period economy $T = \{0, 1\}$ within the [Single-Period Investment Model](single-period-investment-model.ipynb).

Suppose an investor allocates initial capital across two risky assets (Stock 1 and Stock 2) with known initial prices $S_1(0) > 0$ and $S_2(0) > 0$ at time $T=0$.

::: {#def-portfolio-weights}
## Definition: Portfolio Value and Weights
Let $x_1, x_2 \in \mathbb{R}$ denote the number of shares purchased in Stock 1 and Stock 2 respectively.

The total initial value of the portfolio $V_p(0)$ at time $T=0$ is:

$$
V_p(0) = x_1 S_1(0) + x_2 S_2(0)
$$

The **portfolio weights** $w_1, w_2 \in \mathbb{R}$ represent the proportion of total initial capital invested in each asset:

$$
w_1 \triangleq \frac{x_1 S_1(0)}{V_p(0)}, \qquad w_2 \triangleq \frac{x_2 S_2(0)}{V_p(0)}
$$
:::

::: {#prp-weight-budget-constraint}
## Proposition: Capital Budget Constraint
The portfolio weights satisfy the normalization identity:

$$
w_1 + w_2 = 1
$$
:::

::: {.callout-tip collapse="true"}
## Proof of Capital Budget Constraint
Summing the definitions of $w_1$ and $w_2$:

$$
\begin{aligned}
w_1 + w_2 &= \frac{x_1 S_1(0)}{V_p(0)} + \frac{x_2 S_2(0)}{V_p(0)} \\
&= \frac{x_1 S_1(0) + x_2 S_2(0)}{V_p(0)} \\
&= \frac{V_p(0)}{V_p(0)} = 1 \quad \blacksquare
\end{aligned}
$$
:::

### 1.1\. Convex vs. Affine Combinations (Short-Selling)

- **No Short-Selling ($w_1, w_2 \ge 0$):** If short positions are prohibited, weights must be non-negative ($0 \le w_1, w_2 \le 1$). The portfolio allocation forms a **convex combination** of the two assets.
- **Short-Selling Allowed ($w_1 < 0$ or $w_2 < 0$):** Short-selling involves borrowing shares of an asset at $T=0$, selling them immediately for cash, and agreeing to buy back the shares at $T=1$. In this scenario, $x_i < 0 \implies w_i < 0$ (and the complementary weight $w_j > 1$). The allocation forms an **affine combination** of the assets.

::: {#prp-portfolio-return-linear}
## Proposition: Linear Portfolio Net Return
Let $R_1 = \frac{S_1(1) - S_1(0)}{S_1(0)}$ and $R_2 = \frac{S_2(1) - S_2(0)}{S_2(0)}$ be the net returns of Stock 1 and Stock 2. The net return of the portfolio $R_p = \frac{V_p(1) - V_p(0)}{V_p(0)}$ is a weighted linear combination of individual asset returns:

$$
R_p = w_1 R_1 + w_2 R_2
$$
:::

::: {.callout-tip collapse="true"}
## Proof of Linear Portfolio Return
At time $T=1$, asset prices realization yields terminal portfolio value $V_p(1) = x_1 S_1(1) + x_2 S_2(1)$.

Substitute share quantities $x_1 = \frac{w_1 V_p(0)}{S_1(0)}$ and $x_2 = \frac{w_2 V_p(0)}{S_2(0)}$ into $V_p(1)$:

$$
\begin{aligned}
V_p(1) &= \left(\frac{w_1 V_p(0)}{S_1(0)}\right) S_1(0)(1 + R_1) + \left(\frac{w_2 V_p(0)}{S_2(0)}\right) S_2(0)(1 + R_2) \\
&= w_1 V_p(0)(1 + R_1) + w_2 V_p(0)(1 + R_2) \\
&= V_p(0) \Big[ w_1(1 + R_1) + w_2(1 + R_2) \Big] \\
&= V_p(0) \Big[ (w_1 + w_2) + w_1 R_1 + w_2 R_2 \Big]
\end{aligned}
$$

Since $w_1 + w_2 = 1$:

$$
V_p(1) = V_p(0) (1 + w_1 R_1 + w_2 R_2)
$$

Dividing both sides by $V_p(0)$ gives gross return $1 + R_p = 1 + w_1 R_1 + w_2 R_2$, so:

$$
R_p = w_1 R_1 + w_2 R_2 \quad \blacksquare
$$
:::

::: {.callout-note}
## Intuition: Capital Weighting
Portfolio return is simply the dollar-weighted average of individual asset returns. The financial leverage or tilt towards one asset is directly controlled by weight $w_1$.
:::



## 2\. Expected Portfolio Return and Variance

Let asset returns $R_1$ and $R_2$ be [random variables](random-variables.ipynb) with expected returns $\mu_1 = \mathbb{E}[R_1], \mu_2 = \mathbb{E}[R_2]$, variances $\sigma_1^2 = \text{Var}(R_1), \sigma_2^2 = \text{Var}(R_2)$, and covariance $\sigma_{12} = \text{Cov}(R_1, R_2) = \rho_{12} \sigma_1 \sigma_2$, where $\rho_{12} \in [-1, 1]$ is the correlation coefficient.

::: {#thm-portfolio-moments}
## Theorem: Portfolio Expected Return and Variance
For a two-asset portfolio with weight vector $\mathbf{w} = [w_1, w_2]^T$ where $w_2 = 1 - w_1$:

1. **Expected Portfolio Return:**
$$
\mu_w = \mathbb{E}[R_p] = w_1 \mu_1 + w_2 \mu_2 = w_1 \mu_1 + (1 - w_1) \mu_2
$$

2. **Portfolio Variance:**
$$
\sigma_w^2 = \text{Var}(R_p) = w_1^2 \sigma_1^2 + w_2^2 \sigma_2^2 + 2 w_1 w_2 \sigma_{12} = w_1^2 \sigma_1^2 + (1-w_1)^2 \sigma_2^2 + 2 w_1 (1-w_1) \rho_{12} \sigma_1 \sigma_2
$$
:::

::: {.callout-tip collapse="true"}
## Proof of Portfolio Moments
Applying the [Linearity of Expectation](expectation.ipynb):

$$
\mathbb{E}[R_p] = \mathbb{E}[w_1 R_1 + w_2 R_2] = w_1 \mathbb{E}[R_1] + w_2 \mathbb{E}[R_2] = w_1 \mu_1 + w_2 \mu_2
$$

Applying variance operator properties for linear combinations of random variables:

$$
\begin{aligned}
\text{Var}(R_p) &= \text{Var}(w_1 R_1 + w_2 R_2) \\
&= \text{Cov}(w_1 R_1 + w_2 R_2, w_1 R_1 + w_2 R_2) \\
&= w_1^2 \text{Var}(R_1) + w_2^2 \text{Var}(R_2) + 2 w_1 w_2 \text{Cov}(R_1, R_2) \\
&= w_1^2 \sigma_1^2 + w_2^2 \sigma_2^2 + 2 w_1 w_2 \sigma_{12} \\
&= w_1^2 \sigma_1^2 + w_2^2 \sigma_2^2 + 2 w_1 w_2 \rho_{12} \sigma_1 \sigma_2 \quad \blacksquare
\end{aligned}
$$
:::

### 2.1\. Compact Matrix Formulation

Define weight vector $\mathbf{w}$, mean vector $\boldsymbol{\mu}$, and symmetric positive semi-definite [Covariance Matrix](covariance-and-correlation.ipynb) $\mathbf{C}$:

$$
\mathbf{w} = \begin{bmatrix} w_1 \\ w_2 \end{bmatrix}, \qquad \boldsymbol{\mu} = \begin{bmatrix} \mu_1 \\ \mu_2 \end{bmatrix}, \qquad \mathbf{C} = \begin{bmatrix} \sigma_1^2 & \sigma_{12} \\ \sigma_{21} & \sigma_2^2 \end{bmatrix}
$$

The expected return and variance expressed in matrix form are:

$$
\mu_w = \mathbf{w}^T \boldsymbol{\mu}, \qquad \sigma_w^2 = \mathbf{w}^T \mathbf{C} \mathbf{w}
$$

::: {.callout-note}
## Intuition: The Power of Diversification
While expected portfolio return $\mu_w$ is strictly linear in weights, portfolio variance $\sigma_w^2$ includes the cross-coupling term $2 w_1 w_2 \rho_{12} \sigma_1 \sigma_2$. If $\rho_{12} < 1$, portfolio volatility $\sigma_w$ is strictly less than the weighted average of individual volatilities $w_1 \sigma_1 + w_2 \sigma_2$. Diversification eliminates un-correlated risk!
:::


```python
import numpy as np
import matplotlib.pyplot as plt

# Asset parameters
mu1, sigma1 = 0.14, 0.25   # Asset 1: 14% return, 25% volatility
mu2, sigma2 = 0.08, 0.15   # Asset 2: 8% return, 15% volatility

# Range of portfolio weights w1 (including short selling)
w1 = np.linspace(-0.4, 1.4, 500)
w2 = 1.0 - w1

mu_p = w1 * mu1 + w2 * mu2

correlations = [1.0, 0.5, 0.0, -0.5, -1.0]
colors = ['#d62728', '#ff7f0e', '#2ca02c', '#1f77b4', '#9467bd']

plt.figure(figsize=(9.5, 6), dpi=120)

for rho, color in zip(correlations, colors):
    sigma_p = np.sqrt(w1**2 * sigma1**2 + w2**2 * sigma2**2 + 2 * w1 * w2 * rho * sigma1 * sigma2)
    plt.plot(sigma_p, mu_p, label=rf"$\rho_{{12}} = {rho:.1f}$", color=color, linewidth=2)

# Mark individual asset positions
plt.scatter([sigma1, sigma2], [mu1, mu2], color='black', s=80, zorder=5)
plt.annotate(r"Asset 1 $(\sigma_1, \mu_1)$", xy=(sigma1, mu1), xytext=(sigma1+0.01, mu1-0.005), fontweight='bold')
plt.annotate(r"Asset 2 $(\sigma_2, \mu_2)$", xy=(sigma2, mu2), xytext=(sigma2+0.01, mu2-0.005), fontweight='bold')

plt.title(r"Two-Asset Attainable Portfolios under Varying Correlations $\rho_{12}$", fontsize=13, pad=12)
plt.xlabel(r"Portfolio Risk / Volatility ($\sigma_w$)", fontsize=11)
plt.ylabel(r"Expected Portfolio Return ($\mu_w$)", fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=10, loc="upper left")
plt.tight_layout()
plt.show()
```


## 3\. The Attainable Set and Hyperbola Geometry

::: {#def-attainable-set}
## Definition: Attainable Set
The **attainable set** of portfolios is the set of all $(\sigma, \mu)$ pairs in risk-return space for which there exists a valid portfolio weight vector $\mathbf{w} = [w_1, w_2]^T$ satisfying $w_1 + w_2 = 1$ such that:

$$
\mu = \mu_w = w_1 \mu_1 + w_2 \mu_2, \qquad \sigma = \sigma_w = \sqrt{w_1^2 \sigma_1^2 + w_2^2 \sigma_2^2 + 2 w_1 w_2 \sigma_{12}}
$$
:::

::: {#thm-hyperbola-geometry}
## Theorem: Hyperbolic Frontier Geometry
If asset expected returns are distinct ($\mu_1 \neq \mu_2$) and asset returns are imperfectly correlated ($\rho_{12} \in (-1, 1)$), the attainable set of portfolios in $(\sigma, \mu)$ space traces out the right branch of a **hyperbola** centered on the vertical axis.
:::

::: {.callout-tip collapse="true"}
## Proof of Hyperbolic Frontier Geometry
Assume $\mu_1 \neq \mu_2$. Express expected portfolio return $\mu_w$ in terms of $w_1$ (using $w_2 = 1 - w_1$):

$$
\mu_w = w_1 \mu_1 + (1 - w_1) \mu_2 = \mu_2 + w_1 (\mu_1 - \mu_2)
$$

Since $\mu_1 - \mu_2 \neq 0$, we can invert this affine relationship to express weight $w_1$ as an explicit linear function of target expected return $\mu$:

$$
w_1 = \frac{\mu - \mu_2}{\mu_1 - \mu_2}, \qquad 1 - w_1 = \frac{\mu_1 - \mu}{\mu_1 - \mu_2}
$$

Substitute $w_1$ and $1-w_1$ directly into the portfolio variance formula $\sigma_w^2$:

$$
\sigma^2 = \left(\frac{\mu - \mu_2}{\mu_1 - \mu_2}\right)^2 \sigma_1^2 + \left(\frac{\mu_1 - \mu}{\mu_1 - \mu_2}\right)^2 \sigma_2^2 + 2 \left(\frac{\mu - \mu_2}{\mu_1 - \mu_2}\right) \left(\frac{\mu_1 - \mu}{\mu_1 - \mu_2}\right) \sigma_{12}
$$

Expanding terms in powers of $\mu$ yields a quadratic polynomial in $\mu$:

$$
\sigma^2 = A \mu^2 + B \mu + C
$$

where coefficient $A > 0$ whenever $\rho_{12} \in (-1, 1)$. Completing the square with respect to $\mu$:

$$
\sigma^2 = A \left( \mu + \frac{B}{2A} \right)^2 + \left( C - \frac{B^2}{4A} \right)
$$

Denoting $\mu_{\text{mvp}} = -\frac{B}{2A}$ and $\sigma_{\text{mvp}}^2 = C - \frac{B^2}{4A}$, we rearrange to standard hyperbolic form:

$$
\frac{\sigma^2}{\sigma_{\text{mvp}}^2} - \frac{(\mu - \mu_{\text{mvp}})^2}{d^2} = 1
$$

where $d^2 = \frac{\sigma_{\text{mvp}}^2}{A} > 0$. In $(\sigma, \mu)$ space with $\sigma \ge 0$, this defines the right branch of a horizontal hyperbola centered at $(0, \mu_{\text{mvp}})$. $\blacksquare$
:::



## 4\. The Global Minimum Variance Portfolio (MVP)

The **Global Minimum Variance Portfolio (MVP)** is the unique portfolio composition that minimizes total risk $\sigma_w^2$ across all attainable weight combinations.

::: {#thm-mvp-solution}
## Theorem: Global Minimum Variance Portfolio Formula
For a two-asset universe with $\sigma_1^2 + \sigma_2^2 - 2\sigma_{12} > 0$, the weight $w^* = w_1^*$ of Stock 1 in the Global Minimum Variance Portfolio is:

$$
w^* = \frac{\sigma_2^2 - \sigma_{12}}{\sigma_1^2 + \sigma_2^2 - 2\sigma_{12}} = \frac{\sigma_2^2 - \rho_{12}\sigma_1 \sigma_2}{\sigma_1^2 + \sigma_2^2 - 2\rho_{12}\sigma_1 \sigma_2}
$$

The corresponding weight of Stock 2 is:

$$
w_2^* = 1 - w^* = \frac{\sigma_1^2 - \sigma_{12}}{\sigma_1^2 + \sigma_2^2 - 2\sigma_{12}}
$$
:::

::: {.callout-tip collapse="true"}
## Derivation of the Global MVP
To minimize variance $\sigma_w^2 = w^2 \sigma_1^2 + (1-w)^2 \sigma_2^2 + 2 w (1-w) \sigma_{12}$, differentiate with respect to $w$:

$$
\frac{\partial \sigma_w^2}{\partial w} = 2 w \sigma_1^2 - 2(1-w) \sigma_2^2 + 2(1 - 2w) \sigma_{12} = 0
$$

Divide by 2 and group terms containing $w$:

$$
w \sigma_1^2 - \sigma_2^2 + w \sigma_2^2 + \sigma_{12} - 2w \sigma_{12} = 0
$$

$$
w \left( \sigma_1^2 + \sigma_2^2 - 2\sigma_{12} \right) = \sigma_2^2 - \sigma_{12}
$$

Solving for $w$ yields:

$$
w^* = \frac{\sigma_2^2 - \sigma_{12}}{\sigma_1^2 + \sigma_2^2 - 2\sigma_{12}}
$$

To verify that this critical point is a strict minimum, compute the second derivative:

$$
\frac{\partial^2 \sigma_w^2}{\partial w^2} = 2 (\sigma_1^2 + \sigma_2^2 - 2\sigma_{12}) = 2 \text{Var}(R_1 - R_2)
$$

Since asset returns $R_1 \neq R_2$ are not identical, $\text{Var}(R_1 - R_2) > 0$. The second derivative is strictly positive, confirming a global minimum. $\blacksquare$
:::

### 4.1\. Minimum Variance Portfolio under No Short-Selling

When short selling is prohibited ($0 \le w \le 1$), the optimization problem becomes:

$$
\min_{w \in [0, 1]} \sigma_w^2
$$

The constrained optimal weight $w_{\text{constrained}}^*$ is bounded by corner solutions:

$$
w_{\text{constrained}}^* = \max\left(0, \min\left(1, w^*\right)\right)
$$


```python
import numpy as np
import matplotlib.pyplot as plt

# Asset parameters
mu1, sigma1 = 0.14, 0.25
mu2, sigma2 = 0.08, 0.15
rho12 = 0.20
cov12 = rho12 * sigma1 * sigma2

# Calculate analytical MVP weights
w1_mvp = (sigma2**2 - cov12) / (sigma1**2 + sigma2**2 - 2 * cov12)
w2_mvp = 1.0 - w1_mvp

mu_mvp = w1_mvp * mu1 + w2_mvp * mu2
sigma_mvp = np.sqrt(w1_mvp**2 * sigma1**2 + w2_mvp**2 * sigma2**2 + 2 * w1_mvp * w2_mvp * cov12)

# Generate attainable hyperbola curve
w1_vals = np.linspace(-0.5, 1.4, 600)
w2_vals = 1.0 - w1_vals
mu_vals = w1_vals * mu1 + w2_vals * mu2
sigma_vals = np.sqrt(w1_vals**2 * sigma1**2 + w2_vals**2 * sigma2**2 + 2 * w1_vals * w2_vals * cov12)

plt.figure(figsize=(9.5, 6), dpi=120)

# Plot hyperbola
plt.plot(sigma_vals, mu_vals, color='#1f77b4', linewidth=2.5, label="Attainable Hyperbola Curve")

# Highlight MVP point (moved % formatting outside of math mode)
plt.scatter([sigma_mvp], [mu_mvp], color='#d62728', s=120, edgecolors='black', zorder=5, 
            label=rf"Global MVP: $\sigma =$ {sigma_mvp:.1%}, $\mu =$ {mu_mvp:.1%}")

# Highlight Assets
plt.scatter([sigma1, sigma2], [mu1, mu2], color='black', s=80, zorder=5)
plt.annotate("Asset 1", xy=(sigma1, mu1), xytext=(sigma1+0.008, mu1-0.005))
plt.annotate("Asset 2", xy=(sigma2, mu2), xytext=(sigma2+0.008, mu2-0.005))

plt.title("Global Minimum Variance Portfolio (MVP) on Attainable Frontier", fontsize=13, pad=12)
plt.xlabel(r"Portfolio Volatility ($\sigma_w$)", fontsize=11)
plt.ylabel(r"Expected Portfolio Return ($\mu_w$)", fontsize=11)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=10, loc="upper left")
plt.tight_layout()
plt.show()
```


## 5\. The Efficient Frontier

::: {#def-efficient-frontier}
## Definition: Two-Asset Efficient Frontier
The **Efficient Frontier** is the upper branch of the attainable hyperbola curve, consisting of all portfolios with expected return $\mu_w \ge \mu_{\text{mvp}}$.
:::

::: {.callout-note}
## Intuition: Dominance Screening
For any portfolio on the lower branch of the hyperbola (where $\mu_w < \mu_{\text{mvp}}$), there exists a corresponding portfolio on the upper branch with the exact same risk $\sigma_w$ but strictly higher expected return $\mu_w$. Therefore, all portfolios below the MVP are dominated and ruled out by rational investors.
:::



## 6\. Risk-Free Asset and the Capital Allocation Line

Now consider introducing a risk-free asset paying deterministic return $R_{rf}$.

Let $\lambda \ge 0$ be the weight invested in a risky portfolio $(\sigma_w, \mu_w)$, and $(1-\lambda)$ be the weight invested in the risk-free asset $R_{rf}$.

::: {#thm-capital-allocation-line}
## Theorem: Capital Allocation Line (CAL)
The expected return $\mu_{\tilde{w}}$ and risk $\sigma_{\tilde{w}}$ of a portfolio combining risk-free capital with risky portfolio $(\sigma_w, \mu_w)$ form a straight line in $(\sigma, \mu)$ space:

$$
\mu_{\tilde{w}} = R_{rf} + \left( \frac{\mu_w - R_{rf}}{\sigma_w} \right) \sigma_{\tilde{w}}
$$

where $S(w) = \frac{\mu_w - R_{rf}}{\sigma_w}$ is the **Sharpe Ratio** (slope of the CAL).
:::

::: {.callout-tip collapse="true"}
## Proof of Capital Allocation Line
Combined portfolio expected return is:

$$
\mu_{\tilde{w}} = \lambda \mu_w + (1-\lambda) R_{rf} = R_{rf} + \lambda (\mu_w - R_{rf})
$$

Combined portfolio variance is (since $R_{rf}$ has zero variance and zero covariance with risky assets):

$$
\sigma_{\tilde{w}}^2 = \text{Var}(\lambda R_w + (1-\lambda) R_{rf}) = \lambda^2 \text{Var}(R_w) = \lambda^2 \sigma_w^2
$$

For non-leveraged or long-risky positions ($\lambda \ge 0$), taking the square root yields $\sigma_{\tilde{w}} = \lambda \sigma_w \implies \lambda = \frac{\sigma_{\tilde{w}}}{\sigma_w}$.

Substituting $\lambda$ into the expected return equation:

$$
\mu_{\tilde{w}} = R_{rf} + \left(\frac{\sigma_{\tilde{w}}}{\sigma_w}\right) (\mu_w - R_{rf}) = R_{rf} + \left( \frac{\mu_w - R_{rf}}{\sigma_w} \right) \sigma_{\tilde{w}} \quad \blacksquare
$$
:::

### 6.1\. The Tangency Portfolio (Market Portfolio)

Assuming $R_{rf} < \mu_{\text{mvp}}$, the optimal risky portfolio selection maximizes the slope of the CAL. Geometrically, this ray originates from $(0, R_{rf})$ and is strictly **tangent** to the risky efficient hyperbola.

::: {#def-market-portfolio}
## Definition: Tangency (Market) Portfolio
The **Tangency Portfolio** (or **Market Portfolio**) $m$ is the unique risky portfolio that maximizes the Sharpe Ratio:

$$
w_m^* = \arg\max_{w_1} \frac{\mu_w - R_{rf}}{\sigma_w}
$$
:::

::: {#thm-market-portfolio-weights}
## Theorem: Analytical Weights of the Tangency Portfolio
For a two-asset risky universe, the optimal tangency weight $w_1^m$ of Asset 1 is:

$$
w_1^m = \frac{(\mu_1 - R_{rf})\sigma_2^2 - (\mu_2 - R_{rf})\sigma_{12}}{(\mu_1 - R_{rf})\sigma_2^2 + (\mu_2 - R_{rf})\sigma_1^2 - (\mu_1 + \mu_2 - 2R_{rf})\sigma_{12}}
$$
:::

::: {.callout-note}
## Intuition: Mutual Fund Separation & Market Equilibrium
Why is the tangency portfolio called the **Market Portfolio**? In economic equilibrium under identical investor expectations, ALL investors hold risky assets in the exact same proportional weights ($w_1^m, w_2^m$) corresponding to the tangency point (adjusting only their risk-free allocation $\lambda$). Aggregating across the economy, the relative holdings $w_1^m : w_2^m$ must equal the total relative market capitalization of Stock 1 and Stock 2!
:::



## 7\. Capital Market Line (CML) and Optimal Investor Choice

The CAL passing through the Tangency Market Portfolio $m = (\sigma_m, \mu_m)$ is called the **Capital Market Line (CML)**.

::: {#def-cml}
## Definition: Capital Market Line (CML)
The **Capital Market Line (CML)** equation is:

$$
\mu = R_{rf} + \left( \frac{\mu_m - R_{rf}}{\sigma_m} \right) \sigma
$$

where:
- $\mu_m - R_{rf}$ is the **Market Excess Return** (Market Risk Premium).
- $\frac{\mu_m - R_{rf}}{\sigma_m}$ is the **Sharpe Ratio of the Market Portfolio** (Price of Risk).
:::

### 7.1\. Investor Preferences and Indifference Curves

To select a specific portfolio along the CML, an investor's preferences are modeled using a mean-variance utility function:

$$
U(\mu, \sigma) = \mu - \frac{\gamma}{2} \sigma^2
$$

where $\gamma > 0$ denotes the investor's coefficient of risk aversion.

**Indifference Curves** represent level sets in $(\sigma, \mu)$ space where utility $U(\mu, \sigma)$ remains constant:

$$
\mu = U_0 + \frac{\gamma}{2} \sigma^2
$$

The **Optimal Portfolio** for a given investor occurs at the point of tangency between the investor's convex indifference curve and the Capital Market Line.


```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
mu1, sigma1 = 0.14, 0.25
mu2, sigma2 = 0.08, 0.15
rho12 = 0.20
cov12 = rho12 * sigma1 * sigma2
R_rf = 0.035

# Compute Tangency Market Portfolio
num = (mu1 - R_rf) * sigma2**2 - (mu2 - R_rf) * cov12
den = (mu1 - R_rf) * sigma2**2 + (mu2 - R_rf) * sigma1**2 - (mu1 + mu2 - 2 * R_rf) * cov12
w1_m = num / den
w2_m = 1.0 - w1_m

mu_m = w1_m * mu1 + w2_m * mu2
sigma_m = np.sqrt(w1_m**2 * sigma1**2 + w2_m**2 * sigma2**2 + 2 * w1_m * w2_m * cov12)
sharpe_m = (mu_m - R_rf) / sigma_m

# Hyperbola curve
w1_vals = np.linspace(-0.5, 1.4, 600)
w2_vals = 1.0 - w1_vals
mu_vals = w1_vals * mu1 + w2_vals * mu2
sigma_vals = np.sqrt(w1_vals**2 * sigma1**2 + w2_vals**2 * sigma2**2 + 2 * w1_vals * w2_vals * cov12)

# CML Line
sigma_cml = np.linspace(0, 0.35, 400)
mu_cml = R_rf + sharpe_m * sigma_cml

# Indifference curves for risk aversion gamma = 4
gamma = 4.0
sigma_opt = sharpe_m / gamma
mu_opt = R_rf + sharpe_m * sigma_opt
u_opt = mu_opt - 0.5 * gamma * sigma_opt**2

sig_ind = np.linspace(0.02, 0.30, 200)
mu_ind1 = (u_opt - 0.02) + 0.5 * gamma * sig_ind**2
mu_ind2 = u_opt + 0.5 * gamma * sig_ind**2
mu_ind3 = (u_opt + 0.02) + 0.5 * gamma * sig_ind**2

plt.figure(figsize=(10, 6), dpi=120)

# Attainable hyperbola
plt.plot(sigma_vals, mu_vals, color='#7f7f7f', linestyle='--', linewidth=1.8, label="Risky Attainable Frontier")

# CML
plt.plot(sigma_cml, mu_cml, color='#1f77b4', linewidth=2.5, label=f"Capital Market Line (CML, Slope = {sharpe_m:.2f})")

# Indifference curves
plt.plot(sig_ind, mu_ind2, color='#2ca02c', linewidth=2, label="Investor Tangent Indifference Curve")
plt.plot(sig_ind, mu_ind1, color='#2ca02c', linestyle=':', alpha=0.6)
plt.plot(sig_ind, mu_ind3, color='#2ca02c', linestyle=':', alpha=0.6)

# Key points (moved % formatting outside math mode and used raw f-strings)
plt.scatter([0], [R_rf], color='blue', s=80, zorder=5, label=rf"Risk-Free Rate ($R_{{rf}} =$ {R_rf:.1%})")
plt.scatter([sigma_m], [mu_m], color='#d62728', s=120, edgecolors='black', zorder=6, label=rf"Tangency Market Portfolio ($w_1 = {w1_m:.2f}$)")
plt.scatter([sigma_opt], [mu_opt], color='purple', s=100, edgecolors='black', zorder=6, label="Optimal Investor Allocation")

plt.title("Markowitz Two-Asset Portfolio Theory: CML and Investor Optimal Choice", fontsize=13, pad=12)
plt.xlabel(r"Portfolio Risk / Volatility ($\sigma$)", fontsize=11)
plt.ylabel(r"Expected Return ($\mu$)", fontsize=11)
plt.xlim(-0.01, 0.35)
plt.ylim(0.01, 0.22)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=9.5, loc="upper left")
plt.tight_layout()
plt.show()
```
