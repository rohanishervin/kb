# Fourier Series

**Prerequisites:** [Integration by Parts and Substitution](integration-by-parts-and-substitution.ipynb) · [Properties of the Riemann Integral](properties-of-the-riemann-integral.ipynb) · [Infinite Series and Convergence Tests](series.ipynb) · [Power Series](power-series.ipynb)

Taylor polynomials approximate a function **near a point**. Trigonometric polynomials approximate a **periodic** function **on a whole period**. This note collects the basic facts; it is the starting point for a fuller treatment of Fourier analysis.

## 1\. Trigonometric Series and Orthogonality

Let $T > 0$ and $\omega = \frac{2\pi}{T}$. The functions $\cos m\omega x$, $\sin m\omega x$ ($m \in \mathbb{N}$) have period $T$. A **trigonometric series** of period $T$ is
$$\frac{a_0}{2} + \sum_{m=1}^\infty\big(a_m\cos m\omega x + b_m\sin m\omega x\big);$$
a finite sum (up to $m = N$) is a **trigonometric polynomial** of order $N$.

::: {#prp-orthogonality}
## Orthogonality Relations

For $m, n \ge 1$, integrating over any interval of length $T$ (e.g. $[-\frac T2, \frac T2]$):
$$\int\cos m\omega x\,\sin n\omega x\,dx = 0, \qquad \int\cos m\omega x\cos n\omega x\,dx = \int\sin m\omega x\sin n\omega x\,dx = \begin{cases}0 & m \ne n\\ \frac T2 & m = n,\end{cases}$$
and $\int\cos m\omega x\,dx = \int\sin m\omega x\,dx = 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-orthogonality
The last two: $\frac{1}{m\omega}\sin m\omega x$ and $-\frac{1}{m\omega}\cos m\omega x$ are $T$-periodic antiderivatives, so their increments over a period vanish. For the products use
$$2\cos A\cos B = \cos(A - B) + \cos(A + B), \quad 2\sin A\sin B = \cos(A - B) - \cos(A + B), \quad 2\sin A\cos B = \sin(A + B) + \sin(A - B):$$
every term is $\cos k\omega x$ or $\sin k\omega x$ with $k \in \mathbb{Z}$, which integrates to $0$ unless $k = 0$; $k = 0$ occurs only in $\cos(A - B)$ with $m = n$, giving $\int 1 = T$, halved. $\blacksquare$
:::

## 2\. Fourier Coefficients

::: {#def-fourier-coefficients}
## Fourier Series of a Function

For a $T$-periodic $f$, integrable on a period, the **Fourier coefficients** are
$$a_n = \frac2T\int_{-T/2}^{T/2}f(x)\cos n\omega x\,dx \quad (n \ge 0), \qquad b_n = \frac2T\int_{-T/2}^{T/2}f(x)\sin n\omega x\,dx \quad (n \ge 1),$$
and the trigonometric series with these coefficients is the **Fourier series** of $f$. Its partial sums $S_Nf$ are the **Fourier approximations** of order $N$.
:::

**Where the formulas come from.** If $f$ *is* a trigonometric polynomial $\frac{a_0}{2} + \sum_{m\le N}(a_m\cos m\omega x + b_m\sin m\omega x)$, multiply by $\cos n\omega x$ and integrate over a period: by orthogonality only the term $m = n$ survives, giving $\int f\cos n\omega x = a_n\frac T2$; similarly for $b_n$; and $\int f = \frac{a_0}{2}T$ (which is why the constant term is written $\frac{a_0}{2}$). So $\frac{a_0}{2}$ is the **mean value** of $f$ over a period, whereas the constant term of a Taylor series is the value at a single point. For infinite series the same computation needs termwise integration, which is valid e.g. under [uniform convergence](function-spaces.ipynb) but not in general.

A function $f : [a, b] \to \mathbb{R}$ is **piecewise $C^1$** if there is a partition $a = a_0 < a_1 < \dots < a_k = b$ such that on each $(a_{i-1}, a_i)$, $f$ has a continuous derivative, and $f$ and $f'$ have one-sided limits at the $a_i$.

::: {#thm-dirichlet}
## Dirichlet's Theorem (stated without proof)

Let $f$ be $T$-periodic and piecewise $C^1$ on a period. Then its Fourier series converges at every $x$, to $f(x)$ where $f$ is continuous and to $\frac{f(x^+) + f(x^-)}{2}$ where $f$ jumps.
:::

::: {#prp-even-odd-coefficients}
## Even and Odd Functions

If $f$ is even, $b_n = 0$ for all $n$ (a **cosine series**) and $a_n = \frac4T\int_0^{T/2}f\cos n\omega x$; if $f$ is odd, $a_n = 0$ (a **sine series**) and $b_n = \frac4T\int_0^{T/2}f\sin n\omega x$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-even-odd-coefficients
The integral over $[-\frac T2, \frac T2]$ of an odd function is $0$ and of an even function is twice the integral over $[0, \frac T2]$ (substitute $x \mapsto -x$). For even $f$, $f\sin n\omega x$ is odd and $f\cos n\omega x$ even; for odd $f$ the reverse. $\blacksquare$
:::

**Functions on an interval.** A piecewise $C^1$ function $\varphi$ on $[0, A]$ can be extended to $[-A, A]$ evenly ($\varphi(-x) = \varphi(x)$) or oddly ($\varphi(-x) = -\varphi(x)$), and then $2A$-periodically. This gives the **cosine series** and the **sine series** of $\varphi$ on $[0, A]$:
$$a_n = \frac2A\int_0^A\varphi(x)\cos\frac{n\pi x}{A}dx, \qquad b_n = \frac2A\int_0^A\varphi(x)\sin\frac{n\pi x}{A}dx,$$
each representing $\varphi$ on $(0, A)$ at continuity points (the sine series is $0$ at $0$ and $A$).

## 3\. Examples

::: {#exm-square-wave}
## Square Wave

$f = 1$ on $(2k\pi, (2k+1)\pi)$ and $f = 0$ on $((2k-1)\pi, 2k\pi)$, $k \in \mathbb{Z}$ (values at multiples of $\pi$ are irrelevant). $T = 2\pi$, $\omega = 1$:
$$a_0 = \frac1\pi\int_0^\pi dx = 1, \quad a_n = \frac1\pi\int_0^\pi\cos nx\,dx = 0, \quad b_n = \frac1\pi\int_0^\pi\sin nx\,dx = \frac{1 - (-1)^n}{n\pi},$$
so the Fourier series is $\frac12 + \frac2\pi\Big(\sin x + \frac{\sin 3x}{3} + \frac{\sin 5x}{5} + \cdots\Big)$. Each term is infinitely differentiable, but the sum is $1$ or $0$ off the jumps and $\frac12$ at the jumps (every $\sin n\cdot k\pi = 0$), as Dirichlet's theorem predicts.
:::

::: {#exm-triangle-wave}
## Triangle Wave and $\sum\frac{1}{n^2}$

Extend $|x|$ on $[-1, 1]$ with period $2$ ($\omega = \pi$). It is even, so $b_n = 0$; $a_0 = 2\int_0^1x\,dx = 1$, and for $n \ge 1$, by parts,
$$a_n = 2\int_0^1x\cos n\pi x\,dx = 2\Big[\frac{x\sin n\pi x}{n\pi}\Big]_0^1 - \frac{2}{n\pi}\int_0^1\sin n\pi x\,dx = \frac{2\big((-1)^n - 1\big)}{n^2\pi^2},$$
which is $0$ for even $n$ and $-\frac{4}{n^2\pi^2}$ for odd $n$. The function is continuous, so for all $x$
$$|x| = \frac12 - \frac4{\pi^2}\Big(\cos\pi x + \frac{\cos 3\pi x}{9} + \frac{\cos 5\pi x}{25} + \cdots\Big) \quad (|x| \le 1).$$
At $x = 0$: $\sum_{k\ge0}\frac{1}{(2k+1)^2} = \frac{\pi^2}{8}$. Writing $S = \sum_{n\ge1}\frac{1}{n^2}$, the even terms sum to $\frac S4$, so $S = \frac{\pi^2}{8} + \frac S4$, i.e. $\sum\frac{1}{n^2} = \frac{\pi^2}{6}$, and $\sum\frac{(-1)^{n-1}}{n^2} = \frac{\pi^2}{8} - \frac{S}{4} = \frac{\pi^2}{12}$.
:::

::: {#thm-mean-square}
## Mean-Square Convergence (stated without proof)

For $f$ as in Dirichlet's theorem, $\displaystyle\lim_{N\to\infty}\int_{-T/2}^{T/2}\big|f(x) - S_Nf(x)\big|^2dx = 0$: the area between $f$ and its Fourier approximations, measured in mean square, tends to $0$.
:::

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

x = np.linspace(-2 * np.pi, 2 * np.pi, 4000)
square = (np.mod(x, 2 * np.pi) < np.pi).astype(float)
def S_square(x, N):
    return 0.5 + 2 / np.pi * sum(np.sin(n * x) / n for n in range(1, N + 1, 2))

xt = np.linspace(-3, 3, 2000)
tri = np.abs((xt + 1) % 2 - 1)
def S_tri(x, N):
    return 0.5 - 4 / np.pi**2 * sum(np.cos(n * np.pi * x) / n**2 for n in range(1, N + 1, 2))

fig, (ax, ax2) = plt.subplots(1, 2, figsize=(13, 4))
ax.plot(x, square, color=INK, lw=2, label="square wave")
for N, col in zip([1, 5, 25], [BLUE, ORANGE, VIOLET]):
    ax.plot(x, S_square(x, N), color=col, lw=1.2, label=f"$S_{{{N}}}$")
ax.legend(fontsize=9, ncol=4, loc="upper center"); ax.set_ylim(-0.3, 1.45)
ax.set_title("jumps: convergence to the midpoint, with overshoot near the jumps")
ax2.plot(xt, tri, color=INK, lw=2, label="triangle wave")
for N, col in zip([1, 3, 9], [BLUE, ORANGE, VIOLET]):
    ax2.plot(xt, S_tri(xt, N), color=col, lw=1.2, label=f"$S_{{{N}}}$")
ax2.legend(fontsize=9, ncol=4, loc="upper center"); ax2.set_ylim(-0.1, 1.3)
ax2.set_title(r"continuous: uniform convergence, coefficients $\sim 1/n^2$")
plt.tight_layout(); plt.show()

N = np.arange(1, 200001)
print("sum 1/n^2 (2e5 terms):", np.sum(1.0 / N**2), "| pi^2/6 =", np.pi**2 / 6)
print("sum over odd n of 1/n^2:", np.sum(1.0 / N[::2]**2), "| pi^2/8 =", np.pi**2 / 8)
```
