# Existence of $n^{\text{th}}$ Roots in $\mathbb{R}$

## 1. Theorem Statement

**Theorem:** For any [real number](real-numbers.ipynb) $a > 0$ and any positive integer $n \in \mathbb{N}$, there exists a unique positive real number $x > 0$ such that:
$$x^n = a$$

This positive real number $x$ is called the **$n^{\text{th}}$ root of $a$**, denoted by $x = \sqrt[n]{a}$ or $x = a^{1/n}$.

-----

## 2. Proof of Existence and Uniqueness

The proof of this theorem relies on two key results proved in the following problems:

1. **Local Power Continuity Lemma** ([Pugh, Ch. 1, P. 15](pset-real-numbers-01.ipynb#problem-2-pugh-ch.-1-p.-15)): Demonstrates that power functions $u \mapsto u^n$ are locally continuous in $\mathbb{R}$, allowing us to bound $|u^n - y^n|$ by $|u - y|$.
2. **Supremum & Trichotomy Proof** ([Pugh, Ch. 1, P. 16](pset-real-numbers-01.ipynb#problem-3-pugh-ch.-1-p.-16)): Uses the [Axiom of Completeness](axiom-of-completeness.ipynb) to construct the candidate root $x = \sup \{s \in \mathbb{R} : s^n \le a\}$ and eliminates $x^n < a$ and $x^n > a$ via continuity.
