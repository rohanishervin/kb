# Intervals, Open Sets, and Borel Sets Are Measurable

**Prerequisites:** [Lebesgue Measurable Sets](measurable-sets.ipynb) · [Measurable Sets Form a σ-Algebra](sigma-algebra-of-measurable-sets.ipynb) · [σ-Algebras](sigma-algebra.ipynb)

Carathéodory's definition is easy to check for very simple sets: [half-lines and half-spaces](measurable-sets.ipynb#prp-half-space-measurable). Now that the measurable sets are known to be a [σ-algebra](sigma-algebra-of-measurable-sets.ipynb#cor-lebesgue-sigma-algebra), countable set operations do the rest: from half-lines we get every interval, every open and closed set, and every Borel set. So the σ-algebra $\mathcal{L}$ of Lebesgue measurable sets is rich enough for all practical purposes.

## 1\. Intervals in $\mathbb{R}$

::: {#thm-intervals-measurable}
## Intervals Are Measurable

Every interval of $\mathbb{R}$ (open, closed, half-open, bounded or not) and every singleton is Lebesgue measurable.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-intervals-measurable
We use complements, countable unions and countable intersections, under which $\mathcal{L}$ is closed.

1. $[a, \infty) \in \mathcal{L}$ for every $a$, by [the half-line proposition](measurable-sets.ipynb#prp-half-line-measurable).
2. $(-\infty, b) = [b, \infty)^c \in \mathcal{L}$. (Intersecting half-lines $[b, \infty)$ alone would never produce an open end; a complement is needed.)
3. $[a, b) = [a, \infty) \cap (-\infty, b) \in \mathcal{L}$.
4. $\{a\} = \bigcap_{n=1}^\infty \big[a, a + \tfrac1n\big) \in \mathcal{L}$.
5. $[a, b] = [a, b) \cup \{b\}$, $(a, b) = [a, b) \setminus \{a\}$, $(a, b] = (a, b) \cup \{b\}$, all in $\mathcal{L}$.
6. Unbounded intervals: $(a, \infty) = (-\infty, a]^c$ with $(-\infty, a] = (-\infty, a) \cup \{a\}$, and $\mathbb{R} = \emptyset^c$. $\blacksquare$
:::

::: {#cor-open-closed-measurable-R}
## Open and Closed Subsets of $\mathbb{R}$

Every open subset of $\mathbb{R}$ and every closed subset of $\mathbb{R}$ is measurable. In particular the [Cantor set](cantor-set.ipynb) is measurable.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-open-closed-measurable-R
An open set is a [countable union of open intervals](sigma-algebra.ipynb#lem-open-sets-of-R), and open intervals are measurable by @thm-intervals-measurable. A closed set is the complement of an open set. The Cantor set is closed. $\blacksquare$
:::

## 2\. Boxes and Open Sets in $\mathbb{R}^n$

In the plane the same process starts from half-planes. A rectangle is the intersection of four half-planes (a left, a right, an upper and a lower one); intersecting shrinking rectangles gives segments, adding or removing segments gives closed and open rectangles. In $\mathbb{R}^n$ boxes replace rectangles. Written out:

::: {#prp-boxes-measurable}
## Boxes Are Measurable

Every box in $\mathbb{R}^n$, open, closed or neither, is Lebesgue measurable.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-boxes-measurable
By the [half-space proposition](measurable-sets.ipynb#prp-half-space-measurable), $\{x : x_i \ge a\} \in \mathcal{L}$ for all $i$ and $a$. Taking complements and countable unions,
$$\{x_i < a\} = \{x_i \ge a\}^c, \qquad \{x_i > a\} = \bigcup_{k=1}^\infty \{x_i \ge a + \tfrac1k\}, \qquad \{x_i \le a\} = \{x_i > a\}^c$$
are all measurable. A box $I_1 \times \dots \times I_n$ is the intersection of the $2n$ "slabs" bounding its sides, each of one of these four forms; for example
$$(a_1, b_1) \times [a_2, b_2] = \{x_1 > a_1\} \cap \{x_1 < b_1\} \cap \{x_2 \ge a_2\} \cap \{x_2 \le b_2\}.$$
A finite intersection of measurable sets is measurable. $\blacksquare$
:::

::: {#lem-open-sets-of-Rn}
## Open Subsets of $\mathbb{R}^n$

Every open set $O \subseteq \mathbb{R}^n$ is a countable union of open boxes; the boxes can be chosen with rational corners.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-open-sets-of-Rn
Every open set is a union of open balls. Inside each ball around a point $x \in O$ one can fit a small open cube containing $x$, and by the density of $\mathbb{Q}$ its corners can be taken rational: if the ball $B_r(x) \subseteq O$, choose rationals $p_i < x_i < q_i$ with $q_i - p_i < r/\sqrt n$; then $x \in \prod_i (p_i, q_i) \subseteq B_r(x) \subseteq O$. So $O$ is the union of all open boxes with rational corners contained in $O$. There are countably many boxes with rational corners. (The boxes may overlap; this does not matter here, since we only need $O$ to be a countable union of measurable sets.) $\blacksquare$
:::

::: {#cor-open-closed-measurable}
## Open and Closed Sets Are Measurable

Every open and every closed subset of $\mathbb{R}^n$ is Lebesgue measurable.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-open-closed-measurable
Open sets are countable unions of open boxes (@lem-open-sets-of-Rn), which are measurable (@prp-boxes-measurable). Closed sets are complements of open sets. $\blacksquare$
:::

## 3\. Borel Sets

::: {#def-borel-sets}
## Borel Sets

The **Borel σ-algebra** $\mathcal{B}$ of $\mathbb{R}^n$ is the [σ-algebra generated](sigma-algebra.ipynb#def-generated-sigma-algebra) by the open sets. Its members, the **Borel sets**, are all the sets that can be built from open sets by complements, countable unions and countable intersections. For $n = 1$ this is the [σ-algebra generated by the open intervals](sigma-algebra.ipynb#cor-borel-generated-by-open-sets).
:::

Examples are the **$G_\delta$ sets**, countable intersections of open sets, and the **$F_\sigma$ sets**, countable unions of closed sets; they appear in the [approximation theorems](approximation-of-measurable-sets.ipynb).

::: {#thm-borel-sets-measurable}
## Borel Sets Are Lebesgue Measurable

$\mathcal{B} \subseteq \mathcal{L}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-borel-sets-measurable
$\mathcal{L}$ is a σ-algebra ([σ-algebra of measurable sets](sigma-algebra-of-measurable-sets.ipynb#cor-lebesgue-sigma-algebra)) containing all open sets (@cor-open-closed-measurable). $\mathcal{B}$ is the [smallest](sigma-algebra.ipynb#prp-generated-is-smallest) σ-algebra containing the open sets, so $\mathcal{B} \subseteq \mathcal{L}$. $\blacksquare$
:::

::: {.callout-note}
## Why Carathéodory's Definition Pays Off
The definition is hard to check for complicated sets, but easy for half-lines and half-spaces, and the σ-algebra property then delivers every set built from those by countable operations. In practice one never meets a set outside $\mathcal{L}$.
:::

::: {.callout-warning}
## Not Every Set Is Measurable
$\mathcal{L}$ is not all of $2^{\mathbb{R}^n}$: non-measurable sets exist (the classical example is the Vitali set, built with the axiom of choice). Their existence proofs are not constructive, and they are not needed here. One can also show that $\mathcal{B} \ne \mathcal{L}$, so that $\mathcal{B} \subsetneq \mathcal{L} \subsetneq 2^{\mathbb{R}^n}$.
:::

The figure illustrates @lem-open-sets-of-Rn for an open annulus. Dyadic squares with rational corners and side $2^{-k}$ that fit inside it are drawn by generation $k$, and they exhaust the annulus. (The lemma uses *all* open boxes with rational corners inside $O$; these overlap, and so they also cover the grid lines between the squares.)

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

from matplotlib.patches import Rectangle, Circle

def inside(x0, y0, h):
    """Is the closed square [x0,x0+h]x[y0,y0+h] inside the open annulus 0.45 < r < 1.15?"""
    xs = np.array([x0, x0 + h]); ys = np.array([y0, y0 + h])
    far = max(abs(xs)) ** 2 + max(abs(ys)) ** 2
    nx = 0 if x0 <= 0 <= x0 + h else min(abs(xs)); ny = 0 if y0 <= 0 <= y0 + h else min(abs(ys))
    return far < 1.15 ** 2 and nx ** 2 + ny ** 2 > 0.45 ** 2

cols = [BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED]
fig, ax = plt.subplots(figsize=(6.2, 6.2))
prev, counts = set(), []
for k in range(1, 7):
    h = 2.0 ** -k
    cur = {(i * h, j * h) for i in range(-2 ** (k + 1), 2 ** (k + 1)) for j in range(-2 ** (k + 1), 2 ** (k + 1))
           if inside(i * h, j * h, h)}
    new = [(x0, y0) for (x0, y0) in cur
           if (np.floor(x0 * 2 ** (k - 1)) / 2 ** (k - 1), np.floor(y0 * 2 ** (k - 1)) / 2 ** (k - 1)) not in prev]
    for (x0, y0) in new:
        ax.add_patch(Rectangle((x0, y0), h, h, facecolor=cols[k - 1], edgecolor="white", lw=0.3))
    counts.append(len(new)); prev = cur
for rad in (0.45, 1.15):
    ax.add_patch(Circle((0, 0), rad, fill=False, edgecolor=INK, lw=1.2, ls="--"))
ax.set_xlim(-1.2, 1.2); ax.set_ylim(-1.2, 1.2); ax.set_aspect("equal"); ax.axis("off")
ax.set_title("An open annulus filled by countably many squares with rational corners")
plt.tight_layout(); plt.show()
print("new squares per generation k = 1..6:", counts)
```
