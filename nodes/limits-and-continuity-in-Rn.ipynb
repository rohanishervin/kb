# Limits and Continuity of Functions of Several Variables

**Prerequisites:** [Parametrized Curves and Arc Length](parametrized-curves.ipynb#lem-norm-components) · [Point Set Topology](metric-space-topology.ipynb) · [Continuity and Homeomorphisms](continuity-topology.ipynb) · [Continuous Functions](function-continuity.ipynb)

Let $S \subseteq \mathbb{R}^n$ and $f : S \to \mathbb{R}^m$; distances are Euclidean. Continuity is the metric-space notion ([$\varepsilon$–$\delta$ form](continuity-topology.ipynb)): $f$ is **continuous at $a \in S$** if for every $\varepsilon > 0$ there is $\delta > 0$ with $|f(x) - f(a)| < \varepsilon$ for all $x \in S$ with $|x - a| < \delta$. By the [comparison of lengths with components](parametrized-curves.ipynb#lem-norm-components), balls may be replaced by boxes $\{|x_j - a_j| < \delta\}$ in this definition.

## 1\. Continuous Functions

::: {#prp-continuity-componentwise}
## Componentwise Continuity

$f = (f_1, \dots, f_m)$ is continuous at $a$ iff every $f_i$ is.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-continuity-componentwise
$|f_i(x) - f_i(a)| \le |f(x) - f(a)| \le \sqrt m\max_i|f_i(x) - f_i(a)|$. Given $\varepsilon$, if $|x - a| < \delta_i$ gives $|f_i(x) - f_i(a)| < \varepsilon/\sqrt m$, take $\delta = \min_i\delta_i$. $\blacksquare$
:::

::: {#prp-basic-continuous}
## Basic Continuous Functions

The following are continuous: affine maps $\mathbb{R}^n \to \mathbb{R}^m$ (in particular constants and the projections $\pi_j(x) = x_j$); the sum $s(x, y) = x + y$ and product $p(x, y) = xy$ on $\mathbb{R}^2$; the quotient $q(x, y) = x/y$ on $\{y \ne 0\}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-basic-continuous
*Affine.* By componentwise continuity it suffices to take $f(x) = \sum_jc_jx_j + d$: $|f(x) - f(a)| \le \sum_j|c_j|\,|x_j - a_j| \le C|x - a|$ with $C = \sum_j|c_j|$; take $\delta = \varepsilon/(C + 1)$. The sum is affine.
*Product.* If $|x - a| < 1$ then $|x| < |a| + 1$, and $|xy - ab| \le |x|\,|y - b| + |b|\,|x - a| < (|a| + 1)|y - b| + |b|\,|x - a|$; both terms are $< \varepsilon/2$ when $|x - a| < \min\{1, \frac{\varepsilon}{2(|b| + 1)}\}$ and $|y - b| < \frac{\varepsilon}{2(|a| + 1)}$.
*Quotient.* Let $b \ne 0$. If $|y - b| < |b|/2$ then $|y| > |b|/2$, and $\big|\frac xy - \frac ab\big| = \frac{|(x - a)b - a(y - b)|}{|y|\,|b|} \le \frac{2}{|b|}|x - a| + \frac{2|a|}{b^2}|y - b|$, which is small when $|x - a|$, $|y - b|$ are. $\blacksquare$
:::

::: {#thm-composition-continuous}
## Compositions

If $f : S \to \mathbb{R}^m$ is continuous at $a$ and $g : T \to \mathbb{R}^p$ (with $f(S) \subseteq T$) is continuous at $f(a)$, then $g\circ f$ is continuous at $a$. Consequently polynomials in $x_1, \dots, x_n$ are continuous on $\mathbb{R}^n$, rational functions are continuous where the denominator is non-zero, and so are compositions such as $e^{x^2 + y^2}$ or $\sin(xy)$ with the continuous functions of one variable.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-composition-continuous
Given $\varepsilon$, choose $\eta$ with $|g(y) - g(f(a))| < \varepsilon$ for $|y - f(a)| < \eta$, $y \in T$, then $\delta$ with $|f(x) - f(a)| < \eta$ for $|x - a| < \delta$. A product $fg$ of continuous real functions is $p\circ(f, g)$, continuous by @prp-continuity-componentwise and @prp-basic-continuous; similarly sums and quotients. Monomials are products of projections, polynomials sums of monomials. $\blacksquare$
:::

## 2\. Limits

::: {#def-limit-Rn}
## Limit

Let $a$ be a [limit point](metric-space-topology.ipynb) of $S$ (every deleted ball $\{0 < |x - a| < r\}$ meets $S$; $a$ need not belong to $S$). Then $\lim_{x\to a}f(x) = L$ means: for every $\varepsilon > 0$ there is $\delta > 0$ with $|f(x) - L| < \varepsilon$ for all $x \in S$ with $0 < |x - a| < \delta$.
:::

The value $f(a)$, if defined, plays no role. The limit is unique: if $L \ne L'$ were both limits, take $\varepsilon = \frac12|L - L'|$; points of $S$ in a small deleted ball (they exist since $a$ is a limit point) would be within $\varepsilon$ of both.

::: {#prp-limit-rules}
## Limits, Continuity and the Limit Rules

1. If $a \in S$ is a limit point of $S$, $f$ is continuous at $a$ iff $\lim_{x\to a}f(x) = f(a)$.
2. $\lim_{x\to a}f = L$ iff $\lim_{x\to a}f_i = L_i$ for all $i$.
3. If $\lim f = L$ and $\lim g = L'$ (real-valued), then $\lim(f + g) = L + L'$, $\lim fg = LL'$, and if $L' \ne 0$, $\lim f/g = L/L'$.
4. If $\lim_{x\to a}f(x) = L$ and $\gamma : I \to S\setminus\{a\}$ satisfies $\gamma(t) \to a$ as $t \to t_0$, then $f(\gamma(t)) \to L$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-limit-rules
Let $F$ be defined on $S\cup\{a\}$ by $F = f$ off $a$ and $F(a) = L$. Directly from the definitions, $\lim_{x\to a}f = L$ iff $F$ is continuous at $a$. (1) is this with $L = f(a)$. (2) follows from @prp-continuity-componentwise. (3) Apply @thm-composition-continuous to $s\circ(F, G)$, $p\circ(F, G)$, $q\circ(F, G)$; for the quotient, $|g(x) - L'| < |L'|/2$ near $a$, so $g \ne 0$ on $S$ near $a$ and $a$ is a limit point of the domain of $f/g$. (4) $F\circ\gamma$ (extended by $a$ at $t_0$) is continuous at $t_0$. $\blacksquare$
:::

Part 4 gives the standard test for non-existence: if $f$ has different limits along two paths into $a$, or no limit along one path, then $\lim_{x\to a}f$ does not exist. Agreement along all lines through $a$ is **not** enough (@exm-parabola-path). For limits at $0$ in $\mathbb{R}^2$, polar coordinates $x = r\cos\theta$, $y = r\sin\theta$ are convenient: $\lim f = L$ iff $\sup_\theta|f(r\cos\theta, r\sin\theta) - L| \to 0$ as $r \to 0^+$.

::: {#exm-polar-limits}
## $x^my^n/(x^2 + y^2)$

Let $f(x, y) = \dfrac{x^my^n}{x^2 + y^2}$ on $\mathbb{R}^2\setminus\{0\}$, $m, n \ge 0$ integers. In polar coordinates $f = r^{m+n-2}\cos^m\theta\sin^n\theta$.

* $m + n > 2$: $|f| \le r^{m+n-2} \to 0$, so $\lim_{(x,y)\to0}f = 0$; $f$ extended by $f(0) = 0$ is continuous.
* $m + n = 2$: $f = \cos^m\theta\sin^n\theta$ is constant on each ray and the constant depends on $\theta$ (e.g. $\frac{xy}{x^2 + y^2} = \frac12\sin2\theta$), so there is no limit.
* $m + n < 2$: on the ray $\theta = \frac\pi4$, $|f| = 2^{-(m+n)/2}r^{m+n-2} \to \infty$; no limit.
:::

::: {#exm-parabola-path}
## Lines Are Not Enough

$f(x, y) = \dfrac{x^2y}{x^4 + y^2}$, $(x, y) \ne 0$. On the line $y = kx$, $f = \frac{kx}{x^2 + k^2} \to 0$; on the $y$-axis $f = 0$. But on the parabola $y = x^2$, $f = \frac12$. So $\lim_{(x,y)\to0}f$ does not exist although every line gives $0$.
:::

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

fig, axs = plt.subplots(1, 3, figsize=(14, 4.3))
x = np.linspace(-1, 1, 401); X, Y = np.meshgrid(x, x)
with np.errstate(invalid="ignore", divide="ignore"):
    F = [X**2 * Y / (X**2 + Y**2), X * Y / (X**2 + Y**2), X**2 * Y / (X**4 + Y**2)]
titles = [r"$\frac{x^2y}{x^2+y^2}$: limit $0$", r"$\frac{xy}{x^2+y^2}$: constant on rays", r"$\frac{x^2y}{x^4+y^2}$: $1/2$ on $y = x^2$"]
for ax, Fk, tt in zip(axs, F, titles):
    im = ax.pcolormesh(X, Y, Fk, cmap="coolwarm", vmin=-0.5, vmax=0.5, shading="auto")
    ax.set_aspect("equal"); ax.set_title(tt)
axs[2].plot(x, x**2, color=INK, lw=1, ls="--")
fig.colorbar(im, ax=axs, shrink=0.8)
plt.show()
r = 10.0 ** -np.arange(1, 7)
print("x^2y/(x^4+y^2) on y = 2x:", [round(float(v), 8) for v in (r**2 * 2 * r / (r**4 + 4 * r**2))])
print("x^2y/(x^4+y^2) on y = x^2:", [float(v) for v in (r**4 / (2 * r**4))])
```
