# Linear Programs

A **linear program (LP)** is a fundamental class of mathematical optimization problems in which both the objective function to be optimized and all constraints are linear functions. The feasible region of a linear program forms a [polyhedron](polyhedra-and-polytopes.ipynb), which is a closed [convex set](convex-sets-and-convex-hulls.ipynb) formed by the intersection of [halfspaces](hyperplane-separation-theorems.ipynb).

::: {#def-linear-program}
## Linear Program (Standard Inequality Form)
A linear program in standard inequality form is defined as:
$$
\begin{aligned}
\text{maximize} \quad & c^T x \\
\text{subject to} \quad & A x \le b
\end{aligned}
$$
where:

- $x \in \mathbb{R}^n$ is the vector of decision variables.
- $c \in \mathbb{R}^n$ is the objective coefficient vector.
- $A \in \mathbb{R}^{m \times n}$ is the constraint matrix.
- $b \in \mathbb{R}^m$ is the right-hand side bounding vector.
:::

## 1\. Solutions and Feasibility

::: {#def-feasible-solution}
## Feasible Solution
A vector $x \in \mathbb{R}^n$ is called a **feasible solution** if it satisfies all given constraints (e.g., $A x \le b$). The set of all feasible solutions forms the feasible region, which is a [polyhedron](polyhedra-and-polytopes.ipynb).
:::

::: {#def-optimal-solution}
## Optimal Solution
A feasible solution $x^*$ is called an **optimal solution** if it maximizes (or minimizes) the objective function among all feasible solutions. That is, $c^T x^* \ge c^T x$ for all feasible $x$ (in a maximization problem). If the feasible region is a [polytope](polyhedra-and-polytopes.ipynb), an optimal solution always occurs at an extreme point (vertex).
:::

## 2\. Types of LP Outcomes

Every linear program resolves into exactly one of three mutually exclusive states:

1. **Infeasible:** The feasible region is entirely empty (i.e., no solution $x \in \mathbb{R}^n$ satisfies all constraints simultaneously).
2. **Optimal Solution Exists:** The LP possesses an optimal value, and there is either:
   - A **unique** optimal solution $x^*$.
   - **Infinitely many** optimal solutions (which occurs when a level set of the objective function is parallel to a boundary face of the feasible region).
3. **Unbounded:** The feasible region is non-empty, but the objective function can be made arbitrarily large (in a maximization problem). Thus, no feasible solution maximizes $c^T x$.

## 3\. Modeling Tricks

Several seemingly non-linear or non-standard constraints can be elegantly reformulated into standard linear inequalities:

### 3.1\. Equality Constraints
An equality constraint $a^T x = b$ can be seamlessly replaced by a pair of opposing inequalities:
$$
\begin{aligned}
a^T x &\le b \\
-a^T x &\le -b
\end{aligned}
$$

### 3.2\. Absolute Values in Constraints
A bounded absolute value constraint of the form $|a^T x| \le b$ is equivalent to two standard linear inequalities:
$$
\begin{aligned}
a^T x &\le b \\
-a^T x &\le b
\end{aligned}
$$

### 3.3\. Absolute Values in the Objective
If the objective involves minimizing an absolute value, such as $\min |x|$, this can be reformulated into a standard linear program by introducing an auxiliary decision variable $t$:
$$
\begin{aligned}
\text{minimize} \quad & t \\
\text{subject to} \quad & x \le t \\
& -x \le t
\end{aligned}
$$
*(Note: This trick only works for minimization of absolute values, as maximizing an absolute value leads to a non-convex feasible region).*
