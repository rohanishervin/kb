# Irrational Numbers and the Euclidean Algorithm

**Prerequisites:** [Real Numbers](real-numbers.ipynb) · [Archimedean Property of $\mathbb{R}$](archimedean-property-of-R.ipynb)

A real number is **rational** if it equals $p/q$ with $p \in \mathbb{Z}$, $q \in \mathbb{N}$, and **irrational** otherwise. Two tests for irrationality are proved here: a direct one for square roots and a general one based on the Euclidean algorithm.

## 1\. Square Roots

::: {#thm-sqrt-irrational}
## Irrationality of $\sqrt{n}$

Let $n \in \mathbb{N}$. If a rational number $x > 0$ satisfies $x^2 = n$, then $x \in \mathbb{N}$, so $n$ is a perfect square. Hence the positive real root $\sqrt{n}$ ([it exists](existence-of-nth-roots-in-R.ipynb)) is irrational unless $n = m^2$ with $m \in \mathbb{N}$. In particular $\sqrt{2}, \sqrt{3}, \sqrt{5}, \sqrt{6}$ are irrational.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-sqrt-irrational
Since $x \in \mathbb{Q}$, $qx \in \mathbb{Z}$ for some $q \in \mathbb{N}$; let $q$ be the **smallest** such positive integer. Let $m = \lfloor x \rfloor$, so $m \le x < m + 1$, and put
$$q' = q(x - m) = qx - qm .$$
Then $q' \in \mathbb{Z}$ and $0 \le q' < q$. Moreover, using $x^2 = n$,
$$q'x = qx^2 - m\,(qx) = qn - m\,(qx) \in \mathbb{Z}.$$
If $q' > 0$ this contradicts the minimality of $q$. Hence $q' = 0$, that is $x = m \in \mathbb{N}$ and $n = m^2$. $\blacksquare$
:::

## 2\. The Euclidean Algorithm for Two Lengths

Two numbers $a, b > 0$ are **commensurable** if $a = Mu$ and $b = Nu$ for some $u > 0$ and $M, N \in \mathbb{N}$; such a $u$ is a **common measure**. Clearly $a, b$ are commensurable if and only if $a/b \in \mathbb{Q}$.

::: {#def-euclidean-algorithm}
## The Euclidean Algorithm

Let $a_0, a_1 > 0$. Put $n_1 = \lfloor a_0 / a_1 \rfloor$ and $a_2 = a_0 - n_1 a_1$, so $0 \le a_2 < a_1$. If $a_2 > 0$, repeat with $a_1, a_2$, and so on:
$$a_{k-1} = n_k\, a_k + a_{k+1}, \qquad n_k = \lfloor a_{k-1}/a_k \rfloor \in \mathbb{Z}_{\ge 0}, \qquad 0 \le a_{k+1} < a_k .$$
The algorithm **stops** at step $k$ if $a_{k+1} = 0$; otherwise it produces an infinite strictly decreasing sequence $a_1 > a_2 > a_3 > \dots > 0$.
:::

The integers $n_k$ exist by the [Archimedean property](archimedean-property-of-R.ipynb).

::: {#thm-euclid-rational}
## Rationality Criterion

Let $a_0, a_1 > 0$. Then $a_0 / a_1 \in \mathbb{Q}$ if and only if the Euclidean algorithm stops. In that case the last nonzero remainder $a_k$ is a common measure of $a_0, a_1$, and every common measure $u$ satisfies $a_k / u \in \mathbb{N}$; that is, $a_k$ is the **greatest common measure**. Moreover
$$\frac{a_0}{a_1} = n_1 + \cfrac{1}{n_2 + \cfrac{1}{\ddots + \cfrac{1}{n_k}}} .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-euclid-rational
*If the algorithm stops at step $k$*, then $a_{k-1} = n_k a_k$. Going backwards with $a_{j-1} = n_j a_j + a_{j+1}$: if $a_j$ and $a_{j+1}$ are integer multiples of $a_k$, so is $a_{j-1}$. By induction $a_0 = M a_k$ and $a_1 = N a_k$ with $M, N \in \mathbb{N}$, so $a_k$ is a common measure and $a_0/a_1 = M/N \in \mathbb{Q}$.

*If $a_0/a_1 \in \mathbb{Q}$*, let $u$ be any common measure: $a_0 = Mu$, $a_1 = Nu$. Going forwards with $a_{j+1} = a_{j-1} - n_j a_j$: if $a_{j-1}, a_j$ are integer multiples of $u$, so is $a_{j+1}$. Hence every $a_j$ is a nonnegative integer multiple of $u$, say $a_j = m_j u$. The integers $m_1 > m_2 > \dots \ge 0$ are strictly decreasing as long as the algorithm runs, so after at most $N$ steps some remainder is $0$: the algorithm stops. The same induction shows that the last nonzero remainder $a_k = m_k u$ is an integer multiple of $u$.

*Continued fraction.* Dividing $a_{j-1} = n_j a_j + a_{j+1}$ by $a_j$ gives $\dfrac{a_{j-1}}{a_j} = n_j + \dfrac{1}{a_j / a_{j+1}}$ for $j < k$, and $a_{k-1}/a_k = n_k$. Substituting successively gives the formula. $\blacksquare$
:::

::: {#exm-euclid-integers}
## Two Integers

For $a_0 = 1071$, $a_1 = 462$:
$$1071 = 2 \cdot 462 + 147, \qquad 462 = 3 \cdot 147 + 21, \qquad 147 = 7 \cdot 21 + 0 .$$
The greatest common measure is $21$ ($1071 = 51 \cdot 21$, $462 = 22 \cdot 21$), and $\dfrac{1071}{462} = \dfrac{51}{22} = 2 + \cfrac{1}{3 + \cfrac{1}{7}}$.
:::

::: {#exm-golden-ratio}
## The Golden Ratio Is Irrational

Let $\varphi = \frac{1 + \sqrt{5}}{2}$, the positive root of $x^2 = x + 1$, so $\varphi = 1 + \frac{1}{\varphi}$ and $1 < \varphi < 2$. Run the algorithm on $a_0 = \varphi$, $a_1 = 1$:
$$\varphi = 1 \cdot 1 + \tfrac{1}{\varphi}, \qquad 1 = 1 \cdot \tfrac{1}{\varphi} + \tfrac{1}{\varphi^2}, \qquad \tfrac{1}{\varphi} = 1 \cdot \tfrac{1}{\varphi^2} + \tfrac{1}{\varphi^3}, \ \dots$$
Each line is the first one multiplied by $\varphi^{-j}$, so $a_j = \varphi^{1-j} > 0$ and $n_j = 1$ for all $j$. The algorithm never stops, so $\varphi \notin \mathbb{Q}$ (and the continued fraction of $\varphi$ is $1 + \frac{1}{1 + \frac{1}{1 + \cdots}}$).
:::

::: {.callout-note}
## Self-Similarity Detects Irrationality
Whenever the ratio $a_{j-1}/a_j$ repeats itself at some step, the algorithm repeats forever and the ratio is irrational. For $\varphi$ the ratio is the same at every step; for $\sqrt{2}$ (start with $a_0 = \sqrt{2}$, $a_1 = 1$) the ratio $a_1/a_2 = \sqrt{2} + 1$ appears at every later step.
:::
