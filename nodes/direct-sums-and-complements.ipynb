# Complements, Direct Sums, and Direct Products

**Prerequisites:** [Basis and Dimension](basis-and-dimension.ipynb) · [Sum of Subspaces](subspaces.ipynb#def-sum-subspaces)

When a space is the sum of two subspaces and every vector splits **uniquely** into a piece from each, the space is the **direct sum** of the two. Direct sums let us break a problem into independent smaller problems; the decomposition of $\mathbb{R}^3$ into a plane and a line, of matrices into symmetric and antisymmetric parts, and of functions into even and odd parts are all instances.

Throughout, $V$ is a [vector space](vector-spaces.ipynb) over a field $K$ and $U, W$ are [subspaces](subspaces.ipynb#def-subspace) of $V$.

## 1\. Complements and Direct Sums

::: {#def-direct-sum}
## Direct Sum and Complement

We say that $V$ is the **direct sum** of $U$ and $W$, and write $V = U \oplus W$, if every $v \in V$ can be written **uniquely** as

$$v = u + w, \qquad u \in U,\ w \in W.$$

In this case $W$ is called a **(linear) complement** of $U$ in $V$. The vectors $u$ and $w$ are the **components** of $v$.
:::

$V = U + W$ only says that a decomposition exists; $V = U \oplus W$ adds that it is unique. Uniqueness has a simple test.

::: {#prp-direct-sum-intersection}
## Direct Sum via Intersection

Suppose $V = U + W$. Then $V = U \oplus W$ if and only if $U \cap W = \{0\}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-direct-sum-intersection
($\Rightarrow$) If $0 \ne h \in U \cap W$, then $h = h + 0 = 0 + h$ are two different decompositions of $h$ (first component in $U$, second in $W$).

($\Leftarrow$) If $u + w = u' + w'$ with $u, u' \in U$ and $w, w' \in W$, then $u - u' = w' - w$. The left side lies in $U$ and the right side in $W$, so both lie in $U \cap W = \{0\}$; hence $u = u'$ and $w = w'$. $\blacksquare$
:::

::: {#exm-direct-sums}
## Three Direct Sums

1. **Plane and line.** In $\mathbb{R}^3$, the $xy$-plane $U$ and the $z$-axis $W$ satisfy $\mathbb{R}^3 = U \oplus W$: $(x, y, z) = (x, y, 0) + (0, 0, z)$, and $U \cap W = \{0\}$. The tilted line $W' = \operatorname{span}(1, 0, 1)$ is another complement of $U$: $(x, y, z) = (x - z, y, 0) + z\,(1, 0, 1)$.
2. **Symmetric and antisymmetric matrices.** Let $\operatorname{Sym}_n$ and $\operatorname{Alt}_n$ be the subspaces of $M_n(\mathbb{R})$ with $A^{\mathsf T} = A$ and $A^{\mathsf T} = -A$. Every matrix splits as
   $$A = \tfrac12 (A + A^{\mathsf T}) + \tfrac12 (A - A^{\mathsf T}),$$
   with the first part symmetric and the second antisymmetric, and a matrix that is both satisfies $A = A^{\mathsf T} = -A$, so $A = 0$. Hence $M_n(\mathbb{R}) = \operatorname{Sym}_n \oplus \operatorname{Alt}_n$.
3. **Even and odd functions.** Every $f : \mathbb{R} \to \mathbb{R}$ is $f(x) = \frac{f(x) + f(-x)}{2} + \frac{f(x) - f(-x)}{2}$, an even plus an odd function, and only $0$ is both. For $f = \exp$ this is $e^x = \cosh x + \sinh x$.
:::

::: {#prp-complement-exists}
## Complements Exist and Are Not Unique

Let $V$ be finite-dimensional and $U \le V$.

1. $U$ has a complement in $V$.
2. If $\{0\} \ne U \ne V$ and $K$ is infinite, then $U$ has infinitely many different complements.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-complement-exists
1. Take a basis $u_1, \dots, u_k$ of $U$ and [extend it](basis-and-dimension.ipynb#thm-extend-to-basis) to a basis $u_1, \dots, u_k, w_1, \dots, w_m$ of $V$; put $W = \operatorname{span}(w_1, \dots, w_m)$. Every $v \in V$ is $\sum a_i u_i + \sum b_j w_j \in U + W$. If $x \in U \cap W$, then $x = \sum a_i u_i = \sum b_j w_j$, and the independence of the whole basis gives all $a_i = b_j = 0$, so $x = 0$. By @prp-direct-sum-intersection, $V = U \oplus W$.

2. With $W$ as in part 1, we have $m \ge 1$ because $U \ne V$; fix $0 \ne u \in U$ (possible as $U \ne \{0\}$). For $t \in K$ let $W_t = \operatorname{span}(w_1 + t u, w_2, \dots, w_m)$. The list $u_1, \dots, u_k, w_1 + t u, w_2, \dots, w_m$ still spans $V$ (since $w_1 = (w_1 + tu) - tu$) and has $\dim V$ elements, so it is a basis; the argument of part 1 shows $V = U \oplus W_t$. If $W_s = W_t$ with $s \ne t$, then $(w_1 + su) - (w_1 + tu) = (s - t) u$ would lie in $W_s \cap U = \{0\}$, which is false. So the $W_t$, $t \in K$, are infinitely many different complements. $\blacksquare$
:::

The animation shows part 2 in the plane. The blue line $U$ is fixed and the orange line $W$ is a complement. The yellow vector $v$ is split as $v = u + w$ by drawing a parallel to $W$ through the tip of $v$. Turning $W$ gives a different complement and a different $U$-component $u$: the component depends on the complement, not only on $U$.


```python

from la_manim import *

class ComplementsAreNotUnique(GridScene):
    def construct(self):
        self.plane.set_stroke(opacity=0.18)
        v = np.array([2.0, 1.5, 0.0])
        theta = ValueTracker(90 * DEGREES)
        d = lambda: np.array([np.cos(theta.get_value()), np.sin(theta.get_value()), 0.0])
        def parts():
            t = v[1] / d()[1]                     # v = s e1 + t d
            return np.array([v[0] - t * d()[0], 0, 0]), t * d()
        style = dict(buff=0, stroke_width=5, max_tip_length_to_length_ratio=0.2, tip_length=0.22)

        U = Line([-6, 0, 0], [6, 0, 0], color=C_BLUE, stroke_width=4)
        U_lab = MathTex("U", color=C_BLUE).move_to([4.4, -0.35, 0])
        W = always_redraw(lambda: Line(-7 * d(), 7 * d(), color=C_ORANGE, stroke_width=4))
        W_lab = always_redraw(lambda: MathTex("W", color=C_ORANGE).move_to(2.3 * d() + 0.35 * np.array([d()[1], -d()[0], 0])))
        u_arrow = always_redraw(lambda: Arrow(ORIGIN, parts()[0], color=C_BLUE, **style))
        w_arrow = always_redraw(lambda: Arrow(parts()[0], v, color=C_ORANGE, **style))
        v_arrow = Arrow(ORIGIN, v, color=C_YELLOW, **style)
        v_lab = MathTex("v", color=C_YELLOW).next_to(v, UR, buff=0.08)
        u_lab = always_redraw(lambda: MathTex("u", color=C_BLUE).next_to(parts()[0], DOWN, buff=0.15))
        w_lab = always_redraw(lambda: MathTex("w", color=C_ORANGE).next_to((parts()[0] + v) / 2, RIGHT, buff=0.12))

        readout = VGroup(MathTex(r"v = u + w,\quad u =", color=C_INK),
                         DecimalNumber(2.0, num_decimal_places=2, color=C_BLUE),
                         MathTex(r"\,e_1", color=C_INK)).arrange(RIGHT, buff=0.1).scale(0.75).to_corner(UL, buff=0.3)
        readout[1].add_updater(lambda m: m.set_value(parts()[0][0]))
        box = BackgroundRectangle(readout, fill_opacity=0.85, buff=0.15, color=config.background_color)

        self.add(U, U_lab, W, W_lab, u_arrow, w_arrow, v_arrow, v_lab, u_lab, w_lab, box, readout)
        self.wait(1)
        for ang in (55, 125, 70, 90):
            self.play(theta.animate.set_value(ang * DEGREES), run_time=2.2)
            self.wait(0.6)
        msg = text_caption(r"Each line $W \neq U$ is a complement, with its own $u$")
        self.play(FadeIn(msg)); self.wait(2)

show(ComplementsAreNotUnique)
```


## 2\. Dimension of a Direct Sum

::: {#prp-dim-direct-sum}
## Bases and Dimension of a Direct Sum

Let $V = U \oplus W$ with $U, W$ finite-dimensional. If $u_1, \dots, u_k$ is a basis of $U$ and $w_1, \dots, w_m$ a basis of $W$, then $u_1, \dots, u_k, w_1, \dots, w_m$ is a basis of $V$. In particular
$$\dim (U \oplus W) = \dim U + \dim W.$$
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-dim-direct-sum
**Spanning.** Every $v = u + w$ with $u \in U = \operatorname{span}(u_i)$ and $w \in W = \operatorname{span}(w_j)$.

**Independence.** If $\sum a_i u_i + \sum b_j w_j = 0$, then $\sum a_i u_i = -\sum b_j w_j$ lies in $U \cap W = \{0\}$. So $\sum a_i u_i = 0$ and $\sum b_j w_j = 0$, and all coefficients vanish because each list is independent. $\blacksquare$
:::

## 3\. Direct Sums of Several Subspaces

::: {#def-direct-sum-multiple}
## Direct Sum of $s$ Subspaces

Let $U_1, \dots, U_s \le V$. We write $V = U_1 \oplus \dots \oplus U_s$ if every $v \in V$ can be written uniquely as $v = u_1 + \dots + u_s$ with $u_i \in U_i$. The sum $U_1 + \dots + U_s$ is called **direct** if every vector of it has a unique such decomposition.
:::

::: {#prp-direct-sum-criterion}
## Criteria for a Direct Sum

For subspaces $U_1, \dots, U_s \le V$ the following are equivalent:

1. The sum $U_1 + \dots + U_s$ is direct.
2. The only decomposition of $0$ is $0 = 0 + \dots + 0$: if $u_1 + \dots + u_s = 0$ with $u_i \in U_i$, then all $u_i = 0$.
3. $U_i \cap \sum_{j \ne i} U_j = \{0\}$ for every $i$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-direct-sum-criterion
(1 $\Rightarrow$ 2) $0 = 0 + \dots + 0$ is one decomposition of $0$; by uniqueness it is the only one.

(2 $\Rightarrow$ 1) If $\sum u_i = \sum u_i'$ with $u_i, u_i' \in U_i$, then $\sum (u_i - u_i') = 0$ with $u_i - u_i' \in U_i$, so $u_i = u_i'$ for all $i$.

(2 $\Rightarrow$ 3) If $x \in U_i \cap \sum_{j \ne i} U_j$, write $x = \sum_{j \ne i} u_j$. Then $(-x) + \sum_{j \ne i} u_j = 0$ is a decomposition of $0$ with $i$-th component $-x$, so $x = 0$.

(3 $\Rightarrow$ 2) If $\sum_j u_j = 0$, then for each $i$, $u_i = -\sum_{j \ne i} u_j \in U_i \cap \sum_{j \ne i} U_j = \{0\}$. $\blacksquare$
:::

::: {.callout-warning}
## Pairwise Trivial Intersections Are Not Enough
For three or more subspaces, $U_i \cap U_j = \{0\}$ for all $i \ne j$ does **not** make the sum direct. Three different lines through the origin in a plane meet pairwise only in $0$, but their sum is the plane, of dimension $2 < 3$, and $0$ has non-trivial decompositions. Condition 3 above compares each $U_i$ with the sum of **all** the others.
:::

::: {#prp-dim-direct-sum-multiple}
## Dimension of a Multiple Direct Sum

If $V = U_1 \oplus \dots \oplus U_s$ with each $U_i$ finite-dimensional, then putting bases of the $U_i$ together gives a basis of $V$, and
$$\dim V = \dim U_1 + \dots + \dim U_s.$$
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-dim-direct-sum-multiple
The combined list spans $V$, since every $v$ is a sum of elements of the $U_i$. If a linear combination of the combined list is $0$, group it as $x_1 + \dots + x_s = 0$ where $x_i \in U_i$ collects the terms from the basis of $U_i$. By @prp-direct-sum-criterion all $x_i = 0$, and then all coefficients in $x_i$ vanish because the basis of $U_i$ is independent. $\blacksquare$
:::

## 4\. The Dimension Formula for Sums

::: {#thm-dim-sum}
## Dimension of a Sum of Two Subspaces

If $U$ and $W$ are finite-dimensional subspaces of $V$, then
$$\dim (U + W) = \dim U + \dim W - \dim (U \cap W).$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-dim-sum
Let $z_1, \dots, z_r$ be a basis of $U \cap W$. [Extend it](basis-and-dimension.ipynb#thm-extend-to-basis) to a basis $z_1, \dots, z_r, u_1, \dots, u_p$ of $U$ and, separately, to a basis $z_1, \dots, z_r, w_1, \dots, w_q$ of $W$. We show that

$$\mathcal{B} = (z_1, \dots, z_r, u_1, \dots, u_p, w_1, \dots, w_q)$$

is a basis of $U + W$.

**Spanning.** Each $u + w \in U + W$ is a combination of the basis of $U$ plus a combination of the basis of $W$, hence a combination of $\mathcal{B}$.

**Independence.** Suppose $\sum a_i z_i + \sum b_j u_j + \sum c_k w_k = 0$. Then
$$y := \sum c_k w_k = -\sum a_i z_i - \sum b_j u_j$$
lies in $W$ (left expression) and in $U$ (right expression), hence in $U \cap W$, so $y = \sum d_i z_i$ for some $d_i$. Then $\sum d_i z_i - \sum c_k w_k = 0$ is a relation in the basis of $W$, so all $c_k = 0$. Now $\sum a_i z_i + \sum b_j u_j = 0$ is a relation in the basis of $U$, so all $a_i = b_j = 0$.

Counting, $\dim(U + W) = r + p + q = (r + p) + (r + q) - r$. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Inclusion-Exclusion for Dimensions
The formula mirrors $|A \cup B| = |A| + |B| - |A \cap B|$ for finite sets: the common part is counted twice and must be subtracted once. Two planes through the origin in $\mathbb{R}^3$ that are not equal have $\dim(U + W) = 3$, hence $\dim(U \cap W) = 2 + 2 - 3 = 1$: they always meet in a line. When the sum is direct, $U \cap W = \{0\}$ and the formula becomes $\dim(U \oplus W) = \dim U + \dim W$.
:::

## 5\. The Direct Product

A direct sum splits one given space. The direct product does the opposite: it builds a new space out of two given ones.

::: {#def-direct-product}
## Direct Product

For vector spaces $V$ and $W$ over $K$, the **direct product** $V \times W = \{(v, w) \mid v \in V,\ w \in W\}$ is a vector space with componentwise operations
$$(v, w) + (v', w') = (v + v', w + w'), \qquad a\,(v, w) = (a v, a w).$$
:::

::: {#prp-dim-product}
## Dimension of a Direct Product

If $v_1, \dots, v_n$ is a basis of $V$ and $w_1, \dots, w_m$ a basis of $W$, then $(v_1, 0), \dots, (v_n, 0), (0, w_1), \dots, (0, w_m)$ is a basis of $V \times W$. Hence $\dim(V \times W) = \dim V + \dim W$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-dim-product
**Spanning.** $(v, w) = (v, 0) + (0, w) = \sum a_i (v_i, 0) + \sum b_j (0, w_j)$ where $v = \sum a_i v_i$ and $w = \sum b_j w_j$.

**Independence.** $\sum a_i (v_i, 0) + \sum b_j (0, w_j) = (\sum a_i v_i, \sum b_j w_j) = (0, 0)$ forces $\sum a_i v_i = 0$ and $\sum b_j w_j = 0$, so all $a_i = b_j = 0$. $\blacksquare$
:::

::: {#prp-product-vs-sum}
## Direct Sum Versus Direct Product

1. If $V = U \oplus W$, then $U \times W \to V$, $(u, w) \mapsto u + w$ is a [linear isomorphism](coordinates-and-isomorphism-with-Kn.ipynb#def-linear-isomorphism).
2. In $V \times W$, the subspaces $V \times \{0\}$ and $\{0\} \times W$ satisfy $V \times W = (V \times \{0\}) \oplus (\{0\} \times W)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-product-vs-sum
1. The map respects sums and multiples because addition in $V$ is commutative and associative and scalar multiplication distributes. It is surjective because $V = U + W$ and injective because the decomposition is unique.
2. $(v, w) = (v, 0) + (0, w)$, and $(V \times \{0\}) \cap (\{0\} \times W) = \{(0, 0)\}$; apply @prp-direct-sum-intersection. $\blacksquare$
:::
