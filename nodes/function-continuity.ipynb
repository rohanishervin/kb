# Continuous Functions

**Prerequisites:** [The Limit of a Sequence](sequence-limit.ipynb) · [Algebraic Limit Theorem](composite-sequences-limit.ipynb) · [Triangle Inequality](triangle-inequality.ipynb)

Throughout, $S \subseteq \mathbb{R}$ and $f, g : S \to \mathbb{R}$.

## 1\. Definition

::: {#def-continuity}
## Continuity

$f$ is **continuous at** $a \in S$ if
$$\forall \varepsilon > 0\ \ \exists \delta > 0\ \ \forall x \in S: \quad |x - a| < \delta \ \Rightarrow\ |f(x) - f(a)| < \varepsilon .$$
$f$ is **continuous** (on $S$) if it is continuous at every point of $S$.
:::

The order of quantifiers matters: $\delta$ may depend on $\varepsilon$ **and** on $a$. If one $\delta$ works for all $a \in S$ simultaneously, $f$ is [uniformly continuous](uniform-continuity.ipynb).

::: {#exm-continuity-delta}
## Finding $\delta$

1. $f(x) = 5x + 3$: $|f(x) - f(a)| = 5|x - a|$, so $\delta = \varepsilon / 5$ works at every $a$.
2. $f(x) = x^2$ at $a$: if $|x - a| < 1$ then $|x + a| \le 2|a| + 1$, so $|x^2 - a^2| = |x - a||x + a| \le (2|a| + 1)|x - a|$. Hence $\delta = \min\big(1, \frac{\varepsilon}{2|a| + 1}\big)$. No $\delta$ works for all $a$: for $x = a + \delta/2$, $|x^2 - a^2| = a\delta + \delta^2/4 \ge 1$ once $a \ge 1/\delta$.
3. $f(x) = 1/x$ on $(0, \infty)$ at $a$: for $|x - a| < a/2$, $x > a/2$ and $\big|\frac1x - \frac1a\big| = \frac{|x - a|}{xa} < \frac{2}{a^2}|x - a|$, so $\delta = \min\big(\frac a2, \frac{\varepsilon a^2}{2}\big)$.
:::

::: {#exm-continuity-pathological}
## Continuity at Single Points

1. $f(x) = x$ for $x \in \mathbb{Q}$ and $f(x) = 1 - x$ otherwise, on $[0, 1]$, is continuous **only** at $\frac12$. At $\frac12$: $|f(x) - \frac12| = |x - \frac12|$ in both cases, so $\delta = \varepsilon$. At $a \ne \frac12$: rationals and irrationals arbitrarily close to $a$ have values close to $a$ and to $1 - a \ne a$ respectively, so with $\varepsilon = |2a - 1|/2$ no $\delta$ works.
2. Every $f : \mathbb{Z} \to \mathbb{R}$ is continuous: with $\delta = \frac12$ the only $x \in \mathbb{Z}$ with $|x - a| < \delta$ is $a$.
3. The step function $f(x) = x + 1$ ($x \le 0$), $f(x) = x$ ($x > 0$) is discontinuous at $0$: points $x > 0$ near $0$ have $|f(x) - f(0)| \approx 1$.
:::

::: {#prp-continuity-oscillation}
## Continuity Without Mentioning $f(a)$

$f$ is continuous at $a$ if and only if for every $\varepsilon > 0$ there is $\delta > 0$ such that $|f(x_1) - f(x_2)| < \varepsilon$ for all $x_1, x_2 \in S \cap (a - \delta, a + \delta)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-continuity-oscillation
($\Rightarrow$) Take $\delta$ for $\varepsilon/2$; then $|f(x_1) - f(x_2)| \le |f(x_1) - f(a)| + |f(a) - f(x_2)| < \varepsilon$. ($\Leftarrow$) Take $x_2 = a$. $\blacksquare$
:::

## 2\. Sequences and Algebra

::: {#thm-sequential-continuity}
## Sequential Criterion

$f$ is continuous at $a$ if and only if $f(x_n) \to f(a)$ for every sequence $(x_n)$ in $S$ with $x_n \to a$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-sequential-continuity
($\Rightarrow$) Given $\varepsilon$, take $\delta$ from continuity and $N$ with $|x_n - a| < \delta$ for $n \ge N$; then $|f(x_n) - f(a)| < \varepsilon$ for $n \ge N$.

($\Leftarrow$) If $f$ is not continuous at $a$, there is $\varepsilon > 0$ such that for every $\delta = \frac1n$ some $x_n \in S$ has $|x_n - a| < \frac1n$ but $|f(x_n) - f(a)| \ge \varepsilon$. Then $x_n \to a$ while $f(x_n) \not\to f(a)$. $\blacksquare$
:::

::: {#thm-continuity-algebra}
## Sums, Products, Quotients

If $f, g$ are continuous at $a$, so are $f + g$, $cf$, $fg$, and $f/g$ provided $g(a) \ne 0$ (as a function on $\{x \in S : g(x) \ne 0\}$).
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-continuity-algebra
For $x_n \to a$ in the domain, $f(x_n) \to f(a)$ and $g(x_n) \to g(a)$, so the [limit laws](composite-sequences-limit.ipynb#thm-limit-laws) give $f(x_n) + g(x_n) \to f(a) + g(a)$ etc. Apply @thm-sequential-continuity. $\blacksquare$
:::

::: {#thm-continuity-composition}
## Composition

If $f : S \to T$ is continuous at $a$ and $g : T \to \mathbb{R}$ is continuous at $f(a)$, then $g \circ f$ is continuous at $a$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-continuity-composition
Given $\varepsilon$, choose $\eta > 0$ with $|g(y) - g(f(a))| < \varepsilon$ for $y \in T$, $|y - f(a)| < \eta$; then $\delta > 0$ with $|f(x) - f(a)| < \eta$ for $x \in S$, $|x - a| < \delta$. For such $x$, $|g(f(x)) - g(f(a))| < \varepsilon$. $\blacksquare$
:::

Constants and $x \mapsto x$ are continuous, hence so are all polynomials and all **rational functions** $P/Q$ on $\{Q \ne 0\}$.

## 3\. Trigonometric Functions

Angles are measured in radians: for $0 < \theta < \frac{\pi}{2}$, the triangle $0, (1,0), (\cos\theta, \sin\theta)$, the circular sector of angle $\theta$, and the triangle $0, (1, 0), (1, \tan\theta)$ are nested, with areas $\frac12 \sin\theta < \frac{\theta}{2} < \frac12\tan\theta$. Hence
$$\sin\theta < \theta < \tan\theta \qquad \Big(0 < \theta < \frac{\pi}{2}\Big). \tag{1}$$

::: {#lem-sin-bounds}
## Basic Inequalities

For all $h \in \mathbb{R}$: $|\sin h| \le |h|$ and $0 \le 1 - \cos h \le |h|$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-sin-bounds
For $0 < h < \frac{\pi}{2}$ this is (1); for $h \ge \frac{\pi}{2}$, $|\sin h| \le 1 < h$; $\sin$ is odd. Then $1 - \cos h = 2\sin^2\frac h2 \le 2 \cdot \big|\frac h2\big| \cdot 1 = |h|$. $\blacksquare$
:::

::: {#thm-trig-continuous}
## Continuity of the Trigonometric Functions

$\sin$ and $\cos$ are continuous on $\mathbb{R}$; $\tan, \cot, \sec, \csc$ are continuous on their domains.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-trig-continuous
By the addition formula and @lem-sin-bounds,
$$|\sin(a + h) - \sin a| = |\sin a(\cos h - 1) + \cos a \sin h| \le |1 - \cos h| + |\sin h| \le 2|h|,$$
so $\delta = \varepsilon/2$ works; similarly $|\cos(a+h) - \cos a| \le 2|h|$. The other four are quotients of these (@thm-continuity-algebra). $\blacksquare$
:::
