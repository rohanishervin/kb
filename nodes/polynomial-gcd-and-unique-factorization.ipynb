# Greatest Common Divisors and Unique Factorization of Polynomials

**Prerequisites:** [Rings and Polynomial Rings](rings-and-polynomial-rings.ipynb) · [The Division Algorithm for Polynomials](polynomial-division-algorithm.ipynb)

Polynomials with coefficients in a field behave remarkably like the integers. Both admit [division with remainder](polynomial-division-algorithm.ipynb#thm-polynomial-division), and from that one fact follow the same three results: every pair has a greatest common divisor that is a combination of the pair, "prime" elements cannot divide a product without dividing a factor, and every element factors into primes in essentially one way. The proofs below work word for word for integers.

For operators these results are the algebraic engine of the [Spectral Theorem](generalized-eigenspaces-and-spectral-theorem.ipynb): coprime factors of the characteristic polynomial split the space into independent pieces.

Throughout, $K$ is a [field](field.ipynb) and $K[x]$ its [polynomial ring](rings-and-polynomial-rings.ipynb#def-polynomial-ring).

## 1\. Divisibility

::: {#def-divisibility}
## Divisibility and Monic Polynomials

Let $d, p \in K[x]$. We say $d$ **divides** $p$, written $d \mid p$, if $p = d\, e$ for some $e \in K[x]$; then $d$ is a **divisor** of $p$ and $p$ a **multiple** of $d$. The division is exact: the [remainder](polynomial-division-algorithm.ipynb#thm-polynomial-division) is $0$.

A non-zero polynomial is **monic** if its leading coefficient is $1$:
$$x^n + a_{n-1}x^{n-1} + \dots + a_1 x + a_0 .$$
:::

::: {#prp-units-polynomials}
## Constants Are the Only Invertible Polynomials

1. If $d \mid p$ and $p \ne 0$, then $\deg d \le \deg p$.
2. $u \in K[x]$ has a multiplicative inverse if and only if $u$ is a non-zero constant.
3. Every non-zero $p$ can be written uniquely as $p = c\, \tilde p$ with $c \in K \setminus \{0\}$ and $\tilde p$ monic.
4. If $d \mid e$ and $e \mid d$, then $e = c\, d$ for a non-zero constant $c$; if both are monic, $d = e$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-units-polynomials
Over a field the [degree of a product](rings-and-polynomial-rings.ipynb#prp-degree-of-product) is the sum of the degrees, because the product of the two leading coefficients is non-zero.

1. $p = d e$ with $p \ne 0$ forces $e \ne 0$, so $\deg p = \deg d + \deg e \ge \deg d$.
2. A non-zero constant $c$ has the inverse $c^{-1}$. Conversely, if $u v = 1$, then $\deg u + \deg v = 0$, so $\deg u = 0$.
3. Take $c$ to be the leading coefficient of $p$ and $\tilde p = c^{-1} p$. If also $p = c' \tilde p'$, comparing leading coefficients gives $c = c'$, and then $\tilde p = \tilde p'$.
4. $e = d\, s$ and $d = e\, t$ give $d = d\, s t$. If $d = 0$ then $e = 0$ and any $c$ works. Otherwise $\deg s + \deg t = 0$, so $s$ is a non-zero constant $c$. If $d$ and $e$ are monic, comparing leading coefficients in $e = c d$ gives $c = 1$. $\blacksquare$
:::

So multiplying by a non-zero constant changes nothing as far as divisibility is concerned; insisting on **monic** polynomials removes this ambiguity. It plays the role that insisting on **positive** integers plays for $\mathbb{Z}$, where the invertible elements are $\pm 1$.

## 2\. The Greatest Common Divisor

::: {#thm-gcd}
## Existence of the Greatest Common Divisor (Bézout)

Let $p, q \in K[x]$, not both zero. There is a **unique** monic polynomial $d$ with the following properties.

1. $d = a\, p + b\, q$ for some polynomials $a, b \in K[x]$.
2. $d \mid p$ and $d \mid q$.
3. Every common divisor $s$ of $p$ and $q$ divides $d$.

It is called the **greatest common divisor** of $p$ and $q$, written $\gcd(p, q)$. "Greatest" refers to divisibility (property 3), and therefore also to degree: every common divisor has degree at most $\deg d$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-gcd
**Choosing $d$.** Let $S = \{\, a p + b q \mid a, b \in K[x] \,\} \setminus \{0\}$. It is non-empty: if $p \ne 0$, then $p = 1 \cdot p + 0 \cdot q \in S$, and similarly for $q$. Degrees are non-negative integers, so $S$ contains an element $d_0 = a_0 p + b_0 q$ of **smallest degree**. Let $c$ be its leading coefficient and $d = c^{-1} d_0 = (c^{-1}a_0) p + (c^{-1} b_0) q$. Then $d$ is monic, lies in $S$, has the same smallest degree, and satisfies property 1 with $a = c^{-1} a_0$, $b = c^{-1} b_0$.

**Property 2.** [Divide](polynomial-division-algorithm.ipynb#thm-polynomial-division) $p$ by the monic $d$: $p = e\, d + r$ with $\deg r < \deg d$. Then
$$r = p - e\,(a p + b q) = (1 - e a)\, p + (-e b)\, q$$
is again a combination of $p$ and $q$. If $r \ne 0$, it would be an element of $S$ of degree smaller than $\deg d$, contradicting the choice of $d$. So $r = 0$ and $d \mid p$. The same argument with $q$ gives $d \mid q$.

**Property 3.** If $p = s\, p'$ and $q = s\, q'$, then $d = a s p' + b s q' = s\,(a p' + b q')$, so $s \mid d$. In particular $\deg s \le \deg d$ by @prp-units-polynomials.

**Uniqueness.** If $d$ and $d'$ are monic and both have properties 2 and 3, then $d$ is a common divisor, so $d \mid d'$ by property 3 for $d'$; symmetrically $d' \mid d$. Both are monic, so $d = d'$ by @prp-units-polynomials. $\blacksquare$
:::

::: {.callout-note}
## Intuition: The Same Proof as for Integers
For integers the same argument shows that $\gcd(m, n)$ is the **smallest positive** number of the form $a m + b n$ with $a, b \in \mathbb{Z}$. For example $\gcd(12, 18) = 6 = 1 \cdot 18 - 1 \cdot 12$: one coefficient is typically negative. Remainders of division are again combinations of $m$ and $n$, so the smallest positive combination cannot leave a remainder. Replace "smallest positive" by "smallest degree, normalized to be monic" and the proof becomes the one above.
:::

::: {#exm-gcd}
## Computing a GCD by Repeated Division

Repeating the division (the **Euclidean algorithm**) computes the gcd: $\gcd(p, q) = \gcd(q, r)$ whenever $p = e q + r$, because the common divisors of $p, q$ and of $q, r$ are the same. For $p = x^2 - 1$ and $q = x^2 - 3x + 2$:
$$x^2 - 1 = 1 \cdot (x^2 - 3x + 2) + (3x - 3), \qquad x^2 - 3x + 2 = \big(\tfrac13 x - \tfrac23\big)(3x - 3) + 0 .$$
The last non-zero remainder is $3x - 3$; made monic it is $\gcd(p, q) = x - 1$, and reading the first line backwards gives
$$x - 1 = \tfrac13\, (x^2 - 1) - \tfrac13\, (x^2 - 3x + 2).$$
This matches the factorizations $p = (x - 1)(x + 1)$ and $q = (x - 1)(x - 2)$.
:::

::: {#def-coprime}
## Coprime Polynomials

$p$ and $q$ are **coprime** (relatively prime) if $\gcd(p, q) = 1$, that is, if their only common divisors are the non-zero constants.
:::

::: {#cor-coprime-bezout}
## Coprimality and the Equation $a p + b q = 1$

1. $p$ and $q$ are coprime if and only if $a\, p + b\, q = 1$ for some $a, b \in K[x]$.
2. If $K$ is [algebraically closed](eigenvalues-and-eigenvectors.ipynb#def-algebraically-closed-field), then $p$ and $q$ are coprime if and only if they have no common root.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-coprime-bezout
1. If $\gcd(p, q) = 1$, property 1 of @thm-gcd gives $a, b$. Conversely, if $a p + b q = 1$, every common divisor $s$ of $p$ and $q$ divides $1$, so $\deg s = 0$; the monic gcd is therefore $1$.
2. If $c$ is a common root, then $x - c$ divides both by the [factor theorem](polynomial-division-algorithm.ipynb#cor-factor-theorem), so they are not coprime. Conversely, if $d = \gcd(p, q)$ is not constant, it has a root $c \in K$ because $K$ is algebraically closed; since $d$ divides $p$ and $q$, $c$ is a root of both. $\blacksquare$
:::

::: {.callout-warning}
## "No Common Root" Depends on the Field
Over $\mathbb{R}$, $p = x^2 + 1$ and $q = (x^2 + 1)(x - 5)$ have no common **real** root, yet $\gcd(p, q) = x^2 + 1 \ne 1$. Part 2 of @cor-coprime-bezout needs every non-constant polynomial to have a root, which is exactly algebraic closure.
:::

::: {#exm-bezout-spectral}
## A Bézout Identity Used Later

$p = (x - 2)^3$ and $q = x^2 + 1$ have no common root in $\mathbb{C}$ (the roots are $2$ and $\pm i$), so they are coprime, and
$$1 = -\tfrac{1}{125}\,(11x + 2)\,(x - 2)^3 + \tfrac{1}{125}\,(11x^2 - 64x + 109)\,(x^2 + 1).$$
The coefficients $a, b$ have degree less than $\deg q$ and $\deg p$; the code below finds them by running the Euclidean algorithm backwards. The identity reappears in the [Spectral Theorem](generalized-eigenspaces-and-spectral-theorem.ipynb#exm-spectral-5x5), where substituting a matrix for $x$ splits $\mathbb{C}^5$ into two invariant pieces.
:::



## 3\. Irreducible Polynomials

::: {#def-irreducible-polynomial}
## Irreducible Polynomial

A polynomial $p \in K[x]$ of degree at least $1$ is **irreducible** (over $K$) if it cannot be written as a product $p = g h$ of two polynomials of smaller degree. Equivalently, whenever $p = g h$, one of $g, h$ is a non-zero constant. Equivalently again, the only monic divisors of $p$ are $1$ and $c^{-1} p$, where $c$ is the leading coefficient of $p$.
:::

::: {.callout-warning}
## Constants Are Neither Irreducible Nor Reducible
Every non-zero constant divides every polynomial and is invertible, so constants are excluded from the definition, just as $1$ is not counted as a prime number. Similarly, $3x - 3 = 3 (x - 1)$ is still irreducible: splitting off a constant is not a genuine factorization.
:::

::: {#exm-irreducible}
## Irreducibility Depends on the Field

1. Every polynomial of degree $1$ is irreducible: in $p = g h$, the degrees add up to $1$, so one of them is $0$.
2. A polynomial of degree $2$ or $3$ is irreducible over $K$ if and only if it has no root in $K$: any non-trivial factorization contains a factor of degree $1$, which has a root.
3. $x^2 + 1$ is irreducible in $\mathbb{R}[x]$, because it has no real root ($x^2 + 1 \ge 1$). In $\mathbb{C}[x]$ it is reducible: $x^2 + 1 = (x - i)(x + i)$.
4. Over an algebraically closed field such as $\mathbb{C}$, the irreducible polynomials are **exactly** those of degree $1$: a polynomial of degree $\ge 2$ has a root $c$, hence the factor $x - c$.
5. Over $\mathbb{R}$, the irreducible polynomials are those of degree $1$ and those of degree $2$ with negative discriminant (see @prp-real-irreducibles).
:::

::: {#prp-real-irreducibles}
## Real Polynomials Factor into Linear and Quadratic Pieces

Every non-constant $p \in \mathbb{R}[x]$ is a product of real polynomials of degree $1$ and real quadratics $x^2 + \beta x + \gamma$ with $\beta^2 - 4\gamma < 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-real-irreducibles
Induction on $\deg p$. If $p$ has a real root $c$, then $p = (x - c)\, s$ with $s$ real of smaller degree ([factor theorem](polynomial-division-algorithm.ipynb#cor-factor-theorem)). Otherwise $\deg p \ge 2$ (degree-1 polynomials have real roots) and $p$ has a non-real complex root $z$. Since the coefficients are real, $\overline{p(z)} = p(\bar z)$, so $\bar z \ne z$ is also a root, and
$$g = (x - z)(x - \bar z) = x^2 - 2\operatorname{Re}(z)\, x + |z|^2$$
has real coefficients and discriminant $4\operatorname{Re}(z)^2 - 4|z|^2 = -4 \operatorname{Im}(z)^2 < 0$. Dividing $p$ by the monic $g$ in $\mathbb{R}[x]$ gives $p = s g + r$ with real $s, r$; this is also the division in $\mathbb{C}[x]$, where $g \mid p$ (as $x - z$ and $x - \bar z$ are distinct linear factors of $p$). By the [uniqueness of the remainder](polynomial-division-algorithm.ipynb#thm-polynomial-division), $r = 0$. So $p = s g$ with $s$ real of degree $\deg p - 2$, and the induction applies to $s$. A real quadratic with negative discriminant has no real root, so it is irreducible by @exm-irreducible. $\blacksquare$
:::

The key property of irreducible polynomials is the analogue of Euclid's lemma for primes: a prime that divides a product must divide one of the factors.

::: {#lem-euclid-polynomials}
## Euclid's Lemma for Polynomials

Let $p \in K[x]$ be irreducible and $q, r \in K[x]$. If $p \mid q r$, then $p \mid q$ or $p \mid r$. More generally, if $p \mid q_1 q_2 \cdots q_m$, then $p \mid q_j$ for some $j$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-euclid-polynomials
Suppose $p \nmid q$; we show $p \mid r$. Let $d = \gcd(p, q)$. As a monic divisor of the irreducible $p$, $d$ is either $1$ or $c^{-1} p$ ($c$ the leading coefficient of $p$). The second case is impossible: $d \mid q$ would give $p = c\, d \mid q$. Hence $d = 1$, and by @thm-gcd
$$1 = a\, p + b\, q \qquad\text{for some } a, b \in K[x].$$
Multiplying by $r$ gives
$$r = a\, p\, r + b\, (q r).$$
Both terms on the right are multiples of $p$: the first obviously, the second because $p \mid q r$. Therefore $p \mid r$.

For $m$ factors use induction on $m$: write $q_1 \cdots q_m = (q_1 \cdots q_{m-1})\, q_m$. If $p \nmid q_m$, the case $m = 2$ gives $p \mid q_1 \cdots q_{m-1}$, and the induction hypothesis gives some $j \le m - 1$ with $p \mid q_j$. $\blacksquare$
:::

::: {.callout-warning}
## Irreducibility Is Essential
$x^2 \mid x \cdot x$, but $x^2$ divides neither factor. A reducible polynomial can split its divisibility across several factors; an irreducible one, like a prime number, cannot.
:::

## 4\. Unique Factorization

::: {#thm-unique-factorization}
## Unique Factorization in $K[x]$

Every non-constant $p \in K[x]$ can be written as
$$p = c\; p_1\, p_2 \cdots p_m,$$
with $c \in K$ the leading coefficient of $p$ and $p_1, \dots, p_m$ **monic irreducible** polynomials ($m \ge 1$). This factorization is unique up to the order of the factors: if also $p = c'\, q_1 \cdots q_l$ with monic irreducible $q_j$, then $c' = c$, $l = m$, and after renumbering $q_i = p_i$ for all $i$.
:::

Without the normalization to monic factors, uniqueness holds "up to order and up to multiplying the factors by non-zero constants whose product is $1$", for example $x^2 - 1 = (x - 1)(x + 1) = (2x - 2)\big(\tfrac12 x + \tfrac12\big)$.

::: {.callout-tip collapse="true"}
## Proof of @thm-unique-factorization (Existence)
Strong induction on $n = \deg p \ge 1$. Write $p = c\, \tilde p$ with $\tilde p$ monic (@prp-units-polynomials). If $\tilde p$ is irreducible, we are done with $m = 1$. Otherwise $\tilde p = g h$ with $1 \le \deg g, \deg h < n$; dividing $g$ by its leading coefficient and multiplying $h$ by it, we may take $g$ and $h$ monic (their leading coefficients multiply to $1$). By the induction hypothesis both $g$ and $h$ are products of monic irreducible polynomials, hence so is $\tilde p = g h$. $\blacksquare$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-unique-factorization (Uniqueness)
Comparing leading coefficients (all $p_i, q_j$ are monic) gives $c = c'$; cancelling the non-zero constant leaves
$$p_1 p_2 \cdots p_m = q_1 q_2 \cdots q_l .$$
We prove by induction on $m$ that $l = m$ and the $q_j$ are a rearrangement of the $p_i$.

The irreducible $p_1$ divides the right-hand side, so by @lem-euclid-polynomials it divides some $q_j$; renumber so that $j = 1$. Since $q_1$ is irreducible, its only monic divisors are $1$ and $q_1$, and $p_1 \ne 1$ (it is non-constant); hence $p_1 = q_1$. $K[x]$ has no zero divisors (degrees add), so $p_1\, (p_2 \cdots p_m - q_2 \cdots q_l) = 0$ implies
$$p_2 \cdots p_m = q_2 \cdots q_l ,$$
where an empty product means $1$.

If $m = 1$, the left side is $1$; a product of $l - 1 \ge 1$ non-constant polynomials has positive degree, so $l = 1$. If $m \ge 2$, the right side cannot be empty (the left side has positive degree), and the induction hypothesis applied to the $m - 1$ factors gives $l - 1 = m - 1$ and $\{q_2, \dots, q_l\} = \{p_2, \dots, p_m\}$ as lists up to order. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Primes Cannot Be Split Between Two Factorizations
Uniqueness comes entirely from Euclid's lemma. If one irreducible factor $p_1$ appears on one side, it must "live" inside a single factor on the other side, and there it can only be that factor itself. Cancel and repeat. The same argument proves the uniqueness of prime factorization of integers.
:::

::: {#cor-factorization-algebraically-closed}
## Factorization over an Algebraically Closed Field

Let $K$ be algebraically closed and $p \in K[x]$ monic of degree $n \ge 1$.

1. There are unique pairwise distinct $\mu_1, \dots, \mu_k \in K$ and unique exponents $r_1, \dots, r_k \ge 1$ with
$$p = (x - \mu_1)^{r_1} (x - \mu_2)^{r_2} \cdots (x - \mu_k)^{r_k}, \qquad r_1 + \dots + r_k = n .$$
2. The monic divisors of $p$ are exactly the polynomials $(x - \mu_1)^{s_1} \cdots (x - \mu_k)^{s_k}$ with $0 \le s_j \le r_j$.
3. For $i \ne j$, the polynomials $(x - \mu_i)^{r_i}$ and $\prod_{l \ne i} (x - \mu_l)^{r_l}$ are coprime.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-factorization-algebraically-closed
1. The monic irreducible polynomials over $K$ are the $x - \mu$ (@exm-irreducible). Group equal factors in @thm-unique-factorization; uniqueness of the factorization gives uniqueness of the $\mu_j$ and of their multiplicities $r_j$. The degrees add up to $n$.
2. If $d$ is monic and $p = d e$, factor $d$ and $e$ (made monic; their leading coefficients multiply to $1$) into linear factors. Their product is a factorization of $p$, so by uniqueness only the $x - \mu_j$ occur, and $x - \mu_j$ appears in $d$ at most $r_j$ times. Conversely every such product divides $p$, with the complementary product as quotient.
3. The two polynomials have no common root, since the $\mu_l$ are distinct; apply @cor-coprime-bezout. $\blacksquare$
:::

This is the shape in which the factorization is used for the [characteristic polynomial](eigenvalues-and-eigenvectors.ipynb#thm-characteristic-equation) of an operator: $\mu_j$ are the distinct eigenvalues and $r_j$ their [algebraic multiplicities](diagonalization.ipynb#def-multiplicities).

The code runs the extended Euclidean algorithm with exact rational arithmetic, checks the Bézout identities of this note, and shows how $x^4 + 1$ factors over $\mathbb{Q}$ (not at all), $\mathbb{R}$ (two quadratics) and $\mathbb{C}$ (four linear factors). The figure then shows the real factorization of $x^6 - 1$.


```python

import sympy as sp

x = sp.symbols("x")

def extended_euclid(p, q):
    # Returns (d, a, b) with d = gcd(p, q) monic and d = a p + b q.
    # Invariant: r_i = a_i p + b_i q for the two most recent remainders.
    r0, r1 = sp.Poly(p, x, domain="QQ"), sp.Poly(q, x, domain="QQ")
    a0, a1 = sp.Poly(1, x, domain="QQ"), sp.Poly(0, x, domain="QQ")
    b0, b1 = sp.Poly(0, x, domain="QQ"), sp.Poly(1, x, domain="QQ")
    while not r1.is_zero:
        e, r = r0.div(r1)                       # r0 = e r1 + r,  deg r < deg r1
        r0, r1 = r1, r
        a0, a1 = a1, a0 - e * a1
        b0, b1 = b1, b0 - e * b1
    c = r0.LC()                                 # make the gcd monic
    return r0 * (1 / c), a0 * (1 / c), b0 * (1 / c)

pairs = [(x**2 - 1, x**2 - 3*x + 2),           # gcd x - 1
         ((x - 2)**3, x**2 + 1),               # coprime: the identity of the Bezout example
         (x**4 - 1, x**6 - 1)]                 # gcd x^2 - 1
for p, q in pairs:
    d, a, b = extended_euclid(sp.expand(p), sp.expand(q))
    check = sp.expand(a.as_expr() * p + b.as_expr() * q - d.as_expr())
    print(f"gcd({sp.factor(p)}, {sp.factor(q)}) = {sp.factor(d.as_expr())}")
    print(f"   a = {sp.factor(a.as_expr())},   b = {sp.factor(b.as_expr())},   a p + b q - d = {check}\n")

p = x**4 + 1                                   # irreducibility depends on the field
print("x^4 + 1 over Q:", sp.factor(p))
print("        over R:", sp.factor(p, extension=sp.sqrt(2)), "  (both quadratics have discriminant -2)")
print("        over C:", sp.factor(p, extension=[sp.sqrt(2), sp.I]))
```

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW = "#2a78d6", "#eb6834", "#1baf7a", "#eda100"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "xtick.color": MUTED, "ytick.color": MUTED,
    "legend.frameon": False, "font.size": 10})

# x^6 - 1 = (x - 1)(x + 1)(x^2 + x + 1)(x^2 - x + 1) over R
t = np.linspace(-1.6, 1.6, 400)
factors = [(t - 1, ORANGE, r"$x-1$"), (t + 1, ORANGE, r"$x+1$"),
           (t**2 + t + 1, AQUA, r"$x^2+x+1$"), (t**2 - t + 1, AQUA, r"$x^2-x+1$")]

fig, axes = plt.subplots(1, 2, figsize=(10, 3.6))
ax = axes[0]
ax.plot(t, t**6 - 1, color=BLUE, lw=2.5, label=r"$p(x)=x^6-1$")
ax.scatter([-1, 1], [0, 0], color=YELLOW, zorder=5, s=50, label="real roots")
ax.axhline(0, color=MUTED, lw=0.8)
ax.set_ylim(-2, 6); ax.legend(loc="upper center"); ax.set_title("the polynomial and its real roots")

ax = axes[1]
for y, col, lab in factors:
    ax.plot(t, y, color=col, lw=2, ls="-" if col == AQUA else "--", label=lab)
ax.axhline(0, color=MUTED, lw=0.8)
ax.set_ylim(-3, 5); ax.legend(ncol=2, loc="upper center")
ax.set_title("irreducible real factors: lines and zero-free quadratics")
plt.tight_layout(); plt.show()
```


The two quadratic factors never touch the axis: they carry the four non-real sixth roots of unity $e^{\pm i\pi/3}, e^{\pm 2i\pi/3}$ in conjugate pairs, exactly as in the proof of @prp-real-irreducibles.
