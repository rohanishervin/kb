# Rings and Polynomial Rings

**Prerequisites:** [Fields](field.ipynb) · [Linear Maps](linear-maps.ipynb)

The general proof of the [Cayley-Hamilton Theorem](cayley-hamilton-theorem.ipynb#thm-cayley-hamilton) works with **polynomials whose coefficients are matrices**. Matrices can be added and multiplied but not always divided, and their multiplication is not commutative, so we need the algebraic structure that captures exactly this: a **ring**. This note introduces rings, the polynomial ring $R[x]$, the degree of a polynomial, and the delicate point of substituting a value for $x$ when the coefficients do not commute.

## 1\. Rings

::: {#def-ring}
## Ring

A **ring** is a set $R$ with two operations, addition $+$ and multiplication $\cdot$, such that

1. $(R, +)$ is an abelian group, with neutral element $0$ (associative, commutative, with negatives);
2. multiplication is associative, $(ab)c = a(bc)$, and has a neutral element $1$: $1a = a1 = a$;
3. multiplication distributes over addition on both sides: $a(b + c) = ab + ac$ and $(a + b)c = ac + bc$.

$R$ is **commutative** if $ab = ba$ for all $a, b$. An element $u$ is a **unit** (is **invertible**) if $uv = vu = 1$ for some $v \in R$, written $u^{-1}$.
:::

::: {#exm-rings}
## Rings

1. **Fields** $\mathbb{Q}, \mathbb{R}, \mathbb{C}$ are commutative rings in which every non-zero element is a unit.
2. **The integers** $\mathbb{Z}$: commutative; the only units are $\pm 1$.
3. **Integers modulo $6$**, $\mathbb{Z}/6 = \{0, 1, \dots, 5\}$ with arithmetic mod $6$: commutative, but $2 \cdot 3 = 0$ although $2 \ne 0$ and $3 \ne 0$. The units are $1$ and $5$.
4. **Subsets of a set.** For a set $X$, the power set $\mathcal{P}(X)$ with symmetric difference $A \,\triangle\, B$ as addition and intersection $A \cap B$ as multiplication is a commutative ring with $0 = \emptyset$ and $1 = X$, in which $A + A = 0$ and $A \cdot A = A$ for every $A$.
5. **Matrices.** $M_n(K)$ with matrix addition and multiplication is a ring, **not commutative** for $n \ge 2$. Its units are the invertible matrices; and $E^2 = 0$ for $E = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix} \ne 0$. Likewise the operators $\operatorname{End}(V)$ form a ring under composition ([algebra of operators](linear-maps.ipynb#prp-endomorphism-algebra)).
:::

## 2\. The Polynomial Ring $R[x]$

::: {#def-polynomial-ring}
## Polynomial Ring

Let $R$ be a ring. A **polynomial** over $R$ is a sequence $(a_0, a_1, a_2, \dots)$ of elements of $R$ with only finitely many non-zero terms, written
$$p(x) = a_0 + a_1 x + \dots + a_m x^m = \sum_k a_k x^k.$$
The set of polynomials is $R[x]$, with operations
$$\sum_k a_k x^k + \sum_k b_k x^k = \sum_k (a_k + b_k)\, x^k, \qquad \Big(\sum_i a_i x^i\Big)\Big(\sum_j b_j x^j\Big) = \sum_k \Big(\sum_{i + j = k} a_i b_j\Big) x^k.$$
With these, $R[x]$ is a ring; it is commutative when $R$ is. In the product, coefficients keep their order ($a_i$ before $b_j$), while the symbol $x$ commutes with everything.
:::

::: {.callout-note}
## Polynomials Are Formal
A polynomial here is its list of coefficients, not a function. Over $\mathbb{Z}/2$, the polynomials $x$ and $x^2$ are different (different coefficient lists), although $a = a^2$ for both elements $a \in \{0, 1\}$. Writing $x$ instead of a sequence is a notational convenience; it becomes essential for polynomials in several variables, where sequences of sequences would be unreadable.
:::

## 3\. Degree

::: {#def-polynomial-degree}
## Degree and Leading Coefficient

The **degree** $\deg p$ of a non-zero polynomial $p = \sum_k a_k x^k$ is the largest $m$ with $a_m \ne 0$, and $a_m$ is its **leading coefficient**; $p$ is **monic** if $a_m = 1$. Non-zero constants have degree $0$; the zero polynomial is given degree $-\infty$.
:::

::: {#prp-degree-of-sum}
## Degree of a Sum

$\deg(p + q) \le \max(\deg p, \deg q)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-degree-of-sum
If $N = \max(\deg p, \deg q)$ and $k > N$, both $k$-th coefficients are $0$, so the $k$-th coefficient of $p + q$ is $0$. $\blacksquare$
:::

The degree can drop: if $p$ and $q$ have the same degree and opposite leading coefficients, the top terms cancel.

::: {#prp-degree-of-product}
## Degree of a Product

Let $p, q \ne 0$ with $\deg p = m$, $\deg q = n$ and leading coefficients $a_m$, $b_n$.

1. $\deg(pq) \le m + n$, and the coefficient of $x^{m+n}$ in $pq$ is $a_m b_n$.
2. If $a_m$ or $b_n$ is a unit, then $\deg(pq) = m + n$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-degree-of-product
1. The coefficient of $x^k$ in $pq$ is $r_k = \sum_{i+j=k} a_i b_j$. If $k > m + n$, every term has $i > m$ or $j > n$ and vanishes. For $k = m + n$ the only possibly non-zero term is $i = m$, $j = n$.
2. Suppose $a_m$ is a unit and $a_m b_n = 0$. Then $b_n = (a_m^{-1} a_m) b_n = a_m^{-1}(a_m b_n) = 0$, contradicting $b_n \ne 0$. If $b_n$ is a unit, multiply $a_m b_n = 0$ by $b_n^{-1}$ on the right. So $a_m b_n \ne 0$. $\blacksquare$
:::

::: {.callout-warning}
## Zero Divisors Make the Degree Drop
In $(\mathbb{Z}/6)[x]$: $(2x + 1)(3x) = 6x^2 + 3x = 3x$, of degree $1 < 2$. In $M_2(K)[x]$: with $E = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}$, $(Ex)(Ex) = E^2 x^2 = 0$. Over a [field](field.ipynb) this cannot happen, since every non-zero coefficient is a unit: there $\deg(pq) = \deg p + \deg q$ always.
:::

## 4\. Substituting a Value for $x$

::: {#def-evaluation-map}
## Evaluation

For $c \in R$, the **evaluation at $c$** is the map $E_c : R[x] \to R$,
$$E_c\Big(\sum_k a_k x^k\Big) = \sum_k a_k c^k = a_0 + a_1 c + \dots + a_m c^m,$$
with each coefficient written to the **left** of the power of $c$. We also write $p(c) = E_c(p)$.
:::

::: {#prp-evaluation-additive}
## Evaluation Respects Addition

$E_c(p + q) = E_c(p) + E_c(q)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-evaluation-additive
Padding with zero coefficients, $E_c(p + q) = \sum_k (a_k + b_k) c^k = \sum_k a_k c^k + \sum_k b_k c^k$, using right distributivity and the commutativity and associativity of addition. $\blacksquare$
:::

Multiplication is different. Take $p(x) = a_0 + a_1 x$ and the constant $q(x) = b_0$. Then $pq = a_0 b_0 + a_1 b_0\, x$, so
$$E_c(pq) = a_0 b_0 + a_1 b_0\, c, \qquad E_c(p)\,E_c(q) = (a_0 + a_1 c)\, b_0 = a_0 b_0 + a_1 c\, b_0.$$
These agree only if $a_1 b_0 c = a_1 c\, b_0$, which fails in general when $b_0$ and $c$ do not commute. Substituting first and multiplying afterwards is not the same as multiplying first. The following lemma isolates exactly what is needed.

::: {#lem-commutation-evaluation}
## Commutation Lemma

Let $p, q \in R[x]$ and $c \in R$. If $c$ commutes with every coefficient of $q$ (that is, $c\, b_j = b_j\, c$ for all $j$), then
$$E_c(pq) = E_c(p)\, E_c(q).$$
No condition on the coefficients of $p$ is needed.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-commutation-evaluation
Write $p = \sum_i a_i x^i$ and $q = \sum_j b_j x^j$. By the definition of the product,
$$E_c(pq) = \sum_k \Big(\sum_{i + j = k} a_i b_j\Big) c^k = \sum_{i, j} a_i\, b_j\, c^{i} c^{j}, \qquad E_c(p)\,E_c(q) = \sum_{i, j} a_i\, c^i\, b_j\, c^j.$$
From $c b_j = b_j c$ it follows by induction on $i$ that $c^i b_j = b_j c^i$. Hence $a_i c^i b_j c^j = a_i b_j c^i c^j$ for every pair $(i, j)$, and the two sums agree term by term. $\blacksquare$
:::

::: {.callout-note}
## Commutative Coefficients
If $R$ is commutative (for instance a field), every $c$ commutes with every coefficient, so $E_c(pq) = E_c(p)E_c(q)$ always holds. The lemma matters for $R = M_n(K)$: we will substitute a matrix $A$ for $x$ in a product of matrix polynomials whose right-hand factor is $x I_n - A$, whose coefficients $I_n$ and $-A$ do commute with $A$.
:::

The code checks the degree drop in $(\mathbb{Z}/6)[x]$ and the commutation lemma for matrix coefficients, with and without the commutation hypothesis.


```python

import numpy as np

def polymul_mod(p, q, m):
    """Multiply coefficient lists (lowest degree first) with coefficients in Z/m."""
    return [int(c) % m for c in np.convolve(p, q)]

print("(2x + 1)(3x) in Z/6[x] has coefficients", polymul_mod([1, 2], [0, 3], 6), "-> 3x, degree 1")

rng = np.random.default_rng(42)
n = 3
A = rng.integers(-3, 4, (n, n)).astype(float)
P1, P0 = rng.normal(size=(n, n)), rng.normal(size=(n, n))        # p(x) = P1 x + P0, arbitrary coefficients

def check(Q1, Q0):
    # p q = (P1 Q1) x^2 + (P1 Q0 + P0 Q1) x + P0 Q0 ; substitute A on the right of each coefficient
    lhs = P1 @ Q1 @ A @ A + (P1 @ Q0 + P0 @ Q1) @ A + P0 @ Q0     # E_A(pq)
    rhs = (P1 @ A + P0) @ (Q1 @ A + Q0)                             # E_A(p) E_A(q)
    return np.allclose(lhs, rhs)

print("q = I x - A (coefficients commute with A):  E_A(pq) == E_A(p)E_A(q) ?", check(np.eye(n), -A))
print("q = I x + B (B random, does not commute):  E_A(pq) == E_A(p)E_A(q) ?", check(np.eye(n), rng.normal(size=(n, n))))
```
