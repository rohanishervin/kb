# Infinite Series and Convergence Tests

**Prerequisites:** [The Limit of a Sequence](sequence-limit.ipynb) · [Algebraic Limit Theorem](composite-sequences-limit.ipynb) · [Geometric Sequences and the Geometric Series](geometric-sequence.ipynb)

## 1\. Definitions and First Facts

::: {#def-series}
## Series

For a real or complex sequence $(a_n)_{n \ge 1}$, the **series** $\sum a_n$ is the sequence of **partial sums** $s_n = a_1 + \dots + a_n$. If $s_n \to S$, the series **converges** and $S = \sum_{n=1}^\infty a_n$ is its **sum**; otherwise it diverges.
:::

Changing finitely many terms changes all partial sums from some index on by a fixed constant, so it does not affect convergence (only the sum). By the [limit laws](composite-sequences-limit.ipynb#thm-limit-laws), if $\sum a_n = A$ and $\sum b_n = B$ then $\sum (a_n + b_n) = A + B$ and $\sum c\,a_n = cA$. There is no such rule for products: $\sum_{n\ge0} 2^{-n} = 2$, but $\sum_{n \ge 0} 2^{-n}2^{-n} = \frac43 \ne 4$ (the correct product is the [Cauchy product](absolute-and-conditional-convergence.ipynb#thm-cauchy-product)).

::: {#thm-term-test}
## Terms of a Convergent Series Tend to $0$

If $\sum a_n$ converges then $a_n \to 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-term-test
$a_n = s_n - s_{n-1} \to S - S = 0$. $\blacksquare$
:::

The converse is false:

::: {#thm-harmonic-diverges}
## The Harmonic Series Diverges

$\sum_{n=1}^\infty \frac1n = +\infty$, although $\frac1n \to 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-harmonic-diverges
Group the terms in blocks $\frac{1}{2^{j-1}+1} + \dots + \frac{1}{2^j}$, $j \ge 1$: each block has $2^{j-1}$ terms, each $\ge 2^{-j}$, so it is $\ge \frac12$. Hence $s_{2^k} \ge 1 + \frac{k}{2} \to \infty$. $\blacksquare$
:::

## 2\. Series with Nonnegative Terms

::: {#thm-nonnegative-series}
## Bounded Partial Sums

If $a_n \ge 0$, then $\sum a_n$ converges if and only if its partial sums are bounded; otherwise $\sum a_n = +\infty$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-nonnegative-series
The partial sums increase; apply the [monotone convergence theorem](sequence-limit.ipynb#thm-monotone-convergence). A convergent sequence is bounded. $\blacksquare$
:::

::: {#thm-comparison-test}
## Comparison Test

Let $0 \le a_n \le b_n$ for all $n \ge N$. If $\sum b_n$ converges, so does $\sum a_n$; if $\sum a_n$ diverges, so does $\sum b_n$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-comparison-test
Dropping the first $N - 1$ terms, the partial sums of $\sum a_n$ are bounded by those of $\sum b_n$, which are bounded if it converges. Apply @thm-nonnegative-series. The second statement is the contrapositive. $\blacksquare$
:::

::: {#cor-ratio-comparison}
## Comparison Through Quotients

Let $a_n, b_n > 0$.

1. If $a_n / b_n \le M$ for $n \ge N$ and $\sum b_n$ converges, then $\sum a_n$ converges.
2. If $a_n / b_n \ge m > 0$ for $n \ge N$ and $\sum b_n$ diverges, then $\sum a_n$ diverges.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-ratio-comparison
(1): $a_n \le M b_n$ and $\sum M b_n$ converges. (2): $a_n \ge m b_n$ and $\sum m b_n$ diverges. Apply @thm-comparison-test. $\blacksquare$
:::

::: {#thm-limit-comparison}
## Limit Comparison Test

Let $a_n \ge 0$, $b_n > 0$ and $a_n / b_n \to L$.

1. If $L < \infty$ and $\sum b_n$ converges, then $\sum a_n$ converges.
2. If $0 < L < \infty$, then $\sum a_n$ and $\sum b_n$ converge or diverge together.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-limit-comparison
For $n$ large, $a_n/b_n \le L + 1$; apply @cor-ratio-comparison (1). If $L > 0$, also $a_n/b_n \ge L/2$ for $n$ large; apply (2), and (1) with the roles exchanged ($b_n / a_n \to 1/L$). $\blacksquare$
:::

::: {#thm-p-series}
## $p$-Series

For $p > 0$, $\sum_{n=1}^\infty \frac{1}{n^p}$ converges if and only if $p > 1$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-p-series
If $p \le 1$, $\frac{1}{n^p} \ge \frac1n$ and the harmonic series diverges. If $p > 1$, group as before: the block $2^j \le n < 2^{j+1}$ has $2^j$ terms, each $\le 2^{-jp}$, so its sum is $\le (2^{1-p})^j$. With $q = 2^{1-p} < 1$,
$$s_{2^{k+1} - 1} \le \sum_{j=0}^{k} q^j \le \frac{1}{1 - q},$$
and the partial sums are bounded. $\blacksquare$
:::

Here $p$ may be any real number once real powers are defined ([real powers](exponential-and-logarithm.ipynb#def-real-power)); the proof only uses that $x \mapsto x^{-p}$ is decreasing and $(2^j)^{-p} = (2^{-p})^j$. For $p = 2$ there is also a direct bound: $\frac{1}{n^2} < \frac{1}{n(n-1)} = \frac{1}{n-1} - \frac1n$ for $n \ge 2$, so $s_n < 1 + 1 = 2$.

::: {#exm-degree-rule}
## Rational Terms

If $P, Q$ are polynomials, $Q(n) \neq 0$ for $n \ge 1$, and $d = \deg Q - \deg P$, then $\sum \frac{P(n)}{Q(n)}$ converges if and only if $d \ge 2$: $\big|\frac{P(n)}{Q(n)}\big| \big/ \frac{1}{n^d}$ tends to the quotient of the leading coefficients' absolute values, so @thm-limit-comparison and @thm-p-series apply (for large $n$ the terms have constant sign). For example
$$\sum \frac{2n^3 - n + 3}{n^4 + 1} \ \text{diverges } (d = 1), \qquad \sum \frac{n + 1}{n^3 - 2n + 5}\ \text{converges } (d = 2).$$
:::

## 3\. Ratio and Root Tests

::: {#thm-ratio-test}
## Ratio Test

Let $a_n > 0$.

1. If $a_{n+1}/a_n \le \rho < 1$ for all $n \ge N$, then $\sum a_n$ converges.
2. If $a_{n+1}/a_n \ge 1$ for all $n \ge N$, then $\sum a_n$ diverges.

In particular, if $a_{n+1}/a_n \to L$: the series converges if $L < 1$ and diverges if $L > 1$. If $L = 1$ the test decides nothing.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-ratio-test
(1) By induction $a_n \le a_N \rho^{n - N}$ for $n \ge N$; compare with the [geometric series](geometric-sequence.ipynb#thm-geometric-series). (2) $a_n \ge a_N > 0$ for $n \ge N$, so $a_n \not\to 0$. If $L < 1$, take $\rho = \frac{1+L}{2}$; if $L > 1$, eventually $a_{n+1}/a_n > 1$. For $L = 1$: both $\sum \frac1n$ and $\sum \frac{1}{n^2}$ have ratio limit $1$. $\blacksquare$
:::

::: {#lem-nth-root-n}
## The $n$-th Root of $n$

$\sqrt[n]{n} \to 1$ as $n \to \infty$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-nth-root-n
Write $\sqrt[n]{n} = 1 + h_n$, $h_n \ge 0$. For $n \ge 2$, $n = (1 + h_n)^n \ge \binom{n}{2}h_n^2$, so $h_n^2 \le \frac{2}{n-1}$ and $h_n \to 0$. $\blacksquare$
:::

::: {#thm-root-test}
## Root Test

Let $a_n \ge 0$.

1. If $\sqrt[n]{a_n} \le \rho < 1$ for all $n \ge N$, then $\sum a_n$ converges.
2. If $\sqrt[n]{a_n} \ge 1$ for infinitely many $n$, then $\sum a_n$ diverges.

In particular, if $\sqrt[n]{a_n} \to L$: convergence for $L < 1$, divergence for $L > 1$, no information for $L = 1$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-root-test
(1) $a_n \le \rho^n$ for $n \ge N$; compare with $\sum \rho^n$. (2) $a_n \ge 1$ infinitely often, so $a_n \not\to 0$. The limit form follows as in the ratio test. For $L = 1$: $\sqrt[n]{1/n^p} = (\sqrt[n]{n})^{-p} \to 1$ by @lem-nth-root-n, for both $p = 1$ and $p = 2$. $\blacksquare$
:::

::: {#exm-ratio-root}
## Examples

1. $\sum \frac{x^n}{n!}$ converges for every $x \ge 0$: the ratio is $\frac{x}{n+1} \to 0$.
2. $\sum \frac{x^n}{n^n}$ converges for every $x \ge 0$: the $n$-th root is $\frac{x}{n} \to 0$ (or compare with the previous series, $n^n \ge n!$).
3. $\sum \frac{n^2}{2^n}$ converges: the ratio is $\frac{(n+1)^2}{2n^2} \to \frac12$.
:::
