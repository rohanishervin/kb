## Notation

The same symbols mean the same thing in every note.

| Symbol | Meaning |
|---|---|
| $X$ | the underlying set; $E, F, A, B$ are subsets of $X$ |
| $E^c = X \setminus E$, $E \setminus F$ | complement, difference |
| $2^X$ | the family of all subsets of $X$ |
| $\mathcal{M}$ | a σ-algebra on $X$; its members are the measurable sets |
| $\sigma(\mathcal{S})$ | the σ-algebra generated by a family $\mathcal{S}$ of sets |
| $\mathcal{B}$ | the Borel σ-algebra, generated by the open sets |
| $(X, \mathcal{M})$, $(X, \mathcal{M}, \mu)$ | measurable space, measure space |
| $\mu$ | a measure $\mathcal{M} \to [0, \infty]$ |
| $\delta_{x_0}$ | the Dirac measure at $x_0$ |
| $E_i \nearrow E$, $E_i \searrow E$ | increasing union $E = \bigcup_i E_i$, decreasing intersection $E = \bigcap_i E_i$ |
| a.e. | almost everywhere: outside a set of measure $0$ |
| $\lvert I\rvert$, $\lvert R\rvert$, $\lvert B\rvert$ | length of an interval, area of a rectangle, volume of a box |
| $m^*$ | Lebesgue outer measure, defined on all of $2^{\mathbb{R}^n}$ |
| $\mathcal{L}$ | the σ-algebra of Lebesgue measurable subsets of $\mathbb{R}^n$ |
| $m$ | Lebesgue measure, the restriction of $m^*$ to $\mathcal{L}$ |
| $G_\delta$, $F_\sigma$ | countable intersection of open sets, countable union of closed sets |
| $\overline{\mathbb{R}} = \mathbb{R} \cup \{\pm\infty\}$ | the extended real line |
| $\mathbf{1}_E$ | the indicator (characteristic) function of $E$ |
| $f^+ = \max(f, 0)$, $f^- = \max(-f, 0)$ | positive and negative parts, $f = f^+ - f^-$, $\lvert f\rvert = f^+ + f^-$ |
| $\varphi, \psi$ | simple functions $\sum_k c_k \mathbf{1}_{E_k}$ |
| $\int_E f\, d\mu = \int_X f \mathbf{1}_E\, d\mu$ | the integral of $f$ over $E$; $\int f$ means $\int_X f\, d\mu$ |
| $\mathcal{R}(f)$ | the Riemann integral of $f$ |
| $\lVert f\rVert_1$, $\lVert f\rVert_p$ | $\int \lvert f\rvert\, d\mu$ and $\big(\int \lvert f\rvert^p\, d\mu\big)^{1/p}$ |
| $L^1(\mu)$, $L^p(\mu)$ | integrable ($p$-th power integrable) functions, up to equality a.e. |
