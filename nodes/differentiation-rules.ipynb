# Rules of Differentiation

**Prerequisites:** [The Derivative](derivative.ipynb) · [Limits of Functions](limit.ipynb#thm-sin-limit)

::: {#thm-sum-product-quotient}
## Sum, Product and Quotient Rules

Let $f, g$ be differentiable at an interior point $x_0$ of their domain $S$. Then

1. $(f + g)'(x_0) = f'(x_0) + g'(x_0)$ and $(cf)'(x_0) = cf'(x_0)$;
2. **Leibniz rule:** $(fg)'(x_0) = f'(x_0)g(x_0) + f(x_0)g'(x_0)$;
3. if $g(x_0) \ne 0$, then $x_0$ is interior to $\{g \ne 0\}$ and
$$\Big(\frac{f}{g}\Big)'(x_0) = \frac{f'(x_0)g(x_0) - f(x_0)g'(x_0)}{g(x_0)^2} .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-sum-product-quotient
(1) The difference quotient of $f + g$ is the sum of those of $f$ and $g$.

(2) $\dfrac{f(x)g(x) - f(x_0)g(x_0)}{x - x_0} = \dfrac{f(x) - f(x_0)}{x - x_0}\,g(x) + f(x_0)\,\dfrac{g(x) - g(x_0)}{x - x_0}$, and $g(x) \to g(x_0)$ because $g$ is [continuous](derivative.ipynb#thm-differentiable-continuous) at $x_0$.

(3) By continuity $g \ne 0$ near $x_0$. Then
$$\frac{\frac{f}{g}(x) - \frac{f}{g}(x_0)}{x - x_0} = \Big[\frac{f(x) - f(x_0)}{x - x_0}g(x_0) - f(x_0)\frac{g(x) - g(x_0)}{x - x_0}\Big]\frac{1}{g(x)g(x_0)}$$
and the limit laws give the formula. $\blacksquare$
:::

::: {#cor-polynomial-derivative}
## Polynomials and Rational Functions

$p(x) = a_0 + a_1 x + \dots + a_m x^m$ is differentiable everywhere with $p'(x) = a_1 + 2a_2x + \dots + m a_m x^{m-1}$. A rational function $p/q$ is differentiable on $\{q \ne 0\}$. In particular $(x^{n})' = nx^{n-1}$ for every $n \in \mathbb{Z}$ ($x \ne 0$ if $n < 0$), since $(x^{-n})' = \frac{-nx^{n-1}}{x^{2n}} = -nx^{-n-1}$.
:::

::: {#thm-trig-derivatives}
## Derivatives of the Trigonometric Functions

$$\sin' = \cos, \quad \cos' = -\sin, \quad \tan' = \sec^2 = 1 + \tan^2, \quad \cot' = -\csc^2 = -(1 + \cot^2), \quad \sec' = \sec\tan, \quad \csc' = -\csc\cot,$$
each on the domain of the function.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-trig-derivatives
By the addition formula,
$$\frac{\sin(x + h) - \sin x}{h} = \sin x\,\frac{\cos h - 1}{h} + \cos x\,\frac{\sin h}{h} \to \sin x \cdot 0 + \cos x \cdot 1$$
by the [two trigonometric limits](limit.ipynb#thm-sin-limit). Similarly $\frac{\cos(x+h) - \cos x}{h} = \cos x\frac{\cos h - 1}{h} - \sin x\frac{\sin h}{h} \to -\sin x$. The rest follow from the quotient rule, e.g. $\tan' = \frac{\cos\cdot\cos - \sin\cdot(-\sin)}{\cos^2} = \frac{1}{\cos^2}$ and $\sec' = \big(\frac{1}{\cos}\big)' = \frac{\sin}{\cos^2}$. $\blacksquare$
:::

```python

import sympy as sp

x = sp.symbols("x")
table = {
    "sin": (sp.sin(x), sp.cos(x)),
    "cos": (sp.cos(x), -sp.sin(x)),
    "tan": (sp.tan(x), 1 + sp.tan(x)**2),
    "cot": (sp.cot(x), -(1 + sp.cot(x)**2)),
    "sec": (sp.sec(x), sp.sec(x) * sp.tan(x)),
    "csc": (sp.csc(x), -sp.csc(x) * sp.cot(x)),
}
for name, (f, claimed) in table.items():
    print(f"{name}' = {claimed!s:28}  check: {sp.simplify(sp.diff(f, x) - claimed) == 0}")
```
