# Monotone Functions and Continuous Inverses

**Prerequisites:** [The Intermediate Value Theorem](intermediate-value-theorem.ipynb) · [Existence of $n^{\text{th}}$ Roots](existence-of-nth-roots-in-R.ipynb)

$f : I \to \mathbb{R}$ on an interval $I$ is **increasing** if $x < y \Rightarrow f(x) \le f(y)$, **strictly increasing** if $x < y \Rightarrow f(x) < f(y)$; similarly (strictly) decreasing. **(Strictly) monotone** means one of these. A strictly monotone function is injective.

::: {#thm-injective-monotone}
## Continuous Injective Functions Are Strictly Monotone

If $f : I \to \mathbb{R}$ is continuous and injective on an interval $I$, then $f$ is strictly monotone.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-injective-monotone
Fix $a < b$ in $I$; say $f(a) < f(b)$ (otherwise replace $f$ by $-f$). Let $u < v$ be any points of $I$. For $t \in [0, 1]$ put
$$x(t) = (1-t)u + ta, \qquad y(t) = (1-t)v + tb, \qquad g(t) = f(y(t)) - f(x(t)).$$
Since $u < v$ and $a < b$, $x(t) < y(t)$, and both lie in $I$. By injectivity $g(t) \ne 0$ for all $t$. $g$ is continuous, so by the [intermediate value theorem](intermediate-value-theorem.ipynb#thm-ivt) it cannot change sign; $g(1) = f(b) - f(a) > 0$ gives $g(0) = f(v) - f(u) > 0$. So $f$ is strictly increasing. $\blacksquare$
:::

::: {#thm-continuous-inverse}
## The Inverse of a Continuous Monotone Function Is Continuous

Let $f : I \to \mathbb{R}$ be continuous and strictly monotone on an interval $I$. Then $J = f(I)$ is an interval and $f^{-1} : J \to I$ is continuous and strictly monotone (in the same direction).
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-continuous-inverse
$J$ is an interval by [this corollary](intermediate-value-theorem.ipynb#cor-image-interval). Let $f$ be strictly increasing (else use $-f$); then $f^{-1}$ is strictly increasing. Let $y_0 = f(x_0) \in J$ and $\varepsilon > 0$.

If $x_0$ is interior to $I$, pick $\varepsilon' \le \varepsilon$ with $x_0 \pm \varepsilon' \in I$, and let $\delta = \min\big(y_0 - f(x_0 - \varepsilon'),\ f(x_0 + \varepsilon') - y_0\big) > 0$. For $y \in J$ with $|y - y_0| < \delta$ we have $f(x_0 - \varepsilon') < y < f(x_0 + \varepsilon')$, hence, applying the increasing $f^{-1}$, $x_0 - \varepsilon' < f^{-1}(y) < x_0 + \varepsilon'$.

If $x_0$ is an endpoint of $I$, say the left one, use only $x_0 + \varepsilon'$: for $y \in J$, $y \ge y_0$, and $y < f(x_0 + \varepsilon')$ gives $x_0 \le f^{-1}(y) < x_0 + \varepsilon'$. $\blacksquare$
:::

::: {#exm-root-continuous}
## Roots Are Continuous

$x \mapsto x^n$ is continuous and strictly increasing on $[0, \infty)$ with image $[0, \infty)$ ([roots exist](existence-of-nth-roots-in-R.ipynb#thm-nth-root)), so $x \mapsto x^{1/n}$ is continuous on $[0, \infty)$; hence so is $x \mapsto x^{m/n} = (x^{1/n})^m$ on $(0, \infty)$.
:::

## Inverse Trigonometric Functions

Each trigonometric function is strictly monotone and continuous on a suitable interval; the theorem gives a continuous inverse.

::: {#def-inverse-trig}
## Inverse Trigonometric Functions

| function | restricted to | inverse | domain $\to$ range of the inverse |
|---|---|---|---|
| $\sin$ (increasing) | $[-\frac{\pi}{2}, \frac{\pi}{2}]$ | $\arcsin$ | $[-1, 1] \to [-\frac\pi2, \frac\pi2]$ |
| $\cos$ (decreasing) | $[0, \pi]$ | $\arccos$ | $[-1, 1] \to [0, \pi]$ |
| $\tan$ (increasing) | $(-\frac\pi2, \frac\pi2)$ | $\arctan$ | $\mathbb{R} \to (-\frac\pi2, \frac\pi2)$ |
| $\cot$ (decreasing) | $(0, \pi)$ | $\operatorname{arccot}$ | $\mathbb{R} \to (0, \pi)$ |
| $\sec$ | $[0, \pi] \setminus \{\frac\pi2\}$ | $\operatorname{arcsec}$ | $\vert x \vert \ge 1 \to [0, \pi] \setminus \{\frac\pi2\}$ |
| $\csc$ | $[-\frac\pi2, \frac\pi2] \setminus \{0\}$ | $\operatorname{arccsc}$ | $\vert x \vert \ge 1 \to [-\frac\pi2, \frac\pi2] \setminus \{0\}$ |
:::

$\tan$ maps $(-\frac\pi2, \frac\pi2)$ **onto** $\mathbb{R}$: given $y > 0$, for small $t > 0$ we have $\tan(\frac\pi2 - t) = \frac{\cos t}{\sin t} \ge \frac{\cos t}{t} > y$ (as $0 < \sin t \le t$ by [this lemma](function-continuity.ipynb#lem-sin-bounds) and $\cos t \to 1$), while $\tan 0 = 0$; the [IVT](intermediate-value-theorem.ipynb#thm-ivt) gives $\tan\theta = y$. Negative $y$ follow since $\tan$ is odd. On the two pieces of its restricted domain, $\sec$ and $\csc$ are strictly monotone, and their inverses are continuous on each of the two pieces $x \le -1$, $x \ge 1$.

::: {#prp-inverse-trig-identities}
## Complementary Angles

$\arcsin x + \arccos x = \frac{\pi}{2}$ for $x \in [-1, 1]$, and $\arctan x + \operatorname{arccot} x = \frac\pi2$ for $x \in \mathbb{R}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-inverse-trig-identities
$\theta = \frac\pi2 - \arcsin x$ lies in $[0, \pi]$ and $\cos\theta = \sin(\arcsin x) = x$, so $\theta = \arccos x$. Similarly $\frac\pi2 - \arctan x \in (0, \pi)$ has cotangent $\tan(\arctan x) = x$. $\blacksquare$
:::

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

fig, (ax, ax2) = plt.subplots(1, 2, figsize=(11, 4.6))
t = np.linspace(-np.pi / 2, np.pi / 2, 300)
ax.plot(t, np.sin(t), color=BLUE, lw=2, label=r"$\sin$ on $[-\pi/2, \pi/2]$")
ax.plot(np.sin(t), t, color=ORANGE, lw=2, label=r"$\arcsin$ on $[-1, 1]$")
ax.plot([-1.7, 1.7], [-1.7, 1.7], color=MUTED, ls="--", lw=1, label="$y = x$")
ax.set_aspect("equal"); ax.legend(loc="upper left", fontsize=9)
ax.set_title("the graph of the inverse is the mirror image")

x = np.linspace(-6, 6, 400)
ax2.plot(x, np.arctan(x), color=BLUE, lw=2, label=r"$\arctan$")
ax2.plot(x, np.pi / 2 - np.arctan(x), color=ORANGE, lw=2, label=r"$\mathrm{arccot} = \pi/2 - \arctan$")
for c in (-np.pi / 2, np.pi / 2, np.pi):
    ax2.axhline(c, color=MUTED, ls=":", lw=1)
ax2.legend(loc="center right", fontsize=9)
ax2.set_title(r"$\arctan: \mathbb{R} \to (-\pi/2, \pi/2)$")
plt.tight_layout(); plt.show()
```
