# Change of Basis and Similar Matrices

**Prerequisites:** [The Matrix of a Linear Map](matrix-of-a-linear-map.ipynb) · [Inverses](invertible-linear-maps.ipynb)

The [matrix of a linear map](matrix-of-a-linear-map.ipynb#def-matrix-linear-map) depends on the bases. Changing the basis changes the matrix but not the map. This note gives the exact rule, $[f]_{\mathcal{B}'} = P^{-1} [f]_{\mathcal{B}} P$, and shows how a clever basis can make a matrix much simpler. Matrices related in this way are called **similar**; they are different pictures of the same operator.

Throughout, $V$ is a [vector space](vector-spaces.ipynb) of dimension $n$ over a field $K$, with ordered bases $\mathcal{B} = (v_1, \dots, v_n)$ ("old") and $\mathcal{B}' = (v_1', \dots, v_n')$ ("new").

## 1\. The Index Chain Rule

For linear maps $f : V_1 \to V_2$ and $g : V_2 \to V_3$ with bases $\mathcal{B}_1, \mathcal{B}_2, \mathcal{B}_3$, the [composition theorem](matrix-of-a-linear-map.ipynb#thm-matrix-multiplication-composition) reads
$$[g \circ f]_{\mathcal{B}_1}^{\mathcal{B}_3} = [g]_{\mathcal{B}_2}^{\mathcal{B}_3}\, [f]_{\mathcal{B}_1}^{\mathcal{B}_2}.$$
The middle basis $\mathcal{B}_2$ appears once as an upper and once as a lower index and cancels, like a chain rule. Everything in this note follows by applying this rule to the identity map with different bases on its two sides.

## 2\. The Transition Matrix

::: {#def-transition-matrix}
## Transition Matrix

The **transition matrix** (or **change-of-basis matrix**) from $\mathcal{B}'$ to $\mathcal{B}$ is
$$P = [\mathrm{id}_V]_{\mathcal{B}'}^{\mathcal{B}}, \qquad\text{that is}\qquad v_j' = \sum_{i=1}^n p_{ij}\, v_i \quad (j = 1, \dots, n).$$
Its $j$-th column is the old coordinate vector $[v_j']_{\mathcal{B}}$ of the $j$-th new basis vector.
:::

::: {#prp-coordinate-change}
## How Coordinates Change

For every $v \in V$: $\ [v]_{\mathcal{B}} = P\, [v]_{\mathcal{B}'}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-coordinate-change
Apply the [action theorem](matrix-of-a-linear-map.ipynb#thm-matrix-action) to $\mathrm{id}_V$ with input basis $\mathcal{B}'$ and output basis $\mathcal{B}$: $[v]_{\mathcal{B}} = [\mathrm{id}_V(v)]_{\mathcal{B}} = [\mathrm{id}_V]_{\mathcal{B}'}^{\mathcal{B}} [v]_{\mathcal{B}'}$. $\blacksquare$
:::

::: {#thm-transition-invertibility}
## The Transition Matrix Is Invertible

$P$ is invertible and $P^{-1} = [\mathrm{id}_V]_{\mathcal{B}}^{\mathcal{B}'}$ is the transition matrix in the other direction. Hence $[v]_{\mathcal{B}'} = P^{-1} [v]_{\mathcal{B}}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-transition-invertibility
By the chain rule applied to $\mathrm{id}_V = \mathrm{id}_V \circ \mathrm{id}_V$ and the [matrix of the identity](matrix-of-a-linear-map.ipynb#prp-identity-matrix),
$$[\mathrm{id}_V]_{\mathcal{B}'}^{\mathcal{B}}\,[\mathrm{id}_V]_{\mathcal{B}}^{\mathcal{B}'} = [\mathrm{id}_V]_{\mathcal{B}}^{\mathcal{B}} = I_n, \qquad [\mathrm{id}_V]_{\mathcal{B}}^{\mathcal{B}'}\,[\mathrm{id}_V]_{\mathcal{B}'}^{\mathcal{B}} = [\mathrm{id}_V]_{\mathcal{B}'}^{\mathcal{B}'} = I_n. \qquad \blacksquare$$
:::

::: {.callout-warning}
## Which Way Does $P$ Go?
$P$ has the **new** basis vectors written in the **old** basis, yet it converts **new** coordinates into **old** coordinates. To get new coordinates from old ones, one needs $P^{-1}$, i.e. one has to solve a linear system. In @exm-coordinates-plane below, $v = (3, 1)$ has new coordinates $P^{-1}(3, 1)^{\mathsf T} = (2, 1)^{\mathsf T}$.
:::

::: {#exm-coordinates-plane}
## Changing Coordinates in the Plane

In $\mathbb{R}^2$ take the standard basis $\mathcal{E} = (e_1, e_2)$ and $\mathcal{B}' = (b_1, b_2)$ with $b_1 = (1, 1)$, $b_2 = (1, -1)$. Then
$$P = [\mathrm{id}]_{\mathcal{B}'}^{\mathcal{E}} = \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}, \qquad P^{-1} = \frac12 \begin{pmatrix} 1 & 1 \\ 1 & -1 \end{pmatrix}.$$
For $v = (3, 1)$: $[v]_{\mathcal{B}'} = P^{-1} (3, 1)^{\mathsf T} = (2, 1)^{\mathsf T}$, and indeed $2 b_1 + b_2 = (3, 1)$.
:::

## 3\. How the Matrix of a Map Changes

::: {#thm-operator-change-of-basis}
## Transformation Law for Operators

Let $f : V \to V$ be linear and $P = [\mathrm{id}_V]_{\mathcal{B}'}^{\mathcal{B}}$. Then
$$[f]_{\mathcal{B}'} = P^{-1}\, [f]_{\mathcal{B}}\, P.$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-operator-change-of-basis
Write $f = \mathrm{id}_V \circ f \circ \mathrm{id}_V$ and use the bases $(V, \mathcal{B}') \to (V, \mathcal{B}) \to (V, \mathcal{B}) \to (V, \mathcal{B}')$:
$$[f]_{\mathcal{B}'}^{\mathcal{B}'} = [\mathrm{id}_V]_{\mathcal{B}}^{\mathcal{B}'}\,[f]_{\mathcal{B}}^{\mathcal{B}}\,[\mathrm{id}_V]_{\mathcal{B}'}^{\mathcal{B}} = P^{-1} [f]_{\mathcal{B}} P. \qquad \blacksquare$$
:::

::: {.callout-note}
## Reading $P^{-1} A P$ from Right to Left
To apply $f$ to a vector given in new coordinates: first convert to old coordinates ($P$), then apply the old matrix ($A = [f]_{\mathcal{B}}$), then convert back ($P^{-1}$).
:::

::: {#prp-general-change-of-basis}
## Maps Between Two Spaces

If $f : V \to W$, $\mathcal{C}, \mathcal{C}'$ are bases of $W$, $P = [\mathrm{id}_V]_{\mathcal{B}'}^{\mathcal{B}}$ and $Q = [\mathrm{id}_W]_{\mathcal{C}'}^{\mathcal{C}}$, then
$$[f]_{\mathcal{B}'}^{\mathcal{C}'} = Q^{-1}\, [f]_{\mathcal{B}}^{\mathcal{C}}\, P.$$
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-general-change-of-basis
As before, $f = \mathrm{id}_W \circ f \circ \mathrm{id}_V$ gives $[f]_{\mathcal{B}'}^{\mathcal{C}'} = [\mathrm{id}_W]_{\mathcal{C}}^{\mathcal{C}'} [f]_{\mathcal{B}}^{\mathcal{C}} [\mathrm{id}_V]_{\mathcal{B}'}^{\mathcal{B}}$, and $[\mathrm{id}_W]_{\mathcal{C}}^{\mathcal{C}'} = Q^{-1}$ by @thm-transition-invertibility. $\blacksquare$
:::

::: {#exm-change-of-basis-diagonal}
## A Better Basis Makes the Matrix Diagonal

Let $f : \mathbb{R}^2 \to \mathbb{R}^2$ have standard matrix $A = \begin{pmatrix} 2 & 1 \\ 1 & 2 \end{pmatrix}$, and use the basis $b_1 = (1, 1)$, $b_2 = (1, -1)$ of @exm-coordinates-plane. Since $f(b_1) = (3, 3) = 3 b_1$ and $f(b_2) = (1, -1) = b_2$,
$$[f]_{\mathcal{B}'} = P^{-1} A P = \begin{pmatrix} 3 & 0 \\ 0 & 1 \end{pmatrix}.$$
In the new basis $f$ just stretches the $b_1$-direction by $3$ and leaves the $b_2$-direction alone; in the standard basis this simple behaviour is hidden. Finding such bases is the subject of [eigenvalues](eigenvalues-and-eigenvectors.ipynb) and [diagonalization](diagonalization.ipynb).
:::

The animation applies this $f$. The standard basis vectors $e_1, e_2$ (blue, orange) are moved to the columns of $A$ and the grid is sheared and stretched. The yellow and green vectors (multiples of $b_1$ and $b_2$) stay on their dashed lines: the first triples in length and the second does not move.


```python

from la_manim import *

class SameMapTwoBases(GridScene):
    def construct(self):
        for d in ([1, 1, 0], [1, -1, 0]):
            self.add(DashedLine(-6 * np.array(d), 6 * np.array(d), color=C_INK, stroke_width=1.5, dash_length=0.12))
        self.vec([1, 0], C_BLUE, r"e_1", DOWN)
        self.vec([0, 1], C_ORANGE, r"e_2", LEFT)
        self.vec([0.8, 0.8], C_YELLOW, r"b_1", UL)
        self.vec([1, -1], C_AQUA, r"b_2", DR)
        cap = VGroup(MathTex(r"[f]_{\mathcal{E}} = \begin{pmatrix} 2 & 1 \\ 1 & 2 \end{pmatrix}", color=C_INK),
                     MathTex(r"[f]_{\mathcal{B}'} = \begin{pmatrix} 3 & 0 \\ 0 & 1 \end{pmatrix}", color=C_INK)
                     ).arrange(DOWN, aligned_edge=LEFT).scale(0.7).to_corner(UL, buff=0.3)
        box = BackgroundRectangle(cap, fill_opacity=0.85, buff=0.15, color=config.background_color)
        self.play(FadeIn(box), FadeIn(cap[0]))
        self.wait(1)
        self.transform([[2, 1], [1, 2]], run_time=3.5)
        self.play(FadeIn(cap[1]))
        msg = text_caption(r"$e_1, e_2$ are moved off their lines; $b_1 \mapsto 3b_1$ and $b_2 \mapsto b_2$")
        self.play(FadeIn(msg)); self.wait(2.5)

show(SameMapTwoBases)
```


## 4\. Similar Matrices

::: {#def-matrix-similarity}
## Similar Matrices

$A, B \in M_n(K)$ are **similar**, written $A \sim B$, if $B = P^{-1} A P$ for some invertible $P \in GL_n(K)$. (Here $GL_n(K)$ denotes the set of invertible $n \times n$ matrices.)
:::

::: {#prp-similarity-equivalence}
## Similarity Is an Equivalence Relation, and Similar Means "Same Operator"

1. $\sim$ is reflexive, symmetric and transitive.
2. $A \sim B$ if and only if there are an operator $f$ on an $n$-dimensional space and two bases $\mathcal{B}, \mathcal{B}'$ with $[f]_{\mathcal{B}} = A$ and $[f]_{\mathcal{B}'} = B$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-similarity-equivalence
1. $A = I^{-1} A I$. If $B = P^{-1} A P$ then $A = (P^{-1})^{-1} B P^{-1}$. If $B = P^{-1} A P$ and $C = Q^{-1} B Q$ then $C = (PQ)^{-1} A (PQ)$.
2. ($\Leftarrow$) is @thm-operator-change-of-basis. ($\Rightarrow$) Let $B = P^{-1} A P$. Take $V = K^n$, $f(x) = A x$, $\mathcal{B}$ the standard basis (so $[f]_{\mathcal{B}} = A$), and let $\mathcal{B}'$ consist of the columns of $P$. These columns form a basis because $P$ is [invertible](invertible-linear-maps.ipynb#cor-equal-dim-invertibility), and the transition matrix from $\mathcal{B}'$ to $\mathcal{B}$ is $P$ itself. By @thm-operator-change-of-basis, $[f]_{\mathcal{B}'} = P^{-1} A P = B$. $\blacksquare$
:::

::: {.callout-note}
## Invariants
Because similar matrices describe the same operator, any quantity defined from a matrix that does not change under $A \mapsto P^{-1} A P$ is really a property of the operator. The [determinant](determinant.ipynb#def-det-operator), the [trace](trace.ipynb#def-trace-operator) and the [characteristic polynomial](eigenvalues-and-eigenvectors.ipynb#thm-characteristic-equation) are the main examples.
:::

The code computes $[f]_{\mathcal{B}'}$ for a random operator on $\mathbb{R}^3$ in two ways: column by column from the definition (coordinates of $f(v_j')$ in the new basis), and by the transformation law.


```python

import numpy as np

rng = np.random.default_rng(7)
A = rng.integers(-3, 4, size=(3, 3)).astype(float)       # [f] in the standard basis
P = np.array([[1.0, 1.0, 0.0],                           # columns: the new basis vectors
              [0.0, 1.0, 1.0],
              [1.0, 0.0, 1.0]])

# definition: column j = coordinates of f(v'_j) in the new basis, i.e. the solution c of P c = A v'_j
by_definition = np.column_stack([np.linalg.solve(P, A @ P[:, j]) for j in range(3)])
by_law = np.linalg.inv(P) @ A @ P
print("[f]_B =\n", A)
print("[f]_B' from the definition =\n", np.round(by_definition, 4))
print("P^{-1} [f]_B P =\n", np.round(by_law, 4))
print("equal:", np.allclose(by_definition, by_law))
```
