# Vector Moving Average Models (VMA)

## 1\. The VMA(q) Model

We can generalize the univariate [Moving Average (MA)](moving-average-models.ipynb) model to multivariate time series. Instead of scalar innovations, we use vector innovations (shocks), and instead of scalar parameters $\theta$, we use parameter matrices $\Theta$.

::: {#def-vma-q}
## Vector Moving Average Model: VMA(q)
Let $\vec{r}_t$ be a $k$-dimensional vector of asset returns. A VMA(q) model represents $\vec{r}_t$ as a moving average of current and past vector shocks:
$$
\vec{r}_t = \vec{\Theta}_0 + \vec{a}_t - \Theta_1 \vec{a}_{t-1} - \dots - \Theta_q \vec{a}_{t-q}
$$
where:
- $\vec{\Theta}_0$ is a $k \times 1$ constant vector (mean).
- $\vec{a}_t = (a_{1t}, a_{2t}, \dots, a_{kt})^T$ is a sequence of iid random shock vectors (multivariate white noise).
- $E[\vec{a}_t] = \vec{0}$.
- $\Theta_i$ are $k \times k$ parameter matrices.
:::

## 2\. Concurrent Correlation

In a VMA model, the shock vector $\vec{a}_t$ describes the unpredictable innovations at time $t$. The individual components of $\vec{a}_t$ do not have to be independent of each other. 

::: {#def-concurrent-correlation}
## Concurrent Correlation Matrix
The covariance matrix of the shock vector $\vec{a}_t$ is denoted by $\Sigma$:
$$
\text{Cov}(\vec{a}_t) = \Sigma
$$
If $\Sigma$ is not a diagonal matrix (i.e., its off-diagonal elements are non-zero), it means the components of $\vec{a}_t$ are concurrently correlated. A shock to one asset at time $t$ is correlated with a shock to another asset at the same time $t$.
:::

::: {.callout-note}
## Intuition: Structural Form vs Reduced Form
By making a small variable transformation, it is possible to rewrite the VMA model in a "structural form" such that the new noise vectors have no concurrent correlation (i.e., the transformed $\Sigma$ is diagonal). However, in its standard "reduced form", concurrent correlation is expected and common in finance because macroeconomic news impacts multiple assets simultaneously.
:::

## 3\. Properties of VMA Models

Similar to univariate [MA(q)](moving-average-models.ipynb) models, VMA(q) models are **always stationary**.

The cross-covariance matrix $\Gamma(\ell)$ for a VMA(q) model will cut off after lag $q$. That is:
$$
\Gamma(\ell) = \vec{0} \quad \text{for } \ell > q
$$
where $\vec{0}$ is a $k \times k$ matrix of zeros.

## 4\. Model Fitting

Fitting a VMA(q) model involves two steps:

1. **Identification of hyperparameter q:** We analyze the sample [cross-correlation matrices](multivariate-time-series.ipynb). We look for the lag $q$ after which the cross-correlation matrices are effectively zero.
2. **Estimation of parameters:** We estimate the parameter matrices $\Theta_1, \dots, \Theta_q$ and the covariance matrix $\Sigma$ using **Maximum Likelihood Estimation (MLE)**. Because the VMA model depends on unobserved past errors, the likelihood function is highly non-linear and requires numerical optimization methods.
