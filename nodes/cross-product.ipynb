# The Cross Product

**Prerequisites:** [Orthogonality, Projections and the Gram–Schmidt Process](orthogonality-and-gram-schmidt.ipynb) · [The Determinant](determinant.ipynb) · [Determinants, Volume, and Orientation](determinant-volume-and-orientation.ipynb)

In $\mathbb{R}^3$ write $\mathbf i = e_1$, $\mathbf j = e_2$, $\mathbf k = e_3$, and $[u\ v\ w]$ for the matrix with columns $u, v, w$.

::: {#def-cross-product}
## Cross Product

For $u, v \in \mathbb{R}^3$,
$$u\times v = (u_2v_3 - u_3v_2,\ u_3v_1 - u_1v_3,\ u_1v_2 - u_2v_1) = \det\begin{bmatrix}\mathbf i & \mathbf j & \mathbf k\\ u_1 & u_2 & u_3\\ v_1 & v_2 & v_3\end{bmatrix},$$
the determinant being expanded formally along the first row.
:::

::: {#prp-cross-algebra}
## Algebraic Rules

1. $u\times v$ is bilinear, $v\times u = -u\times v$, $u\times u = 0$; $\mathbf i\times\mathbf j = \mathbf k$, $\mathbf j\times\mathbf k = \mathbf i$, $\mathbf k\times\mathbf i = \mathbf j$.
2. **Triple product:** $w\cdot(u\times v) = \det[u\ v\ w]$; hence $u\cdot(v\times w) = v\cdot(w\times u) = w\cdot(u\times v)$, and $u\times v$ is orthogonal to $u$ and to $v$.
3. $a\times(b\times c) = (a\cdot c)\,b - (a\cdot b)\,c$.
4. $(a\times b)\cdot(c\times d) = (a\cdot c)(b\cdot d) - (a\cdot d)(b\cdot c)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-cross-algebra
(1) Each component is a bilinear expression that changes sign when $u, v$ are swapped; the basis products are read off the formula.
(2) $w\cdot(u\times v) = w_1(u_2v_3 - u_3v_2) - w_2(u_1v_3 - u_3v_1) + w_3(u_1v_2 - u_2v_1)$ is the [cofactor expansion](laplace-expansion-adjugate-and-cramers-rule.ipynb) of $\det[w\ u\ v] = \det[u\ v\ w]$ (two column swaps). Cyclic permutations of the columns do not change the determinant, and $\det[u\ v\ u] = 0$.
(3), (4) Both sides are linear in each of the vectors, so it suffices to check them on basis vectors. In (3) with $b = e_j$, $c = e_k$: if $j = k$ both sides vanish; if $j \ne k$, $e_j\times e_k = \pm e_m$ ($m$ the third index), and $a\times e_m$ for $a = e_j, e_k, e_m$ is checked directly to agree with $(a\cdot e_k)e_j - (a\cdot e_j)e_k$. In (4) with $a = e_i$, $b = e_j$, $c = e_k$, $d = e_l$: the left side is $\pm1$ only if $\{i, j\} = \{k, l\}$ with $i \ne j$, namely $+1$ if $(i, j) = (k, l)$ and $-1$ if $(i, j) = (l, k)$, and $0$ otherwise; the right side $\delta_{ik}\delta_{jl} - \delta_{il}\delta_{jk}$ has the same values. $\blacksquare$
:::

::: {#thm-cross-geometry}
## Geometry of the Cross Product

1. **Lagrange's identity:** $|u\times v|^2 + (u\cdot v)^2 = |u|^2|v|^2$. Hence for $u, v \ne 0$, $|u\times v| = |u|\,|v|\sin\angle(u, v)$, the area of the parallelogram $P(u, v) = \{su + tv : 0 \le s, t \le 1\}$, and $u\times v = 0$ iff $u, v$ are linearly dependent.
2. $|w\cdot(u\times v)| = |\det[u\ v\ w]|$ is the volume of the parallelepiped spanned by $u, v, w$.
3. If $u, v$ are independent, $(u, v, u\times v)$ is a positively oriented basis: $\det[u\ v\ u\times v] = |u\times v|^2 > 0$. Thus $u\times v$ is the unique vector orthogonal to $u$ and $v$, of length equal to the area of $P(u, v)$, that makes $(u, v, u\times v)$ positively oriented.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-cross-geometry
(1) Put $a = c = u$, $b = d = v$ in @prp-cross-algebra (4). Then $|u\times v|^2 = |u|^2|v|^2(1 - \cos^2\angle(u, v))$, and $\sin \ge 0$ on $[0, \pi]$. The parallelogram has base $|u|$ and height $|v|\sin\angle(u, v)$. Equality in Cauchy–Schwarz characterizes dependence.
(2) This is the [volume formula](determinant-volume-and-orientation.ipynb#def-parallelotope) together with @prp-cross-algebra (2).
(3) $\det[u\ v\ u\times v] = (u\times v)\cdot(u\times v)$ by (2). Orthogonality to $u, v$ leaves the line $\mathbb{R}(u\times v)$; the length fixes the vector up to sign, and the orientation fixes the sign. $\blacksquare$
:::

In the plane the same computation gives the signed area: for $u, v \in \mathbb{R}^2$, $\det[u\ v] = u_1v_2 - u_2v_1 = |u|\,|v|\sin\theta$, where $\theta$ is the angle from $u$ to $v$ measured counterclockwise; it is the third component of $(u_1, u_2, 0)\times(v_1, v_2, 0)$.

::: {#exm-plane-three-points}
## A Plane Through Three Points

For $P = (1, 0, 0)$, $Q = (0, 2, 0)$, $R = (0, 0, 3)$: $(Q - P)\times(R - P) = (-1, 2, 0)\times(-1, 0, 3) = (6, 3, 2)$ is normal to the plane, which is therefore $6x + 3y + 2z = 6$; the triangle $PQR$ has area $\frac12|(6, 3, 2)| = \frac72$, and the tetrahedron $OPQR$ has volume $\frac16|\det[P\ Q\ R]| = 1$.
:::

```python

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

a, b, c, d = (sp.Matrix(sp.symbols(f"{s}1:4")) for s in "abcd")
print("BAC-CAB:", sp.simplify(a.cross(b.cross(c)) - (a.dot(c) * b - a.dot(b) * c)) == sp.zeros(3, 1))
print("(a x b).(c x d) identity:", sp.expand(a.cross(b).dot(c.cross(d)) - (a.dot(c) * b.dot(d) - a.dot(d) * b.dot(c))) == 0)
print("triple product = det:", sp.expand(c.dot(a.cross(b)) - sp.Matrix.hstack(a, b, c).det()) == 0)
P, Q, R = sp.Matrix([1, 0, 0]), sp.Matrix([0, 2, 0]), sp.Matrix([0, 0, 3])
n = (Q - P).cross(R - P)
print("normal:", list(n), "| triangle area:", sp.sqrt(n.dot(n)) / 2, "| tetrahedron volume:", sp.Matrix.hstack(P, Q, R).det() / 6)

u, v = np.array([2.0, 0.3, 0.0]), np.array([0.6, 1.8, 0.4])
w = np.cross(u, v)
fig = plt.figure(figsize=(6.5, 5.2))
ax = fig.add_subplot(projection="3d")
ax.add_collection3d(Poly3DCollection([[np.zeros(3), u, u + v, v]], color=BLUE, alpha=0.25))
for vec, col, lab in [(u, INK, "$u$"), (v, INK, "$v$"), (w / np.linalg.norm(w) * 2, ORANGE, r"$u\times v$")]:
    ax.quiver(0, 0, 0, *vec, color=col, lw=2, arrow_length_ratio=0.08)
    ax.text(*(vec * 1.08), lab, color=col)
ax.set_xlim(0, 2.6); ax.set_ylim(-0.8, 2.2); ax.set_zlim(0, 2.2); ax.view_init(elev=20, azim=-55)
ax.set_title(r"$|u\times v|$ = area of the parallelogram, $u\times v \perp u, v$")
plt.tight_layout(); plt.show()
print("|u x v| =", np.linalg.norm(w), "| |u||v|sin =", np.linalg.norm(u) * np.linalg.norm(v) * np.sin(np.arccos(u @ v / np.linalg.norm(u) / np.linalg.norm(v))))
```
