# Vertex Cover and Independent Set

**Prerequisites:** [Integer Programs and LP Relaxation](integer-programming.ipynb) · [Maximum-Weight Bipartite Matching](bipartite-matching-lp.ipynb)

Throughout, $G = (V, E)$ is a finite graph without loops or multiple edges, $n = |V|$, and $uv$ denotes the edge $\{u, v\}$.

## 1\. The Two Problems

::: {#def-vertex-cover-independent-set}
## Vertex Cover and Independent Set

A set $C \subseteq V$ is a **vertex cover** if every edge has at least one endpoint in $C$; the smallest size of a vertex cover is $\tau(G)$. A set $S \subseteq V$ is **independent** if no edge has both endpoints in $S$; the largest size of an independent set is $\alpha(G)$. Computing $\tau(G)$ and computing $\alpha(G)$ are NP-hard problems.
:::

::: {#prp-cover-complement}
## Complements of Covers Are Independent

$S \subseteq V$ is independent if and only if $V \setminus S$ is a vertex cover. Hence $\alpha(G) = n - \tau(G)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-cover-complement
$V \setminus S$ fails to be a vertex cover iff some edge has no endpoint in $V \setminus S$, i.e. both endpoints in $S$, iff $S$ is not independent. The map $S \mapsto V \setminus S$ is a bijection between independent sets and vertex covers with $|V \setminus S| = n - |S|$; the largest independent sets correspond to the smallest covers. $\blacksquare$
:::

::: {#def-approximation-algorithm}
## Approximation Algorithm

For a minimization problem with optimum $\mathrm{OPT}$, an algorithm is a **$\rho$-approximation** ($\rho \ge 1$) if it always returns a feasible solution of cost $\le \rho \cdot \mathrm{OPT}$. For a maximization problem, a **$\rho$-approximation** ($0 < \rho \le 1$) always returns a feasible solution of value $\ge \rho \cdot \mathrm{OPT}$. Usually the algorithm is required to run in polynomial time.
:::

## 2\. Vertex Cover: Relaxation and Rounding

::: {#prp-vertex-cover-ip}
## Vertex Cover as an Integer Program

$\tau(G)$ is the optimal value of
$$
\text{minimize } \sum_{v \in V} x_v \quad \text{subject to } x_u + x_v \ge 1\ \ (uv \in E), \quad x \in \{0, 1\}^V .
$$
Let $\tau_{\mathrm{LP}}(G)$ be the optimal value of its relaxation (with $0 \le x_v \le 1$). Then $\tau_{\mathrm{LP}}(G) \le \tau(G)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-vertex-cover-ip
A vector $x \in \{0,1\}^V$ is the indicator of $C = \{v : x_v = 1\}$, and $\sum_v x_v = |C|$. For an edge $uv$, $x_u + x_v \ge 1$ says that $u \in C$ or $v \in C$. So the feasible $x$ are exactly the indicators of vertex covers. The relaxation has an optimum because its feasible set is nonempty ($x \equiv 1$), closed and bounded; the inequality is [the relaxation bound](integer-programming.ipynb#prp-relaxation-bound). $\blacksquare$
:::

::: {#exm-complete-graph-cover}
## The Complete Graph

For the complete graph $K_n$ ($n \ge 2$): $\tau(K_n) = n - 1$ and $\tau_{\mathrm{LP}}(K_n) = \frac n2$.
:::

::: {.callout-tip collapse="true"}
## Proof of @exm-complete-graph-cover
All vertices but one form a cover. A set missing two vertices $u, v$ does not cover the edge $uv$. So $\tau = n - 1$.

The vector $x \equiv \frac12$ is feasible with value $\frac n2$. Conversely, add the constraints $x_u + x_v \ge 1$ over all $\binom n2$ edges. Each vertex lies on $n - 1$ edges, so $(n - 1)\sum_v x_v \ge \binom n2 = \frac{n(n-1)}2$, i.e. $\sum_v x_v \ge \frac n2$ for every feasible $x$. $\blacksquare$
:::

So $\tau / \tau_{\mathrm{LP}} = 2(n-1)/n$, which tends to $2$: the relaxation can be off by a factor close to $2$. It is never worse than that.

::: {#thm-vertex-cover-rounding}
## Rounding at $\frac12$ Is a 2-Approximation

Let $x^*$ be an optimal solution of the relaxation and $C = \{v \in V : x^*_v \ge \frac12\}$. Then $C$ is a vertex cover and
$$
|C| \le 2\,\tau_{\mathrm{LP}}(G) \le 2\,\tau(G) .
$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-vertex-cover-rounding
For an edge $uv$, $x^*_u + x^*_v \ge 1$, so $x^*_u$ and $x^*_v$ cannot both be $< \frac12$; hence $u \in C$ or $v \in C$. For $v \in C$ we have $1 \le 2x^*_v$, and $2x^*_v \ge 0$ for all $v$, so
$$
|C| = \sum_{v \in C} 1 \le \sum_{v \in C} 2x^*_v \le \sum_{v \in V} 2x^*_v = 2\,\tau_{\mathrm{LP}}(G) \le 2\,\tau(G). \qquad \blacksquare
$$
:::

Solving an LP takes polynomial time, so this is a polynomial-time 2-approximation algorithm for an NP-hard problem.

## 3\. Bipartite Graphs

::: {#thm-bipartite-cover-integral}
## For Bipartite Graphs the Relaxation Is Exact

If $G$ is bipartite, the relaxation of @prp-vertex-cover-ip has an integral optimal solution, so $\tau_{\mathrm{LP}}(G) = \tau(G)$. An integral optimal solution is obtained from any optimal solution by the following step, repeated at most $n$ times: shift all fractional values on one side up by $\varepsilon$ and all fractional values on the other side down by $\varepsilon$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-bipartite-cover-integral
Let $V = U \cup W$ be the bipartition and $x$ an optimal solution with some fractional coordinate. Let $F_U = \{u \in U : 0 < x_u < 1\}$, $F_W = \{w \in W : 0 < x_w < 1\}$, and for $\varepsilon \in \mathbb{R}$ define $x^\varepsilon$ by $x^\varepsilon_u = x_u + \varepsilon$ on $F_U$, $x^\varepsilon_w = x_w - \varepsilon$ on $F_W$, unchanged elsewhere. Let $I$ be the set of $\varepsilon$ for which all shifted coordinates stay in $[0, 1]$; it is a closed interval containing $0$ in its interior.

*Claim: $x^\varepsilon$ is feasible for every $\varepsilon \in I$.* Take an edge $uw$, $u \in U$, $w \in W$.

- If $u \in F_U$ and $w \in F_W$: $x^\varepsilon_u + x^\varepsilon_w = x_u + x_w \ge 1$.
- If $u \in F_U$ and $w \notin F_W$: $x_w \in \{0, 1\}$. If $x_w = 0$, then $x_u \ge 1$, impossible for $u \in F_U$. So $x_w = 1$ and $x^\varepsilon_u + x^\varepsilon_w \ge 0 + 1$.
- If $u \notin F_U$ and $w \in F_W$: symmetrically $x_u = 1$.
- If neither is fractional, nothing changed.

*Optimality forces $|F_U| = |F_W|$.* The objective of $x^\varepsilon$ is $\sum_v x_v + \varepsilon(|F_U| - |F_W|)$. If $|F_U| \ne |F_W|$, a small $\varepsilon \in I$ of the appropriate sign would give a smaller value, contradicting optimality. So all $x^\varepsilon$, $\varepsilon \in I$, are optimal.

*Fewer fractional coordinates.* At an endpoint $\varepsilon_1$ of $I$ some shifted coordinate equals $0$ or $1$, while the coordinates in $\{0,1\}$ did not move. So $x^{\varepsilon_1}$ is optimal with fewer fractional coordinates. Repeating at most $n$ times gives an integral optimal solution, the indicator of a vertex cover of size $\tau_{\mathrm{LP}}(G)$; hence $\tau(G) \le \tau_{\mathrm{LP}}(G)$, and equality holds by @prp-vertex-cover-ip. $\blacksquare$
:::

::: {#exm-odd-cycle-cover}
## An Odd Cycle

For the cycle $C_{2k+1}$: $x \equiv \frac12$ is feasible, and adding the $2k+1$ edge constraints gives $2\sum_v x_v \ge 2k + 1$; so $\tau_{\mathrm{LP}} = k + \frac12$. A vertex cover must contain at least one endpoint of each of the $2k+1$ edges and each vertex covers two edges, so $2\tau \ge 2k + 1$ and $\tau \ge k + 1$; with vertices $0, 1, \dots, 2k$ in cyclic order, the $k + 1$ vertices $0, 2, \dots, 2k$ form a cover, so $\tau = k + 1$. The relaxation is not exact, as for $K_n$; both graphs contain odd cycles and are not bipartite.
:::

## 4\. Independent Set

::: {#prp-independent-set-ip}
## Independent Set as an Integer Program

$\alpha(G)$ is the optimal value of
$$
\text{maximize } \sum_{v} x_v \quad \text{subject to } x_u + x_v \le 1\ \ (uv \in E), \quad x \in \{0,1\}^V .
$$
Let $\alpha_{\mathrm{LP}}(G)$ be the value of its relaxation. Then $\alpha_{\mathrm{LP}}(G) \ge \alpha(G)$, and $x \mapsto \mathbf 1 - x$ maps the feasible set of the vertex cover relaxation bijectively onto that of the independent set relaxation, so $\alpha_{\mathrm{LP}}(G) = n - \tau_{\mathrm{LP}}(G)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-independent-set-ip
For $x \in \{0,1\}^V$, $x_u + x_v \le 1$ says that not both $u, v$ are chosen; so the feasible $x$ are the indicators of independent sets. With $y = \mathbf 1 - x$: $0 \le x \le 1 \iff 0 \le y \le 1$, and $x_u + x_v \ge 1 \iff y_u + y_v \le 1$; also $\sum y_v = n - \sum x_v$. So minimizing $\sum x_v$ over one set is maximizing $\sum y_v$ over the other. $\blacksquare$
:::

::: {#exm-complete-graph-independent}
## The Relaxation Can Be Useless for Independent Sets

In $K_n$, $\alpha(K_n) = 1$ but $\alpha_{\mathrm{LP}}(K_n) = n - \frac n2 = \frac n2$. The LP bound is off by the factor $\frac n2$, which grows with $n$.
:::

::: {.callout-warning}
## Equivalent Problems, Different Approximability
Although $\alpha = n - \tau$, a good approximation for one problem does not give one for the other. If $C$ is a cover with $|C| \le 2\tau$, the independent set $V \setminus C$ has size $n - |C| \ge n - 2\tau = 2\alpha - n$, which says nothing when $\alpha \le \frac n2$. For $K_n$, the optimal LP solution $x^* \equiv \frac12$ gives $C = V$ and the empty independent set, while $\alpha = 1$. In fact it is known (not proved here) that, unless P $=$ NP, no polynomial-time algorithm approximates $\alpha(G)$ within any constant factor.
:::

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID, FAINT = "#0b0b0b", "#52514e", "#d9d8d4", "#a3a29a"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})
from itertools import combinations
from scipy.optimize import linprog

def cover_lp(n, edges):
    A = np.zeros((len(edges), n))
    for k, (u, v) in enumerate(edges): A[k, u] = A[k, v] = 1
    res = linprog(np.ones(n), A_ub=-A, b_ub=-np.ones(len(edges)), bounds=[(0, 1)] * n)
    return res.fun, res.x

def tau_brute(n, edges):
    for k in range(n + 1):
        for C in combinations(range(n), k):
            if all(u in C or v in C for u, v in edges):
                return k

rng = np.random.default_rng(5)
n = 12
edges = [(u, v) for u, v in combinations(range(n), 2) if rng.random() < 0.3]
val, x = cover_lp(n, edges)
C = [v for v in range(n) if x[v] >= 0.5 - 1e-9]
print(f"random graph: tau_LP = {val:.2f}, rounded cover size = {len(C)}, tau = {tau_brute(n, edges)}")
for k in [2, 3, 4]:
    cyc = [(i, (i + 1) % (2 * k + 1)) for i in range(2 * k + 1)]
    print(f"odd cycle C_{2*k+1}: tau_LP = {cover_lp(2*k+1, cyc)[0]:.2f}, tau = {tau_brute(2*k+1, cyc)}")

def bipartite_round(x, U, W, edges, tol=1e-9):
    """Shifting procedure of the theorem for a bipartite graph with sides U, W."""
    x = x.copy()
    while True:
        FU = [u for u in U if tol < x[u] < 1 - tol]; FW = [w for w in W if tol < x[w] < 1 - tol]
        if not FU and not FW:
            return np.round(x)
        up = min([1 - x[u] for u in FU] + [x[w] for w in FW])   # largest eps >= 0
        x[FU] += up; x[FW] -= up

# an even cycle (bipartite): the LP solver may return a fractional optimum
m = 8
cyc = [(i, (i + 1) % m) for i in range(m)]
val, x = cover_lp(m, cyc)
xf = np.full(m, 0.5)                                  # another optimal solution, fully fractional
print("even cycle C_8: tau_LP =", val, "| fractional optimum value", xf.sum(),
      "-> rounded by shifting:", bipartite_round(xf, list(range(0, m, 2)), list(range(1, m, 2)), cyc))
```
