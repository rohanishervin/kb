# Parametrized Curves and Arc Length

**Prerequisites:** [The Cross Product](cross-product.ipynb) · [Local Extrema and the Mean Value Theorem](mean-value-theorem.ipynb#thm-mvt) · [The Fundamental Theorem of Calculus](fundamental-theorem-of-calculus.ipynb) · [Uniform Continuity and Compactness](uniform-continuity.ipynb) · [Properties of the Riemann Integral](properties-of-the-riemann-integral.ipynb#thm-riemann-sums)

## 1\. Curves, Velocity, Smoothness

::: {#def-curve}
## Parametrized Curve

A **parametrized curve** is a map $\gamma : I \to \mathbb{R}^n$, $\gamma(t) = (\gamma_1(t), \dots, \gamma_n(t))$, on an interval $I$. Its **image** (trace) is $\gamma(I) \subseteq \mathbb{R}^n$; its **graph** is $\{(t, \gamma(t))\} \subseteq \mathbb{R}^{n+1}$. Limits, continuity and the derivative
$$\gamma'(t_0) = \lim_{h\to0}\frac{\gamma(t_0 + h) - \gamma(t_0)}{h}$$
are defined with the Euclidean distance; $\gamma'(t_0)$ is the **velocity** and $|\gamma'(t_0)|$ the **speed**. An integral $\int_a^b\gamma = \big(\int_a^b\gamma_1, \dots, \int_a^b\gamma_n\big)$ is taken componentwise.
:::

::: {#lem-norm-components}
## Length Versus Components

For $z \in \mathbb{R}^n$: $\ |z_j| \le |z| \le \sqrt n\,\max_j|z_j|$ for every $j$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-norm-components
$z_j^2 \le \sum_i z_i^2 \le n\max_i z_i^2$. $\blacksquare$
:::

::: {#prp-curve-componentwise}
## Everything Is Componentwise

$\lim_{t\to t_0}\gamma(t) = L$ iff $\lim_{t\to t_0}\gamma_j(t) = L_j$ for every $j$. Hence $\gamma$ is continuous iff all $\gamma_j$ are, and $\gamma'(t_0)$ exists iff all $\gamma_j'(t_0)$ exist, with $\gamma'(t_0) = (\gamma_1'(t_0), \dots, \gamma_n'(t_0))$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-curve-componentwise
By @lem-norm-components, $|\gamma_j(t) - L_j| \le |\gamma(t) - L| \le \sqrt n\max_j|\gamma_j(t) - L_j|$: the middle tends to $0$ iff every component difference does. Apply this to $\gamma$ itself and to the difference quotients. $\blacksquare$
:::

::: {#def-smooth-curve}
## Smooth (Regular) Curve

$\gamma$ is **$C^1$** if $\gamma'$ exists and is continuous, and **smooth** (regular) if moreover $\gamma'(t) \ne 0$ for all $t$. Then $\gamma'(t_0)$ spans the **tangent line** $\{\gamma(t_0) + s\gamma'(t_0)\}$.
:::

Differentiability of $\gamma$ alone does not make the image look smooth; the condition $\gamma' \ne 0$ is what matters.

::: {#exm-curves}
## Four Curves

1. $\gamma(t) = (t^2, t^3)$ is $C^\infty$ with $\gamma'(t) = (2t, 3t^2)$, but $\gamma'(0) = 0$: the image $y^2 = x^3$ has a **cusp** at the origin.
2. $\gamma(t) = (t^2 - 1, t^3 - t)$ is smooth, yet $\gamma(1) = \gamma(-1) = (0, 0)$: the image crosses itself, with tangent directions $\gamma'(\pm1) = (\pm2, 2)$.
3. The **helix** $\gamma(t) = (a\cos t, a\sin t, bt)$ ($a > 0$) is smooth: $|\gamma'(t)| = \sqrt{a^2 + b^2}$.
4. $\gamma(t) = (t|t|, t^2)$ is $C^1$ with image the graph of $y = |x|$; its corner is possible because $\gamma'(0) = 0$.
:::

## 2\. Reparametrization and Arc Length

::: {#def-reparametrization}
## Reparametrization

If $\alpha : [c, d] \to [a, b]$ is a $C^1$ bijection with $\alpha' \ne 0$, then $\tilde\gamma = \gamma\circ\alpha$ is a **reparametrization** of $\gamma : [a, b] \to \mathbb{R}^n$; it has the same image and $\tilde\gamma'(\tau) = \alpha'(\tau)\,\gamma'(\alpha(\tau))$ (chain rule, componentwise). It is **orientation-preserving** if $\alpha' > 0$ and **orientation-reversing** if $\alpha' < 0$.
:::

::: {#lem-integral-norm}
## Norm of an Integral

For continuous $\gamma : [a, b] \to \mathbb{R}^n$, $\ \big|\int_a^b\gamma\big| \le \int_a^b|\gamma|$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-integral-norm
Let $v = \int_a^b\gamma$; assume $v \ne 0$ and put $u = v/|v|$. By linearity and Cauchy–Schwarz, $|v| = u\cdot v = \int_a^b u\cdot\gamma(t)\,dt \le \int_a^b|\gamma(t)|\,dt$. $\blacksquare$
:::

::: {#def-arc-length}
## Arc Length

The **length** of a $C^1$ curve $\gamma : [a, b] \to \mathbb{R}^n$ is $L(\gamma) = \int_a^b|\gamma'(t)|\,dt$, and $s(t) = \int_a^t|\gamma'(u)|\,du$ is the **arc length function**, so $\frac{ds}{dt} = |\gamma'(t)|$.
:::

The definition is justified by polygonal approximation. For a partition $\mathcal{P} = \{t_0 < \dots < t_k\}$ of $[a, b]$ let $\Lambda(\mathcal{P}) = \sum_{i=1}^k|\gamma(t_i) - \gamma(t_{i-1})|$ be the length of the inscribed polygon.

::: {#thm-arc-length-polygons}
## Arc Length as a Limit of Polygons

For a $C^1$ curve, $\Lambda(\mathcal{P}) \le L(\gamma)$ for every partition, and $\Lambda(\mathcal{P}) \to L(\gamma)$ as $\|\mathcal{P}\| \to 0$. So $L(\gamma) = \sup_{\mathcal{P}}\Lambda(\mathcal{P})$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-arc-length-polygons
*Upper bound.* By the FTC componentwise and @lem-integral-norm, $|\gamma(t_i) - \gamma(t_{i-1})| = \big|\int_{t_{i-1}}^{t_i}\gamma'\big| \le \int_{t_{i-1}}^{t_i}|\gamma'|$; sum over $i$.

*Convergence.* By the MVT for each component, $\gamma_j(t_i) - \gamma_j(t_{i-1}) = \gamma_j'(\xi_{ij})\,\Delta t_i$ for some $\xi_{ij} \in (t_{i-1}, t_i)$. Let $\varepsilon > 0$. Each $\gamma_j'$ is uniformly continuous on $[a, b]$, so there is $\delta > 0$ with $|\gamma_j'(\xi) - \gamma_j'(\eta)| < \varepsilon$ whenever $|\xi - \eta| < \delta$. If $\|\mathcal{P}\| < \delta$, the vector $w_i = (\gamma_1'(\xi_{i1}), \dots, \gamma_n'(\xi_{in}))$ satisfies $|w_i - \gamma'(t_i)| \le \sqrt n\,\varepsilon$ (@lem-norm-components), so
$$\Lambda(\mathcal{P}) = \sum_i|w_i|\,\Delta t_i \ge \sum_i|\gamma'(t_i)|\,\Delta t_i - \sqrt n\,\varepsilon(b - a).$$
The Riemann sums $\sum_i|\gamma'(t_i)|\Delta t_i$ of the continuous function $|\gamma'|$ tend to $L(\gamma)$. Hence $\liminf\Lambda(\mathcal{P}) \ge L(\gamma) - \sqrt n\,\varepsilon(b - a)$ for every $\varepsilon$, and with the upper bound $\Lambda(\mathcal{P}) \to L(\gamma)$. $\blacksquare$
:::

::: {#prp-arc-length-parametrization}
## Invariance and the Arc-Length Parameter

1. Length is unchanged by reparametrization.
2. A smooth curve $\gamma : [a, b] \to \mathbb{R}^n$ has an orientation-preserving reparametrization $\tilde\gamma : [0, L] \to \mathbb{R}^n$ with $|\tilde\gamma'| \equiv 1$; then the parameter of $\tilde\gamma$ is arc length. Conversely, if $|\gamma'| \equiv 1$ then $s(t) = t - a$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-arc-length-parametrization
(1) With $t = \alpha(\tau)$, substitution gives $\int_c^d|\alpha'(\tau)|\,|\gamma'(\alpha(\tau))|\,d\tau = \int_a^b|\gamma'(t)|\,dt$ (for $\alpha' < 0$ the limits are reversed and $|\alpha'| = -\alpha'$).
(2) $s : [a, b] \to [0, L]$ is $C^1$ with $s' = |\gamma'| > 0$, hence a bijection with $C^1$ inverse ([inverse function rule](chain-rule-and-inverse-functions.ipynb#thm-inverse-derivative)); let $\tilde\gamma = \gamma\circ s^{-1}$. Then $\tilde\gamma'(\sigma) = \gamma'(t)/s'(t) = \gamma'(t)/|\gamma'(t)|$ with $t = s^{-1}(\sigma)$, a unit vector. The converse is $s(t) = \int_a^t 1$. $\blacksquare$
:::

::: {#exm-limacon}
## Length of a Limaçon

The curve $r = 1 + 2\cos\theta$ in polar coordinates is $\gamma(\theta) = \big((1 + 2\cos\theta)\cos\theta, (1 + 2\cos\theta)\sin\theta\big)$, $0 \le \theta \le 2\pi$. Then $\gamma'(\theta) = (-\sin\theta - 2\sin2\theta,\ \cos\theta + 2\cos2\theta)$ and $|\gamma'|^2 = 5 + 4\cos\theta$, so
$$L = \int_0^{2\pi}\sqrt{5 + 4\cos\theta}\,d\theta \approx 13.36,$$
an elliptic integral with no elementary antiderivative. For the helix, $s(t) = \sqrt{a^2 + b^2}\,t$, and $t = s/\sqrt{a^2 + b^2}$ gives the arc-length parametrization.
:::

## 3\. Differentiation Rules

::: {#prp-curve-derivative-rules}
## Rules for Vector Functions

For differentiable $\gamma, \lambda : I \to \mathbb{R}^n$ and $f : I \to \mathbb{R}$:
$$(\gamma + \lambda)' = \gamma' + \lambda', \quad (f\gamma)' = f'\gamma + f\gamma', \quad (\gamma\cdot\lambda)' = \gamma'\cdot\lambda + \gamma\cdot\lambda',$$
and for $n = 3$, $(\gamma\times\lambda)' = \gamma'\times\lambda + \gamma\times\lambda'$. In particular $|\gamma|$ is constant iff $\gamma\cdot\gamma' \equiv 0$: a vector of constant length is orthogonal to its derivative.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-curve-derivative-rules
Each component of each expression is a sum of products of differentiable real functions; apply the [sum and product rules](differentiation-rules.ipynb) componentwise and regroup. For the last claim, $(|\gamma|^2)' = 2\gamma\cdot\gamma'$. $\blacksquare$
:::

```python

from scipy.integrate import quad
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

fig, axs = plt.subplots(1, 4, figsize=(14, 3.6))
t = np.linspace(-1.4, 1.4, 400)
axs[0].plot(t**2, t**3, color=BLUE, lw=2); axs[0].set_title(r"$(t^2, t^3)$: cusp")
t2 = np.linspace(-1.6, 1.6, 400)
axs[1].plot(t2**2 - 1, t2**3 - t2, color=ORANGE, lw=2); axs[1].set_title(r"$(t^2-1, t^3-t)$: crossing")
th = np.linspace(0, 2 * np.pi, 600)
r = 1 + 2 * np.cos(th)
axs[2].plot(r * np.cos(th), r * np.sin(th), color=VIOLET, lw=2); axs[2].set_title(r"limaçon $r = 1 + 2\cos\theta$")
t3 = np.linspace(-1.3, 1.3, 400)
axs[3].plot(t3 * np.abs(t3), t3**2, color=AQUA, lw=2); axs[3].set_title(r"$(t|t|, t^2)$: corner")
for ax in axs:
    ax.set_aspect("equal"); ax.plot(0, 0, "o", color=INK, ms=3)
plt.tight_layout(); plt.show()

L, _ = quad(lambda x: np.sqrt(5 + 4 * np.cos(x)), 0, 2 * np.pi)
g = np.stack([r * np.cos(th), r * np.sin(th)])
print("limacon length:", round(L, 6), "| inscribed polygon with 600 points:", round(np.sum(np.linalg.norm(np.diff(g, axis=1), axis=0)), 6))
```
