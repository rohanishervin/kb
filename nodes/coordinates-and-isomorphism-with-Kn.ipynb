# Coordinates and Linear Isomorphisms

**Prerequisites:** [Basis and Dimension](basis-and-dimension.ipynb)

Once a basis is fixed, every vector is described by a list of numbers, its **coordinates**. This turns an abstract $n$-dimensional space into the concrete space $K^n$. The precise statement uses **linear isomorphisms**: bijections that respect addition and scalar multiplication, so that the two spaces are "the same space with different names for the vectors".

Throughout, $V$ and $W$ are [vector spaces](vector-spaces.ipynb) over a field $K$.

## 1\. Unique Representation and Coordinates

::: {#prp-unique-representation}
## Unique Representation

Let $v_1, \dots, v_n$ span $V$. Then $v_1, \dots, v_n$ are linearly independent if and only if every $v \in V$ can be written **in exactly one way** as $v = c_1 v_1 + \dots + c_n v_n$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-unique-representation
($\Rightarrow$) Spanning gives at least one representation. If $\sum_i c_i v_i = \sum_i c_i' v_i$, subtracting gives $\sum_i (c_i - c_i') v_i = 0$, so $c_i = c_i'$ for all $i$ by independence.

($\Leftarrow$) The zero vector has the representation $0 = 0 v_1 + \dots + 0 v_n$. If it is the only one, then $\sum_i c_i v_i = 0$ forces all $c_i = 0$, which is independence. $\blacksquare$
:::

So a [basis](basis-and-dimension.ipynb#def-basis) can be described either as "independent and spanning" or as "every vector has unique coefficients".

::: {#def-coordinates}
## Coordinates

Let $\mathcal{B} = (v_1, \dots, v_n)$ be an ordered basis of $V$ and $v = c_1 v_1 + \dots + c_n v_n$. The column vector

$$[v]_{\mathcal{B}} = \begin{pmatrix} c_1 \\ \vdots \\ c_n \end{pmatrix} \in K^n$$

is the **coordinate vector** of $v$ with respect to $\mathcal{B}$, and $c_i$ is its $i$-th **coordinate**. The map $v \mapsto [v]_{\mathcal{B}}$ is the **coordinate map** of $\mathcal{B}$.
:::

::: {#exm-coordinates-polynomials}
## The Same Polynomial in Two Bases

In $P_2$ take $\mathcal{B} = (1, x, x^2)$ and $\mathcal{B}' = (1, x - 1, (x - 1)^2)$; the second is a basis because its elements have degrees $0, 1, 2$ (so they are independent) and there are $3 = \dim P_2$ of them. For $p(x) = x^2$,

$$x^2 = 1 + 2(x - 1) + (x - 1)^2, \qquad\text{so}\qquad [p]_{\mathcal{B}} = \begin{pmatrix} 0 \\ 0 \\ 1 \end{pmatrix}, \quad [p]_{\mathcal{B}'} = \begin{pmatrix} 1 \\ 2 \\ 1 \end{pmatrix}.$$

The coordinates with respect to $\mathcal{B}'$ are $p(1)$, $p'(1)$ and $p''(1)/2$: the coefficients of the Taylor expansion at $1$. Different bases, different coordinates, same vector.
:::

The animation shows the same effect in the plane. The yellow vector $v$ never moves. First it is measured with the standard basis $e_1, e_2$, which gives $[v] = (3, 1)$. Then the grid is redrawn for the basis $b_1 = (1, 1)$, $b_2 = (-1, 1)$, and the same arrow now reads $v = 2 b_1 - b_2$, so its coordinates are $(2, -1)$.


```python

from la_manim import *

class CoordinatesInTwoBases(GridScene):
    def construct(self):
        v = np.array([3.0, 1.0, 0.0])
        style = dict(buff=0, stroke_width=5, max_tip_length_to_length_ratio=0.2, tip_length=0.22)
        v_arrow = Arrow(ORIGIN, v, color=C_YELLOW, **style)
        v_lab = MathTex("v", color=C_YELLOW).next_to(v, UR, buff=0.08)
        b1 = self.vec([1, 0], C_BLUE, r"e_1", DOWN)
        b2 = self.vec([0, 1], C_ORANGE, r"e_2", LEFT)
        self.add(v_arrow, v_lab)

        path1 = VGroup(DashedLine(ORIGIN, [3, 0, 0], color=C_BLUE), DashedLine([3, 0, 0], v, color=C_ORANGE))
        cap1 = caption(r"v = 3\,e_1 + 1\,e_2 \quad [v]_{\mathcal{E}} = (3,\ 1)")
        self.play(Create(path1), FadeIn(cap1), run_time=1.5)
        self.wait(1.5)
        self.play(FadeOut(path1), FadeOut(cap1))

        # redraw the grid for the basis b1 = (1, 1), b2 = (-1, 1); v stays where it is
        self.transform([[1, -1], [1, 1]], run_time=3)
        new_labels = VGroup(MathTex("b_1", color=C_BLUE).next_to([1, 1, 0], RIGHT, buff=0.1),
                            MathTex("b_2", color=C_ORANGE).next_to([-1, 1, 0], LEFT, buff=0.1))
        self.play(FadeOut(b1[1]), FadeOut(b2[1]), FadeIn(new_labels))
        path2 = VGroup(DashedLine(ORIGIN, [2, 2, 0], color=C_BLUE), DashedLine([2, 2, 0], v, color=C_ORANGE))
        cap2 = caption(r"v = 2\,b_1 - 1\,b_2 \quad [v]_{\mathcal{B}} = (2,\ -1)")
        self.play(Create(path2), FadeIn(cap2), run_time=1.5)
        self.wait(2.5)

show(CoordinatesInTwoBases)
```


## 2\. Linear Isomorphisms

::: {#def-linear-isomorphism}
## Linear Isomorphism

A map $f : V \to W$ is a **linear isomorphism** if

1. $f$ is bijective,
2. $f(u + v) = f(u) + f(v)$ for all $u, v \in V$, and
3. $f(a v) = a f(v)$ for all $a \in K$, $v \in V$.

If such an $f$ exists, $V$ and $W$ are **isomorphic**, written $V \cong W$. (Maps satisfying 2 and 3 are the [linear maps](linear-maps.ipynb#def-linear-map); an isomorphism is a bijective linear map.)
:::

::: {.callout-note}
## Intuition: Same Structure, Different Labels
An isomorphism renames the vectors of $V$ as vectors of $W$ in such a way that sums go to sums and multiples go to multiples. Any statement about $V$ expressed only through addition and scalar multiplication is therefore true in $V$ exactly when the corresponding statement is true in $W$. For example, $P_1 = \{a + bx\} \cong \mathbb{R}^2$ via $a + bx \mapsto (a, b)$.
:::

::: {#prp-isomorphism-basic}
## Inverses and Composites of Isomorphisms

1. If $f : V \to W$ is a linear isomorphism, so is its inverse $f^{-1} : W \to V$.
2. If $f : V \to W$ and $g : W \to U$ are linear isomorphisms, so is $g \circ f$.
3. Consequently $\cong$ is reflexive, symmetric and transitive.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-isomorphism-basic
1. $f^{-1}$ is a bijection. For $w, w' \in W$ put $v = f^{-1}(w)$ and $v' = f^{-1}(w')$. Then $f(v + v') = w + w'$, so $f^{-1}(w + w') = v + v' = f^{-1}(w) + f^{-1}(w')$. Similarly $f(a v) = a w$ gives $f^{-1}(a w) = a f^{-1}(w)$.
2. A composite of bijections is a bijection, and $g(f(u + v)) = g(f(u) + f(v)) = g(f(u)) + g(f(v))$, and likewise for scalar multiples.
3. $\mathrm{id}_V$ is an isomorphism; symmetry is part 1; transitivity is part 2. $\blacksquare$
:::

::: {#prp-isomorphism-preserves-bases}
## Isomorphisms Preserve Independence, Spanning and Dimension

Let $f : V \to W$ be a linear isomorphism and $v_1, \dots, v_n \in V$.

1. $v_1, \dots, v_n$ are independent if and only if $f(v_1), \dots, f(v_n)$ are independent.
2. $v_1, \dots, v_n$ span $V$ if and only if $f(v_1), \dots, f(v_n)$ span $W$.
3. Hence $f$ maps bases to bases, and if $V$ is finite-dimensional then $\dim W = \dim V$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-isomorphism-preserves-bases
Because $f$ preserves sums and multiples, $f(c_1 v_1 + \dots + c_n v_n) = c_1 f(v_1) + \dots + c_n f(v_n)$, and $f(0) = f(0 \cdot 0) = 0$.

1. If the $v_i$ are independent and $\sum c_i f(v_i) = 0$, then $f(\sum c_i v_i) = 0 = f(0)$, so $\sum c_i v_i = 0$ by injectivity and all $c_i = 0$. If the $f(v_i)$ are independent and $\sum c_i v_i = 0$, then $\sum c_i f(v_i) = f(0) = 0$, so all $c_i = 0$.
2. If the $v_i$ span $V$, every $w \in W$ is $f(v)$ for some $v = \sum c_i v_i$ (surjectivity), hence $w = \sum c_i f(v_i)$. Conversely apply this to the isomorphism $f^{-1}$ ([Proposition](#prp-isomorphism-basic)).
3. Combine 1 and 2; a basis of $n$ vectors goes to a basis of $n$ vectors. $\blacksquare$
:::

## 3\. Every $n$-Dimensional Space Is Isomorphic to $K^n$

::: {#prp-fin-dim-isomorphic-Kn}
## The Coordinate Isomorphism

If $\mathcal{B} = (v_1, \dots, v_n)$ is an ordered basis of $V$, then the coordinate map
$$V \to K^n, \qquad v \mapsto [v]_{\mathcal{B}}$$
is a linear isomorphism. In particular every $n$-dimensional space over $K$ is isomorphic to $K^n$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-fin-dim-isomorphic-Kn
**Well defined and injective.** By @prp-unique-representation each $v$ has exactly one coefficient list, so $[v]_{\mathcal{B}}$ is well defined, and different vectors have different coefficient lists.

**Surjective.** Any $(c_1, \dots, c_n)$ is the coordinate vector of $c_1 v_1 + \dots + c_n v_n$.

**Linear.** If $v = \sum c_i v_i$ and $w = \sum d_i v_i$, then $v + w = \sum (c_i + d_i) v_i$ and $a v = \sum (a c_i) v_i$. By uniqueness these are the representations of $v + w$ and $a v$, so $[v + w]_{\mathcal{B}} = [v]_{\mathcal{B}} + [w]_{\mathcal{B}}$ and $[a v]_{\mathcal{B}} = a [v]_{\mathcal{B}}$. $\blacksquare$
:::

::: {#thm-iso-iff-same-dim}
## Classification of Finite-Dimensional Spaces

Two finite-dimensional vector spaces over $K$ are isomorphic if and only if they have the same dimension.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-iso-iff-same-dim
If $V \cong W$, then $\dim V = \dim W$ by @prp-isomorphism-preserves-bases. If $\dim V = \dim W = n$, then $V \cong K^n$ and $W \cong K^n$ by @prp-fin-dim-isomorphic-Kn, so $V \cong W$ by @prp-isomorphism-basic. $\blacksquare$
:::

::: {.callout-warning}
## The Isomorphism Depends on the Basis
The isomorphism $V \cong K^n$ is not canonical: each ordered basis gives a different one, as the animation shows. For example $M_2(\mathbb{R}) \cong \mathbb{R}^4$, but nothing singles out one particular way of listing the four entries. This is why it pays to keep working with abstract spaces and to choose coordinates only when they help.
:::
