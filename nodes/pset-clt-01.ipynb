# Problem Set: Central Limit Theorem

## Problem 1: Elevator Capacity

**Problem Statement:**

An elevator has a capacity of $4500\text{ kg}$. The weight of an individual is a random variable with mean $\mu = 80\text{ kg}$ and standard deviation $\sigma = 20\text{ kg}$.

Assuming $50$ people enter the elevator and their weights are independent and identically distributed, what is the probability that their total weight exceeds $4500\text{ kg}$?

::: {.callout-tip collapse="true"}
## Solution

Let $W_1, \dots, W_{50}$ be the weights of the passengers, and $S_{50} = \sum_{i=1}^{50} W_i$ be the total weight.

- **Mean:** $E(S_{50}) = 50 \times 80 = 4000\text{ kg}$
- **Standard Deviation:** $\text{SD}(S_{50}) = \sqrt{50} \times 20 = 100\sqrt{2} \approx 141.42\text{ kg}$

By the [Central Limit Theorem](law-of-large-numbers-and-central-limit-theorem.ipynb), as $n \to \infty$, the standardized sum converges in distribution to the standard [Normal distribution](normal-distribution.ipynb):

$$
Z_n = \frac{\sum_{i=1}^n W_i - n\mu}{\sqrt{n}\sigma} \xrightarrow{d} \mathcal{N}(0, 1)
$$

The probability that a standard normal variable $Z$ exceeds a threshold $a$ is given by the integral of the standard normal density $\varphi(t) = \frac{1}{\sqrt{2\pi}} e^{-t^2/2}$:

$$
\lim_{n \to \infty} \mathbb{P}\left( \frac{\sum_{i=1}^n W_i - n\mu}{\sqrt{n}\sigma} \ge a \right) = \frac{1}{\sqrt{2\pi}} \int_a^\infty e^{-t^2/2} \, dt = 1 - \Phi(a)
$$

where $\Phi(a) = \frac{1}{\sqrt{2\pi}} \int_{-\infty}^a e^{-t^2/2} \, dt$ is the standard normal cumulative distribution function (CDF).

Standardizing our problem with $n = 50$:

$$
a = \frac{4500 - n\mu}{\sqrt{n}\sigma} = \frac{4500 - 50 \times 80}{\sqrt{50} \times 20} = \frac{500}{100\sqrt{2}} = \frac{5}{\sqrt{2}} \approx 3.54
$$

Therefore:

$$
\begin{aligned}
\mathbb{P}(S_{50} \ge 4500) &= \mathbb{P}\left( \frac{S_{50} - 4000}{100\sqrt{2}} \ge \frac{4500 - 4000}{100\sqrt{2}} \right) \\
&\approx \frac{1}{\sqrt{2\pi}} \int_{3.54}^\infty e^{-t^2/2} \, dt \\
&= 1 - \Phi(3.54) \\
&\approx 0.0002
\end{aligned}
$$

The probability is approximately **$0.0002$** (or $0.02\%$). $\blacksquare$
:::
