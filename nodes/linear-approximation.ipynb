# Linear Approximation and Its Error

**Prerequisites:** [The Chain Rule and Derivatives of Inverse Functions](chain-rule-and-inverse-functions.ipynb) · [Local Extrema and the Mean Value Theorem](mean-value-theorem.ipynb)

## 1\. Linear Approximation

If $f$ is differentiable at $a$, the [tangent line](derivative.ipynb#thm-linear-approximation-char) is the best affine approximation near $a$:
$$f(a + h) \approx f(a) + f'(a)h, \qquad \text{equivalently} \qquad \Delta y \approx dy = f'(a)\,\Delta x .$$
The **relative error** of a quantity $x$ is $\frac{\Delta x}{x}$. For $y = kx^n$, $\frac{dy}{y} = \frac{nkx^{n-1}dx}{kx^n} = n\frac{dx}{x}$: the relative error is multiplied by $n$.

::: {#exm-linear-approximation}
## Examples

1. $\sqrt[3]{1.012} \approx 1 + \frac13(0.012) = 1.004$ (tangent to $x^{1/3}$ at $1$, slope $\frac13$).
2. $\sin\theta \approx \theta$ for small $\theta$ (tangent at $0$, slope $\cos 0 = 1$).
3. *Measurement error.* In the conical tank $V = \frac{\pi}{12}h^3$, $h = 6 \pm 0.1$ cm: $dV = \frac\pi4 h^2\,dh = 9\pi\,dh$, so the volume error is about $|dV| \le 0.9\pi \approx 2.83\ \text{cm}^3$. A relative error of $1\%$ in $h$ gives about $3\%$ in $V$.
4. *Area of a square.* $(a \pm h)^2 - a^2 = \pm 2ah + h^2 \approx \pm 2ah$; relatively, $\pm 2\frac{h}{a}$.
:::

## 2\. The Error and the Second Derivative

If $f$ is differentiable on an open set $S'$, the derivative of $f' : S' \to \mathbb{R}$ at $a$, if it exists, is the **second derivative** $f''(a) = \frac{d^2y}{dx^2}$. The size of $f''$ measures how fast the slope turns, i.e. how far the graph departs from its tangent.

::: {#thm-rolle-second-order}
## Rolle's Theorem for $f''$

Let $f$ be twice differentiable on an open interval $I$, $a \ne b$ in $I$, $f(a) = f(b) = 0$ and $f'(a) = 0$. Then $f''(c) = 0$ for some $c$ strictly between $a$ and $b$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-rolle-second-order
[Rolle](mean-value-theorem.ipynb#thm-rolle) for $f$ between $a$ and $b$ gives $c_1$ strictly between them with $f'(c_1) = 0$. Rolle for $f'$ between $a$ and $c_1$ ($f'(a) = f'(c_1) = 0$) gives $c$ with $f''(c) = 0$. $\blacksquare$
:::

::: {#thm-linear-approx-error}
## Error of the Linear Approximation

Let $f$ be twice differentiable on an open interval $I$ and $a, b \in I$. Then for some $c$ between $a$ and $b$
$$f(b) - \big[f(a) + f'(a)(b - a)\big] = \tfrac12 f''(c)(b - a)^2 .$$
In particular, if $|f''| \le M$ between $a$ and $a + h$: $\ \big|f(a + h) - f(a) - f'(a)h\big| \le \frac M2 h^2$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-linear-approx-error
For $b = a$ there is nothing to prove. Let $\varphi(x) = f(a) + f'(a)(x - a) + C(x - a)^2$ with $C$ chosen so that $\varphi(b) = f(b)$:
$$C = \frac{f(b) - f(a) - f'(a)(b - a)}{(b - a)^2}.$$
Then $g = f - \varphi$ satisfies $g(a) = g(b) = 0$ and $g'(a) = f'(a) - f'(a) = 0$. By @thm-rolle-second-order, $0 = g''(c) = f''(c) - 2C$, i.e. $C = \frac12 f''(c)$. $\blacksquare$
:::

This is the case $k = 1$ of [Taylor's theorem](taylor-polynomials.ipynb#thm-lagrange-remainder). The error is quadratic in $h$: one more correct decimal in $h$ gives about two more in the approximation. Its sign is the sign of $f''$: for $f'' < 0$ the tangent lies above the graph and the approximation is too large.

::: {#exm-cube-root-error}
## How Good Is $\sqrt[3]{1.012} \approx 1.004$?

$f(x) = x^{1/3}$, $f''(x) = -\frac29 x^{-5/3}$. With $a = 1$, $h = 0.012$:
$$\sqrt[3]{1.012} - 1.004 = \tfrac12\Big(-\tfrac29\Big)c^{-5/3}(0.012)^2, \qquad 1 < c < 1.012 .$$
So the error is negative and its size is less than $\frac19 (0.012)^2 = 1.6\cdot 10^{-5}$. Indeed $\sqrt[3]{1.012} = 1.0039841\dots$ For larger $h$ the bound deteriorates: $f''$ grows near $0$ and decays for large $x$, so the approximation is better for $h > 0$ than for $-h$ (see the table below).
:::

```python

import numpy as np

f = lambda x: np.cbrt(x)
lin = lambda h: 1 + h / 3
print(f"{'h':>7} {'cbrt(1+h)':>12} {'1+h/3':>9} {'error':>12} {'bound |h|^2/9 * max c^(-5/3)':>30}")
for h in [0.012, 0.21, -0.21, 1.0, -1.0]:
    exact = f(1 + h)
    c_min = min(1, 1 + h)                       # |f''| is largest at the smaller endpoint
    bound = h**2 / 9 * c_min ** (-5 / 3) if c_min > 0 else np.inf
    print(f"{h:7.3f} {exact:12.7f} {lin(h):9.5f} {exact - lin(h):12.3e} {bound:30.3e}")
```
