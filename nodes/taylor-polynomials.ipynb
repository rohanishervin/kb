# Taylor Polynomials and Taylor's Theorem

**Prerequisites:** [Linear Approximation and Its Error](linear-approximation.ipynb) · [Local Extrema and the Mean Value Theorem](mean-value-theorem.ipynb) · [Convexity, Inflection Points, and Curve Sketching](convexity-and-curve-sketching.ipynb#thm-second-derivative-test)

## 1\. Higher Derivatives

If $f^{(k)}$ exists on an open interval around $a$ and is differentiable at $a$, its derivative is $f^{(k+1)}(a)$ ($f^{(0)} = f$, $f^{(1)} = f'$, $f^{(2)} = f''$). $f$ is **$k$ times differentiable** if $f^{(k)}$ exists on its domain, and **infinitely differentiable** if all $f^{(k)}$ exist. Polynomials, $\sin$ and $\cos$ are infinitely differentiable; $\sin^{(4)} = \sin$, and $p^{(k)} = 0$ for $k > \deg p$.

::: {#exm-finitely-differentiable}
## Exactly $n - 1$ Times Differentiable

Let $f(x) = x^n$ for $x \ge 0$ and $f(x) = -x^n$ for $x < 0$ ($n \ge 1$), i.e. $f(x) = x^{n-1}|x|$. Away from $0$ it is a polynomial. By induction, for $k \le n - 1$, $f^{(k)}(x) = \pm\, n(n-1)\cdots(n-k+1)x^{n-k}$ (sign $+$ for $x \ge 0$, $-$ for $x < 0$): at $0$ the difference quotient of $f^{(k-1)}$ is $\pm c\, h^{n-k}$, which tends to $0$ when $k < n$. For $k = n - 1$ this gives $f^{(n-1)}(x) = n!\,|x|$, which is not differentiable at $0$. So $f$ is exactly $n - 1$ times differentiable at $0$.
:::

::: {#lem-general-leibniz}
## General Leibniz Rule

If $f, g$ are $n$ times differentiable, then $(fg)^{(n)} = \sum_{k=0}^n \binom{n}{k} f^{(k)}g^{(n-k)}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-general-leibniz
Induction on $n$: differentiate the formula for $n$ with the product rule and use $\binom{n}{k} + \binom{n}{k-1} = \binom{n+1}{k}$, exactly as in the proof of the [binomial theorem](binomial-theorem.ipynb). $\blacksquare$
:::

## 2\. Taylor Polynomials

::: {#thm-taylor-polynomial}
## The Taylor Polynomial

Let $f$ be $k$ times differentiable at $a$. There is exactly one polynomial $p_k$ of degree $\le k$ with $p_k^{(j)}(a) = f^{(j)}(a)$ for $j = 0, \dots, k$, namely
$$p_k(x) = \sum_{j=0}^k \frac{f^{(j)}(a)}{j!}(x - a)^j = f(a) + f'(a)(x - a) + \frac{f''(a)}{2!}(x - a)^2 + \dots + \frac{f^{(k)}(a)}{k!}(x-a)^k,$$
the **Taylor polynomial of degree $k$** of $f$ at $a$. For $k = 1$ it is the linear approximation.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-taylor-polynomial
Write any polynomial of degree $\le k$ in powers of $x - a$: $p(x) = \sum_{j=0}^k c_j (x-a)^j$. Differentiating $j$ times and putting $x = a$ kills every term except $j!\,c_j$, so $p^{(j)}(a) = j!\,c_j$. The conditions force $c_j = f^{(j)}(a)/j!$. $\blacksquare$
:::

::: {#thm-taylor-peano}
## Taylor's Theorem with the Peano Remainder

Let $f$ be $(k-1)$ times differentiable on an open interval $I \ni a$ and $k$ times differentiable at $a$. Then
$$\lim_{x \to a} \frac{f(x) - p_k(x)}{(x - a)^k} = 0 .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-taylor-peano
Induction on $k$. For $k = 1$ this is the definition of $f'(a)$. Let $k \ge 2$ and assume the statement for $k - 1$. Put $\varphi = f - p_k$. Then $\varphi(a) = 0$, and $\varphi' = f' - p_k'$ where $p_k'$ is the Taylor polynomial of degree $k - 1$ of $f'$ at $a$ (differentiate the formula). $f'$ satisfies the hypotheses with $k - 1$, so for $\varepsilon > 0$ there is $\delta$ with $|\varphi'(x)| < \varepsilon|x - a|^{k-1}$ for $0 < |x - a| < \delta$. By the [MVT](mean-value-theorem.ipynb#thm-mvt), $\varphi(x) = \varphi(x) - \varphi(a) = \varphi'(c)(x - a)$ with $c$ between $a$ and $x$, so $|c - a| < |x - a|$ and
$$|\varphi(x)| < \varepsilon|c - a|^{k-1}|x - a| \le \varepsilon|x - a|^k . \ \blacksquare$$
:::

::: {#cor-taylor-unique}
## Uniqueness of the Approximating Polynomial

If $q$ is a polynomial of degree $\le k$ with $\frac{f(x) - q(x)}{(x-a)^k} \to 0$, then $q = p_k$. So any algebraic way of finding such a $q$ yields the Taylor polynomial.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-taylor-unique
$r = p_k - q = \sum_{j \le k} c_j (x - a)^j$ satisfies $r(x)/(x-a)^k \to 0$. If some $c_j \ne 0$, let $j$ be the smallest; then $r(x)/(x-a)^j \to c_j \neq 0$, while $r(x)/(x-a)^j = (x-a)^{k-j} r(x)/(x-a)^k \to 0$: contradiction. $\blacksquare$
:::

::: {#thm-lagrange-remainder}
## Taylor's Theorem with the Lagrange Remainder

Let $f$ be $(k+1)$ times differentiable on an open interval $I$, and $a, x \in I$. Then for some $c$ between $a$ and $x$
$$f(x) - p_k(x) = \frac{f^{(k+1)}(c)}{(k+1)!}(x - a)^{k+1}.$$
:::

::: {#lem-generalized-rolle}
## Generalized Rolle

If $g$ is $(k+1)$ times differentiable on $I$, $a \ne b$ in $I$, $g(a) = g'(a) = \dots = g^{(k)}(a) = 0$ and $g(b) = 0$, then $g^{(k+1)}(c) = 0$ for some $c$ strictly between $a$ and $b$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-generalized-rolle and @thm-lagrange-remainder
*Lemma.* Rolle gives $b_1$ between $a$ and $b$ with $g'(b_1) = 0$; then Rolle for $g'$ on $[a, b_1]$ gives $b_2$ with $g''(b_2) = 0$; after $k+1$ steps, $g^{(k+1)}(b_{k+1}) = 0$.

*Theorem.* Let $x \ne a$ and $Q(t) = p_k(t) + C(t - a)^{k+1}$ with $C$ chosen so that $Q(x) = f(x)$. Then $g = f - Q$ satisfies $g^{(j)}(a) = 0$ for $j \le k$ (as $p_k^{(j)}(a) = f^{(j)}(a)$) and $g(x) = 0$. By the lemma, $0 = g^{(k+1)}(c) = f^{(k+1)}(c) - (k+1)!\,C$. So $f(x) - p_k(x) = C(x-a)^{k+1}$ with $C = \frac{f^{(k+1)}(c)}{(k+1)!}$. $\blacksquare$
:::

::: {#exm-taylor-polynomials}
## Examples

1. At $a = 0$: $\sin$ has derivatives $0, 1, 0, -1$ repeating, so its Taylor polynomials of degree $2m+1$ and $2m+2$ coincide: $x - \frac{x^3}{3!} + \dots + (-1)^m\frac{x^{2m+1}}{(2m+1)!}$. Similarly $\cos$: $1 - \frac{x^2}{2!} + \frac{x^4}{4!} - \dots$.
2. $f(x) = \frac1x$ at $a = 1$: $f^{(j)}(x) = (-1)^j j!\,x^{-j-1}$, so $p_k(x) = \sum_{j=0}^k (-1)^j (x - 1)^j$. It approximates $\frac1x$ well only for $x$ near $1$: as $x \to 0^+$, $\frac1x \to \infty$ while $p_k(0) = k + 1$; at $x = 2$, $p_k(2)$ is $1$ or $0$ while $\frac1x = \frac12$.
3. For a polynomial, expand in powers of $x - a$: $1 - x + x^4$ at $a = -1$ is $3 - 5(x+1) + 6(x+1)^2 - 4(x+1)^3 + (x+1)^4$; its Taylor polynomials are the truncations, and $p_4$ is the polynomial itself.
4. $\sin\frac{1}{10} \approx \frac1{10} - \frac{1}{6000}$: this is also $p_4$, so the error is $\frac{|\sin^{(5)} c|}{5!}10^{-5} \le \frac{10^{-5}}{120} < 10^{-7}$.
5. $\frac{1}{1.01} \approx 1 - \frac{1}{100} + \dots + \frac{(-1)^k}{100^k}$ (example 2 with $x = 1.01$): the error is $\big|(-1)^{k+1}c^{-k-2}\,100^{-k-1}\big| < 10^{-2k-2}$ since $c > 1$. Here the exact error is a geometric tail: $\frac{100^{-k-1}}{1.01}$.
:::

## 3\. Critical Points Revisited

::: {#thm-higher-derivative-test}
## The $k$-th Derivative Test

Let $k \ge 2$, $f$ be $(k-1)$ times differentiable near $a$ and $k$ times at $a$, with $f'(a) = \dots = f^{(k-1)}(a) = 0 \ne f^{(k)}(a)$.

1. If $k$ is even, $a$ is a strict local minimum when $f^{(k)}(a) > 0$ and a strict local maximum when $f^{(k)}(a) < 0$.
2. If $k$ is odd, $a$ is not a local extremum.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-higher-derivative-test
Here $p_k(x) = f(a) + \frac{f^{(k)}(a)}{k!}(x-a)^k$, so by @thm-taylor-peano
$$\frac{f(x) - f(a)}{(x - a)^k} \to \frac{f^{(k)}(a)}{k!} \neq 0 .$$
Hence for $x$ near $a$, $x \ne a$, $\frac{f(x) - f(a)}{(x-a)^k}$ has the sign of $f^{(k)}(a)$. If $k$ is even, $(x-a)^k > 0$, so $f(x) - f(a)$ has that sign on both sides. If $k$ is odd, $(x - a)^k$ changes sign at $a$, so $f(x) - f(a)$ does too. $\blacksquare$
:::

::: {#exm-higher-derivative-test}
## Examples

1. $f(x) = x^6\cos x - x^5\sin x$ at $0$. By the Lagrange remainder, $\cos x = 1 - \frac{x^2}{2} + \frac{x^4}{24}\cos c_1$ and $\sin x = x - \frac{x^3}{6} + \frac{x^5}{120}\cos c_2$, so
$$f(x) = -\tfrac13 x^8 + x^{10}\Big(\tfrac{\cos c_1}{24} - \tfrac{\cos c_2}{120}\Big),$$
and the bracket is bounded. By @cor-taylor-unique, $p_9 = -\frac13 x^8$: $f^{(j)}(0) = 0$ for $j < 8$ and $f^{(8)}(0) = -\frac{8!}{3} < 0$. So $0$ is a local maximum.
2. $f(x) = (x^2 - 2x)^{100}$ has critical points $0, 1, 2$ ($f' = 100(x^2 - 2x)^{99}(2x - 2)$). At $0$: $f = x^{100}(x - 2)^{100} = 2^{100}x^{100} + (\text{higher powers})$, so $f^{(100)}(0) = 2^{100}\,100! > 0$ and the lower derivatives vanish: local minimum (obvious anyway, as $f \ge 0 = f(0)$). Likewise at $2$. At $1$: $f = \big((x-1)^2 - 1\big)^{100} = 1 - 100(x - 1)^2 + \dots$, so $f''(1) = -200 < 0$: local maximum.
:::

```python

import numpy as np
import matplotlib.pyplot as plt
from math import factorial

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

def p_sin(x, k):
    return sum((-1) ** m * x ** (2 * m + 1) / factorial(2 * m + 1) for m in range((k + 1) // 2))

x = np.linspace(-2 * np.pi, 2 * np.pi, 600)
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.plot(x, np.sin(x), color=INK, lw=2.5, label=r"$\sin x$")
for k, col in zip([1, 3, 5, 7, 9], [BLUE, ORANGE, AQUA, VIOLET, YELLOW]):
    ax.plot(x, p_sin(x, k), color=col, lw=1.4, label=f"$p_{{{k}}}$")
ax.set_ylim(-2.5, 2.5); ax.legend(ncol=3, fontsize=9, loc="lower left")
ax.set_title(r"Taylor polynomials of $\sin$ at $0$: good near $0$, worse far away")
plt.tight_layout(); plt.show()

import sympy as sp
X = sp.symbols("x")
u = sp.Symbol("u")   # u = x + 1
print("1 - x + x^4 in powers of u = x + 1:", sp.Poly(sp.expand((1 - X + X**4).subs(X, u - 1)), u).as_expr())
print("Taylor of x^6 cos x - x^5 sin x at 0:", sp.series(X**6*sp.cos(X) - X**5*sp.sin(X), X, 0, 11))
```
