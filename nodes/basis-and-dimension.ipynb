# Basis and Dimension

**Prerequisites:** [Linear Independence](linear-independence.ipynb) · [Span](subspaces.ipynb#def-span)

A **basis** is a set of vectors that is large enough to build every vector (it spans) and small enough to have no redundancy (it is independent). The central fact of this note is that all bases of a space have the same number of elements, the **dimension**. Everything rests on one combinatorial lemma: in the span of $n$ vectors there is no room for more than $n$ independent vectors.

Throughout, $V$ is a [vector space](vector-spaces.ipynb) over a field $K$.

## 1\. The Exchange Lemma (Steinitz)

::: {#lem-exchange}
## Exchange Lemma

Let $x_1, \dots, x_n \in V$, and let $y_1, \dots, y_m$ be linearly independent vectors in $\operatorname{span}(x_1, \dots, x_n)$. Then $m \le n$. Moreover, after renumbering the $x_j$, one can replace $x_1, \dots, x_m$ by $y_1, \dots, y_m$ without changing the span:

$$\operatorname{span}(y_1, \dots, y_m, x_{m+1}, \dots, x_n) = \operatorname{span}(x_1, \dots, x_n).$$
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-exchange
Write $X = \operatorname{span}(x_1, \dots, x_n)$. We prove by induction on $k = 0, 1, \dots, m$ the statement

> $(E_k)$: $k \le n$ and, after renumbering the $x_j$, $\operatorname{span}(y_1, \dots, y_k, x_{k+1}, \dots, x_n) = X$.

$(E_0)$ is trivial. Assume $(E_k)$ for some $k < m$. Since $y_{k+1} \in X$, we can write

$$y_{k+1} = b_1 y_1 + \dots + b_k y_k + c_{k+1} x_{k+1} + \dots + c_n x_n.$$

If all $c_j$ were zero (this includes the case $k = n$, where there are no $x_j$ left), then $y_{k+1}$ would lie in $\operatorname{span}(y_1, \dots, y_k)$, contradicting the [independence](linear-independence.ipynb#prp-dependence-redundancy) of $y_1, \dots, y_m$. Hence $k < n$, so $k + 1 \le n$, and some $c_j \neq 0$; renumber $x_{k+1}, \dots, x_n$ so that $c_{k+1} \neq 0$. Solving for $x_{k+1}$,

$$x_{k+1} = \frac{1}{c_{k+1}}\Big(y_{k+1} - \sum_{i \le k} b_i y_i - \sum_{j \ge k+2} c_j x_j\Big) \in \operatorname{span}(y_1, \dots, y_{k+1}, x_{k+2}, \dots, x_n).$$

So the list $y_1, \dots, y_{k+1}, x_{k+2}, \dots, x_n$ spans every vector of the list in $(E_k)$, hence all of $X$; conversely all its vectors lie in $X$. This proves $(E_{k+1})$. For $k = m$ we obtain $m \le n$ and the exchanged spanning list. $\blacksquare$
:::

::: {.callout-note}
## Intuition
On a line (the span of one vector) any two vectors are dependent; in a plane (the span of two vectors) any three are dependent. The proof makes this precise: each new independent $y$ can be traded for one of the $x$'s without losing anything, so the $x$'s cannot run out before the $y$'s do.
:::

## 2\. Bases

::: {#def-basis}
## Basis

A **basis** of $V$ is a linearly independent subset $\mathcal{B} \subseteq V$ that spans $V$. When the basis is finite and its elements are listed in a fixed order, $\mathcal{B} = (v_1, \dots, v_n)$, we call it an **ordered basis**.
:::

::: {#def-finite-dimensional}
## Finite-Dimensional Space

$V$ is **finite-dimensional** if it has a finite basis. The zero space $\{0\}$ is finite-dimensional, with the empty set as basis. A space that is not finite-dimensional is **infinite-dimensional**.
:::

::: {#cor-finitely-generated}
## Every Finite Spanning Set Contains a Basis

If $V = \operatorname{span}(x_1, \dots, x_n)$, then some subset of $\{x_1, \dots, x_n\}$ is a basis of $V$. In particular $V$ is finite-dimensional and has a basis with at most $n$ elements.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-finitely-generated
Among all subsets of $\{x_1, \dots, x_n\}$ that span $V$ (there is at least one, the whole set), choose one, $S$, with the fewest elements. If $S$ were dependent, then by [redundancy](linear-independence.ipynb#prp-dependence-redundancy) some $x \in S$ would lie in $\operatorname{span}(S \setminus \{x\})$. Then $\operatorname{span}(S \setminus \{x\})$ is a subspace containing all of $S$, so it contains $\operatorname{span}(S) = V$; thus $S \setminus \{x\}$ is a smaller spanning subset, a contradiction. Hence $S$ is independent and spans $V$: it is a basis with at most $n$ elements. $\blacksquare$
:::

## 3\. Dimension

::: {#thm-basis-cardinality}
## All Bases Have the Same Size

If $(x_1, \dots, x_n)$ and $(y_1, \dots, y_m)$ are bases of $V$, then $m = n$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-basis-cardinality
The $y_i$ are independent and lie in $V = \operatorname{span}(x_1, \dots, x_n)$, so $m \le n$ by the [exchange lemma](#lem-exchange). Exchanging the roles of the two bases gives $n \le m$. $\blacksquare$
:::

::: {#def-dimension}
## Dimension

The **dimension** $\dim V$ of a finite-dimensional space $V$ is the number of elements of any basis. It is well defined by @thm-basis-cardinality. We write $\dim V = \infty$ if $V$ is infinite-dimensional.
:::

::: {#cor-independent-full-size}
## Counting Criteria for Bases

Let $\dim V = n$.

1. Every linearly independent subset of $V$ has at most $n$ elements, and every spanning subset has at least $n$ elements.
2. A linearly independent set with exactly $n$ elements is a basis.
3. A spanning set with exactly $n$ elements is a basis.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-independent-full-size
Fix a basis $(v_1, \dots, v_n)$.

1. Independent vectors lie in $V = \operatorname{span}(v_1, \dots, v_n)$, so there are at most $n$ of them by the [exchange lemma](#lem-exchange). If $x_1, \dots, x_m$ span $V$, the $n$ independent vectors $v_i$ lie in their span, so $n \le m$.
2. Let $y_1, \dots, y_n$ be independent. By the exchange lemma with $m = n$, all of $v_1, \dots, v_n$ can be exchanged: $\operatorname{span}(y_1, \dots, y_n) = \operatorname{span}(v_1, \dots, v_n) = V$. So the $y_i$ also span.
3. A spanning set with $n$ elements contains a basis by @cor-finitely-generated; that basis has $n$ elements, so it is the whole set. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Dimension Counts Degrees of Freedom
Dimension is an invariant of the space, not of a choice of basis. It is both the **largest** number of independent vectors and the **smallest** number of spanning vectors that $V$ admits; a basis sits exactly at the point where the two meet.
:::

::: {#thm-extend-to-basis}
## Extending an Independent Set to a Basis

Let $\dim V = n$. Every linearly independent list $y_1, \dots, y_m$ in $V$ can be extended to a basis $y_1, \dots, y_m, y_{m+1}, \dots, y_n$ of $V$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-extend-to-basis
By @cor-independent-full-size, $m \le n$. If $\operatorname{span}(y_1, \dots, y_m) = V$, the list is already a basis. Otherwise pick $y_{m+1} \in V \setminus \operatorname{span}(y_1, \dots, y_m)$; by the [lemma on adding a vector](linear-independence.ipynb#lem-extend-independent), $y_1, \dots, y_{m+1}$ is independent. Repeat. Each step produces a longer independent list, and such lists have at most $n$ elements, so the process stops, and it can only stop when the list spans $V$. Its length is then $n$ by @thm-basis-cardinality. $\blacksquare$
:::

::: {#thm-subspace-dimension}
## Dimension of a Subspace

Let $V$ be finite-dimensional and $W \le V$ a subspace. Then $W$ is finite-dimensional, $\dim W \le \dim V$, and $\dim W = \dim V$ holds only if $W = V$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-subspace-dimension
Let $n = \dim V$. Start with the empty list and, as long as the current independent list $w_1, \dots, w_k$ of vectors of $W$ does not span $W$, add a vector of $W$ outside its span; by the [lemma on adding a vector](linear-independence.ipynb#lem-extend-independent) the list stays independent. Independent lists in $V$ have at most $n$ elements, so this stops after at most $n$ steps with a basis of $W$. Hence $\dim W \le n$.

If $\dim W = n$, a basis of $W$ is an independent subset of $V$ with $n$ elements, hence a basis of $V$ by @cor-independent-full-size. It spans $V$, and its span is contained in $W$, so $W = V$. $\blacksquare$
:::

## 4\. Examples

::: {#exm-standard-basis}
## The Standard Basis of $K^n$

Let $e_i \in K^n$ have $1$ in position $i$ and $0$ elsewhere. Then $(e_1, \dots, e_n)$ is a basis, so $\dim K^n = n$:

- **Spanning:** $(x_1, \dots, x_n) = x_1 e_1 + \dots + x_n e_n$.
- **Independence:** $c_1 e_1 + \dots + c_n e_n = (c_1, \dots, c_n)$, which is $0$ only if all $c_i = 0$.
:::

::: {#exm-dimension-examples}
## More Dimensions

1. **Polynomials.** $(1, x, x^2, \dots, x^d)$ is a basis of $P_d$ (it spans by definition and is [independent](linear-independence.ipynb#exm-independence-functions)), so $\dim P_d = d + 1$.
2. **Matrices.** The matrices $E_{ij}$ with a single entry $1$ in position $(i, j)$ form a basis of $M_{m \times n}(K)$, since $A = \sum_{i,j} a_{ij} E_{ij}$ and this sum is $0$ only if all $a_{ij} = 0$. So $\dim M_{m \times n}(K) = mn$.
3. **A plane.** $W = \{(x, y, z) \in \mathbb{R}^3 \mid x + y + z = 0\}$ has basis $(1, -1, 0), (0, 1, -1)$: every $(x, y, z) \in W$ equals $x\,(1, -1, 0) - z\,(0, 1, -1)$ (use $y = -x - z$), and the two vectors are not multiples of each other. So $\dim W = 2$. The vectors $(1, 0, -1), (1, -2, 1)$ form another basis of $W$; it also has two elements, as it must.
4. **The oscillator.** In the solution space $S$ of $x'' + \omega^2 x = 0$ from [Vector Spaces](vector-spaces.ipynb#prp-ode-solution-space), every solution $x$ equals $x(0) \cos \omega t + \frac{x'(0)}{\omega} \sin \omega t$: both sides solve the equation with the same initial values, so they agree by the [uniqueness theorem for ODEs](ode-and-picards-thm.ipynb). Hence $(\cos \omega t, \sin \omega t)$ spans $S$, it is independent, and $\dim S = 2$.
:::

::: {#prp-polynomials-infinite-dimensional}
## $\mathbb{R}[x]$ Is Infinite-Dimensional

The space $\mathbb{R}[x]$ of all real polynomials has no finite spanning set.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-polynomials-infinite-dimensional
Suppose $p_1, \dots, p_k$ span $\mathbb{R}[x]$ and let $d$ be the largest of their degrees. Every linear combination of the $p_i$ has degree at most $d$, so $x^{d+1}$ is not in their span, a contradiction. $\blacksquare$
:::

The figure shows example 3: the same plane $W$ with two different bases. Each basis is a pair of independent vectors in $W$, and no basis of $W$ can have a third vector.


```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6"
INK = "#0b0b0b"
plt.rcParams.update({"axes.titlesize": 11, "axes.titleweight": "bold", "font.size": 10})

bases = [((1, -1, 0), (0, 1, -1), (BLUE, ORANGE)), ((1, 0, -1), (1, -2, 1), (AQUA, YELLOW))]
fig = plt.figure(figsize=(11, 4.8))
for k, (a, b, cols) in enumerate(bases):
    ax = fig.add_subplot(1, 2, k + 1, projection="3d")
    s, t = np.meshgrid(np.linspace(-1.6, 1.6, 12), np.linspace(-1.6, 1.6, 12))
    X, Z = s, t
    ax.plot_surface(X, -X - Z, Z, color=VIOLET, alpha=0.12, linewidth=0)   # x + y + z = 0
    for v, c in zip((a, b), cols):
        ax.quiver(0, 0, 0, *v, color=c, lw=2.4, arrow_length_ratio=0.15)
        ax.text(*(1.15 * np.array(v)), str(v), color=INK, fontsize=10)
    ax.set_xlim(-1.6, 1.6); ax.set_ylim(-2, 2); ax.set_zlim(-1.6, 1.6)
    ax.set_xlabel("$x$"); ax.set_ylabel("$y$"); ax.set_zlabel("$z$")
    ax.view_init(elev=22, azim=35)
    ax.set_title(f"Basis {k + 1} of the plane $x+y+z=0$")
plt.tight_layout(); plt.show()
```
