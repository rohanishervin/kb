# Critical Points and the Second Derivative Test

**Prerequisites:** [Higher Partial Derivatives and Taylor's Theorem in Several Variables](higher-partials-and-taylor.ipynb) · [The Gradient and Level Sets](gradient.ipynb#def-critical-point) · [Orthogonality, Projections and the Gram–Schmidt Process](orthogonality-and-gram-schmidt.ipynb#thm-real-spectral) · [Local Extrema and the Mean Value Theorem](mean-value-theorem.ipynb#thm-fermat)

Let $f : S \to \mathbb{R}$, $S \subseteq \mathbb{R}^n$. A point $a \in S$ is a **local maximum** (minimum) of $f$ if $f(x) \le f(a)$ ($f(x) \ge f(a)$) for all $x \in S$ near $a$, **strict** if the inequality is strict for $x \ne a$.

## 1\. Critical Points

::: {#thm-fermat-several}
## Interior Extrema Are Critical Points

If $a$ is an interior point of $S$, a local extremum of $f$, and $f$ is differentiable at $a$, then $\nabla f(a) = 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-fermat-several
$t \mapsto f(a + te_i)$ has a local extremum at $t = 0$, so its derivative $\frac{\partial f}{\partial x_i}(a)$ vanishes by [Fermat's theorem](mean-value-theorem.ipynb#thm-fermat). $\blacksquare$
:::

A critical point that is not a local extremum is a **saddle point**. Near a critical point of a function of two variables the tangent plane of the graph is horizontal; at an extremum the graph stays (locally) on one side of it, at a saddle it crosses it.

::: {#exm-critical-points}
## The Variety of Critical Points

1. $f = x_1^2 + \dots + x_n^2$: the only critical point $0$ is a strict minimum; $-f$ has a strict maximum there.
2. $f = \sum_{i\le k}x_i^2 - \sum_{i>k}x_i^2$ ($0 < k < n$): $0$ is a saddle: a minimum along the first $k$ axes and a maximum along the others. For $n = 2$: $x^2 - y^2$.
3. $f = x^2$ on $\mathbb{R}^2$: every point of the $y$-axis is critical and a (non-strict) minimum; the level sets $f = c > 0$ are the line pairs $x = \pm\sqrt c$.
4. $f = x^2 + y^3$: $0$ is the only critical point; $f$ has a minimum at $0$ along the $x$-axis but an inflection along the $y$-axis, so $0$ is a saddle.
5. **Monkey saddle** $f = x^3 - 3xy^2 = x(x - \sqrt3y)(x + \sqrt3y)$: $\nabla f = (3x^2 - 3y^2, -6xy)$ vanishes only at $0$. The level set $f = 0$ is three lines through $0$, which cut the plane into six sectors where $f$ alternates in sign: three directions go up, three go down.
:::

## 2\. The Hessian

::: {#def-hessian}
## Hessian Matrix

For $f$ of class $C^2$ near $a$, the **Hessian** of $f$ at $a$ is the $n\times n$ matrix $Hf(a) = \Big[\dfrac{\partial^2f}{\partial x_i\partial x_j}(a)\Big]$, symmetric by [Schwarz's theorem](higher-partials-and-taylor.ipynb#thm-schwarz). Its quadratic form $h^{\mathsf T}Hf(a)h = (h\cdot\partial)^2f(a)$ is twice the degree-2 term of the [Taylor polynomial](higher-partials-and-taylor.ipynb#def-taylor-several).
:::

If $\nabla f(a) = 0$, the [Lagrange remainder](higher-partials-and-taylor.ipynb#thm-taylor-several) with $p = 1$ gives $f(a + h) - f(a) = \frac12h^{\mathsf T}Hf(a + ch)h$ for some $c \in (0, 1)$: near a critical point $f$ behaves like its Hessian form. By the [spectral theorem](orthogonality-and-gram-schmidt.ipynb#thm-real-spectral) $Hf(a)$ has real eigenvalues $\lambda_1, \dots, \lambda_n$ (with multiplicity) and an orthonormal eigenbasis.

::: {#thm-second-derivative-test}
## Second Derivative Test

Let $f$ be $C^2$ near a critical point $a$, with Hessian eigenvalues $\lambda_1, \dots, \lambda_n$.

1. If all $\lambda_i > 0$, $a$ is a strict local minimum.
2. If all $\lambda_i < 0$, $a$ is a strict local maximum.
3. If some $\lambda_i > 0$ and some $\lambda_j < 0$, $a$ is a saddle: along the line through $a$ in the direction of an eigenvector for a positive (negative) eigenvalue, $f$ has a strict local minimum (maximum) at $a$.
4. Otherwise (some $\lambda_i = 0$, the others of one sign) the test is inconclusive.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-second-derivative-test
(1) Let $m = \lambda_{\min}(Hf(a)) > 0$, so $h^{\mathsf T}Hf(a)h \ge m|h|^2$ ([quadratic form bounds](orthogonality-and-gram-schmidt.ipynb#cor-quadratic-form-bounds)). For any symmetric $M$, $|h^{\mathsf T}Mh| \le |h|\,|Mh| \le \|M\|\,|h|^2$ ([matrix bound](multivariable-chain-rule.ipynb#lem-matrix-norm)). The second partials are continuous, so $\|Hf(x) - Hf(a)\| < \frac m2$ for $x$ in a ball about $a$. For $h \ne 0$ in that ball's radius,
$$f(a + h) - f(a) = \tfrac12h^{\mathsf T}Hf(a + ch)h \ge \tfrac12\big(m|h|^2 - \tfrac m2|h|^2\big) > 0 .$$
(2) Apply (1) to $-f$.
(3) Let $Hf(a)v = \lambda v$, $|v| = 1$, and $\varphi(t) = f(a + tv)$. By the [segment derivatives](higher-partials-and-taylor.ipynb#lem-derivatives-along-segment), $\varphi'(0) = \nabla f(a)\cdot v = 0$ and $\varphi''(0) = v^{\mathsf T}Hf(a)v = \lambda$. If $\lambda > 0$, $\varphi$ has a strict local minimum at $0$, if $\lambda < 0$ a strict local maximum ([one-variable test](taylor-polynomials.ipynb#thm-higher-derivative-test)). With eigenvalues of both signs, $f$ takes values above and below $f(a)$ arbitrarily close to $a$.
(4) $x^4 + y^4$, $-(x^4 + y^4)$ and $x^4 - y^4$ all have $Hf(0) = 0$, and $0$ is a minimum, a maximum, and a saddle respectively. $\blacksquare$
:::

::: {#cor-two-variable-test}
## Two Variables

Let $f(x, y)$ be $C^2$ near a critical point $a$ and $\Delta = f_{xx}f_{yy} - f_{xy}^2$ at $a$.

1. $\Delta > 0$, $f_{xx} > 0$: strict local minimum. $\quad$ 2. $\Delta > 0$, $f_{xx} < 0$: strict local maximum.
3. $\Delta < 0$: saddle point. $\quad$ 4. $\Delta = 0$: inconclusive.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-two-variable-test
The characteristic polynomial of $Hf(a)$ is $\lambda^2 - (f_{xx} + f_{yy})\lambda + \Delta$, so $\lambda_1\lambda_2 = \Delta$ and $\lambda_1 + \lambda_2 = f_{xx} + f_{yy}$. If $\Delta > 0$ the eigenvalues have the same sign, and $f_{xx}f_{yy} > f_{xy}^2 \ge 0$ shows that $f_{xx}, f_{yy}$ have the same sign, which is then the sign of the trace, i.e. of the eigenvalues. If $\Delta < 0$ they have opposite signs. Apply @thm-second-derivative-test. $\blacksquare$
:::

::: {#exm-classify-2d}
## Classifying Critical Points in the Plane

$f = x^3 + y^3 - 3xy$: $\nabla f = (3x^2 - 3y, 3y^2 - 3x) = 0$ gives $y = x^2$, $x^4 = x$, so $(0, 0)$ and $(1, 1)$. With $f_{xx} = 6x$, $f_{yy} = 6y$, $f_{xy} = -3$: at $(0, 0)$, $\Delta = -9 < 0$, a saddle; at $(1, 1)$, $\Delta = 27 > 0$ and $f_{xx} = 6 > 0$, a strict local minimum, $f(1, 1) = -1$ (not a global one: $f(x, 0) = x^3 \to -\infty$).
:::

::: {#exm-classify-3d}
## A Function of Three Variables

$f = x^2 + 12yz + (y - z)^3$ has $\nabla f = \big(2x,\ 12z + 3(y - z)^2,\ 12y - 3(y - z)^2\big)$. Adding the last two equations gives $y = -z$, then $12y = 12y^2$: the critical points are $P = (0, 0, 0)$ and $Q = (0, 1, -1)$. With $f_{yy} = f_{zz} = 6(y - z)$, $f_{yz} = 12 - 6(y - z)$,
$$Hf(P) = \begin{bmatrix}2 & 0 & 0\\ 0 & 0 & 12\\ 0 & 12 & 0\end{bmatrix}, \qquad Hf(Q) = \begin{bmatrix}2 & 0 & 0\\ 0 & 12 & 0\\ 0 & 0 & 12\end{bmatrix} .$$
$\det(Hf(P) - \lambda I) = (2 - \lambda)(\lambda^2 - 144)$, eigenvalues $2, \pm12$: $P$ is a saddle. $Hf(Q)$ has eigenvalues $2, 12, 12 > 0$: $Q$ is a strict local minimum.
:::

```python

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})
from matplotlib.colors import TwoSlopeNorm

x, y, z = sp.symbols("x y z", real=True)
for f in [x**3 + y**3 - 3 * x * y, x**2 + 12 * y * z + (y - z)**3]:
    vs = sorted(f.free_symbols, key=lambda s: s.name)
    crit = sp.solve([sp.diff(f, v) for v in vs], vs, dict=True)
    for c in crit:
        if all(val.is_real for val in c.values()):
            H = sp.hessian(f, vs).subs(c)
            print(f, "| critical", c, "| Hessian eigenvalues", list(H.eigenvals().keys()))

fig, axs = plt.subplots(1, 3, figsize=(14, 4.4))
g = np.linspace(-1.6, 1.8, 400); X, Y = np.meshgrid(g, g)
for ax, F, tt, pts in [(axs[0], X**2 + Y**3, r"$x^2 + y^3$", [(0, 0)]),
                       (axs[1], X**3 - 3 * X * Y**2, r"monkey saddle $x^3 - 3xy^2$", [(0, 0)]),
                       (axs[2], X**3 + Y**3 - 3 * X * Y, r"$x^3 + y^3 - 3xy$", [(0, 0), (1, 1)])]:
    cf = ax.contourf(X, Y, F, levels=25, cmap="coolwarm", norm=TwoSlopeNorm(0.0))
    ax.contour(X, Y, F, levels=[0], colors=[INK], linewidths=1)
    for p in pts:
        ax.plot(*p, "o", color=YELLOW, ms=6, mec=INK)
    ax.set_aspect("equal"); ax.set_title(tt)
plt.tight_layout(); plt.show()
```
