# Linear Maps

**Prerequisites:** [Vector Spaces](vector-spaces.ipynb) · [Subspaces](subspaces.ipynb) · [Basis and Dimension](basis-and-dimension.ipynb)

A **linear map** is a function between vector spaces that respects addition and scalar multiplication. Rotations, projections, differentiation, integration and evaluation of functions are all linear. Geometrically, a linear map of the plane keeps the origin fixed and sends the square grid to a grid of parallelograms: grid lines stay straight, parallel and evenly spaced.

Throughout, $V$, $W$ and $U$ are [vector spaces](vector-spaces.ipynb) over a field $K$.

## 1\. Definition

::: {#def-linear-map}
## Linear Map

A function $f : V \to W$ is **linear** (a **linear map** or **linear transformation**) if for all $u, v \in V$ and $a \in K$

1. $f(u + v) = f(u) + f(v)$ (**additivity**), and
2. $f(a v) = a f(v)$ (**homogeneity**).

Equivalently, $f(a u + b v) = a f(u) + b f(v)$ for all $u, v \in V$ and $a, b \in K$. A linear map $f : V \to V$ from a space to itself is also called a **linear operator** on $V$.
:::

::: {#prp-linear-map-basic}
## Linear Maps Preserve Linear Combinations

If $f : V \to W$ is linear, then $f(0) = 0$, $f(-v) = -f(v)$, and
$$f(c_1 v_1 + \dots + c_k v_k) = c_1 f(v_1) + \dots + c_k f(v_k).$$
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-linear-map-basic
$f(0) = f(0 \cdot 0) = 0 \cdot f(0) = 0$ and $f(-v) = f((-1)v) = -f(v)$, using the [basic rules](vector-spaces.ipynb#prp-vector-space-consequences). The formula for $k$ terms follows from additivity and homogeneity by induction on $k$. $\blacksquare$
:::

::: {.callout-note}
## Terminology and Two Special Cases
"Linear function", "linear map", "linear mapping" and "linear transformation" are synonyms.

- When $W = K$ (the field viewed as a $1$-dimensional space over itself), a linear map $f : V \to K$ is called a [linear functional](dual-spaces.ipynb#def-linear-functional).
- A bijective linear map is a [linear isomorphism](coordinates-and-isomorphism-with-Kn.ipynb#def-linear-isomorphism).
:::

## 2\. Examples and Non-Examples

::: {#exm-linear-maps}
## Linear Maps

1. **Rotation.** The rotation of $\mathbb{R}^2$ by an angle $\alpha$ about the origin sends $(x, y)$ to
   $$(x \cos\alpha - y \sin\alpha,\ x \sin\alpha + y \cos\alpha).$$
   Each output coordinate is a combination of $x$ and $y$ with fixed coefficients, so the rotation is linear.
2. **Projection.** For $m \le n$, $(x_1, \dots, x_n) \mapsto (x_1, \dots, x_m)$ is a linear map $K^n \to K^m$.
3. **Cross product.** For fixed $a \in \mathbb{R}^3$, $v \mapsto a \times v$ is linear, because the cross product is distributive and $a \times (c v) = c\,(a \times v)$. In coordinates, $a \times v = (a_2 v_3 - a_3 v_2,\ a_3 v_1 - a_1 v_3,\ a_1 v_2 - a_2 v_1)$.
4. **Differentiation.** $D : P_d \to P_{d-1}$, $D(p) = p'$, is linear: $(ap + q)' = a p' + q'$.
5. **Integration and evaluation.** $C([a, b]) \to \mathbb{R}$, $g \mapsto \int_a^b g(x)\,dx$, and $\mathbb{R}^X \to \mathbb{R}$, $g \mapsto g(x_0)$, are linear.
6. **Matrices.** Every $A \in M_{m \times n}(K)$ defines a linear map $K^n \to K^m$, $x \mapsto A x$; this is the prototype of all linear maps between finite-dimensional spaces (see [The Matrix of a Linear Map](matrix-of-a-linear-map.ipynb)).
:::

::: {.callout-warning}
## Not Linear
A **translation** $v \mapsto v + a$ with $a \ne 0$ is not linear, since it moves $0$. The map $x \mapsto x^2$ on $\mathbb{R}$ is not linear, since $(2x)^2 \ne 2 x^2$. And $x \mapsto 3x + 1$, although its graph is a line, is not linear in this sense: it is linear plus a translation (an *affine* map).
:::

The animation shows three linear maps of the plane acting on the grid. The blue and orange arrows are the images $f(e_1)$ and $f(e_2)$ of the standard basis vectors, and the yellow arrow is $f(v)$ for $v = (1, 1)$. In each case the origin stays fixed and the grid stays a grid. The third map, the projection onto the line spanned by $(2, 1)$, squashes the whole plane onto that line.


```python

from la_manim import *

class ThreeLinearMaps(GridScene):
    def construct(self):
        maps = [
            (r"\text{rotation by }60^\circ", [[np.cos(PI / 3), -np.sin(PI / 3)], [np.sin(PI / 3), np.cos(PI / 3)]]),
            (r"\text{shear }\begin{pmatrix}1&1\\0&1\end{pmatrix}", [[1, 1], [0, 1]]),
            (r"\text{projection }\begin{pmatrix}0.8&0.4\\0.4&0.2\end{pmatrix}", [[0.8, 0.4], [0.4, 0.2]]),
        ]
        for k, (name, M) in enumerate(maps):
            e1 = self.vec([1, 0], C_BLUE, r"f(e_1)", DR)
            e2 = self.vec([0, 1], C_ORANGE, r"f(e_2)", UL)
            v = self.vec([1, 1], C_YELLOW, r"f(v)", UR)
            cap = caption(name)
            self.play(FadeIn(cap), run_time=0.6)
            self.wait(0.4)
            self.transform(M, run_time=2.5)
            self.wait(1.2)
            if k < len(maps) - 1:
                self.play(FadeOut(cap), FadeOut(e1), FadeOut(e2), FadeOut(v), run_time=0.6)
                self.vectors = []
                self.reset_grid()
        self.wait(1)

show(ThreeLinearMaps)
```


## 3\. A Linear Map Is Determined by a Basis

::: {#thm-linear-extension}
## Linear Extension

Let $(v_1, \dots, v_n)$ be a [basis](basis-and-dimension.ipynb#def-basis) of $V$ and let $w_1, \dots, w_n \in W$ be **arbitrary** vectors. There is exactly one linear map $f : V \to W$ with $f(v_i) = w_i$ for $i = 1, \dots, n$, namely
$$f(c_1 v_1 + \dots + c_n v_n) = c_1 w_1 + \dots + c_n w_n.$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-linear-extension
**Uniqueness.** If $f$ is linear with $f(v_i) = w_i$, then for $v = \sum c_i v_i$, [linearity](#prp-linear-map-basic) forces $f(v) = \sum c_i f(v_i) = \sum c_i w_i$.

**Existence.** Every $v \in V$ has [unique coordinates](coordinates-and-isomorphism-with-Kn.ipynb#prp-unique-representation) $c_i$, so the displayed formula defines a function $f$. If $v = \sum c_i v_i$ and $v' = \sum c_i' v_i$, then $a v + v' = \sum (a c_i + c_i') v_i$, hence
$$f(a v + v') = \sum (a c_i + c_i') w_i = a \sum c_i w_i + \sum c_i' w_i = a f(v) + f(v').$$
Finally $f(v_i) = w_i$ because $v_i$ has coordinates $e_i$. $\blacksquare$
:::

::: {.callout-note}
## Intuition
To specify a linear map it suffices to say where the basis vectors go, and there is no restriction on that choice. In the animation, the whole deformed grid is determined by the two arrows $f(e_1)$ and $f(e_2)$.
:::

## 4\. Images and Preimages of Subspaces

::: {#thm-subspace-image}
## Image of a Subspace

If $f : V \to W$ is linear and $V' \le V$, then $f(V') = \{f(x) \mid x \in V'\}$ is a subspace of $W$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-subspace-image
$0 = f(0) \in f(V')$. If $w = f(x)$ and $w' = f(x')$ with $x, x' \in V'$, then $a w + w' = f(a x + x')$ with $a x + x' \in V'$. $\blacksquare$
:::

::: {#thm-subspace-preimage}
## Preimage of a Subspace

If $f : V \to W$ is linear and $W' \le W$, then $f^{-1}(W') = \{x \in V \mid f(x) \in W'\}$ is a subspace of $V$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-subspace-preimage
$f(0) = 0 \in W'$, so $0 \in f^{-1}(W')$. If $f(x), f(x') \in W'$, then $f(a x + x') = a f(x) + f(x') \in W'$. $\blacksquare$
:::

::: {.callout-note}
## Intuition
A linear map sends lines and planes through the origin to lines and planes through the origin (or squashes them to something smaller), never to curved or shifted sets. The preimage statement is the less obvious one: **all** points that land in a subspace of $W$ form a subspace of $V$. The two most important cases, the preimage of $\{0\}$ and the image of $V$, are the [kernel and image](kernel-image-and-rank-nullity.ipynb#def-kernel-image) of $f$.
:::

## 5\. The Space of Linear Maps

::: {#def-hom-space}
## The Space $\mathcal{L}(V, W)$

$\mathcal{L}(V, W)$ denotes the set of all linear maps $V \to W$, with pointwise operations
$$(f + g)(v) = f(v) + g(v), \qquad (a f)(v) = a\, f(v).$$
We write $\operatorname{End}(V) = \mathcal{L}(V, V)$.
:::

::: {#prp-hom-is-vector-space}
## $\mathcal{L}(V, W)$ Is a Vector Space of Dimension $mn$

$\mathcal{L}(V, W)$ is a vector space; its zero vector is the zero map. If $\dim V = n$ and $\dim W = m$, then $\dim \mathcal{L}(V, W) = mn$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-hom-is-vector-space
**Closure.** If $f, g$ are linear, then
$$(f + g)(a u + v) = f(a u + v) + g(a u + v) = a\,(f + g)(u) + (f + g)(v),$$
and similarly for $c f$. So $\mathcal{L}(V, W)$ is a [subspace](subspaces.ipynb#prp-subspace-criterion) of the function space $W^V$, hence a vector space.

**Dimension.** Fix bases $(v_1, \dots, v_n)$ of $V$ and $(w_1, \dots, w_m)$ of $W$. By @thm-linear-extension there is a unique linear map $E_{ij}$ with
$$E_{ij}(v_k) = \begin{cases} w_i & k = j, \\ 0 & k \ne j, \end{cases} \qquad (1 \le i \le m,\ 1 \le j \le n).$$
*Spanning:* given $f$, write $f(v_j) = \sum_i a_{ij} w_i$. Then $g = \sum_{i,j} a_{ij} E_{ij}$ satisfies $g(v_k) = \sum_i a_{ik} w_i = f(v_k)$ for every $k$, so $g = f$ by uniqueness in @thm-linear-extension.
*Independence:* if $\sum_{i,j} a_{ij} E_{ij} = 0$, applying it to $v_k$ gives $\sum_i a_{ik} w_i = 0$, so all $a_{ik} = 0$.
Hence the $mn$ maps $E_{ij}$ form a basis. $\blacksquare$
:::

::: {#prp-composition-linearity}
## Composites of Linear Maps

If $f \in \mathcal{L}(V, W)$ and $g \in \mathcal{L}(W, U)$, then $g \circ f \in \mathcal{L}(V, U)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-composition-linearity
$(g \circ f)(a u + v) = g(a f(u) + f(v)) = a\, g(f(u)) + g(f(v))$. $\blacksquare$
:::

::: {#prp-endomorphism-algebra}
## The Algebra of Operators

On $\operatorname{End}(V)$, composition is associative, has the identity map $\mathrm{id}_V$ as neutral element, and distributes over addition:
$$h \circ (f + g) = h \circ f + h \circ g, \qquad (f + g) \circ h = f \circ h + g \circ h,$$
and $(a f) \circ g = a (f \circ g) = f \circ (a g)$. So $\operatorname{End}(V)$ is a [ring](rings-and-polynomial-rings.ipynb#def-ring) (indeed an associative algebra over $K$) with composition as multiplication.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-endomorphism-algebra
Associativity and the identity hold for composition of any functions. For distributivity evaluate at $v$: $h((f + g)(v)) = h(f(v) + g(v)) = h(f(v)) + h(g(v))$ uses linearity of $h$, while $((f + g) \circ h)(v) = f(h(v)) + g(h(v))$ is the definition of the sum. The scalar rules are checked the same way. $\blacksquare$
:::

::: {.callout-warning}
## Composition Is Not Commutative
In general $f \circ g \ne g \circ f$. In $\mathbb{R}^2$, let $r$ be the rotation by $90^\circ$ and $s$ the reflection $(x, y) \mapsto (x, -y)$. Then $r(s(1, 0)) = r(1, 0) = (0, 1)$ but $s(r(1, 0)) = s(0, 1) = (0, -1)$.
:::
