# Absolute and Conditional Convergence

**Prerequisites:** [Infinite Series and Convergence Tests](series.ipynb) · [The Limit of a Sequence](sequence-limit.ipynb#thm-complex-limit-parts)

## 1\. Absolute Convergence

::: {#def-absolute-convergence}
## Absolute and Conditional Convergence

A real or complex series $\sum a_n$ is **absolutely convergent** if $\sum |a_n|$ converges, and **conditionally convergent** if it converges but not absolutely.
:::

::: {#thm-absolute-implies-convergent}
## Absolute Convergence Implies Convergence

If $\sum |a_n|$ converges, then $\sum a_n$ converges and $\big|\sum a_n\big| \le \sum |a_n|$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-absolute-implies-convergent
*Real terms.* $0 \le a_n + |a_n| \le 2|a_n|$, so $\sum (a_n + |a_n|)$ converges by [comparison](series.ipynb#thm-comparison-test), and $\sum a_n = \sum (a_n + |a_n|) - \sum |a_n|$ converges.

*Complex terms.* $|\operatorname{Re} a_n|, |\operatorname{Im} a_n| \le |a_n|$, so the real and imaginary parts converge absolutely, hence converge, and the partial sums converge [componentwise](sequence-limit.ipynb#thm-complex-limit-parts).

The inequality: $|s_n| \le \sum_{k \le n} |a_k|$ by the triangle inequality; let $n \to \infty$. $\blacksquare$
:::

Consequently the [ratio and root tests](series.ipynb#thm-ratio-test) apply to complex series through $|a_n|$: if $|a_{n+1}/a_n| \to L < 1$ or $\sqrt[n]{|a_n|} \to L < 1$, the series converges absolutely. For example $\sum \frac{1}{(n + i)^2}$ converges absolutely, since $|n + i|^2 = n^2 + 1$ gives $\big|\frac{1}{(n+i)^2}\big| < \frac{1}{n^2}$.

## 2\. Alternating Series

::: {#thm-alternating-series}
## Leibniz Test

If $a_1 \ge a_2 \ge \dots \ge 0$ and $a_n \to 0$, then $\sum_{n \ge 1} (-1)^{n+1} a_n = a_1 - a_2 + a_3 - \dots$ converges, and its sum $S$ satisfies $|S - s_n| \le a_{n+1}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-alternating-series
$s_{2n+2} = s_{2n} + (a_{2n+1} - a_{2n+2}) \ge s_{2n}$ and $s_{2n+1} = s_{2n-1} - (a_{2n} - a_{2n+1}) \le s_{2n-1}$. Also $s_{2n} \le s_{2n} + a_{2n+1} = s_{2n+1} \le s_1$. So $(s_{2n})$ increases and is bounded above, $(s_{2n+1})$ decreases and is bounded below; both converge ([monotone convergence](sequence-limit.ipynb#thm-monotone-convergence)), and their limits agree because $s_{2n+1} - s_{2n} = a_{2n+1} \to 0$. Hence $s_n \to S$. Finally $s_{2n} \le S \le s_{2n+1}$ for all $n$, so $S$ lies between any two consecutive partial sums, which differ by $a_{n+1}$. $\blacksquare$
:::

For example the **alternating harmonic series** $1 - \frac12 + \frac13 - \dots$ converges, but only conditionally. (Its sum is $\ln 2$; see [power series](power-series.ipynb#exm-log-series).)

## 3\. Rearrangements

A **rearrangement** of $\sum a_n$ is $\sum a_{\sigma(n)}$ for a bijection $\sigma : \mathbb{N} \to \mathbb{N}$.

::: {#thm-rearrangement-absolute}
## Absolutely Convergent Series Can Be Rearranged

If $\sum a_n$ converges absolutely with sum $S$, every rearrangement converges absolutely with the same sum $S$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-rearrangement-absolute
Partial sums of $\sum |a_{\sigma(n)}|$ are bounded by $\sum |a_n|$, so the rearrangement converges absolutely. Given $\varepsilon > 0$, choose $N$ with $\sum_{n > N} |a_n| < \varepsilon$, and $M$ with $\{1, \dots, N\} \subseteq \{\sigma(1), \dots, \sigma(M)\}$. For $m \ge M$, the difference $\sum_{k \le m} a_{\sigma(k)} - \sum_{n \le N} a_n$ is a sum of distinct terms $a_n$ with $n > N$, so its modulus is $< \varepsilon$. Since also $|S - \sum_{n \le N} a_n| < \varepsilon$, the rearranged partial sums are within $2\varepsilon$ of $S$ for $m \ge M$. $\blacksquare$
:::

::: {#thm-riemann-rearrangement}
## Riemann's Rearrangement Theorem

If a real series $\sum a_n$ converges conditionally, then for every $L \in \mathbb{R}$ some rearrangement converges to $L$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-riemann-rearrangement
Let $p_1, p_2, \dots$ be the terms $a_n \ge 0$ and $q_1, q_2, \dots$ the absolute values of the terms $a_n < 0$, each list in its original order. If $\sum p_k$ and $\sum q_k$ both converged, $\sum |a_n|$ would converge; if exactly one converged, $\sum a_n$ would diverge. So $\sum p_k = \sum q_k = +\infty$; and $p_k, q_k \to 0$ since $a_n \to 0$.

Take $p$'s in order until the partial sum exceeds $L$ (possible, as $\sum p_k = \infty$); then $q$'s until it drops below $L$; then $p$'s again, and so on. Every run is nonempty and every term is eventually used, so this is a rearrangement. A run stops as soon as it crosses $L$, so the sum at the end of a run differs from $L$ by at most the last term added. Within a run the partial sums move monotonically from the previous end value toward $L$ and just past it. Hence, after the first run, every partial sum differs from $L$ by at most the larger of the last terms of the current and the previous run. These terms tend to $0$, so the partial sums tend to $L$. $\blacksquare$
:::

For instance, taking from $1 - \frac12 + \frac13 - \dots$ positive terms until the sum passes $2$, then one negative term, and repeating, gives a rearrangement with sum $2$.

## 4\. Double Series and the Cauchy Product

::: {#lem-double-series}
## Summing a Double Series in Either Order

Let $a_{jk}$ ($j, k \ge 0$) be real or complex with $\sum_{j \le J,\, k \le K} |a_{jk}| \le B$ for all $J, K$. Then all the series below converge absolutely and
$$\sum_{j} \Big(\sum_k a_{jk}\Big) = \sum_k \Big(\sum_j a_{jk}\Big) = \sum_{m} a_{\tau(m)}$$
for every enumeration $\tau : \mathbb{N} \to \mathbb{N}_0^2$ of the index pairs.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-double-series
Any finite set of pairs lies in a rectangle, so $\sum_m |a_{\tau(m)}| \le B$: the enumerated series converges absolutely, to a sum $T$ independent of $\tau$ by @thm-rearrangement-absolute. Each row $\sum_k a_{jk}$ converges absolutely (bounded partial sums). Given $\varepsilon$, choose a finite set $F$ of pairs with $\sum_{(j,k) \notin F} |a_{jk}| < \varepsilon$, and $J_0, K_0$ with $F \subseteq [0, J_0] \times [0, K_0]$. For $J \ge J_0$, $K \ge K_0$, the rectangle sum $R_{J,K} = \sum_{j \le J, k \le K} a_{jk}$ satisfies
$$|T - R_{J,K}| \le \sum_{(j,k) \notin F} |a_{jk}| < \varepsilon$$
(enumerate the rectangle first; $T - R_{J,K}$ is the sum of the remaining terms). Letting $K \to \infty$ gives $\big|T - \sum_{j \le J}\sum_k a_{jk}\big| \le \varepsilon$ for all $J \ge J_0$. So the row sums add up to $T$; the columns are the same with $j, k$ exchanged. $\blacksquare$
:::

::: {#thm-cauchy-product}
## Cauchy Product

Let $\sum_{n \ge 0} a_n = A$ and $\sum_{n \ge 0} b_n = B$ converge absolutely, and $c_n = \sum_{k=0}^n a_k b_{n-k}$. Then $\sum c_n$ converges absolutely and $\sum_{n\ge0} c_n = AB$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-cauchy-product
Apply @lem-double-series to $a_{jk} = a_j b_k$: finite sums satisfy $\sum_{j \le J, k \le K} |a_j||b_k| \le \sum |a_j| \sum |b_k|$. Summing by rows gives $\sum_j a_j B = AB$. The enumeration that runs through the diagonals $j + k = 0, 1, 2, \dots$ has partial sums $c_0 + \dots + c_n$ along a subsequence, so $\sum c_n = AB$; and $|c_n| \le \sum_k |a_k||b_{n-k}|$ gives absolute convergence. $\blacksquare$
:::

::: {#exm-cauchy-product}
## Examples

1. For $|x| < 1$: $\Big(\sum_{n\ge0} x^n\Big)^2 = \sum_{n \ge 0} (n+1)x^n$, i.e. $\sum (n+1)x^n = \frac{1}{(1-x)^2}$.
2. With $E(x) = \sum_{n \ge 0} \frac{x^n}{n!}$ (absolutely convergent for all real or complex $x$), the binomial theorem gives $\sum_{k} \frac{x^k}{k!}\frac{y^{n-k}}{(n-k)!} = \frac{(x+y)^n}{n!}$, hence $E(x)E(y) = E(x+y)$.
:::
