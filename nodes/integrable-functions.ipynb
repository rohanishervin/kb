# Integrable Functions and Linearity of the Integral

**Prerequisites:** [Fatou's Lemma and the Monotone Convergence Theorem](fatous-lemma-and-convergence-theorems.ipynb) · [Measurable Functions](measurable-functions.ipynb) · [The Integral of a Nonnegative Measurable Function](lebesgue-integral.ipynb)

So far the integral is defined for nonnegative measurable functions. A general measurable function, taking both signs, is split into its positive and negative parts, each of which is integrated separately. This note defines the general integral and the integrable functions, proves that the integral is linear, monotone and additive over domains, and shows that integrating a nonnegative function over sets produces a new measure. Throughout, $(X, \mathcal{M}, \mu)$ is a measure space.

## 1\. Positive and Negative Parts

::: {#def-positive-negative-parts}
## Positive and Negative Parts

For $f : X \to \overline{\mathbb{R}}$,
$$f^+ = \max(f, 0), \qquad f^- = \max(-f, 0).$$
:::

Both are nonnegative, and [measurable when $f$ is](measurable-functions.ipynb#cor-positive-negative-parts-measurable). The positive part keeps the parts of the graph above the axis and replaces the rest by $0$; the negative part reflects the parts below the axis upwards. At each point at least one of them is $0$, and
$$f = f^+ - f^-, \qquad |f| = f^+ + f^-, \qquad f^+ \le |f|, \quad f^- \le |f|.$$

## 2\. The General Integral

::: {#def-general-integral}
## Integral of a Measurable Function

For measurable $f : X \to \overline{\mathbb{R}}$,
$$\int_X f\, d\mu = \int_X f^+\, d\mu - \int_X f^-\, d\mu,$$
provided at least one of the two integrals on the right is finite. If both are $\infty$, the expression $\infty - \infty$ is undefined and $f$ **has no integral**. For $E \in \mathcal{M}$, $\int_E f\, d\mu = \int_X f \mathbf{1}_E\, d\mu$.
:::

So the integral is $+\infty$ if $\int f^+ = \infty > \int f^-$, it is $-\infty$ if $\int f^- = \infty > \int f^+$, and it is a real number if both are finite.

Why not define the integral of a signed function directly as a supremum over simple functions below it, as for nonnegative functions? Where $f$ is negative, the simple functions below $f$ are negative too, and the natural approximation there is from *above*, by an infimum over simple functions above $f$: the process is reversed. Splitting $f$ into two nonnegative parts, each handled by the supremum construction, avoids this.

::: {#def-integrable-function}
## Integrable Function

A measurable $f$ is **integrable** (over $X$) if $\int_X f^+\, d\mu < \infty$ and $\int_X f^-\, d\mu < \infty$; equivalently, if
$$\int_X |f|\, d\mu < \infty.$$
:::

The two conditions are equivalent because $\int |f| = \int (f^+ + f^-) = \int f^+ + \int f^-$ by [additivity for nonnegative functions](fatous-lemma-and-convergence-theorems.ipynb#thm-additivity-nonnegative). Many books take $\int |f| < \infty$ as the definition.

::: {.callout-warning}
## Integrability Means Absolute Integrability
A Lebesgue integrable function has $\int |f| < \infty$. Improper Riemann integrals that converge only conditionally are not Lebesgue integrals. For example, $f(x) = \sin x / x$ on $[1, \infty)$ has a convergent improper Riemann integral, but on each interval $[k\pi, (k+1)\pi]$ we have $\int |f|\, dm \ge \frac{2}{(k+1)\pi}$, and the harmonic series diverges; so $\int |f| = \infty$. The same estimate on the intervals where $\sin > 0$ and where $\sin < 0$ shows $\int f^+ = \int f^- = \infty$: $f$ has no Lebesgue integral at all.
:::

::: {#prp-integrable-finite-ae}
## Integrable Functions Are Finite Almost Everywhere

If $f$ is integrable, then $|f| < \infty$ a.e.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-integrable-finite-ae
Apply [finiteness](lebesgue-integral.ipynb#prp-finite-zero-integral) (part 1) to $|f|$. $\blacksquare$
:::

Since integrals do not see null sets, an integrable function may therefore be replaced by a real-valued one: $f \mathbf{1}_{\{|f| < \infty\}}$ is measurable, agrees with $f$ a.e., and has the same integrals of its positive and negative parts ([null sets](lebesgue-integral.ipynb#prp-nonnegative-integral-properties)).

::: {#prp-integral-triangle-inequality}
## Triangle Inequality and Comparison Test

1. If $f$ is integrable, then $\displaystyle\Big|\int_X f\, d\mu\Big| \le \int_X |f|\, d\mu$.
2. If $f$ is measurable, $g$ is integrable and $|f| \le g$ a.e., then $f$ is integrable and $\displaystyle\Big|\int f\, d\mu\Big| \le \int |f|\, d\mu \le \int g\, d\mu$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-integral-triangle-inequality

1. By definition and the triangle inequality for the two nonnegative numbers $a = \int f^+$ and $b = \int f^-$, $|a - b| \le a + b$:
$$\Big|\int f\Big| = \Big|\int f^+ - \int f^-\Big| \le \int f^+ + \int f^- = \int (f^+ + f^-) = \int |f|,$$
the last step by additivity for nonnegative functions.
2. By [monotonicity](lebesgue-integral.ipynb#prp-nonnegative-integral-properties), $\int |f| \le \int g < \infty$, so $f$ is integrable; then apply part 1. $\blacksquare$
:::

Part 2 is the "static" version of the [dominated convergence theorem](dominated-convergence-theorem.ipynb): there the same kind of bound $|f_n| \le g$ is imposed on a whole sequence.

## 3\. Linearity, Monotonicity, Additivity

::: {#thm-integral-linear}
## Properties of the Integral

Let $f, g$ be integrable.

1. **Linearity:** for $\alpha, \beta \in \mathbb{R}$, $\alpha f + \beta g$ is integrable and $\int (\alpha f + \beta g)\, d\mu = \alpha \int f\, d\mu + \beta \int g\, d\mu$.
2. **Monotonicity:** if $f \le g$ a.e., then $\int f\, d\mu \le \int g\, d\mu$.
3. **Additivity over domains:** if $A, B \in \mathcal{M}$ are disjoint, then $\int_{A \cup B} f\, d\mu = \int_A f\, d\mu + \int_B f\, d\mu$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-integral-linear
By @prp-integrable-finite-ae and the remark after it, we may assume $f$ and $g$ are real-valued: changing them on a null set changes no integral, and now $f + g$ is defined everywhere.

**3** (from 1). Since $A \cap B = \emptyset$, $\mathbf{1}_{A \cup B} = \mathbf{1}_A + \mathbf{1}_B$, so by part 1
$$\int_{A \cup B} f = \int f \mathbf{1}_{A \cup B} = \int (f \mathbf{1}_A + f \mathbf{1}_B) = \int f \mathbf{1}_A + \int f \mathbf{1}_B = \int_A f + \int_B f .$$
($f\mathbf{1}_A$ is integrable since $|f \mathbf{1}_A| \le |f|$.)

**2** (from 1). Changing $f$ and $g$ on the null set where $f \le g$ fails, we may assume $f \le g$ everywhere. Then $g - f \ge 0$, and by part 1
$$\int g = \int \big(f + (g - f)\big) = \int f + \int (g - f) \ge \int f,$$
since a nonnegative function has $(g - f)^+ = g - f$ and $(g-f)^- = 0$, so its integral is the integral of a nonnegative function, which is $\ge 0$.

**1.** *Integrability.* $\alpha f + \beta g$ is measurable ([operations](measurable-functions.ipynb#thm-measurable-algebra)), and $|\alpha f + \beta g| \le |\alpha| |f| + |\beta| |g|$, whose integral is $|\alpha| \int |f| + |\beta| \int |g| < \infty$ by additivity for nonnegative functions; so $\alpha f + \beta g$ is integrable.

*Scalars.* If $\alpha \ge 0$, then $(\alpha f)^+ = \alpha f^+$ and $(\alpha f)^- = \alpha f^-$; since constants come out of integrals of [nonnegative functions](lebesgue-integral.ipynb#prp-nonnegative-integral-properties), $\int \alpha f = \alpha \int f^+ - \alpha \int f^- = \alpha \int f$. If $\alpha < 0$, the parts swap: $(\alpha f)^+ = (-\alpha) f^-$ and $(\alpha f)^- = (-\alpha) f^+$, so $\int \alpha f = (-\alpha)\int f^- - (-\alpha) \int f^+ = \alpha \int f$.

*Sums.* It remains to show $\int (f + g) = \int f + \int g$. The difficulty is that $(f + g)^+$ is **not** $f^+ + g^+$ in general: where $f > 0 > g$, the sum may be positive while $g^+ = 0$. The trick moves the problem to nonnegative functions. By definition,
$$(f + g)^+ - (f + g)^- = f + g = (f^+ - f^-) + (g^+ - g^-).$$
Move the negative terms to the other side, so that both sides are sums of nonnegative functions:
$$(f + g)^+ + f^- + g^- = (f + g)^- + f^+ + g^+ .$$
Integrate both sides; by additivity for nonnegative functions,
$$\int (f+g)^+ + \int f^- + \int g^- = \int (f+g)^- + \int f^+ + \int g^+ .$$
All six numbers are finite, so they can be rearranged:
$$\int (f+g)^+ - \int (f+g)^- = \Big(\int f^+ - \int f^-\Big) + \Big(\int g^+ - \int g^-\Big),$$
which says $\int (f + g) = \int f + \int g$. $\blacksquare$
:::

The trick puts the whole difficulty on the additivity of the integral for nonnegative functions, which was proved with the monotone convergence theorem. Linearity makes the integrable functions a vector space and $f \mapsto \int f\, d\mu$ a linear functional on it, the starting point of the theory of $L^p$ spaces.

::: {#exm-continuous-signed}
## Continuous Functions of Any Sign

Let $f : [a, b] \to \mathbb{R}$ be continuous. Then $|f| \le M$ for some $M$, so $\int_{[a,b]} |f|\, dm \le M (b - a) < \infty$ and $f$ is integrable. Its parts $f^+$ and $f^-$ are continuous and nonnegative, so by [Riemann = Lebesgue](riemann-and-lebesgue-integrals.ipynb#thm-riemann-equals-lebesgue) and the linearity of the Riemann integral,
$$\int_{[a,b]} f\, dm = \int_{[a,b]} f^+ dm - \int_{[a,b]} f^- dm = \mathcal{R}(f^+) - \mathcal{R}(f^-) = \mathcal{R}(f).$$
For instance, for $f = \sin$ on $[0, \frac{3\pi}{2}]$: $f^+ = \sin$ on $[0, \pi]$ with integral $2$, $f^- = -\sin$ on $[\pi, \frac{3\pi}2]$ with integral $1$, and $\int f\, dm = 2 - 1 = 1 = \big[-\cos x\big]_0^{3\pi/2}$.
:::


## 4\. The Indefinite Integral Is a Measure

The integral was built from a measure; conversely, an integral defines new measures.

::: {#thm-indefinite-integral-measure}
## Integrating a Nonnegative Function Gives a Measure

Let $f : X \to [0, \infty]$ be measurable and define, for $A \in \mathcal{M}$,
$$\nu(A) = \int_A f\, d\mu .$$
Then $\nu$ is a measure on $(X, \mathcal{M})$.
:::

$f$ is required to be nonnegative only so that $\nu \ge 0$, as a measure must be.

::: {.callout-tip collapse="true"}
## Proof of @thm-indefinite-integral-measure
$\nu(A) \ge 0$ because $f \ge 0$, and $\nu(\emptyset) = \int f \mathbf{1}_\emptyset = \int 0 = 0$. The main point is countable additivity: for pairwise disjoint $E_1, E_2, \dots \in \mathcal{M}$ with union $E$, we must show
$$\int_X f \mathbf{1}_E \, d\mu = \sum_{i=1}^\infty \int_X f \mathbf{1}_{E_i}\, d\mu .$$
This asks to move an infinite sum out of an integral. A *finite* sum can be moved out without difficulty ([additivity](fatous-lemma-and-convergence-theorems.ipynb#thm-additivity-nonnegative)), but an infinite sum is a limit, $\sum_{i=1}^\infty = \lim_n \sum_{i=1}^n$, and moving a limit out of an integral needs a convergence theorem.

The partial unions increase, which suggests the monotone convergence theorem. Let
$$g_n = \sum_{i=1}^n \mathbf{1}_{E_i}, \qquad g = \mathbf{1}_E .$$
Because the $E_i$ are disjoint, $g_n = \mathbf{1}_{E_1 \cup \dots \cup E_n}$, so $g_1 \le g_2 \le \cdots$ and $g_n \to g$ pointwise. Multiplying pointwise by the fixed function $f \ge 0$ keeps this: $f g_1 \le f g_2 \le \cdots \to f g$. By the [monotone convergence theorem](fatous-lemma-and-convergence-theorems.ipynb#thm-monotone-convergence), $\int f g_n \to \int f g = \int_E f$. On the other hand the sum in $g_n$ is finite, so by additivity
$$\int f g_n = \int \sum_{i=1}^n f \mathbf{1}_{E_i} = \sum_{i=1}^n \int_{E_i} f .$$
Letting $n \to \infty$, the partial sums converge to $\int_E f$, that is, $\sum_{i=1}^\infty \int_{E_i} f = \int_E f$. $\blacksquare$
:::

::: {.callout-note}
## Intuition: A Density
$\nu$ is a reweighted version of $\mu$: the function $f$ acts as a **density**, telling how much $\nu$-mass sits on each piece of $\mu$-mass. In probability this is exactly how a [probability density function](probability-density-function.ipynb) defines a distribution: $P(A) = \int_A f\, dm$. This is the measure-theoretic answer to the question of [how to describe a distribution](why-measure-theory.ipynb#distributions-of-a-quantity).
:::

For integrable $f$ of any sign the same countable additivity holds, although $\nu$ is then not a measure (it may be negative).

::: {#thm-integral-countably-additive}
## Countable Additivity of the Integral Over Domains

If $f$ is integrable and $X = \bigcup_{n=1}^\infty X_n$ with $X_n \in \mathcal{M}$ pairwise disjoint, then
$$\int_X f\, d\mu = \sum_{n=1}^\infty \int_{X_n} f\, d\mu,$$
and the series converges absolutely.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-integral-countably-additive
Apply @thm-indefinite-integral-measure to $f^+$ and to $f^-$: $\int_X f^\pm = \sum_n \int_{X_n} f^\pm$, and both sums are finite because $f$ is integrable. Subtracting the two convergent series gives the claim, since $\int_{X_n} f = \int_{X_n} f^+ - \int_{X_n} f^-$; and $\sum_n \big|\int_{X_n} f\big| \le \sum_n \int_{X_n} |f| = \int_X |f| < \infty$. $\blacksquare$
:::

The integral even inherits [continuity](continuity-of-measure.ipynb#thm-continuity-of-measure) from the measure.

::: {#cor-continuity-of-integration}
## Continuity of Integration

Let $f$ be integrable and $X_n \in \mathcal{M}$.

1. If $X_1 \subseteq X_2 \subseteq \cdots$, then $\displaystyle\int_{\bigcup_n X_n} f\, d\mu = \lim_{n \to \infty} \int_{X_n} f\, d\mu$.
2. If $X_1 \supseteq X_2 \supseteq \cdots$, then $\displaystyle\int_{\bigcap_n X_n} f\, d\mu = \lim_{n \to \infty} \int_{X_n} f\, d\mu$.

Part 1 also holds for every measurable $f \ge 0$, integrable or not.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-continuity-of-integration
The proofs repeat those for measures. In fact they *are* those for measures: $\nu^\pm(A) = \int_A f^\pm\, d\mu$ are measures by @thm-indefinite-integral-measure, and they are finite since $\nu^\pm(X) = \int f^\pm < \infty$. By [continuity of measure](continuity-of-measure.ipynb#thm-continuity-of-measure) (from below in part 1; from above in part 2, where finiteness is needed), $\nu^\pm(X_n) \to \nu^\pm(\bigcup X_n)$, respectively $\nu^\pm(\bigcap X_n)$. Subtract. For $f \ge 0$ in part 1 only $\nu = \nu^+$ is involved, and continuity from below needs no finiteness. $\blacksquare$
:::

The figure shows the decomposition $f = f^+ - f^-$ for $f = \sin$ on $[0, \frac{3\pi}{2}]$ and checks the computation of @exm-continuous-signed numerically.

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

x = np.linspace(0, 1.5 * np.pi, 3001)
f = np.sin(x)
fp, fm = np.maximum(f, 0), np.maximum(-f, 0)
fig, axes = plt.subplots(1, 3, figsize=(13, 3.6), sharey=True)
for ax, y, col, title in [(axes[0], f, INK, "$f = \\sin$"), (axes[1], fp, BLUE, "$f^+ = \\max(f, 0)$"),
                          (axes[2], fm, ORANGE, "$f^- = \\max(-f, 0)$")]:
    ax.axhline(0, color=MUTED, lw=0.8)
    ax.plot(x, y, color=col, lw=2)
    ax.set_title(title); ax.set_xlabel("$x$")
axes[0].fill_between(x, 0, f, where=f > 0, color=BLUE, alpha=0.3)
axes[0].fill_between(x, 0, f, where=f < 0, color=ORANGE, alpha=0.3)
axes[1].fill_between(x, 0, fp, color=BLUE, alpha=0.3)
axes[2].fill_between(x, 0, fm, color=ORANGE, alpha=0.3)
plt.tight_layout(); plt.show()
Ip, Im = np.trapezoid(fp, x), np.trapezoid(fm, x)
print(f"int f+ = {Ip:.4f},  int f- = {Im:.4f},  int f = {Ip - Im:.4f},  int |f| = {Ip + Im:.4f}")
```
