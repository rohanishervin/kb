## Notation

The same symbols mean the same thing in every note.

| Symbol | Meaning |
|---|---|
| $K$ | the field of scalars ($\mathbb{Q}$, $\mathbb{R}$, $\mathbb{C}$, ...) |
| $V, W, U$ | vector spaces over $K$; $u, v, w, x, y$ are vectors |
| $a, b, c, \lambda, \mu$ | scalars; $c_1, \dots, c_n$ coefficients of a linear combination |
| $W \le V$ | $W$ is a subspace of $V$ |
| $\operatorname{span}(S)$ | all finite linear combinations of elements of $S$ |
| $U + W$, $U \oplus W$ | sum and direct sum of subspaces |
| $V/W$, $[v] = v + W$ | quotient space and the coset of $v$ |
| $\mathcal{B} = (v_1, \dots, v_n)$ | an ordered basis; $\mathcal{B}'$ another basis of the same space; $\mathcal{C}$ a basis of $W$ |
| $e_1, \dots, e_n$ | the standard basis of $K^n$ |
| $[v]_{\mathcal{B}}$ | coordinate column of $v$ in the basis $\mathcal{B}$ |
| $f, g, h$ | linear maps; an operator is a linear map $V \to V$ |
| $\mathcal{L}(V, W)$, $\operatorname{End}(V)$ | linear maps $V \to W$; operators on $V$ |
| $\ker f$, $\operatorname{im} f$, $\operatorname{rank} f$ | kernel, image, $\dim \operatorname{im} f$ |
| $\mathrm{id}_V$, $I_n$ | identity map, identity matrix |
| $M_{m\times n}(K)$, $M_n(K)$, $GL_n(K)$ | matrices, square matrices, invertible matrices |
| $A = (a_{ij})$, $A_j$ | a matrix and its $j$-th column |
| $A^{\mathsf T}$ | transpose |
| $[f]_{\mathcal{B}}^{\mathcal{C}}$, $[f]_{\mathcal{B}}$ | matrix of $f$ from basis $\mathcal{B}$ (domain, lower index) to $\mathcal{C}$ (codomain, upper index); $[f]_{\mathcal{B}} = [f]_{\mathcal{B}}^{\mathcal{B}}$ |
| $P = [\mathrm{id}_V]_{\mathcal{B}'}^{\mathcal{B}}$ | transition matrix: columns are the old coordinates of the new basis vectors |
| $V^*$, $\varphi, \psi$ | dual space and linear functionals |
| $v_1^*, \dots, v_n^*$ | dual basis: $v_i^*(v_j) = \delta_{ij}$ |
| $W^\perp \subseteq V^*$ | annihilator of $W$ |
| $f^* : W^* \to V^*$ | dual map, $f^*(\psi) = \psi \circ f$ |
| $S_n$, $\sigma$, $\tau$, $\operatorname{sgn}$ | permutations, transpositions, sign |
| $\det$, $\operatorname{tr}$ | determinant, trace |
| $A_{ij}$, $C_{ij}$, $\operatorname{adj} A$ | $A$ without row $i$ and column $j$; cofactor $(-1)^{i+j}\det A_{ij}$; adjugate |
| $p_A(\lambda) = \det(\lambda I_n - A)$ | characteristic polynomial (monic) |
| $E_\lambda$, $m_a$, $m_g$ | eigenspace; algebraic and geometric multiplicity |
| $R$, $R[x]$, $\deg$, $E_c$ | a ring, its polynomial ring, degree, evaluation at $c$ |
| $p(A)$, $m_A$ | a polynomial evaluated at a matrix; minimal polynomial |
| $d \mid p$, $\gcd(p, q)$ | $d$ divides $p$; monic greatest common divisor |
| $W$ $f$-invariant, $f\vert_W$ | $f(W) \subseteq W$; restriction of $f$ to $W$ |
| $G_\lambda(f)$, $d(\lambda)$ | generalized eigenspace $\bigcup_m \ker(f - \lambda)^m$; index of $\lambda$ |
