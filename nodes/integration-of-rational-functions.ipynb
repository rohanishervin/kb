# Integration of Rational Functions

**Prerequisites:** [Integration by Parts and Substitution](integration-by-parts-and-substitution.ipynb) · [The Division Algorithm for Polynomials](polynomial-division-algorithm.ipynb) · [Greatest Common Divisors and Unique Factorization of Polynomials](polynomial-gcd-and-unique-factorization.ipynb) · [Complex Numbers](complex-numbers.ipynb)

Every rational function $\frac{P}{Q}$ has an antiderivative built from rational functions, $\ln$ and $\arctan$. The method has three steps.

## 1\. Reduction to Proper Fractions

Divide: $P = AQ + R$ with $\deg R < \deg Q$ ([division algorithm](polynomial-division-algorithm.ipynb#thm-polynomial-division)), so $\frac PQ = A + \frac RQ$. The polynomial $A$ is integrated termwise; it remains to integrate **proper** fractions $\frac RQ$, $R \ne 0$, $\deg R < \deg Q$.

## 2\. Partial Fractions

**Fundamental theorem of algebra** (stated without proof): every non-constant complex polynomial has a complex root, hence factors as $Q(x) = c_n(x - \alpha_1)\cdots(x - \alpha_n)$. If $Q$ has real coefficients and $Q(\alpha) = 0$, then $Q(\bar\alpha) = \overline{Q(\alpha)} = 0$, and $(x - \alpha)(x - \bar\alpha) = x^2 - 2\operatorname{Re}\alpha\,x + |\alpha|^2$ is real. Hence ([details](polynomial-gcd-and-unique-factorization.ipynb#prp-real-irreducibles))
$$Q(x) = c\,(x - \beta_1)^{\rho_1}\cdots(x - \beta_r)^{\rho_r}\,(x^2 + A_1x + B_1)^{\sigma_1}\cdots(x^2 + A_sx + B_s)^{\sigma_s},$$
with distinct real $\beta_i$ and distinct quadratics with $A_j^2 - 4B_j < 0$.

::: {#thm-partial-fractions}
## Partial Fraction Decomposition

With $Q$ factored as above and $\deg R < \deg Q$,
$$\frac{R(x)}{Q(x)} = \sum_{i=1}^{r}\sum_{k=1}^{\rho_i}\frac{c_{ik}}{(x - \beta_i)^k} + \sum_{j=1}^{s}\sum_{k=1}^{\sigma_j}\frac{D_{jk}x + E_{jk}}{(x^2 + A_jx + B_j)^k}$$
for some real constants $c_{ik}, D_{jk}, E_{jk}$ (which are in fact unique).
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-partial-fractions
*Splitting coprime factors.* If $Q = Q_1Q_2$ with $\gcd(Q_1, Q_2) = 1$, there are $a, b$ with $aQ_1 + bQ_2 = 1$ ([Bézout](polynomial-gcd-and-unique-factorization.ipynb#cor-coprime-bezout)), so $\frac{R}{Q} = \frac{Rb}{Q_1} + \frac{Ra}{Q_2}$. Divide: $Rb = sQ_1 + R_1$, $Ra = tQ_2 + R_2$ with $\deg R_i < \deg Q_i$; then $\frac RQ = (s + t) + \frac{R_1}{Q_1} + \frac{R_2}{Q_2}$. The left side and the two fractions tend to $0$ as $x \to \infty$, so the polynomial $s + t$ does too, hence $s + t = 0$. Distinct factors $(x - \beta_i)^{\rho_i}$, $(x^2 + A_jx + B_j)^{\sigma_j}$ are pairwise coprime (distinct irreducibles), so by induction $\frac RQ$ is a sum of proper fractions with denominators $u^m$, $u$ irreducible of degree $d \in \{1, 2\}$ (the constant $c$ goes into the numerators).

*One prime power.* Let $u$ be irreducible of degree $d$ and $\deg N < md$. Dividing repeatedly by $u$ ($N = q_0u + N_0$, $q_0 = q_1u + N_1$, $\dots$, with $\deg N_k < d$; the degrees of the quotients drop by $d$ each time) writes $N = N_0 + N_1u + \dots + N_{m-1}u^{m-1}$. Then $\frac{N}{u^m} = \sum_{k=0}^{m-1}\frac{N_k}{u^{m-k}}$, which has the required form. $\blacksquare$
:::

The constants are found by putting the right side over the common denominator $Q$ and comparing coefficients of $R$, or by evaluating at convenient points (@exm-cover-up).

## 3\. Integrating the Pieces

* $\displaystyle\int\frac{dx}{(x - \beta)^k} = \ln|x - \beta|$ for $k = 1$ and $\dfrac{(x - \beta)^{1-k}}{1 - k}$ for $k \ge 2$.
* For a quadratic piece, complete the square: $x^2 + Ax + B = (x + \frac A2)^2 + \kappa^2$, $\kappa^2 = B - \frac{A^2}{4} > 0$. With $t = \frac{x + A/2}{\kappa}$, $\frac{Dx + E}{(x^2 + Ax + B)^k}\,dx$ becomes a combination of $\frac{2t\,dt}{(t^2 + 1)^k}$ and $\frac{dt}{(t^2 + 1)^k}$, and
$$\int\frac{2t\,dt}{(t^2+1)^k} = \begin{cases}\ln(t^2 + 1) & k = 1\\[2pt] \dfrac{(t^2+1)^{1-k}}{1-k} & k \ge 2\end{cases}, \qquad J_k = \int\frac{dt}{(t^2+1)^k}.$$

::: {#prp-reduction-formula}
## Reduction Formula

$J_1 = \arctan t$ and for $k \ge 1$
$$J_{k+1} = \frac{t}{2k(t^2+1)^k} + \frac{2k - 1}{2k}J_k .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-reduction-formula
By parts with $f = (t^2+1)^{-k}$, $g = t$: $J_k = \frac{t}{(t^2+1)^k} + 2k\int\frac{t^2}{(t^2+1)^{k+1}}dt = \frac{t}{(t^2+1)^k} + 2k(J_k - J_{k+1})$, using $t^2 = (t^2 + 1) - 1$. Solve for $J_{k+1}$. $\blacksquare$
:::

(Equivalently, $t = \tan\theta$ turns $J_k$ into $\int\cos^{2k-2}\theta\,d\theta$.) Hence the antiderivative of every rational function is a sum of a rational function, terms $c\ln|x - \beta|$, $c\ln(x^2 + Ax + B)$ and $c\arctan\frac{x + A/2}{\kappa}$.

::: {#exm-partial-fractions}
## A Complete Example

$\displaystyle\frac{-x^7 + x^5 - x^4 + x + 1}{x^6 + x^4 + x^2} = -x + \frac{2x^5 - x^4 + x^3 + x + 1}{x^2(x^2 + x + 1)(x^2 - x + 1)}$, since $x^6 + x^4 + x^2 = x^2(x^4 + x^2 + 1)$ and $x^4 + x^2 + 1 = (x^2 + x + 1)(x^2 - x + 1)$ (both with negative discriminant). Writing the proper part as $\frac Ax + \frac{B}{x^2} + \frac{Cx + D}{x^2 + x + 1} + \frac{Ex + F}{x^2 - x + 1}$ and comparing coefficients:
$$B = 1,\ A = 1,\ B + D + F = 0,\ A - C + E = 1,\ B - D + F = -1,\ A + C + E = 2,$$
so $A = B = 1$, $C = D = E = \frac12$, $F = -\frac12$. Using $\frac{x + 1}{x^2 + x + 1} = \frac12\frac{2x + 1}{x^2 + x + 1} + \frac{1/2}{(x + \frac12)^2 + \frac34}$ and similarly for the other quadratic,
$$\int = -\frac{x^2}{2} + \ln|x| - \frac1x + \frac14\ln\big((x^2 + x + 1)(x^2 - x + 1)\big) + \frac{1}{2\sqrt3}\Big(\arctan\frac{2x + 1}{\sqrt3} - \arctan\frac{2x - 1}{\sqrt3}\Big) + C .$$
:::

::: {#exm-cover-up}
## Evaluating Instead of Comparing

$\frac{x^2 + 1}{x^3 + x^2} = \frac{A}{x + 1} + \frac{B}{x} + \frac{C}{x^2}$. Multiplying by $x + 1$ and letting $x \to -1$: $A = \lim_{x\to-1}\frac{x^2+1}{x^2} = 2$. Multiplying by $x^2$ and letting $x \to 0$: $C = 1$. Then $\frac{x^2+1}{x^3+x^2} - \frac{2}{x+1} - \frac{1}{x^2} = \frac{-1}{x}$, so $B = -1$, and
$\int\frac{x^2 + 1}{x^3 + x^2}dx = 2\ln|x + 1| - \ln|x| - \frac1x + C$.
:::

```python

import sympy as sp

x = sp.symbols("x")
f = (-x**7 + x**5 - x**4 + x + 1) / (x**6 + x**4 + x**2)
print("apart:", sp.apart(f, x))
F = (-x**2 / 2 + sp.log(x) - 1 / x + sp.Rational(1, 4) * sp.log((x**2 + x + 1) * (x**2 - x + 1))
     + (sp.atan((2*x + 1) / sp.sqrt(3)) - sp.atan((2*x - 1) / sp.sqrt(3))) / (2 * sp.sqrt(3)))
print("F' - f simplifies to:", sp.simplify(sp.diff(F, x) - f))
g = (x**2 + 1) / (x**3 + x**2)
print("apart:", sp.apart(g, x))
t, k = sp.symbols("t k", positive=True)
for kk in [1, 2, 3]:
    J = lambda m: sp.integrate(1 / (t**2 + 1)**m, t)
    rhs = t / (2*kk*(t**2 + 1)**kk) + sp.Rational(2*kk - 1, 2*kk) * J(kk)
    print(f"reduction formula k = {kk}:", sp.simplify(sp.diff(J(kk + 1) - rhs, t)) == 0)
```
