# The Riemann and Lebesgue Integrals of Continuous Functions

**Prerequisites:** [The Integral of a Nonnegative Measurable Function](lebesgue-integral.ipynb) · [Lebesgue Measure](Lebesgue-measure.ipynb) · [Uniform Continuity and Compactness](uniform-continuity.ipynb) · [Continuity and Compactness](continuity-and-compactness.ipynb)

The Lebesgue integral was promised to be a generalization of the Riemann integral, not a different integral. For continuous functions on a closed bounded interval this is proved here. Besides settling the question, it lets us compute Lebesgue integrals of continuous functions with the familiar tools of calculus, which is convenient for examples. Throughout, $m$ is Lebesgue measure on $\mathbb{R}$.

Recall the **Riemann integral** of $f : [a, b] \to \mathbb{R}$ ([The Riemann Integral](riemann-integral.ipynb#def-riemann-integral); for continuous $f$ it is the [limit of Riemann sums](properties-of-the-riemann-integral.ipynb#thm-riemann-sums)), written $\mathcal{R}(f)$:
$$\mathcal{R}(f) = \lim \sum_{i=0}^{n-1} f(x_i)\, \Delta x_i, \qquad \Delta x_i = x_{i+1} - x_i,$$
the limit being taken over partitions $P : a = x_0 < x_1 < \dots < x_n = b$ whose **mesh** $\max_i \Delta x_i$ tends to $0$. Its building blocks are step functions, and step functions are simple functions; that is the heart of the comparison. (The converse is false: simple functions may be constant on sets that are not intervals.)

::: {#thm-riemann-equals-lebesgue}
## Riemann = Lebesgue for Continuous Functions

Let $f : [a, b] \to \mathbb{R}$ be continuous and $f \ge 0$. Then
$$\int_{[a, b]} f\, dm = \mathcal{R}(f).$$
:::

The hypothesis $f \ge 0$ is there only because so far the integral is defined for nonnegative functions; the extension to all continuous functions is in the note on [integrable functions](integrable-functions.ipynb#exm-continuous-signed).

::: {.callout-tip collapse="true"}
## Proof of @thm-riemann-equals-lebesgue
Fix a partition $P$ and write $I_i = [x_i, x_{i+1})$ for $i < n - 1$ and $I_{n-1} = [x_{n-1}, x_n]$; these intervals are disjoint and cover $[a, b]$. (Using closed intervals instead would make neighbours share an endpoint, a set of measure zero, which changes nothing.)

**Step 1: $\mathcal{R}(f) \le \int_{[a,b]} f\, dm$.** The Lebesgue integral is a supremum over simple functions **below** $f$. The Riemann rectangles built on the left endpoints need not lie below the graph: where $f$ increases they stick out above it. To stay below, use on each interval the smallest value of $f$; it exists because $f$ is continuous on the closed interval $[x_i, x_{i+1}]$ ([extreme value theorem](continuity-and-compactness.ipynb#the-extreme-value-theorem-minmax-theorem)):
$$c_i = \min_{[x_i, x_{i+1}]} f = f(x_i^*), \qquad \varphi_P = \sum_{i=0}^{n-1} c_i \mathbf{1}_{I_i}.$$
Then $0 \le \varphi_P \le f$ on $[a,b]$, so $\varphi_P$ is a competitor in the supremum defining $\int_{[a,b]} f\, dm$ (as a function on $\mathbb{R}$ it is $0$ outside $[a,b]$), and
$$\int_{[a,b]} f\, dm \ge \int \varphi_P\, dm = \sum_i c_i\, m(I_i) = \sum_i c_i\, \Delta x_i. \tag{1}$$
The Riemann sum uses $f(x_i)$ instead of $f(x_i^*)$, but by continuity the two values are close. Precisely, $f$ is continuous on the compact interval $[a,b]$, hence [uniformly continuous](uniform-continuity.ipynb#compactness-guarantees-uniform-continuity): for every $\varepsilon > 0$ there is $\delta > 0$ such that $|x - y| < \delta$ implies $|f(x) - f(y)| < \varepsilon$, wherever $x$ and $y$ lie in $[a, b]$. If the mesh of $P$ is less than $\delta$, then $x_i$ and $x_i^*$ lie in the same interval of length $< \delta$, and by the triangle inequality
$$\Big|\sum_i f(x_i)\, \Delta x_i - \sum_i c_i\, \Delta x_i\Big| \le \sum_i \Delta x_i\, \big|f(x_i) - f(x_i^*)\big| < \varepsilon \sum_i \Delta x_i = \varepsilon\, (b - a). \tag{2}$$
So, by (1), every Riemann sum of mesh $< \delta$ is less than $\int_{[a,b]} f\, dm + \varepsilon (b - a)$. Letting the mesh tend to $0$ and then $\varepsilon \to 0$ gives $\mathcal{R}(f) \le \int_{[a,b]} f\, dm$.

**Step 2: $\int_{[a,b]} f\, dm \le \mathcal{R}(f)$.** Now bound $f$ from **above** by a simple function, using the largest values:
$$c_i' = \max_{[x_i, x_{i+1}]} f, \qquad \psi_P = \sum_{i=0}^{n-1} c_i' \mathbf{1}_{I_i}.$$
Then $f \mathbf{1}_{[a,b]} \le \psi_P$, and by [monotonicity of the integral](lebesgue-integral.ipynb#prp-nonnegative-integral-properties) (for the simple function $\psi_P$ the integral is the [simple integral](simple-functions.ipynb#def-simple-integral))
$$\int_{[a,b]} f\, dm \le \int \psi_P\, dm = \sum_i c_i'\, \Delta x_i .$$
The same uniform continuity estimate as in (2) shows that $\sum_i c_i' \Delta x_i$ differs from the Riemann sum $\sum_i f(x_i) \Delta x_i$ by less than $\varepsilon (b - a)$ once the mesh is below $\delta$. So $\int_{[a,b]} f\, dm$ is less than every such Riemann sum plus $\varepsilon(b-a)$; letting the mesh and then $\varepsilon$ tend to $0$ gives $\int_{[a,b]} f\, dm \le \mathcal{R}(f)$.

Steps 1 and 2 together give the equality. $\blacksquare$
:::

::: {.callout-note}
## Intuition: A Sandwich of Simple Functions
The proof uses a **lower** Riemann sum, with the minimum on each interval, and an **upper** Riemann sum, with the maximum. Each is the integral of a simple function: the lower one lies below $f$, so it bounds $\int f$ from below; the upper one lies above $f$, so it bounds $\int f$ from above. Uniform continuity makes the gap between them, at most $\varepsilon(b - a)$, tend to $0$, so $\int f\, dm$ is squeezed between quantities that converge to the Riemann integral. The argument even shows that the Riemann sums converge, since they too are squeezed between the lower and the upper sums.
:::

After this, the Lebesgue integral of a continuous function on an interval can be computed with antiderivatives: for example $\int_{[0,1]} x\, dm = \frac12$ and $\int_{[0, \pi]} \sin x\, dm = 2$.

The figure shows the lower simple function $\varphi_P$ and the upper simple function $\psi_P$ for a continuous function on $[0, 3]$, and how their integrals close in on the integral of $f$ as the partition is refined.

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

f = lambda t: 1.2 + np.sin(2 * t) + 0.35 * t
a, b = 0.0, 3.0
exact = 1.2 * 3 + (1 - np.cos(6)) / 2 + 0.35 * 9 / 2           # antiderivative

def lower_upper(n):
    edges = np.linspace(a, b, n + 1)
    lo, hi = [], []
    for x0, x1 in zip(edges, edges[1:]):
        t = np.linspace(x0, x1, 400)
        lo.append(f(t).min()); hi.append(f(t).max())
    return edges, np.array(lo), np.array(hi)

fig, (ax, ax2) = plt.subplots(1, 2, figsize=(12, 4.2))
edges, lo, hi = lower_upper(6)
xx = np.linspace(a, b, 600)
for x0, x1, l, h in zip(edges, edges[1:], lo, hi):
    ax.add_patch(plt.Rectangle((x0, 0), x1 - x0, h, facecolor=ORANGE, alpha=0.18, edgecolor=ORANGE))
    ax.add_patch(plt.Rectangle((x0, 0), x1 - x0, l, facecolor=BLUE, alpha=0.35, edgecolor=BLUE))
ax.plot(xx, f(xx), color=INK, lw=2, label="$f$")
ax.plot([], [], color=BLUE, lw=6, alpha=0.5, label=r"$\varphi_P$ (minima): below $f$")
ax.plot([], [], color=ORANGE, lw=6, alpha=0.35, label=r"$\psi_P$ (maxima): above $f$")
ax.set_xlabel("$x$"); ax.legend(loc="upper left", fontsize=9)
ax.set_title("Lower and upper simple functions, 6 intervals")

ns = np.array([2, 4, 8, 16, 32, 64, 128, 256])
L, U = [], []
for n in ns:
    e, l, h = lower_upper(n)
    L.append((l * np.diff(e)).sum()); U.append((h * np.diff(e)).sum())
ax2.plot(ns, L, "o-", color=BLUE, label=r"$\int \varphi_P\, dm$")
ax2.plot(ns, U, "o-", color=ORANGE, label=r"$\int \psi_P\, dm$")
ax2.axhline(exact, color=RED, ls="--", lw=1.2, label=f"Riemann integral = {exact:.4f}")
ax2.set_xscale("log", base=2); ax2.set_xlabel("number of intervals")
ax2.set_title(r"$\int\varphi_P \leq \int f\,dm \leq \int\psi_P$, gap $\to 0$")
ax2.legend(fontsize=9)
plt.tight_layout(); plt.show()
print("n = 256: lower =", round(L[-1], 5), " upper =", round(U[-1], 5), " exact =", round(exact, 5))
```
