# The Inverse Function Theorem and Curvilinear Coordinates

**Prerequisites:** [The Chain Rule in Several Variables](multivariable-chain-rule.ipynb#thm-mean-value-inequality) · [The Banach Contraction Principle](banach-fp-thm.ipynb) · [Completeness](completeness.ipynb) · [Laplace Expansion, Adjugate, and Cramer's Rule](laplace-expansion-adjugate-and-cramers-rule.ipynb) · [Invertible Linear Maps](invertible-linear-maps.ipynb)

Given $n$ quantities $y = f(x)$ depending on $n$ quantities $x$, when can one solve for $x$ in terms of $y$, i.e. use $y$ as new coordinates? In one variable, a $C^1$ function with $f'(a) \ne 0$ is strictly monotone near $a$ and has a differentiable [local inverse](chain-rule-and-inverse-functions.ipynb#thm-inverse-derivative) with $g' = 1/f'$. The same holds in $n$ variables with $f'(a) \ne 0$ replaced by $\det Df(a) \ne 0$, but only locally.

## 1\. The Theorem

::: {#thm-inverse-function}
## Inverse Function Theorem

Let $U \subseteq \mathbb{R}^n$ be open, $f : U \to \mathbb{R}^n$ of class $C^1$, $a \in U$, and $\det Df(a) \ne 0$. Then there are open sets $V \ni a$ and $W \ni f(a)$ such that $f$ maps $V$ bijectively onto $W$, and the inverse $g : W \to V$ is of class $C^1$ with
$$Dg(y) = \big[Df(g(y))\big]^{-1} \qquad (y \in W).$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-inverse-function
*Reduction.* Let $A = Df(a)$ and $\tilde f(x) = A^{-1}\big(f(a + x) - f(a)\big)$. Then $\tilde f(0) = 0$, $D\tilde f(0) = I$, and the theorem for $\tilde f$ at $0$ gives it for $f$ at $a$ (compose with the affine bijections $x \mapsto a + x$ and $z \mapsto f(a) + Az$). So assume $a = 0$, $f(0) = 0$, $Df(0) = I$.

*Step 1: $f$ is injective near $0$.* The partials are continuous, so there is $r > 0$ with $\overline B_r = \{|x| \le r\} \subseteq U$ and $\|Df(x) - I\| \le \frac12$ for $x \in \overline B_r$ (norm of [this lemma](multivariable-chain-rule.ipynb#lem-matrix-norm)). Let $\varphi(x) = x - f(x)$; then $\|D\varphi\| \le \frac12$ on the convex set $\overline B_r$, so by the [mean value inequality](multivariable-chain-rule.ipynb#thm-mean-value-inequality) $|\varphi(x) - \varphi(x')| \le \frac12|x - x'|$, and
$$|f(x) - f(x')| \ge |x - x'| - |\varphi(x) - \varphi(x')| \ge \tfrac12|x - x'| \qquad (x, x' \in \overline B_r). \tag{1}$$

*Step 2: $f$ hits every point near $0$.* Let $|y| < \frac r2$ and $T_y(x) = x - f(x) + y = \varphi(x) + y$. For $x \in \overline B_r$, $|T_y(x)| \le |\varphi(x) - \varphi(0)| + |y| < \frac r2 + \frac r2 = r$, so $T_y$ maps the complete metric space $\overline B_r$ (closed in $\mathbb{R}^n$) into itself and is a contraction with constant $\frac12$. By the [contraction principle](banach-fp-thm.ipynb) it has a unique fixed point $x$, i.e. $f(x) = y$, and $|x| < r$. Put $W = \{|y| < \frac r2\}$ and $V = B_r\cap f^{-1}(W)$, open since $f$ is continuous. Then $f : V \to W$ is a bijection; let $g$ be its inverse. By (1), $|g(y) - g(y')| \le 2|y - y'|$, so $g$ is continuous.

*Step 3: $g$ is differentiable.* For $x \in \overline B_r$, $M = Df(x)$ is invertible: $Mh = 0$ gives $|h| = |h - Mh| \le \frac12|h|$. Fix $y = f(x) \in W$, let $k \ne 0$ be small and $h = g(y + k) - g(y)$, so $|h| \le 2|k|$ and $h \ne 0$. Differentiability of $f$ at $x$ gives $k = f(x + h) - f(x) = Mh + |h|\rho(h)$ with $\rho(h) \to 0$ as $h \to 0$. Then
$$\frac{|g(y + k) - g(y) - M^{-1}k|}{|k|} = \frac{|h|\,|M^{-1}\rho(h)|}{|k|} \le 2\|M^{-1}\|\,|\rho(h)| \to 0 \quad (k \to 0),$$
because $h \to 0$ as $k \to 0$. So $Dg(y) = Df(g(y))^{-1}$.

*Step 4: $g$ is $C^1$.* By the [adjugate formula](laplace-expansion-adjugate-and-cramers-rule.ipynb), the entries of $M^{-1}$ are polynomials in the entries of $M$ divided by $\det M \ne 0$; the entries of $Df(g(y))$ are continuous functions of $y$ (compositions of continuous maps). Hence the entries of $Dg(y)$ are continuous. $\blacksquare$
:::

In coordinates the derivative formula reads $\Big[\dfrac{\partial x_i}{\partial y_j}(y)\Big] = \Big[\dfrac{\partial y_i}{\partial x_j}(x)\Big]^{-1}$. A $C^1$ map $g : W \to V$ between open sets that is bijective with $C^1$ inverse is a **($C^1$) change of coordinates**: a function $Z = \psi(x)$ on $V$ becomes the function $\psi\circ g$ of $y$, and by the chain rule differentiable functions stay differentiable.

::: {.callout-warning}
## Only Local

$\det Df \ne 0$ everywhere does not make $f$ injective: $f(x, y) = (e^x\cos y, e^x\sin y)$ has $\det Df = e^{2x} > 0$ but $f(x, y + 2\pi) = f(x, y)$. The theorem only gives injectivity on small neighbourhoods.
:::

## 2\. Polar, Cylindrical and Spherical Coordinates

::: {#exm-polar-coordinates}
## Polar Coordinates

$(x, y) = (r\cos\theta, r\sin\theta)$ has $\dfrac{\partial(x, y)}{\partial(r, \theta)} = \begin{bmatrix}\cos\theta & -r\sin\theta\\ \sin\theta & r\cos\theta\end{bmatrix}$, determinant $r$. Near any $(r_0, \theta_0)$ with $r_0 \ne 0$ there is a local inverse $(r, \theta) = (g_1(x, y), g_2(x, y))$. A domain on which the map is injective must avoid $r = 0$ (all $\theta$ give the origin) and must not contain two points $(r, \theta_1)$, $(r, \theta_2)$ with $\theta_2 - \theta_1 \in 2\pi\mathbb{Z}$; e.g. $r > 0$, $0 < \theta < 2\pi$ is mapped bijectively onto $\mathbb{R}^2$ minus the ray $\{(x, 0) : x \ge 0\}$.
:::

::: {#exm-cylindrical-spherical}
## Cylindrical and Spherical Coordinates

* **Cylindrical:** $(x, y, z) = (r\cos\theta, r\sin\theta, z)$, $\det\dfrac{\partial(x, y, z)}{\partial(r, \theta, z)} = r$.
* **Spherical:** $\rho = |(x, y, z)|$, $\varphi \in [0, \pi]$ the angle from the positive $z$-axis, $\theta$ the polar angle of $(x, y)$:
$$x = \rho\sin\varphi\cos\theta, \quad y = \rho\sin\varphi\sin\theta, \quad z = \rho\cos\varphi,$$
$$\frac{\partial(x, y, z)}{\partial(\rho, \varphi, \theta)} = \begin{bmatrix}\sin\varphi\cos\theta & \rho\cos\varphi\cos\theta & -\rho\sin\varphi\sin\theta\\ \sin\varphi\sin\theta & \rho\cos\varphi\sin\theta & \rho\sin\varphi\cos\theta\\ \cos\varphi & -\rho\sin\varphi & 0\end{bmatrix}, \qquad \det = \rho^2\sin\varphi .$$
The determinant vanishes exactly on the $z$-axis; on $\rho > 0$, $0 < \varphi < \pi$, $0 < \theta < 2\pi$ the map is a change of coordinates onto $\mathbb{R}^3$ minus a closed half-plane.
:::

```python

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

rho, ph, th = sp.symbols("rho varphi theta", positive=True)
S = sp.Matrix([rho * sp.sin(ph) * sp.cos(th), rho * sp.sin(ph) * sp.sin(th), rho * sp.cos(ph)])
print("spherical Jacobian determinant:", sp.simplify(S.jacobian([rho, ph, th]).det()))

# Newton-free inversion by the contraction of the proof: solve f(x) = y near 0 for f = x + small nonlinearity
f = lambda v: np.array([v[0] + 0.3 * np.sin(v[1]), v[1] + 0.2 * v[0]**2])
y = np.array([0.15, -0.1]); x = np.zeros(2)
A = np.array([[1.0, 0.3], [0.0, 1.0]])           # Df(0)
for k in range(30):
    x = x - np.linalg.solve(A, f(x) - y)            # x -> x - Df(0)^{-1}(f(x) - y), the map T_y of the proof
print("solution of f(x) = y:", x, "| residual:", np.linalg.norm(f(x) - y))

fig, ax = plt.subplots(figsize=(6, 5.4))
for r0 in np.linspace(0.2, 1.2, 6):
    t = np.linspace(-1.2, 1.2, 200)
    ax.plot(np.exp(r0 - 0.7) * np.cos(t), np.exp(r0 - 0.7) * np.sin(t), color=BLUE, lw=1)
for t0 in np.linspace(-1.2, 1.2, 9):
    s = np.linspace(0.2, 1.2, 100)
    ax.plot(np.exp(s - 0.7) * np.cos(t0), np.exp(s - 0.7) * np.sin(t0), color=ORANGE, lw=1)
ax.set_aspect("equal"); ax.set_title(r"Image of a grid under $(x, y) \mapsto (e^x\cos y, e^x\sin y)$")
plt.tight_layout(); plt.show()
```
