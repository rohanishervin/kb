# Geometric Sequences and the Geometric Series

**Prerequisites:** [The Limit of a Sequence](sequence-limit.ipynb) · [Algebraic Limit Theorem](composite-sequences-limit.ipynb)

A **geometric sequence** is $a, ar, ar^2, \dots$ with $a, r \in \mathbb{C}$; $r$ is the **ratio**.

::: {#prp-geometric-sum}
## Finite Geometric Sum

For $r \ne 1$: $\displaystyle \sum_{k=0}^{n} a r^k = a\,\frac{1 - r^{n+1}}{1 - r}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-geometric-sum
Let $S = \sum_{k=0}^n r^k$. Then $S - rS = 1 - r^{n+1}$ (all other terms cancel); divide by $1 - r$. $\blacksquare$
:::

::: {#thm-power-limit}
## Limit of $r^n$

1. If $|r| < 1$ then $r^n \to 0$.
2. If $|r| > 1$ then $|r^n| \to +\infty$.
3. If $|r| = 1$ then $(r^n)$ converges only for $r = 1$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-power-limit
**Bernoulli's inequality:** $(1 + h)^n \ge 1 + nh$ for $h \ge 0$ (drop all but two terms of the binomial expansion).

(1) If $r = 0$ it is clear. Otherwise $1/|r| = 1 + h$ with $h > 0$, so $|r^n| = (1+h)^{-n} \le \frac{1}{1 + nh} \to 0$.
(2) $|r| = 1 + h$ with $h > 0$, so $|r^n| \ge 1 + nh \to +\infty$.
(3) If $r^n \to L$ then $|L| = 1$ and also $r^{n+1} \to L$; but $r^{n+1} = r \cdot r^n \to rL$, so $L = rL$, and $L \ne 0$ forces $r = 1$. $\blacksquare$
:::

A series $\sum_{n \ge 0} a_n$ **converges** to $S$ if its partial sums $s_n = a_0 + \dots + a_n$ converge to $S$ (see [Infinite Series](series.ipynb)).

::: {#thm-geometric-series}
## The Geometric Series

Let $a \ne 0$. The series $\sum_{n=0}^{\infty} a r^n$ converges if and only if $|r| < 1$, and then
$$\sum_{n=0}^{\infty} a r^n = \frac{a}{1 - r} .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-geometric-series
If $|r| < 1$, the partial sums $a\frac{1 - r^{n+1}}{1 - r} \to \frac{a}{1-r}$ by @thm-power-limit. If $|r| \ge 1$ and the partial sums $s_n$ converged, then $ar^n = s_n - s_{n-1} \to 0$; but $|ar^n| \ge |a| > 0$. $\blacksquare$
:::

::: {#exm-geometric-series}
## Examples

1. $0.999\dots = \sum_{n \ge 1} 9 \cdot 10^{-n} = \frac{9/10}{1 - 1/10} = 1$.
2. $0.\overline{142857} = 142857 \sum_{k \ge 1} 10^{-6k} = \frac{142857}{10^6 - 1} = \frac17$.
3. $\sum_{n \ge 0} \left(\frac{1+i}{2}\right)^n = \frac{1}{1 - \frac{1+i}{2}} = \frac{2}{1 - i} = 1 + i$, since $\left|\frac{1+i}{2}\right| = \frac{\sqrt2}{2} < 1$.
:::
