# Flux, Divergence and Curl

**Prerequisites:** [Vector Fields, Line Integrals and Conservative Fields](line-integrals-and-conservative-fields.ipynb) · [Integrals of Functions over Curves and Surfaces](integrals-over-curves-and-surfaces.ipynb) · [Parametrized Surfaces and Surface Area](parametrized-surfaces.ipynb#prp-reparametrization-surface)

## 1\. Flux Through a Surface

::: {#def-flux}
## Flux

Let $\varphi : W \to \mathbb{R}^3$ be a smooth parametrized surface with standard unit normal $\mathbf n$, and $F$ a continuous vector field on $\varphi(W)$. The **flux** of $F$ through $\varphi$ is
$$\iint_\varphi F\cdot d\mathbf S = \iint_\varphi(F\cdot\mathbf n)\,dS = \iint_WF(\varphi(u, v))\cdot(\mathbf r_u\times\mathbf r_v)\,du\,dv,$$
using $\mathbf n\,dS = (\mathbf r_u\times\mathbf r_v)\,du\,dv$. If $F$ is the velocity field of a fluid, $F\cdot\mathbf n$ is the rate of flow across a unit of area in the direction $\mathbf n$, and the flux is the volume crossing the surface per unit time.
:::

The integrand $F\cdot\mathbf n$ changes sign with $\mathbf n$, so by the [reparametrization rule](parametrized-surfaces.ipynb#prp-reparametrization-surface) the flux is unchanged under orientation-preserving and changes sign under orientation-reversing reparametrizations: it depends on the surface together with a choice of normal. Expanding the cross product as $2\times2$ minors,
$$\iint_\varphi F\cdot d\mathbf S = \iint_W\Big[F_1\frac{\partial(y, z)}{\partial(u, v)} + F_2\frac{\partial(z, x)}{\partial(u, v)} + F_3\frac{\partial(x, y)}{\partial(u, v)}\Big]du\,dv,$$
written $\iint_\varphi F_1\,dy\,dz + F_2\,dz\,dx + F_3\,dx\,dy$ (note the cyclic order $dz\,dx$).

::: {#exm-flux-hemisphere}
## Flux of a Constant Field

Flux of $F = \mathbf k$ through the upper hemisphere $H$ of radius $R$ with upward (outward) normal. *In spherical coordinates:* $\mathbf n = \frac{(x, y, z)}{R}$, $F\cdot\mathbf n = \frac zR = \cos\varphi$, $dS = R^2\sin\varphi\,d\varphi\,d\theta$, so $\iint_HF\cdot d\mathbf S = \int_0^{2\pi}\int_0^{\pi/2}\cos\varphi\,R^2\sin\varphi\,d\varphi\,d\theta = \pi R^2$. *As a graph* $z = \sqrt{R^2 - x^2 - y^2}$, whose standard normal points up: the flux is $\iint F_3\,dx\,dy = \iint_{x^2+y^2\le R^2}1\,dx\,dy = \pi R^2$. The same amount flows through the flat disk under $H$, as it should for an incompressible flow.
:::

::: {#exm-flux-radial}
## Radial Fields Through Spheres

On the sphere $|x| = R$ with outward normal $\mathbf n = x/R$: for $F = x$, $F\cdot\mathbf n = R$ and the flux is $R\cdot4\pi R^2 = 4\pi R^3$; for $G = \frac{x}{|x|^3}$, $G\cdot\mathbf n = \frac1{R^2}$ and the flux is $4\pi$, independent of $R$.
:::

## 2\. Divergence and Curl

::: {#def-div-curl}
## Divergence and Curl

For a $C^1$ field $F = (F_1, \dots, F_n)$ on an open set in $\mathbb{R}^n$, the **divergence** is the function
$$\operatorname{div}F = \frac{\partial F_1}{\partial x_1} + \dots + \frac{\partial F_n}{\partial x_n},$$
the trace of the Jacobian matrix $DF$. For $n = 3$ the **curl** is the vector field
$$\operatorname{curl}F = \Big(\frac{\partial F_3}{\partial y} - \frac{\partial F_2}{\partial z},\ \frac{\partial F_1}{\partial z} - \frac{\partial F_3}{\partial x},\ \frac{\partial F_2}{\partial x} - \frac{\partial F_1}{\partial y}\Big).$$
With the symbolic vector $\nabla = \big(\frac{\partial}{\partial x}, \frac{\partial}{\partial y}, \frac{\partial}{\partial z}\big)$: $\nabla f = \operatorname{grad}f$, $\nabla\cdot F = \operatorname{div}F$, $\nabla\times F = \operatorname{curl}F$ (the formal [cross product](cross-product.ipynb#def-cross-product)).
:::

The divergence will turn out to be the outward flux per unit volume, measuring expansion ($\operatorname{div}F > 0$) or compression of the flow near a point ([divergence theorem](divergence-theorem.ipynb)); the curl is the circulation per unit area, measuring rotation ([Stokes' theorem](stokes-theorem.ipynb)).

::: {#exm-div-curl}
## Examples

1. $F = (x, y, z)$: $\operatorname{div}F = 3$, $\operatorname{curl}F = 0$.
2. Rotation about the $z$-axis with angular speed $\omega$: $F = -\omega y\,\mathbf i + \omega x\,\mathbf j$: $\operatorname{div}F = 0$, $\operatorname{curl}F = 2\omega\,\mathbf k$, along the axis of rotation with length twice the angular speed.
3. Shear $F = x\,\mathbf j$: $\operatorname{div}F = 0$ but $\operatorname{curl}F = \mathbf k \ne 0$ although the flow lines are straight: a small paddle wheel placed in the flow at $x > 0$ turns, because the push on its far side exceeds that on its near side.
4. $G = \frac{x}{|x|^3}$ on $\mathbb{R}^3\setminus\{0\}$: $\frac{\partial}{\partial x}\frac{x}{r^3} = \frac{1}{r^3} - \frac{3x^2}{r^5}$, so $\operatorname{div}G = \frac{3}{r^3} - \frac{3r^2}{r^5} = 0$.
:::

::: {#thm-curl-grad-div-curl}
## $\operatorname{curl}\operatorname{grad} = 0$ and $\operatorname{div}\operatorname{curl} = 0$

For a $C^2$ function $f$ and a $C^2$ field $F$ on an open set in $\mathbb{R}^3$:
$$\operatorname{curl}(\nabla f) = 0, \qquad \operatorname{div}(\operatorname{curl}F) = 0 .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-curl-grad-div-curl
$\operatorname{curl}\nabla f = (f_{zy} - f_{yz},\ f_{xz} - f_{zx},\ f_{yx} - f_{xy}) = 0$ and
$$\operatorname{div}\operatorname{curl}F = \frac{\partial^2F_3}{\partial x\partial y} - \frac{\partial^2F_2}{\partial x\partial z} + \frac{\partial^2F_1}{\partial y\partial z} - \frac{\partial^2F_3}{\partial y\partial x} + \frac{\partial^2F_2}{\partial z\partial x} - \frac{\partial^2F_1}{\partial z\partial y} = 0$$
by [Schwarz's theorem](higher-partials-and-taylor.ipynb#thm-schwarz). $\blacksquare$
:::

So a gradient field is irrotational (its flow lines climb $f$ and never return), and a curl is incompressible. The first identity is a restatement, in $\mathbb{R}^3$, of the [necessary condition for gradients](line-integrals-and-conservative-fields.ipynb#prp-gradient-necessary).

::: {#prp-vector-identities}
## Product Rules

For $C^1$ functions $f, g$ and fields $F, G$ on an open set in $\mathbb{R}^3$:
$$\nabla(fg) = g\nabla f + f\nabla g, \qquad \nabla\cdot(fG) = f\,\nabla\cdot G + \nabla f\cdot G, \qquad \nabla\times(fG) = f\,\nabla\times G + \nabla f\times G,$$
$$\nabla\cdot(F\times G) = G\cdot(\nabla\times F) - F\cdot(\nabla\times G),$$
and $\nabla$, $\nabla\cdot$, $\nabla\times$ are linear.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-vector-identities
Each identity is checked componentwise with the one-variable product rule. For example $\nabla\cdot(fG) = \sum_i\partial_i(fG_i) = \sum_i(f\,\partial_iG_i + G_i\,\partial_if)$. The first component of $\nabla\times(fG)$ is $\partial_y(fG_3) - \partial_z(fG_2) = f(\partial_yG_3 - \partial_zG_2) + (f_yG_3 - f_zG_2)$, the first component of $f\nabla\times G + \nabla f\times G$. For the last, $\nabla\cdot(F\times G) = \sum_{\text{cyclic}}\partial_x(F_2G_3 - F_3G_2)$; the terms with derivatives of $F$ collect to $G\cdot(\nabla\times F)$ and those with derivatives of $G$ to $-F\cdot(\nabla\times G)$. $\blacksquare$
:::

```python

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})
from matplotlib.patches import FancyArrowPatch

x, y, z, R, ph, th = sp.symbols("x y z R varphi theta", positive=True)
X = sp.Matrix([x, y, z])
def div(F): return sum(sp.diff(F[i], X[i]) for i in range(3))
def curl(F): return sp.Matrix([sp.diff(F[2], y) - sp.diff(F[1], z), sp.diff(F[0], z) - sp.diff(F[2], x), sp.diff(F[1], x) - sp.diff(F[0], y)])
f, g = sp.Function("f")(x, y, z), sp.Function("g")(x, y, z)
F = sp.Matrix([sp.Function(f"F{i}")(x, y, z) for i in range(3)]); G = sp.Matrix([sp.Function(f"G{i}")(x, y, z) for i in range(3)])
grad = lambda h: sp.Matrix([sp.diff(h, v) for v in X])
print("curl grad f = 0:", sp.simplify(curl(grad(f))) == sp.zeros(3, 1), "| div curl F = 0:", sp.simplify(div(curl(F))) == 0)
print("div(F x G) identity:", sp.simplify(div(F.cross(G)) - (G.dot(curl(F)) - F.dot(curl(G)))) == 0)
print("curl(fG) identity:", sp.simplify(curl(f * G) - (f * curl(G) + grad(f).cross(G))) == sp.zeros(3, 1))
r = sp.sqrt(x**2 + y**2 + z**2)
print("div(x/|x|^3) =", sp.simplify(div(X / r**3)), "| curl(-wy, wx, 0) =", list(curl(sp.Matrix([-2 * y, 2 * x, 0]) / 2 * sp.Symbol("omega"))))
print("flux of k through the hemisphere:", sp.integrate(sp.cos(ph) * R**2 * sp.sin(ph), (ph, 0, sp.pi / 2), (th, 0, 2 * sp.pi)))

fig, ax = plt.subplots(figsize=(5.4, 5))
gg = np.linspace(-2, 2, 15); XX, YY = np.meshgrid(gg, gg)
ax.quiver(XX, YY, 0 * XX, XX, color=BLUE, scale=35, width=0.004)
for cx in [-1.2, 1.2]:
    ax.add_patch(plt.Circle((cx, 0), 0.35, fill=False, color=ORANGE, lw=1.5))
    ax.add_patch(FancyArrowPatch((cx + 0.5, -0.2), (cx - 0.2, 0.5), connectionstyle="arc3,rad=0.5",
                                 arrowstyle="-|>", mutation_scale=12, color=ORANGE, lw=1.5))
ax.set_aspect("equal"); ax.set_title(r"Shear $x\,\mathbf{j}$: straight flow lines, $\operatorname{curl} = \mathbf{k}$")
plt.tight_layout(); plt.show()
```
