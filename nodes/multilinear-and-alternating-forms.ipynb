# Multilinear and Alternating Forms

**Prerequisites:** [Linear Maps](linear-maps.ipynb) · [Linear Independence](linear-independence.ipynb)

A **multilinear** map takes several vectors and is linear in each one separately. Products, dot products, cross products and areas of parallelograms are examples. Among them, the **alternating** ones, which vanish as soon as two arguments coincide, are exactly the right tool to measure (signed) area and volume. The [determinant](determinant.ipynb) will be the alternating $n$-linear form on $K^n$, normalised so that the unit cube has volume $1$.

Throughout, $K$ is a field and $V, V_1, \dots, V_p, U$ are [vector spaces](vector-spaces.ipynb) over $K$.

## 1\. Multilinear Maps

::: {#def-multilinear-map}
## Multilinear Map

A map $\Phi : V_1 \times \dots \times V_p \to U$ is **$p$-linear** (or **multilinear**) if it is linear in each argument when the others are held fixed: for each $j$, all fixed $v_k \in V_k$ ($k \ne j$), all $u, v \in V_j$ and $a \in K$,
$$\Phi(v_1, \dots, a u + v, \dots, v_p) = a\,\Phi(v_1, \dots, u, \dots, v_p) + \Phi(v_1, \dots, v, \dots, v_p),$$
the changed argument being in position $j$. For $p = 2$ we say **bilinear**. When $U = K$, $\Phi$ is a **$p$-linear form**.
:::

::: {#exm-multilinear}
## Multilinear Maps

1. **Products of numbers.** $K^p \to K$, $(x_1, \dots, x_p) \mapsto x_1 x_2 \cdots x_p$ is $p$-linear. (It is **not** linear as a map on $K^p$ when $p \ge 2$: doubling all arguments multiplies the value by $2^p$.)
2. **Dot product.** $\mathbb{R}^n \times \mathbb{R}^n \to \mathbb{R}$, $(x, y) \mapsto x_1 y_1 + \dots + x_n y_n$ is bilinear.
3. **Cross product.** $\mathbb{R}^3 \times \mathbb{R}^3 \to \mathbb{R}^3$, $(u, v) \mapsto u \times v$ is bilinear.
4. **Signed area.** $\mathbb{R}^2 \times \mathbb{R}^2 \to \mathbb{R}$, $\omega(u, v) = u_1 v_2 - u_2 v_1$ is bilinear. Its absolute value is the area of the parallelogram spanned by $u$ and $v$.
:::

Expanding a multilinear map one argument at a time turns $\Phi(\sum_i a_i x_i, \dots)$ into a sum over all choices of one term from each argument. This "multiply out every bracket" rule is the source of the Leibniz formula for the determinant.

## 2\. Alternating Maps

::: {#def-alternating-form}
## Alternating Map

A $p$-linear map $\Phi : V \times \dots \times V \to U$ (all arguments from the same space) is **alternating** if
$$\Phi(v_1, \dots, v_p) = 0 \quad \text{whenever } v_i = v_j \text{ for some } i \ne j.$$
:::

The signed area $\omega$ and the cross product are alternating: $\omega(u, u) = u_1 u_2 - u_2 u_1 = 0$ and $u \times u = 0$. The dot product is not, since $\langle u, u \rangle > 0$ for $u \ne 0$.

::: {#prp-alternating-sign-reversal}
## Swapping Two Arguments Changes the Sign

If $\Phi$ is alternating, exchanging two arguments multiplies the value by $-1$:
$$\Phi(\dots, u, \dots, w, \dots) = -\Phi(\dots, w, \dots, u, \dots).$$
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-alternating-sign-reversal
Keep all other arguments fixed and put $u + w$ in both of the two positions. Since $\Phi$ is alternating, the value is $0$. Expanding by linearity in each of the two positions,
$$0 = \Phi(\dots, u, \dots, u, \dots) + \Phi(\dots, u, \dots, w, \dots) + \Phi(\dots, w, \dots, u, \dots) + \Phi(\dots, w, \dots, w, \dots).$$
The first and last terms vanish because $\Phi$ is alternating, which leaves $\Phi(\dots, u, \dots, w, \dots) = -\Phi(\dots, w, \dots, u, \dots)$. $\blacksquare$
:::

::: {.callout-warning}
## Alternating Versus Antisymmetric
A map with the sign-reversal property is called **antisymmetric**. If $1 + 1 \ne 0$ in $K$, antisymmetric implies alternating: putting $u = w$ gives $\Phi(\dots, u, \dots, u, \dots) = -\Phi(\dots, u, \dots, u, \dots)$, so $2\,\Phi(\dots, u, \dots, u, \dots) = 0$, and $2$ is invertible. In a field with $1 + 1 = 0$ (such as $\mathbb{F}_2 = \{0, 1\}$), $-1 = 1$ and the sign-reversal condition says nothing. That is why "alternating" is defined by vanishing on repeated arguments, which works over every field.
:::

::: {#prp-alternating-dependent}
## Alternating Maps and Dependence

Let $\Phi$ be alternating $p$-linear.

1. Adding a multiple of one argument to another does not change the value:
   $$\Phi(\dots, v_i + a v_j, \dots, v_j, \dots) = \Phi(\dots, v_i, \dots, v_j, \dots).$$
2. If $v_1, \dots, v_p$ are [linearly dependent](linear-independence.ipynb#def-linear-independence), then $\Phi(v_1, \dots, v_p) = 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-alternating-dependent
1. By linearity in position $i$, the left side is $\Phi(\dots, v_i, \dots, v_j, \dots) + a\,\Phi(\dots, v_j, \dots, v_j, \dots)$, and the second term vanishes because $v_j$ appears twice.
2. By [redundancy](linear-independence.ipynb#prp-dependence-redundancy) some $v_i$ is a combination $\sum_{j \ne i} c_j v_j$ of the others. Expanding in position $i$, $\Phi(v_1, \dots, v_p) = \sum_{j \ne i} c_j\,\Phi(v_1, \dots, v_j, \dots, v_p)$ with $v_j$ in position $i$; each term has $v_j$ twice and vanishes. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Signed Area Is Bilinear and Alternating
For the signed area $\omega(u, v)$ of the parallelogram spanned by $u$ and $v$:

- **Additivity in $u$** holds because parallelograms with the same base $v$ have areas proportional to their heights, and heights add.
- **Shear invariance** (part 1 above) is Cavalieri's principle: sliding the top edge parallel to the base keeps base and height.
- **Alternating:** a parallelogram spanned by $u$ and $u$ is flat, so its area is $0$; swapping $u$ and $v$ reverses the orientation and hence the sign.

These are exactly the properties that will characterise the determinant.
:::


```python

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

BLUE, ORANGE, AQUA, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({"axes.titlesize": 10.5, "axes.titleweight": "bold", "font.size": 10,
                     "xtick.color": MUTED, "ytick.color": MUTED, "axes.edgecolor": GRID})

area = lambda u, v: u[0] * v[1] - u[1] * v[0]          # signed area omega(u, v)
def para(ax, u, v, color, alpha=0.25, origin=(0, 0)):
    o = np.array(origin, float)
    ax.add_patch(Polygon([o, o + u, o + u + v, o + v], closed=True, fc=color, ec=color, alpha=alpha, lw=1.5))

v = np.array([2.5, 0.0])
u1, u2 = np.array([0.6, 1.0]), np.array([0.4, 0.8])
fig, axes = plt.subplots(1, 3, figsize=(14, 4.2))

ax = axes[0]
para(ax, v, u1, BLUE); para(ax, v, u2, ORANGE, origin=u1)
para(ax, v, u1 + u2, VIOLET, alpha=0.08)
ax.set_title(f"additive: ω(v,u₁+u₂) = {area(v, u1):.1f} + {area(v, u2):.1f} = {area(v, u1 + u2):.1f}")

ax = axes[1]
for t, c in [(0.0, BLUE), (0.6, AQUA), (1.2, ORANGE)]:
    para(ax, v, u1 + t * v, c, alpha=0.2)
ax.set_title(f"shear: ω(v, u + t v) = {area(v, u1):.1f} for every t")

ax = axes[2]
a, b = np.array([2.0, 0.5]), np.array([0.7, 1.6])
para(ax, a, b, AQUA)
ax.annotate("", xy=a, xytext=(0, 0), arrowprops=dict(arrowstyle="-|>", color=BLUE, lw=2))
ax.annotate("", xy=b, xytext=(0, 0), arrowprops=dict(arrowstyle="-|>", color=ORANGE, lw=2))
ax.text(*(a + [0.05, -0.25]), "$u$"); ax.text(*(b + [-0.25, 0.05]), "$v$")
ax.set_title(f"alternating: ω(u,v) = {area(a, b):.2f}, ω(v,u) = {area(b, a):.2f}")

for ax in axes:
    ax.set_aspect("equal"); ax.set_xlim(-0.3, 5.2); ax.set_ylim(-0.3, 2.2)
    ax.axhline(0, color=MUTED, lw=0.8); ax.axvline(0, color=MUTED, lw=0.8)
plt.tight_layout(); plt.show()
```
