# Limits of Functions

**Prerequisites:** [Continuous Functions](function-continuity.ipynb) · [Squeeze Theorem for Sequences](squeeze-theorem-for-sequences.ipynb) · [The Sequential Criterion for Function Limits](func-lim-seq-lim.ipynb)

Throughout, $S \subseteq \mathbb{R}$ and $f, g : S \to \mathbb{R}$.

## 1\. Limit Points

::: {#def-limit-point}
## Limit Point

$a \in \mathbb{R}$ is a **limit point** (accumulation point) of $S$ if every punctured interval $(a - \delta, a + \delta) \setminus \{a\}$, $\delta > 0$, contains a point of $S$. The point $a$ itself may or may not belong to $S$.
:::

::: {#lem-limit-point-sequence}
## Limit Points via Sequences

$a$ is a limit point of $S$ if and only if there is a sequence of **distinct** points $s_n \in S \setminus \{a\}$ with $s_n \to a$. In particular every punctured interval around $a$ contains infinitely many points of $S$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-limit-point-sequence
($\Rightarrow$) Choose $s_1 \in S$ with $0 < |s_1 - a| < 1$. Having chosen $s_1, \dots, s_{n-1}$, let $\delta_n = \min\big(\frac1n, |s_{n-1} - a|\big) > 0$ and choose $s_n \in S$ with $0 < |s_n - a| < \delta_n$. Then $|s_n - a|$ is strictly decreasing, so the $s_n$ are distinct, and $|s_n - a| < \frac1n \to 0$. ($\Leftarrow$) Given $\delta$, $0 < |s_n - a| < \delta$ for $n$ large. $\blacksquare$
:::

For example: $\mathbb{Z}$ has no limit points; the limit points of $(a, b)$, $[a, b)$, $[a,b]$ are all points of $[a, b]$; the only limit point of $\{\frac1n : n \in \mathbb{N}\}$ is $0$.

## 2\. Definition and Basic Properties

::: {#def-function-limit}
## Limit of a Function

Let $a$ be a limit point of $S$. Then $\lim_{x \to a} f(x) = L$ means
$$\forall \varepsilon > 0\ \ \exists \delta > 0\ \ \forall x \in S: \quad 0 < |x - a| < \delta \ \Rightarrow\ |f(x) - L| < \varepsilon .$$
:::

The value $f(a)$ (if defined) plays no role; that $a$ is a limit point guarantees that the condition is not vacuous, and makes $L$ unique (as for sequences).

::: {#prp-limit-and-continuity}
## Continuity in Terms of Limits

Let $a \in S$ be a limit point of $S$. Then $f$ is continuous at $a$ if and only if $\lim_{x \to a} f(x) = f(a)$. (At points of $S$ that are not limit points, every $f$ is continuous.)
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-limit-and-continuity
The two definitions differ only at $x = a$, where $|f(x) - f(a)| = 0 < \varepsilon$ anyway. At an isolated point $a$, some $\delta$-interval meets $S$ only in $a$. $\blacksquare$
:::

::: {#thm-sequential-limit}
## Sequential Criterion

Let $a$ be a limit point of $S$. Then $\lim_{x \to a} f(x) = L$ if and only if $f(x_n) \to L$ for every sequence $x_n \in S \setminus \{a\}$ with $x_n \to a$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-sequential-limit
As for [continuity](function-continuity.ipynb#thm-sequential-continuity): ($\Rightarrow$) for $n$ large $0 < |x_n - a| < \delta$. ($\Leftarrow$) If the limit is not $L$, some $\varepsilon$ admits, for each $\delta = \frac1n$, a point $x_n \in S$ with $0 < |x_n - a| < \frac1n$ and $|f(x_n) - L| \ge \varepsilon$. $\blacksquare$
:::

::: {#thm-function-limit-laws}
## Limit Laws

Let $\lim_{x \to a} f = L_1$, $\lim_{x\to a} g = L_2$. Then $\lim (f + g) = L_1 + L_2$ and $\lim fg = L_1 L_2$. If moreover $L_2 \ne 0$ and $a$ is a limit point of $\{x \in S : g(x) \ne 0\}$ (the domain of $f/g$), then $\lim f/g = L_1/L_2$. If $f \le h \le g$ near $a$ and $L_1 = L_2$, then $\lim h = L_1$ (**squeeze**).
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-function-limit-laws
By @thm-sequential-limit and the [limit laws for sequences](composite-sequences-limit.ipynb#thm-limit-laws) and the [squeeze theorem](squeeze-theorem-for-sequences.ipynb). $\blacksquare$
:::

## 3\. Fundamental Examples

::: {#exm-power-quotient}
## A $0/0$ Quotient

$\lim_{x \to 1} \frac{x^n - 1}{x - 1} = n$: on $S = \mathbb{R} \setminus \{1\}$ the quotient equals the polynomial $1 + x + \dots + x^{n-1}$, which is continuous with value $n$ at $1$.
:::

::: {#thm-sin-limit}
## The Two Trigonometric Limits

$$\lim_{\theta \to 0} \frac{\sin\theta}{\theta} = 1, \qquad \lim_{\theta \to 0} \frac{1 - \cos\theta}{\theta} = 0 .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-sin-limit
For $0 < \theta < \frac\pi2$, the inequality $\sin\theta < \theta < \tan\theta$ ([inequality (1) here](function-continuity.ipynb#trigonometric-functions)) gives $1 < \frac{\theta}{\sin\theta} < \frac{1}{\cos\theta}$; all three expressions are even in $\theta$, so this holds for $0 < |\theta| < \frac\pi2$. Since $\cos$ is continuous with $\cos 0 = 1$, squeeze gives $\frac{\theta}{\sin\theta} \to 1$, hence $\frac{\sin\theta}{\theta} \to 1$. Next $\frac{1 - \cos\theta}{\theta} = \frac{2\sin^2(\theta/2)}{\theta} = \sin\frac\theta2 \cdot \frac{\sin(\theta/2)}{\theta/2} \to 0 \cdot 1$. $\blacksquare$
:::

::: {#exm-limit-examples}
## More Examples

1. *Bounded times null.* If $\lim_{x \to a} f = 0$ and $|g| \le M$, then $\lim fg = 0$: $|fg| \le M|f|$. Hence $\lim_{x\to 0} x^n \sin\frac1x = 0$ for $n \ge 1$.
2. *No limit.* $\lim_{x \to 0}\sin\frac1x$ does not exist: $x_n = \frac{1}{n\pi}$ gives $0$, $x_n = \frac{1}{2n\pi + \pi/2}$ gives $1$ ([sequential criterion](func-lim-seq-lim.ipynb)).
3. *The domain matters.* $h(\theta) = \sin\frac1\theta \big/ \sin\frac1\theta$ is defined on $S = \{\theta \ne 0,\ \theta \ne \frac{1}{n\pi}\}$, where it equals $1$; $0$ is a limit point of $S$ (e.g. $\frac{1}{n\pi + \pi/2} \in S$), so $\lim_{\theta \to 0} h(\theta) = 1$.
:::

## 4\. One-Sided and Infinite Limits

::: {#def-one-sided-limits}
## One-Sided Limits, Limits at Infinity, Infinite Limits

1. $\lim_{x \to a^+} f(x) = L$: the limit of $f$ restricted to $S \cap (a, \infty)$, of which $a$ must be a limit point; similarly $\lim_{x \to a^-}$.
2. $\lim_{x \to +\infty} f(x) = L$ (for $S$ unbounded above): $\forall \varepsilon\ \exists M\ \forall x \in S,\ x > M: |f(x) - L| < \varepsilon$; similarly $x \to -\infty$.
3. $\lim_{x \to a} f(x) = +\infty$: $\forall M\ \exists \delta\ \forall x \in S,\ 0 < |x - a| < \delta: f(x) > M$; similarly for $-\infty$, for one-sided limits and for $x \to \pm\infty$.
:::

If $a$ is a limit point of both $S \cap (a, \infty)$ and $S \cap (-\infty, a)$, then $\lim_{x\to a} f = L$ if and only if both one-sided limits equal $L$ (take the smaller of the two $\delta$'s). The sequential criterion holds in each case, with $x_n \to +\infty$ in (2); for example $\lim_{x\to+\infty} f(x) = L$ if and only if $f(x_n) \to L$ for every sequence $x_n \in S$ with $x_n \to +\infty$. Examples: $\lim_{x\to 0^\pm}\frac{|x|}{x} = \pm 1$, $\lim_{x \to 0^+}\frac1x = +\infty$, $\lim_{x \to +\infty} \frac{\sin x}{x} = 0$.

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

x = np.concatenate([np.linspace(-0.5, -1e-4, 40000), np.linspace(1e-4, 0.5, 40000)])
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(11, 3.8))
ax.plot(x, np.sin(1 / x), color=RED, lw=0.6)
ax.set_title(r"$\sin(1/x)$: no limit at $0$")
ax2.plot(x, x * np.sin(1 / x), color=BLUE, lw=0.8, label=r"$x \sin(1/x)$")
ax2.plot(x, np.abs(x), color=AQUA, lw=1.2, ls="--", label=r"$\pm|x|$")
ax2.plot(x, -np.abs(x), color=AQUA, lw=1.2, ls="--")
ax2.set_title(r"$x\sin(1/x) \to 0$: squeezed between $\pm|x|$"); ax2.legend(fontsize=9)
plt.tight_layout(); plt.show()
```
