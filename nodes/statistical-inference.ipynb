# Statistical Inference

Statistical inference provides a framework for drawing conclusions about a population based on a sample. The core components of this framework include:

- **Sample**: The data collected from the population.
- **Statistic**: A function of the sample data.
- **Point Estimation**: Estimating an unknown parameter with a single value.
- **Confidence Intervals**: Providing a range of plausible values for a parameter.
- **Hypothesis Testing**: Evaluating evidence in favor of or against a claim about the population.


## 1\. Typical Statistical Problem

Given a random variable of interest and its corresponding sample $X_1, \dots, X_n$, the underlying probability mass function (pmf) $p(x)$ or probability density function (pdf) $f(x)$ is generally unknown. 

There are two primary scenarios:

1. We have **no information** about the distribution (non-parametric).
2. We know the **family** of the distribution, but the specific parameters are unknown (parametric).

Our initial focus will be on the second scenario: we know the family, but the parameters are unknown. Thus, our goal is to find the function $f(x; \theta)$ or $p(x; \theta)$ where the parameter $\theta$ belongs to a parameter space $\Omega$.


## 2\. Random Sample

::: {#def-random-sample}
## Random Sample
Let $X_1, \dots, X_n$ be independent and identically distributed (i.i.d.) random variables following a distribution $F$. That is,
$$ X_1, \dots, X_n \overset{\text{i.i.d.}}{\sim} F $$
Then they constitute a **random sample** from the distribution $F$.
:::
The realized values of this random sample are denoted by lowercase letters: $x_1, \dots, x_n$.


## 3\. Statistic

::: {#def-statistic}
## Statistic
Let $X_1, \dots, X_n$ denote a random sample of a random variable $X$. A **statistic** $T$ is any function of the sample:
$$ T = T(X_1, \dots, X_n) $$
:::


## 4\. Unbiasedness

::: {#def-unbiasedness}
## Unbiased Estimator
Let $X_1, \dots, X_n$ be a sample of a random variable $X$ with pdf $f(x; \theta)$ for $\theta \in \Omega$. Let $T = T(X_1, \dots, X_n)$ be a statistic. 

The statistic $T$ is an **unbiased estimator** of $\theta$ if its expected value equals the true parameter:
$$ E[T] = \theta $$
:::
