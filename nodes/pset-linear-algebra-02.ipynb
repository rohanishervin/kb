# Problem Set 2: Linear Algebra

Problems over all 31 notes in [Linear Algebra](index-linear-algebra.ipynb). They follow the order of the notes. The notation is the table in the index. $P_n(K)$ means the polynomials of degree at most $n$ with coefficients in $K$.

Every problem was written for this set. A **Source** line is added only where the statement is a well-known classical result; it names where you can also find it.



## Problem 1: Sum and Intersection of Two Subspaces of $\mathbb{R}^4$

**Covers:** [subspaces](subspaces.ipynb#def-sum-subspaces), [basis and dimension](basis-and-dimension.ipynb#thm-extend-to-basis), [dimension of a sum](direct-sums-and-complements.ipynb#thm-dim-sum)

**Problem Statement:**

In $V = \mathbb{R}^4$ let
$$
U = \operatorname{span}\{(1,0,1,0),\ (0,1,0,1),\ (1,1,1,1)\}, \qquad
W = \{x \in \mathbb{R}^4 : x_1 + x_2 - x_3 - x_4 = 0,\ x_1 - x_2 + x_3 - x_4 = 0\}.
$$

**(a)** Find a basis of $U$ and a basis of $W$.

**(b)** Find $\dim(U + W)$ and $\dim(U \cap W)$ without computing $U \cap W$ directly. Then find a basis of $U \cap W$ and check that it agrees.

**(c)** Is $U + W$ a direct sum? Extend a basis of $U \cap W$ to a basis of $\mathbb{R}^4$.



## Problem 2: When Is a Sum of Several Subspaces Direct?

**Covers:** [direct sum of $s$ subspaces](direct-sums-and-complements.ipynb#prp-direct-sum-criterion), [dimension of a multiple direct sum](direct-sums-and-complements.ipynb#prp-dim-direct-sum-multiple)

**Problem Statement:**

Let $W_1, \dots, W_s$ be subspaces of a finite-dimensional space $V$.

**(a)** Prove that $W_1 + \dots + W_s$ is direct if and only if $\dim(W_1 + \dots + W_s) = \dim W_1 + \dots + \dim W_s$.

**(b)** Give three subspaces $W_1, W_2, W_3$ of $\mathbb{R}^2$ with $W_i \cap W_j = \{0\}$ for all $i \ne j$ whose sum is not direct. Which part of the definition fails?

**(c)** Prove that the sum is direct if and only if $W_j \cap (W_1 + \dots + W_{j-1}) = \{0\}$ for every $j = 2, \dots, s$.

## Problem 7: Rank of a Composite

**Covers:** [rank-nullity](kernel-image-and-rank-nullity.ipynb#thm-rank-nullity), [image of a subspace](linear-maps.ipynb#thm-subspace-image)

**Problem Statement:**

Let $f : V \to W$ and $g : W \to U$ be linear maps between finite-dimensional spaces.

**(a)** Prove $\operatorname{rank}(g \circ f) \le \min(\operatorname{rank} f, \operatorname{rank} g)$.

**(b)** Prove $\dim \ker(g \circ f) \le \dim \ker f + \dim \ker g$. (Hint: restrict $f$ to $\ker(g \circ f)$.)

**(c)** Deduce Sylvester's inequality: $\operatorname{rank} f + \operatorname{rank} g - \dim W \le \operatorname{rank}(g \circ f)$.

**(d)** Show that if $f$ is surjective then $\operatorname{rank}(g \circ f) = \operatorname{rank} g$, and if $g$ is injective then $\operatorname{rank}(g \circ f) = \operatorname{rank} f$.

**Source:** Sylvester's rank inequality (classical).

## Problem 28: Challenge: Synthesis
