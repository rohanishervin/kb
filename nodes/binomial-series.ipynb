# The Binomial Series and Operations on Power Series

**Prerequisites:** [Power Series](power-series.ipynb) · [Taylor Polynomials and Taylor's Theorem](taylor-polynomials.ipynb#lem-general-leibniz) · [Absolute and Conditional Convergence](absolute-and-conditional-convergence.ipynb#thm-cauchy-product) · [The Binomial Theorem](binomial-theorem.ipynb)

## 1\. The Binomial Series

For $\alpha \in \mathbb{R}$ and $n \ge 0$ let $\binom{\alpha}{n} = \frac{\alpha(\alpha-1)\cdots(\alpha-n+1)}{n!}$ ($\binom\alpha0 = 1$). For $\alpha \in \mathbb{N}$ these are the usual binomial coefficients, and $\binom{\alpha}{n} = 0$ for $n > \alpha$.

::: {#thm-binomial-series}
## Binomial Series

For every $\alpha \in \mathbb{R}$ and $|x| < 1$,
$$(1 + x)^\alpha = \sum_{n=0}^\infty\binom{\alpha}{n}x^n = 1 + \alpha x + \frac{\alpha(\alpha-1)}{2!}x^2 + \cdots$$
If $\alpha \in \{0, 1, 2, \dots\}$ the sum is finite and the identity holds for all $x$ ([binomial theorem](binomial-theorem.ipynb)); otherwise the radius of convergence is $1$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-binomial-series
The coefficients of the Taylor series of $(1 + x)^\alpha = e^{\alpha\ln(1+x)}$ at $0$ are $\binom\alpha n$, since the $n$-th derivative is $\alpha(\alpha-1)\cdots(\alpha-n+1)(1+x)^{\alpha-n}$. If $\alpha \notin \mathbb{Z}_{\ge0}$, no coefficient vanishes and $\Big|\binom{\alpha}{n+1}\Big/\binom\alpha n\Big| = \frac{|\alpha - n|}{n+1} \to 1$, so $\rho = 1$ ([formula](power-series.ipynb#prp-radius-formula)).

Let $g(x) = \sum\binom\alpha nx^n$ for $|x| < 1$. Differentiating [term by term](power-series.ipynb#thm-termwise), the coefficient of $x^n$ in $(1 + x)g'(x)$ is
$$(n+1)\binom{\alpha}{n+1} + n\binom\alpha n = \binom\alpha n\big((\alpha - n) + n\big) = \alpha\binom\alpha n,$$
so $(1 + x)g' = \alpha g$. Then $h(x) = g(x)(1 + x)^{-\alpha}$ satisfies $h' = (1+x)^{-\alpha-1}\big((1+x)g' - \alpha g\big) = 0$, hence $h \equiv h(0) = 1$ and $g(x) = (1 + x)^\alpha$. $\blacksquare$
:::

::: {#exm-binomial-series}
## Special Cases

1. $\alpha = -p$, $p \in \mathbb{N}$: $\binom{-p}{n} = (-1)^n\binom{n + p - 1}{n}$, so
$$\frac{1}{(1+x)^p} = \sum_n(-1)^n\binom{n+p-1}{n}x^n, \qquad \frac{1}{(a + h)^p} = \frac{1}{a^p} - \frac{ph}{a^{p+1}} + \frac{p(p+1)h^2}{2a^{p+2}} - \cdots \quad (|h| < |a|).$$
(For $p = 1$ this is the geometric series; the others also follow by differentiating it.)
2. $\alpha = -\frac12$: $\frac{1}{\sqrt{1+x}} = 1 - \frac12x + \frac{1\cdot3}{2\cdot4}x^2 - \cdots$; replacing $x$ by $-x^2$,
$$\frac{1}{\sqrt{1 - x^2}} = 1 + \frac12x^2 + \frac{1\cdot3}{2\cdot4}x^4 + \frac{1\cdot3\cdot5}{2\cdot4\cdot6}x^6 + \cdots \quad (|x| < 1),$$
and integrating term by term,
$$\arcsin x = x + \frac12\cdot\frac{x^3}{3} + \frac{1\cdot 3}{2\cdot4}\cdot\frac{x^5}{5} + \cdots \quad (|x| < 1).$$
:::

## 2\. Algebra of Power Series

::: {#thm-power-series-algebra}
## Sums and Products

Let $f(x) = \sum a_n(x - a)^n$ for $|x - a| < \rho_1$ and $g(x) = \sum b_n(x - a)^n$ for $|x - a| < \rho_2$, and $\rho = \min(\rho_1, \rho_2)$. For $|x - a| < \rho$:
$$f(x) + g(x) = \sum(a_n + b_n)(x - a)^n, \qquad f(x)g(x) = \sum_n c_n(x - a)^n, \quad c_n = \sum_{k=0}^n a_kb_{n-k}.$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-power-series-algebra
The sum: limit laws for partial sums. The product: both series converge absolutely there, so the [Cauchy product](absolute-and-conditional-convergence.ipynb#thm-cauchy-product) of $\sum a_n(x-a)^n$ and $\sum b_n(x-a)^n$, whose $n$-th term is $c_n(x - a)^n$, converges to $f(x)g(x)$. $\blacksquare$
:::

Quotients: if $g(a) \ne 0$, then $f/g$ is also analytic at $a$ (stated without proof; the natural proof uses complex analysis). Its Taylor coefficients can be found from $f = g\cdot\frac fg$ by comparing coefficients. This comparison is justified even without knowing analyticity: by the [general Leibniz rule](taylor-polynomials.ipynb#lem-general-leibniz), the Taylor coefficients of a product $uv$ of infinitely differentiable functions are $\sum_{k\le n}\frac{u^{(k)}(a)}{k!}\frac{v^{(n-k)}(a)}{(n-k)!}$, the Cauchy product of those of $u$ and $v$.

::: {#exm-tan-series}
## The Taylor Series of $\tan$

Let $\tan x = \sum t_nx^n$ near $0$ (Taylor coefficients). From $\sin x = \cos x\tan x$:
$$x - \frac{x^3}{6} + \frac{x^5}{120} - \cdots = \Big(1 - \frac{x^2}{2} + \frac{x^4}{24} - \cdots\Big)\big(t_0 + t_1x + t_2x^2 + \cdots\big).$$
Comparing coefficients of $1, x, \dots, x^5$: $0 = t_0$, $1 = t_1$, $0 = t_2 - \frac{t_0}2$, $-\frac16 = t_3 - \frac{t_1}{2}$, $0 = t_4 - \frac{t_2}2 + \frac{t_0}{24}$, $\frac1{120} = t_5 - \frac{t_3}2 + \frac{t_1}{24}$. Hence
$$\tan x = x + \frac{x^3}{3} + \frac{2x^5}{15} + \cdots$$
(only odd powers appear, as $\tan$ is odd).
:::

::: {#exm-limits-by-series}
## Limits by Series

Many $0/0$ limits are immediate from Taylor expansions ([uniqueness](taylor-polynomials.ipynb#cor-taylor-unique) justifies manipulating them):
$$\frac{x^6\cos x - x^5\sin x}{x^8} = \frac{\big(x^6 - \frac{x^8}{2} + \cdots\big) - \big(x^6 - \frac{x^8}{6} + \cdots\big)}{x^8} = -\frac13 + (\text{terms in } x^2, x^4, \dots) \to -\frac13,$$
whereas l'Hôpital's rule would need eight differentiations. Likewise $\frac{x - \sin x}{x^3} \to \frac16$ and $\frac{\tan x - x}{x^3} \to \frac13$.
:::

```python

import sympy as sp

x, a = sp.symbols("x a")
print("tan x       =", sp.series(sp.tan(x), x, 0, 8))
print("arcsin x    =", sp.series(sp.asin(x), x, 0, 9))
print("1/sqrt(1-x^2) =", sp.series(1 / sp.sqrt(1 - x**2), x, 0, 9))
print("(1+x)^(1/3) =", sp.series((1 + x)**sp.Rational(1, 3), x, 0, 5))
print("binom(1/3, n):", [sp.binomial(sp.Rational(1, 3), n) for n in range(5)])
print("limits:", sp.limit((x**6*sp.cos(x) - x**5*sp.sin(x)) / x**8, x, 0),
      sp.limit((x - sp.sin(x)) / x**3, x, 0), sp.limit((sp.tan(x) - x) / x**3, x, 0))
```
