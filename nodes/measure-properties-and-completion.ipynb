# Almost Everywhere, σ-Finite Measures, and Completeness

**Prerequisites:** [Measures and Measure Spaces](measure.ipynb)

General measure spaces can be very strange. A few definitions single out the well-behaved ones, and they appear among the hypotheses of most theorems. They are rarely restrictive in practice: almost every space met in applications satisfies them. All of them revolve around two themes: sets of measure zero, which measure theory treats as negligible, and sets of finite measure. Throughout, $(X, \mathcal{M}, \mu)$ is a [measure space](measure.ipynb#def-measure).

## 1\. Almost Everywhere

::: {#def-almost-everywhere}
## Almost Everywhere

Let $E \in \mathcal{M}$. A property of points holds **almost everywhere on $E$** (abbreviated **a.e.**), or **for almost all $x \in E$**, if the set $E_0$ of points of $E$ where it fails satisfies $\mu(E_0) = 0$.
:::

When several measures are around, one writes "a.e. $(\mu)$". If the failure set $E_0$ is not known to be measurable, the precise requirement is that $E_0$ is contained in a measurable set of measure $0$; in a complete measure space (Section 4) the two versions agree.

::: {#exm-almost-every-real-is-irrational}
## Almost Every Real Number Is Irrational

For [Lebesgue measure](Lebesgue-measure.ipynb), constructed later, single points have measure $0$. The rationals form a countable union of points, so by countable additivity $\mu(\mathbb{Q}) = 0$: almost every real number is irrational.
:::

## 2\. Finite and σ-Finite Measures

::: {#def-finite-sigma-finite}
## Finite and σ-Finite Measures

- $\mu$ is **finite** if $\mu(X) < \infty$.
- $\mu$ is **σ-finite** if $X$ is a countable union of measurable sets of finite measure: $X = \bigcup_{i=1}^\infty E_i$ with $\mu(E_i) < \infty$.

The same words are used for a set $E \in \mathcal{M}$: $E$ has **finite measure** if $\mu(E) < \infty$, and is **σ-finite** if it is a countable union of measurable sets of finite measure.
:::

In Euclidean space the whole space has infinite measure, but the plane, for instance, is covered by countably many unit squares, each of finite area; so area is σ-finite. By [disjointification](measure.ipynb#lem-disjointification) the covering sets can always be taken pairwise disjoint, and they still have finite measure.

Almost all measures met in practice are σ-finite, which is why so many theorems assume it. Measures that are not σ-finite do exist, but they are rarely useful.

::: {#prp-counting-not-sigma-finite}
## Counting Measure on an Uncountable Set Is Not σ-Finite

If $X$ is uncountable, the [counting measure](measure.ipynb#exm-counting-measure) on $X$ is not σ-finite.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-counting-not-sigma-finite
For the counting measure, a set of finite measure is a finite set. If $X = \bigcup_i E_i$ with every $E_i$ finite, then $X$ is a countable union of finite sets, hence countable, a contradiction. $\blacksquare$
:::

## 3\. Probability Measures

::: {#def-probability-measure}
## Probability Measure

A measure $\mu$ with $\mu(X) = 1$ is a **probability measure**. Probability texts usually write $P$ instead of $\mu$.
:::

Probability theory starts here, and every notion of measure theory gets a probabilistic name. For instance, a property that holds almost everywhere is said to hold **almost surely** (a.s.): the event where it fails has probability $0$. Since $P(X) = 1$, an almost sure event is an event of probability $1$; it need not be the whole sample space.


## 4\. Complete Measure Spaces

Subsets of a set of measure zero should themselves have measure zero; for Lebesgue measure, examples of null sets are $\mathbb{Q}$ and the [Cantor set](cantor-set.ipynb#measure-zero). This is automatic, by monotonicity, *if* the subset is measurable. It may happen that it is not in $\mathcal{M}$ at all. Nothing prevents us from adding such sets to the σ-algebra, and a space where nothing needs to be added is called complete.

::: {#def-complete-measure-space}
## Complete Measure Space

The measure space $(X, \mathcal{M}, \mu)$ is **complete** if, whenever $E \in \mathcal{M}$ has $\mu(E) = 0$, every subset $A \subseteq E$ belongs to $\mathcal{M}$.
:::

Then automatically $\mu(A) = 0$, since $0 \le \mu(A) \le \mu(E) = 0$ by [monotonicity](measure.ipynb#prp-monotonicity). Completeness is a property of the pair $\mathcal{M}$, $\mu$ together, which is why it is attached to the whole triple rather than to $\mu$ alone. The word is related to the completeness of the function spaces $L^p$ built from a measure space (see [$L^1$ and $L^p$](lp-spaces.ipynb)), whose proof needs this condition.

Every measure space can be completed by the obvious recipe: take all subsets of sets of measure zero and add them, together with their unions with measurable sets, to the σ-algebra.

::: {#thm-completion}
## Completion of a Measure Space

Let $\mathcal{M}_0$ be the family of all sets $E \cup A$ with $E \in \mathcal{M}$ and $A \subseteq N$ for some $N \in \mathcal{M}$ with $\mu(N) = 0$, and put $\mu_0(E \cup A) = \mu(E)$. Then

1. $\mu_0$ is well defined;
2. $\mathcal{M}_0$ is a σ-algebra containing $\mathcal{M}$;
3. $\mu_0$ is a measure on $\mathcal{M}_0$ extending $\mu$;
4. $(X, \mathcal{M}_0, \mu_0)$ is complete.

It is called the **completion** of $(X, \mathcal{M}, \mu)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-completion
Call $N \in \mathcal{M}$ with $\mu(N) = 0$ a *null set*.

1. **Well defined.** Suppose $E_1 \cup A_1 = E_2 \cup A_2$ with $A_i \subseteq N_i$, $N_i$ null. Then $E_1 \subseteq E_2 \cup A_2 \subseteq E_2 \cup N_2$, so $\mu(E_1) \le \mu(E_2) + \mu(N_2) = \mu(E_2)$ by monotonicity and subadditivity. By symmetry $\mu(E_2) \le \mu(E_1)$.
2. **σ-algebra.** $\emptyset = \emptyset \cup \emptyset \in \mathcal{M}_0$, and $E = E \cup \emptyset$ shows $\mathcal{M} \subseteq \mathcal{M}_0$.
    - *Complements.* Let $E \cup A$ with $A \subseteq N$. A point outside $E \cup A$ either lies outside $N$, hence outside $E \cup N$, or lies in $N$ but not in $E \cup A$. Conversely both kinds of points avoid $E \cup A$ (recall $A \subseteq N$). So
    $$(E \cup A)^c = (E \cup N)^c \cup \big(N \setminus (E \cup A)\big),$$
    where $(E \cup N)^c \in \mathcal{M}$ and $N \setminus (E \cup A) \subseteq N$. Hence $(E \cup A)^c \in \mathcal{M}_0$.
    - *Countable unions.* $\bigcup_i (E_i \cup A_i) = \big(\bigcup_i E_i\big) \cup \big(\bigcup_i A_i\big)$, with $\bigcup_i E_i \in \mathcal{M}$ and $\bigcup_i A_i \subseteq \bigcup_i N_i$, which is null by [countable monotonicity](measure.ipynb#prp-countable-monotonicity): $\mu(\bigcup_i N_i) \le \sum_i 0 = 0$.
3. **Measure.** $\mu_0(\emptyset) = \mu(\emptyset) = 0$. If the sets $E_i \cup A_i$ are pairwise disjoint, so are the $E_i$, and by the union formula above $\mu_0\big(\bigcup_i (E_i \cup A_i)\big) = \mu\big(\bigcup_i E_i\big) = \sum_i \mu(E_i) = \sum_i \mu_0(E_i \cup A_i)$. It extends $\mu$ because $\mu_0(E \cup \emptyset) = \mu(E)$.
4. **Complete.** If $\mu_0(E \cup A) = 0$, then $\mu(E) = 0$, so $E \cup N$ is null. Any $B \subseteq E \cup A \subseteq E \cup N$ can be written $B = \emptyset \cup B$ with $B$ inside the null set $E \cup N$, so $B \in \mathcal{M}_0$. $\blacksquare$
:::

The code below runs the construction on a small example. On $X = \{0, \dots, 5\}$ take the σ-algebra generated by the blocks $\{0,1\}, \{2,3\}, \{4,5\}$ and the measure with block masses $0, 1, 2$. The block $\{0,1\}$ is null but $\{0\} \notin \mathcal{M}$, so the space is not complete. The code builds $\mathcal{M}_0$ by the recipe of @thm-completion and checks every claim of the theorem by brute force.

```python

from itertools import combinations, chain

X = frozenset(range(6))
blocks = [frozenset({0, 1}), frozenset({2, 3}), frozenset({4, 5})]
mass = {blocks[0]: 0, blocks[1]: 1, blocks[2]: 2}

# the sigma-algebra M generated by the blocks, and the measure mu on it
M = {frozenset(chain.from_iterable(c)) for r in range(4) for c in combinations(blocks, r)}
mu = {E: sum(mass[b] for b in blocks if b <= E) for E in M}
null_sets = [N for N in M if mu[N] == 0]

def subsets(S):
    S = list(S)
    return [frozenset(c) for r in range(len(S) + 1) for c in combinations(S, r)]

# the completion: all E u A with E in M and A inside a null set
M0, mu0 = set(), {}
for E in M:
    for N in null_sets:
        for A in subsets(N):
            F = E | A
            assert mu0.setdefault(F, mu[E]) == mu[E], "mu_0 not well defined"
            M0.add(F)

is_sigma = (frozenset() in M0 and all(X - F in M0 for F in M0)
            and all(F | G in M0 for F in M0 for G in M0))
additive = all(mu0[F | G] == mu0[F] + mu0[G] for F in M0 for G in M0 if not F & G)
complete = all(B in M0 for F in M0 if mu0[F] == 0 for B in subsets(F))
print(f"|M| = {len(M)}, |M_0| = {len(M0)};  {{0}} in M: {frozenset({0}) in M},  {{0}} in M_0: {frozenset({0}) in M0}")
print("M_0 is a sigma-algebra:", is_sigma, "| mu_0 additive:", additive,
      "| mu_0 extends mu:", all(mu0[E] == mu[E] for E in M), "| complete:", complete)
```
