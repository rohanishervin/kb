# Curvature of Plane Curves

**Prerequisites:** [Parametrized Curves and Arc Length](parametrized-curves.ipynb) · [Complex Numbers](complex-numbers.ipynb) · [The Intermediate Value Theorem](intermediate-value-theorem.ipynb)

Throughout, $\gamma$ is a $C^2$ smooth curve in $\mathbb{R}^2$ parametrized by [arc length](parametrized-curves.ipynb#prp-arc-length-parametrization) $s$, so $T(s) = \gamma'(s)$ is a unit vector, the **unit tangent**. Curvature measures how fast $T$ turns.

## 1\. The Turning Angle and Curvature

::: {#lem-angle-function}
## A Continuous Angle for the Tangent

There is a $C^1$ function $\theta$ with $T(s) = (\cos\theta(s), \sin\theta(s))$ for all $s$, namely $\theta(s) = \theta_0 + \int_{s_0}^s(T_1T_2' - T_2T_1')$ where $T(s_0) = (\cos\theta_0, \sin\theta_0)$. Any two continuous such functions differ by a constant multiple of $2\pi$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-angle-function
Identify $\mathbb{R}^2$ with $\mathbb{C}$ and put $z = T_1 + iT_2$, $|z| = 1$. Then $z'\bar z = (T_1T_1' + T_2T_2') + i(T_1T_2' - T_2T_1') = i\theta'$, because $T\cdot T' = 0$ ([constant length](parametrized-curves.ipynb#prp-curve-derivative-rules)); multiplying by $z$ gives $z' = i\theta'z$. For $w = e^{-i\theta}z$ (with $e^{i\theta} = \cos\theta + i\sin\theta$, differentiated componentwise),
$$w' = e^{-i\theta}(z' - i\theta'z) = 0,$$
so $w \equiv w(s_0) = 1$, i.e. $z = e^{i\theta}$. If $\theta, \tilde\theta$ are continuous and both work, $(\theta - \tilde\theta)/2\pi$ is continuous and integer-valued, hence constant by the IVT. $\blacksquare$
:::

::: {#def-plane-curvature}
## Signed Curvature and Normal

The **(signed) curvature** is $\kappa(s) = \theta'(s) = T_1T_2' - T_2T_1' = \det[T\ T']$, and the **unit normal** is $N(s) = (-\sin\theta, \cos\theta)$, the rotation of $T$ by $+\frac\pi2$. $\kappa > 0$ where the curve turns counterclockwise (to the left).
:::

::: {#prp-plane-frenet}
## Frenet Equations in the Plane

$$\frac{dT}{ds} = \kappa N, \qquad \frac{dN}{ds} = -\kappa T, \qquad |\kappa| = \Big|\frac{dT}{ds}\Big| .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-plane-frenet
Differentiate $T = (\cos\theta, \sin\theta)$ and $N = (-\sin\theta, \cos\theta)$ with the chain rule: $T' = \theta'(-\sin\theta, \cos\theta) = \kappa N$, $N' = \theta'(-\cos\theta, -\sin\theta) = -\kappa T$. $\blacksquare$
:::

::: {#exm-curvature-lines-circles}
## Lines and Circles

A line $\gamma(s) = p + sT$ has $\kappa = 0$. The circle $\gamma(s) = c + R(\cos\frac sR, \sin\frac sR)$, traversed counterclockwise, has $T = (-\sin\frac sR, \cos\frac sR)$, $\theta = \frac sR + \frac\pi2$, so $\kappa = \frac1R$; traversed clockwise, $\kappa = -\frac1R$. Conversely, $\kappa \equiv 0$ forces $T$ constant, so the curve is a segment of a line.
:::

## 2\. Arbitrary Parameters

::: {#prp-curvature-formula}
## Curvature and Acceleration for Any Parameter

Let $\gamma(t) = (x(t), y(t))$ be a $C^2$ smooth curve with arc length $s(t)$. Then
$$\gamma'' = s''\,T + \kappa\,(s')^2\,N, \qquad \kappa = \frac{x'y'' - y'x''}{(x'^2 + y'^2)^{3/2}} .$$
In particular for the graph of $y = f(x)$, $\kappa = \dfrac{f''}{(1 + f'^2)^{3/2}}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-curvature-formula
$\gamma' = s'\,T(s(t))$, so by the chain rule and @prp-plane-frenet, $\gamma'' = s''T + s'\cdot s'\,\frac{dT}{ds} = s''T + \kappa(s')^2N$. Then $\det[\gamma'\ \gamma''] = s'\cdot\kappa(s')^2\det[T\ N] = \kappa(s')^3$ and $s' = (x'^2 + y'^2)^{1/2}$. For a graph take $x = t$, $y = f(t)$. $\blacksquare$
:::

The tangential part $s''T$ of the acceleration changes the speed; the normal part $\kappa(s')^2N$ bends the path. The circle of radius $1/|\kappa|$ centred at $\gamma + \frac1\kappa N$ is the **osculating circle**; it has the same tangent and curvature as $\gamma$ at that point.

::: {#exm-ellipse-curvature}
## Curvature of an Ellipse

For $\gamma(t) = (a\cos t, b\sin t)$, $a, b > 0$: $x'y'' - y'x'' = ab$, so $\kappa(t) = \dfrac{ab}{(a^2\sin^2t + b^2\cos^2t)^{3/2}}$. At the ends of the major axis ($t = 0$, if $a > b$) $\kappa = a/b^2$ is largest; at the ends of the minor axis $\kappa = b/a^2$ is smallest.
:::

## 3\. Curvature Determines the Curve

A **rigid motion** of the plane is a map $x \mapsto Rx + c$ with $R$ a rotation ($R$ orthogonal, $\det R = 1$). It preserves arc length and curvature: $T$ becomes $RT$, and $\det[RT\ RT'] = \det R\det[T\ T']$.

::: {#thm-plane-curves-fundamental}
## Fundamental Theorem of Plane Curves

1. If $\gamma_1, \gamma_2 : [0, L] \to \mathbb{R}^2$ are arc-length parametrized $C^2$ curves with the same curvature function, then $\gamma_2 = R\gamma_1 + c$ for a rigid motion.
2. Every continuous $\kappa : [0, L] \to \mathbb{R}$ is the curvature of such a curve: $\theta(s) = \int_0^s\kappa$, $\gamma(s) = \int_0^s(\cos\theta, \sin\theta)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-plane-curves-fundamental
(1) Replace $\gamma_1$ by $R\gamma_1 + c$, where $R$ rotates $T_1(0)$ to $T_2(0)$ and $c$ makes $\gamma_1(0) = \gamma_2(0)$; this keeps $\kappa_1$, and now also $N_1(0) = N_2(0)$, since $N$ is $T$ rotated by $\frac\pi2$ and rotations commute. Let $\varphi = T_1\cdot T_2 + N_1\cdot N_2$. With $\kappa = \kappa_1 = \kappa_2$ and @prp-plane-frenet,
$$\varphi' = \kappa N_1\cdot T_2 + \kappa T_1\cdot N_2 - \kappa T_1\cdot N_2 - \kappa N_1\cdot T_2 = 0,$$
so $\varphi \equiv \varphi(0) = 2$. Since $T_1\cdot T_2 \le 1$ and $N_1\cdot N_2 \le 1$ (unit vectors), both equal $1$, so $T_1 = T_2$; then $(\gamma_1 - \gamma_2)' = 0$ and $\gamma_1 = \gamma_2$.
(2) $\gamma$ is $C^2$ with $\gamma' = (\cos\theta, \sin\theta)$, a unit vector, so $s$ is arc length, $T = (\cos\theta, \sin\theta)$ and the curvature is $\theta' = \kappa$ (FTC). $\blacksquare$
:::

```python

from scipy.integrate import cumulative_trapezoid
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

fig, (ax, ax2) = plt.subplots(1, 2, figsize=(12, 4.6))
a, b = 3.0, 1.5
t = np.linspace(0, 2 * np.pi, 600)
x, y = a * np.cos(t), b * np.sin(t)
kap = a * b / (a**2 * np.sin(t)**2 + b**2 * np.cos(t)**2) ** 1.5
sc = ax.scatter(x, y, c=kap, cmap="viridis", s=6)
fig.colorbar(sc, ax=ax, label=r"$\kappa$", shrink=0.8)
for t0, col in [(0.0, ORANGE), (np.pi / 2, VIOLET)]:
    k0 = a * b / (a**2 * np.sin(t0)**2 + b**2 * np.cos(t0)**2) ** 1.5
    T = np.array([-a * np.sin(t0), b * np.cos(t0)]); T /= np.linalg.norm(T)
    N = np.array([-T[1], T[0]])
    c = np.array([a * np.cos(t0), b * np.sin(t0)]) + N / k0
    ax.plot(c[0] + np.cos(t) / k0, c[1] + np.sin(t) / k0, color=col, lw=1.2, ls="--",
            label=f"osculating circle, $R = {1 / k0:.2f}$")
ax.set_aspect("equal"); ax.set_xlim(-6.5, 6.5); ax.set_ylim(-11, 2.2)
ax.legend(fontsize=8, loc="lower right"); ax.set_title(r"Ellipse $(3\cos t, 1.5\sin t)$ coloured by curvature")

s = np.linspace(-6, 6, 4000)
theta = s**2 / 2                      # kappa(s) = s
gx = cumulative_trapezoid(np.cos(theta), s, initial=0)
gy = cumulative_trapezoid(np.sin(theta), s, initial=0)
ax2.plot(gx - gx[2000], gy - gy[2000], color=BLUE, lw=1.5)
ax2.set_aspect("equal"); ax2.set_title(r"The curve with $\kappa(s) = s$ (clothoid)")
plt.tight_layout(); plt.show()

kn = np.gradient(np.unwrap(np.arctan2(np.gradient(gy, s), np.gradient(gx, s))), s)
print("max |numerical kappa - s| on the clothoid:", np.max(np.abs(kn - s)[50:-50]))
```
