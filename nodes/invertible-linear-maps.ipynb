# Left, Right, and Two-Sided Inverses

**Prerequisites:** [Kernel, Image, and Rank-Nullity](kernel-image-and-rank-nullity.ipynb)

An injective map loses no information, so it should be possible to undo it from the left; a surjective map reaches everything, so it should be possible to undo it from the right. This note makes both statements precise for linear maps and shows that a map with inverses on both sides has exactly one inverse.

Throughout, $V$ and $W$ are [vector spaces](vector-spaces.ipynb) over a field $K$ and $f : V \to W$ is a [linear map](linear-maps.ipynb#def-linear-map).

## 1\. Left and Right Inverses

::: {#def-one-sided-inverses}
## Left and Right Inverses

A linear map $g : W \to V$ is

1. a **left inverse** of $f$ if $g \circ f = \mathrm{id}_V$;
2. a **right inverse** of $f$ if $f \circ g = \mathrm{id}_W$;
3. a **(two-sided) inverse** of $f$ if it is both. Then $f$ is **invertible**.
:::

::: {#exm-one-sided}
## One-Sided Inverses Are Not Unique

Let $f : \mathbb{R}^2 \to \mathbb{R}^3$, $f(x, y) = (x,\ y,\ x + y)$. Both
$$g_1(a, b, c) = (a, b) \qquad\text{and}\qquad g_2(a, b, c) = (c - b,\ b)$$
are left inverses: $g_1(f(x, y)) = (x, y)$ and $g_2(f(x, y)) = ((x + y) - y,\ y) = (x, y)$. Neither is a right inverse, since $f(g_1(0, 0, 1)) = (0, 0, 0) \ne (0, 0, 1)$; in fact $f$ has no right inverse at all, because it is not surjective.

Dually, $p : \mathbb{R}^3 \to \mathbb{R}^2$, $p(a, b, c) = (a, b)$ has the right inverses $s_1(x, y) = (x, y, 0)$ and $s_2(x, y) = (x, y, x + y)$, since $p \circ s_1 = p \circ s_2 = \mathrm{id}$.
:::

::: {#thm-left-inverse-injective}
## Injective Maps Have Left Inverses

Let $V$ and $W$ be finite-dimensional. Then $f$ is injective if and only if $f$ has a linear left inverse.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-left-inverse-injective
($\Rightarrow$) Let $(v_1, \dots, v_n)$ be a basis of $V$ and put $w_i = f(v_i)$. The $w_i$ are independent: if $\sum c_i w_i = 0$, then $f(\sum c_i v_i) = 0$, so $\sum c_i v_i \in \ker f = \{0\}$ by [injectivity](kernel-image-and-rank-nullity.ipynb#prp-injective-kernel), and all $c_i = 0$. [Extend](basis-and-dimension.ipynb#thm-extend-to-basis) them to a basis $w_1, \dots, w_n, w_{n+1}, \dots, w_m$ of $W$ and let $g : W \to V$ be the [linear map determined by](linear-maps.ipynb#thm-linear-extension)
$$g(w_i) = v_i \ (i \le n), \qquad g(w_j) = 0 \ (j > n).$$
Then $(g \circ f)(v_i) = g(w_i) = v_i$ for all $i$, so the linear maps $g \circ f$ and $\mathrm{id}_V$ agree on a basis and are equal.

($\Leftarrow$) If $g \circ f = \mathrm{id}_V$ and $f(u) = f(v)$, then $u = g(f(u)) = g(f(v)) = v$. $\blacksquare$
:::

::: {#thm-right-inverse-surjective}
## Surjective Maps Have Right Inverses

Let $W$ be finite-dimensional. Then $f$ is surjective if and only if $f$ has a linear right inverse.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-right-inverse-surjective
($\Rightarrow$) Let $(w_1, \dots, w_m)$ be a basis of $W$. By surjectivity choose $v_i \in V$ with $f(v_i) = w_i$, and let $g : W \to V$ be the linear map with $g(w_i) = v_i$. Then $(f \circ g)(w_i) = w_i$ for all $i$, so $f \circ g = \mathrm{id}_W$.

($\Leftarrow$) If $f \circ g = \mathrm{id}_W$, every $w \in W$ equals $f(g(w))$. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Where the Freedom Comes From
In the proof for left inverses, the values $g(w_j)$ for $j > n$ were set to $0$, but any choice works: a left inverse only has to undo $f$ on $\operatorname{im} f$ and may do anything on a complement of it. In the proof for right inverses, each preimage $v_i$ may be replaced by $v_i + k$ with $k \in \ker f$. So one-sided inverses are unique only when there is no freedom: $\operatorname{im} f = W$ for left inverses, and $\ker f = \{0\}$ for right inverses.
:::

## 2\. Two-Sided Inverses

::: {#thm-two-sided-inverse-uniqueness}
## Left and Right Inverses Agree

If $g$ is a left inverse and $g'$ a right inverse of $f$, then $g = g'$. Consequently an invertible map has exactly one inverse, denoted $f^{-1}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-two-sided-inverse-uniqueness
Using associativity of composition,
$$g = g \circ \mathrm{id}_W = g \circ (f \circ g') = (g \circ f) \circ g' = \mathrm{id}_V \circ g' = g'.$$
If $g_1, g_2$ are both two-sided inverses, then $g_1$ is a left and $g_2$ a right inverse, so $g_1 = g_2$. $\blacksquare$
:::

::: {#prp-inverse-is-linear}
## The Inverse of a Bijective Linear Map Is Linear

If $f$ is bijective, its inverse function $f^{-1} : W \to V$ is linear. Hence $f$ is invertible in the sense of @def-one-sided-inverses if and only if it is bijective, if and only if it is a [linear isomorphism](coordinates-and-isomorphism-with-Kn.ipynb#def-linear-isomorphism).
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-inverse-is-linear
This is the [first part of the proposition on isomorphisms](coordinates-and-isomorphism-with-Kn.ipynb#prp-isomorphism-basic): for $w, w' \in W$, $f(a f^{-1}(w) + f^{-1}(w')) = a w + w'$, so $f^{-1}(a w + w') = a f^{-1}(w) + f^{-1}(w')$. Conversely, a map with a two-sided inverse is bijective. $\blacksquare$
:::

::: {#cor-equal-dim-invertibility}
## Invertibility in Equal Dimensions

If $\dim V = \dim W$ is finite, the following are equivalent:

1. $f$ is injective;
2. $f$ is surjective;
3. $f$ has a left inverse;
4. $f$ has a right inverse;
5. $f$ is invertible.

Moreover, any one-sided inverse is then automatically the two-sided inverse $f^{-1}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-equal-dim-invertibility
(1) $\iff$ (2) $\iff$ (5) is the [equal-dimension case of rank-nullity](kernel-image-and-rank-nullity.ipynb#cor-dimension-constraints) together with @prp-inverse-is-linear. (1) $\iff$ (3) and (2) $\iff$ (4) are @thm-left-inverse-injective and @thm-right-inverse-surjective. If $g$ is, say, a left inverse, then $f$ is invertible, and $g = g \circ (f \circ f^{-1}) = (g \circ f) \circ f^{-1} = f^{-1}$; similarly for right inverses. $\blacksquare$
:::

::: {.callout-warning}
## Square Matrices
For $A, B \in M_n(K)$, the corollary says that $B A = I$ already implies $A B = I$. This is special to finite dimensions; with the shift operators on sequences (see [Kernel, Image, and Rank-Nullity](kernel-image-and-rank-nullity.ipynb#cor-dimension-constraints)), the left shift is a left inverse of the right shift but not a right inverse.
:::

The code checks @exm-one-sided with matrices: both left inverses undo $f$, neither is a right inverse, and $f \circ g$ is a different projection of $\mathbb{R}^3$ onto the plane $\operatorname{im} f$ in each case.


```python

import numpy as np

F = np.array([[1, 0], [0, 1], [1, 1]])            # f(x, y) = (x, y, x + y)
G1 = np.array([[1, 0, 0], [0, 1, 0]])            # g1(a, b, c) = (a, b)
G2 = np.array([[0, -1, 1], [0, 1, 0]])           # g2(a, b, c) = (c - b, b)

for name, G in [("g1", G1), ("g2", G2)]:
    print(f"{name} o f =\n{G @ F}")
    print(f"f o {name} =\n{F @ G}\n")
P = np.array([[1, 0, 0], [0, 1, 0]])               # p(a, b, c) = (a, b)
S1 = np.array([[1, 0], [0, 1], [0, 0]])
S2 = np.array([[1, 0], [0, 1], [1, 1]])
print("p o s1 =\n", P @ S1, "\np o s2 =\n", P @ S2)
```
