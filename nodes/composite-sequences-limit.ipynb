# Algebraic Limit Theorem for Sequences

**Prerequisites:** [The Limit of a Sequence](sequence-limit.ipynb) · [Triangle Inequality](triangle-inequality.ipynb)

Sequences may be real or complex.

::: {#thm-limit-laws}
## Limit Laws

Let $a_n \to a$, $b_n \to b$, and $c$ a constant. Then

1. $a_n + b_n \to a + b$ and $c\,a_n \to c\,a$;
2. $a_n b_n \to ab$;
3. if $b \ne 0$, then $b_n \ne 0$ for all large $n$ and $\dfrac{a_n}{b_n} \to \dfrac{a}{b}$.
:::

::: {#lem-bounded-away}
## Bounded Away from Zero

If $b_n \to b \ne 0$, there is $N$ with $|b_n| \ge |b|/2$ for all $n \ge N$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-bounded-away
Take $\varepsilon = |b|/2$: for $n \ge N$, $|b_n| \ge |b| - |b - b_n| > |b|/2$. $\blacksquare$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-limit-laws
(1) $|(a_n + b_n) - (a + b)| \le |a_n - a| + |b_n - b|$; given $\varepsilon$, make each term $< \varepsilon/2$ for $n \ge \max(N_1, N_2)$. And $|c a_n - c a| = |c||a_n - a|$.

(2) By [boundedness](sequence-limit.ipynb#prp-convergent-bounded) there is $K_1$ with $|a_n| \le K_1$; let $K_2 = |b|$. Then
$$|a_n b_n - ab| = |a_n(b_n - b) + b(a_n - a)| \le K_1 |b_n - b| + K_2 |a_n - a| . \tag{$*$}$$
Given $\varepsilon$, choose $n$ so large that $|b_n - b| < \frac{\varepsilon}{2(K_1+1)}$ and $|a_n - a| < \frac{\varepsilon}{2(K_2+1)}$.

(3) By @lem-bounded-away, for $n \ge N$:
$$\Big|\frac{1}{b_n} - \frac{1}{b}\Big| = \frac{|b - b_n|}{|b_n||b|} \le \frac{2}{|b|^2}\,|b_n - b| \to 0 . \tag{$**$}$$
So $1/b_n \to 1/b$, and $a_n / b_n = a_n \cdot (1/b_n) \to a/b$ by (2). $\blacksquare$
:::

The inequalities $(*)$ and $(**)$ are **explicit error bounds**: they say how accurately $a_n, b_n$ must be known to know $ab$ or $1/b$ to a prescribed accuracy.

::: {#exm-precision}
## How Many Digits Are Needed?

1. Let $a = 487.r_1r_2\dots$ and $b = 3.s_1s_2\dots$ be known through decimal truncations $a_n, b_n$ ($n$ digits), so $|a - a_n|, |b - b_n| \le 10^{-n}$, $|a_n| \le 488$, $|b| \le 4$. By $(*)$, $|a_n b_n - ab| \le 488 \cdot 10^{-n} + 4\cdot 10^{-n} = 492 \cdot 10^{-n}$. Eight digits ($492 \cdot 10^{-8} < 10^{-5}$) guarantee five correct decimals of the product.
2. Let $b = 0.0321909909\dots$. Every truncation $b_n$ with $n \ge 4$ satisfies $b_n \ge 0.0321 =: k$, and the computation of $(**)$ gives $|1/b_n - 1/b| \le |b_n - b|/k^2$. Since $k^2 > 1.03 \cdot 10^{-3}$, the accuracy $|b_n - b| \le 10^{-6}$ ($n = 6$) gives $1/b$ within $10^{-3}$.
:::

::: {#exm-recursive-limit}
## A Recursive Sequence

$a_1 = 1$, $a_{n+1} = \sqrt{2 + a_n}$. By induction $0 < a_n < 2$ and $a_n < a_{n+1}$ (for $0 < a < 2$, $a^2 < a + 2$, so $a < \sqrt{2 + a}$). By the [monotone convergence theorem](sequence-limit.ipynb#thm-monotone-convergence), $a_n \to L \in (0, 2]$. The limit laws applied to $a_{n+1}^2 = 2 + a_n$ give $L^2 = 2 + L$ (the shifted sequence $a_{n+1}$ has the same limit), so $L = 2$.
:::

::: {.callout-warning}
## The Limit Must Exist First
The laws transfer limits that are known to exist. From $a_{n+1}^2 = 2 + a_n$ alone one cannot conclude anything: e.g. $a_{n+1} = 2a_n$, $a_1 = 1$ "satisfies" $L = 2L$, i.e. $L = 0$, although $a_n = 2^{n-1} \to \infty$.
:::
