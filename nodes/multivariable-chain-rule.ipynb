# The Chain Rule in Several Variables

**Prerequisites:** [Differentiable Maps and the Jacobian Matrix](differentiability-in-Rn.ipynb) · [The Chain Rule and Derivatives of Inverse Functions](chain-rule-and-inverse-functions.ipynb#thm-chain-rule) · [Parametrized Curves and Arc Length](parametrized-curves.ipynb)

## 1\. The Chain Rule

::: {#lem-matrix-norm}
## A Bound for Linear Maps

For an $m\times n$ matrix $M$ let $\|M\| = \big(\sum_{i,j}m_{ij}^2\big)^{1/2}$. Then $|Mh| \le \|M\|\,|h|$ for all $h \in \mathbb{R}^n$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-matrix-norm
The $i$-th entry of $Mh$ is $r_i\cdot h$, $r_i$ the $i$-th row; by Cauchy–Schwarz $|Mh|^2 = \sum_i(r_i\cdot h)^2 \le \sum_i|r_i|^2|h|^2 = \|M\|^2|h|^2$. $\blacksquare$
:::

::: {#thm-multivariable-chain-rule}
## Chain Rule

Let $f : S \to \mathbb{R}^m$ ($S \subseteq \mathbb{R}^n$) be differentiable at the interior point $a$, and $g : T \to \mathbb{R}^p$ ($T \subseteq \mathbb{R}^m$, $f(S) \subseteq T$) differentiable at the interior point $b = f(a)$. Then $g\circ f$ is differentiable at $a$ and
$$D(g\circ f)(a) = Dg(f(a))\,Df(a) \qquad (\text{a } p\times m \text{ times an } m\times n \text{ matrix}).$$
In coordinates, writing $y = f(x)$, $z = g(y)$:
$$\frac{\partial z_i}{\partial x_j}(a) = \sum_{k=1}^m\frac{\partial z_i}{\partial y_k}(b)\,\frac{\partial y_k}{\partial x_j}(a) .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-multivariable-chain-rule
Write $A = Df(a)$, $B = Dg(b)$ and
$$f(a + h) = b + Ah + |h|\,r(h), \qquad g(b + k) = g(b) + Bk + |k|\,s(k),$$
with $r(h) \to 0$ as $h \to 0$ and $s(k) \to 0$ as $k \to 0$; set $r(0) = 0$, $s(0) = 0$ so that $s$ is continuous at $0$. Put $k = f(a + h) - b = Ah + |h|r(h)$. Then
$$g(f(a + h)) - g(b) - BAh = |h|\,Br(h) + |k|\,s(k).$$
By @lem-matrix-norm, $|Br(h)| \le \|B\|\,|r(h)| \to 0$, and $|k| \le (\|A\| + |r(h)|)|h|$, so $\frac{|k|}{|h|}$ stays bounded while $k \to 0$ and hence $s(k) \to 0$. Dividing by $|h|$, the right side tends to $0$. The coordinate formula is the entry $(i, j)$ of the matrix product. $\blacksquare$
:::

For a curve $\gamma : I \to S$ and $f : S \to \mathbb{R}$ differentiable this reads
$$\frac{d}{dt}f(\gamma(t)) = Df(\gamma(t))\,\gamma'(t) = \sum_{i=1}^n\frac{\partial f}{\partial x_i}(\gamma(t))\,\gamma_i'(t),$$
the rate of change of $f$ seen by a moving observer.

::: {#exm-thermometer}
## A Moving Thermometer

The temperature is $T(x, y, z) = \frac{100}{1 + x^2 + y^2 + z^2}$ and a probe moves along $(x, y, z) = (1 + t, t^2, t - t^3)$. At $t = 1$ it is at $(2, 1, 0)$ with velocity $(1, 2, -2)$, and $\frac{\partial T}{\partial x} = \frac{-200x}{(1 + x^2 + y^2 + z^2)^2}$ etc. give $DT(2, 1, 0) = \big[-\frac{100}{9}\ \ -\frac{50}{9}\ \ 0\big]$. Hence $\frac{dT}{dt} = -\frac{100}{9} - \frac{100}{9} + 0 = -\frac{200}{9} \approx -22.2$.
:::

## 2\. Changes of Variables

::: {#exm-polar-chain}
## Polar Coordinates

Let $z = x^2 - y^2$ and $x = r\cos\theta$, $y = r\sin\theta$. Then
$$\Big[\frac{\partial z}{\partial r}\ \ \frac{\partial z}{\partial\theta}\Big] = [2x\ \ -2y]\begin{bmatrix}\cos\theta & -r\sin\theta\\ \sin\theta & r\cos\theta\end{bmatrix} = [2r\cos2\theta\ \ -2r^2\sin2\theta],$$
as also follows from $z = r^2\cos2\theta$ directly. For general $z = f(x, y)$: $z_r = f_x\cos\theta + f_y\sin\theta$, $z_\theta = -f_xr\sin\theta + f_yr\cos\theta$.
:::

::: {#exm-inverse-polar}
## Partial Derivatives of $r$ and $\theta$

On the half-plane $x > 0$, $r = \sqrt{x^2 + y^2}$, $\theta = \arctan\frac yx$, and direct differentiation gives
$$\frac{\partial r}{\partial x} = \cos\theta, \quad \frac{\partial r}{\partial y} = \sin\theta, \quad \frac{\partial\theta}{\partial x} = -\frac{\sin\theta}{r}, \quad \frac{\partial\theta}{\partial y} = \frac{\cos\theta}{r} .$$
This matrix is the inverse of $\frac{\partial(x, y)}{\partial(r, \theta)}$, as the chain rule applied to $(x, y) \mapsto (r, \theta) \mapsto (x, y)$, the identity, predicts. Note $\frac{\partial r}{\partial x} = \cos\theta \ne \frac{1}{\partial x/\partial r} = \frac{1}{\cos\theta}$: a partial derivative depends on which other variables are held fixed, so it is the matrices, not the entries, that are inverse.
:::

## 3\. Two Applications

::: {#exm-energy-conservation}
## Conservation of Energy

Let $x(t) \in \mathbb{R}^n$ be the positions of a mechanical system, $v = x'$ the velocities, masses $m_i > 0$, and suppose the force is $F = -\big(\frac{\partial U}{\partial x_1}, \dots, \frac{\partial U}{\partial x_n}\big)$ for a differentiable potential $U(x)$, with Newton's law $m_ix_i'' = F_i$. For the energy $E = U(x) + \frac12\sum m_iv_i^2$, the chain rule gives
$$\frac{dE}{dt} = \sum_i\frac{\partial U}{\partial x_i}x_i' + \sum_im_iv_iv_i' = \sum_i(-m_ix_i'')v_i + \sum_im_iv_ix_i'' = 0 .$$
:::

::: {#thm-euler-homogeneous}
## Euler's Theorem on Homogeneous Functions

Let $f$ be differentiable on the open orthant $S = \{x : x_i > 0\}$. Then $f$ is **homogeneous of degree $d$**, i.e. $f(tx) = t^df(x)$ for all $t > 0$, $x \in S$, iff
$$\sum_{i=1}^nx_i\frac{\partial f}{\partial x_i}(x) = d\,f(x)\qquad (x \in S).$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-euler-homogeneous
Fix $x$ and let $\varphi(t) = f(tx)$; by the chain rule $\varphi'(t) = \sum_ix_i\partial_if(tx)$. If $f$ is homogeneous, differentiate $f(tx) = t^df(x)$ at $t = 1$. Conversely, if the identity holds, $\psi(t) = t^{-d}\varphi(t)$ has $\psi'(t) = t^{-d-1}\big(t\varphi'(t) - d\varphi(t)\big) = t^{-d-1}\big(\sum_i tx_i\partial_if(tx) - df(tx)\big) = 0$, so $\psi(t) = \psi(1)$, i.e. $f(tx) = t^df(x)$. $\blacksquare$
:::

## 4\. The Mean Value Inequality

There is no mean value *equality* for vector-valued maps ($t \mapsto (\cos t, \sin t)$ on $[0, 2\pi]$), but there is an inequality.

::: {#thm-mean-value-inequality}
## Mean Value Theorem and Inequality

Let $f$ be differentiable on an open set containing the segment $[a, b] = \{a + t(b - a) : 0 \le t \le 1\}$.

1. If $f$ is real-valued, $f(b) - f(a) = Df(c)(b - a)$ for some $c$ on the segment.
2. If $f$ is $\mathbb{R}^m$-valued and $\|Df(x)\| \le M$ on the segment, then $|f(b) - f(a)| \le M|b - a|$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-mean-value-inequality
(1) $\varphi(t) = f(a + t(b - a))$ is differentiable with $\varphi'(t) = Df(a + t(b - a))(b - a)$ (chain rule); apply the [MVT](mean-value-theorem.ipynb#thm-mvt) to $\varphi$ on $[0, 1]$.
(2) We may assume $f(b) \ne f(a)$; let $u = \frac{f(b) - f(a)}{|f(b) - f(a)|}$ and apply (1) to the real function $u\cdot f$: for some $c$, $|f(b) - f(a)| = u\cdot\big(f(b) - f(a)\big) = u\cdot Df(c)(b - a) \le |Df(c)(b - a)| \le M|b - a|$ by @lem-matrix-norm. $\blacksquare$
:::

::: {#cor-zero-derivative}
## Zero Derivative

If $U$ is open, any two points of $U$ can be joined by a polygonal path in $U$ (e.g. $U$ convex), and $Df = 0$ on $U$, then $f$ is constant on $U$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-zero-derivative
By @thm-mean-value-inequality with $M = 0$, $f$ takes the same value at the two ends of every segment in $U$, hence along every polygonal path. $\blacksquare$
:::

```python

import sympy as sp

x, y, z, t, r, th = sp.symbols("x y z t r theta", real=True)
T = 100 / (1 + x**2 + y**2 + z**2)
path = {x: 1 + t, y: t**2, z: t - t**3}
print("dT/dt at t=1 (direct):", sp.diff(T.subs(path), t).subs(t, 1))
J = sp.Matrix([r * sp.cos(th), r * sp.sin(th)]).jacobian([r, th])
print("inverse of d(x,y)/d(r,theta):", sp.simplify(J.inv()))
Zp = (x**2 - y**2).subs({x: r * sp.cos(th), y: r * sp.sin(th)})
print("z_r, z_theta:", sp.simplify(sp.diff(Zp, r)), sp.simplify(sp.diff(Zp, th)))
xs = sp.symbols("x1:4", positive=True)
f = xs[0]**2 * xs[1] / xs[2] + xs[1]**3 / xs[0]          # homogeneous of degree 2
print("Euler: sum x_i f_i - 2 f =", sp.simplify(sum(v * sp.diff(f, v) for v in xs) - 2 * f))
```
