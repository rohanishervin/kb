# Approximating Measurable Sets by Open and Closed Sets

**Prerequisites:** [Lebesgue Measure](Lebesgue-measure.ipynb) · [Intervals, Open Sets, and Borel Sets Are Measurable](Borel-sets-and-measurability.ipynb)

Carathéodory's definition of measurability is relatively recent, and it has a strength and a weakness. Its strength is generality: it only needs an outer measure, so it works on every space where one is available, and it constructs many measures besides Lebesgue's; this is why most books use it. Its weakness is that it is not intuitive: the condition "for every set $A$" does not reveal where it comes from. In $\mathbb{R}^n$ there are equivalent definitions that are much closer to the original idea of outer measure, approximation **from outside**. One of them could have been taken as the definition, with Carathéodory's condition proved as a theorem. Throughout, $m$ is [Lebesgue measure](Lebesgue-measure.ipynb#def-lebesgue-measure) and $m^*$ outer measure on $\mathbb{R}^n$.

## 1\. Approximation From Outside by Open Sets

::: {#thm-outer-approximation}
## Outer Approximation by Open Sets

A set $E \subseteq \mathbb{R}^n$ is measurable if and only if for every $\varepsilon > 0$ there is an open set $O \supseteq E$ with
$$m^*(O \setminus E) < \varepsilon.$$
:::

::: {.callout-note}
## Intuition: A Thin Collar
Approximate $E$ from outside by open sets. $E$ is measurable exactly when the part of the approximating set that sticks out, the "collar" $O \setminus E$, can be made to have arbitrarily small outer measure. Non-measurability is caused by the boundary; this says that the boundary is not so wild that every open set containing $E$ must contain a substantial amount of extra mass.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-outer-approximation
**($\Rightarrow$)** Let $E$ be measurable and $\varepsilon > 0$. Fortunately the definition of outer measure is itself an approximation from outside.

*Case $m(E) < \infty$.* By the definition of $m^*(E)$ as an infimum, there is a cover of $E$ by open boxes $R_k$ with
$$\sum_k |R_k| < m^*(E) + \varepsilon.$$
The union $O = \bigcup_k R_k$ of open boxes is open and contains $E$. By countable subadditivity and $m^*(R_k) = |R_k|$ ([outer measure of a box](lebesgue-outer-measure.ipynb#thm-outer-measure-of-box)),
$$m^*(O) \le \sum_k m^*(R_k) = \sum_k |R_k| < m^*(E) + \varepsilon.$$
Now both $O$ (open) and $E$ are measurable, $m$ is a measure on $\mathcal{L}$, $E \subseteq O$ and $m(E) < \infty$; so [excision](measure.ipynb#prp-excision) gives
$$m^*(O \setminus E) = m(O) - m(E) < \varepsilon.$$

*Case $m(E) = \infty$.* Excision is not available ($\infty - \infty$), so cut $E$ into bounded pieces. Let $Q_k = [-k, k]^n$, $Q_0 = \emptyset$, and $E_k = E \cap (Q_k \setminus Q_{k-1})$. Each $E_k$ is measurable with $m(E_k) \le m(Q_k) < \infty$, so by the first case there are open $O_k \supseteq E_k$ with $m^*(O_k \setminus E_k) < \varepsilon / 2^k$. Then $O = \bigcup_k O_k$ is open, contains $E = \bigcup_k E_k$, and $O \setminus E \subseteq \bigcup_k (O_k \setminus E_k)$, so
$$m^*(O \setminus E) \le \sum_k \frac{\varepsilon}{2^k} = \varepsilon .$$
(Use $\varepsilon/2$ in place of $\varepsilon$ for strict inequality.)

If $E$ is *not* measurable, the first part of the argument still works: there are open $O \supseteq E$ with $m^*(O) < m^*(E) + \varepsilon$. What fails is the last step: $m^*(O \setminus E)$ need not equal $m^*(O) - m^*(E)$, and the collar can have large outer measure.

**($\Leftarrow$)** Suppose that for every $\varepsilon$ there is an open $O \supseteq E$ with $m^*(O \setminus E) < \varepsilon$. We do not check Carathéodory's condition directly. Take $\varepsilon = \frac1n$: there are open sets $O_n \supseteq E$ with
$$m^*(O_n \setminus E) < \frac1n .$$
As $n$ grows these are better and better approximations of $E$. Let
$$G = \bigcap_{n=1}^\infty O_n .$$
$G$ need not be open (a countable intersection of open sets need not be open), but it is measurable, as a countable intersection of open, hence measurable, sets. It contains $E$, and since $G \subseteq O_n$ for every $n$,
$$m^*(G \setminus E) \le m^*(O_n \setminus E) < \frac1n \quad \text{for all } n, \qquad \text{so } m^*(G \setminus E) = 0.$$
So $E$ is approximated from outside by a measurable set whose collar has measure **zero**. Sets of outer measure zero are [measurable](measurable-sets.ipynb#prp-null-sets-measurable), so $G \setminus E$ is measurable, and therefore
$$E = G \setminus (G \setminus E)$$
is measurable, being the difference of two measurable sets. $\blacksquare$
:::

The proof of ($\Leftarrow$) gave more than was asked.

::: {#def-g-delta-f-sigma}
## $G_\delta$ and $F_\sigma$ Sets

A **$G_\delta$ set** is a countable intersection of open sets. An **$F_\sigma$ set** is a countable union of closed sets. By De Morgan's laws, the complement of a $G_\delta$ set is an $F_\sigma$ set and vice versa. Both are Borel sets.
:::

::: {#cor-g-delta-approximation}
## Approximation by a $G_\delta$ Set

A set $E \subseteq \mathbb{R}^n$ is measurable if and only if there is a $G_\delta$ set $G \supseteq E$ with $m^*(G \setminus E) = 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-g-delta-approximation
If $E$ is measurable, the open sets $O_n$ of @thm-outer-approximation with $\varepsilon = \frac1n$ give $G = \bigcap_n O_n$, a $G_\delta$ set with $m^*(G \setminus E) = 0$, exactly as in the proof of ($\Leftarrow$). Conversely, if such a $G$ exists, then $E = G \setminus (G \setminus E)$ is the difference of a Borel set and a null set, hence measurable. $\blacksquare$
:::

## 2\. Approximation From Inside by Closed Sets

Taking complements turns open sets into closed sets and "containing $E$" into "contained in $E$".

::: {#thm-measurability-equivalences}
## Equivalent Forms of Measurability

For $E \subseteq \mathbb{R}^n$ the following are equivalent.

1. $E$ is Lebesgue measurable.
2. For every $\varepsilon > 0$ there is an open set $O \supseteq E$ with $m^*(O \setminus E) < \varepsilon$.
3. There is a $G_\delta$ set $G \supseteq E$ with $m^*(G \setminus E) = 0$.
4. For every $\varepsilon > 0$ there is a closed set $F \subseteq E$ with $m^*(E \setminus F) < \varepsilon$.
5. There is an $F_\sigma$ set $F \subseteq E$ with $m^*(E \setminus F) = 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-measurability-equivalences
(1) $\Leftrightarrow$ (2) is @thm-outer-approximation and (1) $\Leftrightarrow$ (3) is @cor-g-delta-approximation. For the last two, recall that $E$ is measurable if and only if $E^c$ is ([complements](measurable-sets.ipynb#prp-complement-measurable)).

(1) $\Leftrightarrow$ (4): $E$ is measurable $\iff$ $E^c$ is measurable $\iff$ for every $\varepsilon$ there is an open $O \supseteq E^c$ with $m^*(O \setminus E^c) < \varepsilon$. Put $F = O^c$: then $F$ is closed, $F \subseteq E$, and $O \setminus E^c = O \cap E = E \setminus F$. Conversely, from a closed $F \subseteq E$ we get the open $O = F^c \supseteq E^c$ with the same difference.

(1) $\Leftrightarrow$ (5): in the same way, $G \supseteq E^c$ is a $G_\delta$ set with $m^*(G \setminus E^c) = 0$ if and only if $F = G^c$ is an $F_\sigma$ set contained in $E$ with $m^*(E \setminus F) = m^*(G \setminus E^c) = 0$. $\blacksquare$
:::

::: {.callout-warning}
## From Inside: Closed Sets, Not Open Sets
Measurable sets can be approximated from outside by open sets and from inside by **closed** sets. They cannot in general be approximated from inside by open sets: the Cantor set, or the set of irrationals, contains no open interval at all, yet it contains plenty of closed sets. The figure below shows a closed set of irrationals of measure close to $1$.
:::

::: {.callout-note}
## Where These Characterizations Apply
These equivalent definitions use open and closed sets, so they only make sense where there is a topology, as in $\mathbb{R}^n$ (or more generally in metric and topological spaces). Many applications of measure theory, probability in particular, take place on abstract spaces with no notion of open set. There Carathéodory's definition, which only needs an outer measure, is the one that works. The characterizations above give the better intuition for Lebesgue measure.
:::

The figure shows both approximations. Left: the Cantor set $C$ is covered by open sets $O_n$, obtained by slightly enlarging the $2^n$ intervals of $C_n$; the collar $O_n \setminus C$ has measure $m(O_n) \to 0$. Right: removing an open interval of length $\varepsilon/2^k$ around the $k$-th rational from $[0,1]$ (here for the first $60$ rationals, $\varepsilon = 0.2$) leaves a closed set; removing such intervals around **all** rationals leaves a closed set $F$ consisting only of irrationals, with $m(F) \ge 1 - \varepsilon$.

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

from fractions import Fraction as Fr

def cantor_stage(n):
    ivs = [(0.0, 1.0)]
    for _ in range(n):
        ivs = [p for (a, b) in ivs for p in ((a, a + (b - a) / 3), (b - (b - a) / 3, b))]
    return ivs

def union_length(ivs):
    ivs = sorted(ivs); total, (ca, cb) = 0.0, ivs[0]
    for a, b in ivs[1:]:
        if a > cb: total += cb - ca; ca, cb = a, b
        else: cb = max(cb, b)
    return total + cb - ca

fig, (ax, ax2) = plt.subplots(1, 2, figsize=(12, 4.2))
collars = []
for n in range(1, 7):
    delta = 3.0 ** -n / 6
    O = [(a - delta, b + delta) for a, b in cantor_stage(n)]
    collars.append(union_length(O))
    if n <= 5:
        for a, b in O:
            ax.plot([a, b], [-n, -n], color=AQUA, lw=8, solid_capstyle="butt", alpha=0.6)
        for a, b in cantor_stage(6):
            ax.plot([a, b], [-n - 0.16, -n - 0.16], color=BLUE, lw=3.5, solid_capstyle="butt")
        ax.text(1.06, -n, f"$m(O_{n} \\setminus C) = {collars[-1]:.3f}$", va="center", fontsize=9, color=MUTED)
ax.set_xlim(-0.05, 1.45); ax.set_yticks([]); ax.grid(False); ax.spines["left"].set_visible(False)
ax.set_title("Outside: open $O_n \\supseteq C$ (aqua) around the Cantor set (blue)")

rats = []
q = 1
while len(rats) < 60:
    for p in range(q + 1):
        r = Fr(p, q)
        if r not in rats: rats.append(r)
    q += 1
eps = 0.2
removed = [(float(r) - eps / 2 ** (k + 1), float(r) + eps / 2 ** (k + 1)) for k, r in enumerate(rats[:60], start=1)]
ax2.plot([0, 1], [0, 0], color=BLUE, lw=10, solid_capstyle="butt", label="closed set $F$ that remains")
for a, b in removed:
    ax2.plot([max(a, 0), min(b, 1)], [0, 0], color="white", lw=12, solid_capstyle="butt")
    ax2.plot([max(a, 0), min(b, 1)], [0.12, 0.12], color=RED, lw=3, solid_capstyle="butt")
ax2.plot([], [], color=RED, lw=3, label=r"removed intervals of length $\varepsilon/2^k$ around rationals")
left = 1 - union_length([(max(a, 0), min(b, 1)) for a, b in removed])
ax2.set_ylim(-0.4, 0.5); ax2.set_yticks([]); ax2.grid(False); ax2.spines["left"].set_visible(False)
ax2.legend(loc="upper center", fontsize=9)
ax2.set_title(f"Inside: closed set left after 60 removals\nhas measure {left:.3f} $\\geq 1 - \\varepsilon$")
plt.tight_layout(); plt.show()
```
