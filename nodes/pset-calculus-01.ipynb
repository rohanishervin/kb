# Problem Set: Calculus I

Problems over all 38 notes in [Calculus I](index-calculus.ipynb), in the order of the notes. They are meant as a final-exam style review: read the notes once, then solve these without looking back. Most problems mix a computation with a short proof; the last one is a challenge.

Every problem was written for this set. A **Source** line is added only where the statement is a well-known classical result.


## Problem 1: Irrationality and the Euclidean Algorithm

**Covers:** [irrationality of $\sqrt n$](irrationality-and-euclidean-algorithm.ipynb#thm-sqrt-irrational), [rationality criterion](irrationality-and-euclidean-algorithm.ipynb#thm-euclid-rational), [periodic expansions](existence-of-decimal-expansion-in-R.ipynb#thm-rational-periodic)

**Problem Statement:**

**(a)** Prove that $\sqrt2 + \sqrt3$ is irrational.

**(b)** Run the Euclidean algorithm on $a_0 = \sqrt2$, $a_1 = 1$. Show that $\frac{a_k}{a_{k+1}} = \sqrt2 + 1$ for every $k \ge 1$, and conclude again that $\sqrt2 \notin \mathbb{Q}$. What is the continued fraction of $\sqrt2$?

**(c)** Write $0.3\overline{18} = 0.31818\dots$ as a fraction in lowest terms, and find the period of the decimal expansion of $\frac{1}{13}$ by long division.

**(d)** Prove that $0.123456789101112131415\dots$ (the positive integers written one after another) is irrational.


## Problem 2: Suprema and Completeness

**Covers:** [supremum](supremum-infimum.ipynb), [decimal form of completeness](axiom-of-completeness.ipynb#thm-completeness-equivalence), [$n$-th roots](existence-of-nth-roots-in-R.ipynb#thm-nth-root)

**Problem Statement:**

**(a)** Let $A, B \subseteq \mathbb{R}$ be nonempty and bounded above, and $A + B = \{a + b : a \in A, b \in B\}$. Prove $\sup(A + B) = \sup A + \sup B$.

**(b)** Prove that a nonempty set $S$ bounded below has an infimum, and $\inf S = -\sup(-S)$.

**(c)** Find $\sup$ and $\inf$ of $\big\{\frac{n}{n+1} + \frac{(-1)^n}{n} : n \in \mathbb{N}\big\}$; say whether they are attained.

**(d)** Following the proof of the existence of roots, prove directly that $x = \sup\{s \ge 0 : s^3 \le 2\}$ satisfies $x^3 = 2$.

## Problem 4: A Cubic Equation

**Covers:** [solution of the cubic](complex-numbers.ipynb#thm-cubic-formula)

**Problem Statement:**

**(a)** Solve $t^3 - 15t - 4 = 0$ with the cubic formula. (You will need a cube root of $2 + 11i$; try $a + bi$ with small integers.) Check that all three roots are real.
