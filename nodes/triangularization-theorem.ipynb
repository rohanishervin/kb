# The Triangularization Theorem

**Prerequisites:** [Eigenvalues and Eigenvectors](eigenvalues-and-eigenvectors.ipynb) · [Change of Basis](change-of-basis-and-similarity.ipynb) · [Diagonalization](diagonalization.ipynb)

Not every operator can be [diagonalized](diagonalization.ipynb). But over an [algebraically closed field](eigenvalues-and-eigenvectors.ipynb#def-algebraically-closed-field) such as $\mathbb{C}$, every operator can at least be brought to **upper triangular** form, with the eigenvalues on the diagonal. The proof needs only one eigenvector at a time and an induction on the dimension.

Throughout, $V$ is a [vector space](vector-spaces.ipynb) of dimension $n$ over a field $K$ and $f : V \to V$ is a linear operator.

## 1\. Statement

::: {#def-triangular-matrix}
## Triangular Matrices

$M = (m_{ij}) \in M_n(K)$ is **upper triangular** if $m_{ij} = 0$ whenever $i > j$ (all entries below the diagonal vanish), and **lower triangular** if $m_{ij} = 0$ whenever $i < j$:
$$M = \begin{pmatrix} \lambda_1 & * & \cdots & * \\ 0 & \lambda_2 & \cdots & * \\ \vdots & \ddots & \ddots & \vdots \\ 0 & \cdots & 0 & \lambda_n \end{pmatrix}.$$
:::

::: {#thm-triangularization}
## Triangularization Theorem

If $K$ is algebraically closed, then $V$ has an ordered basis $\mathcal{B}$ such that $[f]_{\mathcal{B}}$ is upper triangular. Equivalently, for every $A \in M_n(K)$ there is an invertible $P$ with $P^{-1} A P$ upper triangular.
:::

The two forms are equivalent because changing the basis replaces the matrix by a [similar](change-of-basis-and-similarity.ipynb#prp-similarity-equivalence) one, and every similar matrix arises this way.

::: {#prp-triangular-diagonal-eigenvalues}
## The Diagonal Holds the Eigenvalues

If $[f]_{\mathcal{B}}$ is upper triangular with diagonal entries $\lambda_1, \dots, \lambda_n$, then
$$p_f(\lambda) = (\lambda - \lambda_1)(\lambda - \lambda_2)\cdots(\lambda - \lambda_n).$$
So the diagonal entries are exactly the eigenvalues of $f$, each repeated according to its [algebraic multiplicity](diagonalization.ipynb#def-multiplicities), and $\det f = \lambda_1 \cdots \lambda_n$, $\operatorname{tr} f = \lambda_1 + \dots + \lambda_n$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-triangular-diagonal-eigenvalues
$\lambda I - [f]_{\mathcal{B}}$ is upper triangular with diagonal entries $\lambda - \lambda_i$, and the [determinant of a triangular matrix](determinant.ipynb#prp-det-triangular) is the product of its diagonal entries. The roots of $p_f$ are the [eigenvalues](eigenvalues-and-eigenvectors.ipynb#thm-characteristic-equation). The formulas for $\det$ and $\operatorname{tr}$ follow from the diagonal directly. $\blacksquare$
:::

::: {.callout-note}
## Intuition: A Chain of Invariant Subspaces
If $[f]_{\mathcal{B}}$ is upper triangular for $\mathcal{B} = (v_1, \dots, v_n)$, then $f(v_k)$ is a combination of $v_1, \dots, v_k$ only. So each of the subspaces
$$\operatorname{span}(v_1) \subset \operatorname{span}(v_1, v_2) \subset \dots \subset \operatorname{span}(v_1, \dots, v_n) = V$$
is mapped into itself by $f$. In particular $v_1$ is an eigenvector. The theorem says that over $\mathbb{C}$ every operator has such a chain of invariant subspaces, one of each dimension.
:::

## 2\. Proof

::: {.callout-tip collapse="true"}
## Proof of @thm-triangularization
We prove the matrix form by induction on $n$. For $n = 1$ every matrix is triangular.

Let $n \ge 2$ and assume the result for $(n-1) \times (n-1)$ matrices. Let $A \in M_n(K)$ and $f(x) = Ax$.

**Step 1: one eigenvector.** $p_A$ has degree $n \ge 1$, and $K$ is algebraically closed, so $p_A$ has a root $\lambda \in K$, which is an eigenvalue. Let $v_1$ be an eigenvector, $A v_1 = \lambda v_1$. (This is the only place where algebraic closure is used.)

**Step 2: a first change of basis.** $v_1 \ne 0$, so it [extends to a basis](basis-and-dimension.ipynb#thm-extend-to-basis) $\mathcal{B}' = (v_1, \dots, v_n)$ of $K^n$. The first column of $[f]_{\mathcal{B}'}$ holds the coordinates of $f(v_1) = \lambda v_1$, namely $(\lambda, 0, \dots, 0)$. So, with $Q$ the matrix whose columns are $v_1, \dots, v_n$,
$$Q^{-1} A Q = \begin{pmatrix} \lambda & w \\ 0 & N \end{pmatrix},$$
where $w$ is a row of length $n - 1$ and $N \in M_{n-1}(K)$.

**Step 3: induction.** By the induction hypothesis there is an invertible $R \in M_{n-1}(K)$ with $R^{-1} N R = T$ upper triangular.

**Step 4: enlarge $R$.** Let $\tilde R = \begin{pmatrix} 1 & 0 \\ 0 & R \end{pmatrix}$. Block multiplication shows $\tilde R^{-1} = \begin{pmatrix} 1 & 0 \\ 0 & R^{-1} \end{pmatrix}$, and
$$\tilde R^{-1} \begin{pmatrix} \lambda & w \\ 0 & N \end{pmatrix} \tilde R = \begin{pmatrix} 1 & 0 \\ 0 & R^{-1} \end{pmatrix}\begin{pmatrix} \lambda & w R \\ 0 & N R \end{pmatrix} = \begin{pmatrix} \lambda & w R \\ 0 & R^{-1} N R \end{pmatrix} = \begin{pmatrix} \lambda & w R \\ 0 & T \end{pmatrix}.$$
The first column is $(\lambda, 0, \dots, 0)$ and $T$ is upper triangular, so this matrix is upper triangular.

**Step 5: conclusion.** With $P = Q \tilde R$ (invertible as a product of invertible matrices), $P^{-1} A P = \tilde R^{-1}(Q^{-1} A Q)\tilde R$ is upper triangular. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Peel Off One Eigenvector at a Time
Putting one eigenvector first in the basis clears the first column below the diagonal. What remains is the smaller block $N$, which is handled by induction. The row $w$ is carried along (it becomes $wR$) but never affects the zeros below the diagonal.
:::

::: {#exm-triangularize}
## Triangularizing a $2 \times 2$ Matrix

$A = \begin{pmatrix} 3 & 1 \\ -1 & 1 \end{pmatrix}$ has $p_A(\lambda) = (\lambda - 3)(\lambda - 1) + 1 = (\lambda - 2)^2$. The eigenspace $E_2 = \ker\begin{pmatrix} 1 & 1 \\ -1 & -1 \end{pmatrix} = \operatorname{span}(1, -1)$ is only one-dimensional, so $A$ is [not diagonalizable](diagonalization.ipynb#prp-multiplicity-inequality). Following the proof, take $v_1 = (1, -1)$, extend by $v_2 = (0, 1)$, and let $P$ have columns $v_1, v_2$. Since $A v_2 = (1, 1) = v_1 + 2 v_2$,
$$P^{-1} A P = \begin{pmatrix} 2 & 1 \\ 0 & 2 \end{pmatrix},$$
upper triangular with the double eigenvalue $2$ on the diagonal.
:::

## 3\. Triangular Form Is the Best Possible in General

::: {#prp-shear-not-diagonalizable}
## The Shear Cannot Be Diagonalized

$J = \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}$ is upper triangular but not similar to a diagonal matrix, over any field.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-shear-not-diagonalizable
The only eigenvalue is $1$ (@prp-triangular-diagonal-eigenvalues). $(J - I)v = (v_2, 0)$, so $E_1 = \operatorname{span}(e_1)$ is one-dimensional, and there is no basis of eigenvectors. Alternatively: if $P^{-1} J P = D$ were diagonal, its diagonal would hold the eigenvalues, so $D = I$ and $J = P I P^{-1} = I$, which is false. $\blacksquare$
:::

The animation shows the shear $J$. Every horizontal line slides sideways by an amount equal to its height. The yellow vector on the $x$-axis stays where it is (eigenvalue $1$), and every other direction is tilted towards the $x$-axis. There is only one invariant line, so no basis of eigenvectors exists, and triangular form is the best possible.


```python

from la_manim import *

class ShearHasOneEigenline(GridScene):
    def construct(self):
        self.add(caption(r"J = \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}"))
        for k in range(1, 12):
            t = np.pi * k / 12
            self.vec([1.4 * np.cos(t), 1.4 * np.sin(t)], C_MUTED)
        self.vec([1.6, 0], C_YELLOW, r"e_1", DOWN)
        self.vec([-1.6, 0], C_YELLOW)
        square = self.track(Polygon(ORIGIN, RIGHT, RIGHT + UP, UP, stroke_color=C_AQUA, fill_color=C_AQUA, fill_opacity=0.25))
        self.add(square)
        self.wait(1)
        self.transform([[1, 1], [0, 1]], run_time=3.5)
        msg = text_caption(r"Only the $x$-axis is invariant: $E_1 = \operatorname{span}(e_1)$, $\dim E_1 = 1 < 2$")
        self.play(FadeIn(msg)); self.wait(2.5)

show(ShearHasOneEigenline)
```


::: {.callout-note}
## An Application
Triangularization gives a second proof of the [Cayley-Hamilton Theorem](cayley-hamilton-theorem.ipynb#sec-ch-density) over $\mathbb{C}$, by perturbing the diagonal entries until they are distinct.
:::

The code implements the inductive proof literally (one eigenvector, extend to a basis, recurse on the lower-right block) and checks the result on a random $5 \times 5$ real matrix, triangularized over $\mathbb{C}$.


```python

import numpy as np

def triangularize(A):
    """Return (P, T) with T = P^{-1} A P upper triangular, following the inductive proof."""
    A = np.asarray(A, dtype=complex); n = A.shape[0]
    if n == 1:
        return np.eye(1, dtype=complex), A.copy()
    _, vecs = np.linalg.eig(A)
    v1 = vecs[:, 0]                                         # step 1: one eigenvector
    Q, _ = np.linalg.qr(np.column_stack([v1, np.eye(n)]))   # step 2: extend v1 to a basis
    Q = Q[:, :n]
    B = np.linalg.solve(Q, A @ Q)                           # = [[lambda, w], [0, N]]
    R, _ = triangularize(B[1:, 1:])                         # step 3: recurse on N
    R_big = np.eye(n, dtype=complex); R_big[1:, 1:] = R     # step 4: enlarge R
    P = Q @ R_big                                           # step 5
    return P, np.linalg.solve(P, A @ P)

rng = np.random.default_rng(0)
A = rng.integers(-4, 5, size=(5, 5)).astype(float)
P, T = triangularize(A)
print("below-diagonal part of P^-1 A P is zero:", np.allclose(np.tril(T, -1), 0, atol=1e-9))
print("diagonal of T:   ", np.round(np.sort_complex(np.diag(T)), 4))
print("eigenvalues of A:", np.round(np.sort_complex(np.linalg.eigvals(A)), 4))

A2 = np.array([[3, 1], [-1, 1]]); P2 = np.array([[1, 0], [-1, 1]])
print("\nexample: P^-1 A P =\n", np.linalg.inv(P2) @ A2 @ P2)
```
