# The Logarithm and the Exponential Function

**Prerequisites:** [The Fundamental Theorem of Calculus](fundamental-theorem-of-calculus.ipynb) · [The Chain Rule and Derivatives of Inverse Functions](chain-rule-and-inverse-functions.ipynb#thm-inverse-derivative) · [Monotone Functions and Continuous Inverses](monotone-and-inverse-functions.ipynb)

## 1\. The Logarithm

::: {#def-logarithm}
## Natural Logarithm

For $x > 0$, $\displaystyle \ln x = \int_1^x \frac{dt}{t}$.
:::

::: {#thm-log-properties}
## Properties of $\ln$

1. $\ln$ is differentiable on $(0, \infty)$ with $\ln' x = \frac1x$; it is strictly increasing and $\ln 1 = 0$.
2. $\ln(xy) = \ln x + \ln y$, $\ln\frac{x}{y} = \ln x - \ln y$, and $\ln(x^r) = r\ln x$ for $r \in \mathbb{Q}$.
3. $\ln : (0, \infty) \to \mathbb{R}$ is a bijection; $\ln x \to \pm\infty$ as $x \to \infty$, resp. $x \to 0^+$.
4. $\ln x \le x - 1$, with equality only at $x = 1$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-log-properties
(1) [FTC](fundamental-theorem-of-calculus.ipynb#thm-ftc); $\frac1x > 0$.
(2) Fix $y > 0$; $g(x) = \ln(xy) - \ln x$ has $g'(x) = \frac{y}{xy} - \frac1x = 0$, so $g$ is constant, $g(x) = g(1) = \ln y$. Then $\ln\frac xy + \ln y = \ln x$. For $r = \frac mn$: $\ln(x^{m}) = m\ln x$ by induction (and for $m < 0$ via $\ln\frac1x = -\ln x$), and $n\ln(x^{1/n}) = \ln x$.
(3) $\ln 2 > 0$ and $\ln(2^n) = n\ln 2 \to \infty$; $\ln(2^{-n}) \to -\infty$. Being continuous and increasing on an interval, $\ln$ maps $(0, \infty)$ onto an interval ([IVT](intermediate-value-theorem.ipynb#cor-image-interval)) that is unbounded in both directions, i.e. onto $\mathbb{R}$.
(4) $h(x) = x - 1 - \ln x$ has $h' = 1 - \frac1x$, negative on $(0,1)$ and positive on $(1, \infty)$, so $h > h(1) = 0$ for $x \ne 1$. $\blacksquare$
:::

## 2\. The Exponential Function and Real Powers

::: {#def-exponential}
## Exponential Function

$\exp : \mathbb{R} \to (0, \infty)$ is the inverse of $\ln$, and $e = \exp(1)$, i.e. $\ln e = 1$.
:::

::: {#thm-exp-properties}
## Properties of $\exp$

1. $\exp$ is differentiable with $\exp' = \exp$, strictly increasing, $\exp 0 = 1$.
2. $\exp(x + y) = \exp x\,\exp y$, and $\exp(r) = e^r$ for $r \in \mathbb{Q}$.
3. $\exp x \ge 1 + x$, with equality only at $0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-exp-properties
(1) By the [inverse function theorem](chain-rule-and-inverse-functions.ipynb#thm-inverse-derivative), $\exp'(y) = \frac{1}{\ln'(\exp y)} = \exp y$.
(2) $\ln(\exp x \exp y) = x + y$; apply $\exp$. And $\ln(e^r) = r\ln e = r$.
(3) Put $x = \ln u$ in $\ln u \le u - 1$. $\blacksquare$
:::

::: {#def-real-power}
## Real Powers

For $a > 0$ and $x \in \mathbb{R}$, $a^x = \exp(x\ln a)$. By (2) this agrees with the [rational powers](existence-of-nth-roots-in-R.ipynb#def-rational-power) when $x \in \mathbb{Q}$, and $e^x = \exp x$. The rules $a^{x+y} = a^xa^y$, $(a^x)^y = a^{xy}$, $(ab)^x = a^xb^x$ follow from those of $\exp$ and $\ln$.
:::

Derivatives: $(a^x)' = a^x\ln a$; for $x > 0$ and real $\alpha$, $(x^\alpha)' = \big(e^{\alpha\ln x}\big)' = x^\alpha\frac{\alpha}{x} = \alpha x^{\alpha - 1}$.

::: {#prp-exp-log-limits}
## Growth Rates

For every $n \in \mathbb{N}$ and $\alpha > 0$:
$$\lim_{x\to\infty}\frac{x^n}{e^x} = 0, \qquad \lim_{x\to\infty}\frac{\ln x}{x^\alpha} = 0, \qquad \lim_{x \to 0^+} x^\alpha\ln x = 0, \qquad \lim_{n\to\infty}\Big(1 + \frac1n\Big)^n = e .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-exp-log-limits
*First.* $g_k(x) = e^x - \sum_{j=0}^k \frac{x^j}{j!}$ satisfies $g_k(0) = 0$ and $g_k' = g_{k-1}$, with $g_0 = e^x - 1 \ge 0$ for $x \ge 0$; by induction and the MVT, $g_k \ge 0$ on $[0, \infty)$. So $e^x \ge \frac{x^{n+1}}{(n+1)!}$ and $\frac{x^n}{e^x} \le \frac{(n+1)!}{x} \to 0$.
*Second.* Put $x = e^{t}$: $\frac{\ln x}{x^\alpha} = \frac{t}{e^{\alpha t}} = \frac1\alpha\cdot\frac{\alpha t}{e^{\alpha t}} \to 0$ as $t \to \infty$. *Third.* Put $x = \frac1y$: $x^\alpha\ln x = -\frac{\ln y}{y^\alpha} \to 0$.
*Fourth.* $n\ln(1 + \frac1n) = \frac{\ln(1 + h) - \ln 1}{h}$ with $h = \frac1n$, which tends to $\ln'(1) = 1$; apply the continuous $\exp$. $\blacksquare$
:::

## 3\. Hyperbolic Functions

$$\sinh x = \frac{e^x - e^{-x}}{2}, \qquad \cosh x = \frac{e^x + e^{-x}}{2}, \qquad \tanh x = \frac{\sinh x}{\cosh x}.$$
Then $\cosh^2 - \sinh^2 = 1$, $\sinh' = \cosh$, $\cosh' = \sinh$, $\tanh' = 1 - \tanh^2$. $\sinh$ is a strictly increasing bijection of $\mathbb{R}$, with inverse $\operatorname{arsinh} x = \ln\big(x + \sqrt{x^2 + 1}\big)$ (solve $e^y - e^{-y} = 2x$ for $e^y > 0$); $\cosh$ is a bijection $[0, \infty) \to [1, \infty)$ with inverse $\operatorname{arcosh} x = \ln\big(x + \sqrt{x^2 - 1}\big)$. Their derivatives are $\frac{1}{\sqrt{x^2 + 1}}$ and $\frac{1}{\sqrt{x^2 - 1}}$ ($x > 1$).

## 4\. More Antiderivatives

::: {#exm-log-antiderivatives}
## Antiderivatives Involving $\ln$ and $\exp$

| $f(x)$ | $\dfrac1x$ | $e^x$ | $a^x$ | $\sinh x$ | $\cosh x$ | $\tan x$ | $\dfrac{1}{\sqrt{x^2+1}}$ | $\dfrac{1}{\sqrt{x^2-1}}$ ($x>1$) |
|---|---|---|---|---|---|---|---|---|
| $F(x)$ | $\ln\vert x\vert$ | $e^x$ | $\dfrac{a^x}{\ln a}$ | $\cosh x$ | $\sinh x$ | $-\ln\vert\cos x\vert$ | $\operatorname{arsinh} x$ | $\operatorname{arcosh} x$ |

For $x < 0$, $\big(\ln(-x)\big)' = \frac{-1}{-x} = \frac1x$; so $\ln|x|$ is an antiderivative of $\frac1x$ on each of $(-\infty, 0)$ and $(0, \infty)$, and the general antiderivative on $\mathbb{R} \setminus \{0\}$ is $\ln x + C_1$ ($x > 0$), $\ln(-x) + C_2$ ($x < 0$).
:::

::: {#exm-ln2-sum}
## A Limit of Sums

$\displaystyle\lim_{n\to\infty}\Big(\frac{1}{n+1} + \frac{1}{n+2} + \dots + \frac{1}{2n}\Big) = \lim_{n\to\infty}\frac1n\sum_{i=1}^n\frac{1}{1 + i/n} = \int_1^2\frac{dx}{x} = \ln 2$, by [Riemann sums](properties-of-the-riemann-integral.ipynb#thm-riemann-sums) for $\frac1x$ on $[1, 2]$ with the points $1 + \frac in$.
:::

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

fig, (ax, ax2) = plt.subplots(1, 2, figsize=(11, 4.3))
t = np.linspace(0.3, 3.2, 300)
ax.plot(t, 1 / t, color=INK, lw=2, label="$y = 1/t$")
tt = np.linspace(1, np.e, 200)
ax.fill_between(tt, 1 / tt, color=VIOLET, alpha=0.25, label=r"area $\ln e = 1$")
ax.axvline(np.e, color=VIOLET, lw=1, ls=":")
ax.text(np.e, -0.12, "$e$", ha="center", color=VIOLET)
ax.legend(fontsize=9); ax.set_ylim(0, 3.4); ax.set_title(r"$\ln x = \int_1^x dt/t$")

x = np.linspace(-2.5, 2.5, 400)
xp = np.linspace(0.08, 6, 400)
ax2.plot(x, np.exp(x), color=BLUE, lw=2, label=r"$e^x$")
ax2.plot(xp, np.log(xp), color=ORANGE, lw=2, label=r"$\ln x$")
ax2.plot(x, 1 + x, color=AQUA, lw=1.2, ls="--", label=r"$1 + x \leq e^x$")
ax2.plot([-3, 6], [-3, 6], color=MUTED, lw=1, ls=":")
ax2.set_xlim(-2.5, 6); ax2.set_ylim(-2.5, 6); ax2.set_aspect("equal"); ax2.legend(fontsize=9)
ax2.set_title("$\\exp$ and $\\ln$ are inverse functions")
plt.tight_layout(); plt.show()
print("(1+1/n)^n for n = 10, 1000, 10^6:", [(1 + 1/n) ** n for n in (10, 1000, 10**6)], "| e =", np.e)
```
