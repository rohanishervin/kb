# The Spaces $L^1$ and $L^p$

**Prerequisites:** [Integrable Functions and Linearity of the Integral](integrable-functions.ipynb) · [The Integral of a Nonnegative Measurable Function](lebesgue-integral.ipynb) · [Normed Vector Spaces](normed-vector-spaces.ipynb) · [Function Spaces: Pointwise and Uniform Convergence](function-spaces.ipynb)

With the integral available, measure spaces produce new function spaces, among the most useful in analysis. This note introduces them only far enough to see how they arise and what the one subtlety is; their deeper theory (Hölder's and Minkowski's inequalities, completeness) is not developed here. Throughout, $(X, \mathcal{M}, \mu)$ is a measure space.

## 1\. Distances Between Functions

The kinds of convergence of functions met so far are:

- **pointwise** convergence;
- **uniform** convergence, which comes from the [sup norm](function-spaces.ipynb#uniform-convergence-the-strong-form) $\|f - g\|_\infty = \sup_x |f(x) - g(x)|$, the **largest** vertical distance between the two graphs;
- convergence **almost everywhere**.

Pointwise and almost-everywhere convergence do not come from any norm. With a measure one can define distances that control the **average** vertical distance between two graphs instead of the maximal one: integrate $|f(x) - g(x)|$ rather than taking its supremum.

::: {#def-L1-distance}
## The $L^1$ Distance

For integrable $f, g$,
$$\|f - g\|_1 = \int_X |f - g|\, d\mu, \qquad \text{in particular} \qquad \|f\|_1 = \int_X |f|\, d\mu .$$
:::

For functions on $\mathbb{R}$, $\|f - g\|_1$ is the area of the region between the two graphs.

## 2\. Is It a Norm?

A [norm](normed-vector-spaces.ipynb#definition-of-a-norm) on a vector space must satisfy:

1. $\|f\| \ge 0$, and $\|f\| = 0$ only for $f = 0$;
2. $\|c f\| = |c|\, \|f\|$;
3. $\|f + g\| \le \|f\| + \|g\|$.

::: {#prp-L1-seminorm}
## Properties of $\|\cdot\|_1$

On the vector space of integrable functions, $\|\cdot\|_1$ satisfies 2 and 3, and $\|f\|_1 \ge 0$. However, $\|f\|_1 = 0$ holds if and only if $f = 0$ **almost everywhere**.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-L1-seminorm
$\|f\|_1 \ge 0$ because $|f| \ge 0$. For 2, $|c f| = |c|\,|f|$ and constants come out of the integral. For 3, apply the triangle inequality inside the integral, $|f + g| \le |f| + |g|$, then monotonicity and additivity of the integral: $\int |f + g| \le \int (|f| + |g|) = \int |f| + \int |g|$. Finally, $\int |f| = 0$ if and only if $|f| = 0$ a.e., by the [vanishing criterion](lebesgue-integral.ipynb#prp-finite-zero-integral) (parts 3 and 4). $\blacksquare$
:::

So property 1 fails: a function that is $0$ everywhere except for the value $1$ at one point has $\|f\|_1 = 0$ but is not the zero function.

## 3\. Identifying Functions That Agree Almost Everywhere

The remedy is to stop distinguishing functions that differ only on a null set; they are [the same as far as integration is concerned](limits-of-measurable-functions.ipynb#def-ae-equality).

::: {#def-ae-equivalence}
## Equality a.e. as an Equivalence Relation

For measurable $f, g$ put $f \sim g$ if $f = g$ a.e.
:::

This is an [equivalence relation](pre-set-theory.ipynb#equivalence-relations-and-classes): it is reflexive and symmetric, and transitive because if $f = g$ off a null set $N_1$ and $g = h$ off a null set $N_2$, then $f = h$ off $N_1 \cup N_2$, which is null.

::: {#def-L1-space}
## The Space $L^1(\mu)$

$L^1(\mu)$ is the set of equivalence classes $[f]$ of integrable functions under $\sim$, with
$$[f] + [g] = [f + g], \qquad c[f] = [cf], \qquad \|[f]\|_1 = \int_X |f|\, d\mu .$$
:::

::: {#prp-L1-normed-space}
## $L^1(\mu)$ Is a Normed Vector Space

The operations and the norm on $L^1(\mu)$ are well defined, and $\|\cdot\|_1$ is a norm on $L^1(\mu)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-L1-normed-space
*Well defined.* If $f \sim f'$ and $g \sim g'$, then $f + g = f' + g'$ off the union of two null sets, so $f + g \sim f' + g'$; similarly $cf \sim cf'$. And $|f| = |f'|$ a.e., so $\int |f| = \int |f'|$ because integrals do not see null sets.

*Norm.* Properties 2 and 3 are inherited from @prp-L1-seminorm. For property 1: if $\|[f]\|_1 = 0$, then $f = 0$ a.e., that is $f \sim 0$, so $[f]$ is the zero element $[0]$ of $L^1(\mu)$. $\blacksquare$
:::

In practice one keeps speaking of "functions in $L^1$", remembering that they are only determined up to null sets.

## 4\. The Spaces $L^p$

The same construction works with other powers.

::: {#def-Lp-norm}
## The $L^p$ Norm

For $1 \le p < \infty$ and measurable $f$,
$$\|f\|_p = \Big(\int_X |f|^p\, d\mu\Big)^{1/p} .$$
$L^p(\mu)$ is the set of equivalence classes of measurable $f$ with $\|f\|_p < \infty$.
:::

($|f|^p$ is measurable as a [continuous function of a measurable one](measurable-functions.ipynb#thm-continuous-of-measurable).) Properties 1 and 2 are proved as for $L^1$. The triangle inequality $\|f + g\|_p \le \|f\|_p + \|g\|_p$, Minkowski's inequality, is harder; it follows from Hölder's inequality and is not proved here. These spaces are extremely useful, and their most important property is that they are **complete**: every Cauchy sequence in $L^p(\mu)$ converges in $L^p(\mu)$, so they are Banach spaces. This is one reason why [completeness of the measure space](measure-properties-and-completion.ipynb#def-complete-measure-space) matters.

## 5\. Different Norms, Different Convergence

Convergence in $L^1$ and uniform convergence are genuinely different. For the [escape-of-mass sequences](fatous-lemma-and-convergence-theorems.ipynb#exm-escape-of-mass) and for $x^n$ on $[0,1]$:

| Sequence | $\|f_n\|_\infty$ | $\|f_n\|_1$ | $\|f_n\|_2$ | pointwise limit |
|---|---|---|---|---|
| $\frac1n \mathbf{1}_{[0,n]}$ | $\frac1n \to 0$ | $1$ | $n^{-1/2} \to 0$ | $0$ |
| $2^n \mathbf{1}_{[2^{-n},\, 2^{1-n}]}$ | $2^n \to \infty$ | $1$ | $2^{n/2} \to \infty$ | $0$ |
| $x^n$ on $[0,1]$ | $1$ | $\frac{1}{n+1} \to 0$ | $(2n+1)^{-1/2} \to 0$ | $0$ a.e. |

The first sequence converges to $0$ uniformly but not in $L^1$; the third converges to $0$ in $L^1$ and $L^2$ but not uniformly; the second converges in none of these senses, although it converges pointwise. The code computes these norms numerically.

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

def norms(vals, dx):
    return np.abs(vals).max(), np.abs(vals).sum() * dx, np.sqrt((vals ** 2).sum() * dx)

ns = np.arange(1, 11)
rows = {"(1/n) 1_[0,n]": [], "2^n 1_[2^-n, 2^(1-n)]": [], "x^n on [0,1]": []}
for n in ns:
    x = np.linspace(0, 12, 240001); dx = x[1] - x[0]
    rows["(1/n) 1_[0,n]"].append(norms(np.where(x <= n, 1 / n, 0.0), dx))
    x = np.linspace(0, 1, 400001); dx = x[1] - x[0]
    rows["2^n 1_[2^-n, 2^(1-n)]"].append(norms(np.where((x >= 2.0 ** -n) & (x <= 2.0 ** (1 - n)), 2.0 ** n, 0.0), dx))
    rows["x^n on [0,1]"].append(norms(x ** n, dx))

fig, axes = plt.subplots(1, 3, figsize=(13, 3.8))
for ax, (name, vals) in zip(axes, rows.items()):
    vals = np.array(vals)
    for j, (lab, col) in enumerate([(r"$\|f_n\|_\infty$", RED), (r"$\|f_n\|_1$", BLUE), (r"$\|f_n\|_2$", ORANGE)]):
        ax.plot(ns, vals[:, j], "o-", color=col, ms=4, label=lab)
    ax.set_yscale("log"); ax.set_xlabel("$n$"); ax.set_title(name); ax.legend(fontsize=9)
plt.tight_layout(); plt.show()
for name, vals in rows.items():
    v = np.array(vals)[[0, 4, 9]]
    print(f"{name:24s} n=1,5,10:  sup = {np.round(v[:, 0], 3)},  L1 = {np.round(v[:, 1], 3)},  L2 = {np.round(v[:, 2], 3)}")
```
