# The Dominated Convergence Theorem

**Prerequisites:** [Fatou's Lemma and the Monotone Convergence Theorem](fatous-lemma-and-convergence-theorems.ipynb) · [Integrable Functions and Linearity of the Integral](integrable-functions.ipynb) · [Limits of Measurable Functions](limits-of-measurable-functions.ipynb)

The [monotone convergence theorem](fatous-lemma-and-convergence-theorems.ipynb#thm-monotone-convergence) allows limits and integrals to be interchanged, but its hypothesis is strong: the functions must increase. In most situations where one wants to interchange them, the sequence is not monotone at all. **Lebesgue's dominated convergence theorem** replaces monotonicity by a much weaker and more flexible condition: all the functions, which may now take both signs, stay below one fixed integrable function. It is the most widely used convergence theorem of the theory. Throughout, $(X, \mathcal{M}, \mu)$ is a measure space.

::: {#thm-dominated-convergence}
## Lebesgue's Dominated Convergence Theorem

Let $f_n$ and $f$ be measurable functions on $X$ with $f_n \to f$ a.e. Suppose there is an integrable function $g$, that is $\int g\, d\mu < \infty$, such that for every $n$
$$|f_n| \le g \quad \text{a.e.}$$
Then $f$ is integrable and
$$\lim_{n \to \infty} \int f_n\, d\mu = \int f\, d\mu .$$
:::

If the measure space is complete, the measurability of $f$ is automatic ([a.e. limits](limits-of-measurable-functions.ipynb#thm-measurability-ae)). The function $g$ plays no role in the conclusion; it is only a **controller**. Since $-g \le f_n \le g$, everything happens inside the band between the graphs of $-g$ and $g$, and inside such a band of finite area, pointwise convergence (almost everywhere) forces convergence of the integrals. No monotonicity is required.

::: {.callout-warning}
## Measurable vs. Integrable
Measurability of the $f_n$ cannot be dropped: the integral is only defined for measurable functions. Integrability of the $f_n$, on the other hand, need not be assumed; it follows from $|f_n| \le g$ by the [comparison test](integrable-functions.ipynb#prp-integral-triangle-inequality). The two notions should not be confused: *measurable* concerns preimages of intervals, *integrable* means $\int |f| < \infty$.
:::

::: {#exm-no-integrable-dominator}
## The Escape-of-Mass Examples Cannot Be Dominated

In the [escape-of-mass examples](fatous-lemma-and-convergence-theorems.ipynb#exm-escape-of-mass) the integrals do not converge, so by the theorem no integrable $g$ can dominate them. Indeed, any $g$ with $g \ge f_n$ for all $n$ satisfies $g \ge \sup_n f_n$, and the supremum has infinite integral:

1. for $f_n = \mathbf{1}_{[n, n+1]}$, $\sup_n f_n = \mathbf{1}_{[1, \infty)}$, with integral $\infty$;
2. for $f_n = 2^n \mathbf{1}_{[2^{-n}, 2^{1-n}]}$, the blocks are disjoint except at endpoints, so $\sup_n f_n = \sum_n f_n$ almost everywhere, with integral $\sum_n 1 = \infty$;
3. for $f_n = \frac1n \mathbf{1}_{[0,n]}$, $\sup_n f_n(x) \ge \frac{1}{k}$ on $[k-1, k]$ (take $n = k$), so its integral is at least $\sum_k \frac1k = \infty$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-dominated-convergence
*Preparations.* Removing a null set changes no integral, so we may assume that $f_n \to f$ and $|f_n| \le g$ hold everywhere, and that $g < \infty$ everywhere ($g$ is integrable, hence [finite a.e.](integrable-functions.ipynb#prp-integrable-finite-ae)). Letting $n \to \infty$ in $|f_n| \le g$ gives $|f| \le g$, so $f$ is integrable by the comparison test, and so are all $f_n$.

*Making Fatou's lemma applicable.* [Fatou's lemma](fatous-lemma-and-convergence-theorems.ipynb#thm-fatou) speaks only about **nonnegative** functions, while the $f_n$ may be negative. The bound $-g \le f_n \le g$ provides two nonnegative sequences:
$$g - f_n \ge 0, \qquad g + f_n \ge 0 .$$
Since $f_n \to f$, they converge to $g - f$ and $g + f$ ($g$ is a fixed function and plays no role in the convergence). Fatou's lemma, applied to both, gives for every $\varepsilon > 0$ an $N$ such that for $n > N$
$$\int (g - f_n) > \int (g - f) - \varepsilon, \qquad \int (g + f_n) > \int (g + f) - \varepsilon .$$
All the functions involved are integrable, so by [linearity](integrable-functions.ipynb#thm-integral-linear)
$$\int g - \int f_n > \int g - \int f - \varepsilon, \qquad \int g + \int f_n > \int g + \int f - \varepsilon .$$
Since $\int g$ is **finite**, it can be cancelled from both inequalities. Multiplying the first by $-1$,
$$\int f - \varepsilon < \int f_n < \int f + \varepsilon \qquad (n > N),$$
that is, $\big|\int f_n - \int f\big| < \varepsilon$ for $n > N$. $\blacksquare$
:::

::: {.callout-note}
## Why the Proof Works
The proof is short but not very intuitive. The function $g$ is only an intermediary: adding it to $f_n$ and subtracting $f_n$ from it produces nonnegative functions, so that Fatou's lemma applies twice. Each application gives one half of the convergence, and the two halves together give the limit. The finiteness of $\int g$ is used at exactly one place: to cancel $\int g$ without meeting $\infty - \infty$.
:::

::: {#exm-dct-applications}
## Using the Theorem

1. On $[0, 1]$ with Lebesgue measure, $f_n(x) = x^n \to 0$ almost everywhere ([figure](limits-of-measurable-functions.ipynb)), and $|f_n| \le 1$, an integrable function on $[0,1]$. So $\int_{[0,1]} x^n\, dm \to 0$; indeed it equals $\frac{1}{n+1}$.
2. $f_n(x) = \dfrac{n x}{1 + n^2 x^2}$ on $[0, 1]$ tends to $0$ at every point, but not uniformly: its maximum, at $x = \frac1n$, is $\frac12$ for every $n$. Still $|f_n| \le \frac12$, integrable on $[0,1]$, so $\int_{[0,1]} f_n\, dm \to 0$ (the exact value is $\frac{\ln(1 + n^2)}{2n}$).
3. $f_n(x) = n^2 x e^{-n x}$ on $[0, 1]$ also tends to $0$ at every point, but $\int_{[0,1]} f_n\, dm = 1 - (n+1)e^{-n} \to 1$. Here there is no integrable dominating function: the smallest candidate $\sup_n f_n(x)$ behaves like $\frac{4}{e^2 x}$ near $0$ (maximize $t^2 x e^{-tx}$ over $t$: the maximum is at $t = 2/x$), which is not integrable. This is the spike phenomenon again, now with continuous functions.
:::

The figure shows examples 2 and 3: on the left, a family inside the band $[0, g]$ with $g = \frac12$ and integrals tending to $0$; on the right, a family whose envelope $\sup_n f_n$ is not integrable and whose integrals tend to $1$, not to the integral $0$ of the limit.

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

x = np.linspace(1e-6, 1, 20001)
ns = [1, 2, 4, 8, 16, 32]
cols = [BLUE, ORANGE, AQUA, YELLOW, VIOLET, MUTED]
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(12, 4.2))

ax.axhspan(0, 0.5, color=AQUA, alpha=0.1)
ax.axhline(0.5, color=AQUA, lw=2, label=r"dominating $g = 1/2$")
for n, col in zip(ns, cols):
    ax.plot(x, n * x / (1 + n ** 2 * x ** 2), color=col, lw=1.6, label=f"$n = {n}$")
ax.set_ylim(0, 0.6); ax.set_xlabel("$x$"); ax.legend(fontsize=8, ncol=2, loc="upper right")
ax.set_title(r"$f_n = nx/(1+n^2x^2)$: dominated, $\int f_n \to 0$")

N = np.arange(1, 400)
sup_env = np.max([n ** 2 * x * np.exp(-n * x) for n in N], axis=0)
ax2.plot(x, sup_env, color=RED, lw=2, label=r"envelope $\sup_n f_n$ (not integrable)")
for n, col in zip(ns, cols):
    ax2.plot(x, n ** 2 * x * np.exp(-n * x), color=col, lw=1.4, label=f"$n = {n}$")
ax2.set_xlim(0, 1); ax2.set_ylim(0, 14); ax2.set_xlabel("$x$"); ax2.legend(fontsize=8, ncol=2)
ax2.set_title(r"$f_n = n^2 x e^{-nx}$: not dominated, $\int f_n \to 1$")
plt.tight_layout(); plt.show()

for n in [1, 10, 100, 1000]:
    I1 = np.trapezoid(n * x / (1 + n ** 2 * x ** 2), x)
    I2 = 1 - (n + 1) * np.exp(-n)
    print(f"n = {n:4d}:  int nx/(1+n^2x^2) = {I1:.4f} (exact {np.log(1 + n**2) / (2*n):.4f}),   int n^2 x e^(-nx) = {I2:.4f}")
print("envelope * e^2 x / 4 at x = 0.01, 0.1, 0.3:", np.round((sup_env * np.e ** 2 * x / 4)[[200, 2000, 6000]], 3))
```
