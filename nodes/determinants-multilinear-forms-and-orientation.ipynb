# Multilinear Forms, Determinants, and Orientation

## 1\. Multilinear Forms and Alternating Maps

Let $K$ be a [field](field.ipynb) and let $V_1, V_2, \dots, V_p$ and $V'$ be [vector spaces](vector-spaces.ipynb) over $K$.

::: {#def-multilinear-map}
## Multilinear ($p$-Linear) Map

A function:

$$\alpha: V_1 \times V_2 \times \dots \times V_p \to V'$$

is called **$p$-linear** (or **multilinear**) if it is linear in each argument when the remaining $p - 1$ arguments are held fixed. That is, for each $j \in \{1, \dots, p\}$, fixed vectors $v_k \in V_k$ ($k \neq j$), and any $u, v \in V_j$, $c \in K$:

$$\alpha(v_1, \dots, c u + v, \dots, v_p) = c \alpha(v_1, \dots, u, \dots, v_p) + \alpha(v_1, \dots, v, \dots, v_p)$$
:::

### 1.1\. Examples of Multilinear Maps

1. **Scalar Multiplication:** The multiplication map $m: K \times K \to K$, $m(x, y) = xy$, is bilinear ($2$-linear). More generally, the product map $m: K^p \to K$, $m(x_1, \dots, x_p) = x_1 x_2 \cdots x_p$, is $p$-linear.
2. **Inner Product:** The standard Euclidean inner product on $\mathbb{R}^n$, $\langle x, y \rangle = \sum_{i=1}^n x_i y_i$, is bilinear.
3. **Cross Product:** The vector cross product on $\mathbb{R}^3$, $(u, v) \mapsto u \times v$, is bilinear.

---

::: {#def-alternating-form}
## Alternating (Antisymmetric) Map

Let $V$ and $V'$ be vector spaces over $K$. A $p$-linear map:

$$\alpha: \underbrace{V \times V \times \dots \times V}_{p \text{ copies}} \to V'$$

is called **alternating** (or **antisymmetric**, **skew-symmetric**) if it vanishes whenever any two adjacent or distinct arguments are identical:

$$\alpha(v_1, \dots, v_p) = 0 \qquad \text{whenever } v_i = v_j \text{ for some } i \neq j$$
:::

::: {#prp-alternating-sign-reversal}
## Sign Reversal under Transposition of Arguments

If $\alpha: V^p \to V'$ is an alternating $p$-linear map, swapping any two arguments reverses the sign of the output:

$$\alpha(v_1, \dots, u, \dots, w, \dots, v_p) = -\alpha(v_1, \dots, w, \dots, u, \dots, v_p)$$
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-alternating-sign-reversal
Holding all arguments fixed except the two positions of interest, consider evaluating $\alpha$ with the sum $u + w$ placed in both positions:

$$\alpha(\dots, u + w, \dots, u + w, \dots) = 0$$

By bilinearity in these two arguments, expand the expression:

$$
\begin{aligned}
0 &= \alpha(\dots, u + w, \dots, u + w, \dots) \\
&= \alpha(\dots, u, \dots, u + w, \dots) + \alpha(\dots, w, \dots, u + w, \dots) \\
&= \alpha(\dots, u, \dots, u, \dots) + \alpha(\dots, u, \dots, w, \dots) + \alpha(\dots, w, \dots, u, \dots) + \alpha(\dots, w, \dots, w, \dots)
\end{aligned}
$$

Since $\alpha$ is alternating, the terms with duplicate arguments vanish: $\alpha(\dots, u, \dots, u, \dots) = 0$ and $\alpha(\dots, w, \dots, w, \dots) = 0$. Therefore:

$$\alpha(\dots, u, \dots, w, \dots) + \alpha(\dots, w, \dots, u, \dots) = 0$$

Subtracting yields:

$$\alpha(\dots, u, \dots, w, \dots) = -\alpha(\dots, w, \dots, u, \dots)$$

$\blacksquare$
:::

::: {.callout-warning}
## Remark on Fields of Characteristic 2
When $\text{char}(K) \neq 2$ (meaning $1 + 1 \neq 0$), the sign-reversal property $\alpha(u, w) = -\alpha(w, u)$ implies the alternating property: setting $u = w$ gives $\alpha(u, u) = -\alpha(u, u)$, so $2\alpha(u, u) = 0$, hence $\alpha(u, u) = 0$.

However, over fields where $1 + 1 = 0$ (such as $\mathbb{F}_2$ where $1 = -1$), skew-symmetry does not force $\alpha(u, u) = 0$. Thus, defining alternating maps by the vanishing condition $\alpha(u, u) = 0$ is the universally correct definition over all fields.
:::


## 2\. Permutations and the Sign Homomorphism

::: {#def-permutations}
## Permutations and Transpositions

1. A **permutation** of $\{1, 2, \dots, n\}$ is a bijection $\sigma: \{1, \dots, n\} \to \{1, \dots, n\}$. The set of all $n!$ permutations forms the **symmetric group** $S_n$.
2. A **transposition** $\tau \in S_n$ is a permutation that swaps two distinct elements $\mu \neq \nu$ and leaves all other elements fixed.
:::

::: {#thm-transposition-decomposition}
## Generation by Transpositions and Invariance of Parity

1. Every permutation $\sigma \in S_n$ can be expressed as a finite composition of transpositions:
   $$\sigma = \tau_1 \circ \tau_2 \circ \dots \circ \tau_k$$
2. The parity (even or odd) of the number of transpositions $k$ is an invariant of $\sigma$, independent of the specific decomposition chosen.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-transposition-decomposition
**1. Existence of decomposition (by induction on $n$):**
For $n = 1$, the only permutation is the identity. For $n = 2$, the identity is $(1, 2) \circ (1, 2)$ and the swap is $(1, 2)$.

Assume every permutation of $n$ elements can be written as a product of transpositions. Let $\sigma \in S_{n+1}$. If $\sigma(n+1) = n+1$, $\sigma$ restricts to a permutation of $\{1, \dots, n\}$ and the result holds by induction. If $\sigma(n+1) = j \neq n+1$, let $\tau$ be the transposition swapping $j$ and $n+1$. Then:

$$(\tau \circ \sigma)(n+1) = \tau(j) = n+1$$

Thus $\rho = \tau \circ \sigma$ fixes $n+1$, so $\rho$ can be factored into transpositions: $\rho = \tau_1 \circ \dots \circ \tau_m$. Composing both sides with $\tau$ (using $\tau \circ \tau = \text{id}$):

$$\sigma = \tau \circ \tau_1 \circ \dots \circ \tau_m$$

**2. Parity Invariance via the Vandermonde Product:**
Define the polynomial quotient:

$$\operatorname{sgn}(\sigma) = \frac{\prod_{1 \le i < j \le n} (\sigma(j) - \sigma(i))}{\prod_{1 \le i < j \le n} (j - i)}$$

Since $\sigma$ is a bijection, the factors in the numerator are a permutation of the factors in the denominator up to signs. Therefore $\operatorname{sgn}(\sigma) = \pm 1$.

Let $\tau$ be a transposition swapping $\mu$ and $\nu$ with $\mu < \nu$. In the product $\prod_{i < j} (\sigma(j) - \sigma(i))$, applying $\tau$:

- The single pair $\{i, j\}$ where $\{\sigma(i), \sigma(j)\} = \{\mu, \nu\}$ flips sign: $(\nu - \mu) \mapsto (\mu - \nu) = -(\nu - \mu)$.
- Pairs with neither element in $\{\mu, \nu\}$ are unchanged.
- Pairs containing exactly one element from $\{\mu, \nu\}$ occur in tandem pairs: $(\sigma(k) - \mu)(\sigma(k) - \nu) \mapsto (\sigma(k) - \nu)(\sigma(k) - \mu)$, leaving the combined sign invariant.

Thus, applying any transposition reverses the sign:

$$\operatorname{sgn}(\tau \circ \sigma) = -\operatorname{sgn}(\sigma)$$

For the identity permutation, $\operatorname{sgn}(\text{id}) = +1$. Therefore, for every transposition, $\operatorname{sgn}(\tau) = -1$. If $\sigma = \tau_1 \circ \dots \circ \tau_k$, then:

$$\operatorname{sgn}(\sigma) = (-1)^k$$

Consequently, $k$ must be even whenever $\operatorname{sgn}(\sigma) = +1$, and odd whenever $\operatorname{sgn}(\sigma) = -1$. $\blacksquare$
:::

::: {#cor-alternating-group-partition}
## Equipartition of Permutations

For $n \ge 2$, the set of permutations partitions into exactly $\frac{n!}{2}$ even permutations ($S_n^+$) and $\frac{n!}{2}$ odd permutations ($S_n^-$). The map $\sigma \mapsto \tau \circ \sigma$ for a fixed transposition $\tau$ is a bijection between $S_n^+$ and $S_n^-$.
:::


## 3\. Characterization and Uniqueness of the Determinant

We regard an $n \times n$ [matrix](linear-maps-inverses-and-matrices.ipynb#def-matrix-linear-map) $A \in M_{n \times n}(K)$ as an $n$-tuple of its column vectors:

$$A = (\alpha_1, \alpha_2, \dots, \alpha_n) \in (K^n)^n$$

::: {#thm-alternating-form-uniqueness}
## Complete Characterization of Alternating $n$-Linear Forms

Let $D: (K^n)^n \to K$ be an alternating $n$-linear form. For any vectors $\alpha_1, \dots, \alpha_n \in K^n$ with $\alpha_j = \sum_{i=1}^n a_{ij} e_i$:

$$D(\alpha_1, \dots, \alpha_n) = \left( \sum_{\sigma \in S_n} \operatorname{sgn}(\sigma) a_{\sigma(1), 1} a_{\sigma(2), 2} \cdots a_{\sigma(n), n} \right) D(e_1, \dots, e_n)$$

where $(e_1, \dots, e_n)$ is the standard [basis](linear-independence-basis-dimension.ipynb#def-basis) of $K^n$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-alternating-form-uniqueness
By $n$-linearity, expand $D(\alpha_1, \dots, \alpha_n)$ multilinearly across all columns:

$$
\begin{aligned}
D(\alpha_1, \dots, \alpha_n) &= D\left( \sum_{i_1=1}^n a_{i_1 1} e_{i_1}, \dots, \sum_{i_n=1}^n a_{i_n n} e_{i_n} \right) \\
&= \sum_{i_1=1}^n \sum_{i_2=1}^n \cdots \sum_{i_n=1}^n a_{i_1 1} a_{i_2 2} \cdots a_{i_n n} D(e_{i_1}, e_{i_2}, \dots, e_{i_n})
\end{aligned}
$$

This multilinear expansion contains $n^n$ terms. Because $D$ is alternating:

$$D(e_{i_1}, \dots, e_{i_n}) = 0 \qquad \text{whenever } i_j = i_k \text{ for some } j \neq k$$

Therefore, only terms where $(i_1, i_2, \dots, i_n)$ is a permutation $\sigma = (\sigma(1), \dots, \sigma(n))$ of $(1, \dots, n)$ survive. There are exactly $n!$ such terms:

$$D(\alpha_1, \dots, \alpha_n) = \sum_{\sigma \in S_n} a_{\sigma(1), 1} a_{\sigma(2), 2} \cdots a_{\sigma(n), n} D(e_{\sigma(1)}, \dots, e_{\sigma(n)})$$

By @prp-alternating-sign-reversal, reordering $(e_{\sigma(1)}, \dots, e_{\sigma(n)})$ into the natural order $(e_1, \dots, e_n)$ introduces a factor of $(-1)$ for each transposition, yielding:

$$D(e_{\sigma(1)}, \dots, e_{\sigma(n)}) = \operatorname{sgn}(\sigma) D(e_1, \dots, e_n)$$

Factoring out $D(e_1, \dots, e_n)$ completes the proof: $\blacksquare$
:::

---

::: {#def-determinant}
## The Determinant

The **determinant** is the unique alternating $n$-linear map $\det: M_{n \times n}(K) \to K$ normalized such that the identity matrix has determinant $1$:

$$\det(I) = \det(e_1, e_2, \dots, e_n) = 1$$

For any matrix $A = (a_{ij}) \in M_{n \times n}(K)$, its determinant is given by the **Leibniz formula**:

$$\det(A) = \sum_{\sigma \in S_n} \operatorname{sgn}(\sigma) a_{\sigma(1), 1} a_{\sigma(2), 2} \cdots a_{\sigma(n), n}$$
:::

::: {.callout-note}
## Dimension of Alternating Multilinear Forms
@thm-alternating-form-uniqueness demonstrates that the space of alternating $n$-linear forms on an $n$-dimensional vector space is **$1$-dimensional**. Specifying the value on a single basis tuple $(e_1, \dots, e_n)$ uniquely determines the form everywhere.
:::


## 4\. Multiplicativity and Invertibility of the Determinant

::: {#thm-det-multiplicative}
## Multiplicativity of the Determinant

For any two matrices $A, B \in M_{n \times n}(K)$:

$$\det(AB) = \det(A) \det(B)$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-det-multiplicative via Uniqueness
Fix matrix $A$. Define two functions $D_1, D_2: (K^n)^n \to K$ acting on the columns $b_1, \dots, b_n$ of matrix $B$:

$$D_1(B) = \det(AB), \qquad D_2(B) = \det(A) \det(B)$$

1. **$D_1$ is $n$-linear and alternating in the columns of $B$:**
   The $j$-th column of the product $AB$ is $A b_j$. Since matrix-vector multiplication $b_j \mapsto A b_j$ is linear, each column of $AB$ depends linearly on the corresponding column of $B$. Thus $D_1$ is $n$-linear in the columns of $B$.
   If two columns of $B$ are equal ($b_j = b_k$ for $j \neq k$), then $A b_j = A b_k$, so two columns of $AB$ are equal. Since the determinant is alternating, $\det(AB) = 0$. Hence $D_1$ is alternating.

2. **$D_2$ is $n$-linear and alternating:**
   $D_2(B)$ is simply the scalar $\det(A)$ times the alternating $n$-linear form $\det(B)$. Thus $D_2$ is also an alternating $n$-linear form.

3. **Evaluation at the identity matrix $B = I$:**
   $$
   \begin{aligned}
   D_1(I) &= \det(A I) = \det(A) \\
   D_2(I) &= \det(A) \det(I) = \det(A) \cdot 1 = \det(A)
   \end{aligned}
   $$

By @thm-alternating-form-uniqueness, an alternating $n$-linear form is uniquely determined by its value on $I$. Because $D_1(I) = D_2(I)$, we have $D_1(B) = D_2(B)$ for all $B \in M_{n \times n}(K)$:

$$\det(AB) = \det(A) \det(B)$$

$\blacksquare$
:::

---

::: {#thm-det-invertibility}
## Determinantal Criterion for Invertibility

A matrix $A \in M_{n \times n}(K)$ is [invertible](linear-maps-inverses-and-matrices.ipynb#cor-equal-dim-invertibility) if and only if:

$$\det(A) \neq 0$$

When $A$ is invertible:

$$\det(A^{-1}) = \frac{1}{\det(A)}$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-det-invertibility
**1. $(\Rightarrow)$ Suppose $A$ is invertible:**
Then $A A^{-1} = I$. Taking determinants and applying @thm-det-multiplicative:

$$\det(A) \det(A^{-1}) = \det(I) = 1$$

Since the product is $1$, $\det(A) \neq 0$ and $\det(A^{-1}) = (\det(A))^{-1}$.

**2. $(\Leftarrow)$ Suppose $A$ is not invertible:**
Then the column vectors $\alpha_1, \dots, \alpha_n$ of $A$ are linearly dependent. Thus, one column is a linear combination of the remaining columns. Without loss of generality, let $\alpha_1 = \sum_{j=2}^n c_j \alpha_j$. By multilinearity:

$$\det(A) = \det\left( \sum_{j=2}^n c_j \alpha_j, \alpha_2, \dots, \alpha_n \right) = \sum_{j=2}^n c_j \det(\alpha_j, \alpha_2, \dots, \alpha_n)$$

In each term $\det(\alpha_j, \alpha_2, \dots, \alpha_n)$, the vector $\alpha_j$ appears both in the first column and in the $j$-th column. Because the determinant is alternating, every term is zero. Thus $\det(A) = 0$. $\blacksquare$
:::


## 5\. Geometric Interpretation: Oriented Volume

In real Euclidean space $\mathbb{R}^n$, the geometric counterpart of the determinant is the volume of a parallelotope.

::: {#def-parallelotope}
## Parallelotope

Let $v_1, v_2, \dots, v_n \in \mathbb{R}^n$. The **parallelotope** spanned by these vectors is:

$$P(v_1, \dots, v_n) = \left\{ \sum_{i=1}^n t_i v_i \;\middle|\; 0 \le t_i \le 1 \right\}$$

Its $n$-dimensional **volume** is given by the absolute value of the determinant:

$$\operatorname{Vol}(P(v_1, \dots, v_n)) = |\det(v_1, \dots, v_n)|$$
:::

::: {.callout-note}
## Why Determinant Measures Oriented Volume
- **Scaling:** Scaling one vector by $c$ stretches the parallelotope along that direction by $|c|$, scaling volume by $|c|$.
- **Shearing (Cavalieri's Principle):** Adding a multiple of $v_j$ to $v_i$ ($v_i \mapsto v_i + c v_j$) shears the parallelotope without changing its base or height, preserving volume.
- **Orientation Sign:** The sign of $\det(v_1, \dots, v_n)$ tracks the **orientation** (chirality) of the basis $(v_1, \dots, v_n)$ relative to the standard basis $(e_1, \dots, e_n)$.
:::

```python

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

# Set plot style
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(13, 5))

# 1. Positively oriented basis (Right-handed, det > 0)
v1 = np.array([3.0, 1.0])
v2 = np.array([1.0, 2.5])
det_pos = np.linalg.det(np.column_stack([v1, v2]))

parallelogram1 = np.array([[0, 0], v1, v1 + v2, v2])
poly1 = Polygon(parallelogram1, closed=True, alpha=0.35, facecolor='#2980b9', edgecolor='#1f618d', lw=2)
ax1.add_patch(poly1)
ax1.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='#27ae60', width=0.015, label=r'$v_1$')
ax1.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='#e67e22', width=0.015, label=r'$v_2$')
ax1.set_xlim(-1, 5)
ax1.set_ylim(-1, 4.5)
ax1.set_title(r'Positive Orientation: $\det([v_1, v_2]) = +%.2f$' % det_pos, fontsize=11, fontweight='bold')
ax1.legend(loc='upper left', framealpha=0.9)
ax1.set_aspect('equal')

# 2. Negatively oriented basis (Swapped / Left-handed, det < 0)
det_neg = np.linalg.det(np.column_stack([v2, v1]))
parallelogram2 = np.array([[0, 0], v2, v1 + v2, v1])
poly2 = Polygon(parallelogram2, closed=True, alpha=0.35, facecolor='#c0392b', edgecolor='#922b21', lw=2)
ax2.add_patch(poly2)
ax2.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='#e67e22', width=0.015, label=r'$v_2$')
ax2.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='#27ae60', width=0.015, label=r'$v_1$')
ax2.set_xlim(-1, 5)
ax2.set_ylim(-1, 4.5)
ax2.set_title(r'Negative Orientation (Swapped): $\det([v_2, v_1]) = %.2f$' % det_neg, fontsize=11, fontweight='bold')
ax2.legend(loc='upper left', framealpha=0.9)
ax2.set_aspect('equal')

plt.tight_layout()
plt.show()
```


## 6\. Orientation of Real Vector Spaces

Let $V$ be an $n$-dimensional vector space over the real numbers $\mathbb{R}$.

Let $B = (v_1, \dots, v_n)$ and $B' = (v'_1, \dots, v'_n)$ be two ordered bases of $V$. There exists a unique [invertible](linear-maps-inverses-and-matrices.ipynb#cor-equal-dim-invertibility) transition matrix $M \in M_{n \times n}(\mathbb{R})$ such that:

$$v'_j = \sum_{i=1}^n M_{ij} v_i \qquad (j = 1, \dots, n)$$

Since $M$ maps a basis to a basis, $M$ is invertible, so by @thm-det-invertibility, $\det(M) \neq 0$.

::: {#def-orientation-equivalence}
## Equioriented Bases

Two ordered bases $B$ and $B'$ of an $n$-dimensional real vector space $V$ are called **equioriented** (or **consistently oriented**, written $B \sim B'$) if their transition matrix has positive determinant:

$$B \sim B' \iff \det(M) > 0$$
:::

::: {#prp-orientation-equivalence-relation}
## Equivalence Relation and the Two Orientations

The relation $\sim$ is an equivalence relation on the set of ordered bases of $V$, partitioning all ordered bases into exactly **two equivalence classes**.

Each equivalence class is called an **orientation** of $V$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-orientation-equivalence-relation
We verify the three equivalence axioms:

1. **Reflexivity:** The transition matrix from $B$ to $B$ is the identity matrix $I$. Since $\det(I) = 1 > 0$, $B \sim B$.
2. **Symmetry:** If $B \sim B'$ with transition matrix $M$, then the transition matrix from $B'$ to $B$ is $M^{-1}$. By @thm-det-invertibility:
   $$\det(M^{-1}) = \frac{1}{\det(M)} > 0$$
   Thus $B' \sim B$.
3. **Transitivity:** If $B \sim B'$ with transition matrix $M_1$ and $B' \sim B''$ with transition matrix $M_2$, the transition matrix from $B$ to $B''$ is $M_1 M_2$. By @thm-det-multiplicative:
   $$\det(M_1 M_2) = \det(M_1) \det(M_2) > 0$$
   Thus $B \sim B''$.

Because the determinant of any real invertible transition matrix is either strictly positive or strictly negative, there are precisely two equivalence classes. $\blacksquare$
:::

::: {.callout-note}
## Topological Perspective on Orientation
In the general linear group over $\mathbb{R}$, $GL(n, \mathbb{R})$ has precisely two connected components:

$$GL^+(n, \mathbb{R}) = \{M \in M_{n \times n}(\mathbb{R}) \mid \det(M) > 0\}$$
$$GL^-(n, \mathbb{R}) = \{M \in M_{n \times n}(\mathbb{R}) \mid \det(M) < 0\}$$

Two bases have the same orientation if and only if one can be continuously deformed into the other through a continuous path of bases without ever becoming linearly dependent (crossing $\det = 0$).

In contrast, the complex general linear group $GL(n, \mathbb{C})$ is **connected** (one can loop around $0$ in $\mathbb{C}^*$), so there is no notion of orientation over $\mathbb{C}$.
:::

```python

import numpy as np

# 1. Implementation of Leibniz formula via permutations
from itertools import permutations

def perm_sgn(p):
    inv_count = sum(1 for i in range(len(p)) for j in range(i + 1, len(p)) if p[i] > p[j])
    return -1 if inv_count % 2 else 1

def leibniz_det(A):
    n = A.shape[0]
    det_val = 0.0
    for p in permutations(range(n)):
        term = perm_sgn(p)
        for j in range(n):
            term *= A[p[j], j]
        det_val += term
    return det_val

# Test on a 4x4 matrix
np.random.seed(42)
A = np.random.randint(-3, 4, size=(4, 4)).astype(float)
B = np.random.randint(-3, 4, size=(4, 4)).astype(float)

det_leibniz = leibniz_det(A)
det_numpy = np.linalg.det(A)

print("=== Matrix A (4x4) ===\n", A)
print("Leibniz Formula Determinant:", round(det_leibniz, 4))
print("NumPy Determinant:", round(det_numpy, 4))
print("Matches:", np.isclose(det_leibniz, det_numpy))

# 2. Multiplicativity check: det(AB) = det(A) * det(B)
det_AB = np.linalg.det(A @ B)
det_A_det_B = np.linalg.det(A) * np.linalg.det(B)
print("\n=== Multiplicativity: det(AB) == det(A)*det(B) ===")
print("det(AB):", round(det_AB, 4))
print("det(A) * det(B):", round(det_A_det_B, 4))
print("Matches:", np.isclose(det_AB, det_A_det_B))

# 3. Orientation equivalence check
# Base B1 = standard basis, B2 = rotated, B3 = reflected
theta = np.pi / 3
R = np.array([
    [np.cos(theta), -np.sin(theta)],
    [np.sin(theta),  np.cos(theta)]
])
Refl = np.array([
    [1.0,  0.0],
    [0.0, -1.0]
])

print("\n=== Basis Orientation Classes ===")
print("det(Rotation Matrix) [Equioriented]:", round(np.linalg.det(R), 4), "(det > 0)")
print("det(Reflection Matrix) [Opposite Orientation]:", round(np.linalg.det(Refl), 4), "(det < 0)")
```
