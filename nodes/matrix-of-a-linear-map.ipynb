# The Matrix of a Linear Map

**Prerequisites:** [Linear Maps](linear-maps.ipynb) · [Coordinates](coordinates-and-isomorphism-with-Kn.ipynb)

A linear map is [determined by the images of a basis](linear-maps.ipynb#thm-linear-extension). Writing those images in coordinates produces a rectangular array of numbers, the **matrix** of the map. With matrices, applying a map becomes matrix-vector multiplication, and composing two maps becomes matrix multiplication. This explains the row-times-column rule.

Throughout, $V$, $W$ and $U$ are finite-dimensional [vector spaces](vector-spaces.ipynb) over a field $K$, with ordered bases $\mathcal{B} = (v_1, \dots, v_n)$ of $V$ and $\mathcal{C} = (w_1, \dots, w_m)$ of $W$.

## 1\. Definition

::: {#def-matrix-linear-map}
## Matrix of a Linear Map

Let $f : V \to W$ be linear. Write each $f(v_j)$ in the basis $\mathcal{C}$:
$$f(v_j) = a_{1j} w_1 + a_{2j} w_2 + \dots + a_{mj} w_m \qquad (j = 1, \dots, n).$$
The **matrix of $f$ with respect to $\mathcal{B}$ and $\mathcal{C}$** is the $m \times n$ matrix
$$[f]_{\mathcal{B}}^{\mathcal{C}} = (a_{ij}) = \begin{pmatrix} a_{11} & \cdots & a_{1n} \\ \vdots & & \vdots \\ a_{m1} & \cdots & a_{mn} \end{pmatrix} \in M_{m \times n}(K).$$
Its $j$-th column is the coordinate vector $[f(v_j)]_{\mathcal{C}}$. For an operator $f : V \to V$ we use the same basis on both sides and write $[f]_{\mathcal{B}} = [f]_{\mathcal{B}}^{\mathcal{B}}$.
:::

::: {.callout-note}
## How to Read the Notation
The lower index is the basis of the **domain** (where the input is measured), the upper index the basis of the **codomain** (where the output is measured). Read $[f]_{\mathcal{B}}^{\mathcal{C}}$ as "$f$, from $\mathcal{B}$-coordinates to $\mathcal{C}$-coordinates". For maps $K^n \to K^m$ with the standard bases the matrix is simply called the **standard matrix** of $f$; its columns are $f(e_1), \dots, f(e_n)$.
:::

::: {#exm-matrices-of-maps}
## Reading Off Matrices

1. **Rotation.** The rotation $R_\alpha$ of $\mathbb{R}^2$ by $\alpha$ sends $e_1 \mapsto (\cos\alpha, \sin\alpha)$ and $e_2 \mapsto (-\sin\alpha, \cos\alpha)$, so
   $$[R_\alpha] = \begin{pmatrix} \cos\alpha & -\sin\alpha \\ \sin\alpha & \cos\alpha \end{pmatrix}.$$
2. **Differentiation.** $D : P_3 \to P_2$ with bases $(1, x, x^2, x^3)$ and $(1, x, x^2)$: $D(1) = 0$, $D(x) = 1$, $D(x^2) = 2x$, $D(x^3) = 3x^2$, so
   $$[D] = \begin{pmatrix} 0 & 1 & 0 & 0 \\ 0 & 0 & 2 & 0 \\ 0 & 0 & 0 & 3 \end{pmatrix}.$$
3. **Shift of polynomials.** $S : P_2 \to P_2$, $S(p)(x) = p(x + 1)$, in the basis $(1, x, x^2)$: $S(1) = 1$, $S(x) = 1 + x$, $S(x^2) = 1 + 2x + x^2$, so the columns are $(1, 0, 0)$, $(1, 1, 0)$ and $(1, 2, 1)$.
:::

## 2\. Applying a Map Is Multiplying by Its Matrix

::: {#thm-matrix-action}
## Coordinates of the Image

For every $v \in V$,
$$[f(v)]_{\mathcal{C}} = [f]_{\mathcal{B}}^{\mathcal{C}}\ [v]_{\mathcal{B}}.$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-matrix-action
Let $v = \sum_j x_j v_j$. By linearity,
$$f(v) = \sum_{j=1}^n x_j f(v_j) = \sum_{j=1}^n x_j \sum_{i=1}^m a_{ij} w_i = \sum_{i=1}^m \Big( \sum_{j=1}^n a_{ij} x_j \Big) w_i.$$
So the $i$-th coordinate of $f(v)$ is $\sum_j a_{ij} x_j$, which is the $i$-th entry of the product of the matrix $(a_{ij})$ with the column $(x_1, \dots, x_n)^{\mathsf T}$. $\blacksquare$
:::

::: {.callout-note}
## Linear Maps Are Systems of Homogeneous Linear Expressions
In coordinates, $y = f(v)$ reads
$$y_1 = a_{11} x_1 + \dots + a_{1n} x_n, \quad \dots, \quad y_m = a_{m1} x_1 + \dots + a_{mn} x_n:$$
each output coordinate is a homogeneous expression of degree $1$ in the input coordinates. The abstract definition is better for proofs; this form is what one computes with.
:::

The animation shows the theorem for $A = \begin{pmatrix} 2 & -1 \\ 1 & 1 \end{pmatrix}$. The columns $(2, 1)$ and $(-1, 1)$ are where $e_1$ and $e_2$ land. The vector $v = e_1 + e_2$ is the diagonal of the unit square, and the map carries the whole construction along: $f(v) = f(e_1) + f(e_2) = (2, 1) + (-1, 1) = (1, 2)$, which is $A$ times the column $(1, 1)$.


```python

from la_manim import *

class ColumnsAreImages(GridScene):
    def construct(self):
        self.add(caption(r"A = \begin{pmatrix} 2 & -1 \\ 1 & 1 \end{pmatrix}"))
        self.vec([1, 0], C_BLUE, r"e_1", DOWN)
        self.vec([0, 1], C_ORANGE, r"e_2", LEFT)
        e2_shift = DashedLine([1, 0, 0], [1, 1, 0], color=C_ORANGE, stroke_width=3)
        e1_shift = DashedLine([0, 1, 0], [1, 1, 0], color=C_BLUE, stroke_width=3)
        self.track(e2_shift, e1_shift)
        self.play(Create(e2_shift), Create(e1_shift))
        self.vec([1, 1], C_YELLOW, r"v = e_1 + e_2", RIGHT)
        self.wait(1)
        self.transform([[2, -1], [1, 1]], run_time=3)
        cols = VGroup(MathTex(r"f(e_1) = (2,1)", color=C_BLUE), MathTex(r"f(e_2) = (-1,1)", color=C_ORANGE),
                      MathTex(r"f(v) = f(e_1) + f(e_2) = (1,2)", color=C_YELLOW)).arrange(DOWN, aligned_edge=LEFT)
        cols.scale(0.62).to_corner(DR, buff=0.3)
        self.play(FadeIn(BackgroundRectangle(cols, fill_opacity=0.85, buff=0.12, color=config.background_color)), FadeIn(cols))
        self.wait(2.5)

show(ColumnsAreImages)
```


## 3\. Composition Is Matrix Multiplication

::: {#thm-matrix-multiplication-composition}
## Matrix of a Composite

Let $\mathcal{D} = (u_1, \dots, u_p)$ be an ordered basis of $U$, and let $f : V \to W$ and $g : W \to U$ be linear. Then
$$[g \circ f]_{\mathcal{B}}^{\mathcal{D}} = [g]_{\mathcal{C}}^{\mathcal{D}}\ [f]_{\mathcal{B}}^{\mathcal{C}}.$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-matrix-multiplication-composition
Let $X = [f]_{\mathcal{B}}^{\mathcal{C}} = (x_{ij})$ and $Y = [g]_{\mathcal{C}}^{\mathcal{D}} = (y_{ki})$, so $f(v_j) = \sum_i x_{ij} w_i$ and $g(w_i) = \sum_k y_{ki} u_k$. Then
$$(g \circ f)(v_j) = g\Big(\sum_{i} x_{ij} w_i\Big) = \sum_{i} x_{ij} \sum_{k} y_{ki} u_k = \sum_{k} \Big(\sum_{i} y_{ki} x_{ij}\Big) u_k.$$
The coefficient $\sum_i y_{ki} x_{ij}$ is the $(k, j)$ entry of $Y X$. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Why Rows Times Columns
Matrix multiplication is not an arbitrary rule. It is the only rule that makes "first apply $f$, then $g$" correspond to "multiply the matrices". Read the indices in $[g]_{\mathcal{C}}^{\mathcal{D}} [f]_{\mathcal{B}}^{\mathcal{C}}$: the inner basis $\mathcal{C}$ appears once up and once down and cancels, leaving $[\,\cdot\,]_{\mathcal{B}}^{\mathcal{D}}$.
:::

::: {#exm-rotation-addition}
## The Angle-Addition Formulas

Rotating by $\beta$ and then by $\alpha$ is rotating by $\alpha + \beta$: $R_\alpha \circ R_\beta = R_{\alpha + \beta}$. By the theorem, $[R_\alpha][R_\beta] = [R_{\alpha+\beta}]$:
$$\begin{pmatrix} \cos\alpha & -\sin\alpha \\ \sin\alpha & \cos\alpha \end{pmatrix}\begin{pmatrix} \cos\beta & -\sin\beta \\ \sin\beta & \cos\beta \end{pmatrix} = \begin{pmatrix} \cos\alpha\cos\beta - \sin\alpha\sin\beta & \cdots \\ \sin\alpha\cos\beta + \cos\alpha\sin\beta & \cdots \end{pmatrix}.$$
Comparing the first columns gives $\cos(\alpha + \beta) = \cos\alpha\cos\beta - \sin\alpha\sin\beta$ and $\sin(\alpha + \beta) = \sin\alpha\cos\beta + \cos\alpha\sin\beta$.
:::

::: {#cor-matrix-associativity}
## Associativity of Matrix Multiplication

For $A \in M_{m \times n}(K)$, $B \in M_{n \times p}(K)$ and $C \in M_{p \times q}(K)$: $A(BC) = (AB)C$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-matrix-associativity
Let $f_A : K^n \to K^m$, $f_B$, $f_C$ be the maps $x \mapsto Ax$ etc., whose standard matrices are $A$, $B$, $C$. Composition of functions is associative, $f_A \circ (f_B \circ f_C) = (f_A \circ f_B) \circ f_C$. Taking matrices with @thm-matrix-multiplication-composition gives $A(BC) = (AB)C$. No index manipulation is needed. $\blacksquare$
:::

::: {#prp-identity-matrix}
## Matrix of the Identity

$[\mathrm{id}_V]_{\mathcal{B}}^{\mathcal{B}} = I_n$, the $n \times n$ identity matrix.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-identity-matrix
$\mathrm{id}_V(v_j) = v_j$ has coordinates $e_j$ in $\mathcal{B}$, so the $j$-th column is $e_j$. $\blacksquare$
:::

::: {.callout-warning}
## Two Different Bases
If $\mathcal{B}' \ne \mathcal{B}$ is another basis of $V$, then $[\mathrm{id}_V]_{\mathcal{B}'}^{\mathcal{B}}$ is in general **not** the identity matrix: its columns are the $\mathcal{B}$-coordinates of the vectors of $\mathcal{B}'$. This is the [change-of-basis matrix](change-of-basis-and-similarity.ipynb#def-transition-matrix).
:::

## 4\. Linear Maps and Matrices Are the Same Thing

::: {#prp-maps-matrices-iso}
## $\mathcal{L}(V, W) \cong M_{m \times n}(K)$

The map $\mathcal{L}(V, W) \to M_{m \times n}(K)$, $f \mapsto [f]_{\mathcal{B}}^{\mathcal{C}}$, is a [linear isomorphism](coordinates-and-isomorphism-with-Kn.ipynb#def-linear-isomorphism). In particular $\dim \mathcal{L}(V, W) = mn$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-maps-matrices-iso
**Linear.** The $j$-th column of $[a f + g]$ is $[(a f + g)(v_j)]_{\mathcal{C}} = a [f(v_j)]_{\mathcal{C}} + [g(v_j)]_{\mathcal{C}}$, using that the [coordinate map](coordinates-and-isomorphism-with-Kn.ipynb#prp-fin-dim-isomorphic-Kn) is linear.

**Bijective.** A matrix $(a_{ij})$ prescribes the vectors $\sum_i a_{ij} w_i$ as images of the $v_j$, and by [linear extension](linear-maps.ipynb#thm-linear-extension) there is exactly one linear map with these images. $\blacksquare$
:::

The code builds the matrices of differentiation and of the shift $p(x) \mapsto p(x + 1)$ on $P_2$, and checks on $p = 3 + 2x + 4x^2$ that the matrix of the composite acts like the composite: $S(D(p)) = S(2 + 8x) = 10 + 8x$. Here $D$ and $S$ happen to commute ($\frac{d}{dx} p(x + 1) = p'(x + 1)$ by the chain rule), so $[S][D] = [D][S]$; for most pairs of maps this fails.


```python

import numpy as np

D = np.array([[0, 1, 0],          # D(1) = 0, D(x) = 1, D(x^2) = 2x   (basis 1, x, x^2)
              [0, 0, 2],
              [0, 0, 0]])
S = np.array([[1, 1, 1],          # S(1) = 1, S(x) = 1 + x, S(x^2) = 1 + 2x + x^2
              [0, 1, 2],
              [0, 0, 1]])
p = np.array([3, 2, 4])           # 3 + 2x + 4x^2

print("[S o D] = [S][D] =\n", S @ D)
print("[S o D] p =", S @ D @ p, " -> 10 + 8x, as computed by hand")
print("[D o S] = [D][S] =\n", D @ S, "\n(D o S)(p) = D(p(x+1)) =", D @ S @ p)
print("S D == D S ?", np.array_equal(S @ D, D @ S))
```
