# Iterated Integrals and Fubini's Theorem

**Prerequisites:** [Multiple Integrals](multiple-integrals.ipynb) · [The Fundamental Theorem of Calculus](fundamental-theorem-of-calculus.ipynb) · [Integration by Parts and Substitution](integration-by-parts-and-substitution.ipynb)

Multiple integrals are computed by integrating one variable at a time. Geometrically, for $f \ge 0$ on a rectangle, $\int_c^df(x, y)\,dy$ is the area of the slice of the solid under the graph at a fixed $x$, and integrating these areas over $x$ stacks the slices into the volume.

## 1\. Fubini's Theorem

::: {#thm-fubini}
## Fubini's Theorem

Let $A \subseteq \mathbb{R}^k$, $B \subseteq \mathbb{R}^l$ be boxes and $f$ integrable on $A\times B$. If for every $x \in A$ the function $y \mapsto f(x, y)$ is integrable on $B$, then $F(x) = \int_Bf(x, y)\,dy$ is integrable on $A$ and
$$\int_{A\times B}f = \int_A\Big(\int_Bf(x, y)\,dy\Big)dx .$$
The same holds with the roles of $x$ and $y$ exchanged. In particular, for continuous $f$ on $[a, b]\times[c, d]$,
$$\iint_{[a,b]\times[c,d]}f\,dA = \int_a^b\int_c^df(x, y)\,dy\,dx = \int_c^d\int_a^bf(x, y)\,dx\,dy .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-fubini
A partition of $A\times B$ is a pair $(\mathcal{P}_A, \mathcal{P}_B)$, with subboxes $I\times J$. For $x \in I$,
$$F(x) = \sum_J\int_Jf(x, y)\,dy \ge \sum_Jm_{I\times J}(f)\operatorname{vol}(J),$$
because $f(x, y) \ge m_{I\times J}(f)$ for $y \in J$ (additivity and monotonicity of the integral over $B$). Taking the infimum over $x \in I$, multiplying by $\operatorname{vol}(I)$ and summing over $I$ gives $L(f, \mathcal{P}) \le L(F, \mathcal{P}_A)$; similarly $U(F, \mathcal{P}_A) \le U(f, \mathcal{P})$. Hence
$$L(f, \mathcal{P}) \le L(F, \mathcal{P}_A) \le U(F, \mathcal{P}_A) \le U(f, \mathcal{P}).$$
Since $f$ is integrable, the outer terms are within $\varepsilon$ of each other and of $\int f$ for a suitable $\mathcal{P}$; so $F$ is integrable with $\int_AF = \int_{A\times B}f$. For continuous $f$ every slice is continuous, hence integrable. $\blacksquare$
:::

Iterating, an integral over a box in $\mathbb{R}^n$ is an $n$-fold iterated integral, in any of the $n!$ orders. The notation $\int\cdots\int f(x_1, \dots, x_n)\,dx_1\cdots dx_n$ means: integrate first in $x_1$, then $x_2$, and so on.

::: {#exm-fubini-rectangle}
## Both Orders

On $R = [1, 2]\times[0, 1]$:
$$\iint_R(x^2y + e^{-2y})\,dA = \int_1^2\Big[\tfrac12x^2y^2 - \tfrac12e^{-2y}\Big]_{y=0}^{1}dx = \int_1^2\Big(\tfrac12x^2 + \tfrac{1 - e^{-2}}{2}\Big)dx = \tfrac53 - \tfrac12e^{-2},$$
and in the other order $\int_0^1\big[\frac13x^3y - e^{-2y}x\big]_{x=1}^2dy = \int_0^1\big(\frac73y + e^{-2y}\big)dy = \frac53 - \frac12e^{-2}$. Similarly, on $[1, 2]\times[0, 1]\times[0, 2]$, $\iiint(x + e^{-y+z})\,dV = 3 + (1 - e^{-1})(e^2 - 1) = 2 - e + e^2 + e^{-1}$.
:::

## 2\. Non-Rectangular Regions

::: {#cor-simple-regions}
## Integrals over Regions Between Graphs

Let $\alpha \le \beta$ be continuous on $[a, b]$ and $D = \{(x, y) : a \le x \le b,\ \alpha(x) \le y \le \beta(x)\}$ (a **$y$-simple** region). For $f$ continuous on $D$,
$$\iint_Df\,dA = \int_a^b\Big(\int_{\alpha(x)}^{\beta(x)}f(x, y)\,dy\Big)dx .$$
Likewise for an $x$-simple region $\{c \le y \le d,\ \gamma(y) \le x \le \delta(y)\}$, $\iint_Df = \int_c^d\int_{\gamma(y)}^{\delta(y)}f\,dx\,dy$. In $\mathbb{R}^n$, for $D = \{\alpha_1 \le x_1 \le \beta_1,\ \alpha_2(x_1) \le x_2 \le \beta_2(x_1),\ \dots,\ \alpha_n(x_1, \dots, x_{n-1}) \le x_n \le \beta_n(x_1, \dots, x_{n-1})\}$ with continuous bounds,
$$\int_Df = \int_{\alpha_1}^{\beta_1}\int_{\alpha_2(x_1)}^{\beta_2(x_1)}\cdots\int_{\alpha_n(x_1,\dots,x_{n-1})}^{\beta_n(x_1,\dots,x_{n-1})}f\,dx_n\cdots dx_2\,dx_1 .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-simple-regions
$D$ is admissible (its boundary consists of two graphs and two segments) and $\tilde f$ is [integrable](multiple-integrals.ipynb#thm-continuous-on-region) on a rectangle $[a, b]\times[c, d] \supseteq D$. For fixed $x$, $y \mapsto \tilde f(x, y)$ equals $f(x, y)$ on $[\alpha(x), \beta(x)]$ and $0$ elsewhere, so it is integrable with integral $\int_{\alpha(x)}^{\beta(x)}f(x, y)\,dy$ (it is continuous except at two points). Apply @thm-fubini. The $n$-dimensional statement follows by induction. $\blacksquare$
:::

A special case is **Cavalieri's principle**: if $D \subseteq \mathbb{R}^n$ is admissible and its slices $D_t = \{x' \in \mathbb{R}^{n-1} : (t, x') \in D\}$ are admissible, then $\operatorname{vol}(D) = \int\operatorname{vol}_{n-1}(D_t)\,dt$ (apply @thm-fubini to $\tilde 1$).

::: {#exm-quarter-disk}
## Either Order

Let $D = \{0 \le x \le 1,\ 1 \le y \le 1 + \sqrt{1 - x^2}\}$, a quarter disk. As a $y$-simple region,
$$\iint_Dx\,dA = \int_0^1x\sqrt{1 - x^2}\,dx = \Big[-\tfrac13(1 - x^2)^{3/2}\Big]_0^1 = \tfrac13 ;$$
as an $x$-simple region, $D = \{1 \le y \le 2,\ 0 \le x \le \sqrt{2y - y^2}\}$ and $\int_1^2\frac12(2y - y^2)\,dy = \frac13$.
:::

::: {#exm-order-matters}
## Choosing the Order

On the triangle $T = \{0 \le x \le y \le 1\}$, $\int_0^1\int_x^1e^{-y^2}\,dy\,dx$ cannot be evaluated directly ($e^{-y^2}$ has no elementary antiderivative), but as an $x$-simple region
$$\iint_Te^{-y^2}\,dA = \int_0^1\int_0^ye^{-y^2}\,dx\,dy = \int_0^1ye^{-y^2}\,dy = \tfrac12(1 - e^{-1}) .$$
:::

::: {#exm-volumes-slices}
## Volumes by Slices

1. **Two cylinders.** The solid $x^2 + y^2 \le a^2$, $x^2 + z^2 \le a^2$ has square slices $|y|, |z| \le \sqrt{a^2 - x^2}$, so its volume is $\int_{-a}^a4(a^2 - x^2)\,dx = \frac{16}{3}a^3$.
2. **A four-dimensional ball.** The slice of $x_1^2 + \dots + x_4^2 \le R^2$ at $x_1$ is a 3-ball of radius $\sqrt{R^2 - x_1^2}$, so by Cavalieri and $x_1 = R\sin\theta$,
$$\operatorname{vol}_4 = \int_{-R}^R\tfrac43\pi(R^2 - x_1^2)^{3/2}\,dx_1 = \tfrac43\pi R^4\int_{-\pi/2}^{\pi/2}\cos^4\theta\,d\theta = \tfrac43\pi R^4\cdot\tfrac{3\pi}{8} = \tfrac{\pi^2}{2}R^4 .$$
3. **A paraboloid inside a cylinder.** The region $z \ge x^2 + y^2$ inside $x^2 + (z - 2)^2 \le 1$ is $\{-1 \le x \le 1,\ 2 - \sqrt{1 - x^2} \le z \le 2 + \sqrt{1 - x^2},\ |y| \le \sqrt{z - x^2}\}$, so its volume is $\int_{-1}^1\int_{2-\sqrt{1-x^2}}^{2+\sqrt{1-x^2}}2\sqrt{z - x^2}\,dz\,dx \approx 8.204$.
:::

```python

import sympy as sp
from scipy.integrate import dblquad
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

x, y, z = sp.symbols("x y z", real=True)
print("rectangle:", sp.simplify(sp.integrate(x**2 * y + sp.exp(-2 * y), (y, 0, 1), (x, 1, 2))))
print("box:", sp.simplify(sp.integrate(x + sp.exp(-y + z), (x, 1, 2), (y, 0, 1), (z, 0, 2))))
print("quarter disk:", sp.integrate(x, (y, 1, 1 + sp.sqrt(1 - x**2)), (x, 0, 1)))
print("triangle:", sp.integrate(sp.exp(-y**2), (x, 0, y), (y, 0, 1)))
V, err = dblquad(lambda zz, xx: 2 * np.sqrt(zz - xx**2), -1, 1, lambda xx: 2 - np.sqrt(1 - xx**2), lambda xx: 2 + np.sqrt(1 - xx**2))
print("paraboloid in cylinder volume:", round(V, 6))

fig, axs = plt.subplots(1, 2, figsize=(10, 4.4))
for ax, title, horiz in [(axs[0], r"$y$-simple: $x \leq y \leq 1$ for each $x$", False), (axs[1], r"$x$-simple: $0 \leq x \leq y$ for each $y$", True)]:
    ax.fill([0, 0, 1], [0, 1, 1], color=BLUE, alpha=0.2)
    for c in np.linspace(0.08, 0.92, 8):
        if horiz: ax.plot([0, c], [c, c], color=ORANGE, lw=1.5)
        else: ax.plot([c, c], [c, 1], color=ORANGE, lw=1.5)
    ax.plot([0, 1, 0, 0], [0, 1, 1, 0], color=INK, lw=1)
    ax.set_aspect("equal"); ax.set_title(title)
plt.tight_layout(); plt.show()
```
