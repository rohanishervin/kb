# Decimal Representation of Real Numbers

**Prerequisites:** [The Axiom of Completeness](axiom-of-completeness.ipynb#def-decimal-completeness) · [Archimedean Property of $\mathbb{R}$](archimedean-property-of-R.ipynb)

Notation: for a digit string $c_0.c_1c_2\dots$ ($c_0 \in \mathbb{Z}_{\ge 0}$, $c_k \in \{0,\dots,9\}$) write $C_n = c_0.c_1\dots c_n$. The string is a **decimal expansion** of $c \ge 0$ if $C_n \le c \le C_n + 10^{-n}$ for all $n$; by the [uniqueness lemma](axiom-of-completeness.ipynb#lem-decimal-uniqueness) a string is an expansion of at most one number. For $c < 0$ one uses the expansion of $-c$.

## 1\. Existence and Non-Uniqueness

::: {#thm-decimal-existence}
## Every Real Number Has a Decimal Expansion

Every $c \ge 0$ has a decimal expansion, and $|c - C_n| \le 10^{-n}$, so $C_n \to c$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-decimal-existence
Let $c_0 = \lfloor c \rfloor$, so $C_0 \le c < C_0 + 1$. Suppose $C_{n-1} \le c < C_{n-1} + 10^{-(n-1)}$. Cut $[C_{n-1}, C_{n-1} + 10^{-(n-1)})$ into the ten half-open pieces $[C_{n-1} + d\,10^{-n},\ C_{n-1} + (d+1)10^{-n})$, $d = 0, \dots, 9$; $c$ lies in exactly one of them, and we let $c_n$ be its index. Then $C_n \le c < C_n + 10^{-n}$. By induction this holds for every $n$. $\blacksquare$
:::

::: {#prp-decimal-nonunique}
## When Two Expansions Give the Same Number

If $c_n \le 8$, the strings $c_0.c_1\dots c_n 999\dots$ and $c_0.c_1\dots c_{n-1}(c_n{+}1)000\dots$ are expansions of the same number $C_n + 10^{-n}$; for example $0.999\dots = 1$ and $1.2999\dots = 1.3$. Apart from this, different strings expand different numbers.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-decimal-nonunique
Let $x = C_n + 10^{-n}$. For the first string and $m \ge n$, $C_m = C_n + 9(10^{-n-1} + \dots + 10^{-m}) = C_n + 10^{-n} - 10^{-m}$, so $C_m \le x = C_m + 10^{-m}$; for $m < n$, $C_m \le C_n \le x \le C_m + 10^{-m}$ by nesting. For the second string $C'_m = x$ for $m \ge n$, and for $m < n$ the intervals contain $C'_n = x$ by nesting. So both are expansions of $x$.

Conversely let two different strings expand the same $c$, and let $k$ be the first index with $c_k \neq c'_k$, say $c_k < c'_k$. Then $C_k + 10^{-k} \le C'_k$, and
$$c \le C_k + 10^{-k} \le C'_k \le c,$$
so $c = C_k + 10^{-k} = C'_k$ and $c'_k = c_k + 1$. For $m > k$: $C'_m \le c = C'_k \le C'_m$ forces $c'_m = 0$; and $C_m + 10^{-m} \ge c = C_k + 10^{-k}$ forces $c_{k+1} = \dots = c_m = 9$ (the largest possible value of $C_m$ is $C_k + 10^{-k} - 10^{-m}$, attained only by nines). $\blacksquare$
:::

## 2\. Rational Numbers Have Periodic Expansions

An expansion is **eventually periodic** if there are $N, p \ge 1$ with $c_{k+p} = c_k$ for all $k > N$; we write e.g. $0.1\overline{6}$ for $0.1666\dots$.

::: {#thm-rational-periodic}
## Rational $\iff$ Eventually Periodic

A number $c \ge 0$ is rational if and only if its decimal expansion is eventually periodic.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-rational-periodic
**($\Rightarrow$) Long division.** Let $c = m/q$ with $m \ge 0$, $q \in \mathbb{N}$. Put $c_0 = \lfloor m/q \rfloor$, $r_0 = m - q c_0 \in \{0, \dots, q-1\}$, and for $k \ge 1$
$$c_k = \Big\lfloor \frac{10\, r_{k-1}}{q} \Big\rfloor, \qquad r_k = 10\, r_{k-1} - q\, c_k \in \{0, \dots, q-1\}.$$
Since $r_{k-1} < q$, $c_k \le 9$. By induction $c = C_k + \dfrac{r_k}{q\,10^k}$ (true for $k=0$; and $\frac{r_{k-1}}{q 10^{k-1}} = \frac{c_k}{10^k} + \frac{r_k}{q10^k}$). As $0 \le r_k < q$, $C_k \le c < C_k + 10^{-k}$: these are the digits of $c$. Each digit $c_{k+1}$ and remainder $r_{k+1}$ depend only on $r_k$. The remainders take at most $q$ values, so $r_i = r_j$ for some $i < j$ (pigeonhole), and then the digits repeat with period $p = j - i$ from index $i + 1$ on.

**($\Leftarrow$)** Let the digits be periodic with period $p$ after index $N$, and put $x = 10^N(c - C_N)$, so $c = C_N + 10^{-N}x$. Dividing $C_{N+n} \le c \le C_{N+n} + 10^{-N-n}$ by $10^{-N}$ shows that $x$ has the purely periodic expansion $0.d_1d_2\dots = 0.\overline{d_1 \dots d_p}$ with $d_j = c_{N+j}$; let $X_n = 0.d_1\dots d_n$ be its truncations and $D = d_1 \dots d_p$ read as an integer. The truncations of $x$ satisfy $X_{n+p} = D\,10^{-p} + 10^{-p} X_n$, so multiplying $X_{n+p} \le x \le X_{n+p} + 10^{-n-p}$ by $10^p$:
$$X_n \le 10^p x - D \le X_n + 10^{-n} \quad \text{for all } n.$$
Thus $10^p x - D$ and $x$ have the same expansion, hence are equal: $x = \dfrac{D}{10^p - 1} \in \mathbb{Q}$, and so $c \in \mathbb{Q}$. $\blacksquare$
:::

::: {#exm-decimal-periodic}
## Examples

1. $1/7 = 0.\overline{142857}$: the remainders of $1$ divided by $7$ run through $1, 3, 2, 6, 4, 5$ and then repeat. In agreement with the proof, $142857 / (10^6 - 1) = 1/7$.
2. $0.1\overline{6} = 0.1 + \frac{1}{10}\cdot\frac{6}{9} = \frac{1}{6}$.
3. $0.101001000100001\dots$ (blocks of zeros of lengths $1, 2, 3, \dots$ between ones) is not eventually periodic, so it is **irrational**.
:::

```python

import numpy as np
from fractions import Fraction

def long_division(m, q, ndigits):
    """Digits and remainders of m/q by long division (the proof of the theorem)."""
    c0, r = divmod(m, q)
    digits, rems = [], [r]
    for _ in range(ndigits):
        d, r = divmod(10 * r, q)
        digits.append(d); rems.append(r)
    return c0, digits, rems

for m, q in [(1, 7), (1, 6), (5, 14)]:
    c0, digits, rems = long_division(m, q, 14)
    first_repeat = next(j for j in range(1, len(rems)) if rems[j] in rems[:j])
    i = rems.index(rems[first_repeat])
    print(f"{m}/{q} = {c0}." + "".join(map(str, digits)) + "...",
          f"| remainders {rems[:first_repeat + 1]} | period {first_repeat - i} starting after digit {i}")

# check (<=) on 1/7: D / (10^p - 1)
print("142857 / 999999 =", Fraction(142857, 10**6 - 1))
```
