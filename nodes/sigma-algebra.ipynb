# σ-Algebras

**Prerequisites:** [Why Measure Theory](why-measure-theory.ipynb) · [The Cantor Set](cantor-set.ipynb) · [Logic Preliminaries](pre-logic.ipynb#de-morgans-laws)

A measure cannot be defined on all subsets of a space ([Banach–Tarski](why-measure-theory.ipynb#thm-banach-tarski)), and it must be [countably additive](why-measure-theory.ipynb#exm-finitely-not-countably-additive). So we need a family of subsets on which a measure can live, and this family must be closed under the operations that countable additivity uses. Such a family is a **σ-algebra**. The definition is simple; describing a concrete σ-algebra is not, and much of the technical work of measure theory consists of handling them.

Throughout, $X$ is a set and $2^X$ is the family of all subsets of $X$; $E^c = X \setminus E$ is the complement of $E$.

## 1\. Definition

::: {#def-sigma-algebra}
## σ-Algebra

A **σ-algebra** (or **σ-field**) on $X$ is a family $\mathcal{M} \subseteq 2^X$ such that

1. $\emptyset \in \mathcal{M}$;
2. if $E \in \mathcal{M}$, then $E^c \in \mathcal{M}$ (closed under complements);
3. if $E_1, E_2, \dots \in \mathcal{M}$, then $\bigcup_{i=1}^\infty E_i \in \mathcal{M}$ (closed under countable unions).
:::

::: {#def-algebra-of-sets}
## Algebra of Sets

A family $\mathcal{A} \subseteq 2^X$ is an **algebra** if $\emptyset \in \mathcal{A}$ and it is closed under complements and *finite* unions: $E, F \in \mathcal{A} \implies E \cup F \in \mathcal{A}$.
:::

::: {.callout-note}
## Intuition: What "σ" Means
In measure theory the prefix σ signals *countable*. An algebra must survive finite unions; a σ-algebra must survive countable unions. This is exactly the minimum we need: whatever sets a measure is defined on, we want to be able to take complements and countable unions of them and still be able to measure the result.
:::

::: {#prp-sigma-algebra-closure}
## Closure Properties

Let $\mathcal{M}$ be a σ-algebra on $X$. Then

1. $X \in \mathcal{M}$;
2. if $E_1, E_2, \dots \in \mathcal{M}$, then $\bigcap_{i=1}^\infty E_i \in \mathcal{M}$ (closed under countable intersections);
3. $\mathcal{M}$ is closed under finite unions and finite intersections, so it is an algebra;
4. if $E, F \in \mathcal{M}$, then $E \setminus F \in \mathcal{M}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-sigma-algebra-closure

1. $X = \emptyset^c \in \mathcal{M}$ by axioms 1 and 2.
2. By axiom 2 each $E_i^c \in \mathcal{M}$; by axiom 3, $\bigcup_i E_i^c \in \mathcal{M}$; by axiom 2 again its complement is in $\mathcal{M}$. By [De Morgan's laws](pre-logic.ipynb#de-morgans-laws),
$$\Big(\bigcup_{i=1}^\infty E_i^c\Big)^c = \bigcap_{i=1}^\infty (E_i^c)^c = \bigcap_{i=1}^\infty E_i .$$
3. For $E_1, \dots, E_n$ put $E_{n+1} = E_{n+2} = \dots = \emptyset$; then $\bigcup_{i=1}^n E_i = \bigcup_{i=1}^\infty E_i \in \mathcal{M}$. For intersections put $E_{n+1} = \dots = X$ and use part 2.
4. $E \setminus F = E \cap F^c$ is an intersection of two members of $\mathcal{M}$. $\blacksquare$
:::

::: {.callout-warning}
## Only Countable Operations Are Allowed
A σ-algebra is closed under *countable* unions and intersections, not under arbitrary ones. Taking an uncountable union can leave the σ-algebra. For instance, every subset of $\mathbb{R}$ is a union of singletons, yet not every subset of $\mathbb{R}$ will be measurable.
:::

::: {.callout-note}
## Why σ-Algebras Are Unavoidable
A good part of the technicalities of measure theory comes from keeping track of σ-algebras. There is no way around it: since a measure cannot be defined on all sets, it must be restricted to a family, and that family needs a structure. On a **finite** set $X$ the problem disappears: one simply measures all subsets, and σ-algebras are rarely mentioned.
:::

## 2\. The Two Extreme Examples

::: {#exm-extreme-sigma-algebras}
## The Largest and the Smallest σ-Algebra

1. $2^X$ is a σ-algebra, since it contains every subset. It is the largest σ-algebra on $X$.
2. $\{\emptyset, X\}$ is a σ-algebra, the **trivial σ-algebra**: complements, unions and intersections of $\emptyset$ and $X$ only ever give $\emptyset$ or $X$. Since every σ-algebra contains $\emptyset$ and $X$, it is the smallest one.
:::


## 3\. The σ-Algebra Generated by a Family

Suppose some sets are given, for example all intervals, and we want a σ-algebra containing them. We could start building: take countable unions of the given sets, then complements and countable intersections of what we obtained, then countable unions of those, and so on, putting everything into the σ-algebra. This process **never ends**: after any number of rounds, there are new countable unions of sets already built that must be added. So a σ-algebra generally cannot be listed constructively from its generators, and it is defined from above instead.

::: {#lem-intersection-of-sigma-algebras}
## Intersections of σ-Algebras

If $\{\mathcal{M}_\alpha\}_{\alpha \in J}$ is any non-empty collection of σ-algebras on $X$, then $\bigcap_{\alpha} \mathcal{M}_\alpha$ (the sets belonging to every $\mathcal{M}_\alpha$) is a σ-algebra on $X$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-intersection-of-sigma-algebras
Write $\mathcal{N} = \bigcap_\alpha \mathcal{M}_\alpha$. Each $\mathcal{M}_\alpha$ contains $\emptyset$, so $\emptyset \in \mathcal{N}$. If $E \in \mathcal{N}$, then $E \in \mathcal{M}_\alpha$ for every $\alpha$, hence $E^c \in \mathcal{M}_\alpha$ for every $\alpha$, so $E^c \in \mathcal{N}$. If $E_1, E_2, \dots \in \mathcal{N}$, then for each $\alpha$ all $E_i$ lie in $\mathcal{M}_\alpha$, so $\bigcup_i E_i \in \mathcal{M}_\alpha$; hence $\bigcup_i E_i \in \mathcal{N}$. $\blacksquare$
:::

::: {#def-generated-sigma-algebra}
## Generated σ-Algebra

Let $\mathcal{S} \subseteq 2^X$ be any family of subsets. The **σ-algebra generated by $\mathcal{S}$**, written $\sigma(\mathcal{S})$, is the intersection of all σ-algebras on $X$ that contain $\mathcal{S}$.
:::

::: {#prp-generated-is-smallest}
## $\sigma(\mathcal{S})$ Is the Smallest σ-Algebra Containing $\mathcal{S}$

$\sigma(\mathcal{S})$ is a σ-algebra, it contains $\mathcal{S}$, and it is contained in every σ-algebra that contains $\mathcal{S}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-generated-is-smallest
The collection of σ-algebras containing $\mathcal{S}$ is non-empty, because $2^X$ belongs to it. By @lem-intersection-of-sigma-algebras its intersection $\sigma(\mathcal{S})$ is a σ-algebra. Each member of the collection contains $\mathcal{S}$, so the intersection does. Finally, if $\mathcal{M} \supseteq \mathcal{S}$ is a σ-algebra, it is one of the sets being intersected, so $\sigma(\mathcal{S}) \subseteq \mathcal{M}$. $\blacksquare$
:::

::: {.callout-note}
## How to Show That a Set Belongs to $\sigma(\mathcal{S})$
By definition, $E \in \sigma(\mathcal{S})$ means that $E$ lies in **every** σ-algebra containing $\mathcal{S}$. The practical way to show this is to build $E$ from members of $\mathcal{S}$ using complements, countable unions and countable intersections. Every σ-algebra containing $\mathcal{S}$ is closed under these operations, so it contains $E$, and therefore so does their intersection $\sigma(\mathcal{S})$.
:::


## 4\. The Borel σ-Algebra on $\mathbb{R}$

The most important example starts from the open intervals of $\mathbb{R}$:
$$\mathcal{S} = \{(a, b) : a, b \in \mathbb{R},\ a < b\}.$$

::: {#def-borel-sigma-algebra-R}
## Borel σ-Algebra on $\mathbb{R}$

The σ-algebra $\mathcal{B} = \sigma(\mathcal{S})$ generated by the open intervals is the **Borel σ-algebra** of $\mathbb{R}$; its members are the **Borel sets**.
:::

Let us see how much it contains.

::: {#prp-singletons-borel}
## Singletons Are Borel Sets

For every $x \in \mathbb{R}$, $\{x\} \in \sigma(\mathcal{S})$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-singletons-borel
A single point is a countable intersection of open intervals:
$$\{x\} = \bigcap_{n=1}^\infty \Big(x - \frac1n,\ x + \frac1n\Big).$$
Indeed $x$ lies in every interval, and any $y \ne x$ is excluded once $\frac1n < |y - x|$. Every σ-algebra is closed under countable intersections, so $\{x\}$ belongs to every σ-algebra containing $\mathcal{S}$, hence to their intersection $\sigma(\mathcal{S})$. $\blacksquare$
:::

::: {#cor-countable-sets-borel}
## Countable Sets Are Borel Sets

Every countable subset of $\mathbb{R}$, for example $\mathbb{Q}$, belongs to $\sigma(\mathcal{S})$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-countable-sets-borel
A countable set $E = \{x_1, x_2, \dots\}$ is the countable union $\bigcup_i \{x_i\}$ of singletons, which are in $\sigma(\mathcal{S})$ by @prp-singletons-borel. $\blacksquare$
:::

::: {#prp-closed-intervals-borel}
## Closed Intervals Are Borel Sets

For $a \le b$, $[a, b] \in \sigma(\mathcal{S})$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-closed-intervals-borel
There are infinitely many ways; here are two.

1. As a countable intersection of open intervals: $[a, b] = \bigcap_{n=1}^\infty \big(a - \frac1n,\ b + \frac1n\big)$.
2. As a union of three Borel sets: $[a, b] = \{a\} \cup (a, b) \cup \{b\}$, where $(a,b) \in \mathcal{S}$ (if $a < b$) and the singletons are Borel by @prp-singletons-borel. $\blacksquare$
:::

::: {#prp-cantor-set-borel}
## The Cantor Set Is a Borel Set

The [Cantor set](cantor-set.ipynb#definition-and-construction) $C$ belongs to $\sigma(\mathcal{S})$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-cantor-set-borel
The construction of $C$ is itself a countable process: if $C_n$ is what remains after removing middle thirds $n$ times, then
$$C = \bigcap_{n=1}^\infty C_n .$$
Since σ-algebras are closed under countable intersections, it suffices that each $C_n$ is in $\sigma(\mathcal{S})$. Now $C_n$ is a union of $2^n$ closed intervals, for instance
$$C_2 = [0, \tfrac19] \cup [\tfrac29, \tfrac13] \cup [\tfrac23, \tfrac79] \cup [\tfrac89, 1],$$
and closed intervals are Borel by @prp-closed-intervals-borel. A finite union of Borel sets is Borel. $\blacksquare$
:::

Almost every set one meets is obtained in this way and lies in $\sigma(\mathcal{S})$. Still, $\sigma(\mathcal{S}) \ne 2^{\mathbb{R}}$: not every subset of $\mathbb{R}$ is a Borel set (we do not prove this). The sets that are not Borel cannot be constructed explicitly; we only know that they exist. Anything produced by an explicit procedure is produced by countably many unions, intersections and complements, and so is Borel.

The Borel σ-algebra can equally be described as the σ-algebra generated by the **open sets** of $\mathbb{R}$. This rests on the following fact, which is used many times later.

::: {#lem-open-sets-of-R}
## Open Subsets of $\mathbb{R}$

Every [open set](metric-space-topology.ipynb#open-sets-and-topology) $O \subseteq \mathbb{R}$ is a countable union of open intervals; one can even take intervals with rational endpoints.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-open-sets-of-R
Let $x \in O$. Since $O$ is open there is $r > 0$ with $(x - r, x + r) \subseteq O$, and by the [density of $\mathbb{Q}$](density-theorem-of-R.ipynb) there are rationals $p, q$ with $x - r < p < x < q < x + r$. So $x \in (p, q) \subseteq O$. Hence $O$ is the union of all intervals $(p, q) \subseteq O$ with $p < q$ rational. There are only countably many pairs of rationals, so this is a countable union. $\blacksquare$
:::

::: {#cor-borel-generated-by-open-sets}
## Borel Sets From Open Sets

$\sigma(\{\text{open intervals}\}) = \sigma(\{\text{open subsets of } \mathbb{R}\})$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-borel-generated-by-open-sets
Open intervals are open sets, so the right-hand σ-algebra contains all open intervals and hence contains the left-hand one (@prp-generated-is-smallest). Conversely, by @lem-open-sets-of-R every open set is a countable union of open intervals, so it lies in the left-hand σ-algebra; hence the left-hand one contains the right-hand one. $\blacksquare$
:::

The figure shows the first stages $C_n$ of the Cantor set. Each stage is a finite union of closed intervals, and the code checks that $C_n$ consists of $2^n$ intervals of total length $(2/3)^n$.

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

from fractions import Fraction as Fr

def cantor_stage(n):
    """C_n as a list of closed intervals with exact rational endpoints."""
    ivs = [(Fr(0), Fr(1))]
    for _ in range(n):
        ivs = [piece for (a, b) in ivs for piece in ((a, a + (b - a) / 3), (b - (b - a) / 3, b))]
    return ivs

fig, ax = plt.subplots(figsize=(10, 3.2))
for n in range(7):
    ivs = cantor_stage(n)
    assert len(ivs) == 2 ** n and sum(b - a for a, b in ivs) == Fr(2, 3) ** n
    for a, b in ivs:
        ax.plot([float(a), float(b)], [-n, -n], color=BLUE, lw=7, solid_capstyle="butt")
    ax.text(1.03, -n, f"$C_{n}$: {2**n} intervals, length $(2/3)^{n}$ = {float(Fr(2,3)**n):.3f}",
            va="center", fontsize=9, color=MUTED)
ax.set_xlim(-0.02, 1.6); ax.set_ylim(-6.6, 0.6)
ax.set_yticks([]); ax.grid(False); ax.spines["left"].set_visible(False)
ax.set_title("Each stage $C_n$ is a finite union of closed intervals, so $C = \\bigcap_n C_n$ is a Borel set")
plt.tight_layout(); plt.show()
```


## 5\. The Countable–Co-Countable σ-Algebra

Let $X$ be an arbitrary set, possibly uncountable, and let $\mathcal{S} = \{\{x\} : x \in X\}$ be the family of singletons. Which σ-algebra do the singletons generate? Countable unions of singletons give all **countable** subsets of $X$; their complements are the **co-countable** subsets, those whose complement is countable. Remarkably, nothing more is needed.

::: {#prp-countable-cocountable}
## The Countable–Co-Countable σ-Algebra

The family
$$\mathcal{M} = \{E \subseteq X : E \text{ is countable or } E^c \text{ is countable}\}$$
is a σ-algebra, and it is the σ-algebra generated by the singletons of $X$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-countable-cocountable
We check the three axioms of @def-sigma-algebra.

1. $\emptyset$ is countable (it has $0$ elements), so $\emptyset \in \mathcal{M}$.
2. If $E$ is countable, then $E^c$ has countable complement $E$; if $E^c$ is countable, then $E^c$ itself is countable. Either way $E^c \in \mathcal{M}$.
3. Let $E_1, E_2, \dots \in \mathcal{M}$ (not necessarily disjoint).
    - If every $E_i$ is countable, then $\bigcup_i E_i$ is a countable union of countable sets, hence countable.
    - Otherwise some $E_k$ is not countable; being in $\mathcal{M}$, it has countable complement. Since $E_k \subseteq \bigcup_i E_i$, we get $\big(\bigcup_i E_i\big)^c \subseteq E_k^c$, and a subset of a countable set is countable. So $\bigcup_i E_i$ is co-countable.

    In both cases $\bigcup_i E_i \in \mathcal{M}$.

So $\mathcal{M}$ is a σ-algebra, and it contains every singleton; hence $\sigma(\mathcal{S}) \subseteq \mathcal{M}$ by @prp-generated-is-smallest. Conversely, any σ-algebra containing the singletons contains every countable set (a countable union of singletons) and hence every co-countable set (a complement of a countable set), so it contains $\mathcal{M}$. Therefore $\sigma(\mathcal{S}) = \mathcal{M}$. $\blacksquare$
:::
