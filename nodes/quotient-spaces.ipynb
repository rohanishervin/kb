# Quotient Spaces

**Prerequisites:** [Subspaces](subspaces.ipynb) · [Complements and Direct Sums](direct-sums-and-complements.ipynb)

Given a subspace $W \le V$, we can decide to **ignore** differences that lie in $W$: two vectors are regarded as equal when they differ by an element of $W$. The resulting space $V/W$ is the **quotient** of $V$ by $W$. Geometrically, $W$ is collapsed to a single point and every line (or plane) parallel to $W$ becomes a single point of $V/W$. Quotients are the natural home of the [rank-nullity theorem](kernel-image-and-rank-nullity.ipynb#thm-rank-nullity).

Throughout, $V$ is a [vector space](vector-spaces.ipynb) over a field $K$ and $W \le V$ is a [subspace](subspaces.ipynb#def-subspace).

## 1\. Congruence Modulo a Subspace

Define a relation on $V$ by
$$x \sim y \iff x - y \in W,$$
read "$x$ is congruent to $y$ modulo $W$".

::: {#prp-quotient-equiv}
## Congruence Is an Equivalence Relation

$\sim$ is reflexive, symmetric and transitive.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-quotient-equiv
**Reflexive:** $x - x = 0 \in W$. **Symmetric:** if $x - y \in W$ then $y - x = (-1)(x - y) \in W$. **Transitive:** if $x - y \in W$ and $y - z \in W$ then $x - z = (x - y) + (y - z) \in W$. Each step uses that $W$ is a subspace. $\blacksquare$
:::

::: {#def-coset}
## Cosets

The equivalence class of $x \in V$ is
$$[x] = \{y \in V \mid y \sim x\} = \{x + w \mid w \in W\} = x + W,$$
called the **coset** of $x$ modulo $W$. Any element $y \in [x]$ is a **representative** of the class.
:::

As for every equivalence relation, two cosets are either equal or disjoint, the cosets cover $V$, and
$$[x] = [y] \iff x - y \in W.$$

::: {.callout-note}
## Intuition: Parallel Copies of $W$
The coset $x + W$ is $W$ shifted so that it passes through $x$. If $W$ is a line through the origin in $\mathbb{R}^2$, the cosets are all lines parallel to $W$, and they fill the plane without overlapping. These shifted copies are the [affine subspaces](affine-subspaces.ipynb) with direction $W$.
:::

## 2\. The Vector Space $V/W$

To add two cosets we add representatives. This only makes sense if the answer does not depend on which representatives we pick.

::: {#lem-quotient-compat}
## Compatibility with the Operations

If $x \sim x'$ and $y \sim y'$, then $x + y \sim x' + y'$ and $a x \sim a x'$ for every $a \in K$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-quotient-compat
$(x + y) - (x' + y') = (x - x') + (y - y') \in W$ and $a x - a x' = a (x - x') \in W$, since $W$ is closed under addition and scalar multiplication. $\blacksquare$
:::

::: {#def-quotient-space}
## Quotient Space

The **quotient space** $V/W$ is the set of all cosets $\{[x] \mid x \in V\}$ with the operations
$$[x] + [y] = [x + y], \qquad a\,[x] = [a x].$$
By @lem-quotient-compat these are well defined, and $V/W$ is a vector space over $K$: each axiom follows from the same axiom in $V$ applied to representatives. Its zero vector is $[0] = W$ and $-[x] = [-x]$.
:::

The animation shows why the definition works. The blue and orange lines are two cosets $[x]$ and $[y]$ of a line $W$. The representatives $x$ and $y$ slide freely along their cosets, yet the sum $x + y$ never leaves one fixed parallel line: the coset $[x + y]$.


```python

from la_manim import *

class CosetAddition(GridScene):
    def construct(self):
        self.plane.set_stroke(opacity=0.15)
        w = np.array([2.0, 1.0, 0.0]) / np.sqrt(5)
        x0, y0 = np.array([-0.6, 1.2, 0.0]), np.array([1.4, -0.4, 0.0])
        s, t = ValueTracker(0.0), ValueTracker(0.0)
        X = lambda: x0 + s.get_value() * w
        Y = lambda: y0 + t.get_value() * w
        coset = lambda p, color, width=3: Line(p - 9 * w, p + 9 * w, color=color, stroke_width=width)

        family = VGroup(*[coset(np.array([0, k * 0.5, 0]), C_VIOLET, 1).set_opacity(0.35) for k in range(-10, 11)])
        Wline = coset(ORIGIN, C_INK, 2.5)
        Wlab = MathTex("W = [0]", color=C_INK).scale(0.7).move_to(3.6 * w + np.array([0.3, -0.45, 0]))
        Lx, Ly, Lxy = coset(x0, C_BLUE), coset(y0, C_ORANGE), coset(x0 + y0, C_YELLOW, 4)
        labs = VGroup(MathTex("[x]", color=C_BLUE).move_to(x0 - 2.6 * w + np.array([0, 0.4, 0])),
                      MathTex("[y]", color=C_ORANGE).move_to(y0 + 2.6 * w + np.array([0, -0.4, 0])),
                      MathTex("[x+y]", color=C_YELLOW).move_to(x0 + y0 + 2.4 * w + np.array([-0.2, 0.45, 0])))

        style = dict(buff=0, stroke_width=5, max_tip_length_to_length_ratio=0.2, tip_length=0.2)
        xa = always_redraw(lambda: Arrow(ORIGIN, X(), color=C_BLUE, **style))
        ya = always_redraw(lambda: Arrow(ORIGIN, Y(), color=C_ORANGE, **style))
        sa = always_redraw(lambda: Arrow(ORIGIN, X() + Y(), color=C_YELLOW, **style))
        para = always_redraw(lambda: VGroup(DashedLine(X(), X() + Y(), color=C_ORANGE, stroke_width=2),
                                            DashedLine(Y(), X() + Y(), color=C_BLUE, stroke_width=2)))
        dots = always_redraw(lambda: VGroup(Dot(X(), color=C_BLUE), Dot(Y(), color=C_ORANGE),
                                            Dot(X() + Y(), color=C_YELLOW)))

        self.add(family, Wline, Wlab)
        self.play(Create(Lx), Create(Ly), FadeIn(labs[:2]))
        self.play(FadeIn(xa), FadeIn(ya), FadeIn(dots))
        self.play(FadeIn(para), FadeIn(sa), Create(Lxy), FadeIn(labs[2]))
        cap = caption(r"[x] + [y] = [x + y]")
        self.play(FadeIn(cap))
        for sv, tv in [(1.8, 0.0), (1.8, -2.2), (-1.5, -2.2), (-1.5, 1.6), (0.6, 2.4), (0.0, 0.0)]:
            self.play(s.animate.set_value(sv), t.animate.set_value(tv), run_time=1.4)
        msg = text_caption(r"Other representatives of $[x]$ and $[y]$: the sum stays on $[x+y]$")
        self.play(FadeIn(msg)); self.wait(2)

show(CosetAddition)
```


::: {#prp-quotient-projection}
## The Quotient Map

The map $\pi : V \to V/W$, $\pi(x) = [x]$, preserves sums and scalar multiples, is surjective, and $\pi(x) = [0]$ exactly when $x \in W$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-quotient-projection
$\pi(x + y) = [x + y] = [x] + [y]$ and $\pi(a x) = [a x] = a [x]$ by definition of the operations. Every coset is $[x] = \pi(x)$ for any representative $x$. Finally $[x] = [0]$ if and only if $x - 0 \in W$. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Collapsing $W$ to a Point
Passing from $V$ to $V/W$ makes everything in $W$ equal to zero and keeps only the information "in which parallel copy of $W$ does the vector lie". In the language of [linear maps](linear-maps.ipynb), $\pi$ is a surjective linear map with [kernel](kernel-image-and-rank-nullity.ipynb#def-kernel-image) $W$.
:::

## 3\. Dimension of a Quotient

::: {#thm-dim-quotient}
## Dimension of a Quotient Space

If $V$ is finite-dimensional, then $\dim (V/W) = \dim V - \dim W$. More precisely, if $v_1, \dots, v_k$ is a basis of $W$ extended to a basis $v_1, \dots, v_n$ of $V$, then $[v_{k+1}], \dots, [v_n]$ is a basis of $V/W$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-dim-quotient
**Spanning.** Write $x = \sum_{i=1}^n a_i v_i$. The part $\sum_{i \le k} a_i v_i$ lies in $W$, so $x \sim \sum_{i > k} a_i v_i$ and $[x] = \sum_{i > k} a_i [v_i]$.

**Independence.** If $\sum_{i > k} a_i [v_i] = [0]$, then $\sum_{i > k} a_i v_i \in W = \operatorname{span}(v_1, \dots, v_k)$, say $\sum_{i > k} a_i v_i = \sum_{i \le k} b_i v_i$. This is a linear relation among the basis vectors $v_1, \dots, v_n$, so all coefficients vanish; in particular $a_{k+1} = \dots = a_n = 0$.

So $V/W$ has a basis of $n - k$ elements. $\blacksquare$
:::

::: {#prp-quotient-complement}
## The Quotient Is Isomorphic to Any Complement

If $V = W \oplus U$, then $U \to V/W$, $u \mapsto [u]$, is a [linear isomorphism](coordinates-and-isomorphism-with-Kn.ipynb#def-linear-isomorphism).
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-quotient-complement
The map is the restriction of $\pi$ to $U$, so it preserves sums and multiples.

**Injective.** If $[u] = [u']$ with $u, u' \in U$, then $u - u' \in W \cap U = \{0\}$, so $u = u'$.

**Surjective.** Given $[x]$, decompose $x = w + u$ with $w \in W$, $u \in U$. Then $x - u = w \in W$, so $[x] = [u]$. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Each Coset Meets the Complement Once
A complement $U$ is transverse to $W$, so every parallel copy $x + W$ crosses $U$ in exactly one point. Sending each coset to that point identifies $V/W$ with $U$. The identification depends on the choice of $U$, but $V/W$ itself does not: the quotient is the choice-free version of "a complement of $W$".
:::

## 4\. Examples

::: {#exm-quotients}
## Quotients in Practice

1. **A plane modulo a line.** Let $W = \operatorname{span}(2, 1) \le \mathbb{R}^2$. The cosets are the lines parallel to $W$. The $y$-axis is a complement of $W$, so $\mathbb{R}^2 / W \cong \mathbb{R}$: the coset of $(x, y)$ is determined by the point where its line crosses the $y$-axis, namely $y - x/2$. Indeed $\dim (\mathbb{R}^2/W) = 2 - 1 = 1$.
2. **Forgetting a coordinate.** In $K^3$ with $W = \operatorname{span}(e_3)$, two vectors are congruent when they agree in the first two coordinates. So $K^3 / W \cong K^2$ via $[(x, y, z)] \mapsto (x, y)$.
3. **Polynomials modulo those vanishing at $0$.** In $P_d$ let $W = \{p \mid p(0) = 0\}$, which has dimension $d$ (basis $x, \dots, x^d$). Then $p \sim q$ exactly when $p(0) = q(0)$, so a coset is determined by one number and $P_d / W \cong \mathbb{R}$ via $[p] \mapsto p(0)$, matching $\dim = (d + 1) - d = 1$.
:::
