# Space Curves and the Frenet–Serret Formulas

**Prerequisites:** [Curvature of Plane Curves](curvature-of-plane-curves.ipynb) · [The Cross Product](cross-product.ipynb) · [Ordinary Differential Equations and Picard's Theorem](ode-and-picards-thm.ipynb)

Throughout, $\gamma$ is a $C^3$ smooth curve in $\mathbb{R}^3$ parametrized by arc length $s$, with unit tangent $T = \gamma'$. In space there is no turning angle, so curvature is defined as the speed at which $T$ turns, and a second invariant, the torsion, measures how fast the curve leaves its plane of bending.

## 1\. The Frenet Frame

::: {#def-frenet-frame}
## Curvature, Normal, Binormal

The **curvature** is $\kappa(s) = |T'(s)| \ge 0$. A **Frenet curve** is one with $\kappa(s) > 0$ for all $s$; for it define the **principal normal** $N = T'/\kappa$ and the **binormal** $B = T\times N$.
:::

Since $|T| = 1$, $T' \perp T$ ([constant length](parametrized-curves.ipynb#prp-curve-derivative-rules)); so $(T, N, B)$ is an orthonormal basis, positively oriented by the [geometry of the cross product](cross-product.ipynb#thm-cross-geometry). The plane through $\gamma(s)$ spanned by $T$ and $N$ is the **osculating plane**: the plane in which the curve is bending at that instant.

::: {#thm-frenet-serret}
## Frenet–Serret Formulas

For a Frenet curve there is a continuous function $\tau$, the **torsion**, such that
$$T' = \kappa N, \qquad N' = -\kappa T + \tau B, \qquad B' = -\tau N .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-frenet-serret
Write $(e_1, e_2, e_3) = (T, N, B)$ and $e_i' = \sum_j a_{ij}e_j$ with $a_{ij} = e_i'\cdot e_j$ (expansion in an [orthonormal basis](orthogonality-and-gram-schmidt.ipynb#prp-orthogonal-independent)). Differentiating $e_i\cdot e_j = \delta_{ij}$ gives $a_{ij} + a_{ji} = 0$: the matrix $(a_{ij})$ is antisymmetric, so $a_{ii} = 0$. By definition $T' = \kappa N$, so $a_{12} = \kappa$, $a_{13} = 0$. Put $\tau = a_{23} = N'\cdot B$. Antisymmetry gives $a_{21} = -\kappa$, $a_{31} = 0$, $a_{32} = -\tau$, which are the three formulas. $\tau$ is continuous because $N' = (T'/\kappa)'$ involves $\gamma'''$ continuously. $\blacksquare$
:::

::: {#prp-planar-iff-torsion}
## Zero Torsion Means Planar

A Frenet curve lies in a plane iff $\tau \equiv 0$; the plane is then the osculating plane.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-planar-iff-torsion
If $\tau \equiv 0$, then $B' = 0$, so $B$ is a constant vector. Let $f(s) = (\gamma(s) - \gamma(s_0))\cdot B$; then $f' = T\cdot B = 0$ and $f(s_0) = 0$, so $\gamma$ lies in the plane through $\gamma(s_0)$ with normal $B$.
Conversely, if $(\gamma(s) - p)\cdot n = 0$ for a unit vector $n$, differentiating twice gives $T\cdot n = 0$ and $T'\cdot n = 0$, so $N\cdot n = 0$. Hence $B$ is a unit vector orthogonal to $T$ and $N$, i.e. $B = \pm n$; being continuous, $B$ is constant, so $\tau N = -B' = 0$. $\blacksquare$
:::

## 2\. Arbitrary Parameters

::: {#prp-space-curvature-formulas}
## Curvature and Torsion for Any Parameter

Let $\gamma(t)$ be a $C^3$ smooth curve with arc length $s(t)$ and $\kappa > 0$. Then
$$\gamma'' = s''T + \kappa(s')^2N, \qquad \gamma'\times\gamma'' = \kappa(s')^3B,$$
$$\kappa = \frac{|\gamma'\times\gamma''|}{|\gamma'|^3}, \qquad \tau = \frac{(\gamma'\times\gamma'')\cdot\gamma'''}{|\gamma'\times\gamma''|^2} .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-space-curvature-formulas
$\gamma' = s'T$ and, by the chain rule and @thm-frenet-serret, $\gamma'' = s''T + \kappa(s')^2N$. Then $\gamma'\times\gamma'' = \kappa(s')^3\,T\times N = \kappa(s')^3B$, and $s' = |\gamma'|$ gives $\kappa$. Differentiating once more, the only term of $\gamma'''$ with a $B$-component comes from $\kappa(s')^2\frac{dN}{ds}s'$, whose $B$-component is $\kappa\tau(s')^3$. Hence $(\gamma'\times\gamma'')\cdot\gamma''' = \kappa(s')^3\cdot\kappa\tau(s')^3 = \tau|\gamma'\times\gamma''|^2$. $\blacksquare$
:::

The acceleration always lies in the osculating plane; the binormal direction appears only in the third derivative.

::: {#exm-helix-frenet}
## The Helix

For $\gamma(t) = (a\cos t, a\sin t, bt)$ with $a > 0$, put $c = \sqrt{a^2 + b^2}$; then $s = ct$ and
$$T = \tfrac1c(-a\sin t, a\cos t, b), \quad N = (-\cos t, -\sin t, 0), \quad B = \tfrac1c(b\sin t, -b\cos t, a),$$
$$\kappa = \frac{a}{a^2 + b^2}, \qquad \tau = \frac{b}{a^2 + b^2} .$$
The normal points horizontally toward the axis; the torsion has the sign of $b$ (right- or left-handed helix), and $b = 0$ gives a circle with $\kappa = \frac1a$, $\tau = 0$.
:::

## 3\. Curvature and Torsion Determine the Curve

A rigid motion $x \mapsto Rx + c$ ($R$ orthogonal, $\det R = 1$) maps $T, N, B$ to $RT, RN, RB$ (it preserves lengths, dot products and cross products: $Ru\times Rv = R(u\times v)$ for rotations), so it preserves $\kappa$ and $\tau$.

::: {#thm-space-curves-fundamental}
## Fundamental Theorem of Space Curves

Two Frenet curves $\gamma_1, \gamma_2 : [0, L] \to \mathbb{R}^3$, parametrized by arc length, with $\kappa_1 = \kappa_2$ and $\tau_1 = \tau_2$, differ by a rigid motion. Conversely, for continuous $\kappa > 0$ and $\tau$ on $[0, L]$ there is a Frenet curve with these curvature and torsion.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-space-curves-fundamental
Apply the rigid motion that maps $\gamma_1(0)$ to $\gamma_2(0)$ and the positively oriented orthonormal frame $(T_1, N_1, B_1)(0)$ to $(T_2, N_2, B_2)(0)$. Let $\varphi = T_1\cdot T_2 + N_1\cdot N_2 + B_1\cdot B_2$. By @thm-frenet-serret, with common $\kappa, \tau$,
$$\varphi' = \kappa N_1\cdot T_2 + \kappa T_1\cdot N_2 + (-\kappa T_1 + \tau B_1)\cdot N_2 + N_1\cdot(-\kappa T_2 + \tau B_2) - \tau N_1\cdot B_2 - \tau B_1\cdot N_2 = 0 .$$
So $\varphi \equiv 3$; each term is at most $1$, so $T_1 = T_2$, and $\gamma_1 - \gamma_2$ is constant, equal to $0$.
For existence, the Frenet–Serret formulas are a linear system of ODEs for $(T, N, B) \in \mathbb{R}^9$ with continuous coefficients; it has a solution with a prescribed positively oriented orthonormal initial frame ([Picard's theorem](ode-and-picards-thm.ipynb)), the antisymmetry of the coefficient matrix keeps the frame orthonormal (the dot products satisfy a linear system with the constant solution $\delta_{ij}$), and $\gamma(s) = \int_0^sT$ is the curve. $\blacksquare$
:::

```python

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

t, a, b = sp.symbols("t a b", positive=True)
def kappa_tau(g):
    d1, d2, d3 = g.diff(t), g.diff(t, 2), g.diff(t, 3)
    c = d1.cross(d2)
    return sp.simplify(sp.sqrt(c.dot(c)) / sp.sqrt(d1.dot(d1))**3), sp.simplify(c.dot(d3) / c.dot(c))
print("helix (kappa, tau):", kappa_tau(sp.Matrix([a * sp.cos(t), a * sp.sin(t), b * t])))
k, tt = kappa_tau(sp.Matrix([t, t**2, t**3]))
print("twisted cubic (t, t^2, t^3): kappa =", k, "| tau =", tt)

fig = plt.figure(figsize=(7, 5.6))
ax = fig.add_subplot(projection="3d")
A, Bh = 1.0, 0.25
ts = np.linspace(0, 4 * np.pi, 600)
ax.plot(A * np.cos(ts), A * np.sin(ts), Bh * ts, color=INK, lw=1.4)
c = np.hypot(A, Bh)
for t0 in [1.0, 4.0, 7.5, 10.5]:
    p = np.array([A * np.cos(t0), A * np.sin(t0), Bh * t0])
    T = np.array([-A * np.sin(t0), A * np.cos(t0), Bh]) / c
    N = np.array([-np.cos(t0), -np.sin(t0), 0.0])
    B = np.cross(T, N)
    for v, col in [(T, BLUE), (N, ORANGE), (B, AQUA)]:
        ax.quiver(*p, *(0.6 * v), color=col, lw=1.8, arrow_length_ratio=0.15)
ax.plot([], [], color=BLUE, label="$T$"); ax.plot([], [], color=ORANGE, label="$N$"); ax.plot([], [], color=AQUA, label="$B$")
ax.legend(loc="upper left"); ax.set_title(r"Frenet frames on the helix $(\cos t, \sin t, t/4)$")
plt.tight_layout(); plt.show()
```
