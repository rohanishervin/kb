# Stokes' Theorem

**Prerequisites:** [Green's Theorem](greens-theorem.ipynb) · [Flux, Divergence and Curl](flux-divergence-and-curl.ipynb) · [Parametrized Surfaces and Surface Area](parametrized-surfaces.ipynb) · [The Cross Product](cross-product.ipynb#prp-cross-algebra)

In Green's theorem the integrand $\frac{\partial F_2}{\partial x} - \frac{\partial F_1}{\partial y}$ is the third component of $\operatorname{curl}(F_1, F_2, 0)$, i.e. $\operatorname{curl}F\cdot\mathbf k$, with $\mathbf k$ the normal of the plane. Stokes' theorem replaces the flat region by a curved surface.

## 1\. Simple Surfaces and Their Boundaries

::: {#def-simple-surface}
## Simple Surface, Boundary Orientation

A **simple surface** is $D = \varphi(W)$, where $W$ is a plane region as in [Green's theorem](greens-theorem.ipynb#thm-green) (bounded by finitely many piecewise smooth simple closed curves) and $\varphi$ is a smooth parametrized surface, injective on $W\cup\partial W$, of class $C^2$ on an open set containing it. Its **boundary** is $\partial D = \varphi(\partial W)$, and its standard normal is $\mathbf n = \frac{\mathbf r_u\times\mathbf r_v}{|\mathbf r_u\times\mathbf r_v|}$. The **positive orientation** of $\partial D$ is the image under $\varphi$ of the positive orientation of $\partial W$. Geometrically: walking along $\partial D$ in this direction with one's head pointing along $\mathbf n$, the surface is on the left; equivalently, $(T, \nu, \mathbf n)$ is positively oriented, $\nu$ being the unit vector tangent to $D$, normal to $\partial D$ and pointing into $D$.
:::

(The geometric description follows because $D\varphi$ maps the plane's positive frame $(e_1, e_2)$ to $(\mathbf r_u, \mathbf r_v)$, and $(\mathbf r_u, \mathbf r_v, \mathbf r_u\times\mathbf r_v)$ is [positively oriented](cross-product.ipynb#thm-cross-geometry).)

::: {#exm-simple-surfaces}
## Two Simple Surfaces

1. $\varphi(u, v) = (\cos u, \sin u, v)$ on $W = [0, \pi]\times[-1, 1]$ is the half of the cylinder $x^2 + y^2 = 1$ with $y \ge 0$, $|z| \le 1$; $\mathbf r_u\times\mathbf r_v = (\cos u, \sin u, 0)$, the outward normal. $\partial D$ consists of two half-circles and two segments; on the bottom edge $v = -1$, traversed with $u$ increasing, $T = (-\sin u, \cos u, 0)$; on the edge $u = \pi$ the direction is $T = (0, 0, -1)$.
2. $\varphi(u, v) = (u, v, u^2 + v^2)$ on $W = \{1 \le |u| + |v| \le 2\}$, a piece of the paraboloid $z = x^2 + y^2$ over the region between two squares. $\mathbf n = \frac{(-2u, -2v, 1)}{\sqrt{4u^2 + 4v^2 + 1}}$ points into the bowl; $\partial D$ is the image of the outer square traversed counterclockwise and the inner square traversed clockwise (seen from above).
:::

## 2\. The Theorem

::: {#thm-stokes}
## Stokes' Theorem

Let $D$ be a simple surface with positively oriented boundary and $F$ a $C^1$ vector field on an open set containing $D\cup\partial D$. Then
$$\oint_{\partial D}F\cdot d\mathbf r = \iint_D(\operatorname{curl}F)\cdot\mathbf n\,dS .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-stokes
*Pull back to $W$.* For a curve $c(t) = (u(t), v(t))$ in $W$, $(\varphi\circ c)' = \mathbf r_uu' + \mathbf r_vv'$, so
$$\oint_{\partial D}F\cdot d\mathbf r = \oint_{\partial W}P\,du + Q\,dv, \qquad P = (F\circ\varphi)\cdot\mathbf r_u, \quad Q = (F\circ\varphi)\cdot\mathbf r_v,$$
where $P, Q$ are $C^1$ because $\varphi$ is $C^2$. By [Green's theorem](greens-theorem.ipynb#thm-green) this is $\iint_W(Q_u - P_v)\,du\,dv$.
*Compute the integrand.* With $J = DF$ evaluated at $\varphi(u, v)$, the chain rule gives $Q_u = (J\mathbf r_u)\cdot\mathbf r_v + F\cdot\mathbf r_{vu}$ and $P_v = (J\mathbf r_v)\cdot\mathbf r_u + F\cdot\mathbf r_{uv}$. Since $\mathbf r_{uv} = \mathbf r_{vu}$ ([Schwarz](higher-partials-and-taylor.ipynb#thm-schwarz)),
$$Q_u - P_v = \mathbf r_v^{\mathsf T}(J - J^{\mathsf T})\mathbf r_u .$$
The antisymmetric matrix $J - J^{\mathsf T}$, with entries $\partial_jF_i - \partial_iF_j$, acts as $h \mapsto w\times h$ for $w = \operatorname{curl}F$: e.g. the first entry of $(J - J^{\mathsf T})h$ is $(\partial_2F_1 - \partial_1F_2)h_2 + (\partial_3F_1 - \partial_1F_3)h_3 = w_2h_3 - w_3h_2$. Hence, by the [triple product rule](cross-product.ipynb#prp-cross-algebra),
$$Q_u - P_v = \mathbf r_v\cdot(w\times\mathbf r_u) = w\cdot(\mathbf r_u\times\mathbf r_v),$$
and $\iint_W(\operatorname{curl}F)(\varphi)\cdot(\mathbf r_u\times\mathbf r_v)\,du\,dv = \iint_D\operatorname{curl}F\cdot\mathbf n\,dS$ by the [definition of flux](flux-divergence-and-curl.ipynb#def-flux). $\blacksquare$
:::

For a region in the plane $z = 0$ with $\mathbf n = \mathbf k$ this is Green's theorem. Both sides depend only on $D$ and its orientation, not on $\varphi$; in particular, **surfaces with the same oriented boundary have the same flux of $\operatorname{curl}F$**.

::: {#exm-stokes-paraboloid}
## Choosing a Simpler Surface

Let $M$ be the part of $z = 1 - x^2 - y^2$ with $z \ge 0$, normal pointing up, and $F = (x - y)\,\mathbf i + (x + y^2)\,\mathbf j + z^2\,\mathbf k$. Then $\partial M$ is the unit circle in the plane $z = 0$, traversed counterclockwise (seen from above), and $\operatorname{curl}F = (0, 0, 2)$. The flat unit disk with normal $\mathbf k$ has the same oriented boundary, so
$$\iint_M\operatorname{curl}F\cdot\mathbf n\,dS = \oint_{\partial M}F\cdot d\mathbf r = \iint_{x^2+y^2\le1}(0, 0, 2)\cdot\mathbf k\,dA = 2\pi .$$
:::

## 3\. Orientable Surfaces

::: {#def-orientable-surface}
## Orientable Surface

A set $D = D_1\cup\dots\cup D_k \subseteq \mathbb{R}^3$ is an **orientable surface** if each $D_i$ is a simple surface, distinct $D_i$ meet only along their boundaries in finitely many smooth arcs, each boundary arc of $D_i$ is shared with at most one other $D_j$, and the parametrizations can be chosen so that on each shared arc the two induced boundary orientations are opposite. The **boundary** $\partial D$ is the union of the unshared arcs. If $\partial D = \emptyset$, $D$ is a **closed** surface (e.g. a sphere or a torus, each made of a few simple pieces).
:::

The **Möbius strip** (a strip of paper whose ends are glued after a half twist) is not orientable: carrying a unit normal continuously once around the central circle returns it reversed.

::: {#cor-stokes-orientable}
## Stokes' Theorem for Orientable Surfaces

For an orientable surface $D$ with the induced orientation on $\partial D$ and $F$ $C^1$ near $D$: $\oint_{\partial D}F\cdot d\mathbf r = \iint_D\operatorname{curl}F\cdot\mathbf n\,dS$. If $D$ is closed, $\iint_D\operatorname{curl}F\cdot\mathbf n\,dS = 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-stokes-orientable
Apply @thm-stokes to each $D_i$ and add: the integrals over shared arcs cancel in pairs (opposite orientations), leaving the integral over $\partial D$, which is $0$ if $\partial D = \emptyset$. $\blacksquare$
:::

## 4\. Potentials and the Meaning of Curl

A $C^1$ gradient field on an open set in $\mathbb{R}^3$ has $\operatorname{curl}F = 0$ ([$\operatorname{curl}\nabla = 0$](flux-divergence-and-curl.ipynb#thm-curl-grad-div-curl)). The converse fails: $F = \frac{-y\,\mathbf i + x\,\mathbf j}{x^2 + y^2}$ on $\mathbb{R}^3$ minus the $z$-axis has $\operatorname{curl}F = 0$, but its integral over the unit circle in the $xy$-plane is $2\pi$. What is needed is a condition on the domain $U$: if every piecewise smooth simple closed curve in $U$ is the boundary of an orientable surface lying in $U$, Stokes' theorem gives $\oint_\gamma F\cdot d\mathbf r = 0$ for such curves, and (as in the plane) $F$ is a gradient. $\mathbb{R}^3\setminus\{0\}$ has this property (a surface can always dodge one point), $\mathbb{R}^3$ minus a line does not. For star-shaped domains a potential is written down explicitly by the [Poincaré lemma](divergence-theorem.ipynb#thm-poincare-lemma).

Let $D_\rho$ be the flat disk of radius $\rho$ about $P$ orthogonal to a unit vector $\mathbf n$. By Stokes' theorem and the [mean value theorem for integrals](multiple-integrals.ipynb#thm-mean-value-multiple),
$$(\operatorname{curl}F\cdot\mathbf n)(P) = \lim_{\rho\to0}\frac{1}{\pi\rho^2}\oint_{\partial D_\rho}F\cdot d\mathbf r,$$
the **circulation** of $F$ per unit area in the plane orthogonal to $\mathbf n$. It is largest when $\mathbf n$ points along $\operatorname{curl}F(P)$, the instantaneous axis of rotation of the field. For a rigid rotation $F = \omega(-y, x, 0)$, $\operatorname{curl}F = 2\omega\,\mathbf k$: half its length is the angular speed.

```python

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

x, y, z, t, u, v, r, th = sp.symbols("x y z t u v r theta", real=True)
F = sp.Matrix([x - y, x + y**2, z**2])
X = sp.Matrix([x, y, z])
J = F.jacobian(X)
w = sp.Matrix([sp.diff(F[2], y) - sp.diff(F[1], z), sp.diff(F[0], z) - sp.diff(F[2], x), sp.diff(F[1], x) - sp.diff(F[0], y)])
h = sp.Matrix(sp.symbols("h1:4"))
print("(J - J^T) h = curl F x h:", sp.simplify((J - J.T) * h - w.cross(h)) == sp.zeros(3, 1))
phi = sp.Matrix([r * sp.cos(th), r * sp.sin(th), 1 - r**2])
N = phi.diff(r).cross(phi.diff(th))
flux = sp.integrate(sp.simplify(w.subs({x: phi[0], y: phi[1], z: phi[2]}).dot(N)), (r, 0, 1), (th, 0, 2 * sp.pi))
c = sp.Matrix([sp.cos(t), sp.sin(t), 0])
circ = sp.integrate(F.subs({x: c[0], y: c[1], z: c[2]}).dot(c.diff(t)), (t, 0, 2 * sp.pi))
print("paraboloid cap: flux of curl F =", flux, "| circulation on the boundary =", circ, "| (normal r_r x r_theta points up:", sp.simplify(N[2]) , ")")

fig = plt.figure(figsize=(6.6, 5.2))
ax = fig.add_subplot(projection="3d")
R_, T_ = np.meshgrid(np.linspace(0, 1, 25), np.linspace(0, 2 * np.pi, 50))
ax.plot_surface(R_ * np.cos(T_), R_ * np.sin(T_), 1 - R_**2, color=BLUE, alpha=0.35, linewidth=0)
tt = np.linspace(0, 2 * np.pi, 200)
ax.plot(np.cos(tt), np.sin(tt), 0 * tt, color=ORANGE, lw=2.5)
for k in [0.3, 2.4, 4.5]:
    ax.quiver(np.cos(k), np.sin(k), 0, -np.sin(k) * 0.4, np.cos(k) * 0.4, 0, color=ORANGE, lw=2, arrow_length_ratio=0.4)
for (px, py) in [(0, 0), (0.5, 0.2), (-0.4, -0.5), (0.3, -0.6)]:
    nz = np.array([2 * px, 2 * py, 1.0]); nz /= np.linalg.norm(nz)
    ax.quiver(px, py, 1 - px**2 - py**2, *(0.45 * nz), color=INK, lw=1.2, arrow_length_ratio=0.25)
ax.set_title(r"$z = 1 - x^2 - y^2$: upward $\mathbf{n}$ and the induced orientation of $\partial M$")
plt.tight_layout(); plt.show()
```
