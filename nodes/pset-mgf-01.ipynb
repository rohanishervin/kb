# Problem Set: Moment Generating Functions



## Problem 1: MGF and Moments of the Binomial Distribution

**Problem Statement:**

Let $X \sim \text{Bin}(n, p)$, where $n \in \mathbb{N}$ and $p \in (0, 1)$ with $q = 1 - p$.

1. Derive the [moment generating function](moment-generating-functions.ipynb) $M_X(t) = \mathbb{E}[e^{tX}]$.  
2. Using $M_X(t)$, compute the first moment $\mathbb{E}[X]$ and the second moment $\mathbb{E}[X^2]$.  
3. Determine the [variance](variance.ipynb) $\text{Var}(X) = \mathbb{E}[X^2] - (\mathbb{E}[X])^2$.

::: {.callout-tip collapse="true"}
## Solution

**1. Derivation of the MGF:**

The [probability mass function](probability-mass-function.ipynb) of $X$ is:

$$
P(X = x) = \binom{n}{x} p^x (1-p)^{n-x}, \quad x \in \{0, 1, \dots, n\}
$$

By definition of the [moment generating function](moment-generating-functions.ipynb):

$$
\begin{aligned}
M_X(t) &= \mathbb{E}[e^{tX}] = \sum_{x=0}^n e^{tx} P(X = x) \\
&= \sum_{x=0}^n e^{tx} \binom{n}{x} p^x (1-p)^{n-x} \\
&= \sum_{x=0}^n \binom{n}{x} (p e^t)^x (1-p)^{n-x}
\end{aligned}
$$

Applying the [Binomial Theorem](binomial-theorem.ipynb) $\sum_{x=0}^n \binom{n}{x} a^x b^{n-x} = (a + b)^n$ with $a = p e^t$ and $b = 1 - p = q$:

$$
M_X(t) = (p e^t + 1 - p)^n = (p e^t + q)^n
$$

**2. First and Second Moments:**

The moments are obtained by differentiating $M_X(t)$ and evaluating at $t = 0$:

- **First Moment $\mathbb{E}[X]$:**

  $$
  M_X'(t) = \frac{d}{dt} (p e^t + q)^n = n(p e^t + q)^{n-1} \cdot p e^t
  $$

  Evaluating at $t = 0$:

  $$
  \mathbb{E}[X] = M_X'(0) = n(p e^0 + q)^{n-1} p e^0 = n(p + q)^{n-1} p = n(1)^{n-1} p = np
  $$

- **Second Moment $\mathbb{E}[X^2]$:**

  Applying the product rule to $M_X'(t) = n(p e^t + q)^{n-1} (p e^t)$:

  $$
  M_X''(t) = n(n-1)(p e^t + q)^{n-2} (p e^t)^2 + n(p e^t + q)^{n-1} (p e^t)
  $$

  Evaluating at $t = 0$:

  $$
  \begin{aligned}
  \mathbb{E}[X^2] = M_X''(0) &= n(n-1)(p + q)^{n-2} p^2 + n(p + q)^{n-1} p \\
  &= n(n-1)p^2 + np \\
  &= n^2 p^2 - np^2 + np
  \end{aligned}
  $$

**3. Variance:**

$$
\begin{aligned}
\text{Var}(X) &= \mathbb{E}[X^2] - (\mathbb{E}[X])^2 \\
&= (n^2 p^2 - np^2 + np) - (np)^2 \\
&= np - np^2 \\
&= np(1 - p) = npq
\end{aligned}
$$

$\blacksquare$
:::



## Problem 2: MGF and Moments of the Poisson Distribution

**Problem Statement:**

Let $X \sim \text{Poisson}(\lambda)$ with parameter $\lambda > 0$.

1. Derive the [moment generating function](moment-generating-functions.ipynb) $M_X(t) = \mathbb{E}[e^{tX}]$.  
2. Using $M_X(t)$, compute the first moment $\mathbb{E}[X]$ and the second moment $\mathbb{E}[X^2]$.  
3. Verify that $\mathbb{E}[X] = \lambda$ and $\text{Var}(X) = \lambda$.

::: {.callout-tip collapse="true"}
## Solution

**1. Derivation of the MGF:**

The [probability mass function](probability-mass-function.ipynb) of $X$ is:

$$
P(X = x) = \frac{e^{-\lambda} \lambda^x}{x!}, \quad x \in \{0, 1, 2, \dots\}
$$

By definition of the [moment generating function](moment-generating-functions.ipynb):

$$
M_X(t) = \mathbb{E}[e^{tX}] = \sum_{x=0}^\infty e^{tx} \frac{e^{-\lambda} \lambda^x}{x!} = \sum_{x=0}^\infty \frac{(\lambda e^t)^x e^{-\lambda}}{x!}
$$

Recognizing that for any parameter $\mu > 0$, $\sum_{x=0}^\infty \frac{\mu^x e^{-\mu}}{x!} = 1 \implies \sum_{x=0}^\infty \frac{\mu^x}{x!} = e^\mu$, we set $\mu = \lambda e^t$:

$$
\sum_{x=0}^\infty \frac{(\lambda e^t)^x e^{-\lambda e^t}}{x!} = 1 \implies \sum_{x=0}^\infty \frac{(\lambda e^t)^x e^{-\lambda}}{x!} = \frac{e^{-\lambda}}{e^{-\lambda e^t}} = e^{\lambda(e^t - 1)}
$$

Therefore:

$$
M_X(t) = e^{\lambda(e^t - 1)}
$$

**2. First and Second Moments:**

- **First Moment $\mathbb{E}[X]$:**

  $$
  M_X'(t) = \frac{d}{dt} e^{\lambda(e^t - 1)} = \lambda e^t e^{\lambda(e^t - 1)}
  $$

  Evaluating at $t = 0$:

  $$
  \mathbb{E}[X] = M_X'(0) = \lambda e^0 e^{\lambda(e^0 - 1)} = \lambda
  $$

- **Second Moment $\mathbb{E}[X^2]$:**

  Differentiating $M_X'(t)$ using the product rule:

  $$
  M_X''(t) = \frac{d}{dt}\left( \lambda e^t e^{\lambda(e^t - 1)} \right) = \lambda e^t e^{\lambda(e^t - 1)} + (\lambda e^t)^2 e^{\lambda(e^t - 1)}
  $$

  Evaluating at $t = 0$:

  $$
  \mathbb{E}[X^2] = M_X''(0) = \lambda e^0 e^0 + (\lambda e^0)^2 e^0 = \lambda + \lambda^2
  $$

**3. Variance:**

$$
\text{Var}(X) = \mathbb{E}[X^2] - (\mathbb{E}[X])^2 = (\lambda + \lambda^2) - \lambda^2 = \lambda
$$

Thus, $\mathbb{E}[X] = \lambda$ and $\text{Var}(X) = \lambda$.

$\blacksquare$
:::
