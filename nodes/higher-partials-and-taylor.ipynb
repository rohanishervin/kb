# Higher Partial Derivatives and Taylor's Theorem in Several Variables

**Prerequisites:** [The Chain Rule in Several Variables](multivariable-chain-rule.ipynb) · [Taylor Polynomials and Taylor's Theorem](taylor-polynomials.ipynb#thm-lagrange-remainder) · [Local Extrema and the Mean Value Theorem](mean-value-theorem.ipynb#thm-mvt)

## 1\. Higher Partial Derivatives

::: {#def-higher-partials}
## Higher Partial Derivatives

If $\frac{\partial f}{\partial x_i}$ exists near $a$, its partial derivative $\frac{\partial^2f}{\partial x_j\partial x_i}(a) = \frac{\partial}{\partial x_j}\big(\frac{\partial f}{\partial x_i}\big)(a)$ (also written $f_{x_ix_j}(a)$, $D_{ij}f(a)$) is a **second-order partial derivative**; $\frac{\partial^2 f}{\partial x_i^2}$ means $\frac{\partial^2f}{\partial x_i\partial x_i}$. Third and higher orders are defined in the same way. $f$ is **of class $C^k$** on an open set if all partial derivatives of order $\le k$ exist and are continuous there.
:::

::: {#thm-schwarz}
## Symmetry of Mixed Partials (Schwarz–Clairaut)

If $\frac{\partial^2f}{\partial x_j\partial x_i}$ and $\frac{\partial^2f}{\partial x_i\partial x_j}$ both exist in a ball about $a$ and are continuous at $a$, they are equal at $a$. Consequently, for $f$ of class $C^k$ a partial derivative of order $\le k$ depends only on how many times one differentiates with respect to each variable, not on the order.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-schwarz
Only $x_i, x_j$ vary, so take $n = 2$, variables $(x, y)$, $a = (a, b)$. For small $h, k \ne 0$ let
$$Q(h, k) = f(a + h, b + k) - f(a + h, b) - f(a, b + k) + f(a, b).$$
With $u(x) = f(x, b + k) - f(x, b)$ we have $Q = u(a + h) - u(a)$; by the MVT, $Q = h\,u'(\xi) = h\big[f_x(\xi, b + k) - f_x(\xi, b)\big]$ for some $\xi$ between $a$ and $a + h$, and by the MVT again (in $y$, for $f_x$), $Q = hk\,f_{xy}(\xi, \eta)$ with $\eta$ between $b$ and $b + k$. Starting instead from $v(y) = f(a + h, y) - f(a, y)$, $Q = v(b + k) - v(b)$ gives in the same way $Q = kh\,f_{yx}(\xi', \eta')$. Hence $f_{xy}(\xi, \eta) = f_{yx}(\xi', \eta')$, both points lying in the rectangle with corners $(a, b)$, $(a + h, b + k)$. Let $(h, k) \to 0$ and use continuity at $(a, b)$. Repeated transpositions of adjacent differentiations give the general statement. $\blacksquare$
:::

::: {#exm-mixed-partials}
## Examples

1. $f = xe^y - y^2x^3$: $f_x = e^y - 3x^2y^2$, $f_y = xe^y - 2yx^3$, $f_{xx} = -6xy^2$, $f_{xy} = f_{yx} = e^y - 6x^2y$, $f_{yy} = xe^y - 2x^3$.
2. $f = ye^x$: $\frac{\partial^{k+l}f}{\partial x^k\partial y^l} = 0$ for $l \ge 2$, $\frac{\partial^kf}{\partial x^k} = ye^x$, $\frac{\partial^{k+1}f}{\partial x^k\partial y} = e^x$, in any order of differentiation.
3. Continuity matters: $f = \frac{xy(x^2 - y^2)}{x^2 + y^2}$, $f(0) = 0$, has $f_x(0, y) = -y$ and $f_y(x, 0) = x$, so $\frac{\partial}{\partial y}f_x(0, 0) = -1$ but $\frac{\partial}{\partial x}f_y(0, 0) = 1$.
4. For $C^2$ $z = f(x, y)$ in polar coordinates, the chain rule (twice) and @thm-schwarz give the **Laplacian**
$$z_{xx} + z_{yy} = z_{rr} + \frac1rz_r + \frac1{r^2}z_{\theta\theta} \qquad (r > 0).$$
:::

## 2\. Taylor Polynomials

Let $f$ be $C^{p}$ on a ball $B$ about $a$ and $h$ such that $a + h \in B$. Restrict $f$ to the segment: $\varphi(t) = f(a + th)$, $0 \le t \le 1$. Write $(h\cdot\partial)^kf(x) = \sum_{i_1, \dots, i_k = 1}^n\frac{\partial^kf}{\partial x_{i_1}\cdots\partial x_{i_k}}(x)\,h_{i_1}\cdots h_{i_k}$.

::: {#lem-derivatives-along-segment}
## Derivatives Along a Segment

For $k \le p$, $\varphi^{(k)}(t) = (h\cdot\partial)^kf(a + th)$. For example $\varphi'(t) = \sum_i\partial_if(a + th)h_i$ and $\varphi''(t) = \sum_{i,j}\partial_j\partial_if(a + th)h_ih_j$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-derivatives-along-segment
Induction on $k$. For $k < p$ each $\partial_{i_k}\cdots\partial_{i_1}f$ is $C^1$, hence [differentiable](differentiability-in-Rn.ipynb#thm-c1-differentiable), and by the chain rule $\frac{d}{dt}\partial_{i_k}\cdots\partial_{i_1}f(a + th) = \sum_{i_{k+1}}\partial_{i_{k+1}}\partial_{i_k}\cdots\partial_{i_1}f(a + th)\,h_{i_{k+1}}$; multiply by $h_{i_1}\cdots h_{i_k}$ and sum. $\blacksquare$
:::

::: {#def-taylor-several}
## Taylor Polynomial

The **Taylor polynomial of degree $p$** of $f$ at $a$ is
$$T_p(a + h) = \sum_{k=0}^p\frac{1}{k!}(h\cdot\partial)^kf(a),$$
a polynomial in $h_i = x_i - a_i$. By @thm-schwarz the $n^k$ terms of $(h\cdot\partial)^kf$ group as in the multinomial expansion of $(h_1 + \dots + h_n)^k$; for two variables, with $h = (x - a, y - b)$,
$$\frac{1}{k!}(h\cdot\partial)^kf(a, b) = \frac{1}{k!}\sum_{i=0}^k\binom{k}{i}\frac{\partial^kf}{\partial x^i\partial y^{k-i}}(a, b)\,(x - a)^i(y - b)^{k-i} .$$
The term of degree $2$ is $\frac12\big[f_{xx}(x - a)^2 + 2f_{xy}(x - a)(y - b) + f_{yy}(y - b)^2\big]$, the degree-1 part is the [linear approximation](differentiability-in-Rn.ipynb#def-differentiable-Rn).
:::

::: {#thm-taylor-several}
## Taylor's Theorem in Several Variables

1. **(Lagrange remainder)** If $f$ is $C^{p+1}$ on $B$, then for some $c \in (0, 1)$
$$f(a + h) = T_p(a + h) + \frac{1}{(p+1)!}(h\cdot\partial)^{p+1}f(a + ch).$$
2. **(Peano remainder)** If $f$ is $C^p$ on $B$, then $\dfrac{f(a + h) - T_p(a + h)}{|h|^p} \to 0$ as $h \to 0$.
3. **(Uniqueness)** $T_p$ is the only polynomial $P$ of degree $\le p$ in $h$ with $f(a + h) - P(h) = o(|h|^p)$; in particular all partial derivatives of $T_p$ of order $\le p$ at $a$ equal those of $f$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-taylor-several
(1) Apply the [one-variable Lagrange form](taylor-polynomials.ipynb#thm-lagrange-remainder) to $\varphi$ on $[0, 1]$ and use @lem-derivatives-along-segment: $\varphi(1) = \sum_{k\le p}\frac{\varphi^{(k)}(0)}{k!} + \frac{\varphi^{(p+1)}(c)}{(p+1)!}$.
(2) By (1) with $p - 1$, $f(a + h) - T_p(a + h) = \frac{1}{p!}\big[(h\cdot\partial)^pf(a + ch) - (h\cdot\partial)^pf(a)\big]$, whose absolute value is at most $\frac{|h|^p}{p!}\sum_{i_1,\dots,i_p}\big|\partial_{i_p}\cdots\partial_{i_1}f(a + ch) - \partial_{i_p}\cdots\partial_{i_1}f(a)\big|$ (since $|h_i| \le |h|$); the sum tends to $0$ by continuity.
(3) Let $Q = P - T_p$, so $Q(h) = o(|h|^p)$ by (2). Write $Q = \sum_{k\le p}Q_k$ with $Q_k$ homogeneous of degree $k$. For a unit vector $u$, $Q(tu) = \sum_kt^kQ_k(u)$ is a polynomial in $t$ that is $o(t^p)$, so all its coefficients vanish ([one-variable uniqueness](taylor-polynomials.ipynb#cor-taylor-unique)): $Q_k(u) = 0$ for all unit $u$, hence $Q_k(h) = |h|^kQ_k(h/|h|) = 0$. So $P = T_p$. $\blacksquare$
:::

By uniqueness, a polynomial is its own Taylor polynomial: re-expanding it in powers of $x_i - a_i$ gives all its Taylor polynomials at $a$ at once.

::: {#exm-taylor-cube-root}
## A Second-Order Taylor Polynomial

For $f = (x^2 - y)^{1/3}$ at $(3, 1)$: $f_x = \frac12$, $f_y = -\frac1{12}$ and
$$f_{xx} = \tfrac23(x^2 - y)^{-2/3} - \tfrac{8x^2}{9}(x^2 - y)^{-5/3} = -\tfrac1{12},\quad f_{xy} = \tfrac{4x}{9}(x^2 - y)^{-5/3} = \tfrac1{24},\quad f_{yy} = -\tfrac29(x^2 - y)^{-5/3} = -\tfrac1{144},$$
so $T_2 = 2 + \frac12(x - 3) - \frac1{12}(y - 1) - \frac1{24}(x - 3)^2 + \frac1{24}(x - 3)(y - 1) - \frac1{288}(y - 1)^2$.
:::

::: {#exm-taylor-error-bound}
## Bounding the Error of a Linear Approximation

Estimate $\sqrt[3]{3.1^2 - 0.9}$ by $T_1$ at $(3, 1)$, $h = (0.1, -0.1)$: $T_1 = 2 + \frac1{20} + \frac1{120} = 2.05833$. By (1) with $p = 1$ the error is $\frac12\big[f_{xx}h_1^2 + 2f_{xy}h_1h_2 + f_{yy}h_2^2\big]$ at a point $(\xi, \eta)$ of the segment. There $x^2 - y \ge 8$ and $\xi^2 \le 10$, so $|f_{xx}| \le \frac16 + \frac89\cdot\frac{10}{32} = \frac49$, $|f_{xy}| \le \frac49\cdot\frac{3.1}{32} < \frac{4}{90}$, $|f_{yy}| \le \frac{2}{9\cdot32} = \frac1{144}$, and
$$|\text{error}| \le \frac{1}{200}\Big(\frac49 + \frac{8}{90} + \frac1{144}\Big) < 0.0028 .$$
The true value is $2.05750$, so the actual error is about $0.0008$.
:::

::: {#exm-taylor-polynomial-recentred}
## Re-expanding a Polynomial

For $f = x^2y^3 - xz$ at $(1, 1, 0)$, substitute $x = 1 + u$, $y = 1 + v$ and expand:
$$f = 1 + (2u + 3v - z) + (u^2 + 6uv + 3v^2 - uz) + (3u^2v + 6uv^2 + v^3) + (3u^2v^2 + 2uv^3) + u^2v^3 .$$
The sum of the homogeneous parts of degree $\le p$ is $T_p$ at $(1, 1, 0)$.
:::

```python

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

x, y, r, th = sp.symbols("x y r theta", positive=True)
g = x * y * (x**2 - y**2) / (x**2 + y**2)
fx0 = sp.simplify(sp.diff(g, x).subs(x, 0))
fy0 = sp.simplify(sp.diff(g, y).subs(y, 0))
print("f_x(0, y) =", fx0, "| f_y(x, 0) =", fy0, "=> f_xy(0) = -1, f_yx(0) = 1")
test = (x**3 * y - sp.exp(x) * sp.sin(y))
lhs = (sp.diff(test, x, 2) + sp.diff(test, y, 2)).subs({x: r * sp.cos(th), y: r * sp.sin(th)})
tp = test.subs({x: r * sp.cos(th), y: r * sp.sin(th)})
rhs = sp.diff(tp, r, 2) + sp.diff(tp, r) / r + sp.diff(tp, th, 2) / r**2
print("Laplacian in polar coordinates checked on a test function:", sp.simplify(lhs - rhs) == 0)
f = sp.cbrt(x**2 - y)
T2 = sum(sp.Rational(1, sp.factorial(i + j)) * sp.binomial(i + j, i) * sp.diff(f, x, i, y, j).subs({x: 3, y: 1}) * (x - 3)**i * (y - 1)**j
         for i in range(3) for j in range(3) if i + j <= 2)
print("T2 =", sp.expand(T2))
u, v, w = sp.symbols("u v w")
print("x^2y^3 - xz re-expanded:", sp.expand(((1 + u)**2 * (1 + v)**3 - (1 + u) * w)))
print("cbrt(3.1^2 - 0.9) =", float(sp.cbrt(sp.Rational(31, 10)**2 - sp.Rational(9, 10))), "| T1 =", 2 + 1 / 20 + 1 / 120)

fn = lambda X, Y: np.cbrt(X**2 - Y)
T1n = lambda X, Y: 2 + (X - 3) / 2 - (Y - 1) / 12
T2n = lambda X, Y: T1n(X, Y) - (X - 3)**2 / 24 + (X - 3) * (Y - 1) / 24 - (Y - 1)**2 / 288
hs = np.logspace(-3, 0, 40); d = np.array([0.6, -0.8])
e1 = [abs(fn(3 + s * d[0], 1 + s * d[1]) - T1n(3 + s * d[0], 1 + s * d[1])) for s in hs]
e2 = [abs(fn(3 + s * d[0], 1 + s * d[1]) - T2n(3 + s * d[0], 1 + s * d[1])) for s in hs]
fig, ax = plt.subplots(figsize=(6.4, 4.4))
ax.loglog(hs, e1, color=BLUE, lw=2, label=r"$|f - T_1|$")
ax.loglog(hs, e2, color=ORANGE, lw=2, label=r"$|f - T_2|$")
ax.loglog(hs, 0.03 * hs**2, color=BLUE, ls=":", label=r"slope 2"); ax.loglog(hs, 0.003 * hs**3, color=ORANGE, ls=":", label=r"slope 3")
ax.set_xlabel(r"$|h|$"); ax.legend(fontsize=9); ax.set_title(r"Taylor errors of $\sqrt[3]{x^2 - y}$ at $(3, 1)$")
plt.tight_layout(); plt.show()
```
