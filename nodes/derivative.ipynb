# The Derivative

**Prerequisites:** [Limits of Functions](limit.ipynb) · [Continuous Functions](function-continuity.ipynb)

A point $a \in S$ is **interior** to $S$ if $(a - \delta, a + \delta) \subseteq S$ for some $\delta > 0$.

## 1\. Definition

::: {#def-derivative}
## Derivative

Let $a$ be interior to $S$ and $f : S \to \mathbb{R}$. $f$ is **differentiable at** $a$ if
$$f'(a) = \lim_{h \to 0} \frac{f(a + h) - f(a)}{h} = \lim_{x \to a}\frac{f(x) - f(a)}{x - a}$$
exists (as a real number); $f'(a)$ is the **derivative** of $f$ at $a$. $f$ is differentiable if it is differentiable at every point of its (open) domain. The **one-sided derivatives** $f'_{\pm}(a)$ are the limits for $h \to 0^\pm$; they also make sense at an endpoint of an interval.
:::

$\frac{f(a+h) - f(a)}{h}$ is the slope of the secant through $(a, f(a))$ and $(a + h, f(a + h))$, or the average rate of change of $f$ over $[a, a + h]$; the derivative is its limit, the **instantaneous rate of change** (for $f(t)$ the position of a moving point, $f'(t)$ is its velocity). The line $y = f(a) + f'(a)(x - a)$ is the **tangent line** to the graph at $(a, f(a))$.

::: {#exm-derivatives}
## First Derivatives

1. $f(x) = Ax + B$: the difference quotient is $A$, so $f' = A$; constants have derivative $0$.
2. $f(x) = cx^n$, $n \in \mathbb{N}$: $\frac{x^n - x_0^n}{x - x_0} = x^{n-1} + x^{n-2}x_0 + \dots + x_0^{n-1} \to n x_0^{n-1}$, so $f'(x_0) = c\,n x_0^{n-1}$.
3. $f(x) = |x|$: $f'(x) = 1$ for $x > 0$, $-1$ for $x < 0$. At $0$, $\frac{|h|}{h} = \pm 1$, so $f'_+(0) = 1$, $f'_-(0) = -1$ and $f$ is **not** differentiable at $0$.
4. *Tangent to an ellipse.* The point $(2, -\frac{3\sqrt3}{2})$ lies on $\frac{x^2}{16} + \frac{y^2}{9} = 1$, on the lower half $y = f(x) = -\frac34\sqrt{16 - x^2}$. Rationalizing,
$$\frac{f(x) - f(2)}{x - 2} = -\frac34\cdot\frac{\sqrt{16 - x^2} - \sqrt{12}}{x - 2} = \frac34 \cdot \frac{x + 2}{\sqrt{16 - x^2} + \sqrt{12}} \to \frac34 \cdot \frac{4}{2\sqrt{12}} = \frac{\sqrt3}{4},$$
so the tangent is $y + \frac{3\sqrt3}{2} = \frac{\sqrt3}{4}(x - 2)$.
:::

## 2\. Tangency and Linear Approximation

::: {#def-tangency}
## Tangent Functions

$f, g$ defined near $x_0$ are **tangent** at $x_0$ if $f(x_0) = g(x_0)$ and $\displaystyle\lim_{x \to x_0}\frac{f(x) - g(x)}{x - x_0} = 0$: the vertical distance between the graphs tends to $0$ faster than $x - x_0$.
:::

For example $f(x) = |x| + x^2$ and $g(x) = |x|$ are tangent at $0$ ($\frac{f - g}{x} = x \to 0$), although neither is differentiable there.

::: {#thm-linear-approximation-char}
## Differentiability Means a Tangent Line Exists

$f$ is differentiable at $x_0$ if and only if some affine function $g(x) = f(x_0) + m(x - x_0)$ is tangent to $f$ at $x_0$. Then $m = f'(x_0)$; so the tangent line is unique. $g$ is the **linear approximation** of $f$ at $x_0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-linear-approximation-char
For $x \ne x_0$,
$$\frac{f(x) - g(x)}{x - x_0} = \frac{f(x) - f(x_0)}{x - x_0} - m,$$
which tends to $0$ if and only if the difference quotient tends to $m$. $\blacksquare$
:::

::: {#thm-caratheodory}
## Differentiability with an Error Term

$f$ is differentiable at $a$ if and only if there are $m \in \mathbb{R}$ and a function $\varphi$, defined for small $h \ne 0$, with $\varphi(h) \to 0$ as $h \to 0$ and
$$f(a + h) = f(a) + mh + \varphi(h)\,h .$$
Then $m = f'(a)$. Writing $\Delta x = h$ and $\Delta y = f(a + h) - f(a)$: $\ \Delta y = f'(a)\Delta x + \varphi(\Delta x)\Delta x$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-caratheodory
The identity defines $\varphi(h) = \frac{f(a+h) - f(a)}{h} - m$, which tends to $0$ if and only if $f'(a) = m$. $\blacksquare$
:::

::: {#thm-differentiable-continuous}
## Differentiable Implies Continuous

If $f$ is differentiable at $a$, then $f$ is continuous at $a$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-differentiable-continuous
$f(a + h) - f(a) = (f'(a) + \varphi(h))h \to 0$ as $h \to 0$. $\blacksquare$
:::

The converse fails: $|x|$ is continuous but not differentiable at $0$.

::: {#exm-symmetric-quotient}
## Symmetric Difference Quotient

If $f'(t)$ exists, then $\displaystyle\lim_{h \to 0}\frac{f(t + h) - f(t - h)}{2h} = f'(t)$, because the quotient is the average of $\frac{f(t+h) - f(t)}{h}$ and $\frac{f(t - h) - f(t)}{-h}$. The converse fails: for $|x|$ at $0$ the symmetric quotient is $0$ for all $h$.
:::

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

f = lambda x: 0.5 * x**3 - x + 1.5
fp = lambda x: 1.5 * x**2 - 1
a = 1.2
x = np.linspace(-0.3, 2.4, 300)
fig, ax = plt.subplots(figsize=(8, 4.8))
ax.plot(x, f(x), color=INK, lw=2, label="$y = f(x)$")
for h, col in zip([1.0, 0.6, 0.3], [VIOLET, ORANGE, AQUA]):
    s = (f(a + h) - f(a)) / h
    ax.plot(x, f(a) + s * (x - a), color=col, lw=1.2, ls="--", label=f"secant, $h = {h}$, slope {s:.2f}")
    ax.plot(a + h, f(a + h), "o", color=col, ms=6)
ax.plot(x, f(a) + fp(a) * (x - a), color=BLUE, lw=2, label=f"tangent, slope $f'(a) = {fp(a):.2f}$")
ax.plot(a, f(a), "o", color=YELLOW, ms=9, mec=INK)
ax.set_ylim(-0.5, 5.2); ax.legend(fontsize=9, loc="upper left")
ax.set_title(r"secant slopes $\frac{f(a+h)-f(a)}{h}$ tend to the tangent slope $f'(a)$")
plt.tight_layout(); plt.show()
```
