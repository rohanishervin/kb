# Green's Theorem

**Prerequisites:** [Vector Fields, Line Integrals and Conservative Fields](line-integrals-and-conservative-fields.ipynb) · [Flux, Divergence and Curl](flux-divergence-and-curl.ipynb#def-div-curl) · [Iterated Integrals and Fubini's Theorem](iterated-integrals-and-fubini.ipynb#cor-simple-regions) · [Multiple Integrals](multiple-integrals.ipynb#thm-mean-value-multiple)

Green's theorem expresses a double integral over a plane region as a line integral over its boundary; it is the two-dimensional version of the fundamental theorem of calculus, and the model for the theorems of Stokes and Gauss.

## 1\. The Theorem

Let $D$ be a bounded region in the plane whose boundary $\partial D$ consists of finitely many piecewise smooth simple closed curves. The **positive orientation** of $\partial D$ keeps $D$ on the left: at each non-corner point, the unit tangent $T$ turned by $+\frac\pi2$ points into $D$. For a disk this is counterclockwise; for an annulus, the outer circle is counterclockwise and the inner one clockwise.

::: {#thm-green}
## Green's Theorem

Let $F = (F_1, F_2)$ be $C^1$ on an open set containing $D\cup\partial D$, with $\partial D$ positively oriented. Then
$$\oint_{\partial D}F_1\,dx + F_2\,dy = \iint_D\Big(\frac{\partial F_2}{\partial x} - \frac{\partial F_1}{\partial y}\Big)dA .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-green
*Regions that are both $y$-simple and $x$-simple.* Let $D = \{a \le x \le b,\ \alpha(x) \le y \le \beta(x)\}$ with $\alpha \le \beta$ piecewise $C^1$. By [Fubini](iterated-integrals-and-fubini.ipynb#cor-simple-regions) and the FTC in $y$,
$$-\iint_D\frac{\partial F_1}{\partial y}\,dA = -\int_a^b\big[F_1(x, \beta(x)) - F_1(x, \alpha(x))\big]dx .$$
On the other hand $\partial D$ consists of the lower graph $x \mapsto (x, \alpha(x))$ traversed left to right, contributing $\int_a^bF_1(x, \alpha(x))\,dx$ to $\oint F_1\,dx$; the upper graph traversed right to left, contributing $-\int_a^bF_1(x, \beta(x))\,dx$; and possibly vertical segments, on which $dx = 0$. So $\oint_{\partial D}F_1\,dx = -\iint_D\frac{\partial F_1}{\partial y}\,dA$. If $D$ is also $x$-simple, the same argument with the roles of $x, y$ exchanged gives $\oint_{\partial D}F_2\,dy = \iint_D\frac{\partial F_2}{\partial x}\,dA$ (now the right graph is traversed upward). Adding proves the theorem for such regions: rectangles, disks, triangles, etc.
*Unions.* If $D$ is cut by finitely many arcs into pieces $D_1, \dots, D_k$ of this kind, add the formulas for the pieces. Each cutting arc belongs to the boundaries of two pieces, with opposite orientations, so the line integrals over it cancel, and what remains is the integral over $\partial D$; the double integrals add up by additivity. This covers annuli and most regions met in practice. The general case follows by approximating $D$ with such regions; this step is not carried out here. $\blacksquare$
:::

## 2\. Applications

::: {#cor-green-area}
## Area as a Line Integral

$$\operatorname{area}(D) = \oint_{\partial D}x\,dy = -\oint_{\partial D}y\,dx = \frac12\oint_{\partial D}(-y\,dx + x\,dy).$$
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-green-area
Apply @thm-green to $F = (0, x)$, $(-y, 0)$ and $\frac12(-y, x)$, for which $\frac{\partial F_2}{\partial x} - \frac{\partial F_1}{\partial y} = 1$. $\blacksquare$
:::

For the ellipse $(a\cos t, b\sin t)$: $\frac12\int_0^{2\pi}(ab\sin^2t + ab\cos^2t)\,dt = \pi ab$. This also explains the [circulation computations](line-integrals-and-conservative-fields.ipynb#exm-circulation): $\oint(-y\,dx + x\,dy)$ is twice the enclosed area.

::: {#thm-curl-free-rectangle}
## Curl-Free Fields on a Rectangle Are Conservative

Let $F = (F_1, F_2)$ be $C^1$ on an open rectangle $U$ with $\frac{\partial F_1}{\partial y} = \frac{\partial F_2}{\partial x}$. Then $F$ is a gradient field on $U$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-curl-free-rectangle
Fix $P = (x_0, y_0) \in U$. For $Q = (x, y) \in U$ let $\alpha_Q$ go horizontally from $P$ to $(x, y_0)$ and then vertically to $Q$, and $\beta_Q$ go vertically to $(x_0, y)$ and then horizontally to $Q$; both stay in $U$. They bound a rectangle (possibly degenerate), so by @thm-green (with $\frac{\partial F_2}{\partial x} - \frac{\partial F_1}{\partial y} = 0$) $\int_{\alpha_Q}F\cdot d\mathbf r = \int_{\beta_Q}F\cdot d\mathbf r =: f(Q)$. From $\alpha_Q$, $f(x, y) = \int_{x_0}^xF_1(s, y_0)\,ds + \int_{y_0}^yF_2(x, s)\,ds$, so $\frac{\partial f}{\partial y} = F_2$ by the FTC; from $\beta_Q$, $f(x, y) = \int_{y_0}^yF_2(x_0, s)\,ds + \int_{x_0}^xF_1(s, y)\,ds$, so $\frac{\partial f}{\partial x} = F_1$. $\blacksquare$
:::

The same conclusion holds on any open path-connected set $U$ that is **simply connected** (the interior of every simple closed curve in $U$ lies in $U$): Green's theorem shows that the integral over every piecewise smooth simple closed curve in $U$ is $0$, and one shows that this suffices for [path independence](line-integrals-and-conservative-fields.ipynb#thm-path-independence). The [angle field](line-integrals-and-conservative-fields.ipynb#exm-angle-field) shows that the hole in $\mathbb{R}^2\setminus\{0\}$ cannot be ignored.

::: {#cor-deformation}
## Deforming the Curve

Let $\gamma, \gamma_1, \dots, \gamma_k$ be piecewise smooth simple closed curves, all counterclockwise, the $\gamma_i$ disjoint, lying inside $\gamma$ and outside each other, and let $F$ be $C^1$ with $\frac{\partial F_1}{\partial y} = \frac{\partial F_2}{\partial x}$ on an open set containing the closed region $D$ between them. Then
$$\oint_\gamma F\cdot d\mathbf r = \sum_{i=1}^k\oint_{\gamma_i}F\cdot d\mathbf r .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-deformation
The positively oriented boundary of $D$ is $\gamma$ together with the $\gamma_i$ traversed clockwise. By @thm-green, $\oint_\gamma F\cdot d\mathbf r - \sum_i\oint_{\gamma_i}F\cdot d\mathbf r = \iint_D0\,dA = 0$. $\blacksquare$
:::

For the angle field $\frac{-y\,dx + x\,dy}{x^2 + y^2}$ this shows that its integral over any counterclockwise simple closed curve around $0$ equals its integral over a small circle, $2\pi$; over a curve not enclosing $0$ it is $0$. The **winding number** of a closed curve $\gamma$ in $\mathbb{R}^2\setminus\{p\}$ around $p$ is
$$I_\gamma(p) = \frac{1}{2\pi}\oint_\gamma\frac{-(y - p_2)\,dx + (x - p_1)\,dy}{(x - p_1)^2 + (y - p_2)^2},$$
the net number of counterclockwise turns of $\gamma$ around $p$.

## 3\. Divergence and Curl in the Plane

At a non-corner point of the positively oriented $\partial D$, with unit tangent $T = (\frac{dx}{ds}, \frac{dy}{ds})$, the **outward unit normal** is $\mathbf n = (\frac{dy}{ds}, -\frac{dx}{ds})$ ($T$ turned by $-\frac\pi2$).

::: {#cor-green-divergence}
## Green's Theorem in Divergence Form

For a $C^1$ field $G = (G_1, G_2)$ on an open set containing $D\cup\partial D$,
$$\oint_{\partial D}G\cdot\mathbf n\,ds = \iint_D\operatorname{div}G\,dA :$$
the outward flux across the boundary equals the integral of the divergence.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-green-divergence
Let $F = (-G_2, G_1)$. Then $F\cdot T = -G_2\frac{dx}{ds} + G_1\frac{dy}{ds} = G\cdot\mathbf n$ and $\frac{\partial F_2}{\partial x} - \frac{\partial F_1}{\partial y} = \frac{\partial G_1}{\partial x} + \frac{\partial G_2}{\partial y}$. Apply @thm-green, using $\oint F\cdot d\mathbf r = \oint F\cdot T\,ds$. $\blacksquare$
:::

Let $D_\rho$ be the disk of radius $\rho$ about $P$. By the [mean value theorem for integrals](multiple-integrals.ipynb#thm-mean-value-multiple), $\iint_{D_\rho}\operatorname{div}G\,dA = \pi\rho^2\operatorname{div}G(Q_\rho)$ for some $Q_\rho \in D_\rho$, and continuity gives
$$\operatorname{div}G(P) = \lim_{\rho\to0}\frac{1}{\pi\rho^2}\oint_{\partial D_\rho}G\cdot\mathbf n\,ds, \qquad \Big(\frac{\partial F_2}{\partial x} - \frac{\partial F_1}{\partial y}\Big)(P) = \lim_{\rho\to0}\frac{1}{\pi\rho^2}\oint_{\partial D_\rho}F\cdot d\mathbf r :$$
the divergence is the outward flux per unit area and the (scalar) curl the circulation per unit area.

::: {#exm-area-perimeter}
## Area, Perimeter and Radius

Let $D$ have area $A$ and boundary length $L$. For any point $C$, let $r_C = \max_{q\in\partial D}|q - C|$. Then $A \le \frac12r_CL$, with equality for a disk and its centre. Indeed, $G(x) = x - C$ has $\operatorname{div}G = 2$ and $G\cdot\mathbf n \le |G| \le r_C$ on $\partial D$, so by @cor-green-divergence $2A = \oint_{\partial D}G\cdot\mathbf n\,ds \le r_CL$. (For a convex region the smallest $r_C$ is attained at a point called its centre, and $r_C$ is its radius.)
:::

```python

import sympy as sp
from scipy.integrate import quad
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

x, y, t, r, th = sp.symbols("x y t r theta", real=True)
F1, F2 = x**2 * y + sp.sin(x), x * y**2 + x**3
lhs = sp.integrate((F1 * sp.diff(sp.cos(t), t) + F2 * sp.diff(sp.sin(t), t)).subs({x: sp.cos(t), y: sp.sin(t)}), (t, 0, 2 * sp.pi))
rhs = sp.integrate(((sp.diff(F2, x) - sp.diff(F1, y)).subs({x: r * sp.cos(th), y: r * sp.sin(th)}) * r), (r, 0, 1), (th, 0, 2 * sp.pi))
print("Green on the unit disk:", sp.simplify(lhs), "=", sp.simplify(rhs))

def winding(px, py, gx, gy, dgx, dgy):
    f = lambda s: (-(gy(s) - py) * dgx(s) + (gx(s) - px) * dgy(s)) / ((gx(s) - px)**2 + (gy(s) - py)**2)
    return quad(f, 0, 2 * np.pi, limit=200)[0] / (2 * np.pi)
rr = lambda s: 1 + 2 * np.cos(s)
gx, gy = lambda s: rr(s) * np.cos(s), lambda s: rr(s) * np.sin(s)
dgx = lambda s: -2 * np.sin(s) * np.cos(s) - rr(s) * np.sin(s)
dgy = lambda s: -2 * np.sin(s) * np.sin(s) + rr(s) * np.cos(s)
pts = [(0.5, 0.0), (2.0, 0.0), (-1.0, 0.0)]
print("winding numbers of the limacon r = 1 + 2cos(theta):", [round(winding(px, py, gx, gy, dgx, dgy), 6) for px, py in pts])

fig, ax = plt.subplots(figsize=(6, 4.8))
s = np.linspace(0, 2 * np.pi, 600)
ax.plot(gx(s), gy(s), color=BLUE, lw=2)
for k in [40, 200, 330, 520]:
    ax.annotate("", xy=(gx(s[k + 5]), gy(s[k + 5])), xytext=(gx(s[k]), gy(s[k])), arrowprops=dict(arrowstyle="-|>", color=BLUE))
for (px, py), w in zip(pts, [2, 1, 0]):
    ax.plot(px, py, "o", color=ORANGE); ax.text(px + 0.06, py + 0.1, f"$I = {w}$", color=ORANGE)
ax.set_aspect("equal"); ax.set_title(r"Winding numbers of the limaçon $r = 1 + 2\cos\theta$")
plt.tight_layout(); plt.show()
```
