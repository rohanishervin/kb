# Linear Independence

**Prerequisites:** [Subspaces, Sums, and Span](subspaces.ipynb)

A list of vectors is **linearly independent** when none of them is redundant: no vector can be built from the others. Independence is what makes coordinates unique and is the key ingredient of [bases and dimension](basis-and-dimension.ipynb).

Throughout, $V$ is a [vector space](vector-spaces.ipynb) over a field $K$, and [linear combinations](subspaces.ipynb#def-linear-combination) and [spans](subspaces.ipynb#def-span) are as defined earlier.

## 1\. Definition

::: {#def-linear-independence}
## Linear Independence

A subset $S \subseteq V$ is **linearly independent** if the only way to write $0$ as a linear combination of distinct elements of $S$ is with all coefficients zero: whenever

$$c_1 v_1 + c_2 v_2 + \dots + c_m v_m = 0$$

for distinct $v_1, \dots, v_m \in S$ and $c_1, \dots, c_m \in K$, we have $c_1 = c_2 = \dots = c_m = 0$.

Otherwise $S$ is **linearly dependent**: some combination with not all coefficients zero (a **non-trivial relation**) equals $0$. The same words are used for a finite list $v_1, \dots, v_m$; a list with a repeated vector is dependent ($v - v = 0$).
:::

The next proposition turns "a non-trivial relation exists" into the more intuitive "some vector is redundant".

::: {#prp-dependence-redundancy}
## Dependence Means Redundancy

A set $S \subseteq V$ is linearly dependent if and only if some $v \in S$ lies in the span of the other elements, $v \in \operatorname{span}(S \setminus \{v\})$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-dependence-redundancy
($\Rightarrow$) Take a non-trivial relation $c_1 v_1 + \dots + c_m v_m = 0$ with distinct $v_i \in S$, and choose $j$ with $c_j \ne 0$. Dividing by $c_j$,
$$v_j = -\sum_{i \ne j} \frac{c_i}{c_j}\, v_i \in \operatorname{span}(S \setminus \{v_j\}).$$

($\Leftarrow$) If $v = c_1 w_1 + \dots + c_k w_k$ with distinct $w_i \in S \setminus \{v\}$, then
$$1 \cdot v - c_1 w_1 - \dots - c_k w_k = 0$$
is a relation among distinct elements of $S$ whose coefficient of $v$ is $1 \ne 0$. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Each Independent Vector Adds a New Direction
Two vectors are independent exactly when neither is a multiple of the other, so they point in different directions. Three vectors in $\mathbb{R}^3$ are independent exactly when they do not lie in a common plane through the origin. Each vector of an independent list reaches outside the span of the previous ones and so adds a genuinely new direction.
:::

## 2\. Basic Properties

::: {#prp-independence-basic}
## Basic Properties

1. A set containing $0$ is linearly dependent.
2. A single vector $v$ is linearly independent if and only if $v \neq 0$.
3. If $S$ is linearly dependent and $S \subseteq T$, then $T$ is linearly dependent.
4. If $S$ is linearly independent and $T \subseteq S$, then $T$ is linearly independent.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-independence-basic
1. $1 \cdot 0 = 0$ is a non-trivial relation.
2. If $v \ne 0$ and $cv = 0$, then $c = 0$ by the [basic rules](vector-spaces.ipynb#prp-vector-space-consequences); together with part 1 this proves the claim.
3. A non-trivial relation among elements of $S$ is also a non-trivial relation among elements of $T$.
4. This is the contrapositive of part 3. $\blacksquare$
:::

The following lemma is the step used again and again to build independent sets one vector at a time.

::: {#lem-extend-independent}
## Adding a Vector Outside the Span

If $S$ is linearly independent and $v \notin \operatorname{span}(S)$, then $S \cup \{v\}$ is linearly independent.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-extend-independent
Suppose $c\,v + c_1 v_1 + \dots + c_m v_m = 0$ with distinct $v_i \in S$. If $c \ne 0$, then $v = -\sum_i (c_i / c)\, v_i \in \operatorname{span}(S)$, which is false. So $c = 0$, and the relation becomes $c_1 v_1 + \dots + c_m v_m = 0$, which forces all $c_i = 0$ because $S$ is independent. $\blacksquare$
:::

## 3\. Examples

::: {#exm-independence-kn}
## Testing Independence in $\mathbb{R}^3$

Are $v_1 = (1, 0, 1)$, $v_2 = (0, 1, 1)$, $v_3 = (1, 1, 3)$ independent? A relation $c_1 v_1 + c_2 v_2 + c_3 v_3 = 0$ means

$$c_1 + c_3 = 0, \qquad c_2 + c_3 = 0, \qquad c_1 + c_2 + 3 c_3 = 0.$$

The first two equations give $c_1 = c_2 = -c_3$, and then the third gives $c_3 = 0$. So only the trivial relation exists, and the vectors are independent.

Replacing $v_3$ by $w = (1, 1, 2)$ changes the answer: $v_1 + v_2 - w = 0$, so $w = v_1 + v_2$ lies in the plane spanned by $v_1$ and $v_2$.
:::

::: {#exm-independence-functions}
## Independence of Functions

In the space $\mathbb{R}^{\mathbb{R}}$ of all functions, $\cos$ and $\sin$ are independent: if $a \cos + b \sin = 0$ as functions, then evaluating at $t = 0$ gives $a = 0$ and evaluating at $t = \pi/2$ gives $b = 0$.

Similarly $1, x, x^2, \dots, x^d$ are independent in $\mathbb{R}^{\mathbb{R}}$: if $a_0 + a_1 x + \dots + a_d x^d$ is the zero function, this polynomial has infinitely many roots, which is only possible when all $a_i = 0$. On the other hand, $\cos^2$, $\sin^2$ and the constant function $1$ are dependent, since $\cos^2 + \sin^2 - 1 = 0$.
:::

The figure shows the two triples of the first example: independent vectors point out of each other's plane, while in the dependent triple the third vector lies in the plane of the first two.


```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, RED, VIOLET = "#2a78d6", "#eb6834", "#1baf7a", "#e34948", "#6250d6"
INK, MUTED = "#0b0b0b", "#52514e"
plt.rcParams.update({"axes.titlesize": 11, "axes.titleweight": "bold", "font.size": 10})

v1, v2 = np.array([1, 0, 1.0]), np.array([0, 1, 1.0])
triples = [(np.array([1, 1, 3.0]), AQUA, "independent: $v_3$ leaves the plane"),
           (np.array([1, 1, 2.0]), RED, "dependent: $w = v_1 + v_2$ lies in it")]

fig = plt.figure(figsize=(11, 4.8))
for k, (v3, c3, title) in enumerate(triples):
    ax = fig.add_subplot(1, 2, k + 1, projection="3d")
    s, t = np.meshgrid(np.linspace(-0.4, 1.4, 10), np.linspace(-0.4, 1.4, 10))
    P = s[..., None] * v1 + t[..., None] * v2                 # the plane span(v1, v2)
    ax.plot_surface(P[..., 0], P[..., 1], P[..., 2], color=VIOLET, alpha=0.12, linewidth=0)
    for v, c, name in [(v1, BLUE, "$v_1$"), (v2, ORANGE, "$v_2$"), (v3, c3, "$v_3$" if k == 0 else "$w$")]:
        ax.quiver(0, 0, 0, *v, color=c, lw=2.2, arrow_length_ratio=0.12)
        ax.text(*(1.08 * v), name, color=INK, fontsize=12)
    if k == 1:   # show w as the diagonal of the parallelogram
        for a, b in [(v1, v1 + v2), (v2, v1 + v2)]:
            ax.plot(*zip(a, b), color=MUTED, ls="--", lw=1)
    ax.set_xlim(0, 1.4); ax.set_ylim(0, 1.4); ax.set_zlim(0, 3)
    ax.set_xlabel("$x$"); ax.set_ylabel("$y$"); ax.set_zlabel("$z$")
    ax.view_init(elev=18, azim=-60)
    ax.set_title(title)
plt.tight_layout(); plt.show()
```
