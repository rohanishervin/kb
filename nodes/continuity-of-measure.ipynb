# Continuity of Measure

**Prerequisites:** [Measures and Measure Spaces](measure.ipynb)

A measure is a function on a σ-algebra, so asking whether it is *continuous* first requires saying what it means for a sequence of **sets** to converge. There are several notions; the simplest, and the one needed here, is **monotone** convergence: sets that grow to their union or shrink to their intersection. Continuity of measure says that the measures then converge too, with one important finiteness condition in the shrinking case. Throughout, $(X, \mathcal{M}, \mu)$ is a [measure space](measure.ipynb#def-measure) and all sets are measurable.

## 1\. Monotone Convergence of Sets

::: {#def-monotone-convergence-of-sets}
## Increasing and Decreasing Sequences of Sets

- $E_1, E_2, \dots$ **increase to** $E$, written $E_i \nearrow E$, if $E_1 \subseteq E_2 \subseteq E_3 \subseteq \cdots$ and $E = \bigcup_{i=1}^\infty E_i$.
- $E_1, E_2, \dots$ **decrease to** $E$, written $E_i \searrow E$, if $E_1 \supseteq E_2 \supseteq E_3 \supseteq \cdots$ and $E = \bigcap_{i=1}^\infty E_i$.
:::

In the first case the sets start inside $E$ and grow towards it; in the second they contain $E$ and shrink onto it. With this notion of convergence, continuity of $\mu$ means: $E_i \nearrow E$ or $E_i \searrow E$ implies $\mu(E_i) \to \mu(E)$.

## 2\. Continuity From Below and From Above

::: {#thm-continuity-of-measure}
## Continuity of Measure

1. (**From below**) If $A_1 \subseteq A_2 \subseteq \cdots$, then
$$\mu\Big(\bigcup_{k=1}^\infty A_k\Big) = \lim_{k \to \infty} \mu(A_k).$$
2. (**From above**) If $B_1 \supseteq B_2 \supseteq \cdots$ and $\mu(B_1) < \infty$, then
$$\mu\Big(\bigcap_{k=1}^\infty B_k\Big) = \lim_{k \to \infty} \mu(B_k).$$
:::

In both cases the limit exists, because $\mu(A_k)$ increases and $\mu(B_k)$ decreases by [monotonicity](measure.ipynb#prp-monotonicity).

::: {.callout-tip collapse="true"}
## Proof of @thm-continuity-of-measure
We use only the axioms and the properties already proved.

**1. From below.** Let $A = \bigcup_k A_k$. As in the proof of [countable monotonicity](measure.ipynb#prp-countable-monotonicity), we cut $A$ into disjoint pieces; since the $A_k$ are nested this is simpler. Put $A_0 = \emptyset$ and
$$G_k = A_k \setminus A_{k-1} \qquad (k \ge 1),$$
the "rings" between consecutive sets. (We need not remove $A_1$ from $A_3$, because $A_1 \subseteq A_2$.) The $G_k$ are pairwise disjoint and their union is $A$, so by countable additivity
$$\mu(A) = \sum_{k=1}^\infty \mu(G_k).$$
If some $\mu(A_k) = \infty$, then $\mu(A) = \infty$ and $\mu(A_j) = \infty$ for all $j \ge k$, so both sides of the claim are $\infty$. Otherwise all $\mu(A_k)$ are finite and, since $A_{k-1} \subseteq A_k$, [excision](measure.ipynb#prp-excision) gives $\mu(G_k) = \mu(A_k) - \mu(A_{k-1})$. An infinite series is by definition the limit of its partial sums, and here the partial sums telescope:
$$\mu(A) = \lim_{n \to \infty} \sum_{k=1}^n \big(\mu(A_k) - \mu(A_{k-1})\big) = \lim_{n \to \infty} \big(\mu(A_n) - \mu(A_0)\big) = \lim_{n \to \infty} \mu(A_n),$$
since $\mu(A_0) = \mu(\emptyset) = 0$.

**2. From above.** Let $B = \bigcap_k B_k$ and $G_k = B_k \setminus B_{k+1}$. These rings are pairwise disjoint, and their union is $B_1 \setminus B$: the "hole" $B$ lies in none of them. In detail, each $G_k \subseteq B_1$ and $G_k \cap B = \emptyset$; conversely, if $x \in B_1 \setminus B$, then $x \notin B_m$ for some $m$, and if $k$ is the last index with $x \in B_k$ (it exists, as $x \in B_1$ and the sets decrease), then $x \in B_k \setminus B_{k+1} = G_k$.

Here the hypothesis $\mu(B_1) < \infty$ is used: all $\mu(B_k)$ and $\mu(B)$ are finite, so excision applies to $B \subseteq B_1$ and to $B_{k+1} \subseteq B_k$. Countable additivity gives
$$\mu(B_1) - \mu(B) = \mu(B_1 \setminus B) = \sum_{k=1}^\infty \big(\mu(B_k) - \mu(B_{k+1})\big) = \lim_{n \to \infty} \big(\mu(B_1) - \mu(B_{n+1})\big).$$
The constant $\mu(B_1)$ comes out of the limit and cancels, leaving $\mu(B) = \lim_n \mu(B_{n+1}) = \lim_n \mu(B_n)$. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Adding Up the Rings
Countable additivity only speaks about *disjoint* sets, while the $A_k$ are nested. The rings $A_k \setminus A_{k-1}$ are the layers added at each step; the measure of the union is the total of all layers, and the first $n$ layers add up to $\mu(A_n)$. Going down, the rings are the layers removed at each step, and $\mu(B_1)$ must be finite so that "what is left" can be computed as "what we started with minus what was removed".
:::

::: {.callout-warning}
## The Finiteness Condition Cannot Be Dropped
Take the [counting measure](measure.ipynb#exm-counting-measure) on $\mathbb{N}$ and $B_k = \{k, k+1, k+2, \dots\}$. The sets decrease, $\mu(B_k) = \infty$ for every $k$, but $\bigcap_k B_k = \emptyset$, so $\mu(\bigcap_k B_k) = 0 \ne \infty = \lim_k \mu(B_k)$. The same happens for length on $\mathbb{R}$ with $B_k = [k, \infty)$. It is enough, however, that $\mu(B_k) < \infty$ for *some* $k$: discard the sets before it.
:::

The figure illustrates both halves for area in the plane. On the left, discs $A_k$ of radius $1 - \frac{1}{k+1}$ increase to the open unit disc, cut into rings $G_k$; their areas increase to $\pi$. On the right, the decreasing sets $B_k = \{(x,y) : x \ge k\}$ (half-planes) all have infinite area but intersect in the empty set.

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

from matplotlib.patches import Circle, Wedge

fig, (ax, ax2, ax3) = plt.subplots(1, 3, figsize=(13, 4.2), gridspec_kw={"width_ratios": [1, 1.2, 1]})
cols = [BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED]
radii = [1 - 1 / (k + 1) for k in range(1, 7)]
prev = 0
for k, (r, col) in enumerate(zip(radii, cols), start=1):
    ax.add_patch(Wedge((0, 0), r, 0, 360, width=r - prev, facecolor=col, alpha=0.85, edgecolor="white"))
    prev = r
ax.add_patch(Circle((0, 0), 1, fill=False, edgecolor=INK, lw=1.2, ls="--"))
ax.set_xlim(-1.1, 1.1); ax.set_ylim(-1.1, 1.1); ax.set_aspect("equal"); ax.axis("off")
ax.set_title(r"Rings $G_k = A_k \setminus A_{k-1}$")

k = np.arange(1, 31)
areas = np.pi * (1 - 1 / (k + 1)) ** 2
ax2.plot(k, areas, "o-", color=BLUE, ms=4, label=r"$\mu(A_k)$")
ax2.axhline(np.pi, color=RED, ls="--", lw=1.2, label=r"$\mu(\bigcup_k A_k) = \pi$")
ax2.set_xlabel("$k$"); ax2.set_title("Continuity from below"); ax2.legend(loc="lower right")

ax3.axvspan(1, 6, color=VIOLET, alpha=0.08)
for kk, col in zip(range(1, 5), [BLUE, ORANGE, AQUA, VIOLET]):
    ax3.annotate("", xy=(5.9, 0.9 - 0.12 * kk - 0.04), xytext=(kk, 0.9 - 0.12 * kk - 0.04),
                 xycoords=("data", "axes fraction"), arrowprops=dict(arrowstyle="->", color=col, lw=1.6))
    ax3.axvline(kk, color=col, lw=2)
    ax3.text(kk + 0.08, 0.9 - 0.12 * kk, f"$B_{kk}$", color=col, fontsize=12, transform=ax3.get_xaxis_transform())
ax3.set_xlim(0, 6); ax3.set_ylim(-1, 1)
ax3.set_title(r"$B_k$: area $\infty$ for all $k$, $\bigcap_k B_k = \emptyset$")
ax3.set_xlabel("$x$")
plt.tight_layout(); plt.show()
```
