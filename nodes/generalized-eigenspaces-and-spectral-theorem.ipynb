# Generalized Eigenspaces and the Spectral Theorem

**Prerequisites:** [Direct Sums of Several Subspaces](direct-sums-and-complements.ipynb#def-direct-sum-multiple) · [Eigenvalues and Eigenvectors](eigenvalues-and-eigenvectors.ipynb) · [Diagonalization](diagonalization.ipynb) · [The Cayley-Hamilton Theorem](cayley-hamilton-theorem.ipynb) · [GCD and Unique Factorization of Polynomials](polynomial-gcd-and-unique-factorization.ipynb)

The **spectrum** of an operator is the set of its eigenvalues, the roots of the [characteristic polynomial](eigenvalues-and-eigenvectors.ipynb#thm-characteristic-equation). A **spectral theorem** relates the spectrum to the way the operator acts. The most basic one, proved here, says: over $\mathbb{C}$, the factorization of the characteristic polynomial
$$p_f(\lambda) = (\lambda - \lambda_1)^{r_1} \cdots (\lambda - \lambda_k)^{r_k}$$
is mirrored by a splitting of the space into $k$ pieces, of dimensions $r_1, \dots, r_k$, that $f$ maps into themselves, with $\lambda_i$ the only eigenvalue on the $i$-th piece. When all roots are simple, the pieces are eigenlines and $f$ is [diagonalizable](diagonalization.ipynb#thm-sufficient-diagonalizability). Repeated roots are where all the difficulty lies, and they are handled by **generalized eigenvectors**.

Throughout, $V$ is a vector space of dimension $n \ge 1$ over a field $K$ and $f : V \to V$ is a linear operator. For a polynomial $s = \sum_j a_j x^j \in K[x]$, $s(f) = \sum_j a_j f^j$ with $f^0 = \mathrm{id}_V$, as in the [Cayley-Hamilton Theorem](cayley-hamilton-theorem.ipynb#thm-cayley-hamilton). We write $f - \lambda$ for $f - \lambda\,\mathrm{id}_V$.

::: {.callout-note}
## Polynomials in One Operator Commute
Powers of $f$ commute with each other, so $s(f)\, t(f) = t(f)\, s(f)$ for all polynomials $s, t$, and $(s t)(f) = s(f)\, t(f)$ because the coefficients are scalars ([commutation lemma](rings-and-polynomial-rings.ipynb#lem-commutation-evaluation)). Every identity between polynomials therefore becomes an identity between operators when $f$ is substituted for $x$. The proofs below use this constantly.
:::

## 1\. Invariant Subspaces

::: {#def-invariant-subspace}
## Invariant Subspace and Restriction

A subspace $W \le V$ is **invariant** under $f$ (or **$f$-invariant**) if $f(W) \subseteq W$. Then $f$ restricts to an operator
$$f|_W : W \to W, \qquad w \mapsto f(w),$$
the **restriction** of $f$ to $W$. For every polynomial $s$, $s(f)$ maps $W$ into $W$ and $s(f)|_W = s(f|_W)$.
:::

$\{0\}$, $V$, every [eigenspace](eigenvalues-and-eigenvectors.ipynb#def-eigenvalue-eigenvector) $E_\lambda$, $\ker f$ and $\operatorname{im} f$ are invariant. The last three are special cases of the following.

::: {#prp-kernel-polynomial-invariant}
## Kernels and Images of Polynomials in $f$ Are Invariant

For every $s \in K[x]$, the subspaces $\ker s(f)$ and $\operatorname{im} s(f)$ are $f$-invariant.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-kernel-polynomial-invariant
If $s(f) v = 0$, then $s(f)\big(f(v)\big) = f\big(s(f) v\big) = f(0) = 0$, so $f(v) \in \ker s(f)$. If $w = s(f) v$, then $f(w) = s(f)\big(f(v)\big) \in \operatorname{im} s(f)$. Both use only $f\, s(f) = s(f)\, f$. $\blacksquare$
:::

When the whole space is a [direct sum](direct-sums-and-complements.ipynb#def-direct-sum-multiple) of invariant subspaces, the operator falls apart into independent pieces, and so does its matrix and its characteristic polynomial.

::: {#prp-invariant-direct-sum-block}
## Invariant Direct Sums Give Block-Diagonal Matrices

Let $V = W_1 \oplus \dots \oplus W_k$ with every $W_i$ invariant under $f$, and let $\mathcal{B}$ be the basis of $V$ obtained by [putting together](direct-sums-and-complements.ipynb#prp-dim-direct-sum-multiple) bases $\mathcal{B}_i$ of the $W_i$, in this order. Then
$$[f]_{\mathcal{B}} = \begin{pmatrix} A_1 & & 0 \\ & \ddots & \\ 0 & & A_k \end{pmatrix}, \qquad A_i = [f|_{W_i}]_{\mathcal{B}_i},$$
with zeros outside the diagonal blocks, and
$$p_f = p_{f|_{W_1}} \cdot p_{f|_{W_2}} \cdots p_{f|_{W_k}} .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-invariant-direct-sum-block
**The matrix.** The columns of $[f]_{\mathcal{B}}$ belonging to $\mathcal{B}_i$ hold the [coordinates](matrix-of-a-linear-map.ipynb#def-matrix-linear-map) of $f(v)$ for $v \in \mathcal{B}_i$. Since $f(v) \in W_i$, it is a combination of $\mathcal{B}_i$ alone, so these columns vanish outside the rows of $\mathcal{B}_i$, and the remaining block holds the coordinates of $f|_{W_i}(v)$ in $\mathcal{B}_i$, which is the corresponding column of $A_i$.

**Determinants of block-diagonal matrices.** We show $\det \operatorname{diag}(A_1, \dots, A_k) = \det A_1 \cdots \det A_k$. By induction on $k$ it suffices to treat two blocks, $D = \operatorname{diag}(A, B)$ with $A \in M_a(K)$, $B \in M_b(K)$. Block multiplication gives
$$D = \begin{pmatrix} A & 0 \\ 0 & I_b \end{pmatrix} \begin{pmatrix} I_a & 0 \\ 0 & B \end{pmatrix}.$$
Expanding $\det \operatorname{diag}(A, I_b)$ along its last row by the [Laplace expansion](laplace-expansion-adjugate-and-cramers-rule.ipynb#thm-laplace-expansion): the only non-zero entry is the $1$ in the corner, with sign $(-1)^{2(a+b)} = 1$, and its minor is $\operatorname{diag}(A, I_{b-1})$. By induction on $b$, $\det \operatorname{diag}(A, I_b) = \det A$. Expanding along the first row in the same way gives $\det \operatorname{diag}(I_a, B) = \det B$. By [multiplicativity](determinant.ipynb#thm-det-multiplicative), $\det D = \det A \det B$.

**The characteristic polynomial.** $\lambda I_n - [f]_{\mathcal{B}} = \operatorname{diag}(\lambda I - A_1, \dots, \lambda I - A_k)$, so its determinant is $\prod_i \det(\lambda I - A_i) = \prod_i p_{f|_{W_i}}(\lambda)$. (As polynomial identities these follow by applying the determinant formula with entries in $K[\lambda]$, or by noting that both sides agree for every value of $\lambda$ in an infinite field.) $\blacksquare$
:::

::: {.callout-note}
## Intuition: Pieces That Do Not Talk to Each Other
Invariance of every $W_i$ means that $f$ never moves a vector of one piece into another piece. Writing $v = w_1 + \dots + w_k$, the operator acts piece by piece: $f(v) = f|_{W_1}(w_1) + \dots + f|_{W_k}(w_k)$. Understanding $f$ reduces to understanding the smaller operators $f|_{W_i}$; that is what the block-diagonal matrix records.
:::

## 2\. Generalized Eigenvectors

A vector $v \ne 0$ is an eigenvector for $\lambda$ exactly when $(f - \lambda)v = 0$, that is, $v \in \ker(f - \lambda)$; note that $\ker(f - \lambda) = \ker(\lambda - f)$. Some vectors are not killed by one application of $f - \lambda$, but are killed after several.

::: {#def-generalized-eigenvector}
## Generalized Eigenvector and Generalized Eigenspace

Let $\lambda \in K$. A vector $v \ne 0$ is a **generalized eigenvector** of $f$ for $\lambda$ if
$$(f - \lambda)^m\, v = 0 \qquad\text{for some integer } m \ge 1 .$$
The **generalized eigenspace** of $\lambda$ is
$$G_\lambda(f) = \{\, v \in V \mid (f - \lambda)^m v = 0 \text{ for some } m \ge 1 \,\} = \bigcup_{m \ge 1} \ker (f - \lambda)^m .$$
For a matrix $A$, $G_\lambda(A)$ is the generalized eigenspace of $x \mapsto A x$.
:::

Here $(f - \lambda)^m$ means $f - \lambda$ composed with itself $m$ times; for a matrix, $(A - \lambda I)^m$ is the $m$-th power. Every eigenvector is a generalized eigenvector ($m = 1$), so $E_\lambda \subseteq G_\lambda(f)$.

::: {#prp-kernel-chain}
## The Chain of Kernels

Let $N : V \to V$ be any operator.

1. $\{0\} \subseteq \ker N \subseteq \ker N^2 \subseteq \ker N^3 \subseteq \cdots$
2. If $\ker N^j = \ker N^{j+1}$ for some $j$, then $\ker N^j = \ker N^{j+l}$ for all $l \ge 0$.
3. The chain becomes constant from some index $j \le n$ on, and before that it grows strictly, the dimension increasing by at least $1$ at each step.

Applied to $N = f - \lambda$: $G_\lambda(f) = \ker (f - \lambda)^n$, so it is an $f$-invariant subspace, and $G_\lambda(f) \ne \{0\}$ if and only if $\lambda$ is an eigenvalue of $f$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-kernel-chain
1. If $N^j v = 0$, then $N^{j+1} v = N(N^j v) = N(0) = 0$.
2. By induction on $l$ it suffices to show $\ker N^{j+2} \subseteq \ker N^{j+1}$. If $N^{j+2} v = 0$, then $N v \in \ker N^{j+1} = \ker N^j$, so $N^{j+1} v = N^j (N v) = 0$.
3. By 2, as long as the chain has not become constant, every inclusion is strict, and a strictly larger subspace has strictly larger dimension. The dimensions are at most $n$, so there are at most $n$ strict steps.

For $N = f - \lambda$, part 3 gives $\ker N^m \subseteq \ker N^n$ for every $m$ (for $m \ge n$ the chain is already constant), so the union $G_\lambda(f)$ equals $\ker N^n$, a subspace; it is invariant by @prp-kernel-polynomial-invariant. If $\lambda$ is an eigenvalue, $\{0\} \ne E_\lambda \subseteq G_\lambda(f)$. Conversely, let $v \ne 0$ with $N^m v = 0$ and let $j$ be the largest exponent with $u = N^j v \ne 0$ (so $0 \le j < m$). Then $N u = 0$, so $u$ is an eigenvector for $\lambda$. $\blacksquare$
:::

The last step of the proof is worth remembering: applying $f - \lambda$ repeatedly to a generalized eigenvector produces, just before it reaches $0$, a genuine eigenvector.

::: {#exm-shear-generalized}
## The Shear

$J = \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}$ on $\mathbb{R}^2$ has the single eigenvalue $1$, and with $N = J - I = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}$,
$$E_1 = \ker N = \operatorname{span}(e_1) \quad (\text{the } x\text{-axis}), \qquad N^2 = 0, \qquad G_1(J) = \ker N^2 = \mathbb{R}^2 .$$
So every non-zero vector is a generalized eigenvector, but only those on the $x$-axis are eigenvectors. For $v = (a, b)$ with $b \ne 0$: $N v = (b, 0)$ lies on the eigenline, and $N^2 v = 0$. The chain of kernels has dimensions $0, 1, 2, 2, \dots$

The shift operator $N(x_1, x_2, x_3) = (x_2, x_3, 0)$ on $K^3$, which is $J_3 - I$ for $J_3 = \begin{pmatrix} 1 & 1 & 0 \\ 0 & 1 & 1 \\ 0 & 0 & 1 \end{pmatrix}$, has $\ker N = \operatorname{span}(e_1)$, $\ker N^2 = \operatorname{span}(e_1, e_2)$, $\ker N^3 = K^3$: the dimensions grow by exactly one per step, $e_3 \mapsto e_2 \mapsto e_1 \mapsto 0$.
:::

The animation applies $N = J - I$ twice. The first application flattens the plane onto the eigenline (the kernel, red) and sends the eigenvector $e_1$ (blue) to $0$; the test vector $v$ (yellow), which is only a generalized eigenvector, lands **on** the eigenline. The second application sends everything to $0$.


```python

from la_manim import *

class GeneralizedEigenvector(GridScene):
    def construct(self):
        N = [[0, 1], [0, 0]]                      # N = J - I for the shear J
        eigenline = Line(LEFT * 8, RIGHT * 8, color=C_RED, stroke_width=6, stroke_opacity=0.55)
        self.add(eigenline)
        g1 = self.vec((1, 0), C_BLUE, r"e_1", label_dir=DOWN)
        g2 = self.vec((1, 2), C_YELLOW, r"v", label_dir=UR)
        g1.update(); g2.update()                  # place the labels before the first frame
        cap = caption(r"N = J - I = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}")
        txt = text_caption(r"red: $\ker N = E_1$, the eigenline")
        self.add(cap, txt)
        self.wait(1.2)

        self.transform(N)                         # v = (1, 2) -> (2, 0) on the eigenline, e_1 -> 0
        txt2 = text_caption(r"after one step: $Nv = (2,0)$ is an eigenvector, $Ne_1 = 0$")
        self.play(Transform(txt, txt2))
        self.wait(1.5)

        self.transform(N)                         # everything -> 0
        txt3 = text_caption(r"after two steps: $N^2 = 0$, so $G_1(J) = \mathbb{R}^2$")
        self.play(Transform(txt, txt3))
        self.wait(2)

show(GeneralizedEigenvector)
```


## 3\. The Spectral Theorem

Two lemmas carry the proof. The first is where the algebra of [greatest common divisors](polynomial-gcd-and-unique-factorization.ipynb#thm-gcd) enters linear algebra.

::: {#lem-coprime-kernel-splitting}
## Coprime Factors Split the Space

Let $s, t \in K[x]$ be [coprime](polynomial-gcd-and-unique-factorization.ipynb#def-coprime) and suppose $(s t)(f) = 0$. Then
$$V = \ker s(f) \oplus \ker t(f),$$
and both summands are $f$-invariant. Explicitly, if $a s + b t = 1$, the components of $v \in V$ are $b(f) t(f) v \in \ker s(f)$ and $a(f) s(f) v \in \ker t(f)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-coprime-kernel-splitting
By [Bézout](polynomial-gcd-and-unique-factorization.ipynb#cor-coprime-bezout) there are $a, b \in K[x]$ with $a s + b t = 1$. Substituting $f$ (a polynomial identity remains true after substitution, since the coefficients are scalars):
$$\mathrm{id}_V = a(f)\, s(f) + b(f)\, t(f). \tag{$\ast$}$$

**The sum is all of $V$.** Apply $(\ast)$ to $v$: $v = a(f)s(f)v + b(f)t(f)v$. The first term lies in $\ker t(f)$, because polynomials in $f$ commute:
$$t(f)\, a(f)\, s(f)\, v = a(f)\, (s t)(f)\, v = 0 .$$
Likewise $s(f)\, b(f)\, t(f)\, v = b(f)\,(s t)(f)\, v = 0$, so the second term lies in $\ker s(f)$. (Note the swap: the term containing $s(f)$ lands in the kernel of $t(f)$.)

**The sum is direct.** If $y \in \ker s(f) \cap \ker t(f)$, then by $(\ast)$, $y = a(f)\, s(f) y + b(f)\, t(f) y = 0 + 0 = 0$. For two subspaces, a trivial intersection means the sum is [direct](direct-sums-and-complements.ipynb#prp-direct-sum-intersection).

Invariance is @prp-kernel-polynomial-invariant. $\blacksquare$
:::

::: {#lem-single-eigenvalue-block}
## A Piece Killed by $(f - \mu)^r$

Let $W \ne \{0\}$ be $f$-invariant with $(f - \mu)^r w = 0$ for all $w \in W$, for some $\mu \in K$ and $r \ge 1$. Then:

1. $\mu$ is the only possible eigenvalue of $f|_W$;
2. for every $\lambda \ne \mu$, $(f - \lambda)|_W$ is invertible, hence so is $(f - \lambda)^m|_W$ for every $m$;
3. if $K$ is algebraically closed, $p_{f|_W}(\lambda) = (\lambda - \mu)^{\dim W}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-single-eigenvalue-block
1. If $f(w) = \nu w$ with $0 \ne w \in W$, then $(f - \mu)^r w = (\nu - \mu)^r w$ by the [spectral mapping property](eigenvalues-and-eigenvectors.ipynb#thm-spectral-mapping-polynomial). This is $0$, and $w \ne 0$, so $\nu = \mu$.
2. If $\lambda \ne \mu$, then $\lambda$ is not an eigenvalue of $f|_W$ by 1, so $(f - \lambda)|_W$ has kernel $\{0\}$. An injective operator on the finite-dimensional $W$ is bijective by [rank-nullity](kernel-image-and-rank-nullity.ipynb#cor-dimension-constraints). Compositions of invertible maps are invertible.
3. Over an algebraically closed field $p_{f|_W}$ splits into linear factors $\lambda - \nu$, where every $\nu$ is an eigenvalue of $f|_W$; by 1 each $\nu = \mu$. The degree is $\dim W$. $\blacksquare$
:::

::: {#thm-spectral}
## The Spectral Theorem (Primary Decomposition)

Let $K$ be [algebraically closed](eigenvalues-and-eigenvectors.ipynb#def-algebraically-closed-field) (for example $K = \mathbb{C}$) and factor the characteristic polynomial as
$$p_f(\lambda) = (\lambda - \lambda_1)^{r_1} (\lambda - \lambda_2)^{r_2} \cdots (\lambda - \lambda_k)^{r_k},$$
with $\lambda_1, \dots, \lambda_k$ **pairwise distinct** and $r_i \ge 1$ the [algebraic multiplicities](diagonalization.ipynb#def-multiplicities), so $r_1 + \dots + r_k = n$. Put
$$V_i = \ker (f - \lambda_i)^{r_i}, \qquad i = 1, \dots, k .$$
Then:

1. $V = V_1 \oplus V_2 \oplus \dots \oplus V_k$;
2. every $V_i$ is invariant under $f$;
3. $\dim V_i = r_i$, and the characteristic polynomial of the restriction is $p_{f|_{V_i}}(\lambda) = (\lambda - \lambda_i)^{r_i}$; so $\lambda_i$ is the only eigenvalue of $f|_{V_i}$;
4. $V_i = G_{\lambda_i}(f)$ is the generalized eigenspace of $\lambda_i$; in particular $\ker (f - \lambda_i)^m = V_i$ for every $m \ge r_i$.

Moreover the decomposition is unique: if $V = U_1 \oplus \dots \oplus U_k$ with $f$-invariant $U_i$ and $p_{f|_{U_i}} = (\lambda - \lambda_i)^{r_i}$, then $U_i = V_i$ for all $i$.
:::

::: {.callout-warning}
## Many Theorems Are Called "Spectral"
"Spectral theorem" is a family name for results linking the spectrum of an operator to its structure. The best-known member concerns symmetric and Hermitian matrices, which are diagonalizable by an orthonormal basis. The theorem here is the most elementary member, valid for **every** operator over $\mathbb{C}$; it is also called the **primary decomposition theorem** or the **generalized eigenspace decomposition**.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-spectral
We prove parts 1 to 3 by induction on the number $k$ of distinct eigenvalues, for all spaces and operators at once; part 4 and uniqueness follow afterwards.

**Case $k = 1$.** Then $p_f = (\lambda - \lambda_1)^n$, and by [Cayley-Hamilton](cayley-hamilton-theorem.ipynb#thm-cayley-hamilton) $(f - \lambda_1)^n = p_f(f) = 0$, so $V_1 = \ker(f - \lambda_1)^n = V$. Parts 1 to 3 are immediate.

**Induction step.** Let $k \ge 2$ and assume parts 1 to 3 for operators with $k - 1$ distinct eigenvalues. Split off the first factor:
$$s = (x - \lambda_1)^{r_1}, \qquad t = \prod_{i \ge 2} (x - \lambda_i)^{r_i}, \qquad s t = p_f .$$
$s$ and $t$ have no common root, so they are [coprime](polynomial-gcd-and-unique-factorization.ipynb#cor-factorization-algebraically-closed), and $(s t)(f) = p_f(f) = 0$ by Cayley-Hamilton. By @lem-coprime-kernel-splitting,
$$V = V_1 \oplus W, \qquad V_1 = \ker (f - \lambda_1)^{r_1}, \quad W = \ker t(f),$$
both $f$-invariant.

*Characteristic polynomials of the two pieces.* By @prp-invariant-direct-sum-block, $p_f = p_{f|_{V_1}} \cdot p_{f|_W}$. By @lem-single-eigenvalue-block (with $\mu = \lambda_1$), $p_{f|_{V_1}} = (\lambda - \lambda_1)^{\dim V_1}$ (this also holds if $V_1 = \{0\}$, with the empty determinant $1$). Every eigenvalue $\nu$ of $f|_W$ satisfies $t(\nu) w = t(f) w = 0$ for an eigenvector $w$, so $\nu \in \{\lambda_2, \dots, \lambda_k\}$; hence $p_{f|_W} = \prod_{i \ge 2} (\lambda - \lambda_i)^{e_i}$ for some $e_i \ge 0$. Comparing with $p_f = \prod_i (\lambda - \lambda_i)^{r_i}$, [uniqueness of factorization](polynomial-gcd-and-unique-factorization.ipynb#cor-factorization-algebraically-closed) gives
$$\dim V_1 = r_1, \qquad p_{f|_W} = \prod_{i \ge 2} (\lambda - \lambda_i)^{r_i} .$$
So $f|_{V_1}$ satisfies part 3, and $f|_W$ is an operator with exactly $k - 1$ distinct eigenvalues ($\dim W = n - r_1 \ge 1$).

*Induction.* Applied to $f|_W$, the hypothesis gives $W = W_2 \oplus \dots \oplus W_k$ with $W_i = \ker (f|_W - \lambda_i)^{r_i}$ invariant, of dimension $r_i$, and $p_{f|_{W_i}} = (\lambda - \lambda_i)^{r_i}$.

*The pieces of $W$ are the $V_i$.* Let $i \ge 2$. Clearly $W_i \subseteq V_i = \ker (f - \lambda_i)^{r_i}$. Conversely, let $(f - \lambda_i)^{r_i} v = 0$ and write $v = v_1 + w$ with $v_1 \in V_1$, $w \in W$. Then
$$0 = (f - \lambda_i)^{r_i} v_1 + (f - \lambda_i)^{r_i} w,$$
where the two terms lie in the invariant subspaces $V_1$ and $W$. As the sum is direct, both are $0$. Since $(f - \lambda_1)^{r_1}$ vanishes on $V_1$ and $\lambda_i \ne \lambda_1$, @lem-single-eigenvalue-block says $(f - \lambda_i)^{r_i}$ is invertible on $V_1$, so $v_1 = 0$. Thus $v = w \in W$ and $(f|_W - \lambda_i)^{r_i} w = 0$, i.e. $v \in W_i$. Hence $W_i = V_i$.

*Assembling.* $V = V_1 \oplus (V_2 \oplus \dots \oplus V_k)$. This is a direct sum of all $k$ pieces by [criterion 2](direct-sums-and-complements.ipynb#prp-direct-sum-criterion): if $v_1 + v_2 + \dots + v_k = 0$ with $v_i \in V_i$, then $v_1 = 0$ and $v_2 + \dots + v_k = 0$ because $V = V_1 \oplus W$, and then all $v_i = 0$ because $W = \bigoplus_{i \ge 2} V_i$. The restrictions $f|_{V_i} = (f|_W)|_{W_i}$ have the required characteristic polynomials. This proves parts 1 to 3.

**Part 4.** $V_i = \ker(f - \lambda_i)^{r_i} \subseteq G_{\lambda_i}(f)$ by definition. Conversely let $(f - \lambda_i)^m v = 0$ and decompose $v = v_1 + \dots + v_k$, $v_j \in V_j$. The vectors $(f - \lambda_i)^m v_j \in V_j$ add up to $0$, so each is $0$ by directness. For $j \ne i$, $(f - \lambda_i)^m$ is invertible on $V_j$ by @lem-single-eigenvalue-block (as $(f - \lambda_j)^{r_j}$ vanishes on $V_j$), so $v_j = 0$. Hence $v = v_i \in V_i$. Finally, for $m \ge r_i$: $V_i = \ker(f - \lambda_i)^{r_i} \subseteq \ker (f - \lambda_i)^m \subseteq G_{\lambda_i}(f) = V_i$.

**Uniqueness.** By Cayley-Hamilton for $f|_{U_i}$, $(f - \lambda_i)^{r_i}$ vanishes on $U_i$, so $U_i \subseteq V_i$; both have dimension $r_i$, so $U_i = V_i$. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Why the Kernels Stop Growing at $r_i$
On $V_j$ with $j \ne i$, the operator $f - \lambda_i$ is invertible, so no power of it kills anything there. All growth of the chain $\ker(f - \lambda_i)^m$ happens inside $V_i$, which has dimension $r_i$, and there the power $r_i$ already kills everything. A generalized eigenspace larger than $r_i$ would make $\lambda_i$ a root of $p_f$ more than $r_i$ times, which is impossible. How fast the chain fills $V_i$ (in one step, or in several) is not determined by $r_i$; it is measured by the [minimal polynomial](minimal-polynomial.ipynb#def-index-of-eigenvalue).
:::

::: {#cor-block-diagonal-form}
## Block-Diagonal Form

Under the hypotheses of @thm-spectral, choose a basis $\mathcal{B}_i$ of each $V_i$ and let $\mathcal{B}$ be their concatenation. Then
$$[f]_{\mathcal{B}} = \begin{pmatrix} A_1 & & & 0 \\ & A_2 & & \\ & & \ddots & \\ 0 & & & A_k \end{pmatrix}, \qquad A_i \in M_{r_i}(K), \quad p_{A_i}(\lambda) = (\lambda - \lambda_i)^{r_i}, \quad (A_i - \lambda_i I)^{r_i} = 0 .$$
In matrix form: every $A \in M_n(K)$ is [similar](change-of-basis-and-similarity.ipynb#def-matrix-similarity) to such a block-diagonal matrix, one block per distinct eigenvalue, of size equal to its algebraic multiplicity.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-block-diagonal-form
@prp-invariant-direct-sum-block applied to the invariant direct sum of @thm-spectral gives the block form with $A_i = [f|_{V_i}]_{\mathcal{B}_i}$, of size $\dim V_i = r_i$, and $p_{A_i} = p_{f|_{V_i}} = (\lambda - \lambda_i)^{r_i}$ by part 3. $(A_i - \lambda_i I)^{r_i}$ is the matrix of $(f - \lambda_i)^{r_i}|_{V_i}$, which is $0$ by the definition of $V_i$. For a matrix $A$, apply this to $f(x) = Ax$ and use the [transformation law](change-of-basis-and-similarity.ipynb#thm-operator-change-of-basis). $\blacksquare$
:::

::: {#cor-diagonalizable-generalized}
## Diagonalizable Means "Generalized Eigenvectors Are Eigenvectors"

Over an algebraically closed field, $f$ is diagonalizable if and only if $E_{\lambda_i} = G_{\lambda_i}(f)$ for every eigenvalue, that is, if and only if $(f - \lambda_i)$ already vanishes on $V_i$ for every $i$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-diagonalizable-generalized
$E_{\lambda_i} \subseteq G_{\lambda_i}(f) = V_i$ and $\dim V_i = r_i$, so $E_{\lambda_i} = G_{\lambda_i}(f)$ if and only if $\dim E_{\lambda_i} = r_i$, i.e. geometric and algebraic multiplicity agree. By the [diagonalizability criterion](diagonalization.ipynb#prp-multiplicity-inequality) this holds for all $i$ exactly when $f$ is diagonalizable. $\blacksquare$
:::

So the blocks $A_i$ are scalar matrices $\lambda_i I$ precisely in the diagonalizable case. In general $A_i = \lambda_i I + N_i$ with $N_i^{r_i} = 0$: what remains to be understood is the inner structure of each block, where a well-chosen basis of $V_i$ makes $N_i$ as simple as possible.

## 4\. A Worked Example

::: {#exm-spectral-5x5}
## A $5 \times 5$ Matrix over $\mathbb{C}$

Let $f(x) = A x$ on $\mathbb{C}^5$ with
$$A = \begin{pmatrix} 2 & 0 & 0 & 0 & 0 \\ 0 & 3 & 0 & 2 & -1 \\ 1 & 0 & 2 & 1 & 0 \\ 0 & -2 & 0 & -1 & 0 \\ 0 & 0 & 0 & 1 & 0 \end{pmatrix}.$$

**Characteristic polynomial.** Expand $\det(\lambda I - A)$ along the first row (only $\lambda - 2$ survives), then the remaining $4 \times 4$ determinant along its second column (only $\lambda - 2$ survives again):
$$p_A(\lambda) = (\lambda - 2)^2 \det \begin{pmatrix} \lambda - 3 & -2 & 1 \\ 2 & \lambda + 1 & 0 \\ 0 & -1 & \lambda \end{pmatrix} = (\lambda - 2)^2 \big[(\lambda - 3)(\lambda^2 + \lambda) + 4\lambda - 2\big] .$$
The bracket is $\lambda^3 - 2\lambda^2 + \lambda - 2 = (\lambda - 2)(\lambda^2 + 1)$, so
$$p_A(\lambda) = (\lambda - 2)^3 (\lambda^2 + 1) = (\lambda - 2)^3 (\lambda - i)(\lambda + i) .$$
Over $\mathbb{R}$ the factor $\lambda^2 + 1$ is [irreducible](polynomial-gcd-and-unique-factorization.ipynb#exm-irreducible) and the theorem does not apply as stated; over $\mathbb{C}$ it does.

**What the theorem predicts.** Three generalized eigenspaces: $V_2$ of dimension $3$, and $V_i$, $V_{-i}$ of dimension $1$ (these are just eigenlines). In a basis adapted to $\mathbb{C}^5 = V_i \oplus V_{-i} \oplus V_2$ the matrix becomes $\operatorname{diag}\big(i,\, -i,\, A_3\big)$ with a $3 \times 3$ block $A_3$ whose only eigenvalue is $2$.

**The eigenlines.** Solving $(A - iI)w = 0$ gives $V_i = \operatorname{span}(w)$ with $w = (0,\ 5 - 5i,\ 2 - 4i,\ 10i,\ 10)$. Since $A$ is real, conjugating gives $V_{-i} = \operatorname{span}(\bar w)$.

**The generalized eigenspace of $2$.** With $N = A - 2I$, the equations $N x = 0$ read $x_1 + x_4 = 0$, $-2x_2 - 3x_4 = 0$, $x_4 - 2x_5 = 0$, $x_2 + 2x_4 - x_5 = 0$, so
$$E_2 = \ker N = \operatorname{span}\big(e_3,\ u\big), \qquad u = (-2, -3, 0, 2, 1).$$
Only $2 < 3$ independent eigenvectors: $A$ is **not diagonalizable**. But $N e_1 = e_3$ and $N e_3 = 0$, so $N^2 e_1 = 0$: $e_1$ is a generalized eigenvector that is not an eigenvector. The independent vectors $e_3, e_1, u$ lie in $\ker N^2 \subseteq V_2$, and $\dim V_2 = 3$, so
$$V_2 = \ker N^2 = \ker N^3 = \operatorname{span}(e_3, e_1, u) .$$

**The block form.** In the basis $\mathcal{B} = (w, \bar w, e_3, e_1, u)$: $A e_3 = 2 e_3$, $A e_1 = 2 e_1 + e_3$, $A u = 2 u$, hence
$$[f]_{\mathcal{B}} = \begin{pmatrix} i & 0 & 0 & 0 & 0 \\ 0 & -i & 0 & 0 & 0 \\ 0 & 0 & 2 & 1 & 0 \\ 0 & 0 & 0 & 2 & 0 \\ 0 & 0 & 0 & 0 & 2 \end{pmatrix}.$$
The block $A_3$ is not $2I$, but $(A_3 - 2I)^2 = 0$ already, one power less than the $r = 3$ guaranteed by the theorem. That the exponent $2$ (not $3$) is what matters is explained by the [minimal polynomial](minimal-polynomial.ipynb#exm-minimal-polynomial-5x5).

**The splitting map.** With the [Bézout identity](polynomial-gcd-and-unique-factorization.ipynb#exm-bezout-spectral) $1 = a(x)(x - 2)^3 + b(x)(x^2 + 1)$, @lem-coprime-kernel-splitting shows that $\pi_2 = b(A)(A^2 + I)$ sends every vector to its $V_2$-component, and $I - \pi_2 = a(A)(A - 2I)^3$ sends it to its component in $V_i \oplus V_{-i} = \ker(A^2 + I)$.
:::

The code checks every claim of the example in exact arithmetic, including that $\pi_2$ is the projection onto $V_2$ along $\ker(A^2 + I)$.


```python

import sympy as sp

A = sp.Matrix([[2, 0, 0, 0, 0], [0, 3, 0, 2, -1], [1, 0, 2, 1, 0],
               [0, -2, 0, -1, 0], [0, 0, 0, 1, 0]])
I5, lam, x = sp.eye(5), sp.symbols("lambda"), sp.symbols("x")
print("p_A =", sp.factor(A.charpoly(lam).as_expr(), extension=sp.I))

# the adapted basis (w, conj w, e3, e1, u) and the block-diagonal matrix
w = sp.Matrix([0, 5 - 5*sp.I, 2 - 4*sp.I, 10*sp.I, 10])
e1, e3, u = I5[:, 0], I5[:, 2], sp.Matrix([-2, -3, 0, 2, 1])
P = sp.Matrix.hstack(w, w.conjugate(), e3, e1, u)
print("P^{-1} A P =", sp.simplify(P.inv() * A * P).tolist())

# generalized eigenspace of 2: dim ker (A - 2I)^m for m = 1, 2, 3
N = A - 2*I5
print("dim ker N^m, m = 1..3:", [5 - (N**m).rank() for m in (1, 2, 3)])

# the splitting map from the Bezout identity 1 = a (x-2)^3 + b (x^2+1)
a, b, g = sp.gcdex(sp.expand((x - 2)**3), x**2 + 1, x)
def at(poly, M):                                 # evaluate a polynomial at a matrix (Horner)
    out = sp.zeros(*M.shape)
    for c in sp.Poly(poly, x).all_coeffs():
        out = out * M + c * sp.eye(M.shape[0])
    return out
pi2 = at(b, A) * (A**2 + I5)
zero = lambda M: sp.simplify(M).is_zero_matrix
print("pi_2 is a projection (pi_2^2 = pi_2):", zero(pi2**2 - pi2))
print("pi_2 fixes V_2 = span(e3, e1, u):    ", all(zero(pi2 * y - y) for y in (e3, e1, u)))
print("pi_2 kills V_i and V_-i:             ", zero(pi2 * w) and zero(pi2 * w.conjugate()))
v = sp.Matrix([1, 1, 1, 1, 1])
print("v = pi_2 v + (v - pi_2 v):", (pi2 * v).T, "+", (v - pi2 * v).T)
```


The figure tracks the chain of kernels $\ker (A - \lambda I)^m$ for each eigenvalue. Each staircase rises strictly until it reaches the algebraic multiplicity $r$ (dashed) and then stays flat, as @prp-kernel-chain and part 4 of @thm-spectral say. For $\lambda = 2$ it climbs in two steps, $2 \to 3$, so the plateau is reached at $m = 2 < r = 3$.


```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA = "#2a78d6", "#eb6834", "#1baf7a"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "xtick.color": MUTED, "ytick.color": MUTED,
    "legend.frameon": False, "font.size": 10})

A = np.array([[2, 0, 0, 0, 0], [0, 3, 0, 2, -1], [1, 0, 2, 1, 0],
              [0, -2, 0, -1, 0], [0, 0, 0, 1, 0]], dtype=complex)
n, ms = 5, np.arange(0, 6)
fig, axes = plt.subplots(1, 3, figsize=(11, 3.3), sharey=True)
for ax, (lam, r, col, name) in zip(axes, [(2, 3, BLUE, "A-2I"), (1j, 1, ORANGE, "A-iI"), (-1j, 1, AQUA, "A+iI")]):
    Nl = A - lam * np.eye(n)
    dims = [n - np.linalg.matrix_rank(np.linalg.matrix_power(Nl, m), tol=1e-9) for m in ms]
    ax.step(ms, dims, where="post", color=col, lw=2.5)
    ax.plot(ms, dims, "o", color=col, ms=6)
    ax.axhline(r, color=MUTED, ls="--", lw=1.2, label=f"algebraic multiplicity r = {r}")
    ax.set_title(rf"$\dim\ker({name})^m$")
    ax.set_xlabel("m"); ax.set_ylim(-0.3, 5.3); ax.legend(loc="upper right")
axes[0].set_ylabel("dimension")
plt.tight_layout(); plt.show()
```
