# Improper Integrals

**Prerequisites:** [The Fundamental Theorem of Calculus](fundamental-theorem-of-calculus.ipynb) · [The Logarithm and the Exponential Function](exponential-and-logarithm.ipynb) · [Properties of the Riemann Integral](properties-of-the-riemann-integral.ipynb)

The Riemann integral needs a bounded function on a bounded interval. Removing either restriction leads to limits of integrals.

## 1\. Definitions

::: {#def-improper-integral}
## Improper Integrals

1. **Unbounded interval.** If $f$ is integrable on $[a, A]$ for every $A > a$, put
$$\int_a^\infty f = \lim_{A \to \infty}\int_a^A f$$
if the limit exists; the integral then **converges**, otherwise it **diverges**. Similarly $\int_{-\infty}^a f$.
2. **Unbounded function.** If $f$ is integrable on $[a + \varepsilon, b]$ for every $\varepsilon \in (0, b - a)$ (typically $f$ is unbounded near $a$), put $\int_a^b f = \lim_{\varepsilon\to 0^+}\int_{a + \varepsilon}^b f$; similarly at $b$.
3. **Several singular points.** Split the interval at intermediate points so that each piece has one singular end; the integral converges if **every** piece converges, and is then the sum. (The choice of the splitting points does not matter, by additivity.)
:::

If $f$ is integrable on $[a, b]$ in the ordinary sense, the limit in (2) is the ordinary integral: $\big|\int_a^{a+\varepsilon} f\big| \le \varepsilon\sup|f| \to 0$.

::: {#exm-p-integrals}
## Powers

1. $\int_1^\infty\frac{dx}{x^p}$ converges if and only if $p > 1$, with value $\frac{1}{p - 1}$: $\int_1^A x^{-p}dx = \frac{A^{1-p} - 1}{1 - p}$ for $p \ne 1$ and $\ln A$ for $p = 1$.
2. $\int_0^1\frac{dx}{x^p}$ ($p > 0$) converges if and only if $p < 1$, with value $\frac{1}{1-p}$: $\int_\varepsilon^1 x^{-p}dx = \frac{1 - \varepsilon^{1-p}}{1-p}$ for $p \ne 1$ and $-\ln\varepsilon$ for $p = 1$.
3. Hence $\int_0^\infty\frac{dx}{x^p}$ diverges for every $p > 0$: split at $1$; one of the two pieces diverges.
4. $\int_0^\infty e^{-\alpha x}dx = \lim_A \frac{1 - e^{-\alpha A}}{\alpha} = \frac1\alpha$ for $\alpha > 0$.
:::

## 2\. Comparison

::: {#thm-improper-comparison}
## Comparison Test

Let $0 \le f \le g$ on $[K, \infty)$, both integrable on bounded subintervals of $[a, \infty)$, $K \ge a$. If $\int_a^\infty g$ converges, so does $\int_a^\infty f$; if $\int_a^\infty f$ diverges, so does $\int_a^\infty g$. The same holds at a singular endpoint of a bounded interval.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-improper-comparison
$\int_a^K$ is an ordinary integral, so only $[K, \infty)$ matters. $\Phi(A) = \int_K^A f$ is increasing in $A$ (as $f \ge 0$) and bounded by $\int_K^\infty g$. An increasing bounded function has a limit at $\infty$: with $s = \sup_A \Phi(A)$, for $\varepsilon > 0$ some $A_0$ has $\Phi(A_0) > s - \varepsilon$, and then $s - \varepsilon < \Phi(A) \le s$ for $A \ge A_0$ (as for [monotone sequences](sequence-limit.ipynb#thm-monotone-convergence)). The second statement is the contrapositive. $\blacksquare$
:::

::: {#exm-improper-comparison}
## Examples

1. $\int_1^\infty e^{-x^p}dx$ converges for $p > 1$: $x^p \ge x$ on $[1, \infty)$, so $e^{-x^p} \le e^{-x}$. (For $p = 2$ no elementary antiderivative exists, so comparison is the only way.)
2. $\int_1^2\frac{x^2\,dx}{\sqrt{x^5 - 1}}$ converges: $x^5 - 1 = (x - 1)(x^4 + x^3 + x^2 + x + 1) > 5(x - 1)$ for $x > 1$, so the integrand is $< \frac{4}{\sqrt5}\frac{1}{\sqrt{x - 1}}$, and $\int_1^2\frac{dx}{\sqrt{x - 1}} = \int_0^1\frac{dt}{\sqrt t} = 2$.
3. $\int_0^\infty\frac{dx}{\sqrt{x^3 + x}}$ converges: on $[1, \infty)$ the integrand is $< x^{-3/2}$, on $(0, 1]$ it is $< x^{-1/2}$.
:::

## 3\. Principal Values

$\int_{-1}^1\frac{dx}{x}$ diverges: both $\int_0^1$ and $\int_{-1}^0$ diverge. Removing a symmetric interval $(-\varepsilon, \varepsilon)$ gives $0$ for every $\varepsilon$ by symmetry, but removing $(-\varepsilon_1, \varepsilon_2)$ gives
$$\int_{-1}^{-\varepsilon_1}\frac{dx}{x} + \int_{\varepsilon_2}^1\frac{dx}{x} = \ln\varepsilon_1 - \ln\varepsilon_2 = \ln\frac{\varepsilon_1}{\varepsilon_2},$$
which can tend to any real number $r$ (take $\varepsilon_1 = e^r\varepsilon_2$) or to $\pm\infty$. The symmetric limit is called the **Cauchy principal value**: $\operatorname{PV}\int_{-1}^1\frac{dx}{x} = 0$; it is a convention, not a convergent integral.

## 4\. The Gamma Function

::: {#prp-gamma-converges}
## Convergence of the Gamma Integral

$\Gamma(x) = \int_0^\infty t^{x-1}e^{-t}\,dt$ converges for every $x > 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-gamma-converges
Split at $1$. On $[1, \infty)$: $t^{x-1}e^{-t/2} \to 0$ ([growth rates](exponential-and-logarithm.ipynb#prp-exp-log-limits)), so $t^{x-1}e^{-t} \le e^{-t/2}$ for $t \ge K$, and $\int^\infty e^{-t/2}$ converges. On $(0, 1]$ (singular only if $x < 1$): $t^{x-1}e^{-t} \le t^{x-1} = \frac{1}{t^{1-x}}$ with $1 - x < 1$. $\blacksquare$
:::

Integration by parts gives $\Gamma(x + 1) = x\Gamma(x)$, hence $\Gamma(n + 1) = n!$ since $\Gamma(1) = 1$; and $\Gamma(\frac12) = 2\int_0^\infty e^{-u^2}du = \sqrt\pi$. These are proved in [Gamma Function and Distribution](gamma-function-and-distribution.ipynb#thm-gamma-recurrence) (see also its [half-integer values](gamma-function-and-distribution.ipynb#prp-gamma-half-integer)). So $\Gamma$ extends the factorial to all $x > 0$.
