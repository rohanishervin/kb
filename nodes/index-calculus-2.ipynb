## Notation

The same symbols mean the same thing in every note.

| Symbol | Meaning |
|---|---|
| $x = (x_1, \dots, x_n)$, $e_j$; $\mathbf i, \mathbf j, \mathbf k$ | point or vector of $\mathbb{R}^n$, standard basis; standard basis of $\mathbb{R}^3$ |
| $u\cdot v$, $\lvert u\rvert$, $\angle(u, v)$, $u \perp v$ | dot product, length, angle, orthogonality |
| $W^\perp$, $P_W$ | orthogonal complement, orthogonal projection onto $W$ |
| $A^{\mathsf T}$, $Q$, $\Lambda$, $\lambda_{\min}, \lambda_{\max}$ | transpose, orthogonal matrix, diagonal matrix of eigenvalues, extreme eigenvalues |
| $\lVert M\rVert = (\sum m_{ij}^2)^{1/2}$ | matrix norm, $\lvert Mh\rvert \le \lVert M\rVert\,\lvert h\rvert$ |
| $u\times v$, $[u\ v\ w]$ | cross product, matrix with columns $u, v, w$ |
| $\gamma(t)$, $\gamma'(t)$, $s$, $L(\gamma)$ | curve, velocity, arc length, length |
| $T, N, B$; $\kappa$, $\tau$; $\theta$ | Frenet frame; curvature, torsion; turning angle (plane curves) |
| $\Gamma_f$, $f^{-1}(c)$ | graph, level set |
| $\frac{\partial f}{\partial x_j} = f_{x_j} = \partial_jf$, $D_vf(a)$ | partial derivative, derivative along $v$ |
| $Df(a)$, $\frac{\partial(y_1, \dots, y_m)}{\partial(x_1, \dots, x_n)}$ | derivative = Jacobian matrix |
| $\nabla f$, $Hf(a)$, $(h\cdot\partial)^kf$ | gradient, Hessian, $k$-th order term of Taylor's formula |
| $C^k$; $T_p$ | $k$ times continuously differentiable; Taylor polynomial of degree $p$ |
| $\mathcal{L}$, $\mu_i$ (or $\lambda_i$) | Lagrangian, Lagrange multipliers |
| $R$, $Q$, $\mathcal{P}$, $\operatorname{vol}$ | box, subbox, partition, volume |
| $L(f, \mathcal{P})$, $U(f, \mathcal{P})$, $\int_Rf$, $\tilde f$ | lower and upper sums, integral, extension by $0$ |
| $dA$, $dV$; $r, \theta$; $\rho, \varphi, \theta$ | area and volume elements; polar; spherical coordinates |
| $\varphi(u, v)$, $\mathbf r_u, \mathbf r_v$, $\mathbf n$, $dS$, $E, F, G$ | parametrized surface, tangent vectors, unit normal, area element, first fundamental form |
| $\int_\gamma f\,ds$, $\iint_\varphi f\,dS$ | integrals of functions over curves and surfaces |
| $F$, $\int_\gamma F\cdot d\mathbf r$, $\iint_\varphi F\cdot d\mathbf S$ | vector field, line integral, flux |
| $\operatorname{div}F = \nabla\cdot F$, $\operatorname{curl}F = \nabla\times F$, $\Delta$ | divergence, curl, Laplacian |
| $\partial D$, $I_\gamma(p)$ | boundary (with its positive orientation), winding number |
