# The Chain Rule and Derivatives of Inverse Functions

**Prerequisites:** [The Derivative](derivative.ipynb#thm-caratheodory) · [Rules of Differentiation](differentiation-rules.ipynb) · [Local Extrema and the Mean Value Theorem](mean-value-theorem.ipynb) · [Monotone Functions and Continuous Inverses](monotone-and-inverse-functions.ipynb)

## 1\. The Chain Rule

::: {#thm-chain-rule}
## Chain Rule

Let $f : S \to \mathbb{R}$, $g : T \to \mathbb{R}$, $a$ interior to $S$, $b = f(a)$ interior to $T$. If $f$ is differentiable at $a$ and $g$ at $b$, then $a$ is interior to the domain $S' = \{x \in S : f(x) \in T\}$ of $g \circ f$, and
$$(g \circ f)'(a) = g'(f(a))\, f'(a).$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-chain-rule
*$a$ is interior to $S'$.* Take $\varepsilon$ with $(b - \varepsilon, b + \varepsilon) \subseteq T$. $f$ is [continuous](derivative.ipynb#thm-differentiable-continuous) at $a$, so for $|x - a| < \delta$ (small enough that also $x \in S$) we have $|f(x) - b| < \varepsilon$, i.e. $f(x) \in T$.

*The derivative.* By [the error-term form](derivative.ipynb#thm-caratheodory):
$$f(a + h) = f(a) + \big(f'(a) + \varphi(h)\big)h, \qquad g(b + k) = g(b) + \big(g'(b) + \psi(k)\big)k,$$
with $\varphi(h) \to 0$, $\psi(k) \to 0$. Define $\psi(0) = 0$; the second identity then holds also for $k = 0$, and $\psi$ is continuous at $0$. Put $k(h) = f(a + h) - f(a) = (f'(a) + \varphi(h))h$, so $k(h) \to 0$. Then
$$g(f(a + h)) - g(f(a)) = \big(g'(b) + \psi(k(h))\big)\big(f'(a) + \varphi(h)\big)\,h .$$
Divide by $h \ne 0$ and let $h \to 0$: $\psi(k(h)) \to \psi(0) = 0$ by continuity of $\psi$ at $0$ ([composition](function-continuity.ipynb#thm-continuity-composition)), so the quotient tends to $g'(b)f'(a)$. $\blacksquare$
:::

::: {.callout-warning}
## Why $\psi(0) = 0$ Is Needed
The tempting proof $\frac{\Delta z}{\Delta x} = \frac{\Delta z}{\Delta y}\frac{\Delta y}{\Delta x}$ fails when $\Delta y = f(a+h) - f(a) = 0$ for $h$ arbitrarily small (e.g. $f(x) = x^2\sin\frac1x$ near $0$). The function $\psi$ with $\psi(0) = 0$ avoids dividing by $\Delta y$.
:::

**Leibniz notation.** For $y = f(x)$ one writes $\frac{dy}{dx} = f'(x)$, and $\frac{dy}{dx}\big|_a = f'(a)$. The **differential** $dy = f'(a)\,dx$ is the increment of the linear approximation, as opposed to the true increment $\Delta y$; $\frac{dy}{dx}$ is a limit, not a quotient. With $y = f(x)$, $z = g(y)$, the chain rule reads
$$\frac{dz}{dx} = \frac{dz}{dy}\cdot\frac{dy}{dx},$$
where $\frac{dz}{dy}$ is evaluated at $y = f(a)$ and $\frac{dy}{dx}$ at $a$.

::: {#exm-chain-rule}
## Examples

1. $\big((x^2 + x + 1)^{10}\big)' = 10(x^2 + x + 1)^9(2x + 1)$.
2. $\big(\sin(\cos x)\big)' = -\cos(\cos x)\sin x$.
3. If $f(0) = 0$ and $f'(0) = 2$, then $(f \circ f)'(0) = f'(f(0))f'(0) = 4$.
:::

::: {#exm-related-rates}
## Related Rates

1. *A conical tank* (vertex down, height $20$ cm, top radius $10$ cm) is filled at $2\ \text{cm}^3/\text{s}$. At water height $h$ the radius is $r = h/2$, so $V = \frac13\pi r^2 h = \frac{\pi}{12}h^3$. Then $\frac{dV}{dt} = \frac{\pi}{4}h^2\frac{dh}{dt}$; at $h = 6$: $2 = 9\pi\frac{dh}{dt}$, so $\frac{dh}{dt} = \frac{2}{9\pi}$ cm/s.
2. *A rotating light* at distance $4$ m from a wall turns at $\frac12$ rad/s. With $\theta$ the angle between the beam and the wall, the lit point is at $x = 4\cot\theta$ along the wall, so $\frac{dx}{dt} = -\frac{4}{\sin^2\theta}\frac{d\theta}{dt}$. When the lit point is $5$ m from the light, $\sin\theta = \frac45$ and the speed is $\frac{4}{16/25}\cdot\frac12 = 3.125$ m/s.
:::

## 2\. Inverse Functions

::: {#thm-inverse-derivative}
## Derivative of an Inverse Function

Let $I$ be an open interval and $f : I \to \mathbb{R}$ differentiable with $f' > 0$ on $I$ (or $f' < 0$ on $I$). Then $J = f(I)$ is an open interval, $f^{-1} : J \to I$ is differentiable, and for $b = f(a)$
$$(f^{-1})'(b) = \frac{1}{f'(a)} = \frac{1}{f'(f^{-1}(b))}.$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-inverse-derivative
$f$ is strictly monotone ([MVT](mean-value-theorem.ipynb#cor-mvt-consequences)) and continuous, so $J$ is an interval and $f^{-1}$ is [continuous](monotone-and-inverse-functions.ipynb#thm-continuous-inverse); $J$ is open because a strictly monotone $f$ on an open interval attains no largest or smallest value. For $k \ne 0$ small, let $h(k) = f^{-1}(b + k) - a$; then $h(k) \ne 0$ (injectivity), $h(k) \to 0$ as $k \to 0$ (continuity), and $f(a + h(k)) = b + k$. Hence
$$\frac{f^{-1}(b + k) - f^{-1}(b)}{k} = \frac{h(k)}{f(a + h(k)) - f(a)} \to \frac{1}{f'(a)} .\ \blacksquare$$
:::

Once differentiability is known, the formula also follows by differentiating $f(f^{-1}(y)) = y$ with the chain rule.

::: {#exm-inverse-derivatives}
## Roots and Inverse Trigonometric Functions

1. $(x^{1/n})' = \frac1n x^{\frac1n - 1}$ for $x > 0$: $f(x) = x^n$ has $f'(x) = nx^{n-1} > 0$ on $(0, \infty)$, and $\frac{1}{n(x^{1/n})^{n-1}} = \frac1n x^{\frac1n - 1}$. By the chain rule, $(x^{m/n})' = \big((x^{1/n})^m\big)' = \frac mn x^{\frac mn - 1}$: so $(x^r)' = rx^{r-1}$ for every rational $r$.
2. $\arcsin' x = \dfrac{1}{\sqrt{1 - x^2}}$ on $(-1, 1)$: $\sin' = \cos > 0$ on $(-\frac\pi2, \frac\pi2)$, and $\cos(\arcsin x) = \sqrt{1 - x^2}$ because $\cos \ge 0$ on $[-\frac\pi2, \frac\pi2]$. From $\arcsin + \arccos = \frac\pi2$ ([here](monotone-and-inverse-functions.ipynb#prp-inverse-trig-identities)), $\arccos' x = -\dfrac{1}{\sqrt{1 - x^2}}$.
3. $\arctan' x = \dfrac{1}{1 + x^2}$ on $\mathbb{R}$: $\tan' = 1 + \tan^2 > 0$ and $1 + \tan^2(\arctan x) = 1 + x^2$. Hence $\operatorname{arccot}' x = -\dfrac{1}{1 + x^2}$.

At $x = \pm1$, $\arcsin$ is not differentiable: there $\sin' = \cos$ vanishes, and the graph has a vertical tangent.
:::
