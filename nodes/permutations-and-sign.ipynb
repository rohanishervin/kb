# Permutations and the Sign of a Permutation

**Prerequisites:** [Functions and Bijections](def-functions-mappings.ipynb)

To compute an alternating form we will reorder its arguments, and every reordering costs a sign. This note sets up the bookkeeping: permutations, their decomposition into swaps, and the **sign** $\operatorname{sgn}\sigma = \pm 1$, which records whether a permutation needs an even or an odd number of swaps.

## 1\. Permutations and Transpositions

::: {#def-permutations}
## Permutations and Transpositions

A **permutation** of $\{1, \dots, n\}$ is a [bijection](def-functions-mappings.ipynb) $\sigma : \{1, \dots, n\} \to \{1, \dots, n\}$. The set of all $n!$ permutations is the **symmetric group** $S_n$; its operation is composition $\sigma\tau = \sigma \circ \tau$ and its neutral element is the identity $\mathrm{id}$. We often write $\sigma$ in **one-line notation** $[\sigma(1), \sigma(2), \dots, \sigma(n)]$.

A **transposition** $\tau = (a\ b)$, $a \ne b$, swaps $a$ and $b$ and fixes everything else. Note $\tau^{-1} = \tau$.
:::

::: {#thm-transposition-decomposition}
## Every Permutation Is a Product of Transpositions

Every $\sigma \in S_n$ is a composite $\sigma = \tau_1 \tau_2 \cdots \tau_k$ of transpositions (for $\sigma = \mathrm{id}$, take $k = 0$).
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-transposition-decomposition
Induction on $n$; for $n = 1$ only $\mathrm{id}$ exists. Let $\sigma \in S_{n+1}$. If $\sigma(n + 1) = n + 1$, then $\sigma$ permutes $\{1, \dots, n\}$ and the induction hypothesis applies. Otherwise let $j = \sigma(n + 1)$ and $\tau = (j\ \ n{+}1)$. Then $\tau\sigma$ fixes $n + 1$, so $\tau\sigma = \tau_1 \cdots \tau_m$ by induction, and $\sigma = \tau\tau_1 \cdots \tau_m$ because $\tau\tau = \mathrm{id}$. $\blacksquare$
:::

::: {#exm-permutation-decomposition}
## Decomposing a Permutation

Let $\sigma = [3, 1, 2] \in S_3$, i.e. $1 \mapsto 3$, $2 \mapsto 1$, $3 \mapsto 2$. Following the proof: $\sigma(3) = 2$, so take $\tau = (2\ 3)$; then $\tau\sigma = [2, 1, 3] = (1\ 2)$. Hence $\sigma = (2\ 3)(1\ 2)$. Check: $(2\ 3)(1\ 2)$ sends $1 \mapsto 2 \mapsto 3$, $2 \mapsto 1 \mapsto 1$, $3 \mapsto 3 \mapsto 2$. The decomposition is not unique, e.g. also $\sigma = (1\ 3)(2\ 3)$, but both use an even number of transpositions.
:::

## 2\. The Sign

::: {#def-sign}
## Sign of a Permutation

The **sign** of $\sigma \in S_n$ is
$$\operatorname{sgn}\sigma = \prod_{1 \le i < j \le n} \frac{\sigma(j) - \sigma(i)}{j - i}.$$
A pair $i < j$ with $\sigma(i) > \sigma(j)$ is an **inversion** of $\sigma$.
:::

::: {#prp-sign-inversions}
## The Sign Counts Inversions

$\operatorname{sgn}\sigma = (-1)^{N(\sigma)}$, where $N(\sigma)$ is the number of inversions. In particular $\operatorname{sgn}\sigma \in \{1, -1\}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-sign-inversions
As $\{i, j\}$ runs over all $2$-element subsets of $\{1, \dots, n\}$, so does $\{\sigma(i), \sigma(j)\}$, because $\sigma$ is a bijection. Hence $\prod_{i<j} |\sigma(j) - \sigma(i)| = \prod_{i<j} (j - i)$, and the sign of the product is the sign of $\prod_{i<j}(\sigma(j) - \sigma(i))$, which has one negative factor per inversion. $\blacksquare$
:::

::: {#prp-sign-multiplicative}
## The Sign Is Multiplicative

$\operatorname{sgn}(\sigma\tau) = \operatorname{sgn}\sigma \cdot \operatorname{sgn}\tau$ for all $\sigma, \tau \in S_n$, and $\operatorname{sgn}(\mathrm{id}) = 1$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-sign-multiplicative
The factor $\frac{\sigma(j) - \sigma(i)}{j - i}$ does not change when $i$ and $j$ are exchanged, so $\operatorname{sgn}\sigma = \prod_{\{a, b\}} \frac{\sigma(b) - \sigma(a)}{b - a}$ is a product over **unordered** pairs $\{a, b\}$, each taken once in either order. Now write
$$\operatorname{sgn}(\sigma\tau) = \prod_{i<j} \frac{\sigma(\tau(j)) - \sigma(\tau(i))}{\tau(j) - \tau(i)} \cdot \prod_{i<j} \frac{\tau(j) - \tau(i)}{j - i}.$$
The second product is $\operatorname{sgn}\tau$. In the first, $\{\tau(i), \tau(j)\}$ runs over all unordered pairs exactly once, so the first product is $\prod_{\{a, b\}} \frac{\sigma(b) - \sigma(a)}{b - a} = \operatorname{sgn}\sigma$. For $\mathrm{id}$ every factor is $1$. $\blacksquare$
:::

::: {#prp-sign-transposition}
## Transpositions Are Odd

$\operatorname{sgn}\tau = -1$ for every transposition $\tau$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-sign-transposition
For $\tau_0 = (1\ 2)$ the only inversion is the pair $(1, 2)$, so $\operatorname{sgn}\tau_0 = -1$ by @prp-sign-inversions. A general transposition $(a\ b)$ equals $\pi\tau_0\pi^{-1}$ for any permutation $\pi$ with $\pi(1) = a$, $\pi(2) = b$. By @prp-sign-multiplicative, $\operatorname{sgn}(a\ b) = \operatorname{sgn}\pi \cdot (-1) \cdot \operatorname{sgn}\pi^{-1} = -\operatorname{sgn}(\pi\pi^{-1}) = -1$. $\blacksquare$
:::

::: {#thm-parity}
## Parity Is Well Defined

If $\sigma = \tau_1 \cdots \tau_k$ is any decomposition into transpositions, then $\operatorname{sgn}\sigma = (-1)^k$. Consequently all decompositions of $\sigma$ have a number of factors of the same parity. $\sigma$ is called **even** if $\operatorname{sgn}\sigma = 1$ and **odd** if $\operatorname{sgn}\sigma = -1$. Moreover $\operatorname{sgn}(\sigma^{-1}) = \operatorname{sgn}\sigma$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-parity
By @prp-sign-multiplicative and @prp-sign-transposition, $\operatorname{sgn}\sigma = \prod_i \operatorname{sgn}\tau_i = (-1)^k$. If also $\sigma = \tau_1' \cdots \tau_l'$, then $(-1)^k = (-1)^l$, so $k \equiv l \pmod 2$. Finally $\operatorname{sgn}\sigma \cdot \operatorname{sgn}\sigma^{-1} = \operatorname{sgn}(\mathrm{id}) = 1$, and both factors are $\pm 1$. $\blacksquare$
:::

::: {#cor-alternating-group-partition}
## Half the Permutations Are Even

For $n \ge 2$, exactly $n!/2$ permutations in $S_n$ are even and $n!/2$ are odd.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-alternating-group-partition
Fix the transposition $\tau = (1\ 2)$. The map $\sigma \mapsto \tau\sigma$ is a bijection of $S_n$ (it is its own inverse) and reverses the sign by @prp-sign-multiplicative. So it maps the even permutations bijectively onto the odd ones, and the two sets have the same size $n!/2$. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Counting Crossings
Draw $\sigma$ as a **wiring diagram**: a wire from position $i$ on the left to position $\sigma(i)$ on the right. Two wires cross exactly when they form an inversion, so the sign is $(-1)^{\text{number of crossings}}$. A transposition of neighbours adds or removes exactly one crossing, which is why each swap flips the sign.
:::


```python

import numpy as np
import matplotlib.pyplot as plt
from itertools import permutations

BLUE, ORANGE, AQUA, YELLOW, VIOLET = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6"
INK, MUTED = "#0b0b0b", "#52514e"
plt.rcParams.update({"axes.titlesize": 11, "axes.titleweight": "bold", "font.size": 10})
colors = [BLUE, ORANGE, AQUA, YELLOW, VIOLET]

def inversions(p):
    return [(i, j) for i in range(len(p)) for j in range(i + 1, len(p)) if p[i] > p[j]]

def wiring(ax, p, title):
    n = len(p)
    for i in range(n):
        ax.plot([0, 1], [n - i, n - p[i]], color=colors[i], lw=2.5)
        ax.text(-0.06, n - i, str(i + 1), ha="right", va="center", color=INK)
        ax.text(1.06, n - i, str(i + 1), ha="left", va="center", color=INK)
    inv = inversions(p)
    ax.set_title(f"{title}\n{len(inv)} crossings, sgn = {(-1) ** len(inv):+d}")
    ax.set_xlim(-0.25, 1.25); ax.set_ylim(0.5, n + 0.5); ax.axis("off")

fig, axes = plt.subplots(1, 3, figsize=(12, 3.6))
wiring(axes[0], [2, 0, 1], "σ = [3, 1, 2]")          # zero-based one-line notation
wiring(axes[1], [1, 0, 2], "τ = (1 2)")
wiring(axes[2], [4, 2, 0, 3, 1], "σ = [5, 3, 1, 4, 2]")
plt.tight_layout(); plt.show()

# sign via the product formula agrees with (-1)^(number of inversions) for all of S_5
def sgn_product(p):
    num = np.prod([p[j] - p[i] for i in range(len(p)) for j in range(i + 1, len(p))], dtype=float)
    den = np.prod([j - i for i in range(len(p)) for j in range(i + 1, len(p))], dtype=float)
    return round(num / den)
perms = list(permutations(range(5)))
print("product formula == (-1)^inversions on S_5:",
      all(sgn_product(p) == (-1) ** len(inversions(p)) for p in perms))
print("even permutations in S_5:", sum(sgn_product(p) == 1 for p in perms), "of", len(perms))
```
