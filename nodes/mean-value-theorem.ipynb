# Local Extrema and the Mean Value Theorem

**Prerequisites:** [Rules of Differentiation](differentiation-rules.ipynb) · [Semicontinuity and the Extreme Value Theorem](semicontinuity-and-extreme-value-theorem.ipynb#cor-evt)

## 1\. The Sign of the Derivative at a Point

::: {#prp-derivative-sign}
## Positive Derivative at a Point

If $f'(x_0) > 0$, there is $\delta > 0$ with
$$f(x) < f(x_0) \ \text{ for } x_0 - \delta < x < x_0, \qquad f(x) > f(x_0) \ \text{ for } x_0 < x < x_0 + \delta .$$
If $f'(x_0) < 0$ the inequalities are reversed.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-derivative-sign
With $\varepsilon = f'(x_0)$ in the definition of the limit, for $0 < |x - x_0| < \delta$ we get $\frac{f(x) - f(x_0)}{x - x_0} > 0$: numerator and denominator have the same sign. $\blacksquare$
:::

::: {.callout-warning}
## Not Increasing Near the Point
$f'(x_0) > 0$ only compares $f(x)$ with $f(x_0)$; it does **not** make $f$ increasing on any interval around $x_0$. See [this example](convexity-and-curve-sketching.ipynb#exm-positive-derivative-not-increasing): $\frac{x}{2} + x^2\sin\frac1x$.
:::

::: {#def-local-extremum}
## Local Extrema

$x_0 \in S$ is a **local maximum** of $f : S \to \mathbb{R}$ if $f(x) \le f(x_0)$ for all $x \in S$ with $|x - x_0| < \delta$, for some $\delta > 0$; similarly **local minimum**.
:::

::: {#thm-fermat}
## Fermat's Theorem

If $f$ has a local maximum or minimum at an interior point $x_0$ and $f$ is differentiable at $x_0$, then $f'(x_0) = 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-fermat
If $f'(x_0) \ne 0$, @prp-derivative-sign shows that $f$ takes values both larger and smaller than $f(x_0)$ arbitrarily close to $x_0$. $\blacksquare$
:::

Points with $f'(x_0) = 0$ are **critical points**. A critical point need not be an extremum: $x^3$ at $0$; or $x - \sin x$, whose derivative $1 - \cos x$ vanishes at $2k\pi$, but which is increasing (@exm-increment-estimates).

::: {#exm-quartic-min}
## Locating a Minimum

$f(x) = x^4 - x^3 = x^3(x - 1)$ has $f'(x) = x^2(4x - 3)$, so the critical points are $0$ and $\frac34$. Since $f(0) = f(1) = 0$ and $f < 0$ on $(0, 1)$, the minimum of $f$ on $[0, 1]$ ([it exists](semicontinuity-and-extreme-value-theorem.ipynb#cor-evt)) is attained at an interior point, which must be critical, i.e. at $\frac34$: $f(\frac34) = -\frac{27}{256}$. The critical point $0$ is not an extremum ($x^3(x - 1)$ changes sign there).
:::

## 2\. Rolle and the Mean Value Theorem

::: {#thm-rolle}
## Rolle's Theorem

Let $f : [a, b] \to \mathbb{R}$ be continuous, differentiable on $(a, b)$, with $f(a) = f(b)$. Then $f'(c) = 0$ for some $c \in (a, b)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-rolle
By the [extreme value theorem](semicontinuity-and-extreme-value-theorem.ipynb#cor-evt) $f$ has a maximum and a minimum on $[a, b]$. If both are attained only at the endpoints, then $\max f = \min f = f(a)$ and $f$ is constant, so $f' = 0$ everywhere. Otherwise one of them is attained at an interior $c$, and $f'(c) = 0$ by @thm-fermat. $\blacksquare$
:::

::: {#thm-mvt}
## Mean Value Theorem

Let $f : [a, b] \to \mathbb{R}$ be continuous and differentiable on $(a, b)$. Then for some $c \in (a, b)$
$$f'(c) = \frac{f(b) - f(a)}{b - a} :$$
some tangent is parallel to the chord.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-mvt
Apply Rolle to $g(x) = f(x) - \Big[f(a) + \dfrac{f(b) - f(a)}{b - a}(x - a)\Big]$, the vertical distance from the chord: $g(a) = g(b) = 0$, and $g'(c) = 0$ is the claim. $\blacksquare$
:::

::: {#cor-mvt-consequences}
## Consequences on an Interval

Let $I$ be an interval and $f : I \to \mathbb{R}$ continuous, differentiable at the interior points of $I$.

1. If $f' = 0$ on the interior, $f$ is constant. If $f' = g'$, then $f - g$ is constant.
2. If $f' > 0$ (resp. $\ge 0$), $f$ is strictly increasing (resp. increasing); if $f' < 0$ (resp. $\le 0$), strictly decreasing (resp. decreasing).
3. **Increment estimate:** if $|f'| \le M$, then $|f(b) - f(a)| \le M|b - a|$ for all $a, b \in I$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-mvt-consequences
For $a < b$ in $I$, $f(b) - f(a) = f'(c)(b - a)$ with $c$ interior. All three statements read off the sign or size of $f'(c)$. For $f - g$ apply (1) to $f - g$. $\blacksquare$
:::

::: {#exm-increment-estimates}
## Increment Estimates

1. $|\sin\alpha - \sin\beta| \le |\alpha - \beta|$, since $|\sin'| = |\cos| \le 1$.
2. $\Big|\dfrac{1}{1 + a} - \dfrac{1}{1 + b}\Big| \le |a - b|$ for $a, b > 0$, since $\big|\frac{d}{dx}\frac{1}{1+x}\big| = \frac{1}{(1+x)^2} < 1$ for $x > 0$.
3. $x - \sin x$ is strictly increasing: $1 - \cos x \ge 0$, with equality only at isolated points, so on each $[2k\pi, 2(k+1)\pi]$ it is strictly increasing (the derivative is positive in the interior).
:::

## 3\. Cauchy's Mean Value Theorem and l'Hôpital's Rule

::: {#thm-cauchy-mvt}
## Cauchy's Mean Value Theorem

Let $f, g$ be continuous on $[a, b]$ and differentiable on $(a, b)$. Then for some $c \in (a, b)$
$$\big(f(b) - f(a)\big)\,g'(c) = \big(g(b) - g(a)\big)\,f'(c).$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-cauchy-mvt
Apply Rolle to $h(x) = \big(f(b) - f(a)\big)g(x) - \big(g(b) - g(a)\big)f(x)$; $h(a) = f(b)g(a) - g(b)f(a) = h(b)$. $\blacksquare$
:::

::: {#thm-lhopital}
## l'Hôpital's Rule ($0/0$)

Let $f, g$ be differentiable on $(a, a + r)$ with $g' \ne 0$ there, $\lim_{x\to a^+} f(x) = \lim_{x \to a^+} g(x) = 0$, and $\lim_{x\to a^+}\frac{f'(x)}{g'(x)} = L$. Then $\lim_{x\to a^+}\frac{f(x)}{g(x)} = L$. The same holds for $x \to a^-$, hence for $x \to a$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-lhopital
Set $f(a) = g(a) = 0$; then $f, g$ are continuous on $[a, x]$ for $x \in (a, a + r)$. By Rolle, $g(x) = g(x) - g(a) \ne 0$ (as $g' \ne 0$). By @thm-cauchy-mvt there is $c_x \in (a, x)$ with $\frac{f(x)}{g(x)} = \frac{f'(c_x)}{g'(c_x)}$. As $x \to a^+$, $c_x \to a^+$, so the quotient tends to $L$. $\blacksquare$
:::

For example $\lim_{x\to0}\frac{x - \sin x}{x^3} = \lim \frac{1 - \cos x}{3x^2} = \lim\frac{\sin x}{6x} = \frac16$ (the last limit exists, so by the rule the middle one exists, and then the first). Taylor polynomials usually do such limits faster ([examples](binomial-series.ipynb#exm-limits-by-series)).

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

f = lambda x: x**3 - 3 * x
fp = lambda x: 3 * x**2 - 3
a, b = -2.0, 2.2
slope = (f(b) - f(a)) / (b - a)
xs = np.linspace(a, b, 200001)
cs = xs[np.where(np.diff(np.sign(fp(xs) - slope)))[0]]      # points with f'(c) = slope
x = np.linspace(-2.3, 2.45, 400)
fig, ax = plt.subplots(figsize=(8, 4.5))
ax.plot(x, f(x), color=INK, lw=2, label="$y = f(x)$")
ax.plot([a, b], [f(a), f(b)], color=ORANGE, lw=2, label=f"chord, slope {slope:.3f}")
for c in cs:
    ax.plot(x, f(c) + slope * (x - c), color=BLUE, lw=1.3, ls="--")
    ax.plot(c, f(c), "o", color=YELLOW, ms=9, mec=INK)
ax.plot([], [], color=BLUE, ls="--", label=r"tangents with $f'(c) =$ chord slope")
ax.legend(fontsize=9, loc="upper left"); ax.set_ylim(-4, 12)
ax.set_title("mean value theorem")
plt.tight_layout(); plt.show()
print("points c:", np.round(cs, 4))
```
