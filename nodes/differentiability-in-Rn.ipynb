# Differentiable Maps and the Jacobian Matrix

**Prerequisites:** [Partial and Directional Derivatives](partial-and-directional-derivatives.ipynb) · [The Derivative](derivative.ipynb#thm-linear-approximation-char) · [Local Extrema and the Mean Value Theorem](mean-value-theorem.ipynb#thm-mvt) · [The Matrix of a Linear Map](matrix-of-a-linear-map.ipynb)

In one variable, $f$ is differentiable at $a$ iff it is [tangent to an affine function](derivative.ipynb#thm-linear-approximation-char) there. This characterization, unlike the difference quotient, makes sense in any dimension. Throughout, $f : S \to \mathbb{R}^m$, $S \subseteq \mathbb{R}^n$, and $a$ is an interior point of $S$.

## 1\. The Derivative

::: {#def-differentiable-Rn}
## Differentiability

$f$ is **differentiable at $a$** if there is a linear map $L : \mathbb{R}^n \to \mathbb{R}^m$ with
$$\lim_{h\to0}\frac{|f(a + h) - f(a) - L(h)|}{|h|} = 0,$$
i.e. $f$ is tangent at $a$ to the affine map $A(x) = f(a) + L(x - a)$, the **linear approximation** of $f$ at $a$. Then $L$ is unique (@thm-jacobian), is called the **derivative** (differential) of $f$ at $a$, and is written $Df(a)$; we identify it with its $m\times n$ matrix.
:::

::: {#thm-jacobian}
## The Derivative Is the Jacobian Matrix

If $f$ is differentiable at $a$, then every directional derivative exists and
$$D_vf(a) = Df(a)\,v \quad\text{for all } v \in \mathbb{R}^n .$$
In particular the $j$-th column of $Df(a)$ is $\frac{\partial f}{\partial x_j}(a)$, so $Df(a)$ is the **Jacobian matrix**
$$Df(a) = \Big[\frac{\partial f_i}{\partial x_j}(a)\Big]_{i\le m,\ j\le n} = \begin{bmatrix}\frac{\partial f_1}{\partial x_1}(a) & \cdots & \frac{\partial f_1}{\partial x_n}(a)\\ \vdots & & \vdots\\ \frac{\partial f_m}{\partial x_1}(a) & \cdots & \frac{\partial f_m}{\partial x_n}(a)\end{bmatrix},$$
and $L$ in @def-differentiable-Rn is unique.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-jacobian
For $v = 0$ both sides vanish. For $v \ne 0$ put $h = tv$, $t \to 0$:
$$\Big|\frac{f(a + tv) - f(a)}{t} - L(v)\Big| = |v|\cdot\frac{|f(a + tv) - f(a) - L(tv)|}{|tv|} \to 0 .$$
So $D_vf(a) = L(v)$; any $L$ satisfying the definition is determined by its values $L(e_j) = \frac{\partial f}{\partial x_j}(a)$, which are its columns. $\blacksquare$
:::

::: {#prp-differentiable-basics}
## Basic Facts

1. $f$ is differentiable at $a$ iff each component $f_i$ is; the $i$-th row of $Df(a)$ is $Df_i(a)$.
2. If $f$ is differentiable at $a$, it is continuous at $a$.
3. An affine map $f(x) = Mx + b$ is differentiable everywhere with $Df(a) = M$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-differentiable-basics
(1) [Lengths versus components](parametrized-curves.ipynb#lem-norm-components) applied to the vector $f(a + h) - f(a) - L(h)$. (2) $|f(a + h) - f(a)| \le |f(a + h) - f(a) - Df(a)h| + |Df(a)h|$; the first term is $o(|h|)$ and the second tends to $0$ as $h \to 0$ because linear maps are [continuous](limits-and-continuity-in-Rn.ipynb#prp-basic-continuous). (3) The error is identically $0$. $\blacksquare$
:::

::: {#exm-tangent-plane}
## Tangent Planes to Graphs

For $f(x, y) = x^2 - y^2$ (differentiable by @thm-c1-differentiable below) at $(2, 1)$: $Df(2, 1) = [4\ \ {-2}]$, so the linear approximation is $A(x, y) = 3 + 4(x - 2) - 2(y - 1)$ and the tangent plane to the graph at $(2, 1, 3)$ is $4x - 2y - z = 3$. In general the tangent hyperplane to the graph of a differentiable $f : \mathbb{R}^n \to \mathbb{R}$ at $(a, f(a))$ is
$$x_{n+1} = f(a) + \sum_{i=1}^n\frac{\partial f}{\partial x_i}(a)(x_i - a_i),$$
the analogue of $y = f(a) + f'(a)(x - a)$.
:::

::: {#exm-not-differentiable}
## Existence of Partials Does Not Imply Differentiability

For $f = \frac{x^2y}{x^2 + y^2}$, $f(0) = 0$, all [directional derivatives](partial-and-directional-derivatives.ipynb#exm-all-directions-not-linear) at $0$ exist, but $u \mapsto D_uf(0)$ is not linear, so by @thm-jacobian $f$ is not differentiable at $0$. Directly: the only candidate is $L = [f_x(0)\ \ f_y(0)] = [0\ \ 0]$, and $\frac{|f(h)|}{|h|} = |\cos^2\theta\sin\theta|$ in polar coordinates, which does not tend to $0$. The function of [this example](partial-and-directional-derivatives.ipynb#exm-directional-not-continuous) is not even continuous, so by @prp-differentiable-basics it is not differentiable either.
:::

## 2\. A Practical Criterion

::: {#thm-c1-differentiable}
## Continuous Partials Imply Differentiability

If all partial derivatives $\frac{\partial f}{\partial x_j}$ exist in a ball about $a$ and are continuous at $a$, then $f$ is differentiable at $a$. (In fact continuity of $n - 1$ of them suffices.)
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-c1-differentiable
By @prp-differentiable-basics we may take $f$ real-valued. Let $h = (h_1, \dots, h_n)$ be small, $p_0 = a$, $p_k = a + h_1e_1 + \dots + h_ke_k$ ($p_n = a + h$); all $p_k$ and the segments between them lie in the ball. Telescoping,
$$f(a + h) - f(a) = \sum_{k=1}^n\big[f(p_k) - f(p_{k-1})\big].$$
Only the $k$-th coordinate changes from $p_{k-1}$ to $p_k$, so by the MVT in that variable, $f(p_k) - f(p_{k-1}) = h_k\,\partial_kf(q_k)$ with $q_k$ on the segment $[p_{k-1}, p_k]$. Hence
$$\Big|f(a + h) - f(a) - \sum_kh_k\partial_kf(a)\Big| \le \sum_k|h_k|\,|\partial_kf(q_k) - \partial_kf(a)| \le |h|\sum_k|\partial_kf(q_k) - \partial_kf(a)| .$$
As $h \to 0$, $q_k \to a$, so by continuity the last sum tends to $0$. (For $k = 1$, $f(p_1) - f(a) = h_1\partial_1f(a) + o(|h_1|)$ by the mere existence of $\partial_1f(a)$, which gives the parenthetical remark.) $\blacksquare$
:::

::: {#def-c1-map}
## $C^1$ Maps

$f$ is **of class $C^1$** on an open set $U$ if all partial derivatives exist and are continuous on $U$. By @thm-c1-differentiable, $C^1$ maps are differentiable at every point of $U$; polynomials, rational functions (off their poles) and their compositions with $\exp$, $\sin$, etc. are $C^1$.
:::

In @exm-not-differentiable the partials cannot be continuous at $0$: e.g. $f_y = \cos^2\theta(\cos^2\theta - \sin^2\theta)$ on the ray of angle $\theta$, which does not tend to $f_y(0) = 0$.

## 3\. Linear Approximation

For $h$ small, $f(a + h) \approx f(a) + Df(a)h$, with an error that is $o(|h|)$.

::: {#exm-linear-approx-2var}
## A Numerical Approximation

Estimate $\sqrt[3]{2.9^2 - 1.1}$ with $f(x, y) = (x^2 - y)^{1/3}$ near $(3, 1)$, where $f = 2$ and $f$ is $C^1$ (since $x^2 - y \ne 0$):
$$f_x = \tfrac{2x}{3}(x^2 - y)^{-2/3} = \tfrac12, \qquad f_y = -\tfrac13(x^2 - y)^{-2/3} = -\tfrac{1}{12}\quad\text{at } (3, 1),$$
so $f(2.9, 1.1) \approx 2 + \frac12(-0.1) - \frac1{12}(0.1) = 1.94167$; the true value is $1.94076\ldots$
:::

::: {#exm-relative-error}
## Propagation of Relative Errors

For $f(x) = cx_1^{\alpha_1}\cdots x_n^{\alpha_n}$ on $x_i > 0$ ($c, \alpha_i > 0$), $\frac{\partial f}{\partial x_i} = \frac{\alpha_i}{x_i}f$, so
$$\frac{\Delta f}{f} \approx \sum_{i=1}^n\alpha_i\frac{\Delta x_i}{x_i} :$$
relative errors $r_i$ in the inputs produce a relative error of about $\sum\alpha_ir_i$ in the output.
:::

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

f = lambda x, y: np.cbrt(x**2 - y)
A = lambda x, y: 2 + 0.5 * (x - 3) - (y - 1) / 12
print("f(2.9, 1.1) =", f(2.9, 1.1), "| linear approximation:", A(2.9, 1.1))
hs = 10.0 ** -np.arange(1, 6)
d = np.array([1.0, -2.0]) / np.sqrt(5)
print("error/|h| along a fixed direction:", [float(abs(f(3 + h * d[0], 1 + h * d[1]) - A(3 + h * d[0], 1 + h * d[1])) / h) for h in hs])

g = lambda x, y: np.where((x == 0) & (y == 0), 0.0, x**2 * y / np.where((x == 0) & (y == 0), 1, x**2 + y**2))
th = np.linspace(0, 2 * np.pi, 9)[:-1]
print("x^2y/(x^2+y^2): |f(h)|/|h| at |h| = 1e-6 along angles k*pi/4:", np.round(np.abs(g(1e-6 * np.cos(th), 1e-6 * np.sin(th))) / 1e-6, 4))

fig = plt.figure(figsize=(6.4, 5))
ax = fig.add_subplot(projection="3d")
x = np.linspace(1, 3, 30); y = np.linspace(0, 2, 30); X, Y = np.meshgrid(x, y)
ax.plot_surface(X, Y, X**2 - Y**2, color=BLUE, alpha=0.45, linewidth=0)
ax.plot_surface(X, Y, 3 + 4 * (X - 2) - 2 * (Y - 1), color=ORANGE, alpha=0.35, linewidth=0)
ax.scatter([2], [1], [3], color=INK, s=25)
ax.set_title(r"$z = x^2 - y^2$ and its tangent plane at $(2, 1, 3)$")
plt.tight_layout(); plt.show()
```
