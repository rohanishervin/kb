# Integrals of Functions over Curves and Surfaces

**Prerequisites:** [Parametrized Curves and Arc Length](parametrized-curves.ipynb) · [Parametrized Surfaces and Surface Area](parametrized-surfaces.ipynb) · [Change of Variables in Multiple Integrals](change-of-variables-multiple-integrals.ipynb)

So far integrals were taken over "flat" sets: intervals and regions of $\mathbb{R}^n$. Curves and surfaces are curved, and integrals over them use arc length and surface area as the measure. If $f > 0$ is a density on a wire or a thin shell, these integrals are its mass; their values do not depend on the parametrization or the orientation.

## 1\. Curves

::: {#def-scalar-line-integral}
## Integral with Respect to Arc Length

Let $\gamma : [a, b] \to \mathbb{R}^n$ be a $C^1$ curve and $f$ continuous on its image. Then
$$\int_\gamma f\,ds = \int_a^bf(\gamma(t))\,\frac{ds}{dt}\,dt = \int_a^bf(\gamma(t))\,|\gamma'(t)|\,dt .$$
A **piecewise smooth** curve is a continuous $\gamma$ with a partition $a = t_0 < \dots < t_p = b$ such that each restriction $\gamma_i = \gamma|_{[t_{i-1}, t_i]}$ is $C^1$ and smooth; then $\int_\gamma f\,ds = \sum_i\int_{\gamma_i}f\,ds$. At the points $\gamma(t_i)$ the curve may have corners.
:::

::: {#prp-scalar-line-invariance}
## Independence of the Parametrization

If $\tilde\gamma = \gamma\circ\alpha$ is a reparametrization ($\alpha : [c, d] \to [a, b]$ a $C^1$ bijection with $\alpha' \ne 0$, of either sign), then $\int_{\tilde\gamma}f\,ds = \int_\gamma f\,ds$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-scalar-line-invariance
$|\tilde\gamma'(\tau)| = |\alpha'(\tau)|\,|\gamma'(\alpha(\tau))|$, so substituting $t = \alpha(\tau)$ gives $\int_c^df(\gamma(\alpha(\tau)))\,|\gamma'(\alpha(\tau))|\,|\alpha'(\tau)|\,d\tau = \int_a^bf(\gamma(t))\,|\gamma'(t)|\,dt$; the absolute value of $\alpha'$ compensates for reversed limits when $\alpha' < 0$, exactly as in the [invariance of length](parametrized-curves.ipynb#prp-arc-length-parametrization). $\blacksquare$
:::

For $f = 1$ this is the length of $\gamma$. If $\gamma$ is an injective curve in a plane $P \subseteq \mathbb{R}^3$ and $f \ge 0$, then $\int_\gamma f\,ds$ is the area of the "fence" standing on $\gamma$, perpendicular to $P$, of height $f(\gamma(t))$ at $\gamma(t)$.

::: {#exm-fence-area}
## A Fence on a Circle

The part of the cylinder $x^2 + y^2 = 1$ inside the cylinder $x^2 + z^2 \le 1$ stands on the unit circle $\gamma(t) = (\cos t, \sin t)$ with height $2\sqrt{1 - x^2} = 2|\sin t|$ at $\gamma(t)$, so its area is
$$\int_\gamma2\sqrt{1 - x^2}\,ds = \int_0^{2\pi}2|\sin t|\,dt = 8 .$$
By symmetry the solid $x^2 + y^2 \le 1$, $x^2 + z^2 \le 1$ has surface area $2\cdot8 = 16$, as [computed](parametrized-surfaces.ipynb#exm-surface-areas) with a surface integral.
:::

::: {#exm-helix-mass}
## Mass of a Helical Wire

The helix $\gamma(t) = (a\cos t, a\sin t, bt)$, $0 \le t \le 2\pi$ ($b > 0$), with density $\delta = z$ has $|\gamma'| = \sqrt{a^2 + b^2}$ and mass $\int_0^{2\pi}bt\sqrt{a^2 + b^2}\,dt = 2\pi^2b\sqrt{a^2 + b^2}$.
:::

## 2\. Surfaces

::: {#def-scalar-surface-integral}
## Integral with Respect to Area

Let $\varphi : W \to \mathbb{R}^3$ be a smooth parametrized surface, $W$ an admissible region, and $f$ continuous on $\varphi(W)$. Then
$$\iint_\varphi f\,dS = \iint_Wf(\varphi(u, v))\,|\mathbf r_u\times\mathbf r_v|\,du\,dv .$$
:::

::: {#prp-scalar-surface-invariance}
## Independence of the Parametrization

If $\tau : W' \to W$ is a $C^1$ bijection with $\det D\tau \ne 0$, then $\iint_{\varphi\circ\tau}f\,dS = \iint_\varphi f\,dS$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-scalar-surface-invariance
By the [reparametrization rule](parametrized-surfaces.ipynb#prp-reparametrization-surface), $|\mathbf r_{u'}\times\mathbf r_{v'}| = |\mathbf r_u\times\mathbf r_v|\circ\tau\cdot|\det D\tau|$; the [change of variables formula](change-of-variables-multiple-integrals.ipynb#thm-change-of-variables) turns $\iint_{W'}(f\circ\varphi\circ\tau)\,|\mathbf r_u\times\mathbf r_v|\circ\tau\,|\det D\tau|$ into $\iint_W(f\circ\varphi)\,|\mathbf r_u\times\mathbf r_v|$. $\blacksquare$
:::

Surfaces made of finitely many smooth pieces meeting along curves (such as the boundary of a cube) are handled by adding the integrals over the pieces.

::: {#def-center-of-mass}
## Mass and Centre of Mass

For a wire $\gamma$ or a shell $\varphi$ with density $\delta > 0$, the mass is $m = \int_\gamma\delta\,ds$, resp. $m = \iint_\varphi\delta\,dS$, and the **centre of mass** $(\bar x, \bar y, \bar z)$ has coordinates
$$\bar x = \frac1m\int_\gamma x\,\delta\,ds \quad\text{resp.}\quad \bar x = \frac1m\iint_\varphi x\,\delta\,dS,$$
and similarly for $\bar y$, $\bar z$. For constant density, $\delta$ cancels and the result is the **centroid** of the curve or surface.
:::

::: {#exm-hemisphere-centroid}
## Centroid of a Hemisphere

For $x^2 + y^2 + z^2 = R^2$, $z \ge 0$, symmetry gives $\bar x = \bar y = 0$. With the [spherical parametrization](parametrized-surfaces.ipynb#exm-surfaces), $z = R\cos u$, $dS = R^2\sin u\,du\,dv$, $0 \le u \le \frac\pi2$:
$$\iint z\,dS = \int_0^{2\pi}\int_0^{\pi/2}R^3\cos u\sin u\,du\,dv = \pi R^3, \qquad \bar z = \frac{\pi R^3}{2\pi R^2} = \frac R2 .$$
:::

::: {#exm-sphere-symmetry}
## Using Symmetry

On the unit sphere $S$, $\iint_Sx^2\,dS = \iint_Sy^2\,dS = \iint_Sz^2\,dS$ by symmetry, and their sum is $\iint_S1\,dS = 4\pi$; so $\iint_Sz^2\,dS = \frac{4\pi}{3}$.
:::

```python

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

t, a, b, u, v, R = sp.symbols("t a b u v R", positive=True)
g = sp.Matrix([a * sp.cos(t), a * sp.sin(t), b * t])
speed = sp.sqrt(sp.simplify(g.diff(t).dot(g.diff(t))))
print("helix mass with density z:", sp.simplify(sp.integrate(b * t * speed, (t, 0, 2 * sp.pi))))
print("hemisphere: int z dS =", sp.integrate(R * sp.cos(u) * R**2 * sp.sin(u), (u, 0, sp.pi / 2), (v, 0, 2 * sp.pi)))
print("unit sphere: int z^2 dS =", sp.integrate(sp.cos(u)**2 * sp.sin(u), (u, 0, sp.pi), (v, 0, 2 * sp.pi)))
print("fence area:", sp.integrate(2 * sp.Abs(sp.sin(t)), (t, 0, 2 * sp.pi)))

fig = plt.figure(figsize=(6.4, 5))
ax = fig.add_subplot(projection="3d")
T, Hh = np.meshgrid(np.linspace(0, 2 * np.pi, 120), np.linspace(-1, 1, 30))
X, Y = np.cos(T), np.sin(T)
Z = Hh * np.sqrt(1 - X**2)
ax.plot_surface(X, Y, Z, color=ORANGE, alpha=0.55, linewidth=0)
ax.plot(np.cos(T[0]), np.sin(T[0]), 0 * T[0], color=INK, lw=1.5)
ax.set_box_aspect((1, 1, 0.8)); ax.set_title(r"The fence of height $2\sqrt{1 - x^2}$ over the unit circle")
plt.tight_layout(); plt.show()
```
