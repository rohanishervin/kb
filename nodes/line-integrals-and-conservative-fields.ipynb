# Vector Fields, Line Integrals and Conservative Fields

**Prerequisites:** [Integrals of Functions over Curves and Surfaces](integrals-over-curves-and-surfaces.ipynb) · [The Gradient and Level Sets](gradient.ipynb) · [Higher Partial Derivatives and Taylor's Theorem in Several Variables](higher-partials-and-taylor.ipynb#thm-schwarz) · [Ordinary Differential Equations and Picard's Theorem](ode-and-picards-thm.ipynb)

## 1\. Vector Fields

::: {#def-vector-field}
## Vector Field, Integral Curve

A **vector field** on $U \subseteq \mathbb{R}^n$ is a map $F : U \to \mathbb{R}^n$, pictured as an arrow $F(x)$ attached at each $x$. With the constant fields $e_1, \dots, e_n$ (in $\mathbb{R}^3$: $\mathbf i, \mathbf j, \mathbf k$), $F = \sum_iF_ie_i$. An **integral curve** (flow line) of $F$ is a curve with $\gamma'(t) = F(\gamma(t))$, i.e. a solution of the system $x_i' = F_i(x_1, \dots, x_n)$. If $F$ is $C^1$, exactly one integral curve passes through each point ([Picard's theorem](ode-and-picards-thm.ipynb)).
:::

::: {#exm-plane-fields}
## Four Plane Fields

1. **Source** $F = x\mathbf i + y\mathbf j$: integral curves $\gamma(t) = (x_0e^t, y_0e^t)$, rays leaving $0$ (reaching it only as $t \to -\infty$).
2. **Sink** $G = -x\mathbf i - y\mathbf j$: $\gamma(t) = (x_0e^{-t}, y_0e^{-t})$.
3. **Centre** $R = -y\mathbf i + x\mathbf j$: $R(x)$ is $x$ rotated by $\frac\pi2$; integral curves $\gamma(t) = (r_0\cos(\theta_0 + t), r_0\sin(\theta_0 + t))$, circles about $0$.
4. **Saddle** $S = x\mathbf i - y\mathbf j$: $\gamma(t) = (x_0e^t, y_0e^{-t})$, branches of the hyperbolas $xy = x_0y_0$ (the half-axes if $x_0y_0 = 0$).
:::

::: {#def-gradient-field}
## Gradient Fields and Potentials

$F$ is a **gradient field** (or **conservative**) on $U$ if $F = \nabla f$ for a $C^1$ function $f$ on $U$, a **potential** of $F$.
:::

$F$, $G$, $S$ above are gradient fields, with potentials $\frac12(x^2 + y^2)$, $-\frac12(x^2 + y^2)$, $\frac12(x^2 - y^2)$; their integral curves are orthogonal to the level curves of the potentials ([gradients are orthogonal to level sets](gradient.ipynb#thm-gradient-orthogonal)). $R$ is not:

::: {#prp-gradient-necessary}
## A Necessary Condition

If $F = \nabla f$ is $C^1$ on an open set, then $\dfrac{\partial F_i}{\partial x_j} = \dfrac{\partial F_j}{\partial x_i}$ for all $i, j$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-gradient-necessary
$f$ is then $C^2$ and $\frac{\partial F_i}{\partial x_j} = \frac{\partial^2f}{\partial x_j\partial x_i} = \frac{\partial^2f}{\partial x_i\partial x_j} = \frac{\partial F_j}{\partial x_i}$ by [Schwarz's theorem](higher-partials-and-taylor.ipynb#thm-schwarz). $\blacksquare$
:::

For $R = (-y, x)$: $\frac{\partial R_1}{\partial y} = -1 \ne 1 = \frac{\partial R_2}{\partial x}$. The condition is not sufficient (@exm-angle-field).

## 2\. Line Integrals

::: {#def-line-integral}
## Line Integral of a Vector Field

For a continuous field $F$ and a $C^1$ curve $\gamma : [a, b] \to U$, $\mathbf r = \gamma(t)$,
$$\int_\gamma F\cdot d\mathbf r = \int_a^bF(\gamma(t))\cdot\gamma'(t)\,dt = \int_a^b\Big(F_1\frac{dx_1}{dt} + \dots + F_n\frac{dx_n}{dt}\Big)dt,$$
also written $\int_\gamma F_1\,dx_1 + \dots + F_n\,dx_n$; for piecewise smooth $\gamma$, add over the pieces. If $\gamma$ is smooth with unit tangent $T = \gamma'/|\gamma'|$, then $\int_\gamma F\cdot d\mathbf r = \int_\gamma(F\cdot T)\,ds$: the integral accumulates the tangential component of $F$, e.g. the **work** done by a force field along a path.
:::

::: {#prp-line-integral-orientation}
## Dependence on Orientation

Under an orientation-preserving reparametrization the line integral is unchanged; under an orientation-reversing one it changes sign.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-line-integral-orientation
With $t = \alpha(\tau)$, $\int_c^dF(\gamma(\alpha(\tau)))\cdot\gamma'(\alpha(\tau))\alpha'(\tau)\,d\tau = \int_{\alpha(c)}^{\alpha(d)}F(\gamma(t))\cdot\gamma'(t)\,dt$, which is $\int_a^b$ if $\alpha' > 0$ and $\int_b^a = -\int_a^b$ if $\alpha' < 0$. $\blacksquare$
:::

::: {#exm-circulation}
## Circulation of the Rotation Field

For $R = -y\mathbf i + x\mathbf j$ and the circle $\gamma(t) = (A\cos t, A\sin t)$, $0 \le t \le 2\pi$ (counterclockwise):
$$\int_\gamma R\cdot d\mathbf r = \int_0^{2\pi}\big[(-A\sin t)(-A\sin t) + (A\cos t)(A\cos t)\big]dt = 2\pi A^2 .$$
For the square with vertices $(\pm A, \pm A)$, counterclockwise, the side $\gamma_1(t) = (A, t)$, $-A \le t \le A$, gives $\int_{-A}^A(-t, A)\cdot(0, 1)\,dt = 2A^2$, and the other three sides give the same, total $8A^2$. In both cases the result is twice the enclosed area, a first instance of [Green's theorem](greens-theorem.ipynb).
:::

## 3\. Conservative Fields

::: {#thm-gradient-line-integral}
## Fundamental Theorem for Line Integrals

If $f$ is $C^1$ on $U$ and $\gamma : [a, b] \to U$ is piecewise smooth from $A = \gamma(a)$ to $B = \gamma(b)$, then
$$\int_\gamma\nabla f\cdot d\mathbf r = f(B) - f(A).$$
In particular the integral of a gradient field over a closed curve is $0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-gradient-line-integral
On each smooth piece, $\nabla f(\gamma(t))\cdot\gamma'(t) = \frac{d}{dt}f(\gamma(t))$ by the [chain rule](multivariable-chain-rule.ipynb#thm-multivariable-chain-rule), so by the FTC its integral over $[t_{i-1}, t_i]$ is $f(\gamma(t_i)) - f(\gamma(t_{i-1}))$; the sum telescopes. $\blacksquare$
:::

::: {#thm-path-independence}
## Conservative Means Path-Independent

Let $U$ be open and path-connected (any two points are joined by a piecewise smooth curve in $U$) and $F$ continuous on $U$. Then $F$ is a gradient field iff $\int_\gamma F\cdot d\mathbf r$ depends only on the endpoints of $\gamma$, i.e. iff $\int_\gamma F\cdot d\mathbf r = 0$ for every closed piecewise smooth $\gamma$ in $U$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-path-independence
Necessity is @thm-gradient-line-integral. Conversely, fix $P \in U$ and define $f(Q) = \int_\gamma F\cdot d\mathbf r$ for any piecewise smooth $\gamma$ from $P$ to $Q$; this is well defined by assumption. Let $Q \in U$ and $h \ne 0$ small, so the segment $\lambda(t) = Q + te_i$, $t$ between $0$ and $h$, lies in $U$. Following $\gamma$ and then $\lambda$,
$$f(Q + he_i) - f(Q) = \int_\lambda F\cdot d\mathbf r = \int_0^hF(Q + te_i)\cdot e_i\,dt = \int_0^hF_i(Q + te_i)\,dt .$$
Dividing by $h$ and letting $h \to 0$, continuity of $F_i$ and the [FTC](fundamental-theorem-of-calculus.ipynb#thm-ftc) give $\frac{\partial f}{\partial x_i}(Q) = F_i(Q)$. The partials of $f$ are continuous, so $f$ is $C^1$ and $\nabla f = F$. The two forms of the condition are equivalent because traversing a curve backwards changes the sign (@prp-line-integral-orientation). $\blacksquare$
:::

::: {#exm-radial-field}
## A Radial Field

$F(x) = \dfrac{x}{|x|^2}$ on $\mathbb{R}^n\setminus\{0\}$ is the gradient of $f = \frac12\ln(x_1^2 + \dots + x_n^2)$. Hence for any piecewise smooth $\gamma$ in $\mathbb{R}^n\setminus\{0\}$ from $P$ to $Q$, $\int_\gamma F\cdot d\mathbf r = \ln\frac{|Q|}{|P|}$, which is $0$ if $|P| = |Q|$.
:::

::: {#exm-angle-field}
## A Field That Passes the Test but Is Not Conservative

$F = \dfrac{-y\,\mathbf i + x\,\mathbf j}{x^2 + y^2}$ on $\mathbb{R}^2\setminus\{0\}$ satisfies $\frac{\partial F_1}{\partial y} = \frac{y^2 - x^2}{(x^2 + y^2)^2} = \frac{\partial F_2}{\partial x}$. But on the counterclockwise circle of radius $R$,
$$\int_\gamma F\cdot d\mathbf r = \int_0^{2\pi}\frac{R^2\sin^2t + R^2\cos^2t}{R^2}\,dt = 2\pi \ne 0,$$
so $F$ is not a gradient field on $\mathbb{R}^2\setminus\{0\}$. Locally it is: $\frac{\partial}{\partial x}\arctan\frac yx = \frac{-y}{x^2 + y^2}$ and $\frac{\partial}{\partial y}\arctan\frac yx = \frac{x}{x^2 + y^2}$, so the polar angle $\theta$ is a potential wherever it can be chosen continuously, e.g. on the plane minus a ray from $0$, where $F$ is conservative. Around the full circle $\theta$ increases by $2\pi$ and cannot return to its initial value: the obstruction is the hole at $0$. More generally, the integral of $F$ over a closed curve in $\mathbb{R}^2\setminus\{0\}$ is $2\pi$ times the number of times it winds counterclockwise around $0$.
:::

```python

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

x, y, t, A = sp.symbols("x y t A", real=True)
def line_int(F, g, a, b):
    return sp.simplify(sp.integrate(sum(Fi.subs({x: g[0], y: g[1]}) * sp.diff(gi, t) for Fi, gi in zip(F, g)), (t, a, b)))
R = (-y, x)
print("R on circle:", line_int(R, (A * sp.cos(t), A * sp.sin(t)), 0, 2 * sp.pi))
sides = [(A, t), (-t, A), (-A, -t), (t, -A)]
print("R on square:", sp.simplify(sum(line_int(R, s, -A, A) for s in sides)))
Fang = (-y / (x**2 + y**2), x / (x**2 + y**2))
print("mixed partials equal:", sp.simplify(sp.diff(Fang[0], y) - sp.diff(Fang[1], x)) == 0,
      "| circulation on unit circle:", line_int(Fang, (sp.cos(t), sp.sin(t)), 0, 2 * sp.pi),
      "| twice around:", line_int(Fang, (sp.cos(t), sp.sin(t)), 0, 4 * sp.pi))

fig, axs = plt.subplots(1, 4, figsize=(15, 3.9))
g = np.linspace(-2, 2, 13); X, Y = np.meshgrid(g, g)
fine = np.linspace(-2, 2, 60); FX, FY = np.meshgrid(fine, fine)
fields = [(lambda a, b: (a, b), r"source $x\mathbf{i} + y\mathbf{j}$"),
          (lambda a, b: (-a, -b), r"sink $-x\mathbf{i} - y\mathbf{j}$"),
          (lambda a, b: (-b, a), r"centre $-y\mathbf{i} + x\mathbf{j}$"),
          (lambda a, b: (a, -b), r"saddle $x\mathbf{i} - y\mathbf{j}$")]
for ax, (F, tt) in zip(axs, fields):
    ax.streamplot(FX, FY, *F(FX, FY), color=MUTED, density=0.8, linewidth=0.7, arrowsize=0.7)
    ax.quiver(X, Y, *F(X, Y), color=BLUE, scale=40, width=0.004)
    ax.set_aspect("equal"); ax.set_title(tt); ax.set_xlim(-2, 2); ax.set_ylim(-2, 2)
plt.tight_layout(); plt.show()
```
