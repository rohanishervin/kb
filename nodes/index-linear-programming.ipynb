## Notation

The same symbols mean the same thing in every note.

| Symbol | Meaning |
|---|---|
| $x \in \mathbb{R}^n$, $c^{\mathsf T}x$ | vector of variables (a column), linear objective |
| $A \in \mathbb{R}^{m \times n}$, $a_i^{\mathsf T}$, $A_j$, $b \in \mathbb{R}^m$ | constraint matrix, its $i$-th row, its $j$-th column, right-hand side |
| $u \le v$, $x \ge 0$ | inequality in every coordinate |
| $\max c^{\mathsf T}x$ s.t. $Ax \le b$ | inequality form of an LP |
| $\max c^{\mathsf T}x$ s.t. $Ax = b$, $x \ge 0$ | equational form ($\operatorname{rank} A = m$ assumed) |
| $P$ | feasible set of an LP (a polyhedron) |
| $e_i$, $\mathbf 1$ | standard basis vector, all-ones vector |
| $\lVert x\rVert$, $B(z, r)$ | Euclidean norm, closed ball of radius $r$ about $z$ |
| $[x, y]$ | segment $\{tx + (1-t)y : t \in [0,1]\}$ |
| $\operatorname{conv} S$, $\operatorname{aff} S$, $\dim C$ | convex hull, affine hull, dimension $\dim\operatorname{aff} C$ |
| $H(a, \beta) = \{a^{\mathsf T}x = \beta\}$ | hyperplane with normal vector $a \ne 0$ |
| $\operatorname{ext} C$ | set of extreme points of $C$ |
| $C_n = [-1,1]^n$, $X_n$, $\Delta_n$ | cube, cross-polytope, standard simplex in $\mathbb{R}^{n+1}$ |
| $\operatorname{epi} f$ | epigraph of $f$ |
| $B$, $N$; $A_B$, $x_B$ | basis and nonbasic indices; columns and coordinates indexed by $B$ |
| $\operatorname{supp} x = \{j : x_j \ne 0\}$ | support |
| $x_B = p + Qx_N$, $z = z_0 + r^{\mathsf T}x_N$ | simplex tableau $\mathcal T(B)$, $Q = (q_{ij})$ |
| $u^+ = \max(u, 0)$, $u^- = \max(-u, 0)$ | positive and negative parts |
| $G = (V, E)$, $uv$, $\delta(v)$ | graph, edge $\{u, v\}$, edges containing $v$ |
| $\tau(G)$, $\alpha(G)$; $\tau_{\mathrm{LP}}$, $\alpha_{\mathrm{LP}}$ | smallest vertex cover, largest independent set; their LP relaxation values |
| $\mathrm{OPT}$, $\mathrm{LP}^*$ | optimal value of an integer problem, of its LP relaxation |
| $\Delta$ (set cover) | largest number of sets containing one element |
| $\mathbb{E}$, $\Pr$ | expectation, probability |
