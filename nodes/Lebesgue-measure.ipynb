# Lebesgue Measure

**Prerequisites:** [Measurable Sets Form an Algebra](measurable-sets-algebra.ipynb) · [Measurable Sets Form a σ-Algebra](sigma-algebra-of-measurable-sets.ipynb) · [Intervals, Open Sets, and Borel Sets Are Measurable](Borel-sets-and-measurability.ipynb) · [Almost Everywhere, σ-Finite Measures, and Completeness](measure-properties-and-completion.ipynb#def-complete-measure-space)

Only one step of the construction remains: on the σ-algebra $\mathcal{L}$ of measurable sets, outer measure is **countably additive**. After the work of the previous notes this is the easiest step. The result is a complete measure space $(\mathbb{R}^n, \mathcal{L}, m)$ in which boxes have their usual volume and every Borel set is measurable.

## 1\. Countable Additivity

::: {#thm-countable-additivity-lebesgue}
## Countable Additivity of $m^*$ on $\mathcal{L}$

If $E_1, E_2, \dots$ are measurable and pairwise disjoint, then
$$m^*\Big(\bigcup_{i=1}^\infty E_i\Big) = \sum_{i=1}^\infty m^*(E_i).$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-countable-additivity-lebesgue
Again we turn $m^*$ into a finite sum before letting $n \to \infty$; disjointification is not needed, since the $E_i$ are already disjoint. For every $n$, the finite union is contained in the infinite one, so by monotonicity and [finite additivity](measurable-sets-algebra.ipynb#thm-finite-additivity-test-set),
$$m^*\Big(\bigcup_{i=1}^\infty E_i\Big) \ge m^*\Big(\bigcup_{i=1}^n E_i\Big) = \sum_{i=1}^n m^*(E_i).$$
The left side does not depend on $n$; letting $n \to \infty$ on the right,
$$m^*\Big(\bigcup_{i=1}^\infty E_i\Big) \ge \sum_{i=1}^\infty m^*(E_i).$$
As in all the previous proofs we obtained $\ge$ instead of $=$, and as always the other direction is free: it is [countable subadditivity](lebesgue-outer-measure.ipynb#thm-countable-subadditivity). $\blacksquare$
:::

So the plan of the [outer-measure note](lebesgue-outer-measure.ipynb) is complete.

::: {#def-lebesgue-measure}
## Lebesgue Measure

The restriction of $m^*$ to $\mathcal{L}$,
$$m = m^*|_{\mathcal{L}} : \mathcal{L} \to [0, \infty], \qquad m(E) = m^*(E),$$
is **Lebesgue measure** on $\mathbb{R}^n$.
:::

::: {#thm-lebesgue-measure-space}
## The Lebesgue Measure Space

$(\mathbb{R}^n, \mathcal{L}, m)$ is a [complete](measure-properties-and-completion.ipynb#def-complete-measure-space) measure space. Every Borel set is in $\mathcal{L}$, and every box $B$ has $m(B) = |B|$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-lebesgue-measure-space
Everything has been proved already; we only collect it.

1. $\mathcal{L}$ is a σ-algebra ([σ-algebra of measurable sets](sigma-algebra-of-measurable-sets.ipynb#cor-lebesgue-sigma-algebra)).
2. $m(\emptyset) = m^*(\emptyset) = 0$, and $m$ is countably additive (@thm-countable-additivity-lebesgue). So $m$ is a measure.
3. **Completeness.** Let $m(B) = 0$ and $A \subseteq B$. Then $m^*(B) = 0$, so $m^*(A) = 0$ by monotonicity, and sets of outer measure zero are [measurable](measurable-sets.ipynb#prp-null-sets-measurable). Hence $A \in \mathcal{L}$ (with $m(A) = 0$).
4. Borel sets are measurable ([Borel sets](Borel-sets-and-measurability.ipynb#thm-borel-sets-measurable)), and $m(B) = m^*(B) = |B|$ for boxes ([outer measure of a box](lebesgue-outer-measure.ipynb#thm-outer-measure-of-box)). $\blacksquare$
:::

So all the sets we are likely ever to meet are measurable. Non-measurable sets do exist, but their existence proofs are not constructive, and we do not need them.

::: {.callout-note}
## The Construction Works in Any Space
Nothing in the passage from $m^*$ to $m$ used the geometry of $\mathbb{R}^n$: only that $m^*(\emptyset) = 0$, monotonicity and countable subadditivity. Given *any* outer measure $\mu^*$ on a set $X$, Carathéodory's definition $\mu^*(A) = \mu^*(A \cap E) + \mu^*(A \setminus E)$ for all $A$ defines a family of measurable sets; the same proofs show that it is a σ-algebra, that $\mu^*$ restricted to it is a complete measure, and that sets of outer measure zero are measurable. This **Carathéodory construction** produces many measures besides Lebesgue's, on spaces that are not Euclidean at all.
:::

## 2\. Translation Invariance

::: {#prp-lebesgue-translation-invariant}
## Translation Invariance of Lebesgue Measure

If $E \in \mathcal{L}$ and $y \in \mathbb{R}^n$, then $E + y \in \mathcal{L}$ and $m(E + y) = m(E)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-lebesgue-translation-invariant
Outer measure is [translation invariant](lebesgue-outer-measure.ipynb#prp-outer-measure-translation), because it is defined through boxes and translating a box does not change its volume. For any $A \subseteq \mathbb{R}^n$, apply the measurability of $E$ to the test set $A - y$:
$$m^*(A) = m^*(A - y) = m^*\big((A - y) \cap E\big) + m^*\big((A - y) \cap E^c\big) = m^*\big(A \cap (E + y)\big) + m^*\big(A \cap (E + y)^c\big),$$
since translating by $y$ maps $(A - y) \cap E$ onto $A \cap (E + y)$ and $(A - y) \cap E^c$ onto $A \cap (E+y)^c$. So $E + y$ is measurable, and $m(E + y) = m^*(E + y) = m^*(E) = m(E)$. $\blacksquare$
:::

## 3\. Examples

With a genuine measure available, all the [general properties of measures](measure.ipynb#basic-properties) apply.

::: {#exm-lebesgue-measure-computations}
## Computing With Lebesgue Measure

1. **The rationals.** $\mathbb{Q}$ is countable, so $m(\mathbb{Q}) = 0$. By [excision](measure.ipynb#prp-excision), the irrationals of $[0, 1]$ have measure $m([0,1] \setminus \mathbb{Q}) = m([0,1]) - m([0,1] \cap \mathbb{Q}) = 1 - 0 = 1$. In this precise sense [almost every real number is irrational](measure-properties-and-completion.ipynb#exm-almost-every-real-is-irrational).
2. **The Cantor set.** The stages $C_n$ decrease to $C$, each $C_n$ is a disjoint union of $2^n$ closed intervals of length $3^{-n}$, so $m(C_n) = (2/3)^n$, and $m(C_0) = 1 < \infty$. By [continuity from above](continuity-of-measure.ipynb#thm-continuity-of-measure), $m(C) = \lim_n (2/3)^n = 0$. Together with completeness: **every** subset of the Cantor set is measurable, with measure $0$.
3. **Segments and lines in the plane.** A segment is a null set, so it is measurable with measure $0$. A whole line is a countable union of segments, hence null as well.
4. **Unbounded sets.** $m(\mathbb{R}) = \infty$, but $\mathbb{R} = \bigcup_k [-k, k]$ with $m([-k,k]) = 2k$, so Lebesgue measure is [σ-finite](measure-properties-and-completion.ipynb#def-finite-sigma-finite). Continuity from below gives $m(\mathbb{R}) = \lim_k 2k = \infty$.
:::

The figure shows the computation in example 2: the measures $m(C_n) = (2/3)^n$ of the Cantor stages decrease to $m(C) = 0$, while the uncountably many points of $C$ remain.

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

from fractions import Fraction as Fr

def cantor_stage(n):
    ivs = [(Fr(0), Fr(1))]
    for _ in range(n):
        ivs = [p for (a, b) in ivs for p in ((a, a + (b - a) / 3), (b - (b - a) / 3, b))]
    return ivs

n = np.arange(0, 16)
measures = [float(sum(b - a for a, b in cantor_stage(k))) for k in n]
assert np.allclose(measures, (2 / 3) ** n)
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(11, 3.6), gridspec_kw={"width_ratios": [1.3, 1]})
for k in range(6):
    for a, b in cantor_stage(k):
        ax.plot([float(a), float(b)], [-k, -k], color=BLUE, lw=6, solid_capstyle="butt")
ax.set_yticks(-np.arange(6)); ax.set_yticklabels([f"$C_{k}$" for k in range(6)]); ax.grid(False)
ax.set_title("Cantor stages: decreasing, $m(C_0) = 1 < \\infty$")
ax2.plot(n, measures, "o-", color=BLUE, ms=4, label=r"$m(C_n) = (2/3)^n$")
ax2.axhline(0, color=RED, ls="--", lw=1.2, label="$m(C) = 0$")
ax2.set_xlabel("$n$"); ax2.legend()
ax2.set_title("Continuity from above")
plt.tight_layout(); plt.show()
```
