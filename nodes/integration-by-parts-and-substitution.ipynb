# Integration by Parts and Substitution

**Prerequisites:** [The Fundamental Theorem of Calculus](fundamental-theorem-of-calculus.ipynb) · [The Logarithm and the Exponential Function](exponential-and-logarithm.ipynb)

Each differentiation rule gives an integration rule. Notation: $\int f(x)\,dx$ denotes an antiderivative; in $\int_a^b f(x)\,dx$ the letter $x$ is a dummy variable.

## 1\. Integration by Parts

::: {#thm-integration-by-parts}
## Integration by Parts

If $f, g$ have continuous derivatives on $[a, b]$, then
$$\int_a^b f(x)g'(x)\,dx = \big[f(x)g(x)\big]_a^b - \int_a^b f'(x)g(x)\,dx, \qquad \int f g' \,dx = fg - \int f'g\,dx .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-integration-by-parts
$(fg)' = f'g + fg'$ is continuous, so by the [FTC](fundamental-theorem-of-calculus.ipynb#thm-ftc-evaluation) $\int_a^b (f'g + fg') = [fg]_a^b$; use linearity. The indefinite form says $fg - \int f'g$ is an antiderivative of $fg'$. $\blacksquare$
:::

::: {#exm-by-parts}
## Examples

1. $\int \ln x\,dx = x\ln x - \int x\cdot\frac1x\,dx = x\ln x - x + C$ ($f = \ln x$, $g = x$).
2. $I_n = \int x^ne^x\,dx = x^ne^x - n I_{n-1}$ ($f = x^n$, $g = e^x$); with $I_0 = e^x$, e.g. $I_1 = (x - 1)e^x$, $I_2 = (x^2 - 2x + 2)e^x$.
3. $J = \int_0^{\pi/2} e^x\sin x\,dx = \big[e^x \sin x\big]_0^{\pi/2} - \int_0^{\pi/2} e^x\cos x\,dx = e^{\pi/2} - \Big(\big[e^x\cos x\big]_0^{\pi/2} + J\Big) = e^{\pi/2} + 1 - J$, so $J = \frac{e^{\pi/2} + 1}{2}$.
:::

## 2\. Substitution

::: {#thm-substitution}
## Change of Variables

Let $\varphi : [\alpha, \beta] \to \mathbb{R}$ have a continuous derivative, and $f$ be continuous on an interval containing $\varphi([\alpha, \beta])$. Then
$$\int_\alpha^\beta f(\varphi(t))\,\varphi'(t)\,dt = \int_{\varphi(\alpha)}^{\varphi(\beta)} f(x)\,dx .$$
In indefinite form: if $F' = f$, then $\int f(\varphi(t))\varphi'(t)\,dt = F(\varphi(t)) + C$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-substitution
Let $F$ be an antiderivative of $f$ ([FTC](fundamental-theorem-of-calculus.ipynb#thm-ftc)). By the [chain rule](chain-rule-and-inverse-functions.ipynb#thm-chain-rule), $(F\circ\varphi)' = (f\circ\varphi)\,\varphi'$, which is continuous. By the [FTC](fundamental-theorem-of-calculus.ipynb#thm-ftc-evaluation) both sides equal $F(\varphi(\beta)) - F(\varphi(\alpha))$. $\blacksquare$
:::

In Leibniz notation, $x = \varphi(t)$, $dx = \varphi'(t)\,dt$: the formula is "substitute and change $dx$ to $\frac{dx}{dt}dt$". $\varphi$ need not be injective: for $f = 1$ and $\varphi(t) = t^2$ on $[-1, 2]$, $\int_{-1}^2 2t\,dt = 3 = \int_1^4 dx$; the part $[-1, 0]$, where $\varphi$ decreases, contributes negatively and cancels the part $[0, 1]$. Where $\varphi' > 0$ the factor $\varphi'$ is the stretching of the $t$-axis onto the $x$-axis, which is why it multiplies the integrand.

::: {#exm-substitution}
## Examples

1. $\int\frac{\varphi'(t)}{\varphi(t)}\,dt = \ln|\varphi(t)| + C$ where $\varphi \ne 0$. For example $\int \tan t\,dt = -\ln|\cos t| + C$.
2. $\int\frac{dx}{\sqrt{x^3 + x^4}}$ on $\{x > 0\} \cup \{x < -1\}$: write $\frac{1}{\sqrt{x^3 + x^4}} = \frac{1}{x^2}\big(\frac1x + 1\big)^{-1/2}$ and substitute $u = \frac1x + 1$, $du = -\frac{dx}{x^2}$: the antiderivative is $-2\sqrt{\frac1x + 1} + C$ on each of the two intervals.
3. $\int_0^a \sqrt{a^2 - x^2}\,dx$ with $x = a\sin\theta$, $\theta \in [0, \frac\pi2]$: $\int_0^{\pi/2} a^2\cos^2\theta\,d\theta = a^2\Big[\frac\theta2 + \frac{\sin 2\theta}{4}\Big]_0^{\pi/2} = \frac{\pi a^2}{4}$, a quarter disc.
:::

::: {#prp-standard-substitutions}
## Standard Substitutions for Square Roots

For $a > 0$, the following substitutions remove the root (on suitable intervals):

| expression | substitution | result |
|---|---|---|
| $\sqrt{a^2 - x^2}$ | $x = a\sin\theta$, $\theta \in [-\frac\pi2, \frac\pi2]$ | $a\cos\theta$ |
| $\sqrt{a^2 + x^2}$ | $x = a\tan\theta$ or $x = a\sinh t$ | $a\sec\theta$, resp. $a\cosh t$ |
| $\sqrt{x^2 - a^2}$ | $x = a\sec\theta$ or $x = a\cosh t$ ($x \ge a$) | $a\tan\theta$ ($\theta \in [0, \frac\pi2)$), resp. $a\sinh t$ ($t \ge 0$) |

($\sqrt{-x^2 - a^2}$ is not real.) Every $\sqrt{At^2 + Bt + C}$ reduces to one of these after completing the square. For example $\int\frac{dx}{\sqrt{x^2 + a^2}} = \int dt = t = \operatorname{arsinh}\frac{x}{a} = \ln\big(x + \sqrt{x^2 + a^2}\big) + C$.
:::

```python

import sympy as sp

x, t, a = sp.symbols("x t a", positive=True)
checks = {
    "int ln x": (sp.integrate(sp.log(x), x), x * sp.log(x) - x),
    "int x^2 e^x": (sp.integrate(x**2 * sp.exp(x), x), (x**2 - 2*x + 2) * sp.exp(x)),
    "int_0^{pi/2} e^x sin x": (sp.integrate(sp.exp(x) * sp.sin(x), (x, 0, sp.pi / 2)), (sp.exp(sp.pi / 2) + 1) / 2),
    "int_0^a sqrt(a^2-x^2)": (sp.integrate(sp.sqrt(a**2 - x**2), (x, 0, a)), sp.pi * a**2 / 4),
}
for name, (got, claimed) in checks.items():
    print(f"{name:26} {sp.simplify(got - claimed) == 0}")
X = sp.symbols("X")
F = -2 * sp.sqrt(1 / X + 1)
for x0 in [sp.Rational(1, 2), 3, -2, -5]:
    lhs = sp.diff(F, X).subs(X, x0); rhs = (1 / sp.sqrt(X**3 + X**4)).subs(X, x0)
    print(f"d/dx[-2 sqrt(1/x+1)] at x = {x0}: {sp.N(lhs):.6f} vs 1/sqrt(x^3+x^4) = {sp.N(rhs):.6f}")
```
