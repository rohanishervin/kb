# Linear Programs

**Prerequisites:** [Affine Subspaces](affine-subspaces.ipynb) · [Linear Independence](linear-independence.ipynb)

## 1\. Linear Programs

Vectors in $\mathbb{R}^n$ are columns. For $u, v \in \mathbb{R}^m$, the inequality $u \le v$ means $u_i \le v_i$ for every $i$; similarly $u \ge 0$.

::: {#def-linear-program}
## Linear Program

A **linear function** on $\mathbb{R}^n$ is a map $x \mapsto c^{\mathsf T}x = c_1x_1 + \dots + c_nx_n$ with $c \in \mathbb{R}^n$. A **linear constraint** is a condition of one of the forms
$$
a^{\mathsf T}x \le \beta, \qquad a^{\mathsf T}x \ge \beta, \qquad a^{\mathsf T}x = \beta \qquad (a \in \mathbb{R}^n,\ \beta \in \mathbb{R}).
$$
A **linear program** (LP) asks to maximize, or to minimize, a linear function $c^{\mathsf T}x$ (the **objective function**) over all $x \in \mathbb{R}^n$ satisfying finitely many given linear constraints.

- A vector satisfying all constraints is a **feasible solution**; the set $P$ of all of them is the **feasible set**. The LP is **feasible** if $P \ne \emptyset$ and **infeasible** otherwise.
- For a maximization LP, a feasible $x^*$ is an **optimal solution** if $c^{\mathsf T}x^* \ge c^{\mathsf T}x$ for all $x \in P$; the number $c^{\mathsf T}x^*$ is the **optimal value**. (For minimization reverse the inequality.)
- A feasible maximization LP is **unbounded** if $c^{\mathsf T}x$ is not bounded above on $P$ (for minimization: not bounded below).
:::

::: {#def-inequality-form}
## Inequality Form

An LP is in **inequality form** if it reads
$$
\text{maximize } c^{\mathsf T}x \quad \text{subject to } Ax \le b,
$$
with $A \in \mathbb{R}^{m \times n}$, $b \in \mathbb{R}^m$, $c \in \mathbb{R}^n$. The $i$-th row $a_i^{\mathsf T}$ of $A$ gives the constraint $a_i^{\mathsf T}x \le b_i$.
:::

::: {#prp-inequality-form}
## Every LP Can Be Written in Inequality Form

Apply the following replacements to an LP:

1. "minimize $c^{\mathsf T}x$" $\to$ "maximize $(-c)^{\mathsf T}x$";
2. $a^{\mathsf T}x \ge \beta$ $\to$ $(-a)^{\mathsf T}x \le -\beta$;
3. $a^{\mathsf T}x = \beta$ $\to$ the two constraints $a^{\mathsf T}x \le \beta$ and $(-a)^{\mathsf T}x \le -\beta$.

The resulting LP is in inequality form, has the same feasible set and the same optimal solutions; its optimal value is the original one, with the sign changed if step 1 was used.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-inequality-form
Steps 2 and 3 replace a constraint by constraints with exactly the same solution set: $a^{\mathsf T}x \ge \beta \iff -a^{\mathsf T}x \le -\beta$, and $a^{\mathsf T}x = \beta \iff \big(a^{\mathsf T}x \le \beta$ and $a^{\mathsf T}x \ge \beta\big)$. So the feasible set $P$ does not change. For step 1, $x^* \in P$ satisfies $c^{\mathsf T}x^* \le c^{\mathsf T}x$ for all $x \in P$ if and only if $(-c)^{\mathsf T}x^* \ge (-c)^{\mathsf T}x$ for all $x \in P$; so the optimal solutions are the same and the optimal values are negatives of each other. Finally, stack the rows of all constraints $a^{\mathsf T}x \le \beta$ into a matrix $A$ and the right-hand sides into $b$. $\blacksquare$
:::

## 2\. A Two-Dimensional Example

::: {#exm-first-lp}
## Solving an LP in the Plane

$$
\begin{aligned}
\text{maximize } \ & x_1 + x_2 \\
\text{subject to } \ & x_1 \ge 0,\quad x_2 \ge 0,\\
& x_2 - x_1 \le 1,\quad x_1 + 6x_2 \le 15,\quad 4x_1 - x_2 \le 10.
\end{aligned}
$$
In inequality form, $c = (1, 1)$ and
$$
A = \begin{pmatrix} -1 & 0 \\ 0 & -1 \\ -1 & 1 \\ 1 & 6 \\ 4 & -1 \end{pmatrix}, \qquad b = \begin{pmatrix} 0 \\ 0 \\ 1 \\ 15 \\ 10 \end{pmatrix}.
$$
Each constraint describes a closed half-plane, so $P$ is the intersection of five half-planes: the pentagon with corners $(0, 0)$, $(0, 1)$, $\big(\tfrac97, \tfrac{16}7\big)$, $(3, 2)$, $\big(\tfrac52, 0\big)$.

The level sets $\{x_1 + x_2 = t\}$ are parallel lines orthogonal to $c$, and $t$ grows in the direction of $c$. The largest $t$ whose level line still meets $P$ is $t = 5$, and it meets $P$ only at $(3, 2)$. So the optimal solution is $x^* = (3, 2)$ with optimal value $5$.
:::

::: {.callout-tip collapse="true"}
## Proof that $(3, 2)$ is the unique optimal solution
The point $(3, 2)$ is feasible: $3, 2 \ge 0$, $2 - 3 \le 1$, $3 + 12 = 15$, $12 - 2 = 10$. Add the fourth and fifth constraints with weights $\tfrac15$ each (multiplying an inequality by a positive number keeps its direction):
$$
\tfrac15(x_1 + 6x_2) + \tfrac15(4x_1 - x_2) = x_1 + x_2 \le \tfrac15(15 + 10) = 5 .
$$
So $x_1 + x_2 \le 5$ on $P$, with equality only if both constraints hold with equality: $x_1 + 6x_2 = 15$ and $4x_1 - x_2 = 10$. This system has the unique solution $(3, 2)$. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Maximizing a Linear Function Means Going Far in a Direction
$c^{\mathsf T}x = \lVert c\rVert\,\lVert x\rVert \cos\theta$ is $\lVert c\rVert$ times the signed length of the projection of $x$ onto the line through $c$. Maximizing $c^{\mathsf T}x$ over $P$ means finding a point of $P$ that is as far as possible in the direction $c$: slide the hyperplane $\{c^{\mathsf T}x = t\}$ in the direction of $c$ until it is about to leave $P$.
:::

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID, FAINT = "#0b0b0b", "#52514e", "#d9d8d4", "#a3a29a"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})
from scipy.optimize import linprog

A = np.array([[-1, 0], [0, -1], [-1, 1], [1, 6], [4, -1]], float)
b = np.array([0, 0, 1, 15, 10], float)
c = np.array([1, 1], float)
res = linprog(-c, A_ub=A, b_ub=b, bounds=[(None, None)] * 2)   # linprog minimizes
print("optimal solution:", res.x, " optimal value:", -res.fun)

corners = np.array([[0, 0], [2.5, 0], [3, 2], [9/7, 16/7], [0, 1]])
fig, ax = plt.subplots(figsize=(6.4, 5))
ax.fill(corners[:, 0], corners[:, 1], color=VIOLET, alpha=0.18, label="feasible set $P$")
ax.plot(*np.vstack([corners, corners[:1]]).T, color=VIOLET, lw=1.6)
t = np.linspace(-1, 6, 2)
for level in [1, 2, 3, 4]:
    ax.plot(t, level - t, color=FAINT, lw=1, ls="--")
ax.plot(t, 5 - t, color=ORANGE, lw=2, label="level line $x_1 + x_2 = 5$")
ax.annotate("", xy=(1.7, 1.7), xytext=(1.0, 1.0), arrowprops=dict(arrowstyle="->", color=BLUE, lw=2))
ax.text(1.75, 1.55, "$c$", color=BLUE, fontsize=12)
ax.plot(*res.x, "o", color=YELLOW, mec=INK, ms=9, label="optimum $(3,2)$")
ax.set_xlim(-0.5, 4.5); ax.set_ylim(-0.5, 3.5); ax.set_aspect("equal")
ax.set_xlabel("$x_1$"); ax.set_ylabel("$x_2$"); ax.legend(loc="upper right")
ax.set_title("the last level line that meets $P$")
plt.tight_layout(); plt.show()
```


## 3\. The Possible Outcomes

::: {#exm-lp-outcomes}
## Infinitely Many Optima, Infeasibility, Unboundedness

Let $P$ be the pentagon of @exm-first-lp.

1. *Infinitely many optimal solutions.* Maximize $x_1 + 6x_2$ over $P$. Since $x_1 + 6x_2 \le 15$ is a constraint, the optimal value is at most $15$, and it equals $15$ exactly on the edge $\{x \in P : x_1 + 6x_2 = 15\}$, the segment from $\big(\tfrac97, \tfrac{16}7\big)$ to $(3, 2)$. Every point of this segment is optimal. This happens because $c = (1, 6)$ is orthogonal to that edge, so a level line contains it.
2. *Infeasible.* The constraints $x_1 \ge 0$, $x_2 \ge 0$, $x_1 + x_2 \le -1$ have no common solution: they would give $0 \le x_1 + x_2 \le -1$.
3. *Unbounded.* Keep only $x_1 \ge 0$, $x_2 \ge 0$, $x_2 - x_1 \le 1$. The feasible set contains $(s, s)$ for every $s \ge 0$, and $x_1 + x_2 = 2s \to \infty$, so maximizing $x_1 + x_2$ is unbounded.
4. *Unbounded feasible set, bounded LP.* Over the same unbounded set, maximize $-x_1 - x_2$: it is $\le 0$ on the set and equals $0$ at the origin, which is the unique optimal solution.
:::

::: {#thm-lp-trichotomy-statement}
## The Three Outcomes

Every LP is exactly one of: infeasible; unbounded; feasible with an optimal solution. In particular, if the objective is bounded on a nonempty feasible set, the optimum is attained.
:::

The proof needs the theory of basic feasible solutions; it is given in [Basic Feasible Solutions](basic-feasible-solutions.ipynb#cor-lp-trichotomy).

::: {.callout-warning}
## Attainment Is Special to Linear Programs
For nonlinear constraints a bounded objective need not attain its supremum. Minimize $x_1$ subject to $x_1 x_2 \ge 1$, $x_1 > 0$: the values $x_1 = 1/k$ (with $x_2 = k$) are feasible, so the infimum is $0$, but $x_1 = 0$ is not feasible.
:::

## 4\. Linear Programming and Linear Algebra

| | Linear algebra | Linear programming |
|---|---|---|
| problem | a system of linear equations $Ax = b$ | a system of linear inequalities $Ax \le b$ (plus an objective) |
| solution set | empty or an [affine subspace](affine-subspaces.ipynb#def-affine-subspace) | an intersection of finitely many half-spaces: a [polyhedron](polyhedra-and-polytopes.ipynb#def-polyhedron) |
| basic algorithm | Gaussian elimination | the [simplex method](simplex-method.ipynb) |

A single linear equation in $\mathbb{R}^n$ (with $a \ne 0$) describes a hyperplane, of dimension $n - 1$; a single inequality describes a half-space bounded by such a hyperplane. An affine subspace has no further structure, while a polyhedron has corners, edges, and faces of every dimension between $0$ and its own dimension, and the number of these can grow exponentially with $n$.

::: {.callout-note}
## The Objective Is Not the Hard Part
@prp-feasibility-oracle below shows that deciding whether a system of linear inequalities has a solution is essentially as hard as solving an LP.
:::

## 5\. Geometry in High Dimension

Pictures in the plane suggest how LPs behave, but intuition from $\mathbb{R}^2$ and $\mathbb{R}^3$ can fail badly in $\mathbb{R}^n$, and LPs in practice have thousands of variables.

::: {#exm-ball-in-cube}
## A Ball Surrounded by Corner Balls

In $\mathbb{R}^n$ let $Q = [-2, 2]^n$ and place the $2^n$ unit balls $B(v, 1)$, $v \in \{-1, 1\}^n$, in its corners. Each lies in $Q$ (every coordinate of a point of $B(v, 1)$ is within $1$ of $v_i = \pm 1$), and distinct centers differ in at least one coordinate by $2$, so the balls do not overlap. The largest ball centered at $0$ that does not overlap any corner ball has radius
$$
r_n = \sqrt{n} - 1 .
$$
Thus $r_2 \approx 0.41$, $r_3 \approx 0.73$, $r_4 = 1$, $r_9 = 2$, and for $n \ge 10$ we get $r_n > 2$: the central ball sticks out of the cube $Q$, although it is "enclosed" by the corner balls.
:::

::: {.callout-tip collapse="true"}
## Proof of @exm-ball-in-cube
"Do not overlap" means the open balls $U(0, r) = \{\lVert x\rVert < r\}$ and $U(v, 1)$ are disjoint. Every corner $v$ has $\lVert v\rVert = \sqrt n$. We show: $U(0, r) \cap U(v, 1) = \emptyset \iff r + 1 \le \lVert v\rVert$.

If $r + 1 \le \lVert v\rVert$ and $x$ were in both, the triangle inequality would give $\lVert v\rVert \le \lVert x\rVert + \lVert v - x\rVert < r + 1$, a contradiction. If $r + 1 > \lVert v\rVert$, the point $x = \frac{r}{r+1}v$ has $\lVert x\rVert = \frac{r\lVert v\rVert}{r + 1} < r$ and $\lVert v - x\rVert = \frac{\lVert v\rVert}{r + 1} < 1$, so it lies in both.

Hence the admissible radii are $r \le \sqrt n - 1$, and the largest is $r_n = \sqrt n - 1$. Finally $r_n > 2 \iff n > 9$, and then the point $r_n e_1$ of the closed central ball has first coordinate $> 2$, so it lies outside $Q$. $\blacksquare$
:::

```python

from matplotlib.patches import Circle, Rectangle

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(10, 4.4))
ax1.add_patch(Rectangle((-2, -2), 4, 4, fill=False, ec=MUTED, lw=1.4))
for v in [(1, 1), (1, -1), (-1, 1), (-1, -1)]:
    ax1.add_patch(Circle(v, 1, fc=BLUE, alpha=0.18, ec=BLUE, lw=1.4))
ax1.add_patch(Circle((0, 0), np.sqrt(2) - 1, fc=ORANGE, alpha=0.35, ec=ORANGE, lw=1.6))
ax1.set_xlim(-2.3, 2.3); ax1.set_ylim(-2.3, 2.3); ax1.set_aspect("equal")
ax1.set_title(r"$n = 2$: central radius $\sqrt{2} - 1$")

n = np.arange(1, 21)
ax2.plot(n, np.sqrt(n) - 1, "o-", color=ORANGE, ms=4, label=r"central radius $r_n = \sqrt{n} - 1$")
ax2.axhline(1, color=BLUE, lw=1.2, ls="--", label="corner radius 1")
ax2.axhline(2, color=RED, lw=1.4, label="half side of the cube")
ax2.axvline(9, color=FAINT, lw=1)
ax2.set_xlabel("dimension $n$"); ax2.legend(loc="upper left")
ax2.set_title("for $n \\geq 10$ the central ball leaves the cube")
plt.tight_layout(); plt.show()
print("r_n:", {k: round(float(np.sqrt(k)) - 1, 3) for k in [2, 3, 4, 9, 10, 16]})
```


The cube $[-1, 1]^n$ is described by $2n$ inequalities but has $2^n$ corners (see [Polyhedra and Polytopes](polyhedra-and-polytopes.ipynb#exm-cube)). An optimal solution can always be found among the corners of $P$ (see [Vertices of Polytopes](vertices-of-polytopes.ipynb#cor-lp-optimum-at-vertex)), but listing all corners is hopeless for large $n$.

## 6\. Feasibility Versus Optimization

::: {#prp-feasibility-oracle}
## Optimizing With a Feasibility Test

Suppose we have a procedure that decides, for any finite system of linear inequalities, whether it has a solution. Consider the LP: maximize $c^{\mathsf T}x$ subject to $Ax \le b$.

1. For every $t \in \mathbb{R}$, one call decides whether some feasible $x$ has $c^{\mathsf T}x \ge t$.
2. If the LP has an optimal value $v^*$ and $L \le v^* \le U$ are known, then $k = \lceil \log_2 \frac{U - L}{\varepsilon} \rceil$ calls determine an interval of length $\le \varepsilon$ containing $v^*$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-feasibility-oracle

1. Some feasible $x$ has $c^{\mathsf T}x \ge t$ exactly when the system $Ax \le b$, $(-c)^{\mathsf T}x \le -t$ has a solution.
2. Keep an interval $[L, U] \ni v^*$. Test $t = \frac{L + U}2$: if some feasible $x$ has $c^{\mathsf T}x \ge t$, then $v^* \ge t$ and we replace $L$ by $t$; otherwise $v^* < t$ and we replace $U$ by $t$. The new interval contains $v^*$ and has half the length. After $k$ steps the length is $(U - L)2^{-k} \le \varepsilon$. $\blacksquare$
:::

::: {.callout-note}
## LPs Can Be Solved Efficiently
Despite the difficulties above, LPs can be solved efficiently, both in theory (there are algorithms whose running time is polynomial in the size of the input) and in practice. The classical method is the [simplex method](simplex-method.ipynb), which walks along corners of $P$; the ellipsoid and interior point methods are other algorithms. Solvers are available in standard software; the code cell above uses `scipy.optimize.linprog`, which minimizes, so it is given $-c$.
:::
