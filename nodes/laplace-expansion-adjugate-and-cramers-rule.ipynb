# Laplace Expansion, the Adjugate, and Cramer's Rule

**Prerequisites:** [The Determinant](determinant.ipynb)

The Laplace expansion computes an $n \times n$ determinant from $(n-1) \times (n-1)$ determinants. Collecting these smaller determinants into a matrix gives the **adjugate** $\operatorname{adj} A$, which satisfies $\operatorname{adj}(A)\,A = \det(A)\, I$. This identity yields an explicit formula for $A^{-1}$, Cramer's rule for linear systems, and, later, the key step in the proof of the [Cayley-Hamilton Theorem](cayley-hamilton-theorem.ipynb).

Throughout, $K$ is a field and $A = (a_{ij}) \in M_n(K)$.

## 1\. Minors and Cofactors

::: {#def-minors-and-cofactors}
## Minors and Cofactors

For $1 \le i, j \le n$:

1. $A_{ij} \in M_{n-1}(K)$ denotes the matrix obtained from $A$ by deleting row $i$ and column $j$;
2. $\det A_{ij}$ is the $(i, j)$ **minor**;
3. $C_{ij} = (-1)^{i+j} \det A_{ij}$ is the $(i, j)$ **cofactor**.

The signs $(-1)^{i+j}$ form a checkerboard starting with $+$ in the top-left corner.
:::

## 2\. Laplace Expansion

::: {#lem-unit-column}
## A Column Equal to a Unit Vector

If column $j$ of $A$ is the unit vector $e_i$, then $\det A = (-1)^{i+j} \det A_{ij}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-unit-column
Move row $i$ to the top by $i - 1$ swaps of neighbouring rows, and column $j$ to the front by $j - 1$ swaps of neighbouring columns. Each swap changes the sign ([column operations](determinant.ipynb#prp-det-column-operations) and their [row versions](determinant.ipynb#cor-row-column-duality)), and the other rows and columns keep their relative order. The result is
$$\tilde A = \begin{pmatrix} 1 & * \\ 0 & A_{ij} \end{pmatrix}, \qquad \det A = (-1)^{(i-1) + (j-1)} \det \tilde A = (-1)^{i+j} \det \tilde A.$$
In the Leibniz formula for $\tilde A$, the first column is $e_1$, so every term with $\sigma(1) \ne 1$ contains the factor $\tilde a_{\sigma(1)1} = 0$. The remaining terms are those of the permutations of $\{2, \dots, n\}$, with the same signs (a permutation fixing $1$ has the same inversions as its restriction), and they add up to $\det A_{ij}$. $\blacksquare$
:::

::: {#thm-laplace-expansion}
## Laplace (Cofactor) Expansion

For every column index $j$ and every row index $i$,
$$\det A = \sum_{i=1}^n a_{ij}\, C_{ij} \quad\text{(along column } j\text{)}, \qquad \det A = \sum_{j=1}^n a_{ij}\, C_{ij} \quad\text{(along row } i\text{)}.$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-laplace-expansion
Column $j$ of $A$ is $\sum_i a_{ij} e_i$. By linearity of $\det$ in column $j$,
$$\det A = \sum_{i=1}^n a_{ij} \det(A \text{ with column } j \text{ replaced by } e_i),$$
and by @lem-unit-column the $i$-th determinant is $(-1)^{i+j}\det A_{ij} = C_{ij}$ (deleting row $i$ and column $j$ removes the new column, so the minor is the same as for $A$). The row expansion is the column expansion of $A^{\mathsf T}$, using [$\det A^{\mathsf T} = \det A$](determinant.ipynb#thm-det-transpose). $\blacksquare$
:::

::: {#exm-laplace}
## Expanding Along a Convenient Row

Expand along the second row, which has two zeros:
$$\det\begin{pmatrix} 1 & 2 & 3 \\ 0 & 4 & 0 \\ 5 & 6 & 7 \end{pmatrix} = 4 \cdot (-1)^{2+2} \det\begin{pmatrix} 1 & 3 \\ 5 & 7 \end{pmatrix} = 4\,(7 - 15) = -32.$$
:::

## 3\. The Adjugate Matrix

::: {#def-adjugate-matrix}
## Adjugate

The **adjugate** (or **classical adjoint**) of $A$ is the **transpose** of the cofactor matrix:
$$(\operatorname{adj} A)_{ij} = C_{ji} = (-1)^{i+j} \det A_{ji}.$$
:::

::: {#thm-adjugate-fundamental-identity}
## The Fundamental Identity

For every $A \in M_n(K)$:
$$A\, \operatorname{adj}(A) = \operatorname{adj}(A)\, A = \det(A)\, I_n.$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-adjugate-fundamental-identity
The $(i, j)$ entry of $A \operatorname{adj}(A)$ is $\sum_k a_{ik} C_{jk}$.

- For $i = j$ this is the Laplace expansion of $\det A$ along row $i$.
- For $i \ne j$, let $\tilde A$ be $A$ with row $j$ replaced by a copy of row $i$. Then $\tilde A$ has two equal rows, so $\det \tilde A = 0$ ([rows](determinant.ipynb#cor-row-column-duality)). Expanding $\det \tilde A$ along row $j$, whose entries are $a_{ik}$ and whose cofactors are those of $A$ (deleting row $j$ removes the only changed row), gives $0 = \sum_k a_{ik} C_{jk}$.

So $A \operatorname{adj}(A) = \det(A) I_n$. For $\operatorname{adj}(A)\, A$, whose $(i, j)$ entry is $\sum_k C_{ki} a_{kj}$, argue the same way with columns: for $i = j$ it is the expansion along column $i$, and for $i \ne j$ it is the expansion along column $i$ of the matrix in which column $i$ is replaced by column $j$. $\blacksquare$
:::

::: {.callout-note}
## The Identity Holds over Any Commutative Ring
The Leibniz formula makes sense for matrices with entries in any commutative [ring](rings-and-polynomial-rings.ipynb#def-ring) $R$, for example in the polynomial ring $K[x]$. The proofs of multilinearity, of the alternating property (by cancelling the terms of $\sigma$ and $\sigma\tau$), of $\det A^{\mathsf T} = \det A$, of the Laplace expansion and of the fundamental identity use only addition, multiplication and their commutativity, never division. So $\operatorname{adj}(A)\,A = \det(A)\,I_n$ holds for all $A \in M_n(R)$. This is exactly what the proof of the [Cayley-Hamilton Theorem](cayley-hamilton-theorem.ipynb#thm-cayley-hamilton) needs, with $R = K[x]$ and $A = x I_n - M$.
:::

::: {#cor-matrix-inverse-formula}
## Explicit Inverse

If $\det A \ne 0$, then
$$A^{-1} = \frac{1}{\det A} \operatorname{adj}(A).$$
For $n = 2$: $\begin{pmatrix} a & b \\ c & d \end{pmatrix}^{-1} = \dfrac{1}{ad - bc}\begin{pmatrix} d & -b \\ -c & a \end{pmatrix}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-matrix-inverse-formula
Divide the fundamental identity by the non-zero scalar $\det A$. For $n = 2$, the cofactors are $C_{11} = d$, $C_{12} = -c$, $C_{21} = -b$, $C_{22} = a$; transposing gives the stated adjugate. $\blacksquare$
:::

## 4\. Cramer's Rule

::: {#thm-cramers-rule}
## Cramer's Rule

If $\det A \ne 0$ and $b \in K^n$, the unique solution of $A x = b$ is
$$x_i = \frac{\det A_i(b)}{\det A} \qquad (i = 1, \dots, n),$$
where $A_i(b)$ is $A$ with its $i$-th column replaced by $b$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-cramers-rule
By @cor-matrix-inverse-formula, $x = A^{-1} b = \frac{1}{\det A}\operatorname{adj}(A)\, b$, so
$$x_i = \frac{1}{\det A}\sum_{k} C_{ki}\, b_k.$$
The sum $\sum_k b_k C_{ki}$ is the expansion of $\det A_i(b)$ along its $i$-th column: that column has entries $b_k$, and deleting it leaves the same minors as in $A$. $\blacksquare$
:::

::: {#exm-cramer}
## A $2 \times 2$ System

For $2x + y = 5$, $x + 3y = 10$: $\det A = 2 \cdot 3 - 1 \cdot 1 = 5$, and
$$x = \frac{1}{5}\det\begin{pmatrix} 5 & 1 \\ 10 & 3 \end{pmatrix} = \frac{15 - 10}{5} = 1, \qquad y = \frac{1}{5}\det\begin{pmatrix} 2 & 5 \\ 1 & 10 \end{pmatrix} = \frac{20 - 5}{5} = 3.$$
:::

::: {.callout-warning}
## A Formula, Not an Algorithm
Cramer's rule and the adjugate formula are valuable for proofs: they show, for example, that the entries of $A^{-1}$ and the solution $x$ are rational functions of the entries of $A$ and $b$, hence depend continuously on them. For computation they are hopeless beyond small $n$: they need $n + 1$ determinants of size $n$, while Gaussian elimination solves the system in about $\tfrac23 n^3$ operations.
:::


```python

import numpy as np

def cofactor(M, i, j):
    return (-1) ** (i + j) * np.linalg.det(np.delete(np.delete(M, i, axis=0), j, axis=1))

def adjugate(M):
    n = len(M)
    return np.array([[cofactor(M, j, i) for j in range(n)] for i in range(n)])   # transpose of cofactors

A = np.array([[1.0, 2, 3], [0, 4, 0], [5, 6, 7]])
print("expansion along row 2:", sum(A[1, j] * cofactor(A, 1, j) for j in range(3)), " numpy:", round(np.linalg.det(A), 6))
print("A adj(A) =\n", np.round(A @ adjugate(A), 6))

# a singular matrix: adj(A) A = 0, although adj(A) itself is not zero
S = np.array([[1.0, 2, 3], [2, 4, 6], [1, 0, 1]])
print("\nsingular S: det =", round(np.linalg.det(S), 6), "\nadj(S) =\n", np.round(adjugate(S), 6),
      "\nadj(S) S =\n", np.round(adjugate(S) @ S, 6))

# Cramer's rule on the 2x2 example
M, b = np.array([[2.0, 1], [1, 3]]), np.array([5.0, 10])
x = [np.linalg.det(np.column_stack([b if k == i else M[:, k] for k in range(2)])) / np.linalg.det(M) for i in range(2)]
print("\nCramer:", np.round(x, 6), " solve:", np.linalg.solve(M, b))
```
