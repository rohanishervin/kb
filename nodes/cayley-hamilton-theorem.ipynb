# The Cayley-Hamilton Theorem

**Prerequisites:** [Diagonalization](diagonalization.ipynb) · [Adjugate Identity](laplace-expansion-adjugate-and-cramers-rule.ipynb#thm-adjugate-fundamental-identity) · [Rings and Polynomial Rings](rings-and-polynomial-rings.ipynb) · [Triangularization](triangularization-theorem.ipynb)

Every square matrix satisfies its own characteristic equation: if $p_A(\lambda) = \det(\lambda I - A) = \lambda^n + c_{n-1}\lambda^{n-1} + \dots + c_0$, then
$$A^n + c_{n-1} A^{n-1} + \dots + c_1 A + c_0 I = 0.$$
This note proves the theorem three times: for diagonalizable matrices, in general by a short algebraic argument built on the adjugate and on polynomials with matrix coefficients, and over $\mathbb{C}$ by a limiting argument based on triangularization.

Throughout, $K$ is a field, $A \in M_n(K)$, and for a polynomial $p(x) = \sum_k c_k x^k \in K[x]$ we write $p(A) = \sum_k c_k A^k$ with $A^0 = I_n$.

## 1\. Motivation

### 1\.1\. A Polynomial of Degree $n^2$ Is Easy

$M_n(K)$ has [dimension](basis-and-dimension.ipynb#exm-dimension-examples) $n^2$, so the $n^2 + 1$ matrices $I, A, A^2, \dots, A^{n^2}$ are [linearly dependent](basis-and-dimension.ipynb#cor-independent-full-size): some non-zero polynomial of degree at most $n^2$ annihilates $A$.

### 1\.2\. Degree $n$ for One Vector at a Time

For a fixed $v \in K^n$, the $n + 1$ vectors $v, Av, \dots, A^n v$ in the $n$-dimensional space $K^n$ are dependent, so $q_v(A)\,v = 0$ for some non-zero polynomial $q_v$ of degree at most $n$. But $q_v$ depends on $v$. The theorem says that **one** polynomial of degree $n$, the characteristic polynomial, works for all $v$ at once.

::: {#thm-cayley-hamilton}
## Cayley-Hamilton Theorem

For every $A \in M_n(K)$, $p_A(A) = 0$. Equivalently, every linear operator $f$ on an $n$-dimensional space satisfies $p_f(f) = 0$.
:::

::: {#exm-ch-2x2}
## The $2 \times 2$ Case

For $A = \begin{pmatrix} a & b \\ c & d \end{pmatrix}$, $p_A(\lambda) = \lambda^2 - (a + d)\lambda + (ad - bc)$, and the theorem says $A^2 - (\operatorname{tr} A)\,A + (\det A)\,I = 0$. For $A = \begin{pmatrix} 1 & 2 \\ 3 & 4 \end{pmatrix}$: $A^2 = \begin{pmatrix} 7 & 10 \\ 15 & 22 \end{pmatrix}$ and $5A + 2I = \begin{pmatrix} 7 & 10 \\ 15 & 22 \end{pmatrix}$, so indeed $A^2 - 5A - 2I = 0$.
:::

## 2\. Proof for Diagonalizable Matrices

::: {#prp-cayley-hamilton-diagonalizable}
## The Diagonalizable Case

If $A$ has $n$ linearly independent eigenvectors (for example, if it has $n$ distinct eigenvalues), then $p_A(A) = 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-cayley-hamilton-diagonalizable
Let $v_1, \dots, v_n$ be a basis of eigenvectors, $A v_i = \lambda_i v_i$. By the [spectral mapping property](eigenvalues-and-eigenvectors.ipynb#thm-spectral-mapping-polynomial), $p_A(A) v_i = p_A(\lambda_i) v_i$, and $p_A(\lambda_i) = 0$ because $\lambda_i$ is an eigenvalue. So the linear map $p_A(A)$ vanishes on a basis, hence everywhere: $p_A(A) = 0$. $\blacksquare$
:::

::: {#exm-ch-rotation}
## A Rotation, Checked Directly

The rotation $R_\alpha = \begin{pmatrix} \cos\alpha & -\sin\alpha \\ \sin\alpha & \cos\alpha \end{pmatrix}$ has $p(\lambda) = \lambda^2 - 2\cos\alpha\,\lambda + 1$. Using $R_\alpha^2 = R_{2\alpha}$ and the double-angle formulas,
$$R_\alpha^2 - 2\cos\alpha\, R_\alpha + I = \begin{pmatrix} \cos 2\alpha - 2\cos^2\alpha + 1 & -\sin 2\alpha + 2\sin\alpha\cos\alpha \\ \sin 2\alpha - 2\sin\alpha\cos\alpha & \cos 2\alpha - 2\cos^2\alpha + 1 \end{pmatrix} = 0,$$
for every $\alpha$, including multiples of $\pi$, where $R_\alpha = \pm I$ and the eigenvalue is repeated. Over $\mathbb{R}$, $R_\alpha$ has no eigenvectors unless $\alpha$ is a multiple of $\pi$; the theorem holds regardless.
:::

## 3\. The General Algebraic Proof

The proof applies the [adjugate identity](laplace-expansion-adjugate-and-cramers-rule.ipynb#thm-adjugate-fundamental-identity) to the matrix $x I_n - A$, whose entries are polynomials, and then substitutes $A$ for $x$. Two preparations make this legitimate.

::: {#lem-matrix-polynomials}
## Matrices of Polynomials Are Polynomials with Matrix Coefficients

Every $P(x) \in M_n(K[x])$ can be written uniquely as $P(x) = \sum_k P_k\, x^k$ with $P_k \in M_n(K)$ ($P_k$ collects the coefficients of $x^k$ in all entries). This identifies $M_n(K[x])$ with the polynomial ring $M_n(K)[x]$, and the identification respects addition and multiplication.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-matrix-polynomials
Uniqueness and additivity are clear entry by entry. For products, let $P = \sum_a P_a x^a$ and $Q = \sum_b Q_b x^b$. The $(i, j)$ entry of the matrix product is $\sum_k P_{ik}(x) Q_{kj}(x)$, whose coefficient of $x^m$ is
$$\sum_k \sum_{a + b = m} (P_a)_{ik} (Q_b)_{kj} = \sum_{a + b = m} (P_a Q_b)_{ij},$$
which is the $(i, j)$ entry of the coefficient of $x^m$ in the product $\big(\sum_a P_a x^a\big)\big(\sum_b Q_b x^b\big)$ in $M_n(K)[x]$. $\blacksquare$
:::

The second preparation is the [Commutation Lemma](rings-and-polynomial-rings.ipynb#lem-commutation-evaluation): in a product $P(x)\,Q(x)$ with matrix coefficients, substituting a matrix $C$ for $x$ is compatible with the product as soon as $C$ commutes with the coefficients of the right-hand factor $Q$.

::: {.callout-tip collapse="true"}
## Proof of @thm-cayley-hamilton (general case)
Work in the commutative ring $K[x]$ and let $p_A(x) = \det(x I_n - A) = x^n + c_{n-1}x^{n-1} + \dots + c_0$. The adjugate identity [holds over any commutative ring](laplace-expansion-adjugate-and-cramers-rule.ipynb#thm-adjugate-fundamental-identity), so
$$\operatorname{adj}(x I_n - A)\,(x I_n - A) = p_A(x)\, I_n \qquad \text{in } M_n(K[x]). \qquad (\dagger)$$
By @lem-matrix-polynomials, read $(\dagger)$ in $M_n(K)[x]$:

- $Q(x) := x I_n - A = I_n\, x + (-A)$, a polynomial of degree $1$ with coefficients $I_n$ and $-A$;
- $P(x) := \operatorname{adj}(x I_n - A) = B_{n-1} x^{n-1} + \dots + B_1 x + B_0$ for some $B_k \in M_n(K)$ (each entry is an $(n-1) \times (n-1)$ minor of $x I_n - A$, of degree at most $n - 1$);
- $p_A(x) I_n = I_n x^n + c_{n-1} I_n x^{n-1} + \dots + c_0 I_n$.

Now evaluate both sides of $P(x)\,Q(x) = p_A(x) I_n$ at $x = A$, in the ring $R = M_n(K)$.

1. **The hypothesis of the Commutation Lemma holds:** $A$ commutes with both coefficients of $Q$, since $A I_n = I_n A$ and $A(-A) = (-A)A$. Hence $E_A(PQ) = E_A(P)\, E_A(Q)$.
2. $E_A(Q) = I_n A + (-A) = 0$, so $E_A(PQ) = E_A(P) \cdot 0 = 0$.
3. On the right side, $E_A(p_A(x) I_n) = A^n + c_{n-1}A^{n-1} + \dots + c_0 I_n = p_A(A)$.

Therefore $p_A(A) = 0$. For an operator $f$, apply this to $A = [f]_{\mathcal{B}}$: the map $g \mapsto [g]_{\mathcal{B}}$ respects sums, products and scalars, so $[p_f(f)]_{\mathcal{B}} = p_f(A) = p_A(A) = 0$. $\blacksquare$
:::

::: {.callout-warning}
## The Tempting Wrong Proof
"Substitute $\lambda = A$ in $p_A(\lambda) = \det(\lambda I - A)$ to get $\det(A - A) = \det 0 = 0$" is not a proof: it produces the **number** $0$, while the theorem asserts that a **matrix** is $0$, and $p_A(A)$ means substituting $A$ into the expanded polynomial, not into the determinant. The correct proof above does substitute, but only after rewriting the determinant via the adjugate identity, and only where the Commutation Lemma allows it.
:::

## 4\. A Proof over $\mathbb{C}$ via Triangularization and Density {#sec-ch-density}

The algebraic proof works over every field. Over $\mathbb{C}$, the [Triangularization Theorem](triangularization-theorem.ipynb#thm-triangularization) gives a second proof that leaves pure algebra and uses a limiting argument. Real matrices are complex matrices too, so the result covers them as well.

::: {.callout-tip collapse="true"}
## Proof of @thm-cayley-hamilton for $K = \mathbb{C}$
**Step 1: reduce to a triangular matrix.** By triangularization, $P^{-1} A P = M$ is upper triangular for some invertible $P$, with diagonal entries $\lambda_1, \dots, \lambda_n$. Similar matrices have the [same characteristic polynomial](eigenvalues-and-eigenvectors.ipynb#thm-characteristic-equation), $p := p_A = p_M$, and since $(P^{-1}AP)^k = P^{-1}A^kP$, we get $p(M) = P^{-1}p(A)P$. So it suffices to prove $p_M(M) = 0$.

**Step 2: distinct diagonal entries.** If the $\lambda_i$ are distinct, $M$ has $n$ distinct eigenvalues, so it is [diagonalizable](diagonalization.ipynb#thm-sufficient-diagonalizability), and @prp-cayley-hamilton-diagonalizable applies.

**Step 3: perturb the diagonal.** For $k = 1, 2, \dots$ choose $\varepsilon_{k,1}, \dots, \varepsilon_{k,n} \in [0, \tfrac1k)$ such that $\lambda_1 + \varepsilon_{k,1}, \dots, \lambda_n + \varepsilon_{k,n}$ are distinct. This is possible: take $\varepsilon_{k,1} = 0$, and choose each $\varepsilon_{k,i}$ in turn avoiding the finitely many values $\lambda_j + \varepsilon_{k,j} - \lambda_i$, $j < i$. Let $A_k$ be $M$ with diagonal entries $\lambda_i + \varepsilon_{k,i}$ and the same entries above the diagonal.

**Step 4: Cayley-Hamilton for each $A_k$.** $A_k$ is triangular with distinct diagonal entries, so by Step 2
$$A_k^n + a_{n-1,k} A_k^{n-1} + \dots + a_{0,k} I = 0, \qquad (\ast)$$
where $p_{A_k}(\lambda) = \lambda^n + a_{n-1,k}\lambda^{n-1} + \dots + a_{0,k}$.

**Step 5: let $k \to \infty$.** $A_k \to M$ entrywise. The coefficients $a_{j,k}$ are polynomials in the entries of $A_k$ (expand the determinant), and so are the entries of the powers $A_k^j$. Polynomials are continuous, so the left side of $(\ast)$ converges entrywise to $p_M(M)$. Since it is $0$ for every $k$, $p_M(M) = 0$. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Matrices with Distinct Eigenvalues Are Everywhere
Arbitrarily close to **any** complex matrix there is one with distinct eigenvalues. The expression $p_A(A)$ depends continuously on $A$ and vanishes on all those nearby matrices, so it vanishes at $A$ as well. The argument needs a notion of limit, so it does not apply to arbitrary fields, where the algebraic proof is required.
:::

## 5\. Consequences

::: {#cor-inverse-via-cayley-hamilton}
## The Inverse Is a Polynomial in $A$

If $A$ is invertible, then $c_0 = p_A(0) = (-1)^n \det A \ne 0$ and
$$A^{-1} = -\frac{1}{c_0}\left(A^{n-1} + c_{n-1} A^{n-2} + \dots + c_1 I\right).$$
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-inverse-via-cayley-hamilton
Rewrite $p_A(A) = 0$ as $A\,(A^{n-1} + c_{n-1}A^{n-2} + \dots + c_1 I) = -c_0 I$ and divide by $-c_0$. The factor commutes with $A$, so it is a two-sided inverse. $\blacksquare$
:::

::: {#cor-power-reduction}
## High Powers Reduce to Low Ones

For every $k \ge n$, $A^k = r(A)$ where $r$ is the remainder of $x^k$ upon [division](polynomial-division-algorithm.ipynb#thm-polynomial-division) by the monic polynomial $p_A$; in particular $\deg r \le n - 1$. So every power of $A$, and every polynomial in $A$, is a combination of $I, A, \dots, A^{n-1}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-power-reduction
$x^k = s(x)\,p_A(x) + r(x)$ with $\deg r < n$. Substituting $A$ (allowed, since $K[x]$ is commutative and all coefficients are scalars) gives $A^k = s(A)\,p_A(A) + r(A) = r(A)$. $\blacksquare$
:::

::: {#exm-ch-consequences}
## Using the Theorem

For $A = \begin{pmatrix} 1 & 2 \\ 3 & 4 \end{pmatrix}$ we found $A^2 = 5A + 2I$. Hence $A^{-1} = \tfrac12(A - 5I) = \begin{pmatrix} -2 & 1 \\ 1.5 & -0.5 \end{pmatrix}$, and $A^3 = 5A^2 + 2A = 27A + 10I$, without computing any product of matrices.
:::

The left panel shows the theorem for one vector in the plane: $A^2 v$ is not a new direction but the combination $(\operatorname{tr} A)\,Av - (\det A)\,v$, for every $v$. The other two panels illustrate the density proof for a triangular matrix with eigenvalue $2$ of multiplicity $3$: the perturbed diagonal entries merge as $k \to \infty$, $A_k \to M$, while $p_{A_k}(A_k)$ is zero (up to rounding) at every step.


```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 10.5,
    "axes.titleweight": "bold", "xtick.color": MUTED, "ytick.color": MUTED,
    "legend.frameon": False, "font.size": 10})

def residual(X):
    c = np.poly(X)                                            # coefficients of det(lambda I - X)
    return sum(ci * np.linalg.matrix_power(X, len(c) - 1 - i) for i, ci in enumerate(c))

fig, axes = plt.subplots(1, 3, figsize=(15, 4.4))

# panel 1: A^2 v = tr(A) A v - det(A) v
A = np.array([[1.0, -1.0], [1.0, 0.5]]); v = np.array([1.0, 0.0])   # tr A = 1.5, det A = 1.5
Av, A2v = A @ v, A @ A @ v
ax = axes[0]
for w, c, lab in [(v, BLUE, "$v$"), (Av, ORANGE, "$Av$"), (A2v, VIOLET, "$A^2v$")]:
    ax.annotate("", xy=w, xytext=(0, 0), arrowprops=dict(arrowstyle="-|>", color=c, lw=2.2, mutation_scale=14))
    ax.text(*(w + [0.05, 0.08]), lab, color=INK, fontsize=12)
combo = np.trace(A) * Av - np.linalg.det(A) * v
ax.plot(*combo, "x", color=INK, ms=11, mew=2, label=r"$\operatorname{tr}(A)Av-\det(A)v$")
ax.set_xlim(-0.6, 2.2); ax.set_ylim(-0.3, 2.2); ax.set_aspect("equal"); ax.legend(loc="upper right")
ax.set_title(r"$A^2v$ is a combination of $Av$ and $v$")

# panels 2-3: the density argument
M = np.array([[2.0, 1.0, -1.0], [0.0, 2.0, 3.0], [0.0, 0.0, 2.0]])
ks = np.arange(1, 41)
diag_k, dist, res = [], [], []
for k in ks:
    Ak = M + np.diag(np.array([0.0, 0.5, 0.9]) / k)          # distinct perturbations, each < 1/k
    diag_k.append(np.diag(Ak)); dist.append(np.abs(Ak - M).max()); res.append(np.abs(residual(Ak)).max())
diag_k = np.array(diag_k)
ax = axes[1]
for i, c in enumerate([BLUE, AQUA, VIOLET]):
    ax.plot(ks, diag_k[:, i], "o-", ms=3, color=c, label=rf"$\lambda_{i+1}+\varepsilon_{{k,{i+1}}}$")
ax.axhline(2, color=RED, ls="--", lw=1.2, label="eigenvalue of $M$")
ax.set_xlabel("$k$"); ax.legend(); ax.set_title("distinct eigenvalues of $A_k$ merge")
ax = axes[2]
ax.semilogy(ks, dist, "o-", ms=3, color=BLUE, label=r"$\max|A_k - M|$")
ax.semilogy(ks, np.maximum(res, 1e-17), "s-", ms=3, color=RED, label=r"$\max|p_{A_k}(A_k)|$ (rounding)")
ax.set_xlabel("$k$"); ax.legend(); ax.set_title(r"$A_k \to M$, and $p_{A_k}(A_k) = 0$ throughout")
plt.tight_layout(); plt.show()
print("max |p_M(M)| for the limit matrix itself:", np.abs(residual(M)).max())
```

```python

import numpy as np

rng = np.random.default_rng(42)
n = 4
A = rng.integers(-3, 4, size=(n, n)).astype(float)
c = np.poly(A)                                         # [1, c_{n-1}, ..., c_0]
pA = sum(ci * np.linalg.matrix_power(A, n - i) for i, ci in enumerate(c))
print("characteristic polynomial coefficients:", np.round(c, 6))
print("p_A(A) == 0:", np.allclose(pA, 0))

# inverse as a polynomial in A
c0 = c[-1]
inv = -(1 / c0) * sum(c[i] * np.linalg.matrix_power(A, n - 1 - i) for i in range(n))
print("A^{-1} from Cayley-Hamilton matches numpy:", np.allclose(inv, np.linalg.inv(A)))

# A^10 equals r(A) with r the remainder of x^10 modulo p_A
x10 = np.zeros(11); x10[0] = 1.0                      # numpy.polydiv uses highest degree first
_, r = np.polydiv(x10, c)
rA = sum(ri * np.linalg.matrix_power(A, len(r) - 1 - i) for i, ri in enumerate(r))
print("A^10 == r(A) with deg r =", len(r) - 1, ":", np.allclose(np.linalg.matrix_power(A, 10), rA))
```
