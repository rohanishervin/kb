# Existence of $n^{\text{th}}$ Roots in $\mathbb{R}$

**Prerequisites:** [The Axiom of Completeness](axiom-of-completeness.ipynb) · [Supremum and Infimum](supremum-infimum.ipynb) · [Archimedean Property of $\mathbb{R}$](archimedean-property-of-R.ipynb)

::: {#lem-power-increasing}
## Powers Are Increasing on $[0, \infty)$

Let $n \in \mathbb{N}$. If $0 \le x < y$ then $x^n < y^n$, and
$$0 < y^n - x^n = (y - x)(y^{n-1} + y^{n-2}x + \dots + x^{n-1}) \le n\,y^{n-1}(y - x).$$
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-power-increasing
Multiplying out, the product telescopes to $y^n - x^n$. Each of the $n$ terms satisfies $0 \le y^{n-1-k}x^k \le y^{n-1}$ (as $0 \le x < y$), and the term with $k = 0$ equals $y^{n-1} > 0$. Hence the sum lies in $(0, n y^{n-1}]$, and $y - x > 0$. $\blacksquare$
:::

::: {#thm-nth-root}
## Existence and Uniqueness of $n$-th Roots

For every $a \ge 0$ and $n \in \mathbb{N}$ there is exactly one $x \ge 0$ with $x^n = a$. It is written $\sqrt[n]{a} = a^{1/n}$. The map $a \mapsto a^{1/n}$ is strictly increasing on $[0, \infty)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-nth-root
*Uniqueness and monotonicity* follow from @lem-power-increasing: $x < y$ implies $x^n < y^n$.

*Existence.* The case $a = 0$ is trivial; let $a > 0$ and
$$S = \{s \ge 0 : s^n \le a\}.$$
$0 \in S$, and $S$ is bounded above by $1 + a$, because $s > 1 + a \ge 1$ gives $s^n \ge s > a$. Let $x = \sup S$ ([completeness](axiom-of-completeness.ipynb)); note $x \ge \min(1, a) > 0$ since $\min(1,a)^n \le \min(1, a) \le a$. Put $K = n(x+1)^{n-1}$.

* *$x^n < a$ is impossible.* Choose $0 < h \le 1$ with $hK < a - x^n$. By the lemma, $(x + h)^n - x^n \le n(x+h)^{n-1}h \le Kh < a - x^n$, so $x + h \in S$, contradicting $x = \sup S$.
* *$x^n > a$ is impossible.* Choose $0 < h \le x$ with $hK < x^n - a$. For every $s \in S$, $s^n \le a < x^n - Kh \le (x - h)^n$ (lemma with $y = x$: $x^n - (x-h)^n \le n x^{n-1} h \le Kh$), so $s < x - h$. Then $x - h$ is an upper bound smaller than $x$: contradiction.

Hence $x^n = a$. $\blacksquare$
:::

::: {#def-rational-power}
## Rational Powers

For $a > 0$ and $r = m/n$ ($m \in \mathbb{Z}$, $n \in \mathbb{N}$) put $a^{r} = (a^{1/n})^m$. This does not depend on the representation: if $m/n = m'/n'$ then both $(a^{1/n})^m$ and $(a^{1/n'})^{m'}$ are positive numbers whose $(nn')$-th power is $a^{mn'} = a^{m'n}$, so they coincide by uniqueness. The usual rules $a^{r+s} = a^r a^s$, $(a^r)^s = a^{rs}$, $(ab)^r = a^r b^r$ follow the same way.
:::

The root $a \mapsto a^{1/n}$ is continuous; this is proved in [Monotone Functions and Continuous Inverses](monotone-and-inverse-functions.ipynb#exm-root-continuous), where the existence of roots is also obtained a second time from the intermediate value theorem. A proof of existence that works directly with cuts and the same supremum is in [Problem 2](pset-real-numbers-01.ipynb#problem-2-pugh-ch.-1-p.-15) and [Problem 3](pset-real-numbers-01.ipynb#problem-3-pugh-ch.-1-p.-16) of the real numbers problem set.
