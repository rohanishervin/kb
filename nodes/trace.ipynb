# The Trace

**Prerequisites:** [Change of Basis and Similar Matrices](change-of-basis-and-similarity.ipynb)

Besides the [determinant](determinant.ipynb), a square matrix has a second basic number attached to it: the **trace**, the sum of the diagonal entries. Where the determinant is multiplicative, the trace is additive. Its key property is $\operatorname{tr}(AB) = \operatorname{tr}(BA)$, which makes it independent of the basis, so that every linear operator has a trace.

Throughout, $K$ is a field.

## 1\. Definition and Linearity

::: {#def-trace-matrix}
## Trace of a Matrix

The **trace** of $A = (a_{ij}) \in M_n(K)$ is
$$\operatorname{tr} A = a_{11} + a_{22} + \dots + a_{nn}.$$
:::

::: {#thm-trace-properties}
## Linearity and Transpose Invariance

For $A, B \in M_n(K)$ and $a, b \in K$:

1. $\operatorname{tr}(aA + bB) = a \operatorname{tr} A + b \operatorname{tr} B$; that is, $\operatorname{tr}$ is a [linear functional](dual-spaces.ipynb#def-linear-functional) on $M_n(K)$.
2. $\operatorname{tr}(A^{\mathsf T}) = \operatorname{tr} A$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-trace-properties
1. The $i$-th diagonal entry of $aA + bB$ is $a\,a_{ii} + b\,b_{ii}$; sum over $i$.
2. Transposition does not move diagonal entries. $\blacksquare$
:::

## 2\. The Cyclic Property

::: {#thm-trace-cyclic}
## $\operatorname{tr}(AB) = \operatorname{tr}(BA)$

For $A \in M_{m \times n}(K)$ and $B \in M_{n \times m}(K)$:
$$\operatorname{tr}(AB) = \operatorname{tr}(BA).$$
Note that $AB$ is $m \times m$ and $BA$ is $n \times n$; even for $m = n$ the products $AB$ and $BA$ are usually different matrices.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-trace-cyclic
$$\operatorname{tr}(AB) = \sum_{i=1}^m (AB)_{ii} = \sum_{i=1}^m \sum_{k=1}^n a_{ik} b_{ki}, \qquad \operatorname{tr}(BA) = \sum_{k=1}^n (BA)_{kk} = \sum_{k=1}^n \sum_{i=1}^m b_{ki} a_{ik}.$$
Both are the sum of $a_{ik} b_{ki}$ over all pairs $(i, k)$, since multiplication in $K$ is commutative and finite sums can be reordered. $\blacksquare$
:::

::: {.callout-warning}
## Cyclic, Not Arbitrary, Reordering
By the theorem, $\operatorname{tr}(ABC) = \operatorname{tr}(BCA) = \operatorname{tr}(CAB)$: products may be rotated cyclically. Other reorderings are not allowed in general. For $A = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix}$, $B = \begin{pmatrix} 0 & 0 \\ 1 & 0 \end{pmatrix}$, $C = \begin{pmatrix} 1 & 0 \\ 0 & 0 \end{pmatrix}$: $\operatorname{tr}(ABC) = 1$ but $\operatorname{tr}(BAC) = 0$.
:::

::: {#cor-no-commutator-identity}
## No Two Matrices Satisfy $AB - BA = I$

If $n \cdot 1 \ne 0$ in $K$ (for instance $K = \mathbb{Q}, \mathbb{R}, \mathbb{C}$), there are no $A, B \in M_n(K)$ with $AB - BA = I_n$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-no-commutator-identity
By linearity and @thm-trace-cyclic, $\operatorname{tr}(AB - BA) = \operatorname{tr}(AB) - \operatorname{tr}(BA) = 0$, while $\operatorname{tr} I_n = n \ne 0$. $\blacksquare$
:::

::: {.callout-note}
## Why This Matters
The relation $AB - BA = I$ is satisfied by the operators "multiply by $x$" and "differentiate" on polynomials: $(D \circ X - X \circ D)(p) = (xp)' - x p' = p$. The corollary says this can only happen on infinite-dimensional spaces; it is the linear-algebra reason why the position and momentum operators of quantum mechanics cannot be finite matrices.
:::

## 3\. The Trace of an Operator

::: {#thm-trace-similarity-invariance}
## Similar Matrices Have the Same Trace

If $B = P^{-1} A P$, then $\operatorname{tr} B = \operatorname{tr} A$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-trace-similarity-invariance
By @thm-trace-cyclic with the two factors $P^{-1}A$ and $P$:
$$\operatorname{tr}(P^{-1} A P) = \operatorname{tr}\big((P^{-1}A)\,P\big) = \operatorname{tr}\big(P\,(P^{-1}A)\big) = \operatorname{tr} A. \qquad \blacksquare$$
:::

::: {#def-trace-operator}
## Trace of an Operator

For a linear operator $f$ on a finite-dimensional space $V$, $\operatorname{tr} f := \operatorname{tr} [f]_{\mathcal{B}}$ for any ordered basis $\mathcal{B}$. By the [transformation law](change-of-basis-and-similarity.ipynb#thm-operator-change-of-basis) and @thm-trace-similarity-invariance this does not depend on $\mathcal{B}$.
:::

::: {#exm-trace-operators}
## Traces of Familiar Operators

1. **Projection.** The projection of $\mathbb{R}^3$ onto the $xy$-plane along the $z$-axis has matrix $\operatorname{diag}(1, 1, 0)$, so its trace is $2$, the dimension of the plane it projects onto.
2. **Rotation.** The rotation of $\mathbb{R}^2$ by $\alpha$ has trace $2\cos\alpha$, in every basis.
3. **Differentiation** on $P_d$ has a strictly upper triangular matrix in the basis $1, x, \dots, x^d$, so its trace is $0$.
:::

::: {.callout-note}
## Determinant and Trace Together
$\det$ is multiplicative ($\det(fg) = \det f \det g$) and measures volume change; $\operatorname{tr}$ is linear ($\operatorname{tr}(f + g) = \operatorname{tr} f + \operatorname{tr} g$). Both appear in the [characteristic polynomial](eigenvalues-and-eigenvectors.ipynb#thm-characteristic-equation):
$$\det(\lambda I - A) = \lambda^n - (\operatorname{tr} A)\,\lambda^{n-1} + \dots + (-1)^n \det A,$$
and over $\mathbb{C}$ the trace is the sum and the determinant the product of the eigenvalues ([spectral formulas](eigenvalues-and-eigenvectors.ipynb#thm-characteristic-coefficients)).
:::


```python

import numpy as np

rng = np.random.default_rng(42)
A, B, C = (rng.integers(-3, 4, (3, 3)).astype(float) for _ in range(3))
print("AB == BA ?", np.allclose(A @ B, B @ A), "  tr(AB) =", np.trace(A @ B), " tr(BA) =", np.trace(B @ A))
print("tr(ABC), tr(BCA), tr(CAB) =", np.trace(A @ B @ C), np.trace(B @ C @ A), np.trace(C @ A @ B))
print("tr(BAC) =", np.trace(B @ A @ C), "(not a cyclic rotation)")

# rectangular case: a 2x3 times a 3x2
R = rng.integers(-3, 4, (2, 3)).astype(float); S = rng.integers(-3, 4, (3, 2)).astype(float)
print("tr(RS) (2x2) =", np.trace(R @ S), "  tr(SR) (3x3) =", np.trace(S @ R))

# the same operator in another basis
P = np.array([[1.0, 2, 0], [0, 1, 1], [1, 0, 1]])
print("tr A =", np.trace(A), "  tr(P^-1 A P) =", round(np.trace(np.linalg.inv(P) @ A @ P), 10))

# X and D on polynomials of degree <= 4 (basis 1, x, ..., x^4): XD - DX is not the identity in finite dimension
n = 5
D = np.diag(np.arange(1, n), k=1)          # D(x^k) = k x^(k-1)
X = np.diag(np.ones(n - 1), k=-1)          # X(x^k) = x^(k+1), truncated at degree 4
print("DX - XD =\n", (D @ X - X @ D).astype(int), "\n(trace", np.trace(D @ X - X @ D), "-> the truncation breaks the last entry)")
```
