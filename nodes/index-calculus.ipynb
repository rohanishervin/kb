## Notation

The same symbols mean the same thing in every note.

| Symbol | Meaning |
|---|---|
| $\mathbb{N} = \{1, 2, \dots\}$, $\mathbb{Z}$, $\mathbb{Q}$, $\mathbb{R}$, $\mathbb{C}$ | natural, integer, rational, real, complex numbers |
| $[a, b]$, $(a, b)$, $[a, b)$ | closed, open, half-open intervals |
| $\lfloor x \rfloor$ | the largest integer $\le x$ |
| $\sup S$, $\inf S$ | least upper bound, greatest lower bound |
| $c_0.c_1c_2\dots$, $C_n = c_0.c_1\dots c_n$ | a decimal expansion and its truncations |
| $z = x + iy$, $\operatorname{Re} z$, $\operatorname{Im} z$, $\bar z$, $\lvert z\rvert$ | complex number, real and imaginary parts, conjugate, modulus |
| $\operatorname{cis}\theta = \cos\theta + i\sin\theta = e^{i\theta}$, $\arg z$ | point of the unit circle; argument (angle) of $z$ |
| $\omega$ | a root of unity ($\operatorname{cis}\frac{2\pi}{n}$), or the angular frequency $\frac{2\pi}{T}$ in Fourier series |
| $(a_n)$, $a_n \to a$, $\lim a_n$ | a sequence and its limit |
| $\sum a_n$, $s_n$ | a series and its partial sums |
| $f : S \to \mathbb{R}$, $S \subseteq \mathbb{R}$ | a function on a set of reals; $a$ a point or limit point of $S$ |
| $\lim_{x\to a}f$, $\lim_{x\to a^\pm}f$ | limit, one-sided limits |
| $f'(a)$, $f'_\pm(a)$, $f''$, $f^{(k)}$, $\frac{dy}{dx}$ | derivative, one-sided derivatives, higher derivatives, Leibniz notation |
| $\Delta x$, $\Delta y$; $dx$, $dy = f'(a)\,dx$ | increments; differentials (increments of the linear approximation) |
| $p_k$, $R_k = f - p_k$ | Taylor polynomial of degree $k$ at $a$, remainder |
| $\mathcal{P}$, $\Delta x_i$, $\lVert\mathcal{P}\rVert$ | a partition of $[a, b]$, its subinterval lengths, its mesh |
| $m_i, M_i$; $L(f, \mathcal{P})$, $U(f, \mathcal{P})$ | inf and sup of $f$ on the $i$-th subinterval; lower and upper sums |
| $\underline{\int_a^b} f$, $\overline{\int_a^b} f$, $\int_a^b f = \int_a^b f(x)\,dx$ | lower, upper, Riemann integral |
| $\int f(x)\,dx$, $[F]_a^b = F(b) - F(a)$ | antiderivative; evaluation |
| $\ln$, $\exp$, $e$, $a^x$ | natural logarithm, exponential, $e = \exp 1$, real powers |
| $h = \frac{b-a}{n}$; $L(n), R(n), M(n), T(n), S(n)$ | step; left, right, midpoint, trapezoid, Simpson rules |
| $\rho$ | radius of convergence of a power series |
| $\binom{\alpha}{n}$ | generalized binomial coefficient $\frac{\alpha(\alpha-1)\cdots(\alpha-n+1)}{n!}$ |
| $T$, $a_n$, $b_n$, $S_Nf$ | period, Fourier coefficients, Fourier approximation of order $N$ |
