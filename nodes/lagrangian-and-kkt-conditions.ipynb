# Constrained Optimization and KKT Conditions: First-Order Conditions

Constrained optimization is a fundamental cornerstone of mathematical economics, quantitative finance, engineering, and data science. In real-world applications, decision-makers seek to maximize utility, profit, or performance subject to limited resources, technological boundaries, or budget constraints.

This note provides a systematic, mathematically rigorous development of **First-Order Necessary Conditions (FOC)** for constrained optimization, progressing logically from simple two-variable equality constraints to the general Karush-Kuhn-Tucker (KKT) framework.



## 1\. Equality Constraints: One Constraint and Two Variables

Consider the classical two-variable optimization problem with a single equality constraint:

$$
\max_{x_1, x_2} f(x_1, x_2) \quad \text{subject to} \quad h(x_1, x_2) = c
$$

where objective function $f: \mathbb{R}^2 \to \mathbb{R}$ and constraint function $h: \mathbb{R}^2 \to \mathbb{R}$ are continuously differentiable ($C^1$).

### 1.1\. Geometric Intuition and Tangency

In the $(x_1, x_2)$ plane, the constraint $h(x_1, x_2) = c$ defines a 1-dimensional curve $C$. The objective function $f(x_1, x_2)$ can be visualized through its family of level curves (contours) $f(x_1, x_2) = k$.

Suppose $x^* = (x_1^*, x_2^*)$ is a local constrained maximum. If the level curve $f(x_1, x_2) = f(x^*)$ intersected the constraint curve $h(x_1, x_2) = c$ transversely (at a non-zero angle), we could move along the constraint curve in a direction that crosses into higher level sets of $f$, contradicting the local optimality of $x^*$.

Therefore, at the optimal point $x^*$, the level curve of $f$ must be **tangent** to the constraint curve $h$.

Equating the implicit slopes of the tangent lines to $f$ and $h$ at $x^*$:

$$
-\frac{\frac{\partial f}{\partial x_1}(x^*)}{\frac{\partial f}{\partial x_2}(x^*)} = -\frac{\frac{\partial h}{\partial x_1}(x^*)}{\frac{\partial h}{\partial x_2}(x^*)}
$$

Rearranging ratios yields a common proportionality constant $\mu \in \mathbb{R}$:

$$
\frac{\frac{\partial f}{\partial x_1}(x^*)}{\frac{\partial h}{\partial x_1}(x^*)} = \frac{\frac{\partial f}{\partial x_2}(x^*)}{\frac{\partial h}{\partial x_2}(x^*)} = \mu
$$

In vector gradient notation, this tangency condition means that the gradient vector $\nabla f(x^*)$ is parallel to the constraint gradient vector $\nabla h(x^*)$:

$$
\nabla f(x^*) = \mu \nabla h(x^*)
$$

### 1.2\. The Lagrangian Function and First-Order Conditions

To systematically transform this constrained problem into an unconstrained stationary point search, Joseph-Louis Lagrange formulated the **Lagrangian Function** $\mathcal{L}: \mathbb{R}^2 \times \mathbb{R} \to \mathbb{R}$:

::: {#def-lagrangian-single}
## Definition: Single-Constraint Lagrangian Function
$$
\mathcal{L}(x_1, x_2, \mu) = f(x_1, x_2) - \mu \big( h(x_1, x_2) - c \big)
$$
where $\mu \in \mathbb{R}$ is the **Lagrange Multiplier**.
:::

::: {#thm-lagrange-foc-single}
## Theorem: First-Order Necessary Conditions (1 Constraint, 2 Variables)
Let $f, h: \mathbb{R}^2 \to \mathbb{R}$ be $C^1$ functions. If $x^* = (x_1^*, x_2^*)$ is a local extremum of $f$ subject to $h(x) = c$, and if the **Constraint Qualification (CQ)** holds at $x^*$ (i.e., $\nabla h(x^*) \neq \mathbf{0}$), then there exists a unique multiplier $\mu^* \in \mathbb{R}$ such that $(x_1^*, x_2^*, \mu^*)$ is a stationary point of the Lagrangian:

$$
\nabla \mathcal{L}(x_1^*, x_2^*, \mu^*) = \mathbf{0}
$$

This yields a system of 3 equations in 3 unknowns $(x_1, x_2, \mu)$:

$$
\begin{cases}
\frac{\partial \mathcal{L}}{\partial x_1} = \frac{\partial f}{\partial x_1}(x^*) - \mu^* \frac{\partial h}{\partial x_1}(x^*) = 0 \\
\frac{\partial \mathcal{L}}{\partial x_2} = \frac{\partial f}{\partial x_2}(x^*) - \mu^* \frac{\partial h}{\partial x_2}(x^*) = 0 \\
\frac{\partial \mathcal{L}}{\partial \mu} = -\big( h(x_1^*, x_2^*) - c \big) = 0
\end{cases}
$$
:::

::: {.callout-tip collapse="true"}
## Proof of First-Order Necessary Conditions
Suppose $\nabla h(x^*) \neq \mathbf{0}$. Without loss of generality, assume $\frac{\partial h}{\partial x_2}(x^*) \neq 0$.

By the [Implicit Function Theorem](), there exists an open interval $I \subset \mathbb{R}$ containing $x_1^*$ and a continuously differentiable function $g: I \to \mathbb{R}$ such that $g(x_1^*) = x_2^*$ and $h(x_1, g(x_1)) = c$ for all $x_1 \in I$.

Differentiating $h(x_1, g(x_1)) = c$ with respect to $x_1$ using the Chain Rule:

$$
\frac{\partial h}{\partial x_1} + \frac{\partial h}{\partial x_2} g'(x_1) = 0 \implies g'(x_1) = - \frac{\frac{\partial h}{\partial x_1}}{\frac{\partial h}{\partial x_2}}
$$

Define the single-variable unconstrained objective $F(x_1) = f(x_1, g(x_1))$. Since $x^*$ is a local constrained maximum of $f$, $x_1^*$ is an unconstrained local maximum of $F$. Therefore, Fermat's Theorem requires $F'(x_1^*) = 0$:

$$
F'(x_1^*) = \frac{\partial f}{\partial x_1}(x^*) + \frac{\partial f}{\partial x_2}(x^*) g'(x_1^*) = 0
$$

Substitute $g'(x_1^*)$ into the derivative equation:

$$
\frac{\partial f}{\partial x_1}(x^*) - \frac{\frac{\partial f}{\partial x_2}(x^*)}{\frac{\partial h}{\partial x_2}(x^*)} \frac{\partial h}{\partial x_1}(x^*) = 0
$$

Define $\mu^* = \frac{\frac{\partial f}{\partial x_2}(x^*)}{\frac{\partial h}{\partial x_2}(x^*)}$. Then:

$$
\frac{\partial f}{\partial x_1}(x^*) - \mu^* \frac{\partial h}{\partial x_1}(x^*) = 0 \quad \text{and} \quad \frac{\partial f}{\partial x_2}(x^*) - \mu^* \frac{\partial h}{\partial x_2}(x^*) = 0
$$

Combining these with the feasibility condition $h(x^*) - c = 0$ establishes $\nabla \mathcal{L}(x_1^*, x_2^*, \mu^*) = \mathbf{0}$. $\blacksquare$
:::

::: {.callout-warning}
## Warning: Importance of Constraint Qualification (CQ)
If $\nabla h(x^*) = \mathbf{0}$ at a candidate point, the Constraint Qualification fails. At such irregular points, $\nabla f(x^*)$ might not be expressible as a multiple of $\nabla h(x^*)$, causing Lagrange's method to miss potential optimal solutions. Always verify that $\nabla h(x^*) \neq \mathbf{0}$!
:::



## 2\. Multiple Equality Constraints and Multiple Variables

Now consider generalizing to $n$ decision variables $\mathbf{x} = (x_1, \dots, x_n) \in \mathbb{R}^n$ subject to $m$ equality constraints ($m < n$):

$$
\max_{\mathbf{x} \in \mathbb{R}^n} f(\mathbf{x}) \quad \text{subject to} \quad \begin{cases} h_1(\mathbf{x}) = a_1 \\ h_2(\mathbf{x}) = a_2 \\ \vdots \\ h_m(\mathbf{x}) = a_m \end{cases}
$$

### 2.1\. Geometric Manifold & Linear Dependence

The feasible set $C_h = \{ \mathbf{x} \in \mathbb{R}^n \mid h_1(\mathbf{x}) = a_1, \dots, h_m(\mathbf{x}) = a_m \}$ represents the intersection of $m$ hypersurfaces in $\mathbb{R}^n$, forming an $(n-m)$-dimensional smooth surface (manifold).

At a local constrained optimum $\mathbf{x}^* \in C_h$, the objective gradient $\nabla f(\mathbf{x}^*)$ must be orthogonal to the tangent space of $C_h$. Since the constraint gradients $\nabla h_1(\mathbf{x}^*), \dots, \nabla h_m(\mathbf{x}^*)$ span the $m$-dimensional normal space to $C_h$, $\nabla f(\mathbf{x}^*)$ must lie inside this normal subspace.

Consequently, $\nabla f(\mathbf{x}^*)$ is a linear combination of the constraint gradients:

$$
\nabla f(\mathbf{x}^*) = \mu_1 \nabla h_1(\mathbf{x}^*) + \mu_2 \nabla h_2(\mathbf{x}^*) + \dots + \mu_m \nabla h_m(\mathbf{x}^*)
$$

### 2.2\. Multiple Constraint Lagrangian & FOC

::: {#def-lagrangian-multiple}
## Definition: Multiple-Constraint Lagrangian Function
Let $\boldsymbol{\mu} = (\mu_1, \dots, \mu_m) \in \mathbb{R}^m$. The Lagrangian $\mathcal{L}: \mathbb{R}^n \times \mathbb{R}^m \to \mathbb{R}$ is:

$$
\mathcal{L}(\mathbf{x}, \boldsymbol{\mu}) = f(\mathbf{x}) - \sum_{i=1}^m \mu_i \Big( h_i(\mathbf{x}) - a_i \Big) = f(\mathbf{x}) - \boldsymbol{\mu}^T \big( \mathbf{h}(\mathbf{x}) - \mathbf{a} \big)
$$
:::

::: {#thm-lagrange-foc-multiple}
## Theorem: First-Order Necessary Conditions (m Constraints, n Variables)
Let $f, h_1, \dots, h_m: \mathbb{R}^n \to \mathbb{R}$ be $C^1$ functions. If $\mathbf{x}^* \in \mathbb{R}^n$ is a local constrained extremum of $f$ subject to $\mathbf{h}(\mathbf{x}) = \mathbf{a}$, and if the **Linear Independence Constraint Qualification (LICQ)** holds at $\mathbf{x}^*$:

$$
\text{rank}\left( Dh(\mathbf{x}^*) \right) = \text{rank}\begin{bmatrix} \frac{\partial h_1}{\partial x_1} & \dots & \frac{\partial h_1}{\partial x_n} \\ \vdots & \ddots & \vdots \\ \frac{\partial h_m}{\partial x_1} & \dots & \frac{\partial h_m}{\partial x_n} \end{bmatrix} = m
$$

then there exists a unique multiplier vector $\boldsymbol{\mu}^* = (\mu_1^*, \dots, \mu_m^*) \in \mathbb{R}^m$ such that:

$$
\nabla \mathcal{L}(\mathbf{x}^*, \boldsymbol{\mu}^*) = \mathbf{0}
$$

This provides a system of $n + m$ equations in $n + m$ unknowns $(\mathbf{x}, \boldsymbol{\mu})$:

$$
\begin{cases}
\frac{\partial f}{\partial x_j}(\mathbf{x}^*) - \sum_{i=1}^m \mu_i^* \frac{\partial h_i}{\partial x_j}(\mathbf{x}^*) = 0, & \forall j = 1, \dots, n \\
h_i(\mathbf{x}^*) - a_i = 0, & \forall i = 1, \dots, m
\end{cases}
$$
:::

```python
import numpy as np
import matplotlib.pyplot as plt

# Define grid for 2D visualization
x1 = np.linspace(0.1, 4.0, 400)
x2 = np.linspace(0.1, 4.0, 400)
X1, X2 = np.meshgrid(x1, x2)

# Objective function: f(x1, x2) = x1^0.5 * x2^0.5 (Cobb-Douglas Utility)
F = np.sqrt(X1 * X2)

# Constraint: h(x1, x2) = p1*x1 + p2*x2 = I (Budget line: 2*x1 + 4*x2 = 16)
p1, p2, I = 2.0, 4.0, 16.0
H = p1 * X1 + p2 * X2

# Analytical optimum: x1* = I / (2*p1) = 4, x2* = I / (2*p2) = 2
x1_opt, x2_opt = 4.0, 2.0
f_opt = np.sqrt(x1_opt * x2_opt)
mu_opt = 0.5 * (x1_opt**-0.5 * x2_opt**0.5) / p1  # mu* = 1 / (2*sqrt(8)) = 0.1768

plt.figure(figsize=(9, 6), dpi=120)

# Plot objective contours
contours = plt.contour(X1, X2, F, levels=np.linspace(1.0, 4.0, 10), cmap='viridis', alpha=0.8)
plt.clabel(contours, inline=True, fontsize=9, fmt="f = %.2f")

# Plot constraint line h(x1, x2) = 16
x1_line = np.linspace(0.1, 8.0, 200)
x2_line = (I - p1 * x1_line) / p2
plt.plot(x1_line, x2_line, color='red', linewidth=2.5, label=rf"Constraint: ${p1:.0f}x_1 + {p2:.0f}x_2 = {I:.0f}$")

# Mark optimum
plt.scatter([x1_opt], [x2_opt], color='red', s=120, edgecolors='black', zorder=5, label=rf"Optimum $x^* = ({x1_opt:.0f}, {x2_opt:.0f})$")

# Draw gradient vectors at x*
grad_f = np.array([0.5 * (x2_opt/x1_opt)**0.5, 0.5 * (x1_opt/x2_opt)**0.5])
grad_h = np.array([p1, p2])

plt.quiver(x1_opt, x2_opt, grad_f[0]*4, grad_f[1]*4, angles='xy', scale_units='xy', scale=1, color='blue', width=0.008, label=r"$\nabla f(x^*)$")
plt.quiver(x1_opt, x2_opt, mu_opt*grad_h[0]*4, mu_opt*grad_h[1]*4, angles='xy', scale_units='xy', scale=1, color='purple', width=0.005, linestyle=':', label=r"$\mu^* \nabla h(x^*)$")

plt.title("Equality Constrained Optimization: Level Sets & Tangency", fontsize=13, pad=12)
plt.xlabel("$x_1$", fontsize=11)
plt.ylabel("$x_2$", fontsize=11)
plt.xlim(0, 7.5)
plt.ylim(0, 4.5)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=10, loc="upper right")
plt.tight_layout()
plt.show()
```


## 3\. Inequality Constraints: One Constraint and Two Variables

Consider optimizing an objective function subject to an inequality boundary:

$$
\max_{x_1, x_2} f(x_1, x_2) \quad \text{subject to} \quad g(x_1, x_2) \le b
$$

The feasible set $S = \{ (x_1, x_2) \in \mathbb{R}^2 \mid g(x_1, x_2) \le b \}$ is a 2-dimensional region with boundary $g(x_1, x_2) = b$.

### 3.1\. Disjunctive Analysis: Binding vs. Non-Binding Constraints

Let $x^* = (x_1^*, x_2^*)$ be a local constrained maximum. Exactly one of two mutually exclusive cases must hold:

#### Case 1: Binding (Active) Constraint ($g(x^*) = b$)

- $x^*$ lies directly on the boundary $g(x_1, x_2) = b$.
- The outward gradient vector $\nabla g(x^*)$ points strictly outward into the infeasible region $g(x) > b$.
- To maximize $f$, the direction of increasing $f$ given by $\nabla f(x^*)$ must also point outward into the infeasible region. If $\nabla f(x^*)$ pointed inward into the feasible interior, we could move inside $S$ to increase $f$, contradicting boundary optimality.
- Therefore, $\nabla f(x^*)$ and $\nabla g(x^*)$ must point in the exact same outward direction:
  $$\nabla f(x^*) = \lambda \nabla g(x^*) \quad \text{with} \quad \lambda \ge 0$$

#### Case 2: Non-Binding (Inactive) Constraint ($g(x^*) < b$)

- $x^*$ lies strictly in the interior of the feasible region.
- There exists an open ball $B_\epsilon(x^*) \subset S$. Thus, $x^*$ acts as an unconstrained local maximum of $f$.
- Fermat's Theorem requires $\nabla f(x^*) = \mathbf{0}$.
- We can still express $\nabla f(x^*) = \lambda \nabla g(x^*)$ by assigning multiplier **$\lambda = 0$**.

### 3.2\. The Combined KKT System (Single Inequality)

Combining Case 1 and Case 2 yields the Karush-Kuhn-Tucker (KKT) conditions for a single inequality constraint:

::: {#def-kkt-single}
## Definition: KKT System for Single Inequality Constraint
An optimal point $x^*$ and multiplier $\lambda^*$ must satisfy:

$$
\begin{cases}
\nabla f(x^*) - \lambda^* \nabla g(x^*) = \mathbf{0} & \text{(Stationarity)} \\
g(x^*) \le b & \text{(Primal Feasibility)} \\
\lambda^* \ge 0 & \text{(Dual Feasibility / Non-negativity)} \\
\lambda^* \big( g(x^*) - b \big) = 0 & \text{(Complementary Slackness)}
\end{cases}
$$
:::

::: {.callout-note}
## Intuition: Complementary Slackness
Complementary slackness $\lambda^* (g(x^*) - b) = 0$ enforces the disjunction between active and inactive states:

- If constraint is **inactive** ($g(x^*) < b$), then $\lambda^* = 0$.
- If multiplier is **positive** ($\lambda^* > 0$), then constraint MUST be active ($g(x^*) = b$).
:::



## 4\. Multiple Inequality Constraints

Consider extending to $k$ inequality constraints:

$$
\max_{\mathbf{x} \in \mathbb{R}^n} f(\mathbf{x}) \quad \text{subject to} \quad g_j(\mathbf{x}) \le b_j, \quad \forall j = 1, \dots, k
$$

### 4.1\. Active Constraint Index Set

At any feasible point $\mathbf{x}$, define the **active set** $\mathcal{A}(\mathbf{x})$ of constraints holding at equality:

$$
\mathcal{A}(\mathbf{x}) = \{ j \in \{1, \dots, k\} \mid g_j(\mathbf{x}) = b_j \}
$$

Inactive constraints $j \notin \mathcal{A}(\mathbf{x})$ satisfy $g_j(\mathbf{x}) < b_j$ and do not locally restrict small movements around $\mathbf{x}$.

### 4.2\. KKT First-Order Necessary Conditions (Multiple Inequalities)

::: {#thm-kkt-multiple-inequalities}
## Theorem: KKT Necessary Conditions (k Inequalities)
Let $f, g_1, \dots, g_k: \mathbb{R}^n \to \mathbb{R}$ be $C^1$ functions. If $\mathbf{x}^* \in \mathbb{R}^n$ is a local constrained maximum and **Linear Independence Constraint Qualification (LICQ)** holds for active constraints (i.e. $\{\nabla g_j(\mathbf{x}^*) \mid j \in \mathcal{A}(\mathbf{x}^*)\}$ are linearly independent), then there exists a non-negative multiplier vector $\boldsymbol{\lambda}^* = (\lambda_1^*, \dots, \lambda_k^*) \in \mathbb{R}^k$ satisfying:

$$
\begin{aligned}
\text{1. Stationarity:} & \quad \nabla f(\mathbf{x}^*) - \sum_{j=1}^k \lambda_j^* \nabla g_j(\mathbf{x}^*) = \mathbf{0} \\
\text{2. Primal Feasibility:} & \quad g_j(\mathbf{x}^*) \le b_j, \quad \forall j = 1, \dots, k \\
\text{3. Dual Feasibility:} & \quad \lambda_j^* \ge 0, \quad \forall j = 1, \dots, k \\
\text{4. Complementary Slackness:} & \quad \lambda_j^* \big( g_j(\mathbf{x}^*) - b_j \big) = 0, \quad \forall j = 1, \dots, k
\end{aligned}
$$
:::



## 5\. General Case: Combined Equality and Inequality Constraints

We now present the complete, general Karush-Kuhn-Tucker (KKT) framework incorporating both $m$ equality constraints and $k$ inequality constraints simultaneously.

### 5.1\. General Problem Statement

$$
\max_{\mathbf{x} \in \mathbb{R}^n} f(\mathbf{x}) \quad \text{subject to} \quad \begin{cases} g_j(\mathbf{x}) \le b_j, & j = 1, \dots, k \\ h_i(\mathbf{x}) = a_i, & i = 1, \dots, m \end{cases}
$$

### 5.2\. General KKT First-Order Necessary Conditions

::: {#def-general-kkt-lagrangian}
## Definition: General Lagrangian Function
The general Lagrangian $\mathcal{L}: \mathbb{R}^n \times \mathbb{R}^k \times \mathbb{R}^m \to \mathbb{R}$ is:

$$
\mathcal{L}(\mathbf{x}, \boldsymbol{\lambda}, \boldsymbol{\mu}) = f(\mathbf{x}) - \sum_{j=1}^k \lambda_j \Big( g_j(\mathbf{x}) - b_j \Big) - \sum_{i=1}^m \mu_i \Big( h_i(\mathbf{x}) - a_i \Big)
$$
:::

::: {#thm-general-kkt-conditions}
## Theorem: Complete General KKT First-Order Necessary Conditions
Let $f, g_1, \dots, g_k, h_1, \dots, h_m: \mathbb{R}^n \to \mathbb{R}$ be continuously differentiable ($C^1$).

If $\mathbf{x}^* \in \mathbb{R}^n$ is a local maximum of $f$ subject to the constraints, and if **LICQ** holds at $\mathbf{x}^*$ (i.e. the active constraint gradients $\{ \nabla g_j(\mathbf{x}^*) \mid j \in \mathcal{A}(\mathbf{x}^*) \} \cup \{ \nabla h_i(\mathbf{x}^*) \mid i=1, \dots, m \}$ are linearly independent), then there exist unique multiplier vectors $\boldsymbol{\lambda}^* = (\lambda_1^*, \dots, \lambda_k^*) \in \mathbb{R}^k$ and $\boldsymbol{\mu}^* = (\mu_1^*, \dots, \mu_m^*) \in \mathbb{R}^m$ satisfying:

$$
\begin{aligned}
\text{1. Stationarity:} & \quad \nabla_x \mathcal{L}(\mathbf{x}^*, \boldsymbol{\lambda}^*, \boldsymbol{\mu}^*) = \nabla f(\mathbf{x}^*) - \sum_{j=1}^k \lambda_j^* \nabla g_j(\mathbf{x}^*) - \sum_{i=1}^m \mu_i^* \nabla h_i(\mathbf{x}^*) = \mathbf{0} \\
\text{2. Primal Feasibility:} & \quad g_j(\mathbf{x}^*) \le b_j \; (\forall j=1, \dots, k), \quad h_i(\mathbf{x}^*) = a_i \; (\forall i=1, \dots, m) \\
\text{3. Dual Feasibility:} & \quad \lambda_j^* \ge 0 \; (\forall j=1, \dots, k), \quad (\mu_i^* \text{ unrestricted in sign}) \\
\text{4. Complementary Slackness:} & \quad \lambda_j^* \big( g_j(\mathbf{x}^*) - b_j \big) = 0 \; (\forall j=1, \dots, k)
\end{aligned}
$$
:::

### 5.3\. Maximization vs. Minimization Sign Conventions

The sign of inequality multipliers $\lambda_j$ depends on problem orientation (Maximization vs Minimization) and constraint direction ($g \le b$ vs $g \ge b$).

| Problem Type | Constraint Type | Lagrangian Definition | Multiplier Sign |
| :--- | :--- | :--- | :--- |
| **Maximization** $\max f$ | $g_j(\mathbf{x}) \le b_j$ | $\mathcal{L} = f - \sum \lambda_j (g_j - b_j)$ | $\lambda_j \ge 0$ |
| **Maximization** $\max f$ | $g_j(\mathbf{x}) \ge b_j$ | $\mathcal{L} = f - \sum \lambda_j (g_j - b_j)$ | $\lambda_j \le 0$ |
| **Minimization** $\min f$ | $g_j(\mathbf{x}) \ge b_j$ | $\mathcal{L} = f - \sum \lambda_j (g_j - b_j)$ | $\lambda_j \ge 0$ |
| **Minimization** $\min f$ | $g_j(\mathbf{x}) \le b_j$ | $\mathcal{L} = f - \sum \lambda_j (g_j - b_j)$ | $\lambda_j \le 0$ |


```python
import numpy as np
import matplotlib.pyplot as plt

# Grid setup
x1 = np.linspace(0, 5, 400)
x2 = np.linspace(0, 5, 400)
X1, X2 = np.meshgrid(x1, x2)

# Objective: f(x1, x2) = -(x1 - 4)^2 - (x2 - 4)^2 (Unconstrained max at (4,4))
F = -(X1 - 4)**2 - (X2 - 4)**2

# Feasible Region: g1(x) = x1 + x2 <= 4 and g2(x) = x1 <= 3
feasible = (X1 + X2 <= 4) & (X1 <= 3) & (X1 >= 0) & (X2 >= 0)

plt.figure(figsize=(9, 6), dpi=120)

# Shade feasible region
plt.imshow(feasible.astype(int), extent=(0, 5, 0, 5), origin='lower', cmap='Blues', alpha=0.3)

# Objective contours
contours = plt.contour(X1, X2, F, levels=12, cmap='magma')
plt.clabel(contours, inline=True, fontsize=9, fmt="f = %.1f")

# Constraint boundaries (changed \le to \leq)
plt.plot(x1, 4 - x1, color='darkblue', linewidth=2, label=r"$g_1(x) = x_1 + x_2 \leq 4$")
plt.axvline(3, color='darkgreen', linewidth=2, label=r"$g_2(x) = x_1 \leq 3$")

# Optimal point under constraints: x* = (3, 1)
x1_opt, x2_opt = 3.0, 1.0
plt.scatter([x1_opt], [x2_opt], color='red', s=120, edgecolors='black', zorder=5, label="KKT Optimal $x^* = (3, 1)$")

# Draw KKT Gradient balance vectors at x*
plt.quiver(x1_opt, x2_opt, 2, 6, angles='xy', scale_units='xy', scale=3, color='red', width=0.008, label=r"$\nabla f(x^*)$")
plt.quiver(x1_opt, x2_opt, 1, 1, angles='xy', scale_units='xy', scale=3, color='blue', width=0.006, label=r"$\nabla g_1(x^*)$")
plt.quiver(x1_opt, x2_opt, 1, 0, angles='xy', scale_units='xy', scale=3, color='green', width=0.006, label=r"$\nabla g_2(x^*)$")

plt.title("KKT Inequality Optimization: Feasible Region & Active Constraints", fontsize=13, pad=12)
plt.xlabel("$x_1$", fontsize=11)
plt.ylabel("$x_2$", fontsize=11)
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.grid(True, linestyle=':', alpha=0.6)
plt.legend(fontsize=9.5, loc="upper left")
plt.tight_layout()
plt.show()
```
