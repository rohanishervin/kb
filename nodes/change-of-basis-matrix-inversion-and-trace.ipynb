# Change of Basis, Matrix Inversion, and Trace

## 1\. Change of Basis for Linear Maps and Endomorphisms

Let $V$ be a finite-dimensional [vector space](vector-spaces.ipynb) over a [field](field.ipynb) $K$ with $\dim V = n$. In [matrix representations](linear-maps-inverses-and-matrices.ipynb#def-matrix-representation), every choice of ordered [basis](linear-independence-basis-dimension.ipynb#def-basis) assigns to each [linear operator](linear-maps-and-rank-nullity.ipynb#def-linear-map) $f: V \to V$ a square matrix $[f]_B \in M_{n \times n}(K)$. We now investigate how this matrix transforms when the coordinate system is changed.

### 1.1\. Composition and Index Chain Rule

Let $V_1, V_2, V_3$ be vector spaces over $K$ equipped with ordered bases $B_1, B_2, B_3$ respectively. If $f: V_1 \to V_2$ and $g: V_2 \to V_3$ are linear maps, their composition $g \circ f: V_1 \to V_3$ satisfies the [matrix composition identity](linear-maps-inverses-and-matrices.ipynb#thm-composition-matrix):

$$[g \circ f]_{B_1}^{B_3} = [g]_{B_2}^{B_3} [f]_{B_1}^{B_2}$$

Notice the intermediate basis $B_2$ cancels across the product from northwest to southeast, precisely mirroring the chain rule.

---

::: {#def-transition-matrix}
## Transition Matrix (Change-of-Basis Matrix)

Let $B = (v_1, \dots, v_n)$ and $B' = (v'_1, \dots, v'_n)$ be two ordered bases for $V$.

The **transition matrix** (or **change-of-basis matrix**) from $B'$ to $B$, denoted $P = [\operatorname{id}_V]_{B'}^B$, is the matrix representing the identity operator $\operatorname{id}_V: V \to V$ with domain basis $B'$ and codomain basis $B$:

$$v'_j = \operatorname{id}_V(v'_j) = \sum_{i=1}^n P_{ij} v_i \qquad (j = 1, \dots, n)$$

Equivalently, the $j$-th column of $P$ consists of the coordinate vector of the $j$-th new basis vector $v'_j$ expressed in terms of the old basis $B$.
:::

::: {#thm-transition-invertibility}
## Invertibility of the Transition Matrix

Let $B$ and $B'$ be ordered bases for $V$, and let $P = [\operatorname{id}_V]_{B'}^B$. Then $P$ is [invertible](linear-maps-inverses-and-matrices.ipynb#cor-equal-dim-invertibility), and its inverse is the transition matrix from $B$ to $B'$:

$$P^{-1} = [\operatorname{id}_V]_B^{B'}$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-transition-invertibility
Consider the identity map $\operatorname{id}_V: V \to V$ composed with itself: $\operatorname{id}_V = \operatorname{id}_V \circ \operatorname{id}_V$. Viewing the domain, intermediate, and codomain spaces equipped with bases $B, B', B$ respectively, the composition formula gives:

$$[\operatorname{id}_V]_B^B = [\operatorname{id}_V]_{B'}^B [\operatorname{id}_V]_B^{B'}$$

Now observe the left-hand side. For any basis $B = (v_1, \dots, v_n)$, $\operatorname{id}_V(v_j) = v_j = 0 v_1 + \dots + 1 v_j + \dots + 0 v_n$. Thus the $j$-th column of $[\operatorname{id}_V]_B^B$ is the standard basis vector $e_j$, which means:

$$[\operatorname{id}_V]_B^B = I_n$$

where $I_n$ is the $n \times n$ identity matrix. Similarly, reversing the roles of $B$ and $B'$:

$$[\operatorname{id}_V]_B^{B'} [\operatorname{id}_V]_{B'}^B = [\operatorname{id}_V]_{B'}^{B'} = I_n$$

Setting $P = [\operatorname{id}_V]_{B'}^B$ and $Q = [\operatorname{id}_V]_B^{B'}$, we have $P Q = I_n$ and $Q P = I_n$. Hence $P$ is invertible and $P^{-1} = Q = [\operatorname{id}_V]_B^{B'}$. $\blacksquare$
:::

---

::: {#thm-operator-change-of-basis}
## Transformation Law for Linear Operators

Let $f: V \to V$ be a linear operator, and let $B, B'$ be ordered bases for $V$. Let $P = [\operatorname{id}_V]_{B'}^B$ be the transition matrix from $B'$ to $B$. Then the matrix representations $[f]_B = [f]_B^B$ and $[f]_{B'} = [f]_{B'}^{B'}$ are related by:

$$[f]_{B'} = P^{-1} [f]_B P$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-operator-change-of-basis
We factor the operator $f$ through the identity maps:

$$f = \operatorname{id}_V \circ f \circ \operatorname{id}_V$$

Consider the sequence of spaces and assigned bases:

$$(V, B') \xrightarrow{\operatorname{id}_V} (V, B) \xrightarrow{f} (V, B) \xrightarrow{\operatorname{id}_V} (V, B')$$

Applying the composition rule for matrix representations successively:

$$\begin{aligned}
[f]_{B'}^{B'} &= [\operatorname{id}_V \circ f \circ \operatorname{id}_V]_{B'}^{B'} \\
&= [\operatorname{id}_V]_B^{B'} [f]_B^B [\operatorname{id}_V]_{B'}^B
\end{aligned}$$

By @thm-transition-invertibility, $[\operatorname{id}_V]_B^{B'} = P^{-1}$ and $[\operatorname{id}_V]_{B'}^B = P$. Substituting these matrices yields:

$$[f]_{B'} = P^{-1} [f]_B P \qquad \blacksquare$$
:::

::: {#def-matrix-similarity}
## Similar (Conjugate) Matrices

Two square matrices $A, B \in M_{n \times n}(K)$ are called **similar** (or **conjugate**), written $A \sim B$, if there exists an invertible matrix $P \in GL(n, K)$ such that:

$$B = P^{-1} A P$$

Similarity is an equivalence relation on $M_{n \times n}(K)$ (reflexive via $P = I$, symmetric via $P^{-1}$, and transitive via $P_1 P_2$). Two matrices represent the exact same linear operator $f: V \to V$ under different choices of basis if and only if they are similar.
:::



## 2\. The Intrinsic Determinant of a Linear Operator

In the study of [multilinear forms and determinants](determinants-multilinear-forms-and-orientation.ipynb#def-determinant), the determinant was defined as a function on $n \times n$ matrices. We now demonstrate that the determinant is an intrinsic, coordinate-free invariant of linear endomorphisms.

::: {#thm-det-operator-well-defined}
## Basis Independence of the Determinant

Let $V$ be a finite-dimensional vector space over $K$, and let $f: V \to V$ be a linear operator. If $B$ and $B'$ are any two ordered bases of $V$, then:

$$\det([f]_{B'}) = \det([f]_B)$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-det-operator-well-defined
By @thm-operator-change-of-basis, $[f]_{B'} = P^{-1} [f]_B P$ for the invertible transition matrix $P = [\operatorname{id}_V]_{B'}^B$.

By the [multiplicativity of determinants](determinants-multilinear-forms-and-orientation.ipynb#thm-det-multiplicative), $\det(X Y) = \det(X) \det(Y)$ for all square matrices. Applying this property:

$$\begin{aligned}
\det([f]_{B'}) &= \det(P^{-1} [f]_B P) \\
&= \det(P^{-1}) \det([f]_B) \det(P)
\end{aligned}$$

Because scalars in the field $K$ commute, and because $\det(P^{-1}) = (\det(P))^{-1}$ by the [invertibility criterion](determinants-multilinear-forms-and-orientation.ipynb#thm-det-invertibility):

$$\det([f]_{B'}) = \det(P^{-1}) \det(P) \det([f]_B) = 1 \cdot \det([f]_B) = \det([f]_B) \qquad \blacksquare$$
:::

::: {#def-det-operator}
## Determinant of a Linear Operator

Let $f: V \to V$ be a linear operator on a finite-dimensional vector space $V$. The **determinant** of $f$, denoted $\det(f)$, is defined as:

$$\det(f) := \det([f]_B)$$

where $B$ is any ordered basis of $V$. By @thm-det-operator-well-defined, this value is independent of the choice of $B$.
:::

---

### 2.1\. Geometric Characterization in Real Vector Spaces

When $V$ is an $n$-dimensional vector space over the real field $\mathbb{R}$, $\det(f) \in \mathbb{R}$ conveys two distinct, fundamental geometric properties:

1. **Invertibility and Linear Independence:**
   - $\det(f) = 0 \iff f$ is not invertible (its image has dimension $< n$).
   - $\det(f) \neq 0 \iff f$ is an automorphism (maps any basis to another basis).
   - This provides a mechanical test for linear independence: $n$ vectors $u_1, \dots, u_n \in V$ are linearly independent if and only if $\det([u_1, \dots, u_n]_B) \neq 0$.

2. **Orientation:**
   - If $\det(f) > 0$, $f$ preserves [orientation](determinants-multilinear-forms-and-orientation.ipynb#def-equioriented-bases): for any basis $B$, $f(B)$ belongs to the same orientation class as $B$.
   - If $\det(f) < 0$, $f$ reverses orientation: $f(B)$ and $B$ have opposite chiralities (e.g., right-handed becomes left-handed).

::: {#thm-volume-scaling}
## Volume Scaling by Linear Operators

Let $f: V \to V$ be a linear operator on an $n$-dimensional real vector space $V$. Let $u_1, \dots, u_n \in V$ generate a [parallelotope](determinants-multilinear-forms-and-orientation.ipynb#def-parallelotope) $P(u_1, \dots, u_n) = \left\{ \sum_{i=1}^n t_i u_i \;\middle|\; 0 \le t_i \le 1 \right\}$. Then the image parallelotope $P(f(u_1), \dots, f(u_n))$ has $n$-dimensional volume:

$$\operatorname{Vol}(f(P)) = |\det(f)| \operatorname{Vol}(P)$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-volume-scaling
Fix a standard reference basis $B = (e_1, \dots, e_n)$ for $V$ with respect to which volumes are measured. By definition, $\operatorname{Vol}(P(u_1, \dots, u_n)) = |\det([u_1, \dots, u_n]_B)|$.

Let $g: V \to V$ be the unique linear operator defined by $g(e_i) = u_i$ for $i = 1, \dots, n$. The matrix of $g$ relative to $B$ has columns given by the coordinates of $u_1, \dots, u_n$:

$$[g]_B = [u_1, \dots, u_n]_B \implies \operatorname{Vol}(P(u_1, \dots, u_n)) = |\det(g)|$$

Now consider the image vectors $v_i = f(u_i) = f(g(e_i)) = (f \circ g)(e_i)$. The matrix of the composite operator $f \circ g$ with respect to $B$ has columns $v_1, \dots, v_n$:

$$[f \circ g]_B = [v_1, \dots, v_n]_B$$

Therefore, the volume of the transformed parallelotope is:

$$\operatorname{Vol}(f(P)) = |\det([v_1, \dots, v_n]_B)| = |\det(f \circ g)|$$

By multiplicativity of the determinant of linear operators:

$$\operatorname{Vol}(f(P)) = |\det(f) \det(g)| = |\det(f)| \cdot |\det(g)| = |\det(f)| \operatorname{Vol}(P) \qquad \blacksquare$$
:::

::: {.callout-note}
## Intuition: The Jacobian Determinant in Multivariable Integration
This theorem provides the fundamental linear-algebraic foundation for the change-of-variables formula in multivariable analysis:

$$\int_{\Phi(\Omega)} g(y) \, dy = \int_\Omega g(\Phi(x)) \, |\det(D\Phi(x))| \, dx$$

At each point $x$, the smooth mapping $\Phi$ is approximated to first order by its derivative linear map $D\Phi(x)$. The infinitesimal volume element $dx_1 \dots dx_n$ is a miniature parallelotope whose volume scales precisely by the factor $|\det(D\Phi(x))|$.
:::



## 3\. Determinant of the Transpose

Recall from [dual maps and transposes](dual-maps-transposes-and-affine-subspaces.ipynb#def-transpose-matrix) that for a matrix $A \in M_{n \times n}(K)$ with entries $(a_{ij})$, its transpose $A^T$ has entries $(a^T)_{ij} = a_{ji}$.

::: {#thm-det-transpose}
## Determinant of Transpose

For any square matrix $A \in M_{n \times n}(K)$ over an arbitrary field $K$:

$$\det(A^T) = \det(A)$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-det-transpose
Recall the [Leibniz formula](determinants-multilinear-forms-and-orientation.ipynb#thm-alternating-form-basis-characterization) for the determinant of $A$:

$$\det(A) = \sum_{\sigma \in S_n} \operatorname{sgn}(\sigma) \prod_{j=1}^n a_{\sigma(j), j}$$

Let $B = A^T$, so $b_{ij} = a_{ji}$. Applying the Leibniz formula to $B$:

$$\det(A^T) = \sum_{\sigma \in S_n} \operatorname{sgn}(\sigma) \prod_{j=1}^n b_{\sigma(j), j} = \sum_{\sigma \in S_n} \operatorname{sgn}(\sigma) \prod_{j=1}^n a_{j, \sigma(j)}$$

Consider the product $\prod_{j=1}^n a_{j, \sigma(j)}$ for a fixed permutation $\sigma \in S_n$. Because multiplication in the field $K$ is commutative, we can reorder the $n$ factors by their second index. For each $i \in \{1, \dots, n\}$, because $\sigma$ is a bijection of $\{1, \dots, n\}$, there exists a unique $j$ such that $\sigma(j) = i$, namely $j = \sigma^{-1}(i)$.

Substituting $j = \sigma^{-1}(i)$ into the product:

$$\prod_{j=1}^n a_{j, \sigma(j)} = \prod_{i=1}^n a_{\sigma^{-1}(i), i}$$

Next, examine the signature $\operatorname{sgn}(\sigma)$. Since $\sigma \circ \sigma^{-1} = \operatorname{id}$, we have $\operatorname{sgn}(\sigma) \operatorname{sgn}(\sigma^{-1}) = \operatorname{sgn}(\operatorname{id}) = 1$. Since signatures are $\pm 1$, this requires:

$$\operatorname{sgn}(\sigma^{-1}) = \operatorname{sgn}(\sigma)$$

Therefore:

$$\det(A^T) = \sum_{\sigma \in S_n} \operatorname{sgn}(\sigma^{-1}) \prod_{i=1}^n a_{\sigma^{-1}(i), i}$$

As $\sigma$ ranges over the entire symmetric group $S_n$, the inverse $\tau = \sigma^{-1}$ also ranges over $S_n$ bijectively (the inversion map $\sigma \mapsto \sigma^{-1}$ is a bijection of $S_n$). Substituting $\tau = \sigma^{-1}$:

$$\det(A^T) = \sum_{\tau \in S_n} \operatorname{sgn}(\tau) \prod_{i=1}^n a_{\tau(i), i} = \det(A) \qquad \blacksquare$$
:::

::: {#cor-row-column-duality}
## Row-Column Duality for Determinants

Because transposition swaps the rows and columns of a matrix while preserving its determinant:

1. The determinant is an **alternating $n$-linear form with respect to the rows** of $A$.
2. Swapping any two rows multiplies the determinant by $-1$.
3. If any two rows of $A$ are identical, or if the rows are linearly dependent, $\det(A) = 0$.
4. Adding a scalar multiple of one row to another row leaves $\det(A)$ unchanged.
:::



## 4\. Laplace Expansion (Cofactor Expansion)

The Leibniz formula expresses the determinant as a sum of $n!$ terms. Laplace expansion reduces the computation of an $n \times n$ determinant to $n$ determinants of size $(n-1) \times (n-1)$.

::: {#def-minors-and-cofactors}
## Submatrices, Minors, and Cofactors

Let $A \in M_{n \times n}(K)$ with entries $(a_{ij})$.

1. The **$(i, j)$-submatrix**, denoted $A_{ij} \in M_{(n-1) \times (n-1)}(K)$, is the matrix obtained by deleting the $i$-th row and the $j$-th column from $A$.
2. The **$(i, j)$-minor** of $A$ is the scalar $M_{ij} = \det(A_{ij})$.
3. The **$(i, j)$-cofactor** of $A$, denoted $C_{ij}$, is defined by:
   $$C_{ij} := (-1)^{i+j} \det(A_{ij})$$
:::

::: {#thm-laplace-expansion}
## Laplace Expansion Theorem

Let $A = (a_{ij}) \in M_{n \times n}(K)$.

1. **Expansion along Column $j$:** For any fixed column index $j \in \{1, \dots, n\}$:
   $$\det(A) = \sum_{i=1}^n a_{ij} C_{ij} = \sum_{i=1}^n (-1)^{i+j} a_{ij} \det(A_{ij})$$

2. **Expansion along Row $i$:** For any fixed row index $i \in \{1, \dots, n\}$:
   $$\det(A) = \sum_{j=1}^n a_{ij} C_{ij} = \sum_{j=1}^n (-1)^{i+j} a_{ij} \det(A_{ij})$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-laplace-expansion
**Step 1: The special case $i = 1, j = 1$:**

Consider the terms in the Leibniz formula for $\det(A)$ containing $a_{11}$:

$$\det(A) = \sum_{\sigma \in S_n} \operatorname{sgn}(\sigma) a_{\sigma(1), 1} a_{\sigma(2), 2} \cdots a_{\sigma(n), n}$$

Every permutation $\sigma$ with $\sigma(1) = 1$ restricts to a permutation $\sigma'$ of $\{2, 3, \dots, n\}$. The number of inversions in $\sigma$ is identical to that in $\sigma'$ because $1$ is mapped to $1$ and cannot participate in any inversion with elements $> 1$. Thus $\operatorname{sgn}(\sigma) = \operatorname{sgn}(\sigma')$. Factoring out $a_{11}$ from all such terms:

$$\sum_{\sigma \in S_n, \sigma(1) = 1} \operatorname{sgn}(\sigma) a_{11} \prod_{k=2}^n a_{\sigma(k), k} = a_{11} \sum_{\sigma' \in S_{n-1}} \operatorname{sgn}(\sigma') \prod_{k=2}^n a_{\sigma'(k), k} = a_{11} \det(A_{11})$$

**Step 2: Determining the sign $(-1)^{i+j}$ for general $(i, j)$:**

To evaluate the coefficient of $a_{ij}$, we move entry $(i, j)$ to the top-left position $(1, 1)$ using adjacent row and column swaps:

1. Swap row $i$ with row $i-1$, then with row $i-2$, $\dots$, until it becomes row $1$. This requires exactly $i - 1$ successive adjacent row swaps and preserves the relative order of all other rows.
2. Swap column $j$ with column $j-1$, $\dots$, until it becomes column $1$. This requires exactly $j - 1$ successive adjacent column swaps and preserves the relative order of all other columns.

The resulting matrix $\tilde{A}$ has $a_{ij}$ in position $(1, 1)$, and deleting its first row and first column yields exactly the submatrix $A_{ij}$ of the original matrix $A$.

Each swap of adjacent rows or columns multiplies the determinant by $-1$. The total number of swaps is $(i - 1) + (j - 1) = i + j - 2$. Therefore:

$$\det(A) = (-1)^{(i-1) + (j-1)} \det(\tilde{A}) = (-1)^{i+j} \det(\tilde{A})$$

By Step 1, the contribution of the $(1, 1)$-entry of $\tilde{A}$ (which is $a_{ij}$) to $\det(\tilde{A})$ is $a_{ij} \det(A_{ij})$. Multiplying by the parity factor $(-1)^{i+j}$, the total contribution of $a_{ij}$ to $\det(A)$ is:

$$(-1)^{i+j} a_{ij} \det(A_{ij}) = a_{ij} C_{ij}$$

**Step 3: Column expansion:**

In the Leibniz formula, each permutation $\sigma \in S_n$ assigns column $j$ to a unique row index $i = \sigma(j) \in \{1, \dots, n\}$. Partitioning the symmetric group according to the value of $\sigma(j)$:

$$S_n = \bigsqcup_{i=1}^n \{\sigma \in S_n \mid \sigma(j) = i\}$$

Summing the contributions over all $i = 1, \dots, n$ yields the column expansion:

$$\det(A) = \sum_{i=1}^n a_{ij} C_{ij}$$

Row expansion follows immediately by applying column expansion to the transpose $A^T$ and using @thm-det-transpose. $\blacksquare$
:::



## 5\. The Classical Adjoint, Matrix Inversion, and Cramer's Rule

### 5.1\. The Adjugate Matrix and the Fundamental Identity

::: {#def-adjugate-matrix}
## The Adjugate (Classical Adjoint) Matrix

Let $A \in M_{n \times n}(K)$. The **adjugate matrix** (or **classical adjoint**) of $A$, denoted $\operatorname{adj}(A)$, is the transpose of the cofactor matrix $C = (C_{ij})$:

$$[\operatorname{adj}(A)]_{ij} := C_{ji} = (-1)^{i+j} \det(A_{ji})$$
:::

::: {#thm-adjugate-fundamental-identity}
## Fundamental Identity of the Adjugate

For any square matrix $A \in M_{n \times n}(K)$:

$$A \operatorname{adj}(A) = \operatorname{adj}(A) A = \det(A) I_n$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-adjugate-fundamental-identity
We evaluate the $(i, j)$-entry of the product $A \operatorname{adj}(A)$:

$$(A \operatorname{adj}(A))_{ij} = \sum_{k=1}^n a_{ik} [\operatorname{adj}(A)]_{kj} = \sum_{k=1}^n a_{ik} C_{jk}$$

We analyze two cases:

1. **Diagonal entries ($i = j$):**
   $$\sum_{k=1}^n a_{ik} C_{ik}$$
   By @thm-laplace-expansion, this is precisely the Laplace expansion of $\det(A)$ along row $i$. Hence $(A \operatorname{adj}(A))_{ii} = \det(A)$.

2. **Off-diagonal entries ($i \neq j$):**
   $$\sum_{k=1}^n a_{ik} C_{jk} = \sum_{k=1}^n (-1)^{j+k} a_{ik} \det(A_{jk})$$
   Consider the matrix $\tilde{A}$ obtained from $A$ by replacing row $j$ with row $i$, while leaving all other rows identical. Expanding $\det(\tilde{A})$ along its $j$-th row gives:
   $$\det(\tilde{A}) = \sum_{k=1}^n \tilde{a}_{jk} (-1)^{j+k} \det(\tilde{A}_{jk}) = \sum_{k=1}^n a_{ik} (-1)^{j+k} \det(A_{jk}) = \sum_{k=1}^n a_{ik} C_{jk}$$
   Notice that deleting row $j$ from $\tilde{A}$ leaves the exact same submatrix as deleting row $j$ from $A$, so $\tilde{A}_{jk} = A_{jk}$.
   However, the matrix $\tilde{A}$ has row $i$ and row $j$ completely identical. Because the determinant is an alternating multilinear form on rows (@cor-row-column-duality), any matrix with two identical rows has determinant zero: $\det(\tilde{A}) = 0$.
   Therefore, for $i \neq j$:
   $$\sum_{k=1}^n a_{ik} C_{jk} = 0$$

Combining both cases:

$$(A \operatorname{adj}(A))_{ij} = \det(A) \delta_{ij} \implies A \operatorname{adj}(A) = \det(A) I_n$$

An identical argument expanding along columns shows that $\operatorname{adj}(A) A = \det(A) I_n$. $\blacksquare$
:::

::: {#cor-matrix-inverse-formula}
## Explicit Matrix Inversion Formula

A square matrix $A \in M_{n \times n}(K)$ is invertible if and only if $\det(A) \neq 0$. In this case, its inverse is given explicitly by:

$$A^{-1} = \frac{1}{\det(A)} \operatorname{adj}(A)$$
:::

---

### 5.2\. Cramer's Rule

::: {#thm-cramers-rule}
## Cramer's Rule

Let $A \in M_{n \times n}(K)$ with $\det(A) \neq 0$, and let $b = (b_1, \dots, b_n)^T \in K^n$. The unique solution $x = (x_1, \dots, x_n)^T$ to the linear system $A x = b$ is given by:

$$x_i = \frac{\det(A_i(b))}{\det(A)} \qquad (i = 1, \dots, n)$$

where $A_i(b)$ is the matrix obtained by replacing the $i$-th column of $A$ with the vector $b$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-cramers-rule
Because $\det(A) \neq 0$, $A$ is invertible and the unique solution is $x = A^{-1} b$. By @cor-matrix-inverse-formula:

$$x = \frac{1}{\det(A)} \operatorname{adj}(A) b$$

Examining the $i$-th component $x_i$:

$$x_i = \frac{1}{\det(A)} \sum_{k=1}^n [\operatorname{adj}(A)]_{ik} b_k = \frac{1}{\det(A)} \sum_{k=1}^n C_{ki} b_k = \frac{1}{\det(A)} \sum_{k=1}^n (-1)^{k+i} b_k \det(A_{ki})$$

Now consider the matrix $A_i(b)$ obtained by substituting $b$ into column $i$ of $A$. The $(k, i)$-submatrix of $A_i(b)$ (deleting row $k$ and column $i$) does not involve column $i$ at all, and is therefore identical to the submatrix $A_{ki}$ of $A$.

Expanding $\det(A_i(b))$ along its $i$-th column by @thm-laplace-expansion:

$$\det(A_i(b)) = \sum_{k=1}^n (-1)^{k+i} b_k \det(A_{ki})$$

Substituting this numerator gives:

$$x_i = \frac{\det(A_i(b))}{\det(A)} \qquad \blacksquare$$
:::

::: {.callout-warning}
## Computational Efficiency: Cramer's Rule vs Gaussian Elimination
While Cramer's rule and the adjugate formula provide indispensable closed-form symbolic formulas for proofs and sensitivity analysis, computing an $n \times n$ inverse or solving a system via determinants requires computing $n+1$ determinants, demanding $O(n \cdot n!) = O((n+1)!)$ operations by direct expansion. In contrast, Gaussian elimination requires only $O(n^3)$ arithmetic operations and is the standard algorithm for numerical computation.
:::



## 6\. The Trace of Matrices and Linear Operators

A matrix $A \in M_{n \times n}(K)$ contains $n^2$ scalar entries. Aside from the determinant, there is a second fundamental linear scalar invariant: the **trace**.

::: {#def-trace-matrix}
## Trace of a Square Matrix

Let $A = (a_{ij}) \in M_{n \times n}(K)$. The **trace** of $A$, denoted $\operatorname{Tr}(A)$, is the sum of its main diagonal entries:

$$\operatorname{Tr}(A) := \sum_{i=1}^n a_{ii}$$
:::

::: {#thm-trace-properties}
## Linearity and Transpose Invariance of the Trace

For all $A, B \in M_{n \times n}(K)$ and $\alpha, \beta \in K$:

1. **Linearity:** $\operatorname{Tr}(\alpha A + \beta B) = \alpha \operatorname{Tr}(A) + \beta \operatorname{Tr}(B)$.
2. **Transpose Invariance:** $\operatorname{Tr}(A^T) = \operatorname{Tr}(A)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-trace-properties
1. $(\alpha A + \beta B)_{ii} = \alpha a_{ii} + \beta b_{ii}$. Summing from $i=1$ to $n$:
   $$\sum_{i=1}^n (\alpha a_{ii} + \beta b_{ii}) = \alpha \sum_{i=1}^n a_{ii} + \beta \sum_{i=1}^n b_{ii} = \alpha \operatorname{Tr}(A) + \beta \operatorname{Tr}(B)$$
2. Transposition leaves diagonal entries unchanged: $(A^T)_{ii} = a_{ii}$, so $\operatorname{Tr}(A^T) = \operatorname{Tr}(A)$. $\blacksquare$
:::

---

::: {#thm-trace-cyclic}
## Commutativity under Trace (Cyclic Property)

Let $A \in M_{m \times n}(K)$ and $B \in M_{n \times m}(K)$. Then:

$$\operatorname{Tr}(AB) = \operatorname{Tr}(BA)$$

In particular, when $m = n$, $\operatorname{Tr}(AB) = \operatorname{Tr}(BA)$ holds even though matrix multiplication is generally non-commutative ($AB \neq BA$).
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-trace-cyclic
The product $AB$ is an $m \times m$ matrix with $(i, i)$-entry $(AB)_{ii} = \sum_{k=1}^n a_{ik} b_{ki}$. Thus:

$$\operatorname{Tr}(AB) = \sum_{i=1}^m (AB)_{ii} = \sum_{i=1}^m \sum_{k=1}^n a_{ik} b_{ki}$$

Similarly, $BA$ is an $n \times n$ matrix with $(k, k)$-entry $(BA)_{kk} = \sum_{i=1}^m b_{ki} a_{ik}$. Thus:

$$\operatorname{Tr}(BA) = \sum_{k=1}^n (BA)_{kk} = \sum_{k=1}^n \sum_{i=1}^m b_{ki} a_{ik}$$

Since elements in the field $K$ commute ($a_{ik} b_{ki} = b_{ki} a_{ik}$), and since finite double sums can be interchanged freely:

$$\sum_{i=1}^m \sum_{k=1}^n a_{ik} b_{ki} = \sum_{k=1}^n \sum_{i=1}^m b_{ki} a_{ik} \implies \operatorname{Tr}(AB) = \operatorname{Tr}(BA) \qquad \blacksquare$$
:::

::: {#thm-trace-similarity-invariance}
## Similarity Invariance of the Trace

If $A, B \in M_{n \times n}(K)$ are similar matrices ($B = P^{-1} A P$ for some invertible $P \in GL(n, K)$), then:

$$\operatorname{Tr}(B) = \operatorname{Tr}(A)$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-trace-similarity-invariance
Using @thm-trace-cyclic, set $X = P^{-1} A$ and $Y = P$:

$$\operatorname{Tr}(P^{-1} A P) = \operatorname{Tr}(X Y) = \operatorname{Tr}(Y X) = \operatorname{Tr}(P (P^{-1} A)) = \operatorname{Tr}((P P^{-1}) A) = \operatorname{Tr}(I_n A) = \operatorname{Tr}(A) \qquad \blacksquare$$
:::

::: {#def-trace-operator}
## Trace of a Linear Operator

Let $V$ be a finite-dimensional vector space over $K$, and let $f: V \to V$ be a linear operator. The **trace** of $f$, denoted $\operatorname{Tr}(f)$, is defined as:

$$\operatorname{Tr}(f) := \operatorname{Tr}([f]_B)$$

where $B$ is any ordered basis of $V$. By @thm-trace-similarity-invariance, this value is completely independent of the choice of basis $B$.
:::

::: {.callout-note}
## Perspective: Determinant and Trace as Intrinsic Invariants
The determinant and trace are the two foremost coordinate-free scalar invariants of a linear endomorphism $f \in \mathcal{L}(V, V)$:

- $\det(f)$ is **multiplicative**: $\det(g \circ f) = \det(g) \det(f)$ and geometric (volume scaling).
- $\operatorname{Tr}(f)$ is **additive / linear**: $\operatorname{Tr}(f + g) = \operatorname{Tr}(f) + \operatorname{Tr}(g)$ and algebraic.

Both invariants appear prominently in the **characteristic polynomial** $p_f(\lambda) = \det(\lambda I - f) = \lambda^n - \operatorname{Tr}(f) \lambda^{n-1} + \dots + (-1)^n \det(f)$, where the trace is the sum of the eigenvalues and the determinant is their product.
:::


```python

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Polygon

# Set aesthetic plot style
plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(16, 5), dpi=100)

# Standard basis and unit square
square = np.array([[0, 0], [1, 0], [1, 1], [0, 1]])
e1 = np.array([1, 0])
e2 = np.array([0, 1])

# Panel 1: Original Unit Square
poly1 = Polygon(square, closed=True, alpha=0.35, facecolor='#3498db', edgecolor='#1d6fa5', lw=2)
ax1.add_patch(poly1)
ax1.quiver(0, 0, e1[0], e1[1], angles='xy', scale_units='xy', scale=1, color='#27ae60', width=0.018, label=r'$e_1$')
ax1.quiver(0, 0, e2[0], e2[1], angles='xy', scale_units='xy', scale=1, color='#e67e22', width=0.018, label=r'$e_2$')
ax1.set_xlim(-2.5, 3.5)
ax1.set_ylim(-2.5, 3.5)
ax1.set_aspect('equal')
ax1.set_title('Unit Square (Area = 1, $\\det = 1$)', fontsize=11, fontweight='bold')
ax1.legend(loc='upper left', framealpha=0.9)

# Panel 2: Orientation-Preserving Map T1 (Shear & Scale, det > 0)
T1 = np.array([[1.5, 1.0], [0.5, 1.5]])
det_T1 = np.linalg.det(T1)
square_T1 = (T1 @ square.T).T
v1_T1 = T1 @ e1
v2_T1 = T1 @ e2

poly2 = Polygon(square_T1, closed=True, alpha=0.35, facecolor='#2ecc71', edgecolor='#1e8449', lw=2)
ax2.add_patch(poly2)
ax2.quiver(0, 0, v1_T1[0], v1_T1[1], angles='xy', scale_units='xy', scale=1, color='#27ae60', width=0.018, label=r'$T_1(e_1)$')
ax2.quiver(0, 0, v2_T1[0], v2_T1[1], angles='xy', scale_units='xy', scale=1, color='#e67e22', width=0.018, label=r'$T_1(e_2)$')
ax2.set_xlim(-2.5, 3.5)
ax2.set_ylim(-2.5, 3.5)
ax2.set_aspect('equal')
ax2.set_title(f'Preserves Orientation: $\\det(T_1) = +{det_T1:.2f}$', fontsize=11, fontweight='bold')
ax2.legend(loc='upper left', framealpha=0.9)

# Panel 3: Orientation-Reversing Map T2 (Reflection & Stretch, det < 0)
T2 = np.array([[1.5, 0.5], [1.0, -1.0]])
det_T2 = np.linalg.det(T2)
square_T2 = (T2 @ square.T).T
v1_T2 = T2 @ e1
v2_T2 = T2 @ e2

poly3 = Polygon(square_T2, closed=True, alpha=0.35, facecolor='#e74c3c', edgecolor='#922b21', lw=2)
ax3.add_patch(poly3)
ax3.quiver(0, 0, v1_T2[0], v1_T2[1], angles='xy', scale_units='xy', scale=1, color='#27ae60', width=0.018, label=r'$T_2(e_1)$')
ax3.quiver(0, 0, v2_T2[0], v2_T2[1], angles='xy', scale_units='xy', scale=1, color='#e67e22', width=0.018, label=r'$T_2(e_2)$')
ax3.set_xlim(-2.5, 3.5)
ax3.set_ylim(-2.5, 3.5)
ax3.set_aspect('equal')
ax3.set_title(f'Reverses Orientation: $\\det(T_2) = {det_T2:.2f}$ (Area = {abs(det_T2):.2f})', fontsize=11, fontweight='bold')
ax3.legend(loc='upper left', framealpha=0.9)

plt.tight_layout()
plt.show()
```

```python

import numpy as np

# Set random seed for reproducible verification
np.random.seed(42)
n = 4

# 1. Change of Basis and Operator Invariance
A = np.random.randint(-3, 5, size=(n, n)).astype(float)
# Generate an invertible transition matrix P
while True:
    P = np.random.randint(-2, 4, size=(n, n)).astype(float)
    if abs(np.linalg.det(P)) > 0.5:
        break
P_inv = np.linalg.inv(P)
A_prime = P_inv @ A @ P

det_A = np.linalg.det(A)
det_A_prime = np.linalg.det(A_prime)
tr_A = np.trace(A)
tr_A_prime = np.trace(A_prime)

print("=== 1. Operator Invariants under Change of Basis ===")
print(f"det([f]_B)  = {det_A:.4f},  det([f]_{{B'}}) = {det_A_prime:.4f}  -> Invariant: {np.isclose(det_A, det_A_prime)}")
print(f"Tr([f]_B)   = {tr_A:.4f},   Tr([f]_{{B'}})  = {tr_A_prime:.4f}   -> Invariant: {np.isclose(tr_A, tr_A_prime)}")

# 2. Transpose Determinant
det_A_T = np.linalg.det(A.T)
print("\n=== 2. Transpose Determinant ===")
print(f"det(A) = {det_A:.4f},  det(A^T) = {det_A_T:.4f}  -> Matches: {np.isclose(det_A, det_A_T)}")

# 3. Laplace Expansion & Adjugate Matrix
def cofactor(M, i, j):
    sub_M = np.delete(np.delete(M, i, axis=0), j, axis=1)
    return ((-1) ** (i + j)) * np.linalg.det(sub_M)

def adjugate(M):
    dim = M.shape[0]
    adj = np.zeros((dim, dim))
    for i in range(dim):
        for j in range(dim):
            adj[j, i] = cofactor(M, i, j)  # adj(M) is transpose of cofactors
    return adj

# Laplace expansion along row 0
laplace_row0 = sum(A[0, j] * cofactor(A, 0, j) for j in range(n))
# Laplace expansion along col 1
laplace_col1 = sum(A[i, 1] * cofactor(A, i, 1) for i in range(n))

adj_A = adjugate(A)
A_adj_A = A @ adj_A
expected_diagonal = det_A * np.eye(n)

print("\n=== 3. Laplace Expansion & Adjugate Fundamental Identity ===")
print(f"Laplace Row 0: {laplace_row0:.4f}, NumPy det: {det_A:.4f}  -> Matches: {np.isclose(laplace_row0, det_A)}")
print(f"Laplace Col 1: {laplace_col1:.4f}, NumPy det: {det_A:.4f}  -> Matches: {np.isclose(laplace_col1, det_A)}")
print(f"A @ adj(A) == det(A)*I: {np.allclose(A_adj_A, expected_diagonal)}")

# Inverse via adjugate
A_inv_adj = (1.0 / det_A) * adj_A
A_inv_np = np.linalg.inv(A)
print(f"Inverse via Adjugate matches np.linalg.inv: {np.allclose(A_inv_adj, A_inv_np)}")

# 4. Cramer's Rule
b = np.array([2.0, -1.0, 3.0, 1.0])
x_cramer = np.zeros(n)
for i in range(n):
    A_i = A.copy()
    A_i[:, i] = b
    x_cramer[i] = np.linalg.det(A_i) / det_A

x_solve = np.linalg.solve(A, b)
print("\n=== 4. Cramer's Rule ===")
print(f"Cramer solution: {x_cramer.round(4)}")
print(f"np.linalg.solve: {x_solve.round(4)}")
print(f"Cramer matches solve: {np.allclose(x_cramer, x_solve)}")

# 5. Cyclic Property of Trace
B = np.random.randint(-2, 3, size=(n, n)).astype(float)
tr_AB = np.trace(A @ B)
tr_BA = np.trace(B @ A)
commutes = np.allclose(A @ B, B @ A)
print("\n=== 5. Trace Commutativity ===")
print(f"Does AB == BA? {commutes}")
print(f"Tr(AB) = {tr_AB:.4f},  Tr(BA) = {tr_BA:.4f}  -> Tr(AB) == Tr(BA): {np.isclose(tr_AB, tr_BA)}")
```
