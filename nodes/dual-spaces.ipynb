# Linear Functionals and Dual Spaces

**Prerequisites:** [Linear Maps](linear-maps.ipynb) · [Basis and Dimension](basis-and-dimension.ipynb)

A **linear functional** is a linear map that outputs a single number: a coordinate, an integral, a value at a point, a weighted sum. The functionals on $V$ form a vector space of their own, the **dual space** $V^*$. Each basis of $V$ has a **dual basis** of $V^*$ whose members read off the coordinates. Duality turns questions about vectors into questions about the linear "measurements" one can make on them.

Throughout, $V$ is a [vector space](vector-spaces.ipynb) over a field $K$.

## 1\. Linear Functionals

::: {#def-linear-functional}
## Linear Functional

A **linear functional** (or **linear form**) on $V$ is a [linear map](linear-maps.ipynb#def-linear-map) $\varphi : V \to K$. Here $K$ is regarded as a $1$-dimensional vector space over itself.
:::

::: {#exm-functionals}
## Examples of Linear Functionals

1. **Coordinate functions.** On $K^n$, $\varphi_j(x_1, \dots, x_n) = x_j$.
2. **Weighted sums.** For fixed $a \in \mathbb{R}^n$, $\varphi(x) = a_1 x_1 + \dots + a_n x_n = \langle a, x \rangle$.
3. **Evaluation.** On a space of functions $X \to K$, $\varphi(g) = g(x_0)$ for a fixed point $x_0 \in X$.
4. **Integration.** On integrable functions on $[a, b]$, $\varphi(g) = \int_a^b g(x)\,dx$.
5. **Trace.** On $M_n(K)$, $\varphi(A) = a_{11} + \dots + a_{nn}$, the [trace](trace.ipynb#def-trace-matrix).

Each is linear because the operation involved (taking a coordinate, summing, evaluating, integrating) respects sums and scalar multiples.
:::

::: {#prp-functional-characterisation}
## Functionals in Coordinates

Let $(v_1, \dots, v_n)$ be a basis of $V$ and $\varphi$ a linear functional. For $v = x_1 v_1 + \dots + x_n v_n$,
$$\varphi(v) = c_1 x_1 + \dots + c_n x_n, \qquad c_i = \varphi(v_i).$$
Conversely, for any scalars $c_1, \dots, c_n$ this formula defines a linear functional. So functionals on an $n$-dimensional space are exactly the homogeneous expressions of degree $1$ in the coordinates.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-functional-characterisation
By linearity, $\varphi(\sum_i x_i v_i) = \sum_i x_i \varphi(v_i)$. The converse is the [linear extension theorem](linear-maps.ipynb#thm-linear-extension) with $W = K$ and $w_i = c_i$. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Level Sets
A non-zero functional on $\mathbb{R}^2$, such as $\varphi(x, y) = 2x + y$, is constant along the parallel lines $2x + y = k$. Knowing $\varphi$ is the same as knowing this family of evenly spaced parallel lines (and which way the values increase). In $\mathbb{R}^3$ the level sets are parallel planes.
:::

## 2\. The Dual Space

::: {#def-dual-space}
## Dual Space

The **dual space** of $V$ is $V^* = \mathcal{L}(V, K)$, the set of all linear functionals on $V$ with pointwise operations $(\varphi + \psi)(v) = \varphi(v) + \psi(v)$ and $(a\varphi)(v) = a\,\varphi(v)$. It is a vector space, being a special case of [$\mathcal{L}(V, W)$](linear-maps.ipynb#prp-hom-is-vector-space).
:::

::: {#def-dual-basis}
## Dual Basis

Let $\mathcal{B} = (v_1, \dots, v_n)$ be a basis of $V$. The **dual basis** $(v_1^*, \dots, v_n^*)$ consists of the functionals determined by
$$v_i^*(v_j) = \delta_{ij} = \begin{cases} 1 & i = j, \\ 0 & i \ne j. \end{cases}$$
By linearity, $v_i^*(x_1 v_1 + \dots + x_n v_n) = x_i$: **the $i$-th dual basis vector reads off the $i$-th coordinate** with respect to $\mathcal{B}$.
:::

::: {#thm-dual-basis-is-basis}
## The Dual Basis Is a Basis

$(v_1^*, \dots, v_n^*)$ is a basis of $V^*$. Every $\varphi \in V^*$ expands as
$$\varphi = \varphi(v_1)\, v_1^* + \dots + \varphi(v_n)\, v_n^*.$$
In particular $\dim V^* = \dim V$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-dual-basis-is-basis
**Independence.** If $\sum_i a_i v_i^* = 0$ (the zero functional), evaluate at $v_j$: $0 = \sum_i a_i v_i^*(v_j) = a_j$.

**Spanning.** Both $\varphi$ and $\psi = \sum_j \varphi(v_j) v_j^*$ are linear and $\psi(v_k) = \sum_j \varphi(v_j)\delta_{jk} = \varphi(v_k)$. Two linear maps that agree on a basis are equal ([linear extension](linear-maps.ipynb#thm-linear-extension)), so $\varphi = \psi$. $\blacksquare$
:::

::: {#exm-dual-basis}
## A Dual Basis in the Plane

Take $v_1 = (1, 0)$, $v_2 = (1, 1)$ in $\mathbb{R}^2$. Writing $(x, y) = (x - y)\,v_1 + y\,v_2$ shows
$$v_1^*(x, y) = x - y, \qquad v_2^*(x, y) = y.$$
Note that $v_1^*$ is **not** "the first coordinate $x$": it depends on the whole basis, including $v_2$. Changing $v_2$ changes $v_1^*$.
:::

The figure draws the level lines of the two dual basis functionals of @exm-dual-basis. The lines $v_1^* = k$ (blue) and $v_2^* = k$ (orange) form a skewed grid; the coordinates of a vector with respect to $(v_1, v_2)$ are found by reading which blue and which orange line pass through it. The point $v = 2 v_1 + 1.5 v_2$ lies on $v_1^* = 2$ and on $v_2^* = 1.5$.


```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, YELLOW, VIOLET = "#2a78d6", "#eb6834", "#eda100", "#6250d6"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({"axes.titlesize": 11, "axes.titleweight": "bold", "font.size": 10,
                     "xtick.color": MUTED, "ytick.color": MUTED, "axes.edgecolor": GRID})

def arrow(ax, v, color, label, offset=(0.08, 0.08)):
    ax.annotate("", xy=v, xytext=(0, 0), arrowprops=dict(arrowstyle="-|>", color=color, lw=2.2, mutation_scale=14))
    ax.text(v[0] + offset[0], v[1] + offset[1], label, color=INK, fontsize=12)

v1, v2 = np.array([1.0, 0.0]), np.array([1.0, 1.0])
v = 2 * v1 + 1.5 * v2
xs = np.linspace(-1, 5, 2)
fig, ax = plt.subplots(figsize=(7.5, 5.2))
for k in np.arange(-4, 6):
    ax.plot(xs, xs - k, color=BLUE, lw=1.6 if k == 2 else 0.8, alpha=0.9 if k == 2 else 0.45)   # v1* = x - y = k
    ax.text(4.45, 4.45 - k, f"{k}", color=BLUE, fontsize=8) if -1 < 4.45 - k < 3.8 else None
for k in np.arange(-1, 4, 0.5):
    ax.axhline(k, color=ORANGE, lw=1.6 if k == 1.5 else 0.8, alpha=0.9 if k == 1.5 else 0.45)  # v2* = y = k
arrow(ax, v1, BLUE, "$v_1$", (0.0, -0.3)); arrow(ax, v2, ORANGE, "$v_2$", (-0.35, 0.05))
arrow(ax, v, YELLOW, r"$v = 2v_1 + 1.5v_2$")
ax.plot([2, v[0]], [0, v[1]], ls="--", color=MUTED, lw=1); ax.plot([0, 2], [0, 0], ls="--", color=MUTED, lw=1)
ax.text(-0.9, 3.45, "blue lines: $v_1^* = x - y = k$\norange lines: $v_2^* = y = k$", fontsize=9, color=INK,
        bbox=dict(fc="white", ec=GRID))
ax.set_xlim(-1, 4.8); ax.set_ylim(-0.8, 3.9); ax.set_aspect("equal")
ax.set_title("The dual basis measures coordinates: $v_1^*(v) = 2$, $v_2^*(v) = 1.5$")
plt.tight_layout(); plt.show()
```


## 3\. Application: Exact Quadrature Rules

::: {#prp-quadrature}
## Integrals from Finitely Many Values

Let $x_1, \dots, x_n \in [a, b]$ be distinct. There are constants $c_1, \dots, c_n$ such that
$$\int_a^b p(x)\,dx = c_1\, p(x_1) + \dots + c_n\, p(x_n)$$
for **every** real polynomial $p$ of degree less than $n$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-quadrature
Work in $P_{n-1}$, which has dimension $n$, so $\dim P_{n-1}^* = n$ by @thm-dual-basis-is-basis. Integration $I(p) = \int_a^b p$ and the evaluations $\varepsilon_i(p) = p(x_i)$ are elements of $P_{n-1}^*$.

The $\varepsilon_i$ are independent: suppose $\sum_i a_i \varepsilon_i = 0$. The polynomial $q_j(x) = \prod_{i \ne j}(x - x_i)$ has degree $n - 1$, vanishes at every $x_i$ with $i \ne j$, and $q_j(x_j) \ne 0$. Applying the relation to $q_j$ gives $a_j q_j(x_j) = 0$, so $a_j = 0$.

Being $n$ independent vectors in the $n$-dimensional space $P_{n-1}^*$, the $\varepsilon_i$ form a [basis](basis-and-dimension.ipynb#cor-independent-full-size). Hence $I = \sum_i c_i \varepsilon_i$ for some scalars $c_i$, which is the claim. $\blacksquare$
:::

::: {#exm-simpson}
## Simpson's Rule

On $[0, 1]$ with nodes $0, \tfrac12, 1$ (so $n = 3$, exact on $P_2$), the weights solve $c_1 + c_2 + c_3 = \int_0^1 1 = 1$, $\tfrac12 c_2 + c_3 = \int_0^1 x = \tfrac12$, $\tfrac14 c_2 + c_3 = \int_0^1 x^2 = \tfrac13$. The solution is $(c_1, c_2, c_3) = (\tfrac16, \tfrac46, \tfrac16)$:
$$\int_0^1 p(x)\,dx = \tfrac16\big(p(0) + 4 p(\tfrac12) + p(1)\big) \qquad (\deg p \le 2).$$
The weights are found by testing the identity on the basis $1, x, x^2$, exactly as in the proof.
:::

::: {.callout-note}
## Why the Integral Depends Linearly on the Values
That $n$ values determine a polynomial of degree $< n$ is [interpolation](kernel-image-and-rank-nullity.ipynb#thm-polynomial-interpolation). The extra information supplied by duality is that the integral is a **fixed linear combination** of those values, with weights that do not depend on the polynomial.
:::


```python

import numpy as np

a, b = 0.0, 1.0
nodes = np.array([0.0, 0.5, 1.0])
# weights c solve sum_i c_i x_i^k = int_a^b x^k dx for k = 0, 1, 2 (test on the basis 1, x, x^2)
M = np.vander(nodes, 3, increasing=True).T
moments = np.array([(b**(k + 1) - a**(k + 1)) / (k + 1) for k in range(3)])
c = np.linalg.solve(M, moments)
print("weights:", np.round(c, 6), "= (1/6, 4/6, 1/6)")

rng = np.random.default_rng(0)
for _ in range(3):
    p = rng.normal(size=3)                                   # p(x) = p0 + p1 x + p2 x^2
    exact = sum(p[k] * moments[k] for k in range(3))
    rule = c @ np.polyval(p[::-1], nodes)
    print(f"p = {np.round(p, 3)}:  integral = {exact:.6f},  rule = {rule:.6f}")
```


## 4\. The Double Dual

::: {#def-double-dual}
## Double Dual

The **double dual** of $V$ is $V^{**} = (V^*)^*$, the space of linear functionals on $V^*$. If $\dim V = n$, then $\dim V^{**} = \dim V^* = n$.
:::

Since $V$ and $V^*$ have the same dimension, they are isomorphic; but every such isomorphism (for instance $v_i \mapsto v_i^*$) depends on a choice of basis. For $V^{**}$ there is a better isomorphism, one that needs no choices at all.

::: {#thm-natural-iso}
## The Natural Isomorphism $V \cong V^{**}$

For $v \in V$ define $\iota(v) \in V^{**}$ by **evaluation at $v$**:
$$\iota(v)(\varphi) = \varphi(v) \qquad (\varphi \in V^*).$$
Then $\iota : V \to V^{**}$ is linear and injective, and if $V$ is finite-dimensional it is an isomorphism.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-natural-iso
**$\iota(v)$ lies in $V^{**}$.** $\iota(v)(a\varphi + \psi) = (a\varphi + \psi)(v) = a\varphi(v) + \psi(v)$.

**$\iota$ is linear.** For all $\varphi$: $\iota(a u + v)(\varphi) = \varphi(a u + v) = a\varphi(u) + \varphi(v)$, so $\iota(a u + v) = a\,\iota(u) + \iota(v)$; this uses that each $\varphi$ is linear.

**$\iota$ is injective.** Let $v \ne 0$. [Extend](basis-and-dimension.ipynb#thm-extend-to-basis) $v$ to a basis $v = v_1, v_2, \dots, v_n$ (in the finite-dimensional case). Then $\iota(v)(v_1^*) = v_1^*(v_1) = 1 \ne 0$, so $\iota(v) \ne 0$ and $\ker \iota = \{0\}$.

**$\iota$ is bijective.** $\iota$ is an injective linear map between spaces of the same finite dimension $n$, hence an [isomorphism](kernel-image-and-rank-nullity.ipynb#cor-dimension-constraints). $\blacksquare$
:::

::: {.callout-warning}
## Natural Versus Chosen Isomorphisms

| | $V \cong V^*$ | $V \cong V^{**}$ |
|---|---|---|
| exists (finite dimension) | yes | yes |
| needs a choice of basis | yes | no |
| one distinguished map | no | yes: $v \mapsto (\varphi \mapsto \varphi(v))$ |

$\iota$ is built from the evaluation pairing $(\varphi, v) \mapsto \varphi(v)$ alone. Because of this, finite-dimensional $V$ and $V^{**}$ are routinely identified, and a vector is regarded as a functional on functionals.
:::
