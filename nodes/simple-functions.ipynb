# Simple Functions and Their Integral

**Prerequisites:** [Measurable Functions](measurable-functions.ipynb) · [Measures and Measure Spaces](measure.ipynb) · [Limits of Measurable Functions](limits-of-measurable-functions.ipynb#def-ae-equality)

We now define the integral $\int f \, d\mu$ of a function $f$ with respect to a measure $\mu$. The first step is the integral of the simplest functions, those taking finitely many values on measurable sets. Throughout, $(X, \mathcal{M}, \mu)$ is a measure space.

## 1\. Partitioning the Range Instead of the Domain

The difference between the Riemann and Lebesgue integrals is simple to state, so simple that the Lebesgue integral could almost have come first.

- **Riemann** partitions the **domain** into small intervals, builds step functions on them, and adds the areas of the resulting rectangles.
- **Lebesgue** partitions the **range** into small intervals $[y_k, y_{k+1})$ and groups together all points $x$ whose values $f(x)$ fall in the same interval: the set $E_k = \{x : y_k \le f(x) < y_{k+1}\}$.

For a continuous monotone function the two procedures look much the same. For the [Dirichlet function](measurable-functions.ipynb#exm-dirichlet-measurable) $\mathbf{1}_{\mathbb{Q}}$ they are very different. Partitioning the domain gets nowhere: the values on every subinterval jump between $0$ and $1$. Partitioning the range into $[0, \frac12)$ and $[\frac12, 1]$ is trivial: the function is constant on each of the two preimages, the irrationals and the rationals. The price is that the pieces of the domain are no longer intervals but **arbitrary measurable sets**, possibly interlaced with each other like the rationals and irrationals. Lebesgue needed to know the size of such sets, which is what measure theory provides.

So the building blocks change. For Riemann they are **step functions**, constant on intervals:
$$\sum_{i} c_i \mathbf{1}_{[x_i, x_{i+1})}, \qquad \text{with value } c_i \text{ on } [x_i, x_{i+1}).$$
For Lebesgue they are **simple functions**, constant on arbitrary measurable sets.

## 2\. Simple Functions

::: {#def-simple-function}
## Simple Function

A function $\psi : X \to \mathbb{R}$ is **simple** if there are finitely many measurable sets $E_1, \dots, E_n \in \mathcal{M}$ and real numbers $c_1, \dots, c_n$ with
$$\psi = \sum_{k=1}^n c_k \mathbf{1}_{E_k}.$$
:::

Simple functions are measurable, as linear combinations of the [measurable indicators](measurable-functions.ipynb#prp-indicator-measurable) $\mathbf{1}_{E_k}$. Step functions are simple; the converse is false, since the $E_k$ need not be intervals.

The sets $E_k$ may overlap; on $E_1 \cap E_2$ the function above takes the value $c_1 + c_2$. Overlaps can always be removed: replace $E_1, E_2$ by $E_1 \setminus E_2$, $E_2 \setminus E_1$ and $E_1 \cap E_2$, with values $c_1$, $c_2$ and $c_1 + c_2$. With **disjoint** sets the representation reads like a piecewise definition: $\psi = c_k$ on $E_k$ and $\psi = 0$ outside all of them.

::: {#prp-canonical-representation}
## Canonical Representation

A function $\psi : X \to \mathbb{R}$ is simple if and only if it is measurable and takes only finitely many values. If $a_1, \dots, a_m$ are its distinct values and $A_j = \{\psi = a_j\}$, then the $A_j$ are measurable, pairwise disjoint, cover $X$, and $\psi = \sum_{j=1}^m a_j \mathbf{1}_{A_j}$ (the **canonical representation**).
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-canonical-representation
A simple function $\sum_k c_k \mathbf{1}_{E_k}$ is measurable, and its value at $x$ is the sum of the $c_k$ with $x \in E_k$; there are at most $2^n$ such sums. Conversely, if $\psi$ is measurable with distinct values $a_1, \dots, a_m$, the [level sets](measurable-functions.ipynb#thm-measurability-four-conditions) $A_j = \{\psi = a_j\}$ are measurable; they are disjoint and cover $X$, and $\psi = \sum_j a_j \mathbf{1}_{A_j}$ by checking at each point. $\blacksquare$
:::

Sums, differences and scalar multiples of simple functions are simple: for instance
$$\sum_{k=1}^n c_k \mathbf{1}_{E_k} + \sum_{j=1}^m d_j \mathbf{1}_{F_j}$$
is already a representation with $n + m$ sets.

## 3\. The Integral of a Nonnegative Simple Function

::: {#def-simple-integral}
## Integral of a Nonnegative Simple Function

If $\psi = \sum_{k=1}^n c_k \mathbf{1}_{E_k}$ with all $c_k \ge 0$, its **integral** is
$$\int_X \psi \, d\mu = \sum_{k=1}^n c_k\, \mu(E_k) \in [0, \infty],$$
with the convention $0 \cdot \infty = 0$.
:::

Why this formula? Compare with the Riemann sum $\sum_i f(x_i)\, \Delta x_i$: on a step $E_k$ the function has height $c_k$, and the step has "length" $\mu(E_k)$. When the steps were intervals, the length of $E_k$ was the length of an interval; now $E_k$ is any measurable set, and its length is measured by $\mu$: its Lebesgue measure on $\mathbb{R}$, its area on $\mathbb{R}^2$, and so on. The formula is a **definition**, but it is the only natural one: the area under a function equal to $c_k$ on $E_k$ should be height times base, summed over $k$. It is easy here only because the hard work of measuring arbitrary sets has been done; this is exactly the step the Riemann integral cannot take.

::: {#exm-dirichlet-integral}
## The Integral of the Dirichlet Function

$\mathbf{1}_{\mathbb{Q}}$ is itself a simple function, with a single set $E_1 = \mathbb{Q}$ and value $c_1 = 1$. With respect to Lebesgue measure $m$,
$$\int_{\mathbb{R}} \mathbf{1}_{\mathbb{Q}} \, dm = 1 \cdot m(\mathbb{Q}) = 0 .$$
The function that has no Riemann integral has Lebesgue integral $0$, as the intuition of [area under the graph](measurable-functions.ipynb#why-a-new-integral) predicted.
:::

::: {#def-simple-integral-over-set}
## Integral Over a Measurable Set

For $E \in \mathcal{M}$ and a nonnegative simple $\psi$,
$$\int_E \psi \, d\mu = \int_X \psi\, \mathbf{1}_E \, d\mu .$$
:::

That is, set $\psi$ to $0$ outside $E$ and integrate over the whole space. The Riemann integral can only integrate over intervals; the Lebesgue integral integrates over any measurable set: the irrationals, the Cantor set, and so on.

## 4\. Well-Definedness

A simple function has many representations: a step can be cut into two steps, and two separate steps with the same value can be merged into one set. The value of the integral must not depend on the representation.

::: {#thm-simple-integral-well-defined}
## The Integral Does Not Depend on the Representation

If $\sum_{k=1}^n c_k \mathbf{1}_{E_k} = \sum_{j=1}^{m} c_j' \mathbf{1}_{E_j'}$ with all $c_k, c_j' \ge 0$, then $\sum_k c_k\, \mu(E_k) = \sum_j c_j'\, \mu(E_j')$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-simple-integral-well-defined
It suffices to show that **every** representation $\psi = \sum_{k=1}^n c_k \mathbf{1}_{E_k}$ with $c_k \ge 0$ gives the value $\sum_j a_j\, \mu(A_j)$ of the canonical representation of @prp-canonical-representation. Cut $X$ into the "atoms" generated by $E_1, \dots, E_n$: for each $\varepsilon = (\varepsilon_1, \dots, \varepsilon_n) \in \{0, 1\}^n$ let
$$P_\varepsilon = \bigcap_{k=1}^n E_k^{\varepsilon_k}, \qquad E^1 = E,\ E^0 = E^c .$$
The $P_\varepsilon$ are measurable, pairwise disjoint, and cover $X$; each $E_k$ is the disjoint union of the atoms with $\varepsilon_k = 1$, so by finite additivity $\mu(E_k) = \sum_{\varepsilon_k = 1} \mu(P_\varepsilon)$. On $P_\varepsilon$, $\psi$ has the constant value $v_\varepsilon = \sum_k \varepsilon_k c_k$. All terms below are in $[0, \infty]$, so they may be regrouped freely:
$$\sum_{k} c_k\, \mu(E_k) = \sum_k \sum_{\varepsilon:\, \varepsilon_k = 1} c_k\, \mu(P_\varepsilon) = \sum_{\varepsilon} \Big(\sum_k \varepsilon_k c_k\Big) \mu(P_\varepsilon) = \sum_\varepsilon v_\varepsilon\, \mu(P_\varepsilon).$$
Now group the atoms by value. The atoms with $v_\varepsilon = a_j$ are disjoint and their union is $A_j = \{\psi = a_j\}$, so their measures add up to $\mu(A_j)$; atoms with value $0$ contribute $0$ (using $0 \cdot \infty = 0$). Hence the sum equals $\sum_j a_j\, \mu(A_j)$, the same for every representation. $\blacksquare$
:::

In words: two representations of the same function are related by cutting sets into pieces and gluing pieces with equal values, and additivity of $\mu$ makes both operations harmless.

## 5\. Properties

::: {#thm-simple-integral-properties}
## Properties of the Integral of Nonnegative Simple Functions

Let $\varphi, \psi$ be nonnegative simple functions and $\alpha, \beta \ge 0$.

1. **Linearity:** $\int_X (\alpha \varphi + \beta \psi)\, d\mu = \alpha \int_X \varphi\, d\mu + \beta \int_X \psi \, d\mu$.
2. **Additivity over domains:** if $A, B \in \mathcal{M}$ are disjoint, $\int_{A \cup B} \varphi\, d\mu = \int_A \varphi\, d\mu + \int_B \varphi \, d\mu$.
3. **Null sets do not matter:** if $X_0 \in \mathcal{M}$ and $\mu(X \setminus X_0) = 0$, then $\int_X \varphi\, d\mu = \int_{X_0} \varphi\, d\mu$.
4. **Monotonicity:** if $\psi \le \varphi$ a.e., then $\int_X \psi\, d\mu \le \int_X \varphi \, d\mu$.
:::

Part 3 says that changing a simple function on a set of measure zero does not change its integral. The coefficients are kept nonnegative so that no $\infty - \infty$ can appear.

::: {.callout-tip collapse="true"}
## Proof of @thm-simple-integral-properties
Write $\varphi = \sum_{k=1}^n c_k \mathbf{1}_{E_k}$ and $\psi = \sum_{j=1}^m d_j \mathbf{1}_{F_j}$ with nonnegative coefficients.

**1. Linearity.** Putting the two representations side by side,
$$\alpha \varphi + \beta \psi = \sum_{k=1}^n (\alpha c_k) \mathbf{1}_{E_k} + \sum_{j=1}^m (\beta d_j) \mathbf{1}_{F_j}$$
is a representation of the simple function $\alpha\varphi + \beta\psi$ with nonnegative coefficients. By @thm-simple-integral-well-defined we may use it to compute the integral, which gives $\sum_k \alpha c_k \mu(E_k) + \sum_j \beta d_j \mu(F_j) = \alpha \int \varphi + \beta \int \psi$.

**2. Domains.** Since $A \cap B = \emptyset$, $\mathbf{1}_{A \cup B} = \mathbf{1}_A + \mathbf{1}_B$. By definition and part 1,
$$\int_{A \cup B} \varphi = \int_X \varphi \mathbf{1}_{A \cup B} = \int_X (\varphi \mathbf{1}_A + \varphi \mathbf{1}_B) = \int_X \varphi \mathbf{1}_A + \int_X \varphi \mathbf{1}_B = \int_A \varphi + \int_B \varphi .$$

**3. Null sets.** Using $\mathbf{1}_A \mathbf{1}_B = \mathbf{1}_{A \cap B}$,
$$\varphi\, \mathbf{1}_{X_0} = \sum_k c_k \mathbf{1}_{E_k \cap X_0}, \qquad \text{so} \qquad \int_{X_0} \varphi = \sum_k c_k\, \mu(E_k \cap X_0).$$
Removing a null set from the space does not change the measure of any set: $E_k = (E_k \cap X_0) \cup (E_k \setminus X_0)$ disjointly, and $E_k \setminus X_0 \subseteq X \setminus X_0$ has measure $0$, so $\mu(E_k \cap X_0) = \mu(E_k)$. Hence $\int_{X_0} \varphi = \sum_k c_k \mu(E_k) = \int_X \varphi$.

**4. Monotonicity.** First suppose $\psi \le \varphi$ everywhere. Then $\varphi - \psi$ is simple (concatenate the representations of $\varphi$ and $-\psi$) and nonnegative. A nonnegative simple function has a representation with nonnegative coefficients, namely its canonical one, so its integral is $\ge 0$. By part 1,
$$\int \varphi = \int \big(\psi + (\varphi - \psi)\big) = \int \psi + \int (\varphi - \psi) \ge \int \psi .$$
In general, $X_0 = \{\psi \le \varphi\}$ is measurable and $\mu(X \setminus X_0) = 0$. Then $\psi \mathbf{1}_{X_0} \le \varphi \mathbf{1}_{X_0}$ everywhere, and by part 3
$$\int_X \psi = \int_{X_0} \psi = \int_X \psi \mathbf{1}_{X_0} \le \int_X \varphi \mathbf{1}_{X_0} = \int_{X_0} \varphi = \int_X \varphi. \qquad \blacksquare$$
:::

The figure contrasts the two ideas for the same function on $[0, 10]$. Left: Riemann's step function, built on a partition of the domain into intervals. Right: Lebesgue's simple function, built on a partition of the range into levels; the set $E_k$ where $f$ lies between two consecutive levels is drawn in the colour of its level, and it is a union of several intervals.

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

x = np.linspace(0, 10, 2000)
f = 2.1 + np.sin(x) + 0.6 * np.cos(2.5 * x)
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(12, 4.4))

edges = np.linspace(0, 10, 11)
for a, b in zip(edges, edges[1:]):
    m = f[(x >= a) & (x <= b)].min()
    ax.add_patch(plt.Rectangle((a, 0), b - a, m, facecolor=BLUE, alpha=0.3, edgecolor=BLUE))
ax.plot(x, f, color=INK, lw=1.8)
ax.set_title("Riemann: partition the domain into intervals")
ax.set_xlabel("$x$"); ax.set_ylim(-0.6, 4.1)

levels = np.arange(0, 4.5, 0.75)
cols = [BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED]
for k, (lo, hi) in enumerate(zip(levels, levels[1:])):
    mask = (f >= lo) & (f < hi)
    if not mask.any():
        continue
    col = cols[k % len(cols)]
    ax2.fill_between(x, 0, lo, where=mask, color=col, alpha=0.35, step="mid", lw=0)
    ax2.fill_between(x, -0.45, -0.15, where=mask, color=col, step="mid", lw=0)
    ax2.axhline(lo, color=col, lw=0.8, ls=":")
ax2.plot(x, f, color=INK, lw=1.8)
ax2.text(0.1, -0.55, r"sets $E_k = \{y_k \leq f < y_{k+1}\}$ on the $x$-axis", fontsize=9, color=MUTED, va="top")
ax2.set_ylim(-1.0, 4.1)
ax2.set_title("Lebesgue: partition the range into levels $y_k$")
ax2.set_xlabel("$x$")
plt.tight_layout(); plt.show()
```
