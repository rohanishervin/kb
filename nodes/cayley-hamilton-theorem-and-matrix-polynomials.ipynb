# The Cayley-Hamilton Theorem and Matrix Polynomials

## 1\. Motivation: Dimension Bounds on Matrix Powers

Let $K$ be a [field](field.ipynb) and let $A \in M_{n \times n}(K)$ be an $n \times n$ matrix with entries in $K$. We define powers of $A$ naturally by setting $A^0 = I_n$ (the $n \times n$ identity matrix) and $A^k = A \cdot A^{k-1}$ for $k \ge 1$.

A fundamental inquiry in linear algebra is whether every square matrix satisfies a non-trivial polynomial equation:

$$p(A) = c_m A^m + c_{m-1} A^{m-1} + \dots + c_1 A + c_0 I_n = 0$$

### 1.1\. The Coarse Dimension Bound ($m = n^2$)

The set of all $n \times n$ matrices $M_{n \times n}(K)$ forms a [vector space](vector-spaces.ipynb) over $K$ under entrywise addition and scalar multiplication. Because an $n \times n$ matrix contains $n^2$ independent entries, its [dimension](linear-independence-basis-dimension.ipynb#def-dimension) is:

$$\dim_K M_{n \times n}(K) = n^2$$

Consider the sequence of $n^2 + 1$ matrix powers:

$$I_n, A, A^2, A^3, \dots, A^{n^2}$$

Because the number of vectors ($n^2 + 1$) strictly exceeds the dimension of the space ($n^2$), these matrices must be [linearly dependent](linear-independence-basis-dimension.ipynb#def-linear-independence). Therefore, there exist scalars $c_0, c_1, \dots, c_{n^2} \in K$, not all zero, such that:

$$c_{n^2} A^{n^2} + c_{n^2 - 1} A^{n^2 - 1} + \dots + c_1 A + c_0 I_n = 0$$

This establishes the existence of an annihilating polynomial for $A$ of degree at most $n^2$.

---

### 1.2\. Vectorwise Annihilation ($m = n$)

Can the degree be significantly reduced from $n^2$ to $n$? Consider the action of $A$ on an $n$-dimensional space $V = K^n$. For any fixed vector $v \in K^n$, examine the $n + 1$ vector iterates:

$$v, A v, A^2 v, \dots, A^n v$$

Because $\dim K^n = n$, any $n + 1$ vectors in $K^n$ are linearly dependent. Hence there exist scalars $d_0, d_1, \dots, d_n \in K$, not all zero, such that:

$$(d_n A^n + d_{n-1} A^{n-1} + \dots + d_1 A + d_0 I_n) v = 0$$

However, this linear dependence relation *a priori* depends on the chosen vector $v$. If $v$ is replaced by another vector $w$, the coefficients $d_i$ may change.

The **Cayley-Hamilton Theorem** asserts that a single universal polynomial of degree $n$ annihilates *all* vectors simultaneously—namely, the characteristic polynomial of $A$.

::: {#thm-cayley-hamilton}
## The Cayley-Hamilton Theorem

Let $A \in M_{n \times n}(K)$ be an $n \times n$ matrix over an arbitrary field $K$, and let:

$$p_A(x) = \det(x I_n - A) = x^n + c_{n-1} x^{n-1} + \dots + c_1 x + c_0$$

be its [characteristic polynomial](eigenvalues-eigenvectors-and-diagonalization.ipynb#thm-characteristic-equation). Then $A$ satisfies its own characteristic equation:

$$p_A(A) = A^n + c_{n-1} A^{n-1} + \dots + c_1 A + c_0 I_n = 0_{n \times n}$$
:::



## 2\. Proof for Diagonalizable Matrices and 2D Verification

Before establishing the general algebraic proof, we examine the case where $A$ is [diagonalizable](eigenvalues-eigenvectors-and-diagonalization.ipynb#def-diagonalizable-operator).

::: {#prp-cayley-hamilton-diagonalizable}
## Cayley-Hamilton for Diagonalizable Matrices

If $A \in M_{n \times n}(K)$ possesses $n$ linearly independent [eigenvectors](eigenvalues-eigenvectors-and-diagonalization.ipynb#def-eigenvalue-eigenvector) (in particular, if $p_A(x)$ has $n$ distinct roots in $K$), then $p_A(A) = 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-cayley-hamilton-diagonalizable
Let $v_1, \dots, v_n \in K^n$ be a basis of eigenvectors of $A$, with corresponding eigenvalues $\lambda_1, \dots, \lambda_n \in K$:

$$A v_i = \lambda_i v_i \qquad (i = 1, \dots, n)$$

By the [spectral mapping theorem](eigenvalues-eigenvectors-and-diagonalization.ipynb#thm-spectral-mapping-polynomial), for any polynomial $p(x) \in K[x]$:

$$p(A) v_i = p(\lambda_i) v_i$$

Because each $\lambda_i$ is a root of the characteristic polynomial $p_A(x) = \det(x I_n - A)$, we have $p_A(\lambda_i) = 0$. Therefore:

$$p_A(A) v_i = 0 \cdot v_i = 0 \qquad \text{for every } i = 1, \dots, n$$

Now let $v \in K^n$ be an arbitrary vector. Because $\{v_1, \dots, v_n\}$ forms a basis for $K^n$, there exist scalars $a_1, \dots, a_n \in K$ such that $v = \sum_{i=1}^n a_i v_i$. By linearity of matrix multiplication:

$$p_A(A) v = p_A(A) \left( \sum_{i=1}^n a_i v_i \right) = \sum_{i=1}^n a_i p_A(A) v_i = \sum_{i=1}^n a_i \cdot 0 = 0$$

Since the linear operator $p_A(A)$ maps every vector in $K^n$ to zero, its representing matrix must be the zero matrix: $p_A(A) = 0_{n \times n}$. $\blacksquare$
:::

::: {.callout-note}
## Analytic Density Argument (Over $\mathbb{C}$)
In the complex matrix space $M_{n \times n}(\mathbb{C})$, the set of matrices with $n$ distinct eigenvalues is an open, dense subset (its complement is the algebraic hypersurface defined by the vanishing of the polynomial discriminant $\Delta(p_A) = 0$).

Because the map $A \mapsto p_A(A)$ is a continuous (in fact, polynomial) function of the matrix entries $(a_{ij})$, and because $p_A(A) = 0$ on a dense subset, it follows by topological continuity that $p_A(A) = 0$ for *all* $A \in M_{n \times n}(\mathbb{C})$.

However, this analytic argument relies on topology and the field $\mathbb{C}$. Below, we construct an exact, purely algebraic proof valid over arbitrary fields and commutative rings.
:::

---

### 2.1\. Direct Verification: The 2D Rotation Matrix

Consider the standard rotation matrix $R_\alpha \in M_{2 \times 2}(\mathbb{R})$ by angle $\alpha$:

$$R_\alpha = \begin{pmatrix} \cos\alpha & -\sin\alpha \\ \sin\alpha & \cos\alpha \end{pmatrix}$$

Its characteristic polynomial is:

$$p(\lambda) = \det(\lambda I_2 - R_\alpha) = \det \begin{pmatrix} \lambda - \cos\alpha & \sin\alpha \\ -\sin\alpha & \lambda - \cos\alpha \end{pmatrix} = (\lambda - \cos\alpha)^2 + \sin^2\alpha = \lambda^2 - 2\cos\alpha \lambda + 1$$

Now we evaluate $p(R_\alpha) = R_\alpha^2 - 2\cos\alpha R_\alpha + I_2$ directly:

1. Squaring $R_\alpha$ rotates by $2\alpha$:
   $$R_\alpha^2 = R_{2\alpha} = \begin{pmatrix} \cos 2\alpha & -\sin 2\alpha \\ \sin 2\alpha & \cos 2\alpha \end{pmatrix}$$
2. The middle term is:
   $$-2\cos\alpha R_\alpha = \begin{pmatrix} -2\cos^2\alpha & 2\cos\alpha\sin\alpha \\ -2\cos\alpha\sin\alpha & -2\cos^2\alpha \end{pmatrix}$$
3. Summing all three terms:
   $$\begin{aligned}
   R_\alpha^2 - 2\cos\alpha R_\alpha + I_2 &= \begin{pmatrix} \cos 2\alpha - 2\cos^2\alpha + 1 & -\sin 2\alpha + 2\sin\alpha\cos\alpha \\ \sin 2\alpha - 2\sin\alpha\cos\alpha & \cos 2\alpha - 2\cos^2\alpha + 1 \end{pmatrix} \\
   &= \begin{pmatrix} (2\cos^2\alpha - 1) - 2\cos^2\alpha + 1 & -(2\sin\alpha\cos\alpha) + 2\sin\alpha\cos\alpha \\ 2\sin\alpha\cos\alpha - 2\sin\alpha\cos\alpha & (2\cos^2\alpha - 1) - 2\cos^2\alpha + 1 \end{pmatrix} = \begin{pmatrix} 0 & 0 \\ 0 & 0 \end{pmatrix}
   \end{aligned}$$

The matrix satisfies its characteristic polynomial identically, even when $\alpha$ is a multiple of $\pi$ (where the eigenvalues are not distinct).
:::



## 3\. Algebraic Foundations: Rings and Polynomial Rings

To formulate a general, rigorous proof of the Cayley-Hamilton Theorem, we develop the algebraic theory of rings and formal polynomials.

::: {#def-ring}
## Ring (Unital Ring)

A **ring** is a non-empty set $R$ equipped with two binary operations, **addition** ($+$) and **multiplication** ($\cdot$), satisfying:

1. **Additive Abelian Group:**
   - Associativity: $(a + b) + c = a + (b + c)$ for all $a, b, c \in R$.
   - Identity: There exists an element $0 \in R$ such that $a + 0 = 0 + a = a$ for all $a \in R$.
   - Inverses: For each $a \in R$, there exists $-a \in R$ such that $a + (-a) = (-a) + a = 0$.
   - Commutativity: $a + b = b + a$ for all $a, b \in R$.

2. **Multiplicative Monoid:**
   - Associativity: $(a \cdot b) \cdot c = a \cdot (b \cdot c)$ for all $a, b, c \in R$.
   - Identity: There exists an element $1 \in R$ such that $1 \cdot a = a \cdot 1 = a$ for all $a \in R$.

3. **Distributivity:**
   - Left distributivity: $a \cdot (b + c) = a \cdot b + a \cdot c$ for all $a, b, c \in R$.
   - Right distributivity: $(a + b) \cdot c = a \cdot c + b \cdot c$ for all $a, b, c \in R$.

If multiplication is also commutative ($a \cdot b = b \cdot a$), $R$ is called a **commutative ring**. An element $u \in R$ is called a **unit** (or **invertible**) if there exists $v \in R$ such that $u v = v u = 1$.
:::

### 3.1\. Prototypical Examples of Rings

1. **Fields:** Every field $K$ (such as $\mathbb{Q}, \mathbb{R}, \mathbb{C}$) is a commutative ring where every non-zero element is a unit.
2. **The Integers $\mathbb{Z}$:** A commutative ring with only two units: $\{+1, -1\}$.
3. **The Boolean Ring:** For any set $X$, the power set $P(X)$ is a commutative ring under symmetric difference $A \Delta B = (A \setminus B) \cup (B \setminus A)$ as addition, and intersection $A \cap B$ as multiplication. The additive identity is $\emptyset$, the multiplicative identity is $X$, and every element satisfies $A \Delta A = \emptyset$ and $A \cap A = A$.
4. **The Matrix Ring $M_{n \times n}(R)$:** For any ring $R$ and $n \ge 2$, the set of $n \times n$ matrices with entries in $R$ forms a **non-commutative ring** under standard matrix addition and multiplication. Its units are the invertible matrices $GL_n(R)$.

---

::: {#def-polynomial-ring}
## The Polynomial Ring $R[x]$

Let $R$ be a ring. The **polynomial ring** $R[x]$ in an indeterminate $x$ consists of formal sequences $(a_0, a_1, a_2, \dots)$ of elements of $R$ with **finite support** (that is, there exists $N \in \mathbb{N}$ such that $a_k = 0$ for all $k > N$).

Writing such a sequence as $p(x) = \sum_{k=0}^m a_k x^k$, the ring operations are defined by:

1. **Addition:**
   $$\left( \sum_{k=0}^m a_k x^k \right) + \left( \sum_{k=0}^m b_k x^k \right) := \sum_{k=0}^m (a_k + b_k) x^k$$
2. **Multiplication (Cauchy Product):**
   $$\left( \sum_{i=0}^m a_i x^i \right) \cdot \left( \sum_{j=0}^n b_j x^j \right) := \sum_{k=0}^{m+n} c_k x^k, \qquad c_k = \sum_{i+j = k} a_i b_j = \sum_{i=0}^k a_i b_{k-i}$$
:::

::: {#prp-degree-properties}
## Degree and Leading Coefficients

For $p(x) = \sum_{k=0}^m a_k x^k \in R[x]$ with $a_m \neq 0$, the integer $m$ is the **degree** $\deg(p)$, and $a_m$ is the **leading coefficient**.

1. $\deg(p + q) \le \max(\deg p, \deg q)$.
2. $\deg(p \cdot q) \le \deg p + \deg q$, with equality holding whenever the leading coefficient of $p$ or $q$ is a unit in $R$.
:::



## 4\. Non-commutative Evaluation and the Commutation Lemma

In elementary algebra over a field, substituting a value $c$ into a product of polynomials preserves multiplication: $(p \cdot q)(c) = p(c) q(c)$. In a non-commutative ring, this multiplicativity generally breaks down.

::: {#def-evaluation-map}
## Right Evaluation Map

Let $R$ be a ring and let $c \in R$. The **right evaluation map** at $c$, denoted $E_c: R[x] \to R$, is defined by:

$$E_c\left( \sum_{k=0}^m a_k x^k \right) := \sum_{k=0}^m a_k c^k = a_0 + a_1 c + a_2 c^2 + \dots + a_m c^m$$
:::

While additivity always holds ($E_c(p + q) = E_c(p) + E_c(q)$), consider the following simple counterexample to multiplicativity:

Let $p(x) = a_0 + a_1 x$ and $q(x) = b_0$ (a constant polynomial). Their product in $R[x]$ is:

$$p(x) q(x) = a_0 b_0 + (a_1 b_0) x$$

Evaluating both sides at $c \in R$:

$$E_c(p(x) q(x)) = a_0 b_0 + a_1 b_0 c$$

$$E_c(p(x)) \cdot E_c(q(x)) = (a_0 + a_1 c) b_0 = a_0 b_0 + a_1 c b_0$$

These two values coincide if and only if $a_1 b_0 c = a_1 c b_0$, which fails whenever $b_0$ and $c$ do not commute! Fortunately, when commutation holds, multiplicativity is restored.

::: {#lem-commutation-evaluation}
## The Commutation Evaluation Lemma

Let $R$ be a ring, $c \in R$, and let $p(x), q(x) \in R[x]$.

If $c$ commutes with every coefficient of $q(x)$—that is, if $q(x) = \sum_{j=0}^n b_j x^j$ satisfies:

$$c \cdot b_j = b_j \cdot c \qquad \text{for all } j = 0, 1, \dots, n$$

then the right evaluation map is multiplicative on the product $p(x) q(x)$:

$$E_c(p(x) q(x)) = E_c(p(x)) \cdot E_c(q(x))$$
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-commutation-evaluation
Let $p(x) = \sum_{i=0}^m a_i x^i$ and $q(x) = \sum_{j=0}^n b_j x^j$. By the definition of polynomial multiplication:

$$p(x) q(x) = \sum_{k=0}^{m+n} r_k x^k \qquad \text{where } r_k = \sum_{i+j = k} a_i b_j$$

Evaluating the left-hand side at $c$:

$$E_c(p(x) q(x)) = \sum_{k=0}^{m+n} r_k c^k = \sum_{k=0}^{m+n} \left( \sum_{i+j = k} a_i b_j \right) c^k = \sum_{i=0}^m \sum_{j=0}^n a_i b_j c^{i+j} \qquad (1)$$

Now examine the product of the individual evaluations on the right-hand side:

$$E_c(p(x)) \cdot E_c(q(x)) = \left( \sum_{i=0}^m a_i c^i \right) \cdot \left( \sum_{j=0}^n b_j c^j \right) = \sum_{i=0}^m \sum_{j=0}^n a_i c^i b_j c^j \qquad (2)$$

Compare the inner terms of $(1)$ and $(2)$ for any pair $(i, j)$:

- In $(1)$, the term is $a_i b_j c^{i+j} = a_i b_j c^i c^j$.
- In $(2)$, the term is $a_i c^i b_j c^j$.

By hypothesis, $c$ commutes with $b_j$: $c b_j = b_j c$. By simple induction on $i$, $c^i$ also commutes with $b_j$:

$$c^i b_j = b_j c^i$$

Substituting this commutation relation into $(2)$:

$$a_i (c^i b_j) c^j = a_i (b_j c^i) c^j = a_i b_j c^{i+j}$$

Every corresponding term in the double sums is identical. Therefore:

$$E_c(p(x) q(x)) = E_c(p(x)) \cdot E_c(q(x)) \qquad \blacksquare$$
:::



## 5\. The General Algebraic Proof of the Cayley-Hamilton Theorem

We now synthesize the matrix adjugate identity with the Commutation Evaluation Lemma to deliver a complete, rigorous proof of the Cayley-Hamilton Theorem.

### 5.1\. The Characteristic Matrix and the Adjugate Identity

Let $A \in M_{n \times n}(K)$. Consider the characteristic matrix in the polynomial variable $x$:

$$x I_n - A \in M_{n \times n}(K[x])$$

Because $K[x]$ is a commutative ring, the [fundamental identity of the adjugate](change-of-basis-matrix-inversion-and-trace.ipynb#thm-adjugate-fundamental-identity) holds unconditionally for $x I_n - A$:

$$\operatorname{adj}(x I_n - A) \cdot (x I_n - A) = \det(x I_n - A) I_n = p_A(x) I_n \qquad (\dagger)$$

We identify $M_{n \times n}(K[x])$ with the ring of polynomials whose coefficients are matrices, $(M_{n \times n}(K))[x]$:

1. **The right-hand side:** $p_A(x) I_n = (x^n + c_{n-1} x^{n-1} + \dots + c_1 x + c_0) I_n = I_n x^n + c_{n-1} I_n x^{n-1} + \dots + c_0 I_n$.
2. **The linear factor:** $Q(x) := x I_n - A = I_n x - A$, which is a polynomial of degree $1$ in $(M_{n \times n}(K))[x]$.
3. **The adjugate factor:** $P(x) := \operatorname{adj}(x I_n - A)$. Each entry of $P(x)$ is an $(n-1) \times (n-1)$ minor of $x I_n - A$, which has degree at most $n - 1$ in $x$. Grouping matrix powers of $x$:
   $$P(x) = B_{n-1} x^{n-1} + B_{n-2} x^{n-2} + \dots + B_1 x + B_0$$
   for uniquely determined constant matrices $B_k \in M_{n \times n}(K)$.

---

::: {.callout-tip collapse="true"}
## Rigorous Proof of the Cayley-Hamilton Theorem (@thm-cayley-hamilton)
Consider the polynomial identity in the ring $R[x] = (M_{n \times n}(K))[x]$:

$$p_A(x) I_n = P(x) \cdot Q(x) \qquad \text{where } Q(x) = I_n x - A$$

We apply the right evaluation map $E_A: R[x] \to M_{n \times n}(K)$ at the matrix $c = A$.

**Step 1: Check the Commutation Hypothesis:**

The polynomial $Q(x) = I_n x - A$ has degree $1$ with matrix coefficients:

- Degree $1$ coefficient: $b_1 = I_n$
- Degree $0$ coefficient: $b_0 = -A$

We verify whether $A$ commutes with both coefficients:

$$A \cdot b_1 = A \cdot I_n = I_n \cdot A = b_1 \cdot A = A$$

$$A \cdot b_0 = A \cdot (-A) = -A^2 = (-A) \cdot A = b_0 \cdot A$$

Indeed, $A$ commutes with every coefficient of $Q(x)$!

**Step 2: Apply the Commutation Evaluation Lemma:**

By @lem-commutation-evaluation, the evaluation map is multiplicative on the product $P(x) Q(x)$:

$$E_A(p_A(x) I_n) = E_A(P(x) \cdot Q(x)) = E_A(P(x)) \cdot E_A(Q(x))$$

**Step 3: Evaluate $E_A(Q(x))$:**

Evaluating $Q(x) = I_n x - A$ at $x = A$:

$$E_A(Q(x)) = I_n \cdot A - A = A - A = 0_{n \times n}$$

**Step 4: Conclude Annihilation:**

Substituting $E_A(Q(x)) = 0$:

$$E_A(p_A(x) I_n) = E_A(P(x)) \cdot 0_{n \times n} = 0_{n \times n}$$

On the other hand, directly evaluating the left-hand side:

$$E_A(p_A(x) I_n) = E_A(I_n x^n + c_{n-1} I_n x^{n-1} + \dots + c_0 I_n) = A^n + c_{n-1} A^{n-1} + \dots + c_1 A + c_0 I_n = p_A(A)$$

Therefore:

$$p_A(A) = 0_{n \times n} \qquad \blacksquare$$
:::

---

### 5.2\. Consequences of the Cayley-Hamilton Theorem

::: {#cor-inverse-via-cayley-hamilton}
## Inverting Matrices via Cayley-Hamilton

Let $A \in M_{n \times n}(K)$ be an invertible matrix (so $\det(A) = (-1)^n c_0 \neq 0$). Rearranging $p_A(A) = 0$:

$$A^n + c_{n-1} A^{n-1} + \dots + c_1 A = -c_0 I_n \implies A \left( A^{n-1} + c_{n-1} A^{n-2} + \dots + c_1 I_n \right) = -c_0 I_n$$

Multiplying by $-\frac{1}{c_0}$ yields an explicit polynomial formula for $A^{-1}$:

$$A^{-1} = -\frac{1}{c_0} \left( A^{n-1} + c_{n-1} A^{n-2} + \dots + c_1 I_n \right)$$
:::

::: {#cor-power-reduction}
## Reduction of High Matrix Powers

For any integer $k \ge n$, polynomial division of $x^k$ by $p_A(x)$ yields $x^k = q(x) p_A(x) + r(x)$ with $\deg(r) \le n - 1$. Evaluating at $x = A$:

$$A^k = q(A) p_A(A) + r(A) = q(A) \cdot 0 + r(A) = r(A)$$

Thus *any* arbitrary power $A^k$ can be reduced to a polynomial in $A$ of degree at most $n - 1$.
:::



## 6\. The Minimal Polynomial

The Cayley-Hamilton Theorem shows that the ideal of polynomials annihilating $A$ in $K[x]$ is non-trivial.

::: {#def-minimal-polynomial}
## The Minimal Polynomial

Let $A \in M_{n \times n}(K)$. The **minimal polynomial** of $A$, denoted $m_A(x)$, is the unique monic polynomial in $K[x]$ of **minimal degree** such that:

$$m_A(A) = 0_{n \times n}$$
:::

::: {#thm-minimal-polynomial-properties}
## Structural Properties of the Minimal Polynomial

Let $A \in M_{n \times n}(K)$ with characteristic polynomial $p_A(x)$ and minimal polynomial $m_A(x)$.

1. **Divisibility:** For any polynomial $f(x) \in K[x]$, $f(A) = 0 \iff m_A(x) \mid f(x)$. In particular, by the Cayley-Hamilton Theorem:
   $$m_A(x) \mid p_A(x)$$
2. **Identical Roots:** The roots of $m_A(x)$ in an algebraically closed field are precisely the eigenvalues of $A$. That is, $m_A(x)$ and $p_A(x)$ share the exact same irreducible factors.
3. **Degree Bounds:** $1 \le \deg(m_A) \le n \le n^2$.
:::

::: {.callout-note}
## Looking Ahead: Minimal Polynomial and Diagonalizability
The minimal polynomial provides the sharpest algebraic criterion for diagonalizability: an operator $A$ is **diagonalizable** over $K$ if and only if its minimal polynomial $m_A(x)$ splits into **distinct linear factors** over $K$ (no repeated roots):

$$m_A(x) = (x - \lambda_1)(x - \lambda_2) \cdots (x - \lambda_k), \qquad \lambda_i \neq \lambda_j$$

This criterion will play a pivotal role in deriving the Jordan Normal Form and identifying generalized eigenspaces.
:::


```python

import numpy as np
import matplotlib.pyplot as plt

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(16, 5), dpi=100)

# ── Panel 1: Vector Iterates Closing via Cayley-Hamilton in 2D ───────────────
# Matrix A with tr(A) = 2.5, det(A) = 1.0 -> p(x) = x^2 - 2.5 x + 1.0
A = np.array([[1.5, 0.5], [0.5, 1.0]])
tr_A = np.trace(A)
det_A = np.linalg.det(A)
v0 = np.array([1.0, 0.5])
v1 = A @ v0
v2 = A @ v1
v2_predicted = tr_A * v1 - det_A * v0  # Since A^2 v - tr(A) A v + det(A) v = 0

ax1.quiver(0, 0, v0[0], v0[1], angles='xy', scale_units='xy', scale=1, color='#2980b9', width=0.015, label=r'$v$')
ax1.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='#27ae60', width=0.015, label=r'$A v$')
ax1.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='#e74c3c', width=0.015, label=r'$A^2 v$')
ax1.plot([v2[0]], [v2[1]], 'o', color='#c0392b', markersize=8)
ax1.plot([v2_predicted[0]], [v2_predicted[1]], 'x', color='#2c3e50', markersize=10, mew=2, label='$\\mathrm{Tr}(A)Av - \\det(A)v$')

ax1.set_xlim(-0.5, 4.5)
ax1.set_ylim(-0.5, 4.5)
ax1.set_aspect('equal')
ax1.set_title('Vector Closure: $A^2 v = \\mathrm{Tr}(A) Av - \\det(A) v$', fontsize=11, fontweight='bold')
ax1.legend(loc='upper left', fontsize=8, framealpha=0.9)

# ── Panel 2: Matrix Residual Norm vs Polynomial Degree ─────────────────────
# For an n=3 matrix, examine residual norm ||sum_{k=0}^m c_k A^k||
np.random.seed(42)
A3 = np.array([[1.0, 2.0, -1.0], [0.0, -1.0, 3.0], [1.0, 1.0, 2.0]])
coeffs3 = np.poly(A3)  # [1, -tr, ..., (-1)^n det]

# Build partial polynomial evaluations
degrees = [0, 1, 2, 3]
res_norms = []
for deg in degrees:
    poly_deg = coeffs3[-(deg+1):]
    M_deg = np.zeros_like(A3)
    for p_idx, c_val in enumerate(reversed(poly_deg)):
        M_deg = M_deg + c_val * np.linalg.matrix_power(A3, p_idx)
    res_norms.append(np.linalg.norm(M_deg))

ax2.semilogy(degrees, [max(val, 1e-15) for val in res_norms], 'o-', color='#8e44ad', lw=2, markersize=8)
ax2.axhline(1e-14, color='#e74c3c', linestyle='--', label='Numerical Machine Epsilon')
ax2.set_xticks(degrees)
ax2.set_xlabel('Polynomial Degree $m$', fontsize=10)
ax2.set_ylabel('Residual Norm $||p_m(A)||$', fontsize=10)
ax2.set_title('Annihilation at $m = n = 3$', fontsize=11, fontweight='bold')
ax2.legend(loc='upper right', fontsize=8, framealpha=0.9)

# ── Panel 3: 2D Rotation Trajectory Annihilation ───────────────────────────
theta = np.pi / 4.0
R = np.array([[np.cos(theta), -np.sin(theta)], [np.sin(theta), np.cos(theta)]])
v_rot0 = np.array([1.5, 0.0])
v_rot1 = R @ v_rot0
v_rot2 = R @ v_rot1
v_rot_ch = 2 * np.cos(theta) * v_rot1 - v_rot0

circle_t = np.linspace(0, 2*np.pi, 100)
ax3.plot(1.5*np.cos(circle_t), 1.5*np.sin(circle_t), ':', color='#bdc3c7', label='Orbit radius 1.5')
ax3.quiver(0, 0, v_rot0[0], v_rot0[1], angles='xy', scale_units='xy', scale=1, color='#2980b9', width=0.015, label=r'$v$')
ax3.quiver(0, 0, v_rot1[0], v_rot1[1], angles='xy', scale_units='xy', scale=1, color='#27ae60', width=0.015, label='Rotated $R v$ ($45^\\circ$)')
ax3.quiver(0, 0, v_rot2[0], v_rot2[1], angles='xy', scale_units='xy', scale=1, color='#e67e22', width=0.015, label='Rotated $R^2 v$ ($90^\\circ$)')
ax3.plot([v_rot_ch[0]], [v_rot_ch[1]], 'x', color='#c0392b', markersize=10, mew=2, label='$2\\cos\\alpha R v - v$')

ax3.set_xlim(-2.0, 2.0)
ax3.set_ylim(-2.0, 2.0)
ax3.set_aspect('equal')
ax3.set_title('Rotation Matrix: $R^2 v - 2\\cos\\alpha R v + v = 0$', fontsize=11, fontweight='bold')
ax3.legend(loc='lower left', fontsize=8, framealpha=0.9)

plt.tight_layout()
plt.show()
```

```python

import numpy as np

np.random.seed(42)
n = 4

# 1. Cayley-Hamilton for a 4x4 Matrix
A = np.random.randint(-3, 4, size=(n, n)).astype(float)
char_coeffs = np.poly(A)  # Coefficients c_4, c_3, ..., c_0

# Evaluate p_A(A) = A^4 + c_3 A^3 + c_2 A^2 + c_1 A + c_0 I
p_A = np.zeros((n, n), dtype=float)
for power, coeff in enumerate(reversed(char_coeffs)):
    p_A += coeff * np.linalg.matrix_power(A, power)

norm_p_A = np.linalg.norm(p_A)
print("=== 1. Cayley-Hamilton Annihilation Verification ===")
print(f"Characteristic polynomial coefficients: {char_coeffs.round(4)}")
print(f"Frobenius norm ||p(A)||: {norm_p_A:.4e}")
print(f"Does p(A) == 0? {np.allclose(p_A, 0.0, atol=1e-11)}")

# 2. Adjugate Polynomial Identity: B(x) * (x I - A) == p_A(x) I
# Coefficients of B(x) = B_3 x^3 + B_2 x^2 + B_1 x + B_0
# Matching powers: B_3 = I, B_2 - B_3 A = c_3 I, B_1 - B_2 A = c_2 I, B_0 - B_1 A = c_1 I, -B_0 A = c_0 I
c4, c3, c2, c1, c0 = char_coeffs
B3 = np.eye(n)
B2 = B3 @ A + c3 * np.eye(n)
B1 = B2 @ A + c2 * np.eye(n)
B0 = B1 @ A + c1 * np.eye(n)
remainder_identity = -B0 @ A - c0 * np.eye(n)

print("\n=== 2. Adjugate Recursive Coefficients Identity ===")
print(f"Does -B_0 A == c_0 I? {np.allclose(remainder_identity, 0.0, atol=1e-11)}")

# 3. Matrix Inversion via Cayley-Hamilton
inv_CH = - (1.0 / c0) * (np.linalg.matrix_power(A, 3) + c3 * np.linalg.matrix_power(A, 2) + c2 * A + c1 * np.eye(n))
inv_np = np.linalg.inv(A)
print("\n=== 3. Matrix Inverse via Cayley-Hamilton ===")
print(f"Inverse via CH matches np.linalg.inv: {np.allclose(inv_CH, inv_np)}")

# 4. High Power Reduction: A^7 modulo p_A(x)
# Divide x^7 by p_A(x) to get remainder r(x) of degree <= 3
poly_x7 = np.zeros(8)
poly_x7[0] = 1.0  # x^7
quotient, remainder = np.polydiv(poly_x7, char_coeffs)

A7_direct = np.linalg.matrix_power(A, 7)
A7_reduced = np.zeros((n, n), dtype=float)
for power, coeff in enumerate(reversed(remainder)):
    A7_reduced += coeff * np.linalg.matrix_power(A, power)

print("\n=== 4. High Power Reduction via Remainder ===")
print(f"A^7 matches r(A) with deg(r) <= 3: {np.allclose(A7_direct, A7_reduced)}")

# 5. Non-commutative Evaluation Lemma Check
# P(x) = M1 x + M0, Q(x) = N1 x + N0
M1 = np.random.randn(n, n)
M0 = np.random.randn(n, n)
N1 = np.eye(n)       # Commutes with A
N0 = -A              # Commutes with A
# Cauchy product R(x) = (M1 N1) x^2 + (M1 N0 + M0 N1) x + (M0 N0)
R2 = M1 @ N1
R1 = M1 @ N0 + M0 @ N1
R0 = M0 @ N0

E_A_prod = R2 @ (A @ A) + R1 @ A + R0
E_A_P = M1 @ A + M0
E_A_Q = N1 @ A + N0  # = A - A = 0
E_A_factored = E_A_P @ E_A_Q

print("\n=== 5. Non-commutative Evaluation Lemma Test ===")
print(f"Does E_A(P(x) Q(x)) == E_A(P) * E_A(Q) when A commutes with Q coefficients? {np.allclose(E_A_prod, E_A_factored)}")
print(f"Is E_A(Q(x)) identically zero? {np.allclose(E_A_Q, 0.0)}")
```
