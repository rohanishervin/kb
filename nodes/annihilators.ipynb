# Annihilators

**Prerequisites:** [Dual Spaces](dual-spaces.ipynb) · [Quotient Spaces](quotient-spaces.ipynb) · [Rank-Nullity](kernel-image-and-rank-nullity.ipynb#thm-rank-nullity)

A subspace $W \le V$ can be described in two ways: by the vectors it contains, or by the linear equations its vectors satisfy. The **annihilator** $W^\perp$ collects all those equations: the functionals that vanish on $W$. Its dimension is $\dim V - \dim W$, and taking the annihilator twice gives back $W$. In particular every subspace of $K^n$ is the solution set of $n - \dim W$ independent homogeneous linear equations.

Throughout, $V$ is a finite-dimensional [vector space](vector-spaces.ipynb) over a field $K$ with [dual space](dual-spaces.ipynb#def-dual-space) $V^* = \mathcal{L}(V, K)$; recall that [$\dim V^* = \dim V$](dual-spaces.ipynb#thm-dual-basis-is-basis).

## 1\. The Annihilator of a Subspace

::: {#def-annihilator}
## Annihilator

The **annihilator** of a subspace $W \le V$ is
$$W^\perp = \{\varphi \in V^* \mid \varphi(w) = 0 \text{ for all } w \in W\} \subseteq V^*.$$
:::

::: {#prp-annihilator-subspace}
## The Annihilator Is a Subspace

$W^\perp$ is a subspace of $V^*$. Moreover $\{0\}^\perp = V^*$, $V^\perp = \{0\}$, and $W_1 \subseteq W_2$ implies $W_2^\perp \subseteq W_1^\perp$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-annihilator-subspace
The zero functional vanishes on $W$. If $\varphi, \psi \in W^\perp$ and $a \in K$, then $(a\varphi + \psi)(w) = a\varphi(w) + \psi(w) = 0$ for all $w \in W$. Every functional vanishes on $\{0\}$; only the zero functional vanishes on all of $V$. A functional vanishing on $W_2$ vanishes on the smaller set $W_1$. $\blacksquare$
:::

::: {.callout-note}
## Why the Symbol $\perp$
On $\mathbb{R}^n$ every functional has the form $\varphi_a(x) = \langle a, x \rangle$ for a unique $a \in \mathbb{R}^n$. Then $\varphi_a \in W^\perp$ exactly when $a$ is orthogonal to every vector of $W$. So under the identification $a \leftrightarrow \varphi_a$, the annihilator is the orthogonal complement. The annihilator is the version of this idea that needs no inner product.
:::

::: {#exm-annihilator}
## An Annihilator in $\mathbb{R}^3$

Let $W = \operatorname{span}(1, 1, 0) \le \mathbb{R}^3$ and write functionals as $\varphi(x, y, z) = a x + b y + c z$. Then $\varphi \in W^\perp$ iff $\varphi(1, 1, 0) = a + b = 0$. So
$$W^\perp = \{a x - a y + c z\} = \operatorname{span}(\varphi_1, \varphi_2), \qquad \varphi_1 = x - y, \ \ \varphi_2 = z,$$
which has dimension $2 = 3 - 1$. And $W$ is exactly the set of solutions of the two equations $x - y = 0$, $z = 0$.
:::

## 2\. Annihilators and Quotients

::: {#thm-annihilator-quotient-iso}
## $W^\perp \cong (V/W)^*$

The map $T : W^\perp \to (V/W)^*$ defined by
$$T(\varphi)([v]) = \varphi(v) \qquad ([v] = v + W \in V/W)$$
is a linear isomorphism. (It needs no choice of basis.)
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-annihilator-quotient-iso
1. **$T(\varphi)$ is well defined.** If $[v] = [v']$, then $v - v' \in W$, so $\varphi(v) - \varphi(v') = \varphi(v - v') = 0$.
2. **$T(\varphi)$ is linear on $V/W$.** $T(\varphi)([v] + [v']) = T(\varphi)([v + v']) = \varphi(v + v') = \varphi(v) + \varphi(v')$, and similarly for scalars.
3. **$T$ is linear.** $T(a\varphi + \psi)([v]) = a\varphi(v) + \psi(v) = (a T(\varphi) + T(\psi))([v])$.
4. **$T$ is injective.** If $T(\varphi) = 0$, then $\varphi(v) = 0$ for every $v \in V$, so $\varphi = 0$.
5. **$T$ is surjective.** Given $\beta \in (V/W)^*$, define $\varphi = \beta \circ \pi$, where $\pi(v) = [v]$ is the [quotient map](quotient-spaces.ipynb#prp-quotient-projection). It is linear, and for $w \in W$, $\varphi(w) = \beta([w]) = \beta([0]) = 0$, so $\varphi \in W^\perp$. Finally $T(\varphi)([v]) = \beta([v])$, i.e. $T(\varphi) = \beta$. $\blacksquare$
:::

::: {.callout-note}
## Intuition
A functional that vanishes on $W$ cannot tell apart two vectors that differ by an element of $W$, so it is really a functional on the cosets of $W$, that is, on $V/W$. And every functional on $V/W$ arises this way.
:::

## 3\. Dimension and the Double Annihilator

::: {#cor-annihilator-dim}
## Dimension of the Annihilator

$\dim W^\perp = \dim V - \dim W$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-annihilator-dim
By @thm-annihilator-quotient-iso, the [dimension of a dual space](dual-spaces.ipynb#thm-dual-basis-is-basis) and the [dimension of a quotient](quotient-spaces.ipynb#thm-dim-quotient):
$$\dim W^\perp = \dim (V/W)^* = \dim (V/W) = \dim V - \dim W. \qquad \blacksquare$$
:::

::: {#thm-double-annihilator}
## The Double Annihilator

Identify $V$ with $V^{**}$ by the [natural isomorphism](dual-spaces.ipynb#thm-natural-iso) $\iota$. Then $(W^\perp)^\perp = W$; precisely, $(W^\perp)^\perp = \iota(W)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-double-annihilator
By definition $(W^\perp)^\perp = \{\Phi \in V^{**} \mid \Phi(\varphi) = 0 \text{ for all } \varphi \in W^\perp\}$.

**$\iota(W) \subseteq (W^\perp)^\perp$.** For $w \in W$ and $\varphi \in W^\perp$: $\iota(w)(\varphi) = \varphi(w) = 0$.

**Equal dimensions.** Applying @cor-annihilator-dim in $V$ and in $V^*$,
$$\dim (W^\perp)^\perp = \dim V^* - \dim W^\perp = \dim V - (\dim V - \dim W) = \dim W = \dim \iota(W).$$
A subspace contained in another of the same finite dimension [equals it](basis-and-dimension.ipynb#thm-subspace-dimension). $\blacksquare$
:::

::: {#cor-subspace-as-solutions}
## Every Subspace Is Cut Out by Equations

Let $\dim V = n$, $\dim W = k$, and let $\varphi_1, \dots, \varphi_{n-k}$ be a basis of $W^\perp$. Then
$$W = \{v \in V \mid \varphi_1(v) = \dots = \varphi_{n-k}(v) = 0\}.$$
In $K^n$: every $k$-dimensional subspace is the solution set of $n - k$ independent homogeneous linear equations, and fewer equations cannot suffice.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-subspace-as-solutions
Clearly $W$ is contained in the common zero set $Z$. If $v \in Z$, then every $\varphi \in W^\perp$, being a combination of the $\varphi_i$, vanishes at $v$; so $\iota(v) \in (W^\perp)^\perp = \iota(W)$ and $v \in W$, as $\iota$ is injective.

For the last claim, let $\psi_1, \dots, \psi_r$ be any functionals whose common zero set is $W$. The map $v \mapsto (\psi_1(v), \dots, \psi_r(v))$ from $V$ to $K^r$ is linear with kernel $W$, so by [rank-nullity](kernel-image-and-rank-nullity.ipynb#thm-rank-nullity) $k = \dim W \ge n - r$, i.e. $r \ge n - k$. $\blacksquare$
:::

The figure shows @exm-annihilator: the line $W$ is the intersection of the two planes $x - y = 0$ and $z = 0$, the zero sets of the basis $\varphi_1, \varphi_2$ of $W^\perp$. One equation would only give a plane; two independent equations are needed for a line in $\mathbb{R}^3$, matching $\dim W^\perp = 2$.


```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, RED = "#2a78d6", "#eb6834", "#e34948"
INK = "#0b0b0b"
plt.rcParams.update({"axes.titlesize": 11, "axes.titleweight": "bold", "font.size": 10})

fig = plt.figure(figsize=(7.5, 6))
ax = fig.add_subplot(projection="3d")
s, t = np.meshgrid(np.linspace(-2, 2, 12), np.linspace(-2, 2, 12))
ax.plot_surface(s, s, t, color=BLUE, alpha=0.18, linewidth=0)          # x - y = 0  (phi_1 = 0)
ax.plot_surface(s, t, 0 * s, color=ORANGE, alpha=0.18, linewidth=0)     # z = 0      (phi_2 = 0)
u = np.linspace(-2, 2, 2)
ax.plot(u, u, 0 * u, color=RED, lw=3.5)                                  # W = span(1, 1, 0)
ax.text(1.9, 1.9, 0.25, "$W$", color=INK, fontsize=13)
ax.text(-1.9, -1.9, 1.7, r"$\varphi_1 = x - y = 0$", color=INK, fontsize=10)
ax.text(1.5, -2.4, 0.05, r"$\varphi_2 = z = 0$", color=INK, fontsize=10)
ax.set_xlabel("$x$"); ax.set_ylabel("$y$"); ax.set_zlabel("$z$")
ax.view_init(elev=24, azim=-58)
ax.set_title(r"$W$ is cut out by a basis of $W^\perp$")
plt.tight_layout(); plt.show()
```
