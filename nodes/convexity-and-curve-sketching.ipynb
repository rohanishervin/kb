# Convexity, Inflection Points, and Curve Sketching

**Prerequisites:** [Linear Approximation and Its Error](linear-approximation.ipynb#thm-linear-approx-error) · [Local Extrema and the Mean Value Theorem](mean-value-theorem.ipynb)

Throughout, $f$ is twice differentiable on an open interval $I$.

## 1\. Convexity from the Second Derivative

::: {#prp-above-tangent}
## The Graph Lies Above Its Tangents

If $f'' > 0$ on $I$, then for every $a \in I$ and $x \in I$, $x \ne a$:
$$f(x) > f(a) + f'(a)(x - a).$$
If $f'' < 0$ the inequality is reversed.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-above-tangent
By the [error formula](linear-approximation.ipynb#thm-linear-approx-error), the difference is $\frac12 f''(c)(x - a)^2 > 0$. $\blacksquare$
:::

::: {#prp-chord-above}
## The Graph Lies Below Its Chords

If $f'' > 0$ on $I$ and $a < b$ in $I$, then for $0 < t < 1$
$$f\big((1 - t)a + tb\big) < (1 - t)f(a) + t f(b) .$$
If $f'' < 0$ the inequality is reversed.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-chord-above
Let $c = (1-t)a + tb$, so $c - a = t(b - a)$ and $b - c = (1-t)(b - a)$. By the [MVT](mean-value-theorem.ipynb#thm-mvt) there are $c_1 \in (a, c)$, $c_2 \in (c, b)$ with
$$\frac{f(c) - f(a)}{c - a} = f'(c_1) < f'(c_2) = \frac{f(b) - f(c)}{b - c},$$
the middle inequality because $f'$ is strictly increasing ($f'' > 0$). Multiplying by $t(1-t)(b - a) > 0$: $(1-t)\big(f(c) - f(a)\big) < t\big(f(b) - f(c)\big)$, which rearranges to $f(c) < (1-t)f(a) + tf(b)$. $\blacksquare$
:::

A function satisfying the chord inequality (with $\le$) is [convex](convex-sets-and-convex-hulls.ipynb#def-convex-function), or **concave up**; if the reverse holds, it is **concave** (concave down). So $f'' > 0$ implies strictly convex and $f'' < 0$ strictly concave.

::: {#def-inflection-point}
## Inflection Point

$b \in I$ is an **inflection point** of $f$ if $f''$ has one sign on some $(b - \delta, b)$ and the opposite sign on $(b, b + \delta)$: the graph changes from convex to concave or back.
:::

If $f''$ is continuous at an inflection point $b$, then $f''(b) = 0$: otherwise $f''$ would keep the sign of $f''(b)$ near $b$.

::: {#thm-second-derivative-test}
## Second Derivative Test

Let $f'(a) = 0$. If $f''(a) > 0$, $a$ is a strict local minimum; if $f''(a) < 0$, a strict local maximum. If $f''(a) = 0$ the test decides nothing ($x^4$, $-x^4$, $x^3$ at $0$).
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-second-derivative-test
Let $f''(a) > 0$. [Applied to $f'$](mean-value-theorem.ipynb#prp-derivative-sign), $f'(x) < f'(a) = 0$ on $(a - \delta, a)$ and $f'(x) > 0$ on $(a, a + \delta)$. By the [MVT](mean-value-theorem.ipynb#cor-mvt-consequences), $f$ is strictly decreasing on $(a - \delta, a]$ and strictly increasing on $[a, a + \delta)$, so $f(x) > f(a)$ for $0 < |x - a| < \delta$. The case $f''(a) < 0$ follows by applying this to $-f$. $\blacksquare$
:::

## 2\. Sketching Graphs

::: {#exm-sketch-quartic}
## $f(x) = x^4 - x^3$

$f' = x^2(4x - 3)$ and $f'' = 12x^2 - 6x = 6x(2x - 1)$.

| $x$ | $(-\infty, 0)$ | $0$ | $(0, \frac12)$ | $\frac12$ | $(\frac12, \frac34)$ | $\frac34$ | $(\frac34, 1)$ | $1$ | $(1, \infty)$ |
|---|---|---|---|---|---|---|---|---|---|
| $f$ | $+$ | $0$ | $-$ | $-$ | $-$ | $-\frac{27}{256}$ | $-$ | $0$ | $+$ |
| $f'$ | $-$ | $0$ | $-$ | $-$ | $-$ | $0$ | $+$ | $+$ | $+$ |
| $f''$ | $+$ | $0$ | $-$ | $0$ | $+$ | $+$ | $+$ | $+$ | $+$ |

$f$ decreases until $\frac34$ (global minimum), then increases; inflection points at $0$ and $\frac12$; the critical point $0$ is not an extremum.
:::

::: {#exm-sketch-x-minus-sin}
## $f(x) = x - \sin x$

$f' = 1 - \cos x \ge 0$, zero only at $2k\pi$, so $f$ is strictly increasing with horizontal tangents at $2k\pi$; $f'' = \sin x$ changes sign at every $k\pi$, which are the inflection points. $f(0) = 0$, so $f < 0$ for $x < 0$ and $f > 0$ for $x > 0$.
:::

::: {#exm-x-squared-sin}
## $f(x) = x^2\sin\frac1x$, $f(0) = 0$

$f'(0) = \lim_h h\sin\frac1h = 0$, and $f'(x) = 2x\sin\frac1x - \cos\frac1x$ for $x \ne 0$. As $x \to 0$, $f'(x)$ has no limit (the second term oscillates between $\pm1$), so $f'$ is discontinuous at $0$ and $f''(0)$ does not exist. Writing $f'(x) = 2x\cos\frac1x\big(\tan\frac1x - \frac{1}{2x}\big)$, the zeros of $f'$ are the points $x = \frac1t$ with $\tan t = \frac t2$; the line $\frac t2$ crosses every branch of $\tan$, so there are such points $x_n \to 0$, and $f'$ changes sign at each of them: local maxima and minima accumulate at $0$. The point $0$ is critical but neither an extremum nor an inflection point. The graph lies between $\pm x^2$.
:::

::: {#exm-positive-derivative-not-increasing}
## Positive Derivative, Yet Not Increasing Near the Point

$f(x) = \frac x2 + x^2\sin\frac1x$, $f(0) = 0$: as above $f'(0) = \frac12 > 0$ and $f'(x) = \frac12 + 2x\sin\frac1x - \cos\frac1x$ for $x \ne 0$. At $x_k = \frac{1}{2k\pi}$, $f'(x_k) = \frac12 - 1 = -\frac12 < 0$, so $f$ is decreasing near each $x_k$ ([sign of $f'$](mean-value-theorem.ipynb#prp-derivative-sign)). Since $x_k \to 0$, $f$ is not increasing on any interval containing $0$.
:::

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

fig, axs = plt.subplots(1, 3, figsize=(14, 4))
x = np.linspace(-0.6, 1.15, 400)
ax = axs[0]
ax.plot(x, x**4 - x**3, color=BLUE, lw=2)
ax.plot(0.75, -27 / 256, "o", color=YELLOW, mec=INK, ms=9, label="minimum $x = 3/4$")
ax.plot([0, 0.5], [0, 0.5**4 - 0.5**3], "s", color=VIOLET, ms=7, label="inflection $0,\\ 1/2$")
ax.legend(fontsize=9); ax.set_title(r"$x^4 - x^3$")

x = np.linspace(-0.25, 0.25, 20001); x = x[x != 0]
ax = axs[1]
ax.plot(x, x**2 * np.sin(1 / x), color=BLUE, lw=0.8)
ax.plot(x, x**2, color=AQUA, ls="--", lw=1); ax.plot(x, -x**2, color=AQUA, ls="--", lw=1)
ax.set_title(r"$x^2\sin(1/x)$ between $\pm x^2$")

x = np.linspace(-0.08, 0.08, 20001); x = x[x != 0]
ax = axs[2]
ax.plot(x, x / 2 + x**2 * np.sin(1 / x), color=BLUE, lw=0.8, label=r"$x/2 + x^2\sin(1/x)$")
ax.plot(x, x / 2, color=MUTED, ls="--", lw=1, label="$y = x/2$")
ax.legend(fontsize=9); ax.set_title(r"$f'(0) = 1/2$ but $f$ wiggles down near $0$")
plt.tight_layout(); plt.show()
```
