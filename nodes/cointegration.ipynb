# Cointegration and Error Correction Models

## 1\. Intuitive Concept of Cointegration

In many financial and economic applications, individual time series may be non-stationary (e.g., they follow a random walk), but certain linear combinations of them are [stationary](autocorrelation-and-stationarity.ipynb).

::: {#def-cointegration}
## Cointegration
Two or more non-stationary time series are said to be **cointegrated** if there exists a linear combination of them that is strictly stationary. 

Intuitively, cointegration implies that the series share a common underlying stochastic trend. They might drift apart in the short term, but they are bound together in the long term by an equilibrium relationship. 
:::

::: {.callout-note}
## Intuition: Drunk and her Dog
A classic analogy for cointegration is a drunk walking with her dog on a leash. Both the drunk's path and the dog's path are non-stationary (random walks). However, the distance between them (the linear combination) is bounded by the length of the leash (stationary).
:::

## 2\. Inadequacy of Regular VAR Models

When dealing with non-stationary (integrated) data, a naive approach might be to difference the series to make them stationary and then fit a regular [Vector Autoregressive (VAR)](vector-autoregressive-models.ipynb) model.

::: {.callout-warning}
## Limitations of differenced VAR
If the underlying variables are cointegrated, modeling only their differences with a standard VAR model is a misspecification. Differencing removes the valuable long-term equilibrium relationship (the cointegrating vector) from the data. The model would only capture short-term dynamics and ignore the fact that the variables are drawn back together over time.
:::

## 3\. Error Correction Models (ECM)

To properly model cointegrated series, we use an Error Correction Model (ECM) or a Vector Error Correction Model (VECM).

::: {#def-ecm}
## Error Correction Model
An Error Correction Model connects the short-term dynamics (the differenced series) with the long-term equilibrium (the cointegrating relationship). 

It takes the form:
$$
\Delta \vec{y}_t = \Pi \vec{y}_{t-1} + \sum_{i=1}^{p-1} \Gamma_i \Delta \vec{y}_{t-i} + \vec{a}_t
$$
where:

- $\Delta \vec{y}_t$ is the differenced (stationary) series.
- $\Pi \vec{y}_{t-1}$ is the **error correction term**. The matrix $\Pi$ captures the long-term cointegrating relationships and the speed of adjustment back to equilibrium.
- $\Gamma_i$ captures the short-term dynamics.
:::

The error correction term acts as a penalty: if the system deviates from its long-run equilibrium in period $t-1$, this term pushes it back towards equilibrium in period $t$.
