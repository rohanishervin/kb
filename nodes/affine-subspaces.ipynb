# Affine Subspaces

**Prerequisites:** [Quotient Spaces](quotient-spaces.ipynb) · [Dimension Formula for Sums](direct-sums-and-complements.ipynb#thm-dim-sum)

A [subspace](subspaces.ipynb#def-subspace) must pass through the origin. Lines and planes that do not pass through the origin are **affine subspaces**: subspaces shifted by a fixed vector. They describe solution sets of non-homogeneous linear equations and are exactly the elements of a [quotient space](quotient-spaces.ipynb).

Throughout, $V$ is a [vector space](vector-spaces.ipynb) over a field $K$.

## 1\. Definition and Direction

::: {#def-affine-subspace}
## Affine Subspace

An **affine subspace** (or **flat**) of $V$ is a subset of the form
$$E = a + W = \{a + w \mid w \in W\},$$
where $a \in V$ and $W \le V$ is a subspace. $W$ is the **direction** of $E$, and the **dimension** of $E$ is $\dim E := \dim W$. Flats of dimension $0$, $1$ and $2$ are points, lines and planes.
:::

::: {#prp-affine-direction}
## The Direction Is Determined by the Flat

If $a + W = b + W'$, then $W = W'$ and $b - a \in W$. Conversely, $a + W = b + W$ whenever $b - a \in W$; so any point of a flat can be used as its base point.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-affine-direction
Since $b = b + 0 \in a + W$, we have $b - a \in W$. Then
$$W' = (b + W') - b = (a + W) - b = (a - b) + W = W,$$
where the last step uses that adding a fixed element $a - b$ of $W$ maps $W$ onto itself. Conversely, if $b - a \in W$, then $b + w = a + ((b - a) + w) \in a + W$ for all $w \in W$, and symmetrically, so $a + W = b + W$. $\blacksquare$
:::

::: {#exm-affine}
## Flats in Practice

1. **A line not through the origin.** $\{(x, y) \in \mathbb{R}^2 \mid x + 2y = 3\} = (3, 0) + \operatorname{span}(-2, 1)$: the point $(3, 0)$ satisfies the equation, and two solutions differ by a solution of $x + 2y = 0$.
2. **Solutions of a linear system.** If $x_p$ is one solution of a system of linear equations, every other solution differs from $x_p$ by a solution of the homogeneous system. So the solution set is the flat $x_p + W$, where $W$ is the solution space of the homogeneous system (or it is empty). This is made precise with [kernels](kernel-image-and-rank-nullity.ipynb#def-kernel-image).
3. **Cosets.** The elements $x + W$ of the quotient $V/W$ are precisely the flats with direction $W$.
:::

## 2\. Parallel Flats

::: {#def-parallel-flats}
## Parallel Flats

Two flats $F = a + F_0$ and $G = b + G_0$ are **parallel** if one direction contains the other: $F_0 \subseteq G_0$ or $G_0 \subseteq F_0$.
:::

::: {#prp-parallel-flats}
## Parallel Flats Are Disjoint or Nested

If $F$ and $G$ are parallel, then either $F \cap G = \emptyset$, or one of them contains the other. In particular two distinct parallel flats of the same dimension are disjoint.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-parallel-flats
Say $F_0 \subseteq G_0$ and $p \in F \cap G$. By @prp-affine-direction we may use $p$ as base point of both: $F = p + F_0 \subseteq p + G_0 = G$. If the dimensions are equal, $F_0 \subseteq G_0$ with $\dim F_0 = \dim G_0$ forces $F_0 = G_0$ ([dimension of a subspace](basis-and-dimension.ipynb#thm-subspace-dimension)), so a common point gives $F = G$. $\blacksquare$
:::

## 3\. Intersections of Flats

::: {#prp-affine-intersection}
## Intersection of Two Flats

Let $E_1 = p + V_1$ and $E_2 = q + V_2$. Then

1. $E_1 \cap E_2 \ne \emptyset$ if and only if $q - p \in V_1 + V_2$;
2. if $y_0 \in E_1 \cap E_2$, then $E_1 \cap E_2 = y_0 + (V_1 \cap V_2)$, a flat with direction $V_1 \cap V_2$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-affine-intersection
1. A point lies in both flats iff it can be written $p + x = q + x'$ with $x \in V_1$, $x' \in V_2$, i.e. iff $q - p = x - x' \in V_1 + V_2$ for some such $x, x'$.
2. Using $y_0$ as base point of both flats, $E_1 = y_0 + V_1$ and $E_2 = y_0 + V_2$. A point $y_0 + z$ lies in both iff $z \in V_1$ and $z \in V_2$. $\blacksquare$
:::

Now consider two flats in an $n$-dimensional space whose dimensions add up to $n$, such as a line and a plane in $\mathbb{R}^3$. Whether their intersection survives small shifts depends only on their directions.

::: {#thm-affine-intersection-stability}
## Stable and Unstable Intersections

Let $\dim V = n$ and let $E_1 = p + V_1$, $E_2 = q + V_2$ be flats with $\dim E_1 + \dim E_2 = n$.

1. **Transverse case.** If $E_1 \cap E_2$ is a single point, then for every pair of translates $E_1' = p' + V_1$, $E_2' = q' + V_2$ the intersection $E_1' \cap E_2'$ is again a single point.
2. **Degenerate case.** If $E_1 \cap E_2$ is a flat of dimension $\ge 1$, then some translate $E_1' = p' + V_1$ does not meet $E_2$ at all.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-affine-intersection-stability
By the [dimension formula for sums](direct-sums-and-complements.ipynb#thm-dim-sum),
$$\dim (V_1 + V_2) = \dim V_1 + \dim V_2 - \dim (V_1 \cap V_2) = n - \dim (V_1 \cap V_2).$$

1. By @prp-affine-intersection, the intersection is $y_0 + (V_1 \cap V_2)$, and it is a single point, so $V_1 \cap V_2 = \{0\}$. Then $\dim(V_1 + V_2) = n$, so $V_1 + V_2 = V$ and in fact [$V = V_1 \oplus V_2$](direct-sums-and-complements.ipynb#prp-direct-sum-intersection). For any translates, $q' - p' \in V = V_1 + V_2$, so $E_1' \cap E_2' \neq \emptyset$, and by @prp-affine-intersection it is a flat with direction $V_1 \cap V_2 = \{0\}$, a single point.
2. Now $\dim (V_1 \cap V_2) \ge 1$, so $V_1 + V_2$ is a proper subspace of $V$. Pick $u \notin V_1 + V_2$ and set $p' = p - u$. If $E_1' \cap E_2 \ne \emptyset$, then $q - p' = (q - p) + u \in V_1 + V_2$; since $q - p \in V_1 + V_2$ (the original flats meet), this gives $u \in V_1 + V_2$, a contradiction. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Transversality Is Robust
In $\mathbb{R}^3$, a line that pierces a plane at one point keeps piercing every parallel plane, wherever the line and the plane are moved, as long as their directions stay the same. A line lying **inside** a plane is fragile: lift it slightly in a direction not contained in the plane and the two never meet. The theorem turns this picture into a statement about the directions $V_1$, $V_2$ only.
:::


```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, RED, VIOLET = "#2a78d6", "#eb6834", "#1baf7a", "#e34948", "#6250d6"
INK = "#0b0b0b"
plt.rcParams.update({"axes.titlesize": 11, "axes.titleweight": "bold", "font.size": 10})

fig = plt.figure(figsize=(12, 5))
U, Vg = np.meshgrid(np.linspace(-2, 2, 10), np.linspace(-2, 2, 10))
t = np.linspace(-2, 2, 50)

ax = fig.add_subplot(1, 2, 1, projection="3d")
ax.plot_surface(U, Vg, 0 * U, color=VIOLET, alpha=0.15, linewidth=0)
for shift, c, ls in [((0, 0), BLUE, "-"), ((1.2, 0.8), ORANGE, "--"), ((-1.0, 1.3), AQUA, "--")]:
    d = np.array([0.4, -0.3, 1.0])                       # direction of the line
    P = np.array([shift[0], shift[1], 0.0])[:, None] + d[:, None] * t
    ax.plot(*P, color=c, lw=2.4, ls=ls)
    ax.scatter(*shift, 0, color=c, s=40, depthshade=False)
ax.set_title("Transverse: every translate still meets\nthe plane in exactly one point")

ax = fig.add_subplot(1, 2, 2, projection="3d")
ax.plot_surface(U, Vg, 0 * U, color=VIOLET, alpha=0.15, linewidth=0)
ax.plot(t, 0.5 * t, 0 * t, color=BLUE, lw=2.6, label="line inside the plane")
ax.plot(t, 0.5 * t, 0 * t + 0.8, color=RED, lw=2.4, ls="--", label="lifted copy: no intersection")
ax.quiver(1, 0.5, 0, 0, 0, 0.75, color=INK, lw=1, arrow_length_ratio=0.2)
ax.legend(loc="upper left")
ax.set_title("Degenerate: a small shift destroys\nthe intersection")

for ax in fig.axes:
    ax.set_xlim(-2, 2); ax.set_ylim(-2, 2); ax.set_zlim(-1.5, 1.5)
    ax.set_xlabel("$x$"); ax.set_ylabel("$y$"); ax.set_zlabel("$z$")
    ax.view_init(elev=20, azim=-55)
plt.tight_layout(); plt.show()
```
