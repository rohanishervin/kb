# Graphs, Level Sets and Quadric Surfaces

**Prerequisites:** [Orthogonality, Projections and the Gram–Schmidt Process](orthogonality-and-gram-schmidt.ipynb#thm-real-spectral) · [The Cross Product](cross-product.ipynb#prp-cross-algebra) · [Kernel, Image, and the Rank-Nullity Theorem](kernel-image-and-rank-nullity.ipynb)

## 1\. Functions of Several Variables

A function $f : S \to \mathbb{R}^m$, $S \subseteq \mathbb{R}^n$, has **components** $f = (f_1, \dots, f_m)$, $f_i : S \to \mathbb{R}$. Three pictures of $f$ are useful.

::: {#def-graph-level-set}
## Graph, Image, Level Set

* The **graph** of $f$ is $\Gamma_f = \{(x, f(x)) : x \in S\} \subseteq \mathbb{R}^{n+m}$; for $f : \mathbb{R}^2 \to \mathbb{R}$ it is a surface $z = f(x, y)$ in $\mathbb{R}^3$.
* The **image** $f(S)$, often understood by following the images of lines or curves in $S$.
* For $c \in \mathbb{R}^m$, the **level set** $f^{-1}(c) = \{x \in S : f(x) = c\} = \bigcap_{i=1}^m f_i^{-1}(c_i)$; for $n = 2$, $m = 1$ these are the **level curves**, and $f^{-1}(c)$ is the projection to the $xy$-plane of the section of the graph by the plane $z = c$.
:::

::: {#exm-polar-image}
## Following Lines

$f(x, y) = (x\cos y, x\sin y)$ maps the vertical line $x = a$ onto the circle of radius $|a|$ about $0$ (onto the point $0$ if $a = 0$), and the horizontal line $y = b$ onto the line through $0$ with direction $(\cos b, \sin b)$. Hence the rectangle $[r_1, r_2]\times[\theta_1, \theta_2]$ ($0 < r_1$, $\theta_2 - \theta_1 < 2\pi$) is mapped onto a sector of an annulus: these are polar coordinates.
:::

::: {#exm-level-sets}
## Level Sets

1. For an affine map $f(x) = Lx + b$ ($L$ linear), $f^{-1}(c)$ is empty or a translate $p + \ker L$ of the kernel; e.g. for $f(x) = a\cdot x$ ($a \ne 0$) the level sets are the parallel hyperplanes $a\cdot x = c$ orthogonal to $a$.
2. $f(x, y, z) = x^2 + y^2 + z^2$: empty for $c < 0$, the origin for $c = 0$, the sphere of radius $\sqrt c$ for $c > 0$.
3. $f(x) = a\times x$ on $\mathbb{R}^3$ ($a \ne 0$): $f^{-1}(c)$ is empty unless $c \perp a$; if $c \perp a$ it is the line $\{b + ta\}$ with $b = \frac{c\times a}{|a|^2}$. Indeed $a\cdot(a\times x) = 0$; by the [rule](cross-product.ipynb#prp-cross-algebra) $a\times(c\times a) = |a|^2c - (a\cdot c)a = |a|^2c$ when $c \perp a$, so $a\times b = c$; and $a\times x = a\times b$ iff $a\times(x - b) = 0$ iff $x - b \parallel a$.
4. $f(x, y, z) = (z, x^2 + y^2 + z^2)$: $f^{-1}(c_1, c_2)$ is the circle of radius $\sqrt{c_2 - c_1^2}$ in the plane $z = c_1$ centred on the $z$-axis, a point if $c_2 = c_1^2$, empty if $c_2 < c_1^2$.
:::

## 2\. Quadratic Functions: Conics and Quadrics

A **quadratic function** on $\mathbb{R}^n$ is $q(x) = x^{\mathsf T}Ax + b\cdot x + c_0$ with $A$ real symmetric, $A \ne 0$ ($x^{\mathsf T}Ax = \sum_{i,j}a_{ij}x_ix_j$; a mixed term $2a_{ij}x_ix_j$ is split symmetrically).

::: {#prp-quadratic-normal-form}
## Removing Mixed Terms

There is an orthogonal $Q$ with $\det Q = 1$ such that in the rotated coordinates $x = Qy$,
$$q = \sum_{i=1}^n\lambda_iy_i^2 + \sum_{i=1}^n\beta_iy_i + c_0,$$
where $\lambda_i$ are the eigenvalues of $A$. For each $i$ with $\lambda_i \ne 0$, completing the square ($y_i \mapsto y_i - h_i$, a translation) removes $\beta_iy_i$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-quadratic-normal-form
By the [spectral theorem](orthogonality-and-gram-schmidt.ipynb#thm-real-spectral), $A = Q\Lambda Q^{\mathsf T}$ with $Q$ orthogonal; replacing the first column by its negative if necessary, $\det Q = 1$ (still an eigenbasis). Then $x^{\mathsf T}Ax = y^{\mathsf T}\Lambda y = \sum\lambda_iy_i^2$ and $b\cdot x = (Q^{\mathsf T}b)\cdot y$. Finally $\lambda y^2 + \beta y = \lambda\big(y + \frac{\beta}{2\lambda}\big)^2 - \frac{\beta^2}{4\lambda}$. $\blacksquare$
:::

Rotations and translations do not change the shape of a level set, so it suffices to classify the normal forms.

::: {#exm-conics}
## Conics ($n = 2$)

With $\lambda_1, \lambda_2$ the eigenvalues of $A$:

* **One eigenvalue zero**, say $\lambda_2 = 0 \ne \lambda_1$: $\lambda_1(y_1 - h)^2 + \beta_2y_2 + c = 0$, a **parabola** if $\beta_2 \ne 0$, otherwise two parallel lines, one line, or the empty set.
* **Same sign:** $\frac{(y_1 - h)^2}{a^2} + \frac{(y_2 - k)^2}{b^2} = C$, an **ellipse** for $C > 0$, a point for $C = 0$, empty for $C < 0$.
* **Opposite signs:** $\frac{(y_1 - h)^2}{a^2} - \frac{(y_2 - k)^2}{b^2} = C$, a **hyperbola** for $C \ne 0$ (opening along the $y_1$-axis if $C > 0$, along the $y_2$-axis if $C < 0$), and for $C = 0$ the pair of lines $b(y_1 - h) = \pm a(y_2 - k)$, the common asymptotes of the whole family.

For example the level curves of $x^2 - y^2$ are the hyperbolas $x^2 - y^2 = c$ with vertices $(\pm\sqrt c, 0)$ for $c > 0$, vertices $(0, \pm\sqrt{-c})$ for $c < 0$, and the lines $y = \pm x$ for $c = 0$.
:::

::: {#exm-quadrics}
## Quadric Surfaces ($n = 3$)

The non-degenerate normal forms ($a, b, c > 0$):

| equation | surface | sections $z = k$ |
|---|---|---|
| $\frac{x^2}{a^2} + \frac{y^2}{b^2} + \frac{z^2}{c^2} = 1$ | ellipsoid | ellipses for $\lvert k\rvert < c$ |
| $\frac{x^2}{a^2} + \frac{y^2}{b^2} - \frac{z^2}{c^2} = 1$ | hyperboloid of one sheet | ellipses with semi-axes $a\sqrt{1 + k^2/c^2}$, $b\sqrt{1 + k^2/c^2}$ |
| $\frac{x^2}{a^2} + \frac{y^2}{b^2} - \frac{z^2}{c^2} = -1$ | hyperboloid of two sheets | empty for $\lvert k\rvert < c$, ellipses for $\lvert k\rvert > c$ |
| $\frac{x^2}{a^2} + \frac{y^2}{b^2} = \frac{z^2}{c^2}$ | elliptic cone | ellipses, a point for $k = 0$ |
| $z = \frac{x^2}{a^2} + \frac{y^2}{b^2}$ | elliptic paraboloid | ellipses for $k > 0$ |
| $z = \frac{x^2}{a^2} - \frac{y^2}{b^2}$ | hyperbolic paraboloid (saddle) | hyperbolas, the lines $bx = \pm ay$ for $k = 0$ |

Sections by vertical planes are hyperbolas, parabolas or ellipses accordingly; e.g. the saddle consists of the downward parabolas $z = \frac{x_0^2}{a^2} - \frac{y^2}{b^2}$ hanging on the upward parabola $z = \frac{x^2}{a^2}$. The level sets $\frac{x^2}{a^2} + \frac{y^2}{b^2} - \frac{z^2}{c^2} = k$ of one function are hyperboloids of one sheet ($k > 0$), the cone ($k = 0$) and hyperboloids of two sheets ($k < 0$), all asymptotic to the cone. When a variable is missing, as in $f(x, y, z) = g(x, y)$, the level sets are **cylinders** over the level curves of $g$; e.g. $Ax + By + Cz^2 = 0$ is a parabolic cylinder.
:::

## 3\. Level Sets Do Not Determine the Function

$x + y$, $(x + y)^3$ and $(x + y)^{1/3}$ have the same level curves, the lines $x + y = \text{const}$, with different labels.

::: {#prp-same-level-sets}
## Same Level Sets Means Relabelling

Let $f, g : S \to \mathbb{R}$. If $g = \varphi\circ f$ with $\varphi$ injective on $f(S)$, then $f$ and $g$ have the same level sets. Conversely, if every level set of $f$ is a level set of $g$ and vice versa (non-empty ones), then $g = \varphi\circ f$ for an injective $\varphi : f(S) \to \mathbb{R}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-same-level-sets
If $\varphi$ is injective, $g(x) = g(x')$ iff $f(x) = f(x')$. Conversely define $\varphi(f(x)) = g(x)$: this is well defined, because $f(x) = f(x')$ puts $x, x'$ in one level set of $f$, which is a level set of $g$, so $g(x) = g(x')$; and $\varphi$ is injective by the symmetric argument. $\blacksquare$
:::

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

fig = plt.figure(figsize=(14, 8))
u = np.linspace(0, 2 * np.pi, 60)
def show(k, X, Y, Z, title, col):
    ax = fig.add_subplot(2, 3, k, projection="3d")
    ax.plot_surface(X, Y, Z, color=col, alpha=0.55, linewidth=0.2, edgecolor="white")
    ax.set_title(title); ax.set_box_aspect((1, 1, 0.9))
    return ax
v = np.linspace(0, np.pi, 30); U, V = np.meshgrid(u, v)
show(1, 2 * np.sin(V) * np.cos(U), 1.3 * np.sin(V) * np.sin(U), np.cos(V), "ellipsoid", BLUE)
w = np.linspace(-1.5, 1.5, 30); U, W = np.meshgrid(u, w)
show(2, np.cosh(W) * np.cos(U), np.cosh(W) * np.sin(U), np.sinh(W), "hyperboloid of one sheet", ORANGE)
w2 = np.linspace(0, 1.5, 25); U, W = np.meshgrid(u, w2)
X, Y, Z = np.sinh(W) * np.cos(U), np.sinh(W) * np.sin(U), np.cosh(W)
ax3 = show(3, X, Y, Z, "hyperboloid of two sheets", VIOLET)
ax3.plot_surface(X, Y, -Z, color=VIOLET, alpha=0.55, linewidth=0.2, edgecolor="white")
r = np.linspace(-1.5, 1.5, 30); U, R = np.meshgrid(u, r)
show(4, R * np.cos(U), R * np.sin(U), R, "cone", AQUA)
r2 = np.linspace(0, 1.4, 25); U, R = np.meshgrid(u, r2)
show(5, R * np.cos(U), R * np.sin(U), R**2, "elliptic paraboloid", YELLOW)
x = np.linspace(-1.5, 1.5, 40); X, Y = np.meshgrid(x, x)
show(6, X, Y, X**2 - Y**2, "hyperbolic paraboloid", RED)
plt.tight_layout(); plt.show()

fig, ax = plt.subplots(figsize=(5.2, 5))
cs = ax.contour(X, Y, X**2 - Y**2, levels=[-1.5, -1, -0.5, 0, 0.5, 1, 1.5], cmap="coolwarm")
ax.clabel(cs, fontsize=8); ax.set_aspect("equal"); ax.set_title(r"Level curves $x^2 - y^2 = c$")
plt.tight_layout(); plt.show()
```
