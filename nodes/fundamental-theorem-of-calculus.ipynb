# The Fundamental Theorem of Calculus

**Prerequisites:** [Properties of the Riemann Integral](properties-of-the-riemann-integral.ipynb) · [The Intermediate Value Theorem](intermediate-value-theorem.ipynb) · [Local Extrema and the Mean Value Theorem](mean-value-theorem.ipynb) · [The Chain Rule](chain-rule-and-inverse-functions.ipynb)

::: {#thm-mvt-integrals}
## Mean Value Theorem for Integrals

If $f : [a, b] \to \mathbb{R}$ is continuous, then $\int_a^b f = f(c)(b - a)$ for some $c \in [a, b]$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-mvt-integrals
Let $m = f(c_1)$ and $M = f(c_2)$ be the minimum and maximum ([EVT](semicontinuity-and-extreme-value-theorem.ipynb#cor-evt)). Then $m \le \frac{1}{b - a}\int_a^b f \le M$ ([monotonicity](properties-of-the-riemann-integral.ipynb#thm-integral-order)), and by the [IVT](intermediate-value-theorem.ipynb#thm-ivt) $f$ takes this value at some $c$ between $c_1$ and $c_2$. $\blacksquare$
:::

::: {#thm-ftc}
## Fundamental Theorem of Calculus

Let $f$ be continuous on an interval $I$, $a \in I$, and $F(x) = \int_a^x f$ for $x \in I$. Then $F$ is differentiable and $F' = f$ on $I$ (one-sided at endpoints of $I$).
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-ftc
For $h \ne 0$ with $x + h \in I$, [additivity](properties-of-the-riemann-integral.ipynb#thm-interval-additivity) gives $F(x + h) - F(x) = \int_x^{x+h} f$, and by @thm-mvt-integrals (for $h < 0$ applied on $[x + h, x]$, the signs cancel) this equals $f(c_h)\,h$ with $c_h$ between $x$ and $x + h$. So $\frac{F(x+h) - F(x)}{h} = f(c_h) \to f(x)$ as $h \to 0$, because $c_h \to x$ and $f$ is continuous. $\blacksquare$
:::

::: {#def-antiderivative}
## Antiderivative

$F$ is an **antiderivative** (primitive, indefinite integral) of $f$ on an interval $I$ if $F' = f$ on $I$. The theorem says every continuous $f$ has one. We write $\int f(x)\,dx = F(x) + C$.
:::

::: {#prp-antiderivatives-differ-by-constant}
## Antiderivatives Differ by Constants

If $F$ and $G$ are antiderivatives of $f$ on an interval, then $G - F$ is constant.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-antiderivatives-differ-by-constant
$(G - F)' = 0$ on an interval; apply the [MVT](mean-value-theorem.ipynb#cor-mvt-consequences). $\blacksquare$
:::

::: {.callout-warning}
## Only on Intervals
On a union of disjoint intervals the constants may differ on each piece. For example the antiderivatives of $\frac{1}{x^2}$ on $\mathbb{R} \setminus \{0\}$ are $-\frac1x + C_1$ for $x > 0$ and $-\frac1x + C_2$ for $x < 0$, with $C_1, C_2$ independent.
:::

::: {#thm-ftc-evaluation}
## Evaluating Integrals by Antiderivatives

Let $F : [a, b] \to \mathbb{R}$ be differentiable (one-sided at $a, b$) with $F' = f$ integrable on $[a, b]$. Then
$$\int_a^b f = F(b) - F(a) =: \big[F\big]_a^b .$$
In particular this holds for continuous $f$ and any antiderivative $F$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-ftc-evaluation
For any partition, the [MVT](mean-value-theorem.ipynb#thm-mvt) on each $[x_{i-1}, x_i]$ gives $t_i$ with
$$F(b) - F(a) = \sum_i \big(F(x_i) - F(x_{i-1})\big) = \sum_i f(t_i)\,\Delta x_i \in [L(f, \mathcal{P}), U(f, \mathcal{P})].$$
Since $\int_a^b f$ is the only number between all lower and all upper sums, it equals $F(b) - F(a)$. $\blacksquare$
:::

::: {#prp-variable-limits}
## Variable Limits of Integration

Let $f$ be continuous on an interval $I$ and $\alpha, \beta$ differentiable with values in $I$. Then $F(x) = \int_{\alpha(x)}^{\beta(x)} f$ is differentiable and
$$F'(x) = f(\beta(x))\,\beta'(x) - f(\alpha(x))\,\alpha'(x).$$
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-variable-limits
Fix $c \in I$ and $G(u) = \int_c^u f$, so $G' = f$. By additivity $F(x) = G(\beta(x)) - G(\alpha(x))$; apply the [chain rule](chain-rule-and-inverse-functions.ipynb#thm-chain-rule). $\blacksquare$
:::

::: {#exm-antiderivative-table}
## Basic Antiderivatives

Reading derivative formulas backwards (on intervals where everything is defined):

| $f(x)$ | $x^r$ ($r \in \mathbb{Q}$, $r \ne -1$) | $\sin x$ | $\cos x$ | $\sec^2 x$ | $\sec x\tan x$ | $\dfrac{1}{1 + x^2}$ | $\dfrac{1}{\sqrt{1 - x^2}}$ |
|---|---|---|---|---|---|---|---|
| $F(x)$ | $\dfrac{x^{r+1}}{r + 1}$ | $-\cos x$ | $\sin x$ | $\tan x$ | $\sec x$ | $\arctan x$ | $\arcsin x$ |

The missing case $r = -1$ leads to the [logarithm](exponential-and-logarithm.ipynb).
:::

::: {#exm-sums-as-integrals}
## Computing Integrals and Limits

1. $\int_0^\pi \sin x\,dx = [-\cos x]_0^\pi = 2$, and $\int_0^1 \frac{dx}{1 + x^2} = \arctan 1 = \frac\pi4$.
2. By [Riemann sums](properties-of-the-riemann-integral.ipynb#thm-riemann-sums) with $f(x) = \frac{1}{1 + x^2}$ on $[0, 1]$:
$$\lim_{n\to\infty}\sum_{i=1}^n \frac{n}{n^2 + i^2} = \lim_{n\to\infty}\frac1n\sum_{i=1}^n \frac{1}{1 + (i/n)^2} = \int_0^1\frac{dx}{1 + x^2} = \frac{\pi}{4}.$$
3. $\dfrac{d}{dx}\displaystyle\int_x^{x^2}\sin(t^2)\,dt = 2x\sin(x^4) - \sin(x^2)$.
:::

```python

import numpy as np
from scipy.integrate import quad

for n in [10, 100, 1000, 10000]:
    i = np.arange(1, n + 1)
    print(f"n = {n:6d}:  sum n/(n^2+i^2) = {np.sum(n / (n**2 + i**2)):.8f}")
print(f"pi/4             = {np.pi / 4:.8f}")

# derivative of x -> int_x^{x^2} sin(t^2) dt at x = 1.3, by a difference quotient
F = lambda x: quad(lambda t: np.sin(t**2), x, x**2)[0]
x, h = 1.3, 1e-5
print("difference quotient:", (F(x + h) - F(x - h)) / (2 * h), "| formula:", 2 * x * np.sin(x**4) - np.sin(x**2))
```
