# The Minimal Polynomial

**Prerequisites:** [Cayley-Hamilton Theorem](cayley-hamilton-theorem.ipynb) · [Division Algorithm](polynomial-division-algorithm.ipynb) · [Diagonalization](diagonalization.ipynb)

The [Cayley-Hamilton Theorem](cayley-hamilton-theorem.ipynb#thm-cayley-hamilton) shows that $A$ satisfies a polynomial equation of degree $n$. Often it already satisfies one of lower degree. The annihilating polynomial of **smallest** degree is the **minimal polynomial**. Together with the characteristic polynomial it carries much more information about an operator, especially when eigenvalues are repeated.

Throughout, $K$ is a field and $A \in M_n(K)$; for a polynomial $p$, $p(A)$ is as in [Cayley-Hamilton](cayley-hamilton-theorem.ipynb).

## 1\. The Characteristic Polynomial Is Not Enough

The characteristic polynomial does not always tell matrices apart. The identity $I_2$ and the shear $J = \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}$ have the same characteristic polynomial $(\lambda - 1)^2$, yet
$$I_2 - I = 0, \qquad J - I = \begin{pmatrix} 0 & 1 \\ 0 & 0 \end{pmatrix} \ne 0, \qquad (J - I)^2 = 0.$$
$I_2$ already satisfies the degree-one equation $\lambda - 1 = 0$, while $J$ needs degree two. The [diagonalizable](diagonalization.ipynb#def-diagonalizable-operator) $I_2$ and the [non-diagonalizable](triangularization-theorem.ipynb#prp-shear-not-diagonalizable) $J$ are distinguished by the lowest-degree equation they satisfy.

## 2\. Definition

::: {#def-minimal-polynomial}
## Minimal Polynomial

A polynomial $p \in K[x]$ **annihilates** $A$ if $p(A) = 0$. The **minimal polynomial** $m_A$ of $A$ is the monic polynomial of smallest degree that annihilates $A$. For an operator $f$, $m_f := m_{[f]_{\mathcal{B}}}$ for any basis $\mathcal{B}$.
:::

::: {#thm-minimal-polynomial-properties}
## Existence, Uniqueness and Divisibility

1. $m_A$ exists and is unique, and $1 \le \deg m_A \le n$.
2. For $p \in K[x]$: $p(A) = 0$ if and only if $m_A$ divides $p$. In particular $m_A$ divides the characteristic polynomial $p_A$.
3. Similar matrices have the same minimal polynomial, so $m_f$ is well defined.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-minimal-polynomial-properties
1. By Cayley-Hamilton, $p_A$ is a monic annihilating polynomial of degree $n$, so monic annihilators exist; pick one, $m$, of smallest degree. The constant $1$ does not annihilate $A$ (as $I \ne 0$), so $\deg m \ge 1$. If $m'$ is another monic annihilator of the same degree, then $m - m'$ annihilates $A$ and has smaller degree; if it were non-zero, dividing by its leading coefficient would give a monic annihilator of smaller degree, a contradiction. So $m = m'$.
2. If $p = s\, m_A$, then $p(A) = s(A)\, m_A(A) = 0$ (substitution is multiplicative for polynomials with scalar coefficients). Conversely let $p(A) = 0$ and [divide](polynomial-division-algorithm.ipynb#thm-polynomial-division) by the monic $m_A$: $p = s\, m_A + r$ with $\deg r < \deg m_A$. Then $r(A) = p(A) - s(A) m_A(A) = 0$, so $r = 0$ by the minimality of $\deg m_A$ (as in part 1). Hence $m_A$ divides $p$; for $p = p_A$ use Cayley-Hamilton.
3. If $B = P^{-1}AP$, then $p(B) = P^{-1} p(A) P$ for every polynomial $p$, so $A$ and $B$ have the same annihilating polynomials. $\blacksquare$
:::

## 3\. Roots of the Minimal Polynomial

::: {#thm-minimal-polynomial-roots}
## Same Roots as the Characteristic Polynomial

$\lambda \in K$ is a root of $m_A$ if and only if $\lambda$ is an eigenvalue of $A$. So $m_A$ and $p_A$ have the same roots in $K$; only the multiplicities may differ.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-minimal-polynomial-roots
If $A v = \lambda v$ with $v \ne 0$, then by the [spectral mapping property](eigenvalues-and-eigenvectors.ipynb#thm-spectral-mapping-polynomial) $0 = m_A(A) v = m_A(\lambda) v$, so $m_A(\lambda) = 0$.

Conversely, if $m_A(\mu) = 0$, then $m_A = (x - \mu)\, s$ by the [factor theorem](polynomial-division-algorithm.ipynb#cor-factor-theorem), with $s$ monic of smaller degree. By minimality $s(A) \ne 0$, so $s(A) w \ne 0$ for some $w$. Then $(A - \mu I)\, s(A) w = m_A(A) w = 0$, so $u = s(A) w$ is an eigenvector for $\mu$. $\blacksquare$
:::

::: {#exm-minimal-polynomials}
## Minimal Versus Characteristic Polynomials

| matrix | $p_A$ | $m_A$ |
|---|---|---|
| $I_2$ | $(x - 1)^2$ | $x - 1$ |
| $J = \begin{pmatrix} 1 & 1 \\ 0 & 1 \end{pmatrix}$ | $(x - 1)^2$ | $(x - 1)^2$ |
| $\operatorname{diag}(1, 1, 2)$ | $(x - 1)^2 (x - 2)$ | $(x - 1)(x - 2)$ |
| projection $P$ onto a line in $\mathbb{R}^2$ | $x(x - 1)$ | $x(x - 1)$ |
| $0 \ne N$ with $N^2 = 0$ | $x^2$ | $x^2$ |

For the diagonal matrix, $(D - I)(D - 2I) = \operatorname{diag}(0, 0, -1)\operatorname{diag}(-1, -1, 0) = 0$, while neither $D - I$ nor $D - 2I$ is $0$; by @thm-minimal-polynomial-roots both factors must appear, so $m_D = (x - 1)(x - 2)$.
:::

::: {.callout-note}
## Looking Ahead: The Minimal Polynomial and Diagonalizability
In all the examples, the diagonalizable matrices have a minimal polynomial without repeated roots, and the non-diagonalizable ones ($J$, $N$) have a repeated root. This is a general theorem: $A$ is diagonalizable over $K$ if and only if $m_A$ splits over $K$ into **distinct** linear factors. Together with the characteristic polynomial, the minimal polynomial leads to the decomposition of $V$ into generalized eigenspaces and to the Jordan normal form.
:::

The code computes $m_A$ directly from the definition: it looks for the first power $A^k$ that is a linear combination of $I, A, \dots, A^{k-1}$; the coefficients of that combination give $m_A$.


```python

import numpy as np

def minimal_polynomial(A, tol=1e-9):
    """Coefficients of m_A (highest degree first): the first k with A^k in span(I, A, ..., A^{k-1})."""
    n = len(A)
    powers = [np.eye(n).ravel()]
    for k in range(1, n + 1):
        Ak = np.linalg.matrix_power(A, k).ravel()
        M = np.column_stack(powers)                       # columns: vec(I), vec(A), ..., vec(A^{k-1})
        c, *_ = np.linalg.lstsq(M, Ak, rcond=None)
        if np.linalg.norm(M @ c - Ak) < tol:              # A^k = c_0 I + ... + c_{k-1} A^{k-1}
            return np.concatenate([[1.0], -c[::-1]])      # x^k - c_{k-1} x^{k-1} - ... - c_0
        powers.append(Ak)

examples = {
    "I_2": np.eye(2),
    "J (shear)": np.array([[1.0, 1.0], [0.0, 1.0]]),
    "diag(1, 1, 2)": np.diag([1.0, 1.0, 2.0]),
    "projection onto span(2,1)": np.array([[0.8, 0.4], [0.4, 0.2]]),
    "N, N^2 = 0": np.array([[0.0, 1.0], [0.0, 0.0]]),
}
for name, A in examples.items():
    m = minimal_polynomial(A)
    print(f"{name:28s} p_A = {np.round(np.poly(A), 6)}   m_A = {np.round(m, 6)}")
```
