# The Divergence Theorem and the Poincaré Lemma

**Prerequisites:** [Stokes' Theorem](stokes-theorem.ipynb) · [Green's Theorem](greens-theorem.ipynb#cor-green-divergence) · [Flux, Divergence and Curl](flux-divergence-and-curl.ipynb) · [Uniform Continuity and Compactness](uniform-continuity.ipynb)

## 1\. The Theorem

Let $M \subseteq \mathbb{R}^3$ be a bounded admissible region whose boundary $\partial M$ is a closed orientable surface made of finitely many smooth pieces, oriented by the **outward** unit normal $\mathbf n$.

::: {#thm-divergence}
## Divergence Theorem (Gauss)

For a $C^1$ vector field $G$ on an open set containing $M\cup\partial M$,
$$\iiint_M\operatorname{div}G\,dV = \iint_{\partial M}G\cdot\mathbf n\,dS :$$
the integral of the divergence over a solid equals the outward flux through its boundary.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-divergence
By linearity it suffices to prove $\iiint_M\frac{\partial G_3}{\partial z}\,dV = \iint_{\partial M}G_3\,\mathbf k\cdot\mathbf n\,dS$ and the analogous identities for $G_1\mathbf i$, $G_2\mathbf j$.
*Regions simple in the $z$-direction.* Let $M = \{(x, y) \in D,\ \alpha(x, y) \le z \le \beta(x, y)\}$ with $D$ a plane region and $\alpha \le \beta$ of class $C^1$. By [Fubini](iterated-integrals-and-fubini.ipynb#cor-simple-regions) and the FTC in $z$,
$$\iiint_M\frac{\partial G_3}{\partial z}\,dV = \iint_D\big[G_3(x, y, \beta(x, y)) - G_3(x, y, \alpha(x, y))\big]dA .$$
The top $z = \beta$, parametrized by $(x, y)$, has $\mathbf r_x\times\mathbf r_y = (-\beta_x, -\beta_y, 1)$, the outward normal direction, so its contribution is $\iint_D G_3\,\mathbf k\cdot(\mathbf r_x\times\mathbf r_y)\,dA = \iint_DG_3(x, y, \beta)\,dA$. On the bottom the outward normal is $-(\mathbf r_x\times\mathbf r_y)/|\cdots|$, giving $-\iint_DG_3(x, y, \alpha)\,dA$; on the vertical sides $\mathbf k\cdot\mathbf n = 0$. This proves the $z$-identity; if $M$ is also simple in the $x$- and $y$-directions (boxes, balls, ellipsoids, tetrahedra, ...), the other two follow in the same way.
*Unions.* If $M$ is cut into finitely many such pieces, the flux integrals over the cutting surfaces appear twice with opposite normals and cancel. The general case is obtained by approximation and is not carried out here. $\blacksquare$
:::

In the plane the corresponding statement is [Green's theorem in divergence form](greens-theorem.ipynb#cor-green-divergence). With $M_\rho$ the ball of radius $\rho$ about $P$, the [mean value theorem for integrals](multiple-integrals.ipynb#thm-mean-value-multiple) gives
$$\operatorname{div}G(P) = \lim_{\rho\to0}\frac{1}{\operatorname{vol}(M_\rho)}\iint_{\partial M_\rho}G\cdot\mathbf n\,dS,$$
the outward flux per unit volume: the rate at which the flow expands near $P$.

## 2\. Applications

::: {#exm-volume-by-flux}
## Volumes from Boundary Data

For $G(x) = x - P$, $\operatorname{div}G = 3$, so $\operatorname{vol}(M) = \frac13\iint_{\partial M}(x - P)\cdot\mathbf n\,dS$ for any point $P$.

1. **Cone.** Let $M$ be the cone with apex $P$ over a plane region $D$ (the union of the segments from $P$ to the points of $D$), at height $h$ above the plane of $D$. On the lateral surface, $x - P$ is tangent (the segment from $P$ to $x$ lies in the surface), so $(x - P)\cdot\mathbf n = 0$; on the base, $(x - P)\cdot\mathbf n$ is the length of the projection of $x - P$ on the normal of the plane, which is $h$. Hence $\operatorname{vol}(M) = \frac13h\operatorname{area}(D)$.
2. **Volume and surface.** With $r_P = \max_{q\in\partial M}|q - P|$ and $S$ the area of $\partial M$, $(x - P)\cdot\mathbf n \le r_P$ gives $\operatorname{vol}(M) \le \frac13r_PS$, with equality for a ball and its centre (the analogue of [this inequality](greens-theorem.ipynb#exm-area-perimeter)).
:::

::: {#exm-half-torus-flux}
## Closing Up a Surface

Find the flux of the constant field $G = A\mathbf i + B\mathbf j + C\mathbf k$ through the upper half $T^+$ of the torus $((a + b\cos u)\cos v, (a + b\cos u)\sin v, b\sin u)$, $0 \le u \le \pi$, with the outward normal. $T^+$ does not bound a solid, but together with the annulus $R = \{z = 0,\ (a - b)^2 \le x^2 + y^2 \le (a + b)^2\}$ it bounds the upper half of the solid torus, on whose boundary the outward normal of $R$ is $-\mathbf k$. Since $\operatorname{div}G = 0$,
$$\iint_{T^+}G\cdot\mathbf n\,dS = -\iint_RG\cdot(-\mathbf k)\,dS = C\operatorname{area}(R) = 4\pi abC .$$
(The standard normal of this parametrization points inward, which gives $-4\pi abC$.)
:::

::: {#exm-gauss-law}
## Gauss's Law

$G = \frac{x}{|x|^3}$ on $\mathbb{R}^3\setminus\{0\}$ has $\operatorname{div}G = 0$ ([computed here](flux-divergence-and-curl.ipynb#exm-div-curl)) and flux $4\pi$ through every sphere about $0$ ([computed here](flux-divergence-and-curl.ipynb#exm-flux-radial)). If $\partial M$ is a closed surface as above with $0$ in the interior of $M$, apply the theorem to $M$ minus a small ball about $0$: the outward flux through $\partial M$ equals the flux through the small sphere, $4\pi$. If $0 \notin M\cup\partial M$, the flux through $\partial M$ is $0$.
:::

## 3\. Vector Potentials and the Poincaré Lemma

A field $G$ has a **vector potential** $F$ if $G = \operatorname{curl}F$. Since [$\operatorname{div}\operatorname{curl} = 0$](flux-divergence-and-curl.ipynb#thm-curl-grad-div-curl), a $C^1$ field with a $C^2$ vector potential has $\operatorname{div}G = 0$. Again the converse needs a condition on the domain: $G = x/|x|^3$ has $\operatorname{div}G = 0$ on $\mathbb{R}^3\setminus\{0\}$, but if $G = \operatorname{curl}F$ there, the flux of $G$ through the unit sphere (a closed surface) would be $0$ by [Stokes' theorem](stokes-theorem.ipynb#cor-stokes-orientable), not $4\pi$.

A set $U$ is **star-shaped** with respect to $P \in U$ if for every $x \in U$ the segment from $P$ to $x$ lies in $U$; convex sets, in particular balls and $\mathbb{R}^3$, are star-shaped.

::: {#lem-differentiation-under-integral}
## Differentiating Under the Integral Sign

Let $U \subseteq \mathbb{R}^n$ be open and $g(x, t)$ continuous on $U\times[0, 1]$ with $\frac{\partial g}{\partial x_i}$ continuous there. Then $\varphi(x) = \int_0^1g(x, t)\,dt$ has $\frac{\partial\varphi}{\partial x_i}(x) = \int_0^1\frac{\partial g}{\partial x_i}(x, t)\,dt$, and this is continuous in $x$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-differentiation-under-integral
Fix $x$ and a closed ball $K \subseteq U$ about $x$. On the compact set $K\times[0, 1]$, $\partial_ig$ is [uniformly continuous](uniform-continuity.ipynb): given $\varepsilon$, there is $\delta$ with $|\partial_ig(y, t) - \partial_ig(x, t)| < \varepsilon$ for $|y - x| < \delta$, all $t$. For $0 < |h| < \delta$, the MVT in $x_i$ gives $\frac{g(x + he_i, t) - g(x, t)}{h} = \partial_ig(x + \theta he_i, t)$ with $\theta = \theta(t) \in (0, 1)$, so
$$\Big|\frac{\varphi(x + he_i) - \varphi(x)}{h} - \int_0^1\partial_ig(x, t)\,dt\Big| \le \int_0^1\varepsilon\,dt = \varepsilon .$$
The same uniform continuity shows $\big|\int_0^1\partial_ig(y, t)\,dt - \int_0^1\partial_ig(x, t)\,dt\big| \le \varepsilon$ for $|y - x| < \delta$. $\blacksquare$
:::

::: {#thm-poincare-lemma}
## Poincaré Lemma

Let $U \subseteq \mathbb{R}^3$ be open and star-shaped with respect to $P$, $F$ and $G$ of class $C^1$ on $U$, and $\mathbf r(t) = P + t(x - P)$, $0 \le t \le 1$.

1. If $\operatorname{curl}F = 0$, then $F = \nabla f$ with $f(x) = \displaystyle\int_0^1F(\mathbf r(t))\cdot\mathbf r'(t)\,dt$ (the line integral along the segment from $P$ to $x$).
2. If $\operatorname{div}G = 0$, then $G = \operatorname{curl}H$ with $H(x) = \displaystyle\int_0^1t\,G(\mathbf r(t))\times\mathbf r'(t)\,dt$ (componentwise integral).
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-poincare-lemma
Translate so that $P = 0$; then $\mathbf r(t) = tx$, $\mathbf r'(t) = x$, and @lem-differentiation-under-integral applies to all the integrals below.
(1) $f(x) = \int_0^1F(tx)\cdot x\,dt$, so
$$\frac{\partial f}{\partial x_1} = \int_0^1\Big[F_1(tx) + t\sum_jx_j\,\partial_1F_j(tx)\Big]dt = \int_0^1\Big[F_1(tx) + t\sum_jx_j\,\partial_jF_1(tx)\Big]dt = \int_0^1\frac{d}{dt}\big[tF_1(tx)\big]dt = F_1(x),$$
using $\partial_1F_j = \partial_jF_1$ (from $\operatorname{curl}F = 0$) and the chain rule $\frac{d}{dt}F_1(tx) = \sum_jx_j\partial_jF_1(tx)$. The other components are analogous.
(2) $H = \int_0^1t\,G(tx)\times x\,dt$ has $H_2 = \int_0^1t(G_3x_1 - G_1x_3)\,dt$, $H_3 = \int_0^1t(G_1x_2 - G_2x_1)\,dt$ (with $G = G(tx)$). Differentiating,
$$\partial_2H_3 - \partial_3H_2 = \int_0^1t\Big[2G_1 + t(x_2\partial_2G_1 + x_3\partial_3G_1) - tx_1(\partial_2G_2 + \partial_3G_3)\Big]dt .$$
By $\operatorname{div}G = 0$, $-(\partial_2G_2 + \partial_3G_3) = \partial_1G_1$, so the integrand is $2tG_1(tx) + t^2\,x\cdot\nabla G_1(tx) = \frac{d}{dt}\big[t^2G_1(tx)\big]$, and the integral is $G_1(x)$. The other components are analogous. $\blacksquare$
:::

So on a star-shaped domain (for example all of $\mathbb{R}^3$): a $C^1$ field is a gradient iff its curl vanishes, and is a curl iff its divergence vanishes.

```python

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

x, y, z, t, u, v, a, b, A, B, C = sp.symbols("x y z t u v a b A B C", real=True)
X = sp.Matrix([x, y, z])
def curl(F): return sp.Matrix([sp.diff(F[2], y) - sp.diff(F[1], z), sp.diff(F[0], z) - sp.diff(F[2], x), sp.diff(F[1], x) - sp.diff(F[0], y)])
F = sp.Matrix([2 * x * y + z**3, x**2, 3 * x * z**2])
f = sp.integrate(F.subs({x: t * x, y: t * y, z: t * z}, simultaneous=True).dot(X), (t, 0, 1))
print("curl F =", list(curl(F)), "| potential:", sp.expand(f), "| grad f - F = 0:", sp.simplify(sp.Matrix([sp.diff(f, s) for s in X]) - F) == sp.zeros(3, 1))
G = sp.Matrix([y * z, x**2, x * y])                           # div G = 0
H = sp.integrate(t * G.subs({x: t * x, y: t * y, z: t * z}, simultaneous=True).cross(X), (t, 0, 1))
print("div G =", sum(sp.diff(G[i], X[i]) for i in range(3)), "| H =", list(sp.expand(H)), "| curl H - G = 0:", sp.simplify(curl(H) - G) == sp.zeros(3, 1))
tor = sp.Matrix([(a + b * sp.cos(u)) * sp.cos(v), (a + b * sp.cos(u)) * sp.sin(v), b * sp.sin(u)])
N = tor.diff(u).cross(tor.diff(v))
print("flux of (A, B, C) through T+ with the standard (inward) normal:", sp.simplify(sp.integrate(sp.expand(sp.Matrix([A, B, C]).dot(N)), (u, 0, sp.pi), (v, 0, 2 * sp.pi))))
```
