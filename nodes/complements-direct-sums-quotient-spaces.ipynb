# Complements, Direct Sums, and Quotient Spaces

## 1\. Uniqueness of Basis Representation

Recall that a [basis](linear-independence-basis-dimension.ipynb#def-basis) $\{v_1, \dots, v_n\}$ for a vector space $E$ is a [linearly independent](linear-independence-basis-dimension.ipynb#def-linear-independence) set that [spans](linear-independence-basis-dimension.ipynb#def-span) $E$. Every element of $E$ can be written as a linear combination of the basis vectors. The following proposition strengthens this by showing the representation is unique.

::: {#prp-unique-representation}
## Unique Representation

Let $\{v_1, \dots, v_n\}$ span $E$. Then $\{v_1, \dots, v_n\}$ is linearly independent **if and only if** every element of $E$ has a **unique** representation as a linear combination of $v_1, \dots, v_n$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-unique-representation

**($\Rightarrow$) Independence implies uniqueness.** Suppose $\{v_1, \dots, v_n\}$ is linearly independent and some $x \in E$ has two representations:

$$x = c_1 v_1 + \cdots + c_n v_n = c'_1 v_1 + \cdots + c'_n v_n$$

Subtracting:

$$(c_1 - c'_1)v_1 + \cdots + (c_n - c'_n)v_n = 0$$

By linear independence, $c_i - c'_i = 0$ for all $i$, so $c_i = c'_i$.

**($\Leftarrow$) Uniqueness implies independence.** Suppose every element has a unique representation. If $\{v_1, \dots, v_n\}$ were linearly dependent, there would exist scalars $c_1, \dots, c_n$, not all zero, with:

$$c_1 v_1 + \cdots + c_n v_n = 0$$

But the zero vector already has the trivial representation $0 = 0 \cdot v_1 + \cdots + 0 \cdot v_n$. This gives two distinct representations of $0$, contradicting uniqueness.

$\blacksquare$
:::

::: {.callout-note}
## Intuition: Basis = Spanning + Unique Coordinates
A basis can be characterised in two equivalent ways:

1. Linearly independent + spanning, or
2. Spanning + unique representation.

Either formulation suffices. The unique coefficients $(c_1, \dots, c_n)$ are called the **coordinates** of $x$ with respect to the basis.
:::


## 2\. Well-Definedness of the Isomorphism $V \cong K^n$

We previously showed that every $n$-dimensional space $V$ over $K$ is [isomorphic](vector-spaces.ipynb#def-linear-isomorphism) to $K^n$. The construction deserves a careful revisit, because the key subtlety is **well-definedness**.

Choose a basis $\{v_1, \dots, v_n\}$ for $V$. By @prp-unique-representation, every $v \in V$ has a **unique** representation:

$$v = \alpha_1 v_1 + \cdots + \alpha_n v_n$$

Define $R: V \to K^n$ by $R(v) = (\alpha_1, \dots, \alpha_n)$.

- **Well-defined:** Since the coefficients $\alpha_i$ are uniquely determined, $R(v)$ is unambiguous.
- **Bijective:** $R$ is injective (distinct vectors have distinct coordinates by uniqueness) and surjective (any $(\alpha_1, \dots, \alpha_n) \in K^n$ is the image of $\alpha_1 v_1 + \cdots + \alpha_n v_n$).
- **Preserves operations:** $R(u + v) = R(u) + R(v)$ and $R(cv) = c \, R(v)$ follow from the component-wise definitions of addition and scalar multiplication in $K^n$.

::: {.callout-warning}
## Non-Canonical Isomorphism
The isomorphism $R$ depends on the **choice of basis**. A different basis yields a different isomorphism. There is no single "natural" identification of $V$ with $K^n$; infinitely many isomorphisms exist. This is why abstract vector spaces remain important — for instance, the space of $2 \times 2$ matrices is $4$-dimensional and isomorphic to $\mathbb{R}^4$, but no ordering of the four entries is canonical.
:::


## 3\. Linear Complements

::: {#def-complement}
## Linear Complement

Let $E$ be a [vector space](vector-spaces.ipynb) over $K$ and $F$ a [subspace](vector-spaces.ipynb#def-subspace) of $E$. A subspace $G \subseteq E$ is called a **linear complement** of $F$ in $E$ if every $u \in E$ can be written **uniquely** as:

$$u = f + g, \qquad f \in F,\; g \in G$$
:::

::: {#prp-complement-exists}
## Existence of Complements

Every subspace $F$ of a finite-dimensional space $E$ has a linear complement. Moreover, if $F \neq E$, then $F$ has infinitely many distinct complements.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-complement-exists

**Trivial cases.** If $F = E$, take $G = \{0\}$. If $F = \{0\}$, take $G = E$.

**General case.** Suppose $F \neq \{0\}$ and $F \neq E$. Let $\{v_1, \dots, v_k\}$ be a basis for $F$. By the [basis extension theorem](linear-independence-basis-dimension.ipynb#thm-extend-to-basis), extend it to a basis $\{v_1, \dots, v_k, v_{k+1}, \dots, v_n\}$ for $E$. Set:

$$G = \operatorname{span}(v_{k+1}, \dots, v_n)$$

**Every $u \in E$ decomposes.** Write $u = \alpha_1 v_1 + \cdots + \alpha_n v_n$. Then:

$$u = \underbrace{\alpha_1 v_1 + \cdots + \alpha_k v_k}_{\in F} + \underbrace{\alpha_{k+1} v_{k+1} + \cdots + \alpha_n v_n}_{\in G}$$

**Uniqueness.** Suppose $f + g = f' + g'$ with $f, f' \in F$ and $g, g' \in G$. Then $f - f' = g' - g$. The left side lies in $F = \operatorname{span}(v_1, \dots, v_k)$, the right in $G = \operatorname{span}(v_{k+1}, \dots, v_n)$. Writing both sides in terms of the basis:

$$\alpha_1 v_1 + \cdots + \alpha_k v_k = \alpha_{k+1} v_{k+1} + \cdots + \alpha_n v_n$$

Bringing everything to one side gives a linear combination of the basis $\{v_1, \dots, v_n\}$ equal to zero. By linear independence, all coefficients vanish, so $f = f'$ and $g = g'$.

**Infinitely many complements.** Since the extension of a basis is not unique (different choices of $v_{k+1}, \dots, v_n$ yield different complements), there are infinitely many complements when $F \neq E$.

$\blacksquare$
:::

```python

import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

fig = plt.figure(figsize=(13, 5))

# --- Panel 1: Line F and plane G as complement ---
ax1 = fig.add_subplot(121, projection='3d')
t = np.linspace(-2, 2, 50)

# F = span of (1,0,0) — the x-axis
ax1.plot(t, 0*t, 0*t, color='#4f8df5', linewidth=3, label='$F$ (line)')

# G = span of (0,1,0),(0,0,1) — the yz-plane
yy, zz = np.meshgrid(np.linspace(-1.5, 1.5, 10), np.linspace(-1.5, 1.5, 10))
xx = np.zeros_like(yy)
ax1.plot_surface(xx, yy, zz, alpha=0.15, color='#f5874f')
ax1.text(0, 1.6, 1.6, '$G$ (plane)', fontsize=11, color='#f5874f')

# Decomposition of a point
p = np.array([1.5, 1.0, 0.8])
ax1.scatter(*p, color='#8b5cf6', s=60, zorder=5)
ax1.plot([0, p[0]], [0, 0], [0, 0], '--', color='#4f8df5', alpha=0.6)
ax1.plot([p[0], p[0]], [0, p[1]], [0, p[2]], '--', color='#f5874f', alpha=0.6)
ax1.text(p[0]+0.1, p[1]+0.1, p[2]+0.1, '$u = f + g$', fontsize=10, color='#8b5cf6')

ax1.set_xlabel('$x$'); ax1.set_ylabel('$y$'); ax1.set_zlabel('$z$')
ax1.set_title('$\\mathbb{R}^3 = F \\oplus G$', fontsize=13)
ax1.view_init(elev=20, azim=30)

# --- Panel 2: Two different complements of the same line ---
ax2 = fig.add_subplot(122, projection='3d')
ax2.plot(t, 0*t, 0*t, color='#4f8df5', linewidth=3, label='$F$')

# G1 = yz-plane
ax2.plot_surface(xx, yy, zz, alpha=0.12, color='#f5874f')
ax2.text(0, 1.6, 1.6, "$G_1$", fontsize=11, color='#f5874f')

# G2 = a tilted plane through y and z+x
yy2, tt2 = np.meshgrid(np.linspace(-1.5, 1.5, 10), np.linspace(-1.5, 1.5, 10))
xx2 = tt2 * 0.6  # tilted
zz2 = tt2
ax2.plot_surface(xx2, yy2, zz2, alpha=0.12, color='#10b981')
ax2.text(1.0, 1.6, 1.0, "$G_2$", fontsize=11, color='#10b981')

ax2.set_xlabel('$x$'); ax2.set_ylabel('$y$'); ax2.set_zlabel('$z$')
ax2.set_title('Multiple complements of $F$', fontsize=13)
ax2.view_init(elev=20, azim=30)

plt.tight_layout()
plt.show()
```


## 4\. Direct Sum

### 4\.1\. Direct Sum of Two Subspaces

::: {#def-direct-sum}
## Direct Sum

If $F$ and $G$ are [linear complements](#def-complement) in $E$, we write:

$$E = F \oplus G$$

and say $E$ is the **direct sum** of $F$ and $G$.

- $E = F + G$ means every element of $E$ can be written as $f + g$ with $f \in F$, $g \in G$.
- $E = F \oplus G$ additionally requires the decomposition to be **unique**.
:::

::: {#prp-direct-sum-intersection}
## Direct Sum via Intersection

Let $F$ and $G$ be subspaces with $E = F + G$. Then $G$ is a complement of $F$ (i.e., $E = F \oplus G$) **if and only if**:

$$F \cap G = \{0\}$$
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-direct-sum-intersection

**($\Rightarrow$) Unique decomposition implies trivial intersection.**
Suppose $h \in F \cap G$ with $h \neq 0$. Then $h$ has two distinct decompositions:

$$h = h + 0 \quad (h \in F,\; 0 \in G) \qquad \text{and} \qquad h = 0 + h \quad (0 \in F,\; h \in G)$$

These are different since $h \neq 0$, contradicting uniqueness.

**($\Leftarrow$) Trivial intersection implies unique decomposition.**
Suppose $f + g = f' + g'$ with $f, f' \in F$ and $g, g' \in G$. Then:

$$f - f' = g' - g$$

The left side belongs to $F$, the right to $G$, so this common element lies in $F \cap G = \{0\}$. Hence $f = f'$ and $g = g'$.

$\blacksquare$
:::

---

### 4\.2\. Dimension Formula for Direct Sums

::: {#prp-dim-direct-sum}
## Dimension of a Direct Sum

If $E = F \oplus G$, then:

$$\dim(E) = \dim(F) + \dim(G)$$
:::

---

### 4\.3\. Direct Sum of Multiple Subspaces

::: {#def-direct-sum-multiple}
## Direct Sum of $s$ Subspaces

Let $E_1, \dots, E_s$ be subspaces of $E$. We write:

$$E = E_1 \oplus E_2 \oplus \cdots \oplus E_s$$

if every $v \in E$ can be written **uniquely** as $v = v_1 + \cdots + v_s$ with $v_i \in E_i$.
:::

::: {#prp-dim-direct-sum-multiple}
## Dimension of a Multiple Direct Sum

If $E = E_1 \oplus \cdots \oplus E_s$, then:

$$\dim(E) = \dim(E_1) + \cdots + \dim(E_s)$$
:::

::: {.callout-warning}
## Pairwise Trivial Intersection is Not Sufficient
For three or more subspaces, having $E_i \cap E_j = \{0\}$ for all $i \neq j$ does **not** guarantee a direct sum. For example, three lines through the origin in $\mathbb{R}^3$ may all lie in a single plane — they have pairwise trivial intersections, yet their sum is only $2$-dimensional, not $3$-dimensional, and decompositions are not unique.
:::


## 5\. Dimension Formula for Sums

::: {#thm-dim-sum}
## Dimension of a Sum of Two Subspaces

Let $F$ and $G$ be subspaces of a finite-dimensional space $E$ with $E = F + G$. Then:

$$\dim(F + G) = \dim(F) + \dim(G) - \dim(F \cap G)$$
:::

::: {.callout-tip collapse="true"}
## Proof Sketch of @thm-dim-sum

Let $\{w_1, \dots, w_r\}$ be a basis for $F \cap G$. Extend it to a basis $\{w_1, \dots, w_r, u_1, \dots, u_p\}$ for $F$, and separately extend it to a basis $\{w_1, \dots, w_r, z_1, \dots, z_q\}$ for $G$. One can show that:

$$\{w_1, \dots, w_r, u_1, \dots, u_p, z_1, \dots, z_q\}$$

is a basis for $F + G$. Counting elements:

$$\dim(F+G) = r + p + q = (r + p) + (r + q) - r = \dim(F) + \dim(G) - \dim(F \cap G)$$

$\blacksquare$
:::

::: {.callout-note}
## Intuition: Inclusion-Exclusion for Dimensions
This formula is the linear-algebraic analogue of the inclusion-exclusion principle for sets: $|A \cup B| = |A| + |B| - |A \cap B|$. When $F + G$ is a direct sum ($F \cap G = \{0\}$), the formula reduces to $\dim(F \oplus G) = \dim(F) + \dim(G)$.

Example: In $\mathbb{R}^3$, if $F$ and $G$ are two planes through the origin, then $\dim(F) = \dim(G) = 2$, $\dim(F \cap G) = 1$ (a line), and $\dim(F + G) = 2 + 2 - 1 = 3$.
:::


## 6\. Direct Product of Vector Spaces

::: {#def-direct-product}
## Direct Product

Let $E$ and $E'$ be vector spaces over the same [field](field.ipynb) $K$. Their **direct product** is the set of ordered pairs:

$$E \times E' = \{(v, v') \mid v \in E,\; v' \in E'\}$$

equipped with component-wise operations:

$$
\begin{aligned}
(v_1, v'_1) + (v_2, v'_2) &= (v_1 + v_2,\; v'_1 + v'_2) \\
\alpha(v, v') &= (\alpha v,\; \alpha v')
\end{aligned}
$$

With these operations, $E \times E'$ is a vector space over $K$.
:::

::: {#prp-dim-product}
## Dimension of a Direct Product

$$\dim(E \times E') = \dim(E) + \dim(E')$$
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-dim-product

Let $\{v_1, \dots, v_n\}$ be a basis for $E$ and $\{v'_1, \dots, v'_m\}$ a basis for $E'$. Then:

$$\{(v_1, 0), \dots, (v_n, 0), (0, v'_1), \dots, (0, v'_m)\}$$

is a basis for $E \times E'$. Indeed, any $(u, u') \in E \times E'$ can be written as:

$$
(u, u') = \alpha_1(v_1, 0) + \cdots + \alpha_n(v_n, 0) + \beta_1(0, v'_1) + \cdots + \beta_m(0, v'_m)
$$

where $u = \alpha_1 v_1 + \cdots + \alpha_n v_n$ and $u' = \beta_1 v'_1 + \cdots + \beta_m v'_m$. Linear independence follows from the independence of each basis in its component. The total number of basis vectors is $n + m$.

$\blacksquare$
:::

::: {.callout-note}
## Direct Product $\cong$ Direct Sum
If $E = F \oplus G$, then $E \cong F \times G$ via the natural isomorphism $f + g \mapsto (f, g)$. This is well-defined precisely because the decomposition $u = f + g$ is unique. Conversely, in $E \times E'$ the subspaces $E \times \{0\}$ and $\{0\} \times E'$ give a direct sum decomposition.
:::


## 7\. Quotient Spaces

### 7\.1\. The Equivalence Relation

Given a subspace $F \subseteq E$, we define a relation $\sim$ on $E$:

$$x \sim y \iff x - y \in F$$

::: {#prp-quotient-equiv}
## $\sim$ is an Equivalence Relation

The relation $\sim$ defined above is an equivalence relation on $E$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-quotient-equiv

We verify the three properties:

1. **Reflexive:** $x - x = 0 \in F$ (the zero vector is in every subspace), so $x \sim x$.
2. **Symmetric:** If $x \sim y$, then $x - y \in F$, so $y - x = -(x - y) \in F$ (subspaces are closed under scalar multiplication by $-1$), giving $y \sim x$.
3. **Transitive:** If $x \sim y$ and $y \sim z$, then $x - y \in F$ and $y - z \in F$. Since $F$ is closed under addition, $(x - y) + (y - z) = x - z \in F$, so $x \sim z$.

$\blacksquare$
:::

The **equivalence class** of $x$ is:

$$[x] = \{y \in E \mid y \sim x\} = \{x + f \mid f \in F\} = x + F$$

Geometrically, each class is a **translate** (or "coset") of $F$ — a flat parallel to $F$ passing through $x$. The space $E$ is partitioned into these parallel cosets.

```python

import numpy as np
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(8, 6))

# F is the line y = 0.5x (through origin)
t = np.linspace(-4, 4, 200)
slope = 0.5

# Draw several cosets (parallel translates of F)
offsets = np.linspace(-2.5, 2.5, 11)
colors = plt.cm.viridis(np.linspace(0.15, 0.85, len(offsets)))

for off, col in zip(offsets, colors):
    # coset: y = slope*x + off
    y = slope * t + off
    lw = 3.0 if abs(off) < 0.01 else 1.2
    label = '$F = [0]$' if abs(off) < 0.01 else None
    ax.plot(t, y, color=col, linewidth=lw, alpha=0.8, label=label)

# Highlight one coset
highlight_off = 1.5
ax.plot(t, slope*t + highlight_off, color='#ef4444', linewidth=2.5,
        label=f'$[x]$ (one equivalence class)', zorder=5)
ax.plot(1.0, slope*1.0 + highlight_off, 'o', color='#ef4444',
        markersize=8, zorder=6)
ax.annotate('$x$', (1.1, slope*1.0 + highlight_off + 0.15),
            fontsize=13, color='#ef4444')

ax.set_xlim(-3, 3)
ax.set_ylim(-3, 3)
ax.set_aspect('equal')
ax.grid(True, alpha=0.2)
ax.legend(fontsize=11, loc='upper left')
ax.set_title('Quotient space $E/F$: each line is one "point" of $E/F$',
             fontsize=13)
ax.set_xlabel('$x_1$', fontsize=12)
ax.set_ylabel('$x_2$', fontsize=12)

plt.tight_layout()
plt.show()
```


### 7\.2\. Vector Space Structure on $E/F$

Two key compatibility lemmas enable us to define operations on equivalence classes:

::: {#lem-quotient-compat}
## Compatibility of $\sim$ with Vector Operations

1. If $x \sim x'$ and $y \sim y'$, then $x + y \sim x' + y'$.
2. If $x \sim x'$ and $\alpha \in K$, then $\alpha x \sim \alpha x'$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-quotient-compat

**Part 1.** $x - x' \in F$ and $y - y' \in F$. Since $F$ is a subspace (closed under addition):

$$(x + y) - (x' + y') = (x - x') + (y - y') \in F$$

**Part 2.** $x - x' \in F$. Since $F$ is a subspace (closed under scalar multiplication):

$$\alpha x - \alpha x' = \alpha(x - x') \in F$$

$\blacksquare$
:::

---

::: {#def-quotient-space}
## Quotient Space

The **quotient space** $E/F$ is the set of all equivalence classes $\{[x] \mid x \in E\}$, equipped with:

$$
\begin{aligned}
[x] + [y] &= [x + y] \\
\alpha [x] &= [\alpha x]
\end{aligned}
$$

By @lem-quotient-compat, these operations are **well-defined** (independent of the choice of representatives). With these operations, $E/F$ is a vector space over $K$.

The **zero element** of $E/F$ is $[0] = F$ (the class of all vectors in $F$).
:::

::: {.callout-note}
## Intuition: Collapsing $F$ to Zero
Forming the quotient $E/F$ is the algebraic operation of "collapsing $F$ to a single point (zero)" and adjusting everything else accordingly. All vectors that differ by an element of $F$ become identified. Geometrically, each "point" of $E/F$ is an entire flat parallel to $F$.
:::


### 7\.3\. Dimension of the Quotient

::: {#thm-dim-quotient}
## Dimension of a Quotient Space

If $E$ is finite-dimensional and $F$ is a subspace, then:

$$\dim(E/F) = \dim(E) - \dim(F)$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-dim-quotient

Let $\dim(E) = n$ and $\dim(F) = k$. Take a basis $\{v_1, \dots, v_k\}$ for $F$ and extend it to a basis $\{v_1, \dots, v_k, v_{k+1}, \dots, v_n\}$ for $E$.

**Claim:** $\{[v_{k+1}], \dots, [v_n]\}$ is a basis for $E/F$.

**Linear independence.** Suppose $\alpha_1 [v_{k+1}] + \cdots + \alpha_{n-k} [v_n] = [0]$, i.e.,

$$[\alpha_1 v_{k+1} + \cdots + \alpha_{n-k} v_n] = [0]$$

This means $\alpha_1 v_{k+1} + \cdots + \alpha_{n-k} v_n \in F = \operatorname{span}(v_1, \dots, v_k)$. So:

$$\alpha_1 v_{k+1} + \cdots + \alpha_{n-k} v_n = \beta_1 v_1 + \cdots + \beta_k v_k$$

for some $\beta_j$. Rearranging gives a linear combination of the basis $\{v_1, \dots, v_n\}$ equal to zero. By linear independence of this basis, all coefficients vanish, and in particular $\alpha_1 = \cdots = \alpha_{n-k} = 0$.

**Spanning.** Any $[x] \in E/F$ satisfies:

$$x = \underbrace{\alpha_1 v_1 + \cdots + \alpha_k v_k}_{\in F} + \alpha_{k+1} v_{k+1} + \cdots + \alpha_n v_n$$

Since the first part lies in $F$, we have $x \sim \alpha_{k+1} v_{k+1} + \cdots + \alpha_n v_n$, so:

$$[x] = \alpha_{k+1}[v_{k+1}] + \cdots + \alpha_n [v_n]$$

Therefore $\{[v_{k+1}], \dots, [v_n]\}$ is a basis of size $n - k$.

$\blacksquare$
:::

---

### 7\.4\. Quotient $\cong$ Complement

::: {#prp-quotient-complement}
## Quotient is Isomorphic to any Complement

If $G$ is any linear complement of $F$ in $E$, then:

$$E/F \cong G$$
:::

::: {.callout-note}
## Intuition: Geometric Picture
Each equivalence class in $E/F$ is a coset $x + F$ — a flat parallel to $F$. Any complement $G$ is a subspace transverse to $F$, so each coset intersects $G$ in exactly one point. This bijection between cosets and points of $G$ is the isomorphism. Both spaces have dimension $\dim(E) - \dim(F)$.
:::
