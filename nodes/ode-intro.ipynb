# Ordinary Differential Equations (ODE)

Differential equations establish mathematical relations between unknown functions and their rates of change. When the unknown function depends on a single independent variable, the equation is an [ordinary differential equation]() (ODE), distinguished from a [partial differential equation]() (PDE) which involves multiple independent variables and partial derivatives.

## 1\. Mathematical Definition of an ODE

::: {#def-ode-implicit}
## Ordinary Differential Equation (Implicit Form)
Let $I \subseteq \mathbb{R}$ be an open interval, and let $y: I \to \mathbb{R}$ be an $n$-times differentiable function of the independent variable $x \in I$. An **ordinary differential equation (ODE)** of order $n \ge 1$ is an equation of the form:
$$
F\left(x, y, y', y'', \dots, y^{(n)}\right) = 0
$$
where $F: D \subseteq \mathbb{R}^{n+2} \to \mathbb{R}$ depends non-trivially on the highest derivative $y^{(n)}$.
:::

The **order** of an ODE is the positive integer $n \in \mathbb{N}_{\ge 1}$ representing the highest-order derivative appearing in the equation.

::: {#def-ode-standard}
## Standard Form (Explicit / Normal Form)
An $n$-th order ODE is in **standard form** (or **explicit / normal form**) if the highest derivative is isolated as an explicit function of the independent variable and lower-order derivatives:
$$
y^{(n)} = f\left(x, y, y', \dots, y^{(n-1)}\right)
$$
where $f: \Omega \subseteq \mathbb{R}^{n+1} \to \mathbb{R}$.
:::

::: {.callout-warning}
## Multivalued Ambiguity in Implicit Relations
An implicit equation $F(x, y, y') = 0$ may specify multiple distinct dynamical systems if it cannot be solved globally as a single-valued function for $y'$. For example, the implicit algebraic relation:
$$
(y')^2 - y + x = 0 \implies y' = \pm \sqrt{y - x}
$$
bifurcates on the domain $\{(x, y) \in \mathbb{R}^2 : y \ge x\}$ into two distinct standard-form ODEs:
$$
y' = \sqrt{y - x} \quad \text{and} \quad y' = -\sqrt{y - x}
$$
Each branch defines a different [Direction Field](ode-direction-fields.ipynb) with opposite trajectories. Explicit standard form $y' = f(x, y)$ eliminates this ambiguity.
:::


## 2\. Solutions, Linearity, and Initial Value Problems

::: {#def-ode-solution}
## Solution and General Solution
A function $\phi: I \to \mathbb{R}$ is an **explicit solution** of an $n$-th order ODE on an interval $I$ if $\phi \in C^n(I)$ and:
$$
F\left(x, \phi(x), \phi'(x), \dots, \phi^{(n)}(x)\right) = 0 \quad \forall x \in I
$$
The **general solution** is an $n$-parameter family $\phi(x; c_1, \dots, c_n)$ containing $n$ essential arbitrary constants. Fixing these parameters yields a **particular solution**.
:::

::: {#def-linear-ode}
## Linear Ordinary Differential Equation
An $n$-th order ODE is **linear** if it can be written as:
$$
a_0(x) y^{(n)} + a_1(x) y^{(n-1)} + \dots + a_n(x) y = b(x)
$$
where coefficient functions $a_0, a_1, \dots, a_n, b: I \to \mathbb{R}$ are continuous on $I$ with $a_0(x) \neq 0$ for all $x \in I$. 

The equation is **homogeneous** if $b(x) \equiv 0$, and **inhomogeneous** otherwise. Any ODE failing this affine structure (e.g., featuring $(y')^2$, $y y'$, or $\sin(y)$) is **nonlinear**.
:::

::: {#def-ivp}
## Initial Value Problem (IVP)
An **Initial Value Problem (Cauchy Problem)** pairs the $n$-th order ODE with $n$ initial conditions evaluated at a single base point $x_0 \in I$:
$$
\begin{cases}
y^{(n)} = f\left(x, y, y', \dots, y^{(n-1)}\right) \\
y(x_0) = y_0, \quad y'(x_0) = y_1, \quad \dots, \quad y^{(n-1)}(x_0) = y_{n-1}
\end{cases}
$$
For a first-order system:
$$
\begin{cases}
y' = f(x, y) \\
y(x_0) = y_0
\end{cases}
$$
Under [Picard's Theorem](ode-and-picards-thm.ipynb), if $f$ is continuous and [Lipschitz continuous]() in $y$ on an open set containing $(x_0, y_0)$, a unique local solution exists. When closed-form integration is intractable, solutions are analyzed qualitatively via [Direction Fields](ode-direction-fields.ipynb) or approximated computationally via [Euler's Method](euler-method.ipynb).
:::
