# Change of Variables in Multiple Integrals

**Prerequisites:** [Iterated Integrals and Fubini's Theorem](iterated-integrals-and-fubini.ipynb) · [The Inverse Function Theorem and Curvilinear Coordinates](inverse-function-theorem.ipynb) · [Determinants, Volume, and Orientation](determinant-volume-and-orientation.ipynb#thm-volume-scaling) · [Integration by Parts and Substitution](integration-by-parts-and-substitution.ipynb) · [Improper Integrals](improper-integrals.ipynb)

## 1\. The Formula

In one variable, if $\varphi : [\alpha, \beta] \to [a, b]$ is a $C^1$ bijection, then $\int_a^bf(t)\,dt = \int_\alpha^\beta f(\varphi(\tau))\,|\varphi'(\tau)|\,d\tau$ (the [substitution rule](integration-by-parts-and-substitution.ipynb); for $\varphi' < 0$ the reversal of the limits is absorbed by the absolute value). The factor $|\varphi'(\tau)|$ is the local stretching of lengths. In $n$ variables a $C^1$ map is near $p$ approximately affine, $\varphi(x) \approx \varphi(p) + D\varphi(p)(x - p)$, and a linear map [multiplies volumes](determinant-volume-and-orientation.ipynb#thm-volume-scaling) by $|\det|$; so a small box $Q$ at $p$ is mapped to a region of volume $\approx|\det D\varphi(p)|\operatorname{vol}(Q)$.

::: {#thm-change-of-variables}
## Change of Variables Formula

Let $W'$ and $W$ be admissible regions in $\mathbb{R}^n$ and $\varphi$ a $C^1$ map on an open set containing $\overline{W'}$, with $\varphi(W') = W$, such that $\varphi$ is injective and $\det D\varphi \ne 0$ on $W'$ outside a closed set of measure zero. Then for every bounded continuous $f$ on $W$,
$$\int_Wf = \int_{W'}(f\circ\varphi)\,|\det D\varphi| .$$
With $y = \varphi(x)$: $\displaystyle\int\cdots\int_Wf(y)\,dy_1\cdots dy_n = \int\cdots\int_{W'}f(\varphi(x))\,\Big|\det\frac{\partial(y_1, \dots, y_n)}{\partial(x_1, \dots, x_n)}\Big|\,dx_1\cdots dx_n$.
:::

The theorem is stated without proof. The idea: partition $W'$ into small boxes $Q$ with points $p_Q$; then $\sum_Qf(\varphi(p_Q))\operatorname{vol}(\varphi(Q)) \approx \sum_Qf(\varphi(p_Q))\,|\det D\varphi(p_Q)|\operatorname{vol}(Q)$; the left side approximates $\int_Wf$ and the right side is a Riemann sum for $\int_{W'}(f\circ\varphi)|\det D\varphi|$; making the approximation rigorous as the mesh tends to $0$ is the content of the proof. Unlike one variable, the absolute value is essential (in several variables the limits of iterated integrals are always taken in increasing order), and so is injectivity: a map covering part of $W$ twice would count it twice.

::: {#exm-coordinate-elements}
## Area and Volume Elements

* Polar: $dx\,dy = r\,dr\,d\theta$.
* Cylindrical: $dx\,dy\,dz = r\,dr\,d\theta\,dz$.
* Spherical: $dx\,dy\,dz = \rho^2\sin\varphi\,d\rho\,d\varphi\,d\theta$.

These follow from the [Jacobian determinants](inverse-function-theorem.ipynb#exm-cylindrical-spherical); the sets where the maps fail to be injective or have zero determinant ($r = 0$, $\theta \in \{0, 2\pi\}$, the $z$-axis) have measure zero. For example the ball of radius $R$ has volume $\int_0^{2\pi}\int_0^\pi\int_0^R\rho^2\sin\varphi\,d\rho\,d\varphi\,d\theta = 2\pi\cdot2\cdot\frac{R^3}3 = \frac43\pi R^3$.
:::

## 2\. Examples

::: {#exm-ellipsoid-volume}
## The Ellipsoid

$\varphi(X, Y, Z) = (aX, bY, cZ)$ maps the unit ball onto $\frac{x^2}{a^2} + \frac{y^2}{b^2} + \frac{z^2}{c^2} \le 1$ with $\det D\varphi = abc$; so the ellipsoid has volume $abc\cdot\frac43\pi = \frac43\pi abc$.
:::

::: {#exm-ball-cylinder}
## A Ball Pierced by a Cylinder

The solid inside the sphere $x^2 + y^2 + z^2 = 1$ and the cylinder $(x - \frac12)^2 + y^2 = \frac14$: in cylindrical coordinates the cylinder is $r = \cos\theta$, $-\frac\pi2 \le \theta \le \frac\pi2$, so
$$V = \int_{-\pi/2}^{\pi/2}\int_0^{\cos\theta}\int_{-\sqrt{1-r^2}}^{\sqrt{1-r^2}}r\,dz\,dr\,d\theta = \int_{-\pi/2}^{\pi/2}\tfrac23\big(1 - |\sin\theta|^3\big)d\theta = \tfrac{2\pi}{3} - \tfrac89 .$$
:::

::: {#exm-ice-cream}
## A Spherical Cone

The solid bounded above by the sphere $\rho = R$ and below by the cone $\varphi = \alpha$ ($0 < \alpha \le \pi$) has volume
$$\int_0^{2\pi}\int_0^\alpha\int_0^R\rho^2\sin\varphi\,d\rho\,d\varphi\,d\theta = \tfrac{2\pi}{3}R^3(1 - \cos\alpha),$$
which is $\frac43\pi R^3$ for $\alpha = \pi$.
:::

::: {#exm-parallelogram-area}
## A Parallelogram

The parallelogram between the lines $A_1x + B_1y = C_1, C_1'$ and $A_2x + B_2y = C_2, C_2'$ (with $\Delta = A_1B_2 - A_2B_1 \ne 0$) becomes a rectangle in $u = A_1x + B_1y$, $v = A_2x + B_2y$. Since $\det\frac{\partial(u, v)}{\partial(x, y)} = \Delta$, the [inverse map](inverse-function-theorem.ipynb#thm-inverse-function) has $\big|\det\frac{\partial(x, y)}{\partial(u, v)}\big| = \frac1{|\Delta|}$, and the area is $\dfrac{|C_1' - C_1|\,|C_2' - C_2|}{|A_1B_2 - A_2B_1|}$.
:::

## 3\. Improper Integrals

Let $D \subseteq \mathbb{R}^n$ be, for instance, an admissible region with finitely many points removed, or $\mathbb{R}^n$, or a region bounded by finitely many hyperplanes, and $f \ge 0$ continuous on $D$, possibly unbounded. An **exhaustion** of $D$ is an increasing sequence $D_1 \subseteq D_2 \subseteq \cdots$ of compact admissible sets with union $D$ such that every compact subset of $D$ lies in some $D_k$.

::: {#prp-improper-multiple}
## Improper Integrals of Non-Negative Functions

If $f \ge 0$ is continuous on $D$ and $\lim_k\int_{D_k}f$ is finite for one exhaustion, then it is finite and has the same value for every exhaustion. This common value is the **improper integral** $\int_Df$, which is then said to **converge**.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-improper-multiple
For $f \ge 0$, $I_k = \int_{D_k}f$ increases with $k$, so its limit exists in $[0, \infty]$. Let $(D'_j)$ be another exhaustion. Each $D_k$ is a compact subset of $D$, so $D_k \subseteq D'_j$ for some $j$, whence $I_k \le \int_{D'_j}f \le \lim_jI'_j$. Thus $\lim I_k \le \lim I'_j$, and by symmetry equality holds. $\blacksquare$
:::

The sign condition matters: for $f(x) = \frac1x$ on $[-1, 1]\setminus\{0\}$, the exhaustion $[-1, -\frac1k]\cup[\frac1k, 1]$ gives integrals $0$, while $[-1, -\frac1k]\cup[\frac1{2k}, 1]$ gives $\ln2$.

::: {#exm-gaussian-integral}
## The Gaussian Integral

Exhaust $\mathbb{R}^2$ by the disks $D_k = \{x^2 + y^2 \le k^2\}$: in polar coordinates
$$\iint_{D_k}e^{-(x^2 + y^2)}\,dA = \int_0^{2\pi}\int_0^ke^{-r^2}r\,dr\,d\theta = \pi(1 - e^{-k^2}) \to \pi .$$
Exhausting instead by the squares $[-a, a]^2$ and using Fubini, $\iint_{[-a,a]^2}e^{-x^2}e^{-y^2}\,dA = \big(\int_{-a}^ae^{-x^2}\,dx\big)^2 \to \pi$. Hence
$$\int_{-\infty}^{\infty}e^{-x^2}\,dx = \sqrt\pi,$$
the normalizing constant of the [normal distribution](normal-distribution.ipynb).
:::

```python

from scipy.integrate import tplquad, quad
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

V, _ = tplquad(lambda z, r, t: r, -np.pi / 2, np.pi / 2, lambda t: 0, lambda t: np.cos(t),
               lambda t, r: -np.sqrt(1 - r**2), lambda t, r: np.sqrt(1 - r**2))
print("ball pierced by cylinder:", V, "| 2pi/3 - 8/9 =", 2 * np.pi / 3 - 8 / 9)
a, R = 0.7, 1.3
Vc, _ = tplquad(lambda rho, ph, th: rho**2 * np.sin(ph), 0, 2 * np.pi, 0, a, 0, R)
print("spherical cone:", Vc, "| formula:", 2 * np.pi / 3 * R**3 * (1 - np.cos(a)))
for A in [1, 2, 3, 5]:
    print(f"(int_-{A}^{A} e^(-x^2) dx)^2 = {quad(lambda x: np.exp(-x**2), -A, A)[0]**2:.10f}", end="; ")
print("pi =", np.pi)

fig, (ax, ax2) = plt.subplots(1, 2, figsize=(11, 4.6))
r0, t0, dr, dt = 1.5, 0.6, 0.25, 0.35
ax.add_patch(plt.Rectangle((r0, t0), dr, dt, color=BLUE, alpha=0.35))
ax.set_xlim(0, 2.5); ax.set_ylim(0, 1.5); ax.set_xlabel("$r$"); ax.set_ylabel(r"$\theta$")
ax.set_title(r"box in the $(r, \theta)$-plane, area $\Delta r\,\Delta\theta$")
rr = np.linspace(r0, r0 + dr, 30); tt = np.linspace(t0, t0 + dt, 30)
edge = np.concatenate([np.stack([rr, np.full(30, t0)]), np.stack([np.full(30, r0 + dr), tt]),
                       np.stack([rr[::-1], np.full(30, t0 + dt)]), np.stack([np.full(30, r0), tt[::-1]])], axis=1)
ax2.fill(edge[0] * np.cos(edge[1]), edge[0] * np.sin(edge[1]), color=ORANGE, alpha=0.4)
th = np.linspace(0, np.pi / 2, 100)
for rad in [r0, r0 + dr]:
    ax2.plot(rad * np.cos(th), rad * np.sin(th), color=MUTED, lw=0.8, ls=":")
ax2.set_aspect("equal"); ax2.set_xlim(0, 2); ax2.set_ylim(0, 1.8)
ax2.set_title(r"its image: area $\approx r\,\Delta r\,\Delta\theta$")
plt.tight_layout(); plt.show()
area = quad(lambda r: r * dt, r0, r0 + dr)[0]
print("exact image area:", area, "| r0*dr*dt:", r0 * dr * dt)
```
