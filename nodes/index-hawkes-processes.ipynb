## Notation

The same symbols mean the same thing in every note.

| Symbol | Meaning |
|---|---|
| $T_1 < T_2 < \cdots$, $E_k = T_k - T_{k-1}$ | event times, inter-arrival times ($T_0 = 0$) |
| $N_t$, $N(a, b]$ | number of events in $(0, t]$, in $(a, b]$ (right-continuous) |
| $\mathcal{H}_t$ | history: the information in the event times up to $t$ |
| $\lambda(t)$, $\Lambda(t)$ | deterministic rate and integrated rate of a Poisson process |
| $\lambda_t$, $\Lambda_t = \int_0^t\lambda_s\,ds$ | conditional intensity (left-continuous), compensator |
| $\mu$ (or $\mu(t)$) | baseline (exogenous) intensity |
| $\varphi$, $g = \varphi/\eta$, $\Phi(x) = \int_0^x\varphi$ | excitation kernel, delay density, integrated kernel |
| $\eta = \lVert\varphi\rVert_1$ | branching ratio (expected number of children per event) |
| $\alpha, \beta$; $\kappa = \beta - \alpha$ | exponential kernel $\alpha e^{-\beta t}$; mean-reversion rate of $\lambda$ |
| $\bar\lambda = \mu/(1 - \eta)$ | stationary mean intensity |
| $m(t) = E\,\lambda_t$, $R = \sum_k\varphi^{*k}$ | mean intensity, resolvent |
| $c(u)$, $F(\tau)$ | covariance density of counts, Fano factor (variance/mean of counts in windows of length $\tau$) |
| $\mathcal{A}$ | generator of the Markov process $(N_t, \lambda_t)$ (exponential kernel) |
| $\tau_k = \Lambda_{T_k}$, $r_k = \tau_k - \tau_{k-1}$ | rescaled times, residual gaps |
| $\ell$, $A_i$, $B_i$ | log-likelihood; recursions $\sum_{j<i}e^{-\beta(t_i - t_j)}$ and $\sum_{j<i}(t_i - t_j)e^{-\beta(t_i - t_j)}$ |
| $Z_i$, $p_{i0}$, $p_{ij}$ | parent label of event $i$; probability that it is an immigrant, a child of event $j$ |
| $\varphi_{ij}$, $\Phi = (\lVert\varphi_{ij}\rVert_1)$, $\rho(\Phi)$ | effect of type $j$ on type $i$, branching matrix, spectral radius |
| $M_i$, $\psi$ | marks, impact function of a mark |
| $X_t = N^+_t - N^-_t$, $C(\tau)$ | tick price, signature plot (realized variance per unit time at sampling interval $\tau$) |
