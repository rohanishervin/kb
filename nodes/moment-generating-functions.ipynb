# Moment Generating Functions

## 1. Definition

The **moment generating function (MGF)** of a [random variable](random-variables.ipynb) $X$ is
$$
M(t) = E(e^{tX})
$$
provided this [expectation](expectation.ipynb#the-concept-of-expected-value) is finite on some open interval $(-a,a)$ containing $0$. If no such interval exists, the MGF does not exist.

::: {.callout-note}
## The variable $t$
The variable $t$ has no probabilistic meaning; it is a bookkeeping device that lets us apply calculus (derivatives) to extract discrete sequences of moments.
:::

::: {.callout-warning}
## Quick sanity check
Every valid MGF satisfies $M(0)=E(e^{0\cdot X})=E(1)=1$. Always verify this after computing an MGF.
:::

::: {.callout-warning}
## Existence
Not all random variables have an MGF (moments may grow too fast for convergence). Replacing $t$ with $it$ gives the **characteristic function** $\psi(t)=E(e^{itX})$, which *always* exists.
:::

## 2. Fundamental Properties

**Theorem 2.1 (Moments via derivatives).** $E(X^n) = M^{(n)}(0)$.

::: {.callout-tip collapse="true"}
## Proof
Expand $M(t)$ two ways:

- **Taylor series of $M$:** $\;M(t) = \sum_{n=0}^{\infty} M^{(n)}(0)\,\frac{t^n}{n!}$
- **Definition:** $\;M(t) = E(e^{tX}) = E\!\left(\sum_{n=0}^{\infty} X^n\frac{t^n}{n!}\right) = \sum_{n=0}^{\infty} E(X^n)\frac{t^n}{n!}$

(Swapping $E$ and $\sum$ is justified by finiteness on $(-a,a)$.)

Matching coefficients of $t^n/n!$ gives $E(X^n) = M^{(n)}(0)$. $\blacksquare$
:::

::: {.callout-note}
## Bypassing LOTUS
Computing moments normally requires integrating $x^n f(x)$ via [LOTUS](lotus.ipynb#law-of-the-unconscious-statistician-lotus). This theorem replaces that integration with differentiation.
:::

---

**Theorem 2.2 (Uniqueness).** If two random variables have the same MGF on some interval $(-a,a)$ containing $0$, they have the same distribution. *(Proof omitted — requires advanced analysis.)*

---

**Theorem 2.3 (Sum of independents).** If $X \perp Y$, then $M_{X+Y}(t) = M_X(t)\,M_Y(t)$.

::: {.callout-tip collapse="true"}
## Proof
$M_{X+Y}(t) = E(e^{t(X+Y)}) = E(e^{tX}e^{tY}) = E(e^{tX})E(e^{tY}) = M_X(t)M_Y(t)$,

where the penultimate equality uses independence. $\blacksquare$
:::

---

**Proposition 2.4 (Location-scale).** If $X$ has MGF $M(t)$, then the MGF of $a + bX$ is $e^{at}M(bt)$.

::: {.callout-tip collapse="true"}
## Proof
$M_{a+bX}(t) = E(e^{t(a+bX)}) = E(e^{at}e^{btX}) = e^{at}E(e^{btX}) = e^{at}M(bt)$,

where $e^{at}$ is a constant pulled out of the expectation. $\blacksquare$
:::

## 3. Generating Moments via Taylor Expansion

Instead of differentiating, we can read off *all* moments at once by expanding $M(t)$ as a power series and matching coefficients of $t^n/n!$.

**Example 1 — Expo(1) moments.** $M(t)=\frac{1}{1-t}=\sum_{n=0}^{\infty}t^n = \sum_{n=0}^{\infty}n!\,\frac{t^n}{n!}$ for $|t|<1$, so $E(X^n)=n!$.  
By Prop. 2.4, $Y=X/\lambda\sim\text{Expo}(\lambda)$ gives $E(Y^n)=n!/\lambda^n$.

**Example 2 — $\mathcal{N}(0,1)$ moments.** $M(t)=e^{t^2/2}=\sum_{n=0}^{\infty}\frac{t^{2n}}{2^n\,n!} = \sum_{n=0}^{\infty}\frac{(2n)!}{2^n\,n!}\frac{t^{2n}}{(2n)!}$.  
So $E(Z^{2n})=\frac{(2n)!}{2^n\,n!}$ and all odd moments are $0$ (by symmetry): $E(Z^2)=1,\;E(Z^4)=3,\;E(Z^6)=15,\ldots$

**Example 3 — Log-Normal moments.** If $Y=e^X$ with $X\sim\mathcal{N}(\mu,\sigma^2)$, the MGF of $Y$ does not exist (the integral diverges for $t>0$). However, its moments are still accessible:
$$
E(Y^n) = E(e^{nX}) = M_X(n) = e^{n\mu + \frac{1}{2}n^2\sigma^2}
$$

::: {.callout-warning}
## Nomenclature: Log-Normal
"Log-Normal" means **"the log is Normal"**, not "the log of a Normal". The parameters $\mu,\sigma^2$ describe the underlying Normal, not the Log-Normal itself.
:::

## 4. Distribution of Sums via MGFs

Theorem 2.3 + uniqueness (Thm 2.2) let us prove closure under addition.

**Sum of independent Poissons.** $X\sim\text{Pois}(\lambda),\;Y\sim\text{Pois}(\mu)$, independent.
$$
M_{X+Y}(t) = e^{\lambda(e^t-1)}e^{\mu(e^t-1)} = e^{(\lambda+\mu)(e^t-1)}
$$
This is the $\text{Pois}(\lambda+\mu)$ MGF, so $X+Y\sim\text{Pois}(\lambda+\mu)$.

**Sum of independent Normals.** $X_i\sim\mathcal{N}(\mu_i,\sigma_i^2)$, independent.
$$
M_{X_1+X_2}(t) = e^{\mu_1 t+\frac{1}{2}\sigma_1^2 t^2}\cdot e^{\mu_2 t+\frac{1}{2}\sigma_2^2 t^2} = e^{(\mu_1+\mu_2)t+\frac{1}{2}(\sigma_1^2+\sigma_2^2)t^2}
$$
Hence $X_1+X_2\sim\mathcal{N}(\mu_1+\mu_2,\,\sigma_1^2+\sigma_2^2)$.

::: {.callout-note}
## Cramér's theorem
The converse also holds: if $X_1\perp X_2$ and $X_1+X_2$ is Normal, then $X_1$ and $X_2$ are each individually Normal.
:::

## 5. MGF Table

| Distribution | Notation | MGF $M(t)$ | Domain |
|:---|:---|:---|:---|
| Bernoulli | $\text{Bern}(p)$ | $pe^t + q$ | $\mathbb{R}$ |
| Binomial | $\text{Bin}(n,p)$ | $(pe^t + q)^n$ | $\mathbb{R}$ |
| Geometric | $\text{Geom}(p)$ | $\dfrac{p}{1-qe^t}$ | $t < \log(1/q)$ |
| Negative Binomial | $\text{NBin}(r,p)$ | $\left(\dfrac{p}{1-qe^t}\right)^{\!r}$ | $qe^t < 1$ |
| Poisson | $\text{Pois}(\lambda)$ | $e^{\lambda(e^t-1)}$ | $\mathbb{R}$ |
| Uniform | $\text{Unif}(a,b)$ | $\dfrac{e^{tb}-e^{ta}}{t(b-a)}$ | $t\ne0$ |
| Exponential | $\text{Expo}(\lambda)$ | $\dfrac{\lambda}{\lambda - t}$ | $t<\lambda$ |
| Standard Normal | $\mathcal{N}(0,1)$ | $e^{t^2/2}$ | $\mathbb{R}$ |
| Normal | $\mathcal{N}(\mu,\sigma^2)$ | $e^{\mu t + \frac{1}{2}\sigma^2 t^2}$ | $\mathbb{R}$ |
