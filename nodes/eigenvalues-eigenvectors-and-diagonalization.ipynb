# Eigenvalues, Eigenvectors, and Diagonalization

## 1\. Eigenvalues, Eigenvectors, and the Characteristic Equation

Let $V$ be a finite-dimensional [vector space](vector-spaces.ipynb) of [dimension](linear-independence-basis-dimension.ipynb#def-dimension) $n$ over a [field](field.ipynb) $K$. In the study of [linear operators](linear-maps-and-rank-nullity.ipynb#def-linear-map) $f: V \to V$, we seek an intrinsic coordinate system—an ordered [basis](linear-independence-basis-dimension.ipynb#def-basis)—in which the [matrix representation](linear-maps-inverses-and-matrices.ipynb#def-matrix-representation) $[f]_B$ takes the simplest possible form, ideally a diagonal matrix.

Throughout our discussion, we assume that $K$ is an **algebraically closed field** (such as the field of complex numbers $\mathbb{C}$), ensuring that every non-constant polynomial of degree $n$ splits completely into $n$ linear factors in $K[x]$.

::: {#def-eigenvalue-eigenvector}
## Eigenvalues, Eigenvectors, and Eigenspaces

Let $f: V \to V$ be a linear operator on a vector space $V$ over $K$.

1. A scalar $\lambda \in K$ is called an **eigenvalue** (or **characteristic value**) of $f$ if there exists a **non-zero** vector $v \in V \setminus \{0\}$ such that:
   $$f(v) = \lambda v$$
2. Any non-zero vector $v \in V \setminus \{0\}$ satisfying $f(v) = \lambda v$ is called an **eigenvector** (or **characteristic vector**) of $f$ corresponding to the eigenvalue $\lambda$.
3. For each eigenvalue $\lambda$, the **eigenspace** corresponding to $\lambda$, denoted $E_\lambda$, is the subspace of all vectors satisfying $f(v) = \lambda v$:
   $$E_\lambda := \ker(f - \lambda \operatorname{id}_V) = \{v \in V \mid f(v) = \lambda v\}$$
   The non-zero vectors in $E_\lambda$ are precisely the eigenvectors corresponding to $\lambda$, together with the zero vector $0 \in V$.
:::

::: {.callout-warning}
## The Zero Vector is Never an Eigenvector
By definition, an eigenvector must be non-zero ($v \neq 0$), because the trivial equation $f(0) = 0 = \lambda \cdot 0$ holds for *every* scalar $\lambda \in K$. Conversely, the scalar $\lambda = 0$ is a valid eigenvalue if and only if $\ker(f) \neq \{0\}$, which occurs if and only if $f$ is singular (not [invertible](linear-maps-inverses-and-matrices.ipynb#cor-equal-dim-invertibility)).
:::

---

::: {#thm-characteristic-equation}
## The Characteristic Polynomial and Equation

Let $A \in M_{n \times n}(K)$ be the matrix representation of $f: V \to V$ with respect to an arbitrary basis $B$. A scalar $\lambda \in K$ is an eigenvalue of $A$ if and only if:

$$\det(\lambda I_n - A) = 0$$

The polynomial $p_A(\lambda) := \det(\lambda I_n - A)$ is called the **characteristic polynomial** of $A$, and the equation $p_A(\lambda) = 0$ is called the **characteristic equation**.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-characteristic-equation
A scalar $\lambda \in K$ is an eigenvalue of $A$ if and only if there exists $v \in K^n$ with $v \neq 0$ such that $A v = \lambda v$, which can be rearranged as:

$$(\lambda I_n - A) v = 0$$

This homogeneous linear system has a non-trivial solution $v \neq 0$ if and only if the matrix $\lambda I_n - A$ has a non-trivial [kernel](linear-maps-and-rank-nullity.ipynb#def-kernel-image):

$$\ker(\lambda I_n - A) \neq \{0\}$$

By the fundamental [invertibility criterion](determinants-multilinear-forms-and-orientation.ipynb#thm-det-invertibility), a square matrix is singular (not invertible) if and only if its [determinant](determinants-multilinear-forms-and-orientation.ipynb#def-determinant) is zero:

$$\det(\lambda I_n - A) = 0$$

Because the determinant of a linear operator is [independent of the choice of basis](change-of-basis-matrix-inversion-and-trace.ipynb#thm-det-operator-well-defined) ($\det(\lambda I - P^{-1} A P) = \det(P^{-1} (\lambda I - A) P) = \det(\lambda I - A)$), the characteristic polynomial depends solely on the operator $f$, not on the representing matrix $A$. $\blacksquare$
:::



## 2\. Linear Independence of Eigenvectors

A cornerstone theorem of spectral theory guarantees that eigenvectors associated with distinct eigenvalues cannot interfere with one another.

::: {#thm-distinct-eigenvalues-independent}
## Linear Independence of Eigenvectors for Distinct Eigenvalues

Let $f: V \to V$ be a linear operator on a vector space $V$ over $K$. Let $\lambda_1, \dots, \lambda_k \in K$ be **pairwise distinct** eigenvalues of $f$, and let $v_1, \dots, v_k \in V$ be corresponding non-zero eigenvectors ($f(v_i) = \lambda_i v_i$ with $v_i \neq 0$).

Then the set $\{v_1, \dots, v_k\}$ is [linearly independent](linear-independence-basis-dimension.ipynb#def-linear-independence).
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-distinct-eigenvalues-independent
We argue by contradiction using a minimal dependence relation.

Suppose, for the sake of contradiction, that $\{v_1, \dots, v_k\}$ is linearly dependent. Then there exists a non-trivial linear combination equal to zero. Among all such non-trivial relations, choose one of **minimal length** $l$ (after re-indexing if necessary):

$$c_1 v_1 + c_2 v_2 + \dots + c_l v_l = 0 \qquad (\star)$$

where $c_i \in K$ and $c_i \neq 0$ for all $i = 1, \dots, l$.

Because each eigenvector is non-zero ($v_1 \neq 0$), a single vector cannot be linearly dependent: $c_1 v_1 = 0 \implies c_1 = 0$. Hence the minimal length must satisfy $l \ge 2$.

Now apply the linear operator $f$ to both sides of $(\star)$. By linearity and the eigenvector property $f(v_i) = \lambda_i v_i$:

$$c_1 f(v_1) + c_2 f(v_2) + \dots + c_l f(v_l) = f(0) = 0$$

$$c_1 \lambda_1 v_1 + c_2 \lambda_2 v_2 + \dots + c_l \lambda_l v_l = 0 \qquad (\star\star)$$

Next, multiply the original equation $(\star)$ by the scalar $\lambda_l$:

$$c_1 \lambda_l v_1 + c_2 \lambda_l v_2 + \dots + c_l \lambda_l v_l = 0 \qquad (\star\star\star)$$

Subtract equation $(\star\star\star)$ from equation $(\star\star)$:

$$c_1 (\lambda_1 - \lambda_l) v_1 + c_2 (\lambda_2 - \lambda_l) v_2 + \dots + c_{l-1} (\lambda_{l-1} - \lambda_l) v_{l-1} + 0 \cdot v_l = 0$$

Notice that the $v_l$ term has vanished completely. Furthermore, because the eigenvalues are pairwise distinct, $\lambda_i \neq \lambda_l$ for all $i < l$, which means $\lambda_i - \lambda_l \neq 0$. Because $c_i \neq 0$, the coefficient of $v_1$ is:

$$c_1 (\lambda_1 - \lambda_l) \neq 0$$

We have obtained a linear dependence relation among the vectors $\{v_1, \dots, v_{l-1}\}$ of length $l - 1$ with non-zero coefficients. This contradicts the minimality of $l$.

Hence no non-trivial linear combination can equal zero, and the set $\{v_1, \dots, v_k\}$ is linearly independent. $\blacksquare$
:::

::: {#cor-direct-sum-eigenspaces}
## Direct Sum of Eigenspaces

If $\lambda_1, \dots, \lambda_k$ are pairwise distinct eigenvalues of $f: V \to V$, then the sum of their corresponding eigenspaces is a [direct sum](complements-direct-sums-quotient-spaces.ipynb#def-direct-sum):

$$E_{\lambda_1} + E_{\lambda_2} + \dots + E_{\lambda_k} = E_{\lambda_1} \oplus E_{\lambda_2} \oplus \dots \oplus E_{\lambda_k}$$
:::



## 3\. Diagonalizability

The ultimate goal of spectral decomposition is to represent an operator by a diagonal matrix.

::: {#def-diagonalizable-operator}
## Diagonalizable Operator and Matrix

Let $V$ be an $n$-dimensional vector space over $K$.

1. A linear operator $f: V \to V$ is called **diagonalizable** if there exists an ordered basis $B = (v_1, \dots, v_n)$ of $V$ consisting entirely of eigenvectors of $f$.
2. A square matrix $A \in M_{n \times n}(K)$ is called **diagonalizable** if it is [similar](change-of-basis-matrix-inversion-and-trace.ipynb#def-matrix-similarity) to a diagonal matrix; that is, there exists an invertible matrix $P \in GL(n, K)$ such that:
   $$P^{-1} A P = D = \begin{pmatrix} \lambda_1 & 0 & \dots & 0 \\ 0 & \lambda_2 & \dots & 0 \\ \vdots & \vdots & \ddots & \vdots \\ 0 & 0 & \dots & \lambda_n \end{pmatrix}$$
   The columns of $P$ are the eigenvectors of $A$, and the diagonal entries of $D$ are the corresponding eigenvalues.
:::

::: {#thm-sufficient-diagonalizability}
## Sufficient Condition: $n$ Distinct Eigenvalues

Let $f: V \to V$ be a linear operator on an $n$-dimensional vector space $V$. If $f$ possesses $n$ **pairwise distinct** eigenvalues $\lambda_1, \dots, \lambda_n \in K$, then $f$ is **diagonalizable**.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-sufficient-diagonalizability
For each eigenvalue $\lambda_i$, choose a non-zero eigenvector $v_i \in V \setminus \{0\}$.

By @thm-distinct-eigenvalues-independent, the set $B = (v_1, \dots, v_n)$ is linearly independent. Because $\dim V = n$ and $B$ contains $n$ linearly independent vectors, $B$ is a [basis](linear-independence-basis-dimension.ipynb#thm-basis-equivalent) for $V$.

To compute the matrix $[f]_B$, evaluate $f$ on each basis vector:

$$f(v_j) = \lambda_j v_j = 0 v_1 + \dots + \lambda_j v_j + \dots + 0 v_n$$

The $j$-th column of $[f]_B$ has $\lambda_j$ in the $j$-th entry and zeros elsewhere. Thus $[f]_B = \operatorname{diag}(\lambda_1, \dots, \lambda_n)$ is diagonal. $\blacksquare$
:::

---

### 3.1\. Geometric Action of a Diagonalizable Operator

When an operator $f$ is diagonalized with eigenbasis $B = (v_1, \dots, v_n)$, the space decomposes into $n$ invariant 1-dimensional lines (or complex lines) $L_i = \operatorname{span}(v_i)$. On each line $L_i$, $f$ acts as a **pure homothety (scaling)** by the scalar factor $\lambda_i$:

$$\text{For } v = \sum_{i=1}^n c_i v_i \implies f(v) = \sum_{i=1}^n c_i \lambda_i v_i$$

The action on any arbitrary vector is completely decoupled into independent 1-dimensional scalings along the eigenvector directions.

::: {.callout-note}
## Field Dependence: Diagonalizability over $\mathbb{R}$ vs $\mathbb{C}$
Diagonalizability is critically dependent upon the base field $K$. Consider the $90^\circ$ counterclockwise rotation matrix on $\mathbb{R}^2$:

$$R = \begin{pmatrix} 0 & -1 \\ 1 & 0 \end{pmatrix}$$

Its characteristic polynomial is $p_R(\lambda) = \det(\lambda I - R) = \lambda^2 + 1$.

- **Over $\mathbb{R}$:** $\lambda^2 + 1 = 0$ has no real roots. The operator has **no eigenvalues** and **no eigenvectors** in $\mathbb{R}^2$. It is **not diagonalizable** over $\mathbb{R}$ (no real line is mapped into itself).
- **Over $\mathbb{C}$:** $\lambda^2 + 1 = (\lambda - i)(\lambda + i)$ splits completely. It has $2$ distinct eigenvalues $\lambda_1 = i, \lambda_2 = -i$, with eigenvectors $v_1 = \begin{pmatrix} 1 \\ -i \end{pmatrix}$ and $v_2 = \begin{pmatrix} 1 \\ i \end{pmatrix}$. Hence $R$ is **diagonalizable** over $\mathbb{C}$.
:::



## 4\. Trace and Determinant via the Spectrum (Vieta's Relations)

We previously established that the [determinant](change-of-basis-matrix-inversion-and-trace.ipynb#thm-det-operator-well-defined) and [trace](change-of-basis-matrix-inversion-and-trace.ipynb#thm-trace-similarity-invariance) are similarity invariants. We now reveal their deep algebraic nature: they are the extreme coefficients of the characteristic polynomial.

::: {#thm-characteristic-coefficients}
## Spectral Formulas for Trace and Determinant

Let $A \in M_{n \times n}(K)$ have characteristic polynomial $p_A(\lambda) = \det(\lambda I_n - A)$. Over an algebraically closed field, let $\lambda_1, \dots, \lambda_n \in K$ denote the roots of $p_A(\lambda)$ counted with algebraic multiplicity. Then:

$$p_A(\lambda) = \lambda^n - \operatorname{Tr}(A) \lambda^{n-1} + e_2(\lambda_1, \dots, \lambda_n) \lambda^{n-2} - \dots + (-1)^n \det(A)$$

In particular:

1. **The Trace is the sum of the eigenvalues:**
   $$\operatorname{Tr}(A) = \sum_{i=1}^n \lambda_i$$
2. **The Determinant is the product of the eigenvalues:**
   $$\det(A) = \prod_{i=1}^n \lambda_i$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-characteristic-coefficients
Because $K$ is algebraically closed and $p_A(\lambda)$ is a monic polynomial of degree $n$, it factors completely as:

$$p_A(\lambda) = \prod_{i=1}^n (\lambda - \lambda_i) \qquad (1)$$

Expanding the product $(1)$ by standard polynomial multiplication yields:

$$\prod_{i=1}^n (\lambda - \lambda_i) = \lambda^n - \left( \sum_{i=1}^n \lambda_i \right) \lambda^{n-1} + \left( \sum_{1 \le i < j \le n} \lambda_i \lambda_j \right) \lambda^{n-2} - \dots + (-1)^n \left( \prod_{i=1}^n \lambda_i \right) \qquad (2)$$

Now we analyze the coefficients directly from the determinant definition $p_A(\lambda) = \det(\lambda I_n - A)$ using the [Leibniz formula](determinants-multilinear-forms-and-orientation.ipynb#thm-alternating-form-basis-characterization):

$$\det(\lambda I_n - A) = \sum_{\sigma \in S_n} \operatorname{sgn}(\sigma) B_{1, \sigma(1)} B_{2, \sigma(2)} \cdots B_{n, \sigma(n)}$$

where $B = \lambda I_n - A$, with entries $B_{ii} = \lambda - a_{ii}$ and $B_{ij} = -a_{ij}$ for $i \neq j$.

**1. Constant Term ($\lambda = 0$):**

Evaluating $p_A(\lambda)$ at $\lambda = 0$:

$$p_A(0) = \det(-A) = (-1)^n \det(A)$$

Comparing with the constant term of $(2)$, which is $(-1)^n \prod_{i=1}^n \lambda_i$:

$$(-1)^n \det(A) = (-1)^n \prod_{i=1}^n \lambda_i \implies \det(A) = \prod_{i=1}^n \lambda_i$$

**2. Coefficient of $\lambda^{n-1}$:**

In the Leibniz formula, any term with $\sigma \neq \operatorname{id}$ moves at least two indices, meaning at most $n - 2$ factors come from the diagonal. Such terms have degree at most $n - 2$ in $\lambda$.

Therefore, the coefficient of $\lambda^{n-1}$ comes entirely from the diagonal term (corresponding to $\sigma = \operatorname{id}$):

$$\prod_{i=1}^n (\lambda - a_{ii}) = (\lambda - a_{11})(\lambda - a_{22}) \cdots (\lambda - a_{nn})$$

Expanding this product, the coefficient of $\lambda^{n-1}$ is clearly $-\sum_{i=1}^n a_{ii} = -\operatorname{Tr}(A)$. Comparing with the coefficient of $\lambda^{n-1}$ in $(2)$, which is $-\sum_{i=1}^n \lambda_i$:

$$-\operatorname{Tr}(A) = -\sum_{i=1}^n \lambda_i \implies \operatorname{Tr}(A) = \sum_{i=1}^n \lambda_i \qquad \blacksquare$$
:::

::: {#cor-intermediate-invariants}
## Elementary Symmetric Polynomials of the Spectrum

For each $k \in \{1, \dots, n\}$, the $k$-th elementary symmetric polynomial of the eigenvalues:

$$e_k(\lambda_1, \dots, \lambda_n) = \sum_{1 \le i_1 < i_2 < \dots < i_k \le n} \lambda_{i_1} \lambda_{i_2} \cdots \lambda_{i_k}$$

equals the sum of the principal $k \times k$ minors of $A$. In particular, each $e_k$ is an **intrinsic similarity invariant** of the linear operator, with $e_1(A) = \operatorname{Tr}(A)$ and $e_n(A) = \det(A)$.
:::



## 5\. Polynomials of Operators and the Spectral Mapping Property

A fundamental operational principle in linear algebra is that an operator and any polynomial constructed from it share identical eigenvectors.

::: {#thm-spectral-mapping-polynomial}
## Action of Matrix Polynomials on Eigenvectors

Let $A \in M_{n \times n}(K)$ and let $v \in K^n \setminus \{0\}$ be an eigenvector of $A$ corresponding to the eigenvalue $\lambda \in K$ ($A v = \lambda v$).

1. **Powers:** For any integer $k \ge 0$, $v$ is an eigenvector of $A^k$ corresponding to the eigenvalue $\lambda^k$:
   $$A^k v = \lambda^k v$$
2. **Scalar Multiples:** For any scalar $c \in K$, $v$ is an eigenvector of $c A$ with eigenvalue $c \lambda$:
   $$(c A) v = (c \lambda) v$$
3. **General Polynomial:** For any polynomial $p(x) = c_q x^q + c_{q-1} x^{q-1} + \dots + c_1 x + c_0 \in K[x]$, $v$ is an eigenvector of the matrix polynomial $p(A) := \sum_{j=0}^q c_j A^j$ with eigenvalue $p(\lambda)$:
   $$p(A) v = p(\lambda) v$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-spectral-mapping-polynomial
**1. Powers by induction on $k$:**

For $k = 0$, $A^0 v = I_n v = 1 \cdot v = \lambda^0 v$. For $k = 1$, $A v = \lambda v$ by hypothesis.

Assume inductively that $A^k v = \lambda^k v$ for some $k \ge 1$. Then:

$$A^{k+1} v = A(A^k v) = A(\lambda^k v) = \lambda^k A(v) = \lambda^k (\lambda v) = \lambda^{k+1} v$$

By induction, $A^k v = \lambda^k v$ for all $k \ge 0$.

**2. General polynomial:**

By definition of the matrix polynomial $p(A)$ and linearity:

$$\begin{aligned}
p(A) v &= \left( \sum_{j=0}^q c_j A^j \right) v \\
&= \sum_{j=0}^q c_j (A^j v) \\
&= \sum_{j=0}^q c_j (\lambda^j v) \\
&= \left( \sum_{j=0}^q c_j \lambda^j \right) v = p(\lambda) v \qquad \blacksquare
\end{aligned}$$
:::

::: {.callout-warning}
## Caution: Non-Additivity of Eigenvalues for Arbitrary Matrices
The property $p(A) v = p(\lambda) v$ holds strictly because all powers $A^j$ share the **exact same eigenvector** $v$.

For two *different* matrices $A$ and $B$, even if $\lambda$ is an eigenvalue of $A$ and $\mu$ is an eigenvalue of $B$, $\lambda + \mu$ is **not** in general an eigenvalue of $A + B$. This is because $A$ and $B$ generally do not commute ($A B \neq B A$) and do not possess common eigenvectors.
:::



## 6\. Multiplicity and the Roadmap to Jordan Canonical Form

What happens when the roots of the characteristic polynomial are not all distinct? When multiple roots occur, diagonalizability is no longer guaranteed.

::: {#def-multiplicities}
## Algebraic vs Geometric Multiplicity

Let $\lambda_0$ be an eigenvalue of $f: V \to V$.

1. The **algebraic multiplicity** of $\lambda_0$, denoted $m_a(\lambda_0)$, is its multiplicity as a root of the characteristic polynomial $p_f(\lambda)$:
   $$p_f(\lambda) = (\lambda - \lambda_0)^{m_a(\lambda_0)} q(\lambda), \qquad q(\lambda_0) \neq 0$$
2. The **geometric multiplicity** of $\lambda_0$, denoted $m_g(\lambda_0)$, is the dimension of its eigenspace:
   $$m_g(\lambda_0) := \dim E_{\lambda_0} = \dim \ker(f - \lambda_0 \operatorname{id}_V)$$
:::

::: {#prp-multiplicity-inequality}
## Multiplicity Inequality and Diagonalizability Criterion

For every eigenvalue $\lambda_0$ of a linear operator $f$ on a finite-dimensional vector space:

$$1 \le m_g(\lambda_0) \le m_a(\lambda_0)$$

Furthermore, over an algebraically closed field, $f$ is **diagonalizable** if and only if:

$$m_g(\lambda) = m_a(\lambda) \qquad \text{for every eigenvalue } \lambda$$
:::

::: {.callout-note}
## Looking Ahead: Defective Matrices and Jordan Normal Form
When an operator has $m_g(\lambda) < m_a(\lambda)$ for some eigenvalue, it is called **defective** (non-diagonalizable). A canonical prototype is the non-trivial shear:

$$J = \begin{pmatrix} \lambda & 1 \\ 0 & \lambda \end{pmatrix}$$

Here $p_J(t) = (t - \lambda)^2$ has $m_a(\lambda) = 2$, but $\ker(J - \lambda I) = \operatorname{span}(e_1)$ has dimension $m_g(\lambda) = 1$. It cannot be diagonalized under any change of basis.

The complete structural classification of all linear operators—even defective ones—is achieved via the **minimal polynomial** and the **Jordan Canonical Form**, decomposing $V$ into generalized eigenspaces and Jordan blocks.
:::


```python

import numpy as np
import matplotlib.pyplot as plt
from matplotlib.patches import Ellipse

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
fig, (ax1, ax2, ax3) = plt.subplots(1, 3, figsize=(16, 5), dpi=100)

# ── Panel 1: Diagonalizable 2D Operator Action ─────────────────────────────
# Construct a matrix with known eigenpairs: lambda1=2.0, lambda2=0.5
v1 = np.array([2.0, 1.0]) / np.sqrt(5.0)
v2 = np.array([-1.0, 2.0]) / np.sqrt(5.0)
P = np.column_stack([v1, v2])
D = np.diag([2.0, 0.5])
A = P @ D @ np.linalg.inv(P)

# Plot eigen-axes
t_vals = np.linspace(-3, 3, 100)
ax1.plot(t_vals * v1[0], t_vals * v1[1], '--', color='#27ae60', lw=1.5, label='Eigen-axis 1 ($\\lambda_1=2.0$)')
ax1.plot(t_vals * v2[0], t_vals * v2[1], '--', color='#8e44ad', lw=1.5, label='Eigen-axis 2 ($\\lambda_2=0.5$)')

# Plot eigenvectors and transformed eigenvectors
ax1.quiver(0, 0, v1[0], v1[1], angles='xy', scale_units='xy', scale=1, color='#27ae60', width=0.018, label=r'$v_1$')
ax1.quiver(0, 0, 2*v1[0], 2*v1[1], angles='xy', scale_units='xy', scale=1, color='#2ecc71', width=0.012, label=r'$A v_1 = 2 v_1$')
ax1.quiver(0, 0, v2[0], v2[1], angles='xy', scale_units='xy', scale=1, color='#8e44ad', width=0.018, label=r'$v_2$')
ax1.quiver(0, 0, 0.5*v2[0], 0.5*v2[1], angles='xy', scale_units='xy', scale=1, color='#bb8fce', width=0.012, label=r'$A v_2 = 0.5 v_2$')

ax1.set_xlim(-2.5, 2.5)
ax1.set_ylim(-2.5, 2.5)
ax1.set_aspect('equal')
ax1.set_title('Diagonal Action Along Eigen-axes', fontsize=11, fontweight='bold')
ax1.legend(loc='upper left', fontsize=8, framealpha=0.9)

# ── Panel 2: Transformation of Unit Circle ────────────────────────────────
theta = np.linspace(0, 2*np.pi, 200)
circle = np.array([np.cos(theta), np.sin(theta)])
ellipse = A @ circle

ax2.plot(circle[0], circle[1], color='#3498db', lw=2, label='Unit circle $S^1$')
ax2.plot(ellipse[0], ellipse[1], color='#e74c3c', lw=2, label='Image ellipse $A(S^1)$')
ax2.quiver(0, 0, 2*v1[0], 2*v1[1], angles='xy', scale_units='xy', scale=1, color='#27ae60', width=0.015, label='Major semi-axis ($\\lambda_1=2$)')
ax2.quiver(0, 0, 0.5*v2[0], 0.5*v2[1], angles='xy', scale_units='xy', scale=1, color='#8e44ad', width=0.015, label='Minor semi-axis ($\\lambda_2=0.5$)')

ax2.set_xlim(-2.5, 2.5)
ax2.set_ylim(-2.5, 2.5)
ax2.set_aspect('equal')
ax2.set_title('Circle to Ellipse: $\\det(A) = 1.00$', fontsize=11, fontweight='bold')
ax2.legend(loc='upper left', fontsize=8, framealpha=0.9)

# ── Panel 3: Non-diagonalizable over R (Rotation Matrix) ───────────────────
theta_rot = np.pi / 3.0  # 60 degrees
R = np.array([[np.cos(theta_rot), -np.sin(theta_rot)], [np.sin(theta_rot), np.cos(theta_rot)]])
u = np.array([1.5, 0.5])
Ru = R @ u

ax3.plot(circle[0], circle[1], '--', color='#95a5a6', lw=1.5, label='Unit circle')
ax3.quiver(0, 0, u[0], u[1], angles='xy', scale_units='xy', scale=1, color='#2980b9', width=0.018, label=r'Vector $u$')
ax3.quiver(0, 0, Ru[0], Ru[1], angles='xy', scale_units='xy', scale=1, color='#d35400', width=0.018, label='Rotated $R u$ ($60^\\circ$)')

arc_theta = np.linspace(np.arctan2(u[1], u[0]), np.arctan2(Ru[1], Ru[0]), 30)
ax3.plot(0.6 * np.cos(arc_theta), 0.6 * np.sin(arc_theta), color='#e67e22', lw=2)

ax3.set_xlim(-2.5, 2.5)
ax3.set_ylim(-2.5, 2.5)
ax3.set_aspect('equal')
ax3.set_title('Rotation: No Real Eigen-axes ($\\lambda \\in \\mathbb{C}$)', fontsize=11, fontweight='bold')
ax3.legend(loc='upper left', fontsize=8, framealpha=0.9)

plt.tight_layout()
plt.show()
```

```python

import numpy as np

# Fix seed for deterministic reproduction
np.random.seed(101)
n = 4

# Generate a diagonalizable matrix with 4 distinct real eigenvalues
evals_target = np.array([4.0, 2.5, -1.0, 0.5])
while True:
    P_rand = np.random.randn(n, n)
    if abs(np.linalg.det(P_rand)) > 0.5:
        break
P_mat = P_rand
D_mat = np.diag(evals_target)
A = P_mat @ D_mat @ np.linalg.inv(P_mat)

# 1. Compute Eigenvalues and Eigenvectors via NumPy
evals, evecs = np.linalg.eig(A)
# Sort eigenvalues for clean display
idx = np.argsort(evals)[::-1]
evals = evals[idx]
evecs = evecs[:, idx]

print("=== 1. Eigenvalues and Eigenvectors Verification ===")
print(f"Target eigenvalues:     {evals_target}")
print(f"Computed eigenvalues:   {evals.round(4)}")

# Verify A v_i == lambda_i v_i
all_valid = True
for i in range(n):
    v_i = evecs[:, i]
    lam_i = evals[i]
    valid = np.allclose(A @ v_i, lam_i * v_i)
    if not valid:
        all_valid = False
print(f"Does A v_i == lambda_i v_i for all i? {all_valid}")

# 2. Linear Independence of Eigenvectors and Diagonalization
det_evecs = np.linalg.det(evecs)
print("\n=== 2. Linear Independence and Diagonalization ===")
print(f"det(Eigenvector Matrix) = {det_evecs:.4f} != 0 -> Linearly Independent: {abs(det_evecs) > 1e-5}")
D_reconstructed = np.linalg.inv(evecs) @ A @ evecs
is_diagonal = np.allclose(D_reconstructed, np.diag(evals))
print(f"Is P^(-1) A P equal to diag(evals)? {is_diagonal}")

# 3. Trace and Determinant via the Spectrum (Vieta's Formulas)
tr_A = np.trace(A)
sum_evals = np.sum(evals)
det_A = np.linalg.det(A)
prod_evals = np.prod(evals)

print("\n=== 3. Trace and Determinant via Spectrum ===")
print(f"Tr(A) = {tr_A:.4f},  Sum of eigenvalues = {sum_evals:.4f}  -> Matches: {np.isclose(tr_A, sum_evals)}")
print(f"det(A) = {det_A:.4f}, Prod of eigenvalues = {prod_evals:.4f} -> Matches: {np.isclose(det_A, prod_evals)}")

# 4. Action of Polynomials on Eigenvectors: p(A) v = p(lambda) v
# Let p(x) = 2 x^3 - 3 x^2 + 4 x - 5
p_A = 2 * np.linalg.matrix_power(A, 3) - 3 * np.linalg.matrix_power(A, 2) + 4 * A - 5 * np.eye(n)
poly_action_valid = True
for i in range(n):
    v_i = evecs[:, i]
    lam_i = evals[i]
    p_lam = 2 * (lam_i**3) - 3 * (lam_i**2) + 4 * lam_i - 5
    if not np.allclose(p_A @ v_i, p_lam * v_i):
        poly_action_valid = False

print("\n=== 4. Spectral Mapping: p(A) v_i == p(lambda_i) v_i ===")
print(f"Holds for p(x) = 2x^3 - 3x^2 + 4x - 5? {poly_action_valid}")

# 5. Non-Additivity of Eigenvalues for Independent Non-Commuting Matrices
B = np.random.randn(n, n)
evals_B = np.linalg.eigvals(B)
evals_A_plus_B = np.linalg.eigvals(A + B)
print("\n=== 5. Non-Additivity for Non-Commuting Matrices ===")
print(f"Does AB == BA? {np.allclose(A @ B, B @ A)}")
print(f"Sum of sets of eigenvalues == Eigenvalues(A + B)? {np.allclose(np.sort(evals + evals_B), np.sort(evals_A_plus_B))}")
```
