# Complex Numbers

**Prerequisites:** [Field](field.ipynb) · [Real Numbers](real-numbers.ipynb) · [Existence of $n^{\text{th}}$ Roots](existence-of-nth-roots-in-R.ipynb)

## 1\. The Field $\mathbb{C}$

::: {#def-complex-numbers}
## Complex Numbers

$\mathbb{C} = \mathbb{R}^2$ with
$$(x, y) + (x', y') = (x + x',\ y + y'), \qquad (x, y)(x', y') = (xx' - yy',\ xy' + x'y).$$
Identify $x \in \mathbb{R}$ with $(x, 0)$ and put $i = (0, 1)$. Then $(x, y) = x + iy$, $i^2 = -1$, and the operations are the usual algebra of expressions $x + iy$ with $i^2 = -1$. For $z = x + iy$: $\operatorname{Re} z = x$, $\operatorname{Im} z = y$, the **conjugate** $\bar z = x - iy$, the **modulus** $|z| = \sqrt{x^2 + y^2}$ (distance from $0$).
:::

::: {#thm-complex-field}
## $\mathbb{C}$ Is a Field

$(\mathbb{C}, +, \cdot)$ is a [field](field.ipynb) containing $\mathbb{R}$ as a subfield; the inverse of $z \ne 0$ is $z^{-1} = \bar z / |z|^2$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-complex-field
Addition is that of $\mathbb{R}^2$. Commutativity of multiplication is visible from the formula. Associativity and distributivity are polynomial identities in the real coordinates, checked by expanding both sides; for instance $\big((x,y)(x',y')\big)(x'',y'')$ and $(x,y)\big((x',y')(x'',y'')\big)$ both have real part $xx'x'' - xy'y'' - yx'y'' - yy'x''$. The unit is $1 = (1,0)$. For $z = x + iy \ne 0$, $z \bar z = x^2 + y^2 = |z|^2 > 0$, so $z \cdot \bar z/|z|^2 = 1$. On pairs $(x, 0)$ both operations reduce to those of $\mathbb{R}$. $\blacksquare$
:::

::: {#prp-conjugate-modulus}
## Conjugate and Modulus

For $z, w \in \mathbb{C}$:

1. $\overline{z + w} = \bar z + \bar w$, $\ \overline{zw} = \bar z\, \bar w$, $\ z\bar z = |z|^2$, $\ \operatorname{Re} z = \frac{z + \bar z}{2}$, $\ \operatorname{Im} z = \frac{z - \bar z}{2i}$;
2. $|zw| = |z|\,|w|$ and $|\operatorname{Re} z|, |\operatorname{Im} z| \le |z| \le |\operatorname{Re} z| + |\operatorname{Im} z|$;
3. **triangle inequality** $|z + w| \le |z| + |w|$, and $\big||z| - |w|\big| \le |z - w|$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-conjugate-modulus
(1) is direct computation. (2): $|zw|^2 = zw\overline{zw} = z\bar z\, w \bar w = |z|^2|w|^2$; the inequalities follow from $x^2 \le x^2 + y^2 \le (|x| + |y|)^2$. (3):
$$|z + w|^2 = (z+w)(\bar z + \bar w) = |z|^2 + |w|^2 + 2\operatorname{Re}(z\bar w) \le |z|^2 + |w|^2 + 2|z||w| = (|z| + |w|)^2,$$
using (2). The reverse inequality follows as for [real numbers](triangle-inequality.ipynb): $|z| = |(z - w) + w| \le |z - w| + |w|$, and symmetrically. $\blacksquare$
:::

## 2\. Polar Form and De Moivre's Formula

Write $\operatorname{cis}\theta = \cos\theta + i\sin\theta$ (a point of the unit circle). Every $z \ne 0$ can be written $z = r\operatorname{cis}\theta$ with $r = |z| > 0$; the angle $\theta = \arg z$ is determined up to a multiple of $2\pi$.

::: {#thm-polar-product}
## Multiplication in Polar Form

$\operatorname{cis}\theta\, \operatorname{cis}\varphi = \operatorname{cis}(\theta + \varphi)$. Hence
$$(r\operatorname{cis}\theta)(s\operatorname{cis}\varphi) = rs\operatorname{cis}(\theta + \varphi), \qquad (r\operatorname{cis}\theta)^{-1} = r^{-1}\operatorname{cis}(-\theta):$$
moduli multiply and angles add. Multiplication by $w = s\operatorname{cis}\varphi$ is the rotation by $\varphi$ about $0$ followed by scaling by $s$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-polar-product
$\operatorname{cis}\theta\operatorname{cis}\varphi = (\cos\theta\cos\varphi - \sin\theta\sin\varphi) + i(\sin\theta\cos\varphi + \cos\theta\sin\varphi) = \cos(\theta+\varphi) + i \sin(\theta + \varphi)$ by the addition formulas. The inverse formula follows from $\operatorname{cis}\theta \operatorname{cis}(-\theta) = \operatorname{cis} 0 = 1$. $\blacksquare$
:::

::: {#thm-de-moivre}
## De Moivre's Formula

$(\operatorname{cis}\theta)^n = \operatorname{cis}(n\theta)$ for every $n \in \mathbb{Z}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-de-moivre
For $n \ge 0$ by induction using @thm-polar-product ($z^0 = 1 = \operatorname{cis} 0$). For $n < 0$: $(\operatorname{cis}\theta)^n = \big((\operatorname{cis}\theta)^{-1}\big)^{-n} = \operatorname{cis}(-\theta)^{-n} = \operatorname{cis}(n\theta)$. $\blacksquare$
:::

::: {#exm-triple-angle}
## Multiple-Angle Formulas

Expanding $(\cos\theta + i\sin\theta)^3 = \cos 3\theta + i \sin 3\theta$ and comparing real and imaginary parts:
$$\cos 3\theta = \cos^3\theta - 3\cos\theta\sin^2\theta, \qquad \sin 3\theta = 3\cos^2\theta\sin\theta - \sin^3\theta .$$
:::

::: {#thm-nth-roots}
## $n$-th Roots

Let $w = R\operatorname{cis}\alpha \ne 0$ and $n \in \mathbb{N}$. The equation $z^n = w$ has exactly $n$ solutions
$$z_k = R^{1/n}\operatorname{cis}\Big(\frac{\alpha + 2\pi k}{n}\Big), \qquad k = 0, 1, \dots, n-1 .$$
They are the vertices of a regular $n$-gon centred at $0$. For $w = 1$ they are the **$n$-th roots of unity** $\omega^k$, $\omega = \operatorname{cis}(2\pi/n)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-nth-roots
Write $z = r\operatorname{cis}\theta$. By @thm-de-moivre, $z^n = r^n\operatorname{cis}(n\theta)$, which equals $R\operatorname{cis}\alpha$ if and only if $r^n = R$ and $n\theta = \alpha + 2\pi k$ for some $k \in \mathbb{Z}$, i.e. $r = R^{1/n}$ and $\theta = (\alpha + 2\pi k)/n$. Two values of $k$ give the same $z$ exactly when their angles differ by a multiple of $2\pi$, i.e. when $k \equiv k' \pmod n$. So $k = 0, \dots, n-1$ give all solutions, each once. $\blacksquare$
:::

For example the cube roots of unity are $1$ and $-\frac12 \pm \frac{\sqrt3}{2} i$, and the fourth roots are $\pm 1, \pm i$. If $n \ge 2$, then $1 + \omega + \dots + \omega^{n-1} = \frac{1 - \omega^n}{1 - \omega} = 0$.

## 3\. Application: Cubic Equations

Substituting $x = t - a/3$ turns $x^3 + ax^2 + bx + c = 0$ into a **depressed cubic** $t^3 + pt + q = 0$. Trying $t = u + v$:
$$t^3 + pt + q = u^3 + v^3 + (3uv + p)(u + v) + q,$$
so it suffices that $3uv = -p$ and $u^3 + v^3 = -q$. Then $u^3, v^3$ have sum $-q$ and product $-p^3/27$: they are the roots of the **resolvent** $Z^2 + qZ - \frac{p^3}{27} = 0$.

::: {#thm-cubic-formula}
## Solution of the Cubic

Let $p \ne 0$, let $Z$ be a root of $Z^2 + qZ - p^3/27 = 0$ (it is nonzero), let $u_1$ be any cube root of $Z$, $v_1 = -p/(3u_1)$, and $\omega = \operatorname{cis}(2\pi/3)$. Then
$$t^3 + pt + q = (t - t_0)(t - t_1)(t - t_2), \qquad t_k = \omega^k u_1 + \omega^{-k} v_1 .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-cubic-formula
$Z \ne 0$ because $p \ne 0$. With $u = u_1$, $v = v_1$ we have $3uv = -p$, and $v^3 = -p^3/(27Z)$ is the other root of the resolvent (the product of its roots is $-p^3/27$), so $u^3 + v^3 = -q$. Let $r_k = \omega^k u + \omega^{-k} v$, so $r_0 = u + v$, $r_1 = \omega u + \omega^2 v$, $r_2 = \omega^2 u + \omega v$. Using $\omega^3 = 1$ and $\omega + \omega^2 = -1$:
$$r_0 + r_1 + r_2 = (1 + \omega + \omega^2)(u + v) = 0, \qquad r_1 r_2 = u^2 - uv + v^2, \qquad r_0(r_1 + r_2) = -(u+v)^2,$$
so $r_0r_1 + r_0r_2 + r_1r_2 = -(u+v)^2 + u^2 - uv + v^2 = -3uv$ and $r_0r_1r_2 = (u+v)(u^2 - uv + v^2) = u^3 + v^3$. Therefore
$$(t - r_0)(t - r_1)(t - r_2) = t^3 - 3uv\, t - (u^3 + v^3) = t^3 + pt + q. \ \blacksquare$$
:::

::: {#exm-cubic}
## Real Roots Through Complex Numbers

$t^3 - 6t + 4 = 0$: the resolvent $Z^2 + 4Z + 8 = 0$ has $Z = -2 + 2i$. Since $(1 + i)^3 = -2 + 2i$, take $u_1 = 1 + i$, $v_1 = 2/(1+i) = 1 - i = \bar u_1$. Then $t_k = \omega^k u_1 + \overline{\omega^k u_1} = 2\operatorname{Re}(\omega^k u_1)$:
$$t_0 = 2, \qquad t_1 = -1 - \sqrt3, \qquad t_2 = -1 + \sqrt3 .$$
All three roots are real, yet the formula passes through non-real numbers: the resolvent has non-real roots, so $u_1, v_1$ cannot be real. For $t^3 - 3t + 2$ the resolvent is $(Z+1)^2$, $u_1 = v_1 = -1$, giving the roots $-2, 1, 1$.
:::

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

def arrow(ax, z, color, label):
    ax.annotate("", xy=(z.real, z.imag), xytext=(0, 0),
                arrowprops=dict(arrowstyle="-|>", color=color, lw=2))
    ax.text(z.real * 1.08, z.imag * 1.08, label, color=color, fontsize=11)

fig, (ax, ax2) = plt.subplots(1, 2, figsize=(11, 4.8))
z = 1.2 * np.exp(1j * np.deg2rad(25)); w = 1.4 * np.exp(1j * np.deg2rad(55))
arrow(ax, z, BLUE, r"$z$"); arrow(ax, w, ORANGE, r"$w$"); arrow(ax, z * w, VIOLET, r"$zw$")
for r, a, c in [(0.35, 25, BLUE), (0.5, 55, ORANGE), (0.65, 80, VIOLET)]:
    th = np.linspace(0, np.deg2rad(a), 50); ax.plot(r * np.cos(th), r * np.sin(th), color=c, lw=1)
ax.set_aspect("equal"); ax.set_xlim(-0.6, 2.0); ax.set_ylim(-0.3, 2.0)
ax.set_title(r"$|zw| = |z||w|$, $\arg zw = \arg z + \arg w$")

t = np.linspace(0, 2 * np.pi, 400)
w0 = -8 + 8 * np.sqrt(3) * 1j                     # = 16 cis(2pi/3)
roots = [16 ** 0.25 * np.exp(1j * (2 * np.pi / 3 + 2 * np.pi * k) / 4) for k in range(4)]
ax2.plot(2 * np.cos(t), 2 * np.sin(t), color=GRID, lw=1)
sq = np.array(roots + roots[:1])
ax2.plot(sq.real, sq.imag, color=VIOLET, lw=1.2, alpha=0.7)
for k, r in enumerate(roots):
    ax2.plot(r.real, r.imag, "o", color=YELLOW, ms=8, mec=INK)
    ax2.text(r.real * 1.15, r.imag * 1.15, f"$z_{k}$", fontsize=11, ha="center", va="center")
check = [bool(np.allclose(r ** 4, w0)) for r in roots]
ax2.set_aspect("equal"); ax2.set_xlim(-2.8, 2.8); ax2.set_ylim(-2.6, 2.6)
ax2.set_title(r"the four roots of $z^4 = -8 + 8\sqrt{3}\,i$")
plt.tight_layout(); plt.show()
print("each z_k^4 equals -8+8*sqrt(3) i:", check)
```
