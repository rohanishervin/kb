# Absolute Extrema in Several Variables

**Prerequisites:** [Critical Points and the Second Derivative Test](critical-points-and-hessian.ipynb) · [Constrained Optimization and KKT Conditions](lagrangian-and-kkt-conditions.ipynb#thm-lagrange-foc-multiple) · [Continuity and Compactness](continuity-and-compactness.ipynb) · [Compactness](compactness.ipynb)

## 1\. The Method

Let $f : S \to \mathbb{R}$, $S \subseteq \mathbb{R}^n$. To find the **absolute** (global) maximum and minimum of $f$ on $S$: if an extremum is attained at an interior point where $f$ is differentiable, that point is [critical](critical-points-and-hessian.ipynb#thm-fermat-several). So the candidates are

1. the critical points of $f$ in the interior of $S$,
2. the interior points where $f$ is not differentiable (**singular points**),
3. the boundary points of $S$ that lie in $S$.

If $S$ is closed and bounded and $f$ is continuous, $f$ attains a maximum and a minimum ([Heine–Borel](compactness.ipynb) and the [extreme value theorem](continuity-and-compactness.ipynb)), and comparing the values of $f$ at the candidates gives them. If $S$ is unbounded or not closed, one must also examine $f$ near the missing boundary points and as $|x| \to \infty$; extrema may fail to exist. On a boundary that is a level set $\{g = c\}$ with $\nabla g \ne 0$, the candidates are found by the [Lagrange condition](lagrangian-and-kkt-conditions.ipynb#thm-lagrange-foc-multiple) $\nabla f = \mu\nabla g$; points where $\nabla g = 0$ must be checked separately (see the warning in that note).

## 2\. Examples

::: {#exm-disk-extrema}
## A Closed Disk

$f = x^2 + y^2 + x$ on $x^2 + y^2 \le 1$. Interior: $\nabla f = (2x + 1, 2y) = 0$ only at $(-\frac12, 0)$, $f = -\frac14$. Boundary: on $x^2 + y^2 = 1$, $f = 1 + x$ with $-1 \le x \le 1$, ranging from $0$ at $(-1, 0)$ to $2$ at $(1, 0)$. Hence $\max f = 2$ at $(1, 0)$ and $\min f = -\frac14$ at $(-\frac12, 0)$.
:::

::: {#exm-no-maximum}
## A Supremum That Is Not Attained

$f = \dfrac{|xy|}{x^2 + y^2 + 1}$ on $\mathbb{R}^2$. $f \ge 0$ with $f = 0$ exactly on the axes, so the minimum $0$ is attained there (where $f$ is not differentiable). Off the axes, by symmetry consider $x, y > 0$:
$$\nabla f = \Big(\frac{y(y^2 + 1 - x^2)}{(x^2 + y^2 + 1)^2},\ \frac{x(x^2 + 1 - y^2)}{(x^2 + y^2 + 1)^2}\Big),$$
which vanishes only if $x^2 = y^2 + 1$ and $y^2 = x^2 + 1$, impossible. So there are no critical points and no maximum. Indeed $|xy| \le \frac12(x^2 + y^2) < \frac12(x^2 + y^2 + 1)$ gives $f < \frac12$, while on $y = x$, $f = \frac{x^2}{2x^2 + 1} \to \frac12$: $\sup f = \frac12$ is not attained.
:::

::: {#exm-ellipsoid-nearest}
## Nearest and Farthest Points of an Ellipsoid

Find the points of $g = \frac{x^2}{4} + y^2 + \frac{z^2}{2} = 1$ nearest to and farthest from $(-1, 0, 0)$. Minimize and maximize $f = (x + 1)^2 + y^2 + z^2$ on this compact surface, where $\nabla g = (\frac x2, 2y, z) \ne 0$. The condition $\nabla f = \mu\nabla g$ reads
$$2(x + 1) = \tfrac\mu2x, \qquad 2y = 2\mu y, \qquad 2z = \mu z .$$
If $y = 0$: either $z = 0$, giving $(\pm2, 0, 0)$, or $\mu = 2$, giving $x = -2$ and again $(-2, 0, 0)$. If $y \ne 0$: $\mu = 1$, then $z = 0$, $x = -\frac43$, $y = \pm\frac{\sqrt5}3$. Comparing $f(2, 0, 0) = 9$, $f(-2, 0, 0) = 1$, $f(-\frac43, \pm\frac{\sqrt5}{3}, 0) = \frac23$: the farthest point is $(2, 0, 0)$ (distance $3$), the nearest are $(-\frac43, \pm\frac{\sqrt5}3, 0)$ (distance $\sqrt{2/3}$).
:::

::: {#exm-production}
## Optimal Allocation of a Budget

Maximize the production $f = Kx^\alpha y^\beta z^\gamma$ ($K, \alpha, \beta, \gamma > 0$) subject to the budget $Ax + By + Cz = D$, $x, y, z \ge 0$ ($A, B, C, D > 0$). The feasible set is a closed triangle, $f$ is continuous, and $f = 0$ on its edges, so the maximum is at an interior point of the triangle. There $\nabla f = \lambda(A, B, C)$:
$$\alpha f = \lambda Ax, \qquad \beta f = \lambda By, \qquad \gamma f = \lambda Cz$$
(multiply the equations by $x, y, z$). Hence $\frac{Ax}{\alpha} = \frac{By}{\beta} = \frac{Cz}{\gamma}$, and with the budget, $Ax = \frac{\alpha}{\alpha + \beta + \gamma}D$, etc.: each factor receives the share of the budget equal to its exponent's share.
:::

::: {#exm-cylinder-plane}
## Two Constraints

Find the highest and lowest points of the curve $x^2 + y^2 = 1$, $x + 2y + 3z = 1$. Here $f = z$, $\nabla g_1 = (2x, 2y, 0)$, $\nabla g_2 = (1, 2, 3)$; these are independent on the curve (since $(x, y) \ne 0$ there). From $(0, 0, 1) = \lambda_1(2x, 2y, 0) + \lambda_2(1, 2, 3)$: $\lambda_2 = \frac13$, $2\lambda_1x = -\frac13$, $2\lambda_1y = -\frac23$, so $y = 2x$ and $x = \pm\frac1{\sqrt5}$. The curve is compact, so the maximum $z = \frac{1 + \sqrt5}{3}$ is at $\big(-\frac1{\sqrt5}, -\frac2{\sqrt5}, \frac{1 + \sqrt5}3\big)$ and the minimum $z = \frac{1 - \sqrt5}{3}$ at $\big(\frac1{\sqrt5}, \frac2{\sqrt5}, \frac{1 - \sqrt5}3\big)$.
:::

::: {#exm-shortest-segment}
## The Shortest Segment Between Two Surfaces

Let $g_1(P) = 0$ and $g_2(Q) = 0$ be disjoint surfaces in $\mathbb{R}^3$ with $\nabla g_1, \nabla g_2 \ne 0$ on them, and suppose $PQ$ is a shortest segment from one to the other. Minimize $f(P, Q) = |P - Q|^2$ on $\mathbb{R}^6$ subject to $G_1(P, Q) = g_1(P) = 0$, $G_2(P, Q) = g_2(Q) = 0$. The gradients $\nabla G_1 = (\nabla g_1(P), 0)$ and $\nabla G_2 = (0, \nabla g_2(Q))$ are independent, and $\nabla f = \lambda_1\nabla G_1 + \lambda_2\nabla G_2$ gives
$$2(P - Q) = \lambda_1\nabla g_1(P), \qquad -2(P - Q) = \lambda_2\nabla g_2(Q):$$
the segment $PQ$ is normal to both surfaces.
:::

```python

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

x, y, z, m = sp.symbols("x y z mu", real=True)
f = (x + 1)**2 + y**2 + z**2
g = x**2 / 4 + y**2 + z**2 / 2 - 1
sols = sp.solve([sp.diff(f, v) - m * sp.diff(g, v) for v in (x, y, z)] + [g], [x, y, z, m], dict=True)
print("ellipsoid candidates:", sorted({(s[x], s[y], s[z], f.subs(s)) for s in sols}, key=lambda t: t[3]))

fig, (ax, ax2) = plt.subplots(1, 2, figsize=(12, 4.8))
t = np.linspace(-1.2, 1.2, 400); X, Y = np.meshgrid(t, t)
F = X**2 + Y**2 + X
F[X**2 + Y**2 > 1] = np.nan
cs = ax.contourf(X, Y, F, levels=20, cmap="viridis"); fig.colorbar(cs, ax=ax, shrink=0.8)
ax.plot(np.cos(np.linspace(0, 2 * np.pi, 200)), np.sin(np.linspace(0, 2 * np.pi, 200)), color=INK, lw=1)
ax.plot([-0.5, 1], [0, 0], "o", color=ORANGE, ms=6)
ax.set_aspect("equal"); ax.set_title(r"$x^2 + y^2 + x$ on the unit disk: min at $(-\frac{1}{2}, 0)$, max at $(1, 0)$")

s = np.linspace(0, 6, 300)
ax2.plot(s, s**2 / (2 * s**2 + 1), color=BLUE, lw=2, label=r"$f$ on $y = x$")
ax2.axhline(0.5, color=ORANGE, ls="--", label=r"$\sup f = 1/2$")
ax2.set_xlabel("$x$"); ax2.legend(fontsize=9); ax2.set_title(r"$|xy|/(x^2 + y^2 + 1)$ has no maximum")
plt.tight_layout(); plt.show()
```
