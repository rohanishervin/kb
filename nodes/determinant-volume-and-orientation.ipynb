# Determinants, Volume, and Orientation

**Prerequisites:** [The Determinant](determinant.ipynb)

Over $\mathbb{R}$ the [determinant](determinant.ipynb#def-determinant) has a picture: $|\det A|$ is the volume of the box spanned by the columns of $A$, and the sign of $\det A$ says whether the columns are arranged like the standard basis or like its mirror image. For an operator $f$, $|\det f|$ is the factor by which $f$ scales every volume, and the sign says whether $f$ preserves or reverses orientation.

Throughout, $V$ is a [vector space](vector-spaces.ipynb) of dimension $n$ over $\mathbb{R}$.

## 1\. Volume of a Parallelotope

::: {#def-parallelotope}
## Parallelotope and Its Volume

The **parallelotope** spanned by $u_1, \dots, u_n \in \mathbb{R}^n$ is
$$P(u_1, \dots, u_n) = \Big\{ \sum_{i=1}^n t_i u_i \;\Big|\; 0 \le t_i \le 1 \Big\},$$
a parallelogram for $n = 2$ and a parallelepiped for $n = 3$. Its **volume** is defined as
$$\operatorname{Vol} P(u_1, \dots, u_n) = |\det(u_1, \dots, u_n)|.$$
:::

::: {.callout-note}
## Why $|\det|$ Is the Right Notion of Volume
Any reasonable notion of $n$-dimensional volume of boxes satisfies three rules:

- **Unit cube:** the box spanned by $e_1, \dots, e_n$ has volume $1$.
- **Scaling:** stretching one edge by a factor $c$ multiplies the volume by $|c|$.
- **Shearing:** replacing an edge $u_i$ by $u_i + c\,u_j$ slides the box parallel to a face and does not change the volume (Cavalieri's principle).

$|\det|$ satisfies all three ($\det I_n = 1$, linearity in one column, and [invariance under adding a multiple of a column](determinant.ipynb#prp-det-column-operations)), and with these three rules the volume of every box can be computed by reducing it to a cube with column operations. The **signed** volume $\det(u_1, \dots, u_n)$ adds an orientation, studied next.
:::

::: {#exm-area-volume}
## Area and Volume

The parallelogram spanned by $(3, 1)$ and $(1, 2)$ has area $|3 \cdot 2 - 1 \cdot 1| = 5$. The parallelepiped spanned by $(1, 0, 0)$, $(1, 2, 0)$, $(1, 1, 3)$ has volume $|\det| = 1 \cdot 2 \cdot 3 = 6$, since the matrix with these columns is upper triangular.
:::

## 2\. Orientation

::: {#def-orientation-equivalence}
## Equally Oriented Bases

Let $\mathcal{B} = (v_1, \dots, v_n)$ and $\mathcal{B}' = (v_1', \dots, v_n')$ be ordered bases of $V$ and $M = [\mathrm{id}_V]_{\mathcal{B}'}^{\mathcal{B}}$ the [transition matrix](change-of-basis-and-similarity.ipynb#def-transition-matrix), so $v_j' = \sum_i m_{ij} v_i$. Since $M$ is invertible, $\det M \ne 0$. The bases are **equally oriented**, $\mathcal{B} \sim \mathcal{B}'$, if $\det M > 0$.
:::

::: {#prp-orientation-equivalence-relation}
## There Are Exactly Two Orientations

$\sim$ is an equivalence relation on the ordered bases of $V$ with exactly two classes. Each class is called an **orientation** of $V$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-orientation-equivalence-relation
**Reflexive:** the transition matrix from $\mathcal{B}$ to itself is $I_n$, with determinant $1$. **Symmetric:** the transition matrix from $\mathcal{B}$ to $\mathcal{B}'$ is $M^{-1}$ ([transition matrices](change-of-basis-and-similarity.ipynb#thm-transition-invertibility)), and $\det M^{-1} = 1/\det M > 0$. **Transitive:** if $M_1$ takes $\mathcal{B}$ to $\mathcal{B}'$ and $M_2$ takes $\mathcal{B}'$ to $\mathcal{B}''$, i.e. $v_j' = \sum_i (M_1)_{ij} v_i$ and $v_k'' = \sum_j (M_2)_{jk} v_j'$, then $v_k'' = \sum_i (M_1 M_2)_{ik} v_i$, and $\det(M_1 M_2) = \det M_1 \det M_2 > 0$.

**Two classes.** Fix $\mathcal{B}$. Every basis has a transition matrix with positive or negative determinant, so it is equivalent to $\mathcal{B}$ or not. Swapping $v_1$ and $v_2$ in $\mathcal{B}$ gives a basis with transition matrix of determinant $-1$, so both classes are non-empty. If $\mathcal{B}', \mathcal{B}''$ are both not equivalent to $\mathcal{B}$, their transition matrices $M', M''$ (from $\mathcal{B}$) have negative determinant, and the transition matrix between them, $M'^{-1} M''$, has positive determinant. $\blacksquare$
:::

::: {.callout-note}
## Intuition: Orientation Cannot Be Reversed Continuously
In $\mathbb{R}^2$ the two orientations are "counter-clockwise" ($e_1$ then $e_2$) and "clockwise"; in $\mathbb{R}^3$ they are the right-hand and left-hand rules. Two bases have the same orientation exactly when one can be moved continuously into the other through bases. Along such a path $\det$ is continuous and never $0$, so it cannot change sign. Over $\mathbb{C}$ there is no orientation: one can go from $\det = 1$ to $\det = -1$ through non-zero complex numbers, around the origin.
:::

## 3\. The Geometric Meaning of $\det f$

For an operator $f : V \to V$, [$\det f$](determinant.ipynb#def-det-operator) does not depend on a basis. Over $\mathbb{R}$ it carries two pieces of geometric information:

1. **Invertibility.** $\det f \ne 0$ iff $f$ is invertible, iff $f$ maps bases to bases. In particular $u_1, \dots, u_n$ are independent iff $\det([u_1]_{\mathcal{B}}, \dots, [u_n]_{\mathcal{B}}) \ne 0$.
2. **Orientation.** If $\det f > 0$, then $f(\mathcal{B}) = (f(v_1), \dots, f(v_n))$ has the same orientation as $\mathcal{B}$ for every basis $\mathcal{B}$ (because $[f]_{\mathcal{B}}$ is the transition matrix from $f(\mathcal{B})$ to $\mathcal{B}$); if $\det f < 0$, $f$ reverses orientation.

::: {#thm-volume-scaling}
## Linear Maps Scale Volume by $|\det f|$

Let $f : \mathbb{R}^n \to \mathbb{R}^n$ be linear and $P = P(u_1, \dots, u_n)$ a parallelotope. Then $f(P) = P(f(u_1), \dots, f(u_n))$ and
$$\operatorname{Vol} f(P) = |\det f| \cdot \operatorname{Vol} P.$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-volume-scaling
By linearity $f(\sum t_i u_i) = \sum t_i f(u_i)$, so $f(P) = P(f(u_1), \dots, f(u_n))$. Let $U$ be the matrix with columns $u_i$ and $A$ the standard matrix of $f$. The matrix with columns $f(u_i) = A u_i$ is $AU$, so by [multiplicativity](determinant.ipynb#thm-det-multiplicative)
$$\operatorname{Vol} f(P) = |\det(AU)| = |\det A|\,|\det U| = |\det f| \operatorname{Vol} P. \qquad \blacksquare$$
:::

The animation shows both effects. First the map $\begin{pmatrix} 2 & 1 \\ 0.5 & 1.5 \end{pmatrix}$ turns the unit square into a parallelogram of area $2.5 = \det$. Then a second map is applied gradually. The determinant of the intermediate maps (shown at the top) decreases continuously, passes through $0$, where the square is flattened onto a line, and becomes negative: the square comes back **flipped**, with $f(e_1)$ and $f(e_2)$ in the opposite rotational order.


```python

from la_manim import *

class DeterminantAsArea(GridScene):
    def construct(self):
        def square():
            return Polygon(ORIGIN, RIGHT, RIGHT + UP, UP, stroke_width=2, stroke_color=C_YELLOW,
                           fill_color=C_YELLOW, fill_opacity=0.35)
        sq = self.track(square())
        a1 = self.vec([1, 0], C_BLUE, r"f(e_1)", DR)
        a2 = self.vec([0, 1], C_ORANGE, r"f(e_2)", UL)
        e1, e2 = a1[0], a2[0]
        det_of = lambda: float(e1.get_end()[0] * e2.get_end()[1] - e1.get_end()[1] * e2.get_end()[0])
        readout = VGroup(MathTex(r"\det f =", color=C_INK), DecimalNumber(1.0, num_decimal_places=2, color=C_YELLOW)
                         ).arrange(RIGHT, buff=0.15).scale(0.8).to_corner(UL, buff=0.3)
        readout[1].add_updater(lambda m: m.set_value(det_of()))
        box = BackgroundRectangle(readout, fill_opacity=0.85, buff=0.15, color=config.background_color)
        self.add(sq, box, readout)

        def paint(m):
            m.set_fill(C_YELLOW if det_of() >= 0 else C_RED, opacity=0.35).set_stroke(C_YELLOW if det_of() >= 0 else C_RED)
        sq.add_updater(paint)
        self.wait(1)
        self.transform([[2, 1], [0.5, 1.5]], run_time=3)
        msg = text_caption(r"area of the image of the unit square $= \det f = 2.5$")
        self.play(FadeIn(msg)); self.wait(1.5); self.play(FadeOut(msg))

        # back to the identity, then a map with negative determinant
        self.play(FadeOut(sq), FadeOut(a1), FadeOut(a2), run_time=0.5)
        self.vectors, self.tracked = [], []
        self.reset_grid()
        sq = self.track(square()); sq.add_updater(paint)
        a1 = self.vec([1, 0], C_BLUE, r"f(e_1)", DR)
        a2 = self.vec([0, 1], C_ORANGE, r"f(e_2)", UL)
        e1, e2 = a1[0], a2[0]
        self.add(sq)
        self.wait(0.5)
        self.transform([[1, 0.5], [0.5, -1]], run_time=5)
        msg = text_caption(r"$\det f$ passed through $0$: flattened, then flipped")
        self.play(FadeIn(msg)); self.wait(2.5)

show(DeterminantAsArea)
```


::: {.callout-note}
## The Jacobian in Multivariable Calculus
A smooth map $\Phi : \mathbb{R}^n \to \mathbb{R}^n$ looks, near a point $x$, like its derivative, the linear map $D\Phi(x)$. A tiny box at $x$ is therefore mapped to approximately a parallelotope whose volume is $|\det D\Phi(x)|$ times the original. Summing over tiny boxes gives the change-of-variables formula
$$\int_{\Phi(\Omega)} g(y)\, dy = \int_\Omega g(\Phi(x))\, |\det D\Phi(x)|\, dx,$$
with @thm-volume-scaling as its linear core.
:::

::: {#exm-orientation}
## Rotations and Reflections

A rotation $R_\alpha$ of the plane has $\det R_\alpha = \cos^2\alpha + \sin^2\alpha = 1$: it preserves area and orientation. The reflection $(x, y) \mapsto (x, -y)$ has determinant $-1$: it preserves area but reverses orientation. No rotation can turn a basis into its mirror image.
:::
