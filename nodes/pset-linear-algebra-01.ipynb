# Problem Set: Linear Algebra



## Problem 1: Consequences of Axioms & Isomorphism

**Problem Statement:**

**(a)** Using strictly the abelian group axioms for $(V, +)$ and the scalar multiplication axioms of a [vector space](vector-spaces.ipynb) over a [field](field.ipynb) $K$, prove:

1. $0_K \cdot v = 0_V$ for all $v \in V$.
2. $(-1_K) \cdot v = -v$ for all $v \in V$.
3. For $c \in K$ and $v \in V$, $c \cdot v = 0_V \implies c = 0_K$ or $v = 0_V$.

**(b)** Let $X = \{x_1, \dots, x_n\}$ be a finite set. Show that $K^X = \{f \mid f: X \to K\}$ is isomorphic to $K^n$.

::: {.callout-tip collapse="true"}
## Solution

**(a)**

1. In $K$, $0_K + 0_K = 0_K$. By distributivity of vectors over scalar addition:
   $$
   0_K \cdot v = (0_K + 0_K) \cdot v = 0_K \cdot v + 0_K \cdot v
   $$
   Adding the additive inverse $-(0_K \cdot v) \in V$ to both sides yields:
   $$
   0_V = 0_K \cdot v
   $$

2. By scalar distributivity and the scalar identity axiom $1_K \cdot v = v$:
   $$
   v + (-1_K) \cdot v = 1_K \cdot v + (-1_K) \cdot v = (1_K + (-1_K)) \cdot v = 0_K \cdot v = 0_V
   $$
   Since the additive inverse in an abelian group is unique, $(-1_K) \cdot v = -v$.

3. Suppose $c \cdot v = 0_V$. If $c = 0_K$, the conclusion holds. If $c \ne 0_K$, since $K$ is a field, there exists $c^{-1} \in K$ such that $c^{-1} c = 1_K$. Multiplying both sides by $c^{-1}$:
   $$
   v = 1_K \cdot v = (c^{-1} c) \cdot v = c^{-1} \cdot (c \cdot v) = c^{-1} \cdot 0_V = 0_V
   $$
   Hence $c = 0_K$ or $v = 0_V$.

**(b)**

Define the map $T: K^X \to K^n$ by $T(f) = (f(x_1), \dots, f(x_n))$.

- **Linearity:** For $f, g \in K^X$ and $\alpha, \beta \in K$:
  $$
  T(\alpha f + \beta g) = ((\alpha f + \beta g)(x_1), \dots, (\alpha f + \beta g)(x_n)) = \alpha T(f) + \beta T(g)
  $$
- **Injectivity:** If $T(f) = (0, \dots, 0)$, then $f(x_i) = 0$ for all $i \in \{1, \dots, n\}$, which means $f \equiv 0$. Thus $\ker(T) = \{0\}$.
- **Surjectivity:** For any tuple $(k_1, \dots, k_n) \in K^n$, the function $f \in K^X$ defined by $f(x_i) = k_i$ satisfies $T(f) = (k_1, \dots, k_n)$.

Thus $T$ is a linear bijection, proving $K^X \cong K^n$.

$\blacksquare$
:::



## Problem 2: Subspace Union, Intersection & Span Equivalence

**Problem Statement:**

**(a)** Let $U, W \le V$ be [subspaces](subspaces.ipynb#def-subspace). Show that $U \cup W \le V \iff U \subseteq W$ or $W \subseteq U$.

**(b)** Show that the intersection $\bigcap_{i \in I} W_i$ of an arbitrary family of subspaces is a subspace.

**(c)** For $S \subseteq V$, prove that $\operatorname{span}(S) = \bigcap \{U \le V \mid S \subseteq U\}$.

::: {.callout-tip collapse="true"}
## Solution

**(a)**

- $(\Leftarrow)$ If $U \subseteq W$, then $U \cup W = W \le V$. Symmetrically, if $W \subseteq U$, then $U \cup W = U \le V$.
- $(\Rightarrow)$ Assume for contradiction that $U \cup W \le V$, but $U \not\subseteq W$ and $W \not\subseteq U$. Then there exist $u \in U \setminus W$ and $v \in W \setminus U$. Since $U \cup W$ is a subspace, $u + v \in U \cup W$:
  - If $u + v \in U$, then $v = (u + v) - u \in U$, a contradiction.
  - If $u + v \in W$, then $u = (u + v) - v \in W$, a contradiction.

Hence $U \subseteq W$ or $W \subseteq U$.

**(b)**

Let $F = \bigcap_{i \in I} W_i$.

- Since each $W_i$ is a subspace, $0_V \in W_i$ for all $i \in I$, so $0_V \in F \ne \emptyset$.
- For $u, v \in F$ and $c \in K$, since each $W_i$ is a subspace, $cu + v \in W_i$ for every $i \in I$. Hence $cu + v \in F$.

Thus $F$ is a subspace of $V$.

**(c)**

Let $U_{\min} = \bigcap \{U \le V \mid S \subseteq U\}$ and $L(S) = \left\{ \sum_{i=1}^m \alpha_i v_i \;\middle|\; m \in \mathbb{N}, \alpha_i \in K, v_i \in S \right\}$.

1. **$L(S) \subseteq U_{\min}$:** Any subspace $U$ containing $S$ contains all finite linear combinations of vectors in $S$. Thus $L(S) \subseteq U$ for all such $U$, implying $L(S) \subseteq U_{\min}$.
2. **$U_{\min} \subseteq L(S)$:** $L(S)$ is itself a subspace containing $S$ (since for any $s \in S$, $s = 1_K \cdot s \in L(S)$). Therefore $L(S)$ is one of the subspaces being intersected, which implies $U_{\min} \subseteq L(S)$.

Hence $\operatorname{span}(S) = L(S) = U_{\min}$.

$\blacksquare$
:::



## Problem 3: First-Dependent Vector Lemma

**Problem Statement:**

Let $v_1, \dots, v_m \in V$ with $m \ge 2$. Prove that $\{v_1, \dots, v_m\}$ is [linearly dependent](linear-independence.ipynb#def-linear-independence) if and only if $v_1 = 0_V$ or there exists some index $j \in \{2, \dots, m\}$ such that $v_j \in \operatorname{span}(v_1, \dots, v_{j-1})$.

::: {.callout-tip collapse="true"}
## Solution

**$(\Leftarrow)$:**

- If $v_1 = 0_V$, then $1_K v_1 + 0 v_2 + \dots + 0 v_m = 0_V$ is a non-trivial linear combination.
- If $v_j \in \operatorname{span}(v_1, \dots, v_{j-1})$ for some $j \in \{2, \dots, m\}$, then $v_j = \sum_{i=1}^{j-1} c_i v_i$, so:
  $$
  c_1 v_1 + \dots + c_{j-1} v_{j-1} - 1_K v_j + 0 v_{j+1} + \dots + 0 v_m = 0_V
  $$
  Since the coefficient of $v_j$ is $-1_K \ne 0_K$, this is a non-trivial vanishing combination.

In both cases, $\{v_1, \dots, v_m\}$ is linearly dependent.

**$(\Rightarrow)$:**

Suppose $\sum_{i=1}^m \alpha_i v_i = 0_V$ with at least one non-zero coefficient. Choose the maximal non-zero index:
$$
j = \max \{i \in \{1, \dots, m\} \mid \alpha_i \ne 0_K\}
$$

- If $j = 1$, then $\alpha_1 v_1 = 0_V$ with $\alpha_1 \ne 0_K \implies v_1 = 0_V$.
- If $j \ge 2$, then $\alpha_1 v_1 + \dots + \alpha_j v_j = 0_V$ (since $\alpha_i = 0_K$ for all $i > j$). Multiplying by $\alpha_j^{-1}$:
  $$
  v_j = \sum_{i=1}^{j-1} \left( -\frac{\alpha_i}{\alpha_j} \right) v_i \in \operatorname{span}(v_1, \dots, v_{j-1})
  $$

$\blacksquare$
:::



## Problem 4: Single Exchange Step of Steinitz

**Problem Statement:**

Let $V = \operatorname{span}(w_1, \dots, w_n)$ and let $v \in V$ be a non-zero vector expressed as $v = \sum_{i=1}^n \alpha_i w_i$. If $\alpha_k \ne 0_K$, prove that:
$$
\operatorname{span}(w_1, \dots, w_{k-1}, v, w_{k+1}, \dots, w_n) = V
$$

::: {.callout-tip collapse="true"}
## Solution

Let $S' = \{w_1, \dots, w_{k-1}, v, w_{k+1}, \dots, w_n\}$. Since $S' \subseteq V$ and $V$ is a subspace, $\operatorname{span}(S') \subseteq V$.

For the reverse inclusion, solve for $w_k$ using $\alpha_k \ne 0_K$:
$$
v = \alpha_k w_k + \sum_{i \ne k} \alpha_i w_i \implies w_k = \frac{1}{\alpha_k} v - \sum_{i \ne k} \frac{\alpha_i}{\alpha_k} w_i
$$

Since $v \in S'$ and $w_i \in S'$ for all $i \ne k$, we have $w_k \in \operatorname{span}(S')$. Therefore, the original generating set satisfies $\{w_1, \dots, w_n\} \subseteq \operatorname{span}(S')$, which implies:
$$
V = \operatorname{span}(w_1, \dots, w_n) \subseteq \operatorname{span}(S')
$$

Hence $\operatorname{span}(S') = V$.

$\blacksquare$
:::



## Problem 5: Basis of Subspace Sum $U + W$

**Problem Statement:**

Let $U, W \le V$ be finite-dimensional subspaces. Let $\{v_1, \dots, v_m\}$ be a [basis](basis-and-dimension.ipynb#def-basis) of $U \cap W$, extended to a basis $\{v_1, \dots, v_m, u_1, \dots, u_r\}$ of $U$, and to a basis $\{v_1, \dots, v_m, w_1, \dots, w_s\}$ of $W$. Prove that:
$$
\mathcal{B} = \{v_1, \dots, v_m, u_1, \dots, u_r, w_1, \dots, w_s\}
$$
is a basis of the [subspace sum](subspaces.ipynb#def-sum-subspaces) $U + W$.

::: {.callout-tip collapse="true"}
## Solution

**1. Spanning:**

Any $x \in U + W$ can be written as $x = u + w$ with $u \in U$ and $w \in W$. Expanding $u$ in the basis of $U$ and $w$ in the basis of $W$ yields:
$$
x = \sum_{i=1}^m (a_i + a'_i) v_i + \sum_{j=1}^r b_j u_j + \sum_{k=1}^s c_k w_k \in \operatorname{span}(\mathcal{B})
$$
Thus $\operatorname{span}(\mathcal{B}) = U + W$.

**2. Linear Independence:**

Suppose:
$$
\sum_{i=1}^m \alpha_i v_i + \sum_{j=1}^r \beta_j u_j + \sum_{k=1}^s \gamma_k w_k = 0_V
$$

Rearranging terms:
$$
\underbrace{\sum_{k=1}^s \gamma_k w_k}_{\in W} = - \underbrace{\left( \sum_{i=1}^m \alpha_i v_i + \sum_{j=1}^r \beta_j u_j \right)}_{\in U}
$$

This shows the vector $y = \sum_{k=1}^s \gamma_k w_k$ belongs to both $W$ and $U$, so $y \in U \cap W$. Because $\{v_1, \dots, v_m\}$ is a basis of $U \cap W$, there exist scalars $\delta_i$ such that:
$$
\sum_{k=1}^s \gamma_k w_k = \sum_{i=1}^m \delta_i v_i \implies \sum_{i=1}^m \delta_i v_i - \sum_{k=1}^s \gamma_k w_k = 0_V
$$

Since $\{v_1, \dots, v_m, w_1, \dots, w_s\}$ is a basis of $W$, its vectors are linearly independent, so:
$$
\gamma_1 = \dots = \gamma_s = 0_K
$$

Substituting these back into the original equation leaves:
$$
\sum_{i=1}^m \alpha_i v_i + \sum_{j=1}^r \beta_j u_j = 0_V
$$

Since $\{v_1, \dots, v_m, u_1, \dots, u_r\}$ is a basis of $U$, linear independence implies $\alpha_1 = \dots = \alpha_m = 0_K$ and $\beta_1 = \dots = \beta_r = 0_K$. Thus all coefficients are zero, and $\mathcal{B}$ is linearly independent.

Therefore, $\mathcal{B}$ is a basis of $U + W$.

$\blacksquare$
:::



## Problem 6: Uniqueness of Coordinate Representation

**Problem Statement:**

Let $V$ be an $n$-dimensional [vector space](vector-spaces.ipynb) over $K$ with an ordered [basis](basis-and-dimension.ipynb#def-basis) $\mathcal{B} = (v_1, \dots, v_n)$. Prove that every vector $u \in V$ has a unique representation $u = \sum_{i=1}^n \alpha_i v_i$ with $\alpha_i \in K$.

::: {.callout-tip collapse="true"}
## Solution

- **Existence:** Since $\mathcal{B}$ is a basis, it spans $V$. Therefore at least one representation exists for every $u \in V$.
- **Uniqueness:** Suppose there exists another representation $u = \sum_{i=1}^n \alpha'_i v_i$. Subtracting the two gives:
  $$
  0_V = u - u = \sum_{i=1}^n (\alpha_i - \alpha'_i) v_i
  $$
  Because $\mathcal{B}$ is a basis, the vectors $v_1, \dots, v_n$ are linearly independent. Hence the vanishing linear combination has only the trivial solution:
  $$
  \alpha_i - \alpha'_i = 0_K \implies \alpha_i = \alpha'_i \quad \forall i \in \{1, \dots, n\}
  $$

Thus the coordinate tuple $(\alpha_1, \dots, \alpha_n) \in K^n$ is unique.

$\blacksquare$
:::
