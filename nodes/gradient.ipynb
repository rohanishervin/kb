# The Gradient and Level Sets

**Prerequisites:** [The Chain Rule in Several Variables](multivariable-chain-rule.ipynb) · [Graphs, Level Sets and Quadric Surfaces](graphs-and-level-sets.ipynb) · [Inner Product Spaces and the Cauchy-Schwarz Inequality](Cauchy-Schwarz-inequality.ipynb)

Let $f : S \to \mathbb{R}$, $S \subseteq \mathbb{R}^n$, be differentiable at the interior point $a$. Then $Df(a)$ is a $1\times n$ row; turning it into a vector of $\mathbb{R}^n$ connects derivatives with the geometry of the dot product.

## 1\. Gradient and Steepest Ascent

::: {#def-gradient}
## Gradient

The **gradient** of $f$ at $a$ is $\nabla f(a) = \operatorname{grad}f(a) = \big(\frac{\partial f}{\partial x_1}(a), \dots, \frac{\partial f}{\partial x_n}(a)\big)$. By the [Jacobian formula](differentiability-in-Rn.ipynb#thm-jacobian),
$$D_vf(a) = \nabla f(a)\cdot v \qquad (v \in \mathbb{R}^n).$$
If $f$ is differentiable on $U$, $x \mapsto \nabla f(x)$ is a **vector field** on $U$, drawn as an arrow $\nabla f(x)$ attached at each point $x$.
:::

::: {#thm-steepest-ascent}
## Direction of Steepest Ascent

For unit vectors $u$, $\ -|\nabla f(a)| \le D_uf(a) = |\nabla f(a)|\cos\angle(\nabla f(a), u) \le |\nabla f(a)|$. If $\nabla f(a) \ne 0$, the maximum is attained only at $u = \nabla f(a)/|\nabla f(a)|$ (the direction of fastest increase), the minimum only at the opposite direction, and $D_uf(a) = 0$ exactly for $u \perp \nabla f(a)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-steepest-ascent
$D_uf(a) = \nabla f(a)\cdot u$; apply Cauchy–Schwarz with its equality case: $\nabla f(a)\cdot u = \pm|\nabla f(a)|$ iff $u$ is a positive (negative) multiple of $\nabla f(a)$. $\blacksquare$
:::

::: {#exm-gradient-fields}
## Gradient Fields

1. $f(x, y) = x^2 - y^2$: $\nabla f = (2x, -2y)$; at $(2, 0)$, $\nabla f = (4, 0)$, so $f$ increases fastest in direction $(1, 0)$, at rate $4$, decreases fastest in direction $(-1, 0)$, and is stationary to first order in directions $(0, \pm1)$.
2. $f(x) = |x|^2$: $\nabla f(x) = 2x$, pointing radially outward with length twice the distance from $0$.
3. For the gravitational potential $U = -\frac kr$, $r = |x|$ in $\mathbb{R}^3$: $\nabla U = \frac{k}{r^3}x$, and the force $F = -\nabla U = -\frac{k}{r^3}x$ points to the origin with magnitude $k/r^2$.
4. $x + y$, $(x + y)^3$, $(x + y)^{1/3}$ (off the line $x + y = 0$ for the last) have gradients $(1, 1)$, $3(x + y)^2(1, 1)$, $\frac13(x + y)^{-2/3}(1, 1)$: all orthogonal to the common level lines, with lengths recording how densely the levels are packed.
:::

## 2\. The Gradient Is Orthogonal to Level Sets

::: {#thm-gradient-orthogonal}
## Orthogonality to Level Sets

Let $f$ be differentiable near $a$, $f(a) = c$, and let $\gamma : I \to \mathbb{R}^n$ be a differentiable curve lying in the level set $f^{-1}(c)$ with $\gamma(t_0) = a$. Then $\nabla f(a)\cdot\gamma'(t_0) = 0$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-gradient-orthogonal
$f(\gamma(t)) = c$ for all $t$, so by the [chain rule](multivariable-chain-rule.ipynb#thm-multivariable-chain-rule) $0 = \frac{d}{dt}f(\gamma(t))\big|_{t_0} = \nabla f(a)\cdot\gamma'(t_0)$. $\blacksquare$
:::

::: {#def-critical-point}
## Critical and Regular Points

$a$ is a **critical point** of $f$ if $\nabla f(a) = 0$, and a **regular point** otherwise. At a regular point of a $C^1$ function, the hyperplane $\nabla f(a)\cdot(x - a) = 0$ is the **tangent hyperplane** to the level set through $a$.
:::

Every velocity vector of a curve in the level set lies in this hyperplane (@thm-gradient-orthogonal); conversely, the [implicit function theorem](implicit-function-theorem.ipynb) will show that near a regular point the level set is the graph of a $C^1$ function of $n - 1$ variables, and every vector orthogonal to $\nabla f(a)$ is the velocity of such a curve. At critical points the level set may fail to be smooth. In summary: **the gradient is orthogonal to the level sets and points toward increasing values.** This is the principle behind [gradient descent](gradient-descent-and-variants.ipynb).

::: {#exm-tangent-plane-level}
## Tangent Plane to a Surface

For the ellipsoid $x^2 + 2y^2 + 3z^2 = 6$ at $(1, 1, 1)$: $\nabla f = (2x, 4y, 6z) = (2, 4, 6)$, so the tangent plane is $2(x - 1) + 4(y - 1) + 6(z - 1) = 0$, i.e. $x + 2y + 3z = 6$.
:::

::: {#exm-critical-level-sets}
## Level Sets Through Critical Points

1. $f = x^2 + y^2 - z^2$, $\nabla f = (2x, 2y, -2z)$. The level sets are hyperboloids of one sheet ($c > 0$), the cone ($c = 0$) and hyperboloids of two sheets ($c < 0$). The only critical point is $0$, exactly where the cone is not smooth. Off the $z$-axis the gradient points away from the axis, downward if $z > 0$ and upward if $z < 0$.
2. $f = (x^2 - 1)^2 + y^2$, $\nabla f = (4x(x^2 - 1), 2y)$, with critical points $(\pm1, 0)$ and $(0, 0)$. The level set $f = 0$ consists of the two points $(\pm1, 0)$ (the minima); for $0 < c < 1$ it is two small closed curves around them; $f = 1$ is a figure-eight curve crossing itself at the critical point $(0, 0)$; for $c > 1$ it is one closed curve around everything.
:::

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

fig, axs = plt.subplots(1, 2, figsize=(12, 5))
x = np.linspace(-1.8, 1.8, 400); X, Y = np.meshgrid(x, x)
F = (X**2 - 1)**2 + Y**2
cs = axs[0].contour(X, Y, F, levels=[0.1, 0.4, 0.7, 1.0, 1.5, 2.5, 4], cmap="viridis")
axs[0].contour(X, Y, F, levels=[1.0], colors=[ORANGE], linewidths=2)
axs[0].clabel(cs, fontsize=8)
g = np.linspace(-1.6, 1.6, 15); GX, GY = np.meshgrid(g, g)
U, V = 4 * GX * (GX**2 - 1), 2 * GY
n = np.hypot(U, V) + 1e-9
axs[0].quiver(GX, GY, U / n, V / n, color=MUTED, scale=30, width=0.003)
axs[0].plot([-1, 0, 1], [0, 0, 0], "o", color=RED, ms=5)
axs[0].set_aspect("equal"); axs[0].set_title(r"$(x^2 - 1)^2 + y^2$: level curves and $\nabla f/|\nabla f|$")

F2 = X**2 - Y**2
axs[1].contour(X, Y, F2, levels=np.linspace(-3, 3, 13), cmap="coolwarm")
U2, V2 = 2 * GX, -2 * GY
axs[1].quiver(GX, GY, U2, V2, color=INK, scale=60, width=0.003)
axs[1].set_aspect("equal"); axs[1].set_title(r"$x^2 - y^2$: the gradient field is orthogonal to the levels")
plt.tight_layout(); plt.show()
```
