# Power Series

**Prerequisites:** [Taylor Series](taylor-series.ipynb) · [Absolute and Conditional Convergence](absolute-and-conditional-convergence.ipynb) · [The Fundamental Theorem of Calculus](fundamental-theorem-of-calculus.ipynb)

## 1\. Radius of Convergence

A **power series** about $c$ is $\sum_{n\ge0} c_n(z - c)^n$ with $c, c_n \in \mathbb{C}$ (or all real). Real numbers in base $10$ are "power series in $\frac{1}{10}$" with digit coefficients; power series play the same role for functions.

::: {#thm-radius-of-convergence}
## Radius of Convergence

For every power series there is $\rho \in [0, \infty]$, its **radius of convergence**, such that the series converges absolutely for $|z - c| < \rho$ and diverges for $|z - c| > \rho$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-radius-of-convergence
*Lemma: if the series converges at $z_1$, it converges absolutely for every $z$ with $|z - c| < |z_1 - c|$.* The terms at $z_1$ tend to $0$, so $|c_n(z_1 - c)^n| \le K$. With $\sigma = \frac{|z - c|}{|z_1 - c|} < 1$, $|c_n(z - c)^n| \le K\sigma^n$; compare with the [geometric series](geometric-sequence.ipynb#thm-geometric-series).

Let $\rho = \sup\{|z - c| : \text{the series converges at } z\}$ ($\ge 0$, possibly $\infty$). If $|z - c| < \rho$, there is a point $z_1$ of convergence with $|z_1 - c| > |z - c|$, and the lemma applies. If $|z - c| > \rho$, convergence at $z$ would contradict the definition of $\rho$. $\blacksquare$
:::

The disc $|z - c| < \rho$ (for real series, the interval $(c - \rho, c + \rho)$) is the region of convergence; on its boundary anything can happen.

::: {#prp-radius-formula}
## Computing $\rho$

If $\big|\frac{c_{n+1}}{c_n}\big| \to L$ or $\sqrt[n]{|c_n|} \to L$, with $L \in [0, \infty]$, then $\rho = \frac1L$ (with $\frac10 = \infty$, $\frac1\infty = 0$).
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-radius-formula
The ratio (or root) of the absolute terms tends to $L|z - c|$. By the [ratio and root tests](series.ipynb#thm-ratio-test) the series converges absolutely if $L|z - c| < 1$ and its terms do not tend to $0$ if $L|z - c| > 1$. $\blacksquare$
:::

::: {#exm-radius}
## Examples

1. $\exp z = \sum \frac{z^n}{n!}$: $\frac{1/(n+1)!}{1/n!} = \frac{1}{n+1} \to 0$, so $\rho = \infty$. Likewise $\sin z = \sum\frac{(-1)^kz^{2k+1}}{(2k+1)!}$, $\cos z$, $\sinh z$, $\cosh z$ (defined by the series of the [previous note](taylor-series.ipynb#thm-elementary-taylor-series)) have $\rho = \infty$, by comparison with $\sum \frac{|z|^n}{n!}$.
2. $\sum n!\,z^n$: ratio $n + 1 \to \infty$, so $\rho = 0$; it converges only at $z = 0$.
3. $\sum_{n\ge1}\frac{z^n}{n^p}$ ($p \ge 0$): $\sqrt[n]{1/n^p} \to 1$, so $\rho = 1$. On $|z| = 1$: for $p > 1$ it converges absolutely everywhere; for $p = 1$ it diverges at $z = 1$ (harmonic) but converges at $z = -1$ (Leibniz); for $p = 0$ it diverges at every $|z| = 1$ (terms do not tend to $0$).
:::

::: {#prp-euler-formula}
## Euler's Formula

For real $\theta$, $\exp(i\theta) = \cos\theta + i\sin\theta = \operatorname{cis}\theta$. So the polar form of [complex numbers](complex-numbers.ipynb#thm-polar-product) is $z = re^{i\theta}$, and $\operatorname{cis}\theta\operatorname{cis}\varphi = \operatorname{cis}(\theta + \varphi)$ is $e^{i\theta}e^{i\varphi} = e^{i(\theta+\varphi)}$ ([Cauchy product](absolute-and-conditional-convergence.ipynb#exm-cauchy-product)).
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-euler-formula
$\sum\frac{(i\theta)^n}{n!}$ converges absolutely; by @lem-split-even-odd below we may sum the even and odd terms separately. Since $i^{2k} = (-1)^k$ and $i^{2k+1} = (-1)^ki$, the even terms sum to $\cos\theta$ and the odd ones to $i\sin\theta$. $\blacksquare$
:::

::: {#lem-split-even-odd}
## Splitting a Series

If $\sum a_n$ converges absolutely, then $\sum a_n = \sum_k a_{2k} + \sum_k a_{2k+1}$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-split-even-odd
Both subseries converge absolutely (comparison), and the partial sum $s_{2K+1} = \sum_{k\le K}a_{2k} + \sum_{k\le K}a_{2k+1}$; let $K \to \infty$. $\blacksquare$
:::

## 2\. Differentiation and Integration Term by Term

From now on the series are real: $f(x) = \sum_{n\ge0}a_n(x - a)^n$ on $(a - \rho, a + \rho)$, $\rho > 0$.

::: {#lem-same-radius}
## Derived and Integrated Series Have the Same Radius

$\sum a_n(x - a)^n$, $\sum n a_n(x - a)^{n-1}$ and $\sum \frac{a_n}{n+1}(x - a)^{n+1}$ have the same radius of convergence. Moreover $\sum n^2|a_n|r^{n}$ converges for every $0 \le r < \rho$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-same-radius
Let $0 < r < s < \rho$. The terms $|a_n|s^n$ are bounded by some $K$, so $n^2|a_n|r^n \le Kn^2(r/s)^n$, and $\sum n^2 q^n$ converges for $q < 1$ (ratio test). Hence the derived series (whose terms are $\le n|a_n|r^{n-1}$) converges absolutely for $|x - a| \le r$; so its radius is $\ge \rho$. Conversely $|a_n|r^n \le r\cdot n|a_n|r^{n-1}$, so the original radius is $\ge$ that of the derived series. The integrated series has the original as its derived series, so the same argument applies. $\blacksquare$
:::

::: {#thm-termwise}
## Power Series Can Be Differentiated and Integrated Term by Term

On $(a - \rho, a + \rho)$, $f$ is differentiable and
$$f'(x) = \sum_{n\ge1} n a_n (x - a)^{n-1}, \qquad \int_a^x f = \sum_{n\ge0}\frac{a_n}{n+1}(x - a)^{n+1}.$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-termwise
Take $a = 0$. *Algebraic lemma:* for $|x|, |y| \le r$, $n \ge 2$, $x \ne y$,
$$\Big|\frac{y^n - x^n}{y - x} - nx^{n-1}\Big| \le \frac{n(n-1)}{2}r^{n-2}|y - x| .$$
Indeed the quotient is $\sum_{k=0}^{n-1}y^{n-1-k}x^k$, so the difference is $\sum_{k=0}^{n-1}x^k\big(y^{n-1-k} - x^{n-1-k}\big)$, and $|y^j - x^j| \le jr^{j-1}|y - x|$ (same identity); summing $\sum_k (n-1-k)r^{n-2}|y-x|$ gives the bound.

Fix $|x| < r < \rho$ and let $g(x) = \sum na_nx^{n-1}$ (convergent by @lem-same-radius). For $y \ne x$, $|y| \le r$:
$$\Big|\frac{f(y) - f(x)}{y - x} - g(x)\Big| \le \sum_{n\ge2}|a_n|\frac{n(n-1)}{2}r^{n-2}\,|y - x| = C\,|y - x|,$$
with $C < \infty$ by @lem-same-radius. Letting $y \to x$ gives $f'(x) = g(x)$.

*Integration.* $G(x) = \sum\frac{a_n}{n+1}x^{n+1}$ has radius $\rho$ and, by what was just shown, $G' = f$ with $G(0) = 0$. $f$ is continuous (being differentiable), so by the [FTC](fundamental-theorem-of-calculus.ipynb#thm-ftc-evaluation) $\int_0^x f = G(x) - G(0) = G(x)$. $\blacksquare$
:::

::: {#cor-power-series-taylor}
## A Power Series Is the Taylor Series of Its Sum

$f$ is infinitely differentiable on $(a - \rho, a + \rho)$ with $f^{(k)}(x) = \sum_{n\ge k}n(n-1)\cdots(n-k+1)a_n(x - a)^{n-k}$. In particular $a_k = \frac{f^{(k)}(a)}{k!}$: the coefficients are determined by $f$, and the series is the Taylor series of $f$ at $a$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-power-series-taylor
Apply @thm-termwise $k$ times and put $x = a$: only the term $n = k$ survives, giving $k!\,a_k$. $\blacksquare$
:::

Termwise differentiation is special to power series: a convergent series of differentiable functions may have a non-differentiable, even discontinuous, sum ([Fourier series](fourier-series.ipynb#exm-square-wave)).

## 3\. Analytic Functions

::: {#def-analytic}
## Analytic Function

$f$ is **analytic** at $a$ if it is given by a power series $\sum a_n(x-a)^n$ (necessarily its Taylor series) on some interval $(a - \sigma, a + \sigma)$, $\sigma > 0$.
:::

$e^x, \sin, \cos, \sinh, \cosh$ are analytic at $0$ with $\sigma = \infty$; $\frac1x$ is analytic at $1$ with $\sigma = 1$; $e^{-1/x^2}$ is infinitely differentiable but not analytic at $0$.

::: {#thm-reexpansion}
## Re-expansion About Another Point

If $f(x) = \sum a_n(x - a)^n$ for $|x - a| < \rho$ and $|b - a| < \rho$, then $f$ is analytic at $b$: its Taylor series at $b$ converges to $f$ for $|x - b| < \rho - |b - a|$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-reexpansion
Let $d = |b - a|$ and $|x - b| < \rho - d$. By the binomial theorem,
$$a_n(x - a)^n = a_n\big((x - b) + (b - a)\big)^n = \sum_{k=0}^n a_n\binom nk(b - a)^{n-k}(x - b)^k .$$
The double series of absolute values sums to $\sum_n|a_n|\big(|x - b| + d\big)^n < \infty$, since $|x - b| + d < \rho$. By the [double series lemma](absolute-and-conditional-convergence.ipynb#lem-double-series) we may sum over $k$ first: $f(x) = \sum_k b_k(x - b)^k$ with $b_k = \sum_{n\ge k}a_n\binom nk(b - a)^{n-k}$. By @cor-power-series-taylor this is the Taylor series of $f$ at $b$. $\blacksquare$
:::

## 4\. Behaviour at the Endpoints

::: {#thm-abel}
## Abel's Theorem

Let $f(x) = \sum a_n(x - a)^n$ have radius $\rho \in (0, \infty)$. If the series converges at $x = a + \rho$, then $\displaystyle\lim_{x \to (a+\rho)^-}f(x) = \sum a_n\rho^n$; similarly at $a - \rho$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-abel
Substituting $x = a + \rho u$ we may assume $a = 0$, $\rho = 1$; let $s = \sum a_n$, $s_n = a_0 + \dots + a_n$ ($s_{-1} = 0$). The $s_n$ are bounded, so $\sum s_nx^n$ converges for $|x| < 1$, and
$$f(x) = \sum_n (s_n - s_{n-1})x^n = (1 - x)\sum_n s_nx^n, \qquad f(x) - s = (1 - x)\sum_n (s_n - s)x^n,$$
using $(1 - x)\sum x^n = 1$. Given $\varepsilon$, take $N$ with $|s_n - s| < \varepsilon$ for $n > N$. For $0 < x < 1$,
$$|f(x) - s| \le (1 - x)\sum_{n\le N}|s_n - s| + \varepsilon(1 - x)\sum_{n>N}x^n \le (1 - x)C + \varepsilon,$$
which is $< 2\varepsilon$ when $x$ is close to $1$. $\blacksquare$
:::

::: {#exm-log-series}
## Logarithm and Arctangent

1. $\frac1x = \sum_{n\ge0}(-1)^n(x - 1)^n$ on $(0, 2)$. Termwise differentiation gives $-\frac{1}{x^2} = \sum_{n\ge1}(-1)^nn(x - 1)^{n-1}$, and integration from $1$ gives
$$\ln x = \sum_{n\ge0}\frac{(-1)^n}{n+1}(x - 1)^{n+1}, \qquad \ln(1 + t) = t - \frac{t^2}{2} + \frac{t^3}{3} - \dots \quad (|t| < 1).$$
At $t = 1$ the series converges ([Leibniz](absolute-and-conditional-convergence.ipynb#thm-alternating-series)) and $\ln$ is continuous at $2$, so by Abel's theorem $1 - \frac12 + \frac13 - \frac14 + \dots = \ln 2$. At $t = -1$ it is minus the harmonic series and diverges.
2. More generally, for $a > 0$, $\frac1x = \frac1a\cdot\frac{1}{1 + \frac{x - a}{a}} = \sum_n \frac{(-1)^n}{a^{n+1}}(x - a)^n$ for $|x - a| < a$, so $\ln x$ is analytic at every $a > 0$. The radius cannot exceed $a$: the function is not defined at $0$.
3. $\frac{1}{1 + x^2} = \sum_n(-1)^nx^{2n}$ for $|x| < 1$, so integrating, $\arctan x = x - \frac{x^3}{3} + \frac{x^5}{5} - \dots$ for $|x| < 1$. At $x = \pm1$ the series converges (Leibniz), and Abel's theorem gives
$$\frac{\pi}{4} = \arctan 1 = 1 - \frac13 + \frac15 - \frac17 + \cdots$$
Although $\frac{1}{1 + x^2}$ is analytic on all of $\mathbb{R}$, its series at $0$ has radius $1$: in $\mathbb{C}$, $1 + z^2$ vanishes at $z = \pm i$, at distance $1$ from $0$.
4. $e^x = \sum_n\frac{e^a}{n!}(x - a)^n$ for all $x$, and $\sin x = \sum_k\frac{(-1)^k}{(2k)!}\big(x - \frac\pi2\big)^{2k}$, since $\sin^{(n)}(\frac\pi2)$ is $0$ for odd $n$ and $(-1)^k$ for $n = 2k$.
:::

```python

import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

x = np.linspace(-1.5, 1.5, 600)
fig, (ax, ax2) = plt.subplots(1, 2, figsize=(12, 4.3))
ax.plot(x, np.arctan(x), color=INK, lw=2.5, label=r"$\arctan x$")
for N, col in zip([1, 3, 7, 15], [BLUE, ORANGE, AQUA, VIOLET]):
    ax.plot(x, sum((-1)**k * x**(2*k + 1) / (2*k + 1) for k in range(N + 1)), color=col, lw=1.3,
            label=f"partial sum to $x^{{{2*N + 1}}}$")
ax.axvspan(-1, 1, color=AQUA, alpha=0.08); ax.set_ylim(-2, 2); ax.legend(fontsize=8, loc="upper left")
ax.set_title(r"$\arctan$: radius $1$, although $\arctan$ is smooth on $\mathbb{R}$")

K = np.arange(0, 2000)
leib = np.cumsum((-1.0)**K / (2*K + 1))
alt = np.cumsum((-1.0)**K / (K + 1))
ax2.semilogx(K + 1, np.abs(4 * leib - np.pi), color=BLUE, label=r"$|4\sum_{k\leq n}\frac{(-1)^k}{2k+1} - \pi|$")
ax2.semilogx(K + 1, np.abs(alt - np.log(2)), color=ORANGE, label=r"$|\sum_{k\leq n}\frac{(-1)^k}{k+1} - \ln 2|$")
ax2.semilogx(K + 1, 1 / (K + 2), color=MUTED, ls=":", label=r"$1/(n+2)$")
ax2.set_yscale("log"); ax2.legend(fontsize=9); ax2.set_xlabel("$n$")
ax2.set_title("Abel sums at the endpoints: slow (alternating) convergence")
plt.tight_layout(); plt.show()
print("4 * sum to 2000 terms:", 4 * leib[-1], "| pi =", np.pi)
```
