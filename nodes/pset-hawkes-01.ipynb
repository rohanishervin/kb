# Problem Set: Hawkes Processes

Problems over the seven notes of [Hawkes Processes](index-hawkes-processes.ipynb), in the order of the notes. Most problems combine a computation with a short proof; the last one is a challenge. Problems marked *(computational)* are meant to be checked with a few lines of code.

Every problem was written for this set. A **Source** line is added only where the statement is a well-known classical result.


## Problem 1: Buy and Sell Orders as Poisson Streams

**Covers:** [superposition](point-processes-and-poisson-process.ipynb#thm-superposition), [thinning](point-processes-and-poisson-process.ipynb#thm-thinning), [exponential gaps and conditional uniformity](point-processes-and-poisson-process.ipynb#thm-poisson-from-exponential-gaps)

**Problem Statement:** Buy market orders arrive as a Poisson process with rate $3$ per minute and, independently, sell market orders as a Poisson process with rate $2$ per minute.

**(a)** Find the probability that no order at all arrives during a given 30-second interval.

**(b)** Given that exactly $10$ orders arrived during the first two minutes, find the conditional distribution of the number of buy orders among them, and the conditional expected time of the first order.

**(c)** Prove that the sequence of order *sides* (buy/sell, in arrival order) is i.i.d. with $P(\text{buy}) = 3/5$, independent of the arrival times of the merged stream.

**(d)** Deduce the distribution of the number of buy orders that arrive before the first sell order.


## Problem 2: A Linearly Increasing Rate

**Covers:** [inhomogeneous Poisson process](point-processes-and-poisson-process.ipynb#def-inhomogeneous-poisson), [time change](point-processes-and-poisson-process.ipynb#thm-poisson-time-change)

**Problem Statement:** Let $N$ be an inhomogeneous Poisson process on $[0, 1]$ with rate $\lambda(t) = a + bt$, $a, b > 0$.

**(a)** Compute $P(N(0, 1] = 0)$; evaluate it for $a = 1$, $b = 2$.

**(b)** Find the law of the first event time $T_1$ on $[0, 1]$ (it has a density on $(0, 1)$ and an atom "no event").

**(c)** Using the time-change theorem, give an explicit formula $T_k = \Lambda^{-1}(S_k)$ that turns the event times $S_k$ of a rate-$1$ Poisson process into the event times of $N$.

**(d)** Given $N(0, 1] = n$, what is the joint law of the (unordered) event times? Prove your answer from part 3 of the time-change theorem.


## Problem 3: The Weibull Renewal Process

**Covers:** [conditional intensity](conditional-intensity-and-compensator.ipynb#def-conditional-intensity), [renewal example](conditional-intensity-and-compensator.ipynb#exm-intensities), [likelihood](conditional-intensity-and-compensator.ipynb#thm-point-process-likelihood), [time rescaling](conditional-intensity-and-compensator.ipynb#thm-time-rescaling)

**Problem Statement:** The inter-arrival times of a renewal process are i.i.d. Weibull with shape $k > 0$ and scale $s > 0$: $P(E > x) = e^{-(x/s)^k}$.

**(a)** Write down the conditional intensity $\lambda_t$. For which $k$ do events cluster, and for which are they more regular than Poisson?

**(b)** Write the log-likelihood of the observed events $t_1 < \cdots < t_n$ on $[0, T]$.

**(c)** For fixed $k$, show that the maximum likelihood estimator of the scale satisfies $\hat s^{\,k} = \frac1n\big(\sum_{i=1}^n E_i^k + (T - t_n)^k\big)$, where $E_i = t_i - t_{i-1}$.

**(d)** Compute the residual gaps $\Lambda_{t_k} - \Lambda_{t_{k-1}}$ explicitly and verify directly, without the time-rescaling theorem, that they are i.i.d. $\operatorname{Expo}(1)$.


## Problem 4: The Compensator and the Mean Count

**Covers:** [compensator martingale](conditional-intensity-and-compensator.ipynb#thm-compensator-martingale), [integration against $dN$](conditional-intensity-and-compensator.ipynb#cor-campbell), [mean intensity of the exponential Hawkes process](hawkes-process.ipynb#cor-exponential-mean)

**Problem Statement:** Let $N$ be an exponential Hawkes process with $\lambda_0 = \mu$, kernel $\alpha e^{-\beta t}$, $\alpha < \beta$.

**(a)** Using only $E\,N_t = E\,\Lambda_t$ and the formula for $m(t) = E\,\lambda_t$, prove that
$$
E\,N_t = \bar\lambda t + (\mu - \bar\lambda)\frac{1 - e^{-\kappa t}}{\kappa}, \qquad \kappa = \beta - \alpha,\ \bar\lambda = \frac{\beta\mu}{\kappa}.
$$

**(b)** Evaluate $E\,N_{10}$ for $\mu = 1$, $\alpha = 1$, $\beta = 3$, and compare with the stationary value $\bar\lambda \cdot 10$. Explain the sign of the difference.

**(c)** For $\alpha > \beta$, find the growth rate of $E\,N_t$.


## Problem 5: Computing an Intensity by Hand

**Covers:** [exponential Hawkes process](hawkes-process.ipynb#def-exponential-hawkes), [Markov recursion](hawkes-process.ipynb#prp-exponential-markov), [log-likelihood](hawkes-estimation.ipynb#prp-hawkes-loglik)

**Problem Statement:** An exponential Hawkes process with $\mu = 0.5$, $\alpha = 0.8$, $\beta = 2$ and $\lambda_0 = \mu$ has events at $t = 1$, $1.5$ and $3$ on the window $[0, 4]$.

**(a)** Compute $\lambda_3$ (the left limit) and $\lambda_{3+}$ using the recursion, without summing over all past events.

**(b)** Compute the compensator $\Lambda_4$.

**(c)** Compute the log-likelihood of the data.

**(d)** Sketch $\lambda_t$ on $[0, 4]$, marking the values at the events.


## Problem 6: Clusters

**Covers:** [cluster representation](hawkes-process.ipynb#thm-cluster-representation), [branching structure](hawkes-process.ipynb#prp-branching-structure), [endogenous fraction](hawkes-process.ipynb#cor-endogenous-fraction)

**Problem Statement:** Consider a Hawkes process with branching ratio $\eta = 0.75$.

**(a)** Find the expected cluster size and the long-run fraction of events that are immigrants.

**(b)** Find the probability that an immigrant has no descendants, and the probability that its cluster has exactly two events. Check the second answer against the Borel distribution.

**(c)** Show that the expected number of events in generation $k$ of a cluster is $\eta^k$, and compute the probability that generation $2$ is empty.

**(d)** For $\eta = 1.2$, find numerically the probability that a cluster is infinite *(computational)*.


## Problem 7: A Delayed Kernel and the Renewal Equation

**Covers:** [kernels](hawkes-process.ipynb#def-kernels), [mean intensity](hawkes-process.ipynb#thm-hawkes-mean)

**Problem Statement:** Let $\varphi(t) = \eta\beta^2te^{-\beta t}$ (a gamma kernel with shape $2$).

**(a)** Show that $\|\varphi\|_1 = \eta$ and that $\varphi$ peaks at $t = 1/\beta$.

**(b)** Taking Laplace transforms in the renewal equation, show that $\hat m(s) = \dfrac{\mu}{s\,(1 - \hat\varphi(s))}$ with $\hat\varphi(s) = \eta\beta^2/(s + \beta)^2$.

**(c)** For $\beta = 1$ and $\eta = 1/4$ invert the transform and show
$$
m(t) = \mu\Big(\frac43 - \frac12e^{-t/2} + \frac16e^{-3t/2}\Big).
$$
Check $m(0) = \mu$ and $m(\infty) = \mu/(1 - \eta)$.

**(d)** Explain why $m'(0) = 0$ here, whereas for the exponential kernel $m'(0) = \alpha\mu > 0$.


## Problem 8: Second-Order Statistics

**Covers:** [second-order theorem](hawkes-process.ipynb#thm-hawkes-second-order), [Dynkin's formula](hawkes-process.ipynb#lem-dynkin-hawkes)

**Problem Statement:** Let $N$ be a stationary exponential Hawkes process with $\mu = 0.6$, $\alpha = 1$, $\beta = 2.5$.

**(a)** Compute $\eta$, $\bar\lambda$, $\kappa$ and $\operatorname{Var}\lambda_t$.

**(b)** Compute $\operatorname{Var}N(0, 2]$ and the limit of the Fano factor as the window grows.

**(c)** Prove that the Fano factor $F(\tau)$ is increasing in $\tau$, with $F(0^+) = 1$.

**(d)** Use Dynkin's formula with $f(x) = x^3$ to find $E\,\lambda_t^3$ in terms of $\mu, \alpha, \beta$.


## Problem 9: Simulation Algorithms

**Covers:** [exact simulation](hawkes-simulation.ipynb#prp-exact-exponential), [Ogata's thinning](hawkes-simulation.ipynb#thm-ogata-thinning)

**Problem Statement:**

**(a)** In the exact algorithm with excess intensity $Y = 2$ and $\beta = 1$, compute $E_2$ for the uniform draws $U = 0.5$ and $U = 0.1$, and the probability $P(E_2 = \infty)$.

**(b)** Prove that the waiting time $W = \min(E_1, E_2)$ satisfies $E\,W < \infty$ and compute $P(W = E_2)$ as an integral.

**(c)** For the gamma kernel of Problem 7, explain why the current intensity $\lambda_{t+}$ is *not* a valid ceiling for Ogata's algorithm, and give a valid bound over a look-ahead window $(t, t + L]$.

**(d)** For the exponential kernel with ceiling $M = \lambda_{t+}$, show that the expected number of rejected candidates on $[0, T]$ is $E\int_0^T(M_t - \lambda_t)\,dt$, where $M_t$ is the ceiling in force at time $t$.


## Problem 10: Maximum Likelihood by Hand

**Covers:** [exponential log-likelihood](hawkes-estimation.ipynb#prp-exponential-loglik), [concavity and the compensator identity](hawkes-estimation.ipynb#prp-loglik-concave)

**Problem Statement:** Events are observed at $t = 1.0, 1.1, 1.3, 1.4, 3.5$ on $[0, 4]$. Fix $\beta = 2$.

**(a)** Compute $A_1, \dots, A_5$ with the recursion and $S = \frac1\beta\sum_j(1 - e^{-\beta(4 - t_j)})$.

**(b)** Write $\ell(\mu, \alpha)$ and prove it is strictly concave on $(0, \infty)^2$.

**(c)** *(computational)* Find the maximizer $(\hat\mu, \hat\alpha)$ and verify $\hat\mu\cdot 4 + \hat\alpha S = 5$.

**(d)** Show that for the data $t = 0.5, 1.2, 1.4, 3.0$ the maximum is attained at $\hat\alpha = 0$, and explain why from the data.


## Problem 11: Branching Probabilities

**Covers:** [complete-data likelihood and branching probabilities](hawkes-estimation.ipynb#prp-complete-likelihood)

**Problem Statement:** A Hawkes process with $\mu = 0.5$ and $\varphi(t) = e^{-2t}$ has events at $0$, $0.1$ and $2$.

**(a)** Compute the probabilities that the second event is an immigrant and that it is a child of the first.

**(b)** Compute the three branching probabilities $p_{30}, p_{31}, p_{32}$ for the third event.

**(c)** Compute the expected number of immigrants among the three events.

**(d)** Prove that, given the event times, the parent labels are independent across events, but *not* independent of the event times.


## Problem 12: Parameters from Binned Counts

**Covers:** [method of moments for binned data](hawkes-estimation.ipynb#def-gmm-hawkes), [second-order theorem](hawkes-process.ipynb#thm-hawkes-second-order)

**Problem Statement:** Counts of a stationary exponential Hawkes process in bins of width $\tau = 1$ have mean $2$, variance $8.3918$, lag-1 autocovariance $4.6445$ and lag-2 autocovariance $2.8171$ (population values, to four decimals).

**(a)** Use the ratio of the two autocovariances to find $\kappa = \beta - \alpha$.

**(b)** Use the mean and the variance to find $\beta/\kappa$, hence $\beta$, $\alpha$ and $\mu$.

**(c)** Check your answer against the lag-1 autocovariance.

**(d)** Explain why one bin width cannot identify a kernel time scale much shorter than $\tau$.


## Problem 13: Stability of a Two-Type Network

**Covers:** [stability and mean intensity](multivariate-hawkes-processes.ipynb#thm-multivariate-stability), [expected generation sizes](multivariate-hawkes-processes.ipynb#lem-generation-means)

**Problem Statement:** A bivariate Hawkes process has branching matrix $\Phi = \begin{pmatrix}0.2 & 0.5\\ 0.4 & 0.3\end{pmatrix}$ and baselines $\mu = (0.1, 0.2)$.

**(a)** Compute $\rho(\Phi)$ and decide whether the process is stable.

**(b)** Compute the stationary mean intensities.

**(c)** Find the expected numbers of type-1 and type-2 events in a cluster started by a type-1 immigrant.

**(d)** For $\Phi_c = \begin{pmatrix}c & 0.5\\ 0.5 & c\end{pmatrix}$, find all $c \ge 0$ for which the process is stable. Show by an example that "every row sum is less than $1$" is sufficient for stability but not necessary.


## Problem 14: Marks and the Price Model

**Covers:** [branching ratio with marks](hawkes-processes-in-market-microstructure.ipynb#prp-marked-branching-ratio), [price model](hawkes-processes-in-market-microstructure.ipynb#def-hawkes-price-model), [signature plot](hawkes-processes-in-market-microstructure.ipynb#thm-signature-plot)

**Problem Statement:**

**(a)** Trade volumes are i.i.d. exponential with mean $\bar v$, the impact function is $\psi(v) = (v/\bar v)^\gamma$, $\gamma \ge 0$, and $\|\varphi\|_1 = 0.5$. Find the range of $\gamma$ for which the marked process is stable.

**(b)** In the price model with $\mu = 1$, $\alpha = 0.5$, $\beta = 1$, compute $\bar\lambda$, $C(0^+)$, $C(\infty)$ and $C(1)$.

**(c)** At which sampling interval $\tau$ is $C(\tau)$ halfway between $C(0^+)$ and $C(\infty)$? *(computational)*

**(d)** A trader estimates daily variance by summing squared one-second tick returns. In this model, by what factor is the estimate biased when the kernel time scale is much shorter than a second? Much longer?


## Problem 15: Challenge — Trend and Mean Reversion

**Covers:** [signature plot](hawkes-processes-in-market-microstructure.ipynb#thm-signature-plot), [Dynkin's formula](hawkes-process.ipynb#lem-dynkin-hawkes), [stability](multivariate-hawkes-processes.ipynb#thm-multivariate-stability)

**Problem Statement:** Generalize the price model: up- and down-moves have baseline $\mu$ each, every move excites moves *in the same direction* with kernel $a e^{-\beta t}$ and moves *in the opposite direction* with kernel $b e^{-\beta t}$, $a, b \ge 0$.

**(a)** Show that the total number of moves is a univariate Hawkes process, and state the stability condition and $\bar\lambda$ (per side).

**(b)** Let $Y = \lambda^+ - \lambda^-$. Show that $dY_t = -\beta Y_t\,dt + (a - b)\,dX_t$ and that $\gamma := \beta + b - a > 0$ under the stability condition.

**(c)** Prove that
$$
C(\tau) = 2\bar\lambda\Big[\frac{\beta^2}{\gamma^2} + \Big(1 - \frac{\beta^2}{\gamma^2}\Big)\frac{1 - e^{-\gamma\tau}}{\gamma\tau}\Big].
$$

**(d)** Show that the signature plot decreases if $b > a$ (mean reversion dominates) and increases if $a > b$ (trend following dominates), and interpret $C(\infty)/C(0^+)$.

**(e)** *(computational)* Simulate the model with $\mu = 0.5$, $a = 0.5$, $b = 0.2$, $\beta = 1$ and compare the empirical signature plot with (c).
