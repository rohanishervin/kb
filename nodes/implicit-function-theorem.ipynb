# The Implicit Function Theorem

**Prerequisites:** [The Inverse Function Theorem and Curvilinear Coordinates](inverse-function-theorem.ipynb) · [The Gradient and Level Sets](gradient.ipynb) · [Kernel, Image, and the Rank-Nullity Theorem](kernel-image-and-rank-nullity.ipynb#thm-rank-nullity)

## 1\. Implicit Differentiation

If a relation $F(x, y) = 0$ defines $y = f(x)$ with $f$ differentiable, then $F(x, f(x)) = 0$, and the chain rule gives $F_x + F_y\frac{dy}{dx} = 0$, so
$$\frac{dy}{dx} = -\frac{F_x}{F_y} \quad\text{where } F_y \ne 0 .$$
For the circle $x^2 + y^2 = R^2$ this gives $x + yy' = 0$ at once, without splitting into $y = \pm\sqrt{R^2 - x^2}$. Two questions remain: does $F = 0$ define such a function at all, and near which points? A level set need not be a graph globally (a circle is not), may be empty ($x^2 + y^2 + 1 = 0$; or the system $x^2 + y^2 + z^2 = 1$, $x + y + z = 2$, since the plane is at distance $2/\sqrt3 > 1$ from $0$), or may be a single point.

## 2\. The Theorem

Write points of $\mathbb{R}^{n+m}$ as $(x, y)$ with $x \in \mathbb{R}^n$, $y \in \mathbb{R}^m$, and for $F = (F_1, \dots, F_m)$ let $\frac{\partial F}{\partial x} = \big[\frac{\partial F_i}{\partial x_j}\big]$ ($m\times n$) and $\frac{\partial F}{\partial y} = \big[\frac{\partial F_i}{\partial y_j}\big]$ ($m\times m$).

::: {#thm-implicit-function}
## Implicit Function Theorem

Let $F : U \to \mathbb{R}^m$ be $C^1$ on an open $U \subseteq \mathbb{R}^{n+m}$, $F(a, b) = 0$, and $\det\frac{\partial F}{\partial y}(a, b) \ne 0$. Then there are open sets $A \ni a$ in $\mathbb{R}^n$ and $B \ni b$ in $\mathbb{R}^m$ and a $C^1$ map $\varphi : A \to B$ such that
$$\{(x, y) \in A\times B : F(x, y) = 0\} = \{(x, \varphi(x)) : x \in A\},$$
and
$$D\varphi(x) = -\Big[\frac{\partial F}{\partial y}(x, \varphi(x))\Big]^{-1}\frac{\partial F}{\partial x}(x, \varphi(x)) .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-implicit-function
Let $G(x, y) = (x, F(x, y))$, a $C^1$ map $U \to \mathbb{R}^{n+m}$ with
$$DG = \begin{bmatrix}I & 0\\ \frac{\partial F}{\partial x} & \frac{\partial F}{\partial y}\end{bmatrix}, \qquad \det DG(a, b) = \det\frac{\partial F}{\partial y}(a, b) \ne 0 .$$
By the [inverse function theorem](inverse-function-theorem.ipynb#thm-inverse-function), $G$ maps an open $V \ni (a, b)$ bijectively onto an open $W \ni (a, 0)$ with $C^1$ inverse $H$; since $G$ preserves the first coordinate, $H(x, z) = (x, h(x, z))$ with $h$ of class $C^1$. Choose open $A_0 \ni a$, $B \ni b$ with $A_0\times B \subseteq V$; the set $G(A_0\times B) = H^{-1}(A_0\times B)$ is open, so $A = \{x \in A_0 : (x, 0) \in G(A_0\times B)\}$ is an open set containing $a$. Define $\varphi(x) = h(x, 0)$ for $x \in A$. Then $(x, \varphi(x)) = H(x, 0) \in A_0\times B$ and $F(x, \varphi(x)) = 0$. Conversely, if $(x, y) \in A\times B$ and $F(x, y) = 0$, then $G(x, y) = (x, 0) = G(H(x, 0))$, and injectivity of $G$ on $V$ gives $y = \varphi(x)$. Finally, differentiate $F(x, \varphi(x)) = 0$ with the chain rule: $\frac{\partial F}{\partial x} + \frac{\partial F}{\partial y}D\varphi = 0$. $\blacksquare$
:::

For $m = n = 1$ the condition is $F_y(a, b) \ne 0$ and the formula is $\frac{dy}{dx} = -F_x/F_y$; for one equation $F(x_1, \dots, x_n, y) = 0$ it is $\frac{\partial y}{\partial x_j} = -\frac{\partial F/\partial x_j}{\partial F/\partial y}$.

::: {#exm-ift-condition}
## The Condition Is Sufficient, Not Necessary

1. $F = y^3 - x^4$: $F_y(0, 0) = 0$, yet $F = 0$ is the graph of the $C^1$ function $y = x^{4/3}$.
2. With $F_y(0, 0) = 0$ anything can happen: $x^2 + y^2 = 0$ is a single point; $x + y^2 = 0$ is a parabola that is not a graph $y = f(x)$ near $0$; $x + y^3 = 0$ is the graph of $y = -x^{1/3}$, which is not differentiable at $0$ (vertical tangent).
:::

## 3\. Smoothness of Level Sets

::: {#cor-level-set-smooth}
## Level Sets Near Regular Points

Let $g = (g_1, \dots, g_k) : U \to \mathbb{R}^k$ be $C^1$ on an open $U \subseteq \mathbb{R}^n$, $g(a) = c$, and suppose $\nabla g_1(a), \dots, \nabla g_k(a)$ are linearly independent. Then:

1. After renumbering the coordinates, $M = g^{-1}(c)$ coincides near $a$ with the graph of a $C^1$ function of $n - k$ of the variables (in this sense $M$ is **smooth** near $a$).
2. A vector $v$ is the velocity $\gamma'(0)$ of some $C^1$ curve $\gamma$ in $M$ with $\gamma(0) = a$ iff $Dg(a)v = 0$, i.e. iff $v \perp \nabla g_i(a)$ for all $i$. These vectors form the **tangent space** $\ker Dg(a)$, of dimension $n - k$.
:::

::: {.callout-tip collapse="true"}
## Proof of @cor-level-set-smooth
The $k\times n$ matrix $Dg(a)$ has rank $k$, so some $k$ of its columns are independent; renumber so that they are the last $k$ and write $x = (x', x'')$ with $x'' \in \mathbb{R}^k$. Apply @thm-implicit-function to $F(x', x'') = g(x) - c$: near $a$, $M$ is the graph $x'' = \varphi(x')$, which is (1).
(2) If $\gamma$ lies in $M$, differentiating $g(\gamma(t)) = c$ gives $Dg(a)\gamma'(0) = 0$. Conversely, for $w \in \mathbb{R}^{n-k}$ the curve $\gamma(t) = (a' + tw, \varphi(a' + tw))$ lies in $M$ with $\gamma'(0) = (w, D\varphi(a')w)$. The linear map $w \mapsto (w, D\varphi(a')w)$ is injective, with image in $\ker Dg(a)$, which has dimension $n - k$ by [rank–nullity](kernel-image-and-rank-nullity.ipynb#thm-rank-nullity); so the image is all of $\ker Dg(a)$. $\blacksquare$
:::

For $k = 1$: near a regular point the level set of $f$ is a smooth hypersurface whose tangent hyperplane is $\nabla f(a)^\perp$, as anticipated in [the gradient note](gradient.ipynb#def-critical-point).

## 4\. Examples

::: {#exm-ift-system}
## Choosing the Dependent Variables

The system $x^3y + 2y^2z^2 - xz^3 = 0$, $e^{y+z} - x = 0$ is satisfied at $(1, 0, 0)$. The Jacobian of $(F_1, F_2)$ with respect to $(x, y, z)$ at that point is
$$\begin{bmatrix}3x^2y - z^3 & x^3 + 4yz^2 & 4y^2z - 3xz^2\\ -1 & e^{y+z} & e^{y+z}\end{bmatrix} = \begin{bmatrix}0 & 1 & 0\\ -1 & 1 & 1\end{bmatrix} .$$
The $2\times2$ minors are $\det\frac{\partial(F_1, F_2)}{\partial(x, y)} = 1$, $\det\frac{\partial(F_1, F_2)}{\partial(x, z)} = 0$, $\det\frac{\partial(F_1, F_2)}{\partial(y, z)} = 1$. So near $(1, 0, 0)$ the solution curve can be written as $(x, y) = \varphi(z)$ (and also as $(y, z)$ in terms of $x$), with
$$\begin{bmatrix}x'(z)\\ y'(z)\end{bmatrix} = -\begin{bmatrix}0 & 1\\ -1 & 1\end{bmatrix}^{-1}\begin{bmatrix}0\\ 1\end{bmatrix} = \begin{bmatrix}1\\ 0\end{bmatrix} .$$
It cannot be written as $(x, z)$ in terms of $y$: differentiating with respect to $y$ at the point would give $1 = 0$ from the first equation, since there $\frac{dz}{dy}$ cannot exist (the curve is tangent to the direction $(1, 0, 1)$, along which $y$ is stationary).
:::

::: {#exm-cyclic-rule}
## The Cyclic Rule

If $F(x, y, z) = 0$ with all three partials non-zero at a point, each variable is locally a $C^1$ function of the other two and
$$\Big(\frac{\partial x}{\partial y}\Big)_z\Big(\frac{\partial y}{\partial z}\Big)_x\Big(\frac{\partial z}{\partial x}\Big)_y = \Big(-\frac{F_y}{F_x}\Big)\Big(-\frac{F_z}{F_y}\Big)\Big(-\frac{F_x}{F_z}\Big) = -1,$$
where the subscript names the variable held fixed. For the ideal gas $PV - kT = 0$: $\big(\frac{\partial P}{\partial V}\big)_T\big(\frac{\partial V}{\partial T}\big)_P\big(\frac{\partial T}{\partial P}\big)_V = -1$, not $+1$ as naive cancellation suggests. With $n$ variables the cyclic product is $(-1)^n$.
:::

```python

import sympy as sp
import numpy as np
import matplotlib.pyplot as plt

BLUE, ORANGE, AQUA, YELLOW, VIOLET, RED = "#2a78d6", "#eb6834", "#1baf7a", "#eda100", "#6250d6", "#e34948"
INK, MUTED, GRID = "#0b0b0b", "#52514e", "#d9d8d4"
plt.rcParams.update({
    "figure.facecolor": "white", "axes.facecolor": "white", "savefig.facecolor": "white",
    "axes.edgecolor": GRID, "axes.grid": True, "grid.color": GRID, "grid.linewidth": 0.6,
    "axes.spines.top": False, "axes.spines.right": False, "axes.titlesize": 11,
    "axes.titleweight": "bold", "axes.labelcolor": INK, "xtick.color": MUTED, "ytick.color": MUTED,
    "text.color": INK, "legend.frameon": False, "font.size": 10})

x, y, z = sp.symbols("x y z")
F = sp.Matrix([x**3 * y + 2 * y**2 * z**2 - x * z**3, sp.exp(y + z) - x])
J = F.jacobian([x, y, z]).subs({x: 1, y: 0, z: 0})
print("Jacobian at (1,0,0):", J.tolist())
print("(x'(z), y'(z)) =", list(-J[:, :2].inv() * J[:, 2]))
P, V, T, k = sp.symbols("P V T k", positive=True)
Pf, Vf, Tf = k * T / V, k * T / P, P * V / k
print("ideal gas cyclic product:", sp.simplify(sp.diff(Pf, V) * sp.diff(Vf, T) * sp.diff(Tf, P)).subs(T, P * V / k))

fig, ax = plt.subplots(figsize=(6, 5))
g = np.linspace(-2, 2, 500); X, Y = np.meshgrid(g, g)
Fv = X**3 + Y**3 - 3 * X * Y
ax.contour(X, Y, Fv, levels=[0], colors=[BLUE], linewidths=2)
for p in [(0, 0), (2**(2 / 3), 2**(1 / 3))]:
    ax.plot(*p, "o", color=ORANGE, ms=6)
ax.set_aspect("equal"); ax.set_title(r"$x^3 + y^3 = 3xy$: $y(x)$ fails only where $F_y = 0$")
plt.tight_layout(); plt.show()
```
