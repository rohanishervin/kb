# Properties of the Riemann Integral

**Prerequisites:** [The Riemann Integral](riemann-integral.ipynb)

All functions are bounded on $[a, b]$; "integrable" means Riemann integrable.

## 1\. Linearity and Order

::: {#thm-integral-linear}
## Linearity

1. A constant $c$ is integrable with $\int_a^b c = c(b - a)$.
2. If $f, g$ are integrable, so is $f + g$, and $\int_a^b (f + g) = \int_a^b f + \int_a^b g$.
3. If $f$ is integrable and $c \in \mathbb{R}$, so is $cf$, and $\int_a^b cf = c\int_a^b f$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-integral-linear
(1) Every lower and upper sum equals $c(b - a)$.

(2) On each subinterval $\inf f + \inf g \le \inf (f + g) \le \sup (f + g) \le \sup f + \sup g$, so
$$L(f, \mathcal{P}) + L(g, \mathcal{P}) \le L(f + g, \mathcal{P}) \le U(f + g, \mathcal{P}) \le U(f, \mathcal{P}) + U(g, \mathcal{P}).$$
Choose $\mathcal{P}$ (a common refinement) with $U - L < \frac\varepsilon2$ for both $f$ and $g$; then $U(f+g, \mathcal{P}) - L(f+g, \mathcal{P}) < \varepsilon$, so $f + g$ is integrable by [Riemann's criterion](riemann-integral.ipynb#thm-riemann-criterion). Both $\int (f+g)$ and $\int f + \int g$ lie in $[L(f,\mathcal{P}) + L(g,\mathcal{P}),\ U(f,\mathcal{P}) + U(g,\mathcal{P})]$, an interval of length $< \varepsilon$; so they are equal.

(3) For $c \ge 0$, $\inf cf = c\inf f$ and $\sup cf = c\sup f$, so $L(cf, \mathcal{P}) = cL(f, \mathcal{P})$ and $U(cf, \mathcal{P}) = cU(f, \mathcal{P})$. For $c < 0$, $\inf cf = c \sup f$ and $\sup cf = c\inf f$, so $L(cf, \mathcal{P}) = cU(f, \mathcal{P})$, $U(cf, \mathcal{P}) = cL(f, \mathcal{P})$. In both cases the criterion and the value follow. $\blacksquare$
:::

::: {#thm-integral-order}
## Monotonicity

If $f, g$ are integrable and $f \le g$, then $\int_a^b f \le \int_a^b g$. In particular, if $m \le f \le M$ then $m(b - a) \le \int_a^b f \le M(b - a)$, and if $|f - g| \le \varepsilon$ then $\big|\int_a^b f - \int_a^b g\big| \le \varepsilon(b - a)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-integral-order
If $h \ge 0$ is integrable, all lower sums are $\ge 0$, so $\int h \ge 0$. Apply this to $h = g - f$ and use linearity. The two special cases: compare with constants, and use $-\varepsilon \le f - g \le \varepsilon$. $\blacksquare$
:::

::: {#prp-integral-abs}
## Absolute Value

If $f$ is integrable, so is $|f|$, and $\big|\int_a^b f\big| \le \int_a^b |f|$.
:::

::: {.callout-tip collapse="true"}
## Proof of @prp-integral-abs
For $x, y$ in a subinterval, $\big||f(x)| - |f(y)|\big| \le |f(x) - f(y)| \le M_i - m_i$, so the oscillation of $|f|$ is at most that of $f$, and $U(|f|, \mathcal{P}) - L(|f|, \mathcal{P}) \le U(f, \mathcal{P}) - L(f, \mathcal{P})$. Then $-|f| \le f \le |f|$ and @thm-integral-order. $\blacksquare$
:::

## 2\. Additivity Over Intervals

For $a < b$ define $\int_b^a f = -\int_a^b f$, and $\int_a^a f = 0$. (If $f \ge 0$, reversing the orientation of the $x$-axis reverses the sign of the area.)

::: {#lem-restriction}
## Restriction to a Subinterval

If $f$ is integrable on $[a, b]$ and $[c, d] \subseteq [a, b]$, then $f$ is integrable on $[c, d]$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-restriction
Take $\mathcal{P}$ with $U(f, \mathcal{P}) - L(f, \mathcal{P}) < \varepsilon$ and add the points $c, d$ (this does not increase $U - L$). The points of $\mathcal{P}$ in $[c,d]$ form a partition $\bar{\mathcal{P}}$ of $[c, d]$, and $U(f|_{[c,d]}, \bar{\mathcal{P}}) - L(f|_{[c,d]}, \bar{\mathcal{P}})$ is a part of the sum $\sum (M_i - m_i)\Delta x_i \ge 0$, hence $< \varepsilon$. $\blacksquare$
:::

::: {#lem-extension-by-zero}
## Extension by Zero

If $f$ is integrable on $[a, b] \subseteq [A, B]$ and $\tilde f = f$ on $[a,b]$, $\tilde f = 0$ elsewhere on $[A, B]$, then $\tilde f$ is integrable on $[A, B]$ and $\int_A^B \tilde f = \int_a^b f$.
:::

::: {.callout-tip collapse="true"}
## Proof of @lem-extension-by-zero
For a partition $\mathcal{P}$ of $[a, b]$, let $\tilde{\mathcal{P}} = \mathcal{P} \cup \{A, B\}$. The new subintervals $[A, a]$ and $[b, B]$ contribute $0$ to both sums, so $L(\tilde f, \tilde{\mathcal{P}}) = L(f, \mathcal{P})$ and $U(\tilde f, \tilde{\mathcal{P}}) = U(f, \mathcal{P})$. Riemann's criterion gives integrability and the same value. $\blacksquare$
:::

::: {#thm-interval-additivity}
## Additivity

For any $a, b, c$ (in any order) in an interval on which $f$ is integrable,
$$\int_a^c f = \int_a^b f + \int_b^c f .$$
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-interval-additivity
First let $a \le b \le c$. By @lem-restriction, $f$ is integrable on $[a, b]$ and $[b, c]$. Let $f_1, f_2$ be the extensions by zero to $[a, c]$ of $f|_{[a, b)}$ and $f|_{[b, c]}$ (this is harmless: a function $h$ that vanishes except at one point $x_0$, where $h(x_0) = d$, is integrable with $\int h = 0$, because $x_0$ lies in at most two subintervals, so $|L(h, \mathcal{P})|, |U(h, \mathcal{P})| \le 2|d|\,\|\mathcal{P}\|$; by linearity, changing one value of an integrable function changes neither integrability nor the integral). Then $f = f_1 + f_2$ on $[a, c]$, and by linearity and @lem-extension-by-zero, $\int_a^c f = \int_a^b f + \int_b^c f$. The other orders follow from this one and the convention $\int_b^a = -\int_a^b$; e.g. for $a \le c \le b$: $\int_a^b = \int_a^c + \int_c^b$, i.e. $\int_a^c = \int_a^b - \int_c^b = \int_a^b + \int_b^c$. $\blacksquare$
:::

## 3\. Riemann Sums

For a partition $\mathcal{P}$ and points $x_i^* \in [x_{i-1}, x_i]$, the **Riemann sum** is $\sum_{i=1}^n f(x_i^*)\,\Delta x_i$. Since $m_i \le f(x_i^*) \le M_i$, it lies between $L(f, \mathcal{P})$ and $U(f, \mathcal{P})$.

::: {#thm-riemann-sums}
## Integrals as Limits of Riemann Sums

Let $f$ be continuous on $[a, b]$. For every $\varepsilon > 0$ there is $\delta > 0$ such that
$$\Big|\int_a^b f - \sum_{i=1}^n f(x_i^*)\,\Delta x_i\Big| < \varepsilon$$
for every partition with $\|\mathcal{P}\| < \delta$ and every choice of the points $x_i^*$. In particular, with equal subintervals, $\displaystyle \int_a^b f = \lim_{n\to\infty} \frac{b - a}{n}\sum_{i=1}^n f\Big(a + i\frac{b-a}{n}\Big)$.
:::

::: {.callout-tip collapse="true"}
## Proof of @thm-riemann-sums
By [this theorem](riemann-integral.ipynb#thm-continuous-integrable) there is $\delta$ with $U(f, \mathcal{P}) - L(f, \mathcal{P}) < \varepsilon$ when $\|\mathcal{P}\| < \delta$. Both the integral and the Riemann sum lie in $[L(f, \mathcal{P}), U(f, \mathcal{P})]$. $\blacksquare$
:::

Limits of sums can therefore be computed as integrals, once integrals can be computed; see the [fundamental theorem](fundamental-theorem-of-calculus.ipynb#exm-sums-as-integrals). (The theorem holds for every integrable $f$, but the proof is longer.)
